```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = swings[swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "TUPRS.IS"
    raw = Ticker(symbol).history(period="60d", interval="5m")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=20,
        confirm_candles=3,
    )

    print(swings_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "TUPRS.IS"
    raw = Ticker(symbol).history(period="60d", interval="15m")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 3.0,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.65,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    neckline_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = True,
    require_bullish_structure_break: bool = True,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - neckline is the swing high after B1
    - B2 is close to B1, default within 3%
    - pattern completes only after a close above neckline
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "NecklineBarIdx",
        "NecklineDate",
        "NecklinePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "RetracePct",
        "DowntrendPct",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []

    for pos in range(len(ordered_swings) - 2):
        bottom1 = ordered_swings.iloc[pos]
        neckline = ordered_swings.iloc[pos + 1]
        bottom2 = ordered_swings.iloc[pos + 2]

        if (
            bottom1["SwingType"] != "Low"
            or neckline["SwingType"] != "High"
            or bottom2["SwingType"] != "Low"
        ):
            continue

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        neck_idx = int(neckline["BarIdx"])
        b2_idx = int(bottom2["BarIdx"])
        b1_price = float(bottom1["Price"])
        neck_price = float(neckline["Price"])
        b2_price = float(bottom2["Price"])

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        # B1 should be the lowest point of the current downward leg when formed.
        leg_lows = lows[context_idx : b1_idx + 1]
        if len(leg_lows) and b1_price > float(np.min(leg_lows)) * 1.001:
            continue

        retrace_pct = (neck_price - b1_price) / (context_price - b1_price)
        if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
            continue

        bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
        if bottom_difference_pct > bottom_tolerance_pct:
            continue

        if neck_price <= max(b1_price, b2_price):
            continue

        known_after_idx = max(
            int(bottom1["ConfirmBarIdx"]),
            int(neckline["ConfirmBarIdx"]),
            int(bottom2["ConfirmBarIdx"]),
        )
        breakout_level = neck_price * (1.0 + neckline_break_buffer_pct / 100.0)
        fail_level = min(b1_price, b2_price) * (1.0 - fail_buffer_pct / 100.0)
        entry_idx: int | None = None

        for current_idx in range(known_after_idx, min(len(data), known_after_idx + max_entry_wait_bars + 1)):
            if closes[current_idx] < fail_level:
                break
            if closes[current_idx] > breakout_level:
                entry_idx = current_idx
                break

        if entry_idx is None:
            continue

        bullish_structure_break = _has_bullish_structure_break(
            structure_events,
            start_idx=b2_idx,
            end_idx=entry_idx,
        )
        if require_bullish_structure_break and not bullish_structure_break:
            continue

        entry_volume_ratio = (
            float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
        )
        volume_confirmed = (
            bool(np.isfinite(entry_volume_ratio) and entry_volume_ratio >= volume_multiplier)
            if has_volume
            else False
        )
        if require_volume_confirmation and not volume_confirmed:
            continue

        pattern_height = neck_price - ((b1_price + b2_price) / 2.0)
        target_price = neck_price + pattern_height
        stop_price = min(b1_price, b2_price)
        risk = closes[entry_idx] - stop_price
        reward = target_price - closes[entry_idx]

        patterns.append(
            {
                "PatternId": len(patterns) + 1,
                "ContextHighBarIdx": context_idx,
                "ContextHighDate": context["Date"],
                "ContextHighPrice": context_price,
                "Bottom1BarIdx": b1_idx,
                "Bottom1Date": bottom1["Date"],
                "Bottom1Price": b1_price,
                "NecklineBarIdx": neck_idx,
                "NecklineDate": neckline["Date"],
                "NecklinePrice": neck_price,
                "Bottom2BarIdx": b2_idx,
                "Bottom2Date": bottom2["Date"],
                "Bottom2Price": b2_price,
                "EntryBarIdx": entry_idx,
                "EntryDate": data.index[entry_idx],
                "EntryPrice": float(closes[entry_idx]),
                "TargetPrice": float(target_price),
                "StopPrice": float(stop_price),
                "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                "BottomDifferencePct": float(bottom_difference_pct),
                "RetracePct": float(retrace_pct),
                "DowntrendPct": float(downtrend_pct),
                "BullishStructureBreak": bool(bullish_structure_break),
                "VolumeRatio": entry_volume_ratio,
                "VolumeConfirmed": bool(volume_confirmed),
            }
        )

    return pd.DataFrame(patterns, columns=columns)


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for _, pattern in double_bottoms.iterrows():
            point_indexes = [
                int(pattern["ContextHighBarIdx"]),
                int(pattern["Bottom1BarIdx"]),
                int(pattern["NecklineBarIdx"]),
                int(pattern["Bottom2BarIdx"]),
                int(pattern["EntryBarIdx"]),
            ]
            point_prices = [
                float(pattern["ContextHighPrice"]),
                float(pattern["Bottom1Price"]),
                float(pattern["NecklinePrice"]),
                float(pattern["Bottom2Price"]),
                float(pattern["EntryPrice"]),
            ]
            point_labels = ["Context", "B1", "Neck", "B2", "Entry"]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=[
                        "top center",
                        "bottom center",
                        "top center",
                        "bottom center",
                        "top center",
                    ],
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            line_x = (
                [int(pattern["NecklineBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["NecklineDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Neckline", float(pattern["NecklinePrice"]), "#F8FAFC", "solid"),
                ("Support", min(float(pattern["Bottom1Price"]), float(pattern["Bottom2Price"])), "#22C55E", "dot"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name in {"Neckline", "Support"} else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "TUPRS.IS"
    raw = Ticker(symbol).history(period="60d", interval="5m")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=3.0,
        min_retrace_pct=0.38,
        max_retrace_pct=0.65,
        min_downtrend_pct=2.0,
        neckline_break_buffer_pct=0.03,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.75,
    support_zone_pct: float = 0.50,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.65,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    neckline_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = True,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - neckline is the swing high after B1
    - B2 is close to B1, default within 0.75%
    - B1 support zone and neckline resistance zone stay fixed
    - pattern completes only after a close above neckline
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "NecklineBarIdx",
        "NecklineDate",
        "NecklinePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []

    for pos in range(len(ordered_swings) - 2):
        bottom1 = ordered_swings.iloc[pos]
        neckline = ordered_swings.iloc[pos + 1]
        bottom2 = ordered_swings.iloc[pos + 2]

        if (
            bottom1["SwingType"] != "Low"
            or neckline["SwingType"] != "High"
            or bottom2["SwingType"] != "Low"
        ):
            continue

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        neck_idx = int(neckline["BarIdx"])
        b2_idx = int(bottom2["BarIdx"])
        b1_price = float(bottom1["Price"])
        neck_price = float(neckline["Price"])
        b2_price = float(bottom2["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        bearish_choch_before_b1 = _has_bearish_choch_before(
            structure_events,
            b1_idx=b1_idx,
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        # The neckline peak between bottoms must not exceed the high before B1.
        if neck_price > previous_high_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        # B1 should be the lowest point of the current downward leg when formed.
        leg_lows = lows[context_idx : b1_idx + 1]
        if len(leg_lows) and b1_price > float(np.min(leg_lows)) * 1.001:
            continue

        retrace_pct = (neck_price - b1_price) / (context_price - b1_price)
        if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
            continue

        bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
        if bottom_difference_pct > bottom_tolerance_pct:
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)
        if not support_zone_bottom <= b2_price <= support_zone_top:
            continue

        resistance_zone_bottom = neck_price * (1.0 - resistance_zone_pct / 100.0)
        resistance_zone_top = neck_price * (1.0 + resistance_zone_pct / 100.0)

        if neck_price <= max(b1_price, b2_price):
            continue

        known_after_idx = max(
            int(bottom1["ConfirmBarIdx"]),
            int(neckline["ConfirmBarIdx"]),
            int(bottom2["ConfirmBarIdx"]),
        )
        breakout_level = neck_price * (1.0 + neckline_break_buffer_pct / 100.0)
        fail_level = min(b1_price, b2_price) * (1.0 - fail_buffer_pct / 100.0)
        entry_idx: int | None = None

        for current_idx in range(known_after_idx, min(len(data), known_after_idx + max_entry_wait_bars + 1)):
            if closes[current_idx] < fail_level:
                break
            if closes[current_idx] > breakout_level:
                entry_idx = current_idx
                break

        if entry_idx is None:
            continue

        second_peak_invalidation = _has_second_peak_between_zones(
            ordered_swings,
            start_idx=neck_idx,
            end_idx=entry_idx,
            support_zone_top=support_zone_top,
            resistance_zone_bottom=resistance_zone_bottom,
        )
        if invalidate_second_peak_between_zones and second_peak_invalidation:
            continue

        bullish_structure_break = _has_bullish_structure_break(
            structure_events,
            start_idx=b2_idx,
            end_idx=entry_idx,
        )
        if require_bullish_structure_break and not bullish_structure_break:
            continue

        entry_volume_ratio = (
            float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
        )
        volume_confirmed = (
            bool(np.isfinite(entry_volume_ratio) and entry_volume_ratio >= volume_multiplier)
            if has_volume
            else False
        )
        if require_volume_confirmation and not volume_confirmed:
            continue

        pattern_height = neck_price - ((b1_price + b2_price) / 2.0)
        target_price = neck_price + pattern_height
        stop_price = min(b1_price, b2_price)
        risk = closes[entry_idx] - stop_price
        reward = target_price - closes[entry_idx]

        patterns.append(
            {
                "PatternId": len(patterns) + 1,
                "ContextHighBarIdx": context_idx,
                "ContextHighDate": context["Date"],
                "ContextHighPrice": context_price,
                "PreviousHighBarIdx": previous_high_idx,
                "PreviousHighDate": previous_high["Date"],
                "PreviousHighPrice": previous_high_price,
                "Bottom1BarIdx": b1_idx,
                "Bottom1Date": bottom1["Date"],
                "Bottom1Price": b1_price,
                "NecklineBarIdx": neck_idx,
                "NecklineDate": neckline["Date"],
                "NecklinePrice": neck_price,
                "Bottom2BarIdx": b2_idx,
                "Bottom2Date": bottom2["Date"],
                "Bottom2Price": b2_price,
                "EntryBarIdx": entry_idx,
                "EntryDate": data.index[entry_idx],
                "EntryPrice": float(closes[entry_idx]),
                "TargetPrice": float(target_price),
                "StopPrice": float(stop_price),
                "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                "BottomDifferencePct": float(bottom_difference_pct),
                "SupportZoneBottom": float(support_zone_bottom),
                "SupportZoneTop": float(support_zone_top),
                "ResistanceZoneBottom": float(resistance_zone_bottom),
                "ResistanceZoneTop": float(resistance_zone_top),
                "RetracePct": float(retrace_pct),
                "DowntrendPct": float(downtrend_pct),
                "PriorLHLL": bool(prior_lh_ll),
                "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                "SecondPeakInvalidation": bool(second_peak_invalidation),
                "BullishStructureBreak": bool(bullish_structure_break),
                "VolumeRatio": entry_volume_ratio,
                "VolumeConfirmed": bool(volume_confirmed),
            }
        )

    return pd.DataFrame(patterns, columns=columns)


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _has_bearish_choch_before(
    structure_events: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) < int(b1_idx))
    ]
    return not subset.empty


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if support_zone_top < price < resistance_zone_bottom:
            return True

    return False


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for _, pattern in double_bottoms.iterrows():
            point_indexes = [
                int(pattern["ContextHighBarIdx"]),
                int(pattern["Bottom1BarIdx"]),
                int(pattern["NecklineBarIdx"]),
                int(pattern["Bottom2BarIdx"]),
                int(pattern["EntryBarIdx"]),
            ]
            point_prices = [
                float(pattern["ContextHighPrice"]),
                float(pattern["Bottom1Price"]),
                float(pattern["NecklinePrice"]),
                float(pattern["Bottom2Price"]),
                float(pattern["EntryPrice"]),
            ]
            point_labels = ["Context", "B1", "Neck", "B2", "Entry"]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=[
                        "top center",
                        "bottom center",
                        "top center",
                        "bottom center",
                        "top center",
                    ],
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["NecklineBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["NecklineDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Neckline", float(pattern["NecklinePrice"]), "#F8FAFC", "solid"),
                ("Support", min(float(pattern["Bottom1Price"]), float(pattern["Bottom2Price"])), "#22C55E", "dot"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name in {"Neckline", "Support"} else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "AAPL"
    raw = Ticker(symbol).history(period="60d", interval="5m")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.75,
        support_zone_pct=0.50,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.65,
        min_downtrend_pct=2.0,
        neckline_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.65,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = True,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []

    for pos in range(len(ordered_swings) - 2):
        bottom1 = ordered_swings.iloc[pos]
        resistance = ordered_swings.iloc[pos + 1]
        bottom2 = ordered_swings.iloc[pos + 2]

        if (
            bottom1["SwingType"] != "Low"
            or resistance["SwingType"] != "High"
            or bottom2["SwingType"] != "Low"
        ):
            continue

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        resistance_idx = int(resistance["BarIdx"])
        b2_idx = int(bottom2["BarIdx"])
        b1_price = float(bottom1["Price"])
        resistance_price = float(resistance["Price"])
        b2_price = float(bottom2["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        bearish_choch_before_b1 = _has_bearish_choch_before(
            structure_events,
            b1_idx=b1_idx,
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        # The resistance peak between bottoms must not exceed the high before B1.
        if resistance_price > previous_high_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        # B1 should be the lowest point of the current downward leg when formed.
        leg_lows = lows[context_idx : b1_idx + 1]
        if len(leg_lows) and b1_price > float(np.min(leg_lows)) * 1.001:
            continue

        retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
        if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
            continue

        bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
        if bottom_difference_pct > bottom_tolerance_pct:
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)
        if not support_zone_bottom <= b2_price <= support_zone_top:
            continue

        resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
        resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)

        if resistance_price <= max(b1_price, b2_price):
            continue

        known_after_idx = max(
            int(bottom1["ConfirmBarIdx"]),
            int(resistance["ConfirmBarIdx"]),
            int(bottom2["ConfirmBarIdx"]),
        )
        breakout_level = resistance_zone_top * (
            1.0 + resistance_break_buffer_pct / 100.0
        )
        fail_level = min(b1_price, b2_price) * (1.0 - fail_buffer_pct / 100.0)
        entry_idx: int | None = None

        for current_idx in range(known_after_idx, min(len(data), known_after_idx + max_entry_wait_bars + 1)):
            if closes[current_idx] < fail_level:
                break
            if closes[current_idx] > breakout_level:
                entry_idx = current_idx
                break

        if entry_idx is None:
            continue

        second_peak_invalidation = _has_second_peak_between_zones(
            ordered_swings,
            start_idx=resistance_idx,
            end_idx=entry_idx,
            support_zone_top=support_zone_top,
            resistance_zone_bottom=resistance_zone_bottom,
        )
        if invalidate_second_peak_between_zones and second_peak_invalidation:
            continue

        post_b2_swing_invalidation = _has_any_swing_between(
            ordered_swings,
            start_idx=b2_idx,
            end_idx=entry_idx,
        )
        if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
            continue

        bullish_structure_break = _has_bullish_structure_break(
            structure_events,
            start_idx=b2_idx,
            end_idx=entry_idx,
        )
        if require_bullish_structure_break and not bullish_structure_break:
            continue

        entry_volume_ratio = (
            float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
        )
        volume_confirmed = (
            bool(np.isfinite(entry_volume_ratio) and entry_volume_ratio >= volume_multiplier)
            if has_volume
            else False
        )
        if require_volume_confirmation and not volume_confirmed:
            continue

        pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
        target_price = resistance_zone_top + pattern_height
        stop_price = min(b1_price, b2_price)
        risk = closes[entry_idx] - stop_price
        reward = target_price - closes[entry_idx]

        patterns.append(
            {
                "PatternId": len(patterns) + 1,
                "ContextHighBarIdx": context_idx,
                "ContextHighDate": context["Date"],
                "ContextHighPrice": context_price,
                "PreviousHighBarIdx": previous_high_idx,
                "PreviousHighDate": previous_high["Date"],
                "PreviousHighPrice": previous_high_price,
                "Bottom1BarIdx": b1_idx,
                "Bottom1Date": bottom1["Date"],
                "Bottom1Price": b1_price,
                "ResistanceBarIdx": resistance_idx,
                "ResistanceDate": resistance["Date"],
                "ResistancePrice": resistance_price,
                "Bottom2BarIdx": b2_idx,
                "Bottom2Date": bottom2["Date"],
                "Bottom2Price": b2_price,
                "EntryBarIdx": entry_idx,
                "EntryDate": data.index[entry_idx],
                "EntryPrice": float(closes[entry_idx]),
                "TargetPrice": float(target_price),
                "StopPrice": float(stop_price),
                "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                "BottomDifferencePct": float(bottom_difference_pct),
                "SupportZoneBottom": float(support_zone_bottom),
                "SupportZoneTop": float(support_zone_top),
                "ResistanceZoneBottom": float(resistance_zone_bottom),
                "ResistanceZoneTop": float(resistance_zone_top),
                "RetracePct": float(retrace_pct),
                "DowntrendPct": float(downtrend_pct),
                "PriorLHLL": bool(prior_lh_ll),
                "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                "SecondPeakInvalidation": bool(second_peak_invalidation),
                "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                "BullishStructureBreak": bool(bullish_structure_break),
                "VolumeRatio": entry_volume_ratio,
                "VolumeConfirmed": bool(volume_confirmed),
            }
        )

    return pd.DataFrame(patterns, columns=columns)


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _has_bearish_choch_before(
    structure_events: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) < int(b1_idx))
    ]
    return not subset.empty


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if support_zone_top < price < resistance_zone_bottom:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for _, pattern in double_bottoms.iterrows():
            point_indexes = [
                int(pattern["ContextHighBarIdx"]),
                int(pattern["Bottom1BarIdx"]),
                int(pattern["ResistanceBarIdx"]),
                int(pattern["Bottom2BarIdx"]),
                int(pattern["EntryBarIdx"]),
            ]
            point_prices = [
                float(pattern["ContextHighPrice"]),
                float(pattern["Bottom1Price"]),
                float(pattern["ResistancePrice"]),
                float(pattern["Bottom2Price"]),
                float(pattern["EntryPrice"]),
            ]
            point_labels = ["Context", "B1", "Res", "B2", "Entry"]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=[
                        "top center",
                        "bottom center",
                        "top center",
                        "bottom center",
                        "top center",
                    ],
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance", float(pattern["ResistancePrice"]), "#F8FAFC", "solid"),
                ("Support", min(float(pattern["Bottom1Price"]), float(pattern["Bottom2Price"])), "#22C55E", "dot"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name in {"Resistance", "Support"} else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "GARAN.IS"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.65,
        min_downtrend_pct=2.0,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.65,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = True,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []

    for pos in range(len(ordered_swings) - 2):
        bottom1 = ordered_swings.iloc[pos]
        resistance = ordered_swings.iloc[pos + 1]
        bottom2 = ordered_swings.iloc[pos + 2]

        if (
            bottom1["SwingType"] != "Low"
            or resistance["SwingType"] != "High"
            or bottom2["SwingType"] != "Low"
        ):
            continue

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        resistance_idx = int(resistance["BarIdx"])
        b2_idx = int(bottom2["BarIdx"])
        b1_price = float(bottom1["Price"])
        resistance_price = float(resistance["Price"])
        b2_price = float(bottom2["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        bearish_choch_before_b1 = _has_bearish_choch_before(
            structure_events,
            b1_idx=b1_idx,
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        # The resistance peak between bottoms must not exceed the high before B1.
        if resistance_price > previous_high_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        # B1 should be the lowest point of the current downward leg when formed.
        leg_lows = lows[context_idx : b1_idx + 1]
        if len(leg_lows) and b1_price > float(np.min(leg_lows)) * 1.001:
            continue

        retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
        if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
            continue

        bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
        if bottom_difference_pct > bottom_tolerance_pct:
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)
        if not support_zone_bottom <= b2_price <= support_zone_top:
            continue

        resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
        resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)

        if resistance_price <= max(b1_price, b2_price):
            continue

        known_after_idx = max(
            int(bottom1["ConfirmBarIdx"]),
            int(resistance["ConfirmBarIdx"]),
            int(bottom2["ConfirmBarIdx"]),
        )
        breakout_level = resistance_zone_top * (
            1.0 + resistance_break_buffer_pct / 100.0
        )
        fail_level = min(b1_price, b2_price) * (1.0 - fail_buffer_pct / 100.0)
        entry_idx: int | None = None

        for current_idx in range(known_after_idx, min(len(data), known_after_idx + max_entry_wait_bars + 1)):
            if closes[current_idx] < fail_level:
                break
            if closes[current_idx] > breakout_level:
                entry_idx = current_idx
                break

        if entry_idx is None:
            continue

        second_peak_invalidation = _has_second_peak_between_zones(
            ordered_swings,
            start_idx=resistance_idx,
            end_idx=entry_idx,
            support_zone_top=support_zone_top,
            resistance_zone_bottom=resistance_zone_bottom,
        )
        if invalidate_second_peak_between_zones and second_peak_invalidation:
            continue

        post_b2_swing_invalidation = _has_any_swing_between(
            ordered_swings,
            start_idx=b2_idx,
            end_idx=entry_idx,
        )
        if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
            continue

        bullish_structure_break = _has_bullish_structure_break(
            structure_events,
            start_idx=b2_idx,
            end_idx=entry_idx,
        )
        if require_bullish_structure_break and not bullish_structure_break:
            continue

        pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
        target_price = resistance_zone_top + pattern_height
        follow_through_level = resistance_zone_top + (
            pattern_height * follow_through_pattern_ratio
        )
        follow_through_idx = _find_breakout_follow_through(
            closes=closes,
            entry_idx=entry_idx,
            max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
            follow_through_level=follow_through_level,
            support_zone_bottom=support_zone_bottom,
        )
        if require_breakout_follow_through and follow_through_idx is None:
            continue

        entry_volume_ratio = (
            float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
        )
        volume_confirmed = (
            bool(np.isfinite(entry_volume_ratio) and entry_volume_ratio >= volume_multiplier)
            if has_volume
            else False
        )
        if require_volume_confirmation and not volume_confirmed:
            continue

        stop_price = min(b1_price, b2_price)
        risk = closes[entry_idx] - stop_price
        reward = target_price - closes[entry_idx]

        patterns.append(
            {
                "PatternId": len(patterns) + 1,
                "ContextHighBarIdx": context_idx,
                "ContextHighDate": context["Date"],
                "ContextHighPrice": context_price,
                "PreviousHighBarIdx": previous_high_idx,
                "PreviousHighDate": previous_high["Date"],
                "PreviousHighPrice": previous_high_price,
                "Bottom1BarIdx": b1_idx,
                "Bottom1Date": bottom1["Date"],
                "Bottom1Price": b1_price,
                "ResistanceBarIdx": resistance_idx,
                "ResistanceDate": resistance["Date"],
                "ResistancePrice": resistance_price,
                "Bottom2BarIdx": b2_idx,
                "Bottom2Date": bottom2["Date"],
                "Bottom2Price": b2_price,
                "EntryBarIdx": entry_idx,
                "EntryDate": data.index[entry_idx],
                "EntryPrice": float(closes[entry_idx]),
                "FollowThroughBarIdx": int(follow_through_idx)
                if follow_through_idx is not None
                else np.nan,
                "FollowThroughDate": data.index[int(follow_through_idx)]
                if follow_through_idx is not None
                else pd.NaT,
                "FollowThroughPrice": float(closes[int(follow_through_idx)])
                if follow_through_idx is not None
                else np.nan,
                "FollowThroughLevel": float(follow_through_level),
                "TargetPrice": float(target_price),
                "StopPrice": float(stop_price),
                "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                "BottomDifferencePct": float(bottom_difference_pct),
                "SupportZoneBottom": float(support_zone_bottom),
                "SupportZoneTop": float(support_zone_top),
                "ResistanceZoneBottom": float(resistance_zone_bottom),
                "ResistanceZoneTop": float(resistance_zone_top),
                "RetracePct": float(retrace_pct),
                "DowntrendPct": float(downtrend_pct),
                "PriorLHLL": bool(prior_lh_ll),
                "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                "SecondPeakInvalidation": bool(second_peak_invalidation),
                "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                "BullishStructureBreak": bool(bullish_structure_break),
                "VolumeRatio": entry_volume_ratio,
                "VolumeConfirmed": bool(volume_confirmed),
            }
        )

    return pd.DataFrame(patterns, columns=columns)


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _has_bearish_choch_before(
    structure_events: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) < int(b1_idx))
    ]
    return not subset.empty


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if support_zone_top < price < resistance_zone_bottom:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for _, pattern in double_bottoms.iterrows():
            point_indexes = [
                int(pattern["Bottom1BarIdx"]),
                int(pattern["ResistanceBarIdx"]),
                int(pattern["Bottom2BarIdx"]),
                int(pattern["EntryBarIdx"]),
            ]
            point_prices = [
                float(pattern["Bottom1Price"]),
                float(pattern["ResistancePrice"]),
                float(pattern["Bottom2Price"]),
                float(pattern["EntryPrice"]),
            ]
            point_labels = ["B1", "Resistance", "B2", "Breakout"]
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=[
                        "bottom center",
                        "top center",
                        "bottom center",
                        "top center",
                        "top center",
                    ],
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "GARAN.IS"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.65,
        min_downtrend_pct=2.0,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        bearish_choch_before_b1 = _has_bearish_choch_before(
            structure_events,
            end_idx=int(bottom1["ConfirmBarIdx"]),
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        for b2_pos in range(b1_pos + 1, len(lows_only)):
            bottom2 = lows_only.iloc[b2_pos]
            b2_idx = int(bottom2["BarIdx"])
            bars_between = b2_idx - b1_idx
            if bars_between < min_bars_between_bottoms:
                continue
            if bars_between > max_bars_between_bottoms:
                break

            b2_price = float(bottom2["Price"])
            bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
            if bottom_difference_pct > bottom_tolerance_pct:
                continue
            if not support_zone_bottom <= b2_price <= support_zone_top:
                continue

            highs_between = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                & (ordered_swings["BarIdx"].astype(int) < b2_idx)
            ]
            if highs_between.empty:
                continue

            resistance = (
                highs_between.sort_values(["Price", "BarIdx"], ascending=[False, False])
                .iloc[0]
            )
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= max(b1_price, b2_price):
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)

            second_peak_invalidation = _has_second_peak_between_zones(
                ordered_swings,
                start_idx=resistance_idx,
                end_idx=b2_idx,
                support_zone_top=support_zone_top,
                resistance_zone_bottom=resistance_zone_bottom,
            )
            if invalidate_second_peak_between_zones and second_peak_invalidation:
                continue

            known_after_idx = max(
                int(bottom1["ConfirmBarIdx"]),
                int(resistance["ConfirmBarIdx"]),
                int(bottom2["ConfirmBarIdx"]),
            )
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            fail_level = min(b1_price, b2_price) * (1.0 - fail_buffer_pct / 100.0)
            entry_idx: int | None = None

            for current_idx in range(
                known_after_idx,
                min(len(data), known_after_idx + max_entry_wait_bars + 1),
            ):
                if closes[current_idx] < fail_level:
                    break
                if closes[current_idx] > breakout_level:
                    entry_idx = current_idx
                    break

            if entry_idx is None:
                continue

            post_b2_swing_invalidation = _has_any_swing_between(
                ordered_swings,
                start_idx=b2_idx,
                end_idx=entry_idx,
            )
            if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                continue

            bullish_structure_break = _has_bullish_structure_break(
                structure_events,
                start_idx=b2_idx,
                end_idx=entry_idx,
            )
            if require_bullish_structure_break and not bullish_structure_break:
                continue

            pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
            target_price = resistance_zone_top + pattern_height
            follow_through_level = resistance_zone_top + (
                pattern_height * follow_through_pattern_ratio
            )
            follow_through_idx = _find_breakout_follow_through(
                closes=closes,
                entry_idx=entry_idx,
                max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                follow_through_level=follow_through_level,
                support_zone_bottom=support_zone_bottom,
            )
            if require_breakout_follow_through and follow_through_idx is None:
                continue

            entry_volume_ratio = (
                float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
            )
            volume_confirmed = (
                bool(np.isfinite(entry_volume_ratio) and entry_volume_ratio >= volume_multiplier)
                if has_volume
                else False
            )
            if require_volume_confirmation and not volume_confirmed:
                continue

            stop_price = min(b1_price, b2_price)
            risk = closes[entry_idx] - stop_price
            reward = target_price - closes[entry_idx]

            patterns.append(
                {
                    "PatternId": len(patterns) + 1,
                    "ContextHighBarIdx": context_idx,
                    "ContextHighDate": context["Date"],
                    "ContextHighPrice": context_price,
                    "PreviousHighBarIdx": previous_high_idx,
                    "PreviousHighDate": previous_high["Date"],
                    "PreviousHighPrice": previous_high_price,
                    "Bottom1BarIdx": b1_idx,
                    "Bottom1Date": bottom1["Date"],
                    "Bottom1Price": b1_price,
                    "ResistanceBarIdx": resistance_idx,
                    "ResistanceDate": resistance["Date"],
                    "ResistancePrice": resistance_price,
                    "Bottom2BarIdx": b2_idx,
                    "Bottom2Date": bottom2["Date"],
                    "Bottom2Price": b2_price,
                    "EntryBarIdx": entry_idx,
                    "EntryDate": data.index[entry_idx],
                    "EntryPrice": float(closes[entry_idx]),
                    "FollowThroughBarIdx": int(follow_through_idx)
                    if follow_through_idx is not None
                    else np.nan,
                    "FollowThroughDate": data.index[int(follow_through_idx)]
                    if follow_through_idx is not None
                    else pd.NaT,
                    "FollowThroughPrice": float(closes[int(follow_through_idx)])
                    if follow_through_idx is not None
                    else np.nan,
                    "FollowThroughLevel": float(follow_through_level),
                    "TargetPrice": float(target_price),
                    "StopPrice": float(stop_price),
                    "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                    "BottomDifferencePct": float(bottom_difference_pct),
                    "SupportZoneBottom": float(support_zone_bottom),
                    "SupportZoneTop": float(support_zone_top),
                    "ResistanceZoneBottom": float(resistance_zone_bottom),
                    "ResistanceZoneTop": float(resistance_zone_top),
                    "RetracePct": float(retrace_pct),
                    "DowntrendPct": float(downtrend_pct),
                    "PriorLHLL": bool(prior_lh_ll),
                    "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                    "SecondPeakInvalidation": bool(second_peak_invalidation),
                    "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                    "BullishStructureBreak": bool(bullish_structure_break),
                    "VolumeRatio": entry_volume_ratio,
                    "VolumeConfirmed": bool(volume_confirmed),
                }
            )

    return pd.DataFrame(patterns, columns=columns)


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _has_bearish_choch_before(
    structure_events: pd.DataFrame,
    *,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
    ]
    return not subset.empty


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if support_zone_top < price < resistance_zone_bottom:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for _, pattern in double_bottoms.iterrows():
            point_indexes = [
                int(pattern["Bottom1BarIdx"]),
                int(pattern["ResistanceBarIdx"]),
                int(pattern["Bottom2BarIdx"]),
                int(pattern["EntryBarIdx"]),
            ]
            point_prices = [
                float(pattern["Bottom1Price"]),
                float(pattern["ResistancePrice"]),
                float(pattern["Bottom2Price"]),
                float(pattern["EntryPrice"]),
            ]
            point_labels = ["B1", "Resistance", "B2", "Breakout"]
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=[
                        "bottom center",
                        "top center",
                        "bottom center",
                        "top center",
                        "top center",
                    ],
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "GARAN.IS"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        bearish_choch_before_b1 = _has_bearish_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        for b2_pos in range(b1_pos + 1, len(lows_only)):
            bottom2 = lows_only.iloc[b2_pos]
            b2_idx = int(bottom2["BarIdx"])
            bars_between = b2_idx - b1_idx
            if bars_between < min_bars_between_bottoms:
                continue
            if bars_between > max_bars_between_bottoms:
                break

            b2_price = float(bottom2["Price"])
            bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
            if bottom_difference_pct > bottom_tolerance_pct:
                continue
            if not support_zone_bottom <= b2_price <= support_zone_top:
                continue

            highs_between = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                & (ordered_swings["BarIdx"].astype(int) < b2_idx)
            ]
            if highs_between.empty:
                continue

            resistance = (
                highs_between.sort_values(["Price", "BarIdx"], ascending=[False, False])
                .iloc[0]
            )
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= max(b1_price, b2_price):
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)

            second_peak_invalidation = _has_second_peak_between_zones(
                ordered_swings,
                start_idx=resistance_idx,
                end_idx=b2_idx,
                support_zone_top=support_zone_top,
                resistance_zone_bottom=resistance_zone_bottom,
            )
            if invalidate_second_peak_between_zones and second_peak_invalidation:
                continue

            known_after_idx = max(
                int(bottom1["ConfirmBarIdx"]),
                int(resistance["ConfirmBarIdx"]),
                int(bottom2["ConfirmBarIdx"]),
            )
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            fail_level = min(b1_price, b2_price) * (1.0 - fail_buffer_pct / 100.0)
            entry_idx: int | None = None

            for current_idx in range(
                known_after_idx,
                min(len(data), known_after_idx + max_entry_wait_bars + 1),
            ):
                if closes[current_idx] < fail_level:
                    break
                if closes[current_idx] > breakout_level:
                    entry_idx = current_idx
                    break

            if entry_idx is None:
                continue

            post_b2_swing_invalidation = _has_any_swing_between(
                ordered_swings,
                start_idx=b2_idx,
                end_idx=entry_idx,
            )
            if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                continue

            bullish_structure_break = _has_bullish_structure_break(
                structure_events,
                start_idx=b2_idx,
                end_idx=entry_idx,
            )
            if require_bullish_structure_break and not bullish_structure_break:
                continue

            pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
            target_price = resistance_zone_top + pattern_height
            follow_through_level = resistance_zone_top + (
                pattern_height * follow_through_pattern_ratio
            )
            follow_through_idx = _find_breakout_follow_through(
                closes=closes,
                entry_idx=entry_idx,
                max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                follow_through_level=follow_through_level,
                support_zone_bottom=support_zone_bottom,
            )
            if require_breakout_follow_through and follow_through_idx is None:
                continue

            entry_volume_ratio = (
                float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
            )
            volume_confirmed = (
                bool(np.isfinite(entry_volume_ratio) and entry_volume_ratio >= volume_multiplier)
                if has_volume
                else False
            )
            if require_volume_confirmation and not volume_confirmed:
                continue

            stop_price = min(b1_price, b2_price)
            risk = closes[entry_idx] - stop_price
            reward = target_price - closes[entry_idx]

            patterns.append(
                {
                    "PatternId": len(patterns) + 1,
                    "ContextHighBarIdx": context_idx,
                    "ContextHighDate": context["Date"],
                    "ContextHighPrice": context_price,
                    "PreviousHighBarIdx": previous_high_idx,
                    "PreviousHighDate": previous_high["Date"],
                    "PreviousHighPrice": previous_high_price,
                    "Bottom1BarIdx": b1_idx,
                    "Bottom1Date": bottom1["Date"],
                    "Bottom1Price": b1_price,
                    "ResistanceBarIdx": resistance_idx,
                    "ResistanceDate": resistance["Date"],
                    "ResistancePrice": resistance_price,
                    "Bottom2BarIdx": b2_idx,
                    "Bottom2Date": bottom2["Date"],
                    "Bottom2Price": b2_price,
                    "EntryBarIdx": entry_idx,
                    "EntryDate": data.index[entry_idx],
                    "EntryPrice": float(closes[entry_idx]),
                    "FollowThroughBarIdx": int(follow_through_idx)
                    if follow_through_idx is not None
                    else np.nan,
                    "FollowThroughDate": data.index[int(follow_through_idx)]
                    if follow_through_idx is not None
                    else pd.NaT,
                    "FollowThroughPrice": float(closes[int(follow_through_idx)])
                    if follow_through_idx is not None
                    else np.nan,
                    "FollowThroughLevel": float(follow_through_level),
                    "TargetPrice": float(target_price),
                    "StopPrice": float(stop_price),
                    "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                    "BottomDifferencePct": float(bottom_difference_pct),
                    "SupportZoneBottom": float(support_zone_bottom),
                    "SupportZoneTop": float(support_zone_top),
                    "ResistanceZoneBottom": float(resistance_zone_bottom),
                    "ResistanceZoneTop": float(resistance_zone_top),
                    "RetracePct": float(retrace_pct),
                    "DowntrendPct": float(downtrend_pct),
                    "PriorLHLL": bool(prior_lh_ll),
                    "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                    "SecondPeakInvalidation": bool(second_peak_invalidation),
                    "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                    "BullishStructureBreak": bool(bullish_structure_break),
                    "VolumeRatio": entry_volume_ratio,
                    "VolumeConfirmed": bool(volume_confirmed),
                }
            )

    return pd.DataFrame(patterns, columns=columns)


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _has_bearish_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    return not subset.empty


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if support_zone_top < price < resistance_zone_bottom:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for _, pattern in double_bottoms.iterrows():
            point_indexes = [
                int(pattern["Bottom1BarIdx"]),
                int(pattern["ResistanceBarIdx"]),
                int(pattern["Bottom2BarIdx"]),
                int(pattern["EntryBarIdx"]),
            ]
            point_prices = [
                float(pattern["Bottom1Price"]),
                float(pattern["ResistancePrice"]),
                float(pattern["Bottom2Price"]),
                float(pattern["EntryPrice"]),
            ]
            point_labels = ["B1", "Resistance", "B2", "Breakout"]
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=[
                        "bottom center",
                        "top center",
                        "bottom center",
                        "top center",
                        "top center",
                    ],
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "GARAN.IS"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    max_symmetry_ratio: float = 2.5,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_resistance: bool = True,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        bearish_choch_before_b1 = _has_bearish_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        resistance_candidates = ordered_swings[
            (ordered_swings["SwingType"] == "High")
            & (ordered_swings["BarIdx"].astype(int) > b1_idx)
            & (
                ordered_swings["BarIdx"].astype(int)
                < b1_idx + max_bars_between_bottoms
            )
        ]

        for _, resistance in resistance_candidates.iterrows():
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= b1_price:
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            earliest_breakout_idx = max(
                int(resistance["ConfirmBarIdx"]) + 1,
                resistance_idx + 2,
            )
            latest_breakout_idx = min(
                len(data) - 1,
                earliest_breakout_idx + max_entry_wait_bars,
            )

            for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                if closes[entry_idx] < support_zone_bottom:
                    break

                if closes[entry_idx] <= breakout_level:
                    continue

                right_lows = lows[resistance_idx + 1 : entry_idx]
                if len(right_lows) == 0:
                    continue

                b2_idx = resistance_idx + 1 + int(np.argmin(right_lows))
                b2_price = float(lows[b2_idx])
                bars_between = b2_idx - b1_idx
                if bars_between < min_bars_between_bottoms:
                    continue
                if bars_between > max_bars_between_bottoms:
                    break

                bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
                if bottom_difference_pct > bottom_tolerance_pct:
                    continue
                if not support_zone_bottom <= b2_price <= support_zone_top:
                    continue
                if resistance_price <= max(b1_price, b2_price):
                    continue

                left_width = resistance_idx - b1_idx
                right_width = b2_idx - resistance_idx
                if left_width < 1 or right_width < 1:
                    continue
                symmetry_ratio = max(left_width / right_width, right_width / left_width)
                if symmetry_ratio > max_symmetry_ratio:
                    continue

                left_midpoint = 0.5 * (b1_price + resistance_price)
                right_midpoint = 0.5 * (b2_price + resistance_price)
                if b1_price > right_midpoint or b2_price > left_midpoint:
                    continue

                highs_between = ordered_swings[
                    (ordered_swings["SwingType"] == "High")
                    & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                    & (ordered_swings["BarIdx"].astype(int) < b2_idx)
                ]
                if highs_between.empty:
                    continue
                highest_between = highs_between.sort_values(
                    ["Price", "BarIdx"],
                    ascending=[False, False],
                ).iloc[0]
                if int(highest_between["BarIdx"]) != resistance_idx:
                    continue

                second_peak_invalidation = _has_second_peak_between_zones(
                    ordered_swings,
                    start_idx=resistance_idx,
                    end_idx=b2_idx,
                    support_zone_top=support_zone_top,
                    resistance_zone_bottom=resistance_zone_bottom,
                )
                if invalidate_second_peak_between_zones and second_peak_invalidation:
                    continue

                support_fail_level = min(b1_price, b2_price) * (
                    1.0 - fail_buffer_pct / 100.0
                )
                if np.min(closes[resistance_idx + 1 : entry_idx + 1]) < support_fail_level:
                    continue

                post_b2_swing_invalidation = _has_any_swing_between(
                    ordered_swings,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                    continue

                bullish_structure_break = _has_bullish_structure_break(
                    structure_events,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_bullish_structure_break and not bullish_structure_break:
                    continue

                pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
                target_price = resistance_zone_top + pattern_height
                follow_through_level = resistance_zone_top + (
                    pattern_height * follow_through_pattern_ratio
                )
                follow_through_idx = _find_breakout_follow_through(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                    follow_through_level=follow_through_level,
                    support_zone_bottom=support_zone_bottom,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                )
                if require_breakout_follow_through and follow_through_idx is None:
                    continue

                pattern_start = _find_double_bottom_start(
                    data,
                    b1_idx=b1_idx,
                    resistance_level=resistance_zone_bottom,
                    fallback_idx=context_idx,
                    fallback_price=context_price,
                    lookback_bars=downtrend_lookback_bars,
                )
                pattern_r2 = _compute_double_bottom_r2(
                    closes=closes,
                    start_idx=int(pattern_start["BarIdx"]),
                    start_price=float(pattern_start["Price"]),
                    b1_idx=b1_idx,
                    b1_price=b1_price,
                    resistance_idx=resistance_idx,
                    resistance_price=resistance_price,
                    b2_idx=b2_idx,
                    b2_price=b2_price,
                    entry_idx=entry_idx,
                    entry_price=float(closes[entry_idx]),
                )
                if (
                    min_pattern_r2 is not None
                    and (
                        not np.isfinite(pattern_r2)
                        or float(pattern_r2) < float(min_pattern_r2)
                    )
                ):
                    continue

                entry_volume_ratio = (
                    float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                )
                volume_confirmed = (
                    bool(
                        np.isfinite(entry_volume_ratio)
                        and entry_volume_ratio >= volume_multiplier
                    )
                    if has_volume
                    else False
                )
                if require_volume_confirmation and not volume_confirmed:
                    continue

                pattern_key = (b1_idx, resistance_idx, b2_idx)
                if pattern_key in seen_patterns:
                    continue
                seen_patterns.add(pattern_key)

                stop_price = min(b1_price, b2_price)
                risk = closes[entry_idx] - stop_price
                reward = target_price - closes[entry_idx]

                patterns.append(
                    {
                        "PatternId": len(patterns) + 1,
                        "ContextHighBarIdx": context_idx,
                        "ContextHighDate": context["Date"],
                        "ContextHighPrice": context_price,
                        "PreviousHighBarIdx": previous_high_idx,
                        "PreviousHighDate": previous_high["Date"],
                        "PreviousHighPrice": previous_high_price,
                        "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                        "PatternStartDate": pattern_start["Date"],
                        "PatternStartPrice": float(pattern_start["Price"]),
                        "Bottom1BarIdx": b1_idx,
                        "Bottom1Date": bottom1["Date"],
                        "Bottom1Price": b1_price,
                        "ResistanceBarIdx": resistance_idx,
                        "ResistanceDate": resistance["Date"],
                        "ResistancePrice": resistance_price,
                        "Bottom2BarIdx": b2_idx,
                        "Bottom2Date": data.index[b2_idx],
                        "Bottom2Price": b2_price,
                        "EntryBarIdx": entry_idx,
                        "EntryDate": data.index[entry_idx],
                        "EntryPrice": float(closes[entry_idx]),
                        "FollowThroughBarIdx": int(follow_through_idx)
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughDate": data.index[int(follow_through_idx)]
                        if follow_through_idx is not None
                        else pd.NaT,
                        "FollowThroughPrice": float(closes[int(follow_through_idx)])
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughLevel": float(follow_through_level),
                        "TargetPrice": float(target_price),
                        "StopPrice": float(stop_price),
                        "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                        "BottomDifferencePct": float(bottom_difference_pct),
                        "SupportZoneBottom": float(support_zone_bottom),
                        "SupportZoneTop": float(support_zone_top),
                        "ResistanceZoneBottom": float(resistance_zone_bottom),
                        "ResistanceZoneTop": float(resistance_zone_top),
                        "RetracePct": float(retrace_pct),
                        "DowntrendPct": float(downtrend_pct),
                        "PatternWidth": int(b2_idx - b1_idx),
                        "PatternHeight": float(pattern_height),
                        "SymmetryRatio": float(symmetry_ratio),
                        "PatternR2": float(pattern_r2),
                        "PriorLHLL": bool(prior_lh_ll),
                        "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                        "SecondPeakInvalidation": bool(second_peak_invalidation),
                        "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                        "BullishStructureBreak": bool(bullish_structure_break),
                        "VolumeRatio": entry_volume_ratio,
                        "VolumeConfirmed": bool(volume_confirmed),
                    }
                )
                break

    return pd.DataFrame(patterns, columns=columns)


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _has_bearish_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    return not subset.empty


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if price > support_zone_top:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
    resistance_zone_bottom: float | None = None,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if (
            current_idx > entry_idx
            and resistance_zone_bottom is not None
            and close_price < resistance_zone_bottom
        ):
            return None
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _find_double_bottom_start(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    resistance_level: float,
    fallback_idx: int,
    fallback_price: float,
    lookback_bars: int,
) -> dict[str, Any]:
    search_start = max(0, int(b1_idx) - int(lookback_bars))
    closes = data["close"].to_numpy(dtype=float)
    highs = data["high"].to_numpy(dtype=float)

    for current_idx in range(int(b1_idx) - 1, search_start - 1, -1):
        if closes[current_idx] >= resistance_level or highs[current_idx] >= resistance_level:
            return {
                "BarIdx": current_idx,
                "Date": data.index[current_idx],
                "Price": float(closes[current_idx]),
            }

    return {
        "BarIdx": int(fallback_idx),
        "Date": data.index[int(fallback_idx)],
        "Price": float(fallback_price),
    }


def _compute_double_bottom_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    b1_idx: int,
    b1_price: float,
    resistance_idx: int,
    resistance_price: float,
    b2_idx: int,
    b2_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, b1_idx, resistance_idx, b2_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [start_price, b1_price, resistance_price, b2_price, entry_price],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for _, pattern in double_bottoms.iterrows():
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")
            point_indexes.extend(
                [
                    int(pattern["Bottom1BarIdx"]),
                    int(pattern["ResistanceBarIdx"]),
                    int(pattern["Bottom2BarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["Bottom1Price"]),
                    float(pattern["ResistancePrice"]),
                    float(pattern["Bottom2Price"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["B1", "Resistance", "B2", "Breakout"])
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "TUPRS.IS"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_breakout_hold_resistance=True,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    max_symmetry_ratio: float = 2.5,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_resistance: bool = True,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "LastCHoCHBeforeB1BarIdx",
        "LastCHoCHBeforeB1Date",
        "LastCHoCHBeforeB1Direction",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch_before_b1 = _find_last_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        bearish_choch_before_b1 = (
            last_choch_before_b1 is not None
            and str(last_choch_before_b1["Direction"]) == "Bearish"
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        resistance_candidates = ordered_swings[
            (ordered_swings["SwingType"] == "High")
            & (ordered_swings["BarIdx"].astype(int) > b1_idx)
            & (
                ordered_swings["BarIdx"].astype(int)
                < b1_idx + max_bars_between_bottoms
            )
        ]

        for _, resistance in resistance_candidates.iterrows():
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= b1_price:
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            earliest_breakout_idx = max(
                int(resistance["ConfirmBarIdx"]) + 1,
                resistance_idx + 2,
            )
            latest_breakout_idx = min(
                len(data) - 1,
                earliest_breakout_idx + max_entry_wait_bars,
            )

            for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                if closes[entry_idx] < support_zone_bottom:
                    break

                if closes[entry_idx] <= breakout_level:
                    continue

                right_lows = lows[resistance_idx + 1 : entry_idx]
                if len(right_lows) == 0:
                    continue

                b2_idx = resistance_idx + 1 + int(np.argmin(right_lows))
                b2_price = float(lows[b2_idx])
                bars_between = b2_idx - b1_idx
                if bars_between < min_bars_between_bottoms:
                    continue
                if bars_between > max_bars_between_bottoms:
                    break

                bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
                if bottom_difference_pct > bottom_tolerance_pct:
                    continue
                if not support_zone_bottom <= b2_price <= support_zone_top:
                    continue
                if resistance_price <= max(b1_price, b2_price):
                    continue

                left_width = resistance_idx - b1_idx
                right_width = b2_idx - resistance_idx
                if left_width < 1 or right_width < 1:
                    continue
                symmetry_ratio = max(left_width / right_width, right_width / left_width)
                if symmetry_ratio > max_symmetry_ratio:
                    continue

                left_midpoint = 0.5 * (b1_price + resistance_price)
                right_midpoint = 0.5 * (b2_price + resistance_price)
                if b1_price > right_midpoint or b2_price > left_midpoint:
                    continue

                highs_between = ordered_swings[
                    (ordered_swings["SwingType"] == "High")
                    & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                    & (ordered_swings["BarIdx"].astype(int) < b2_idx)
                ]
                if highs_between.empty:
                    continue
                highest_between = highs_between.sort_values(
                    ["Price", "BarIdx"],
                    ascending=[False, False],
                ).iloc[0]
                if int(highest_between["BarIdx"]) != resistance_idx:
                    continue

                second_peak_invalidation = _has_second_peak_between_zones(
                    ordered_swings,
                    start_idx=resistance_idx,
                    end_idx=b2_idx,
                    support_zone_top=support_zone_top,
                    resistance_zone_bottom=resistance_zone_bottom,
                )
                if invalidate_second_peak_between_zones and second_peak_invalidation:
                    continue

                support_fail_level = min(b1_price, b2_price) * (
                    1.0 - fail_buffer_pct / 100.0
                )
                if np.min(closes[resistance_idx + 1 : entry_idx + 1]) < support_fail_level:
                    continue

                post_b2_swing_invalidation = _has_any_swing_between(
                    ordered_swings,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                    continue

                bullish_structure_break = _has_bullish_structure_break(
                    structure_events,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_bullish_structure_break and not bullish_structure_break:
                    continue

                pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
                target_price = resistance_zone_top + pattern_height
                follow_through_level = resistance_zone_top + (
                    pattern_height * follow_through_pattern_ratio
                )
                follow_through_idx = _find_breakout_follow_through(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                    follow_through_level=follow_through_level,
                    support_zone_bottom=support_zone_bottom,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                )
                if require_breakout_follow_through and follow_through_idx is None:
                    continue

                pattern_start = _find_double_bottom_start(
                    data,
                    b1_idx=b1_idx,
                    resistance_level=resistance_zone_bottom,
                    fallback_idx=context_idx,
                    fallback_price=context_price,
                    lookback_bars=downtrend_lookback_bars,
                )
                pattern_r2 = _compute_double_bottom_r2(
                    closes=closes,
                    start_idx=int(pattern_start["BarIdx"]),
                    start_price=float(pattern_start["Price"]),
                    b1_idx=b1_idx,
                    b1_price=b1_price,
                    resistance_idx=resistance_idx,
                    resistance_price=resistance_price,
                    b2_idx=b2_idx,
                    b2_price=b2_price,
                    entry_idx=entry_idx,
                    entry_price=float(closes[entry_idx]),
                )
                if (
                    min_pattern_r2 is not None
                    and (
                        not np.isfinite(pattern_r2)
                        or float(pattern_r2) < float(min_pattern_r2)
                    )
                ):
                    continue

                entry_volume_ratio = (
                    float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                )
                volume_confirmed = (
                    bool(
                        np.isfinite(entry_volume_ratio)
                        and entry_volume_ratio >= volume_multiplier
                    )
                    if has_volume
                    else False
                )
                if require_volume_confirmation and not volume_confirmed:
                    continue

                pattern_key = (b1_idx, resistance_idx, b2_idx)
                if pattern_key in seen_patterns:
                    continue
                seen_patterns.add(pattern_key)

                stop_price = min(b1_price, b2_price)
                risk = closes[entry_idx] - stop_price
                reward = target_price - closes[entry_idx]

                patterns.append(
                    {
                        "PatternId": len(patterns) + 1,
                        "ContextHighBarIdx": context_idx,
                        "ContextHighDate": context["Date"],
                        "ContextHighPrice": context_price,
                        "PreviousHighBarIdx": previous_high_idx,
                        "PreviousHighDate": previous_high["Date"],
                        "PreviousHighPrice": previous_high_price,
                        "LastCHoCHBeforeB1BarIdx": int(
                            last_choch_before_b1["EventBarIdx"]
                        )
                        if last_choch_before_b1 is not None
                        else np.nan,
                        "LastCHoCHBeforeB1Date": last_choch_before_b1["EventDate"]
                        if last_choch_before_b1 is not None
                        else pd.NaT,
                        "LastCHoCHBeforeB1Direction": str(
                            last_choch_before_b1["Direction"]
                        )
                        if last_choch_before_b1 is not None
                        else None,
                        "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                        "PatternStartDate": pattern_start["Date"],
                        "PatternStartPrice": float(pattern_start["Price"]),
                        "Bottom1BarIdx": b1_idx,
                        "Bottom1Date": bottom1["Date"],
                        "Bottom1Price": b1_price,
                        "ResistanceBarIdx": resistance_idx,
                        "ResistanceDate": resistance["Date"],
                        "ResistancePrice": resistance_price,
                        "Bottom2BarIdx": b2_idx,
                        "Bottom2Date": data.index[b2_idx],
                        "Bottom2Price": b2_price,
                        "EntryBarIdx": entry_idx,
                        "EntryDate": data.index[entry_idx],
                        "EntryPrice": float(closes[entry_idx]),
                        "FollowThroughBarIdx": int(follow_through_idx)
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughDate": data.index[int(follow_through_idx)]
                        if follow_through_idx is not None
                        else pd.NaT,
                        "FollowThroughPrice": float(closes[int(follow_through_idx)])
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughLevel": float(follow_through_level),
                        "TargetPrice": float(target_price),
                        "StopPrice": float(stop_price),
                        "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                        "BottomDifferencePct": float(bottom_difference_pct),
                        "SupportZoneBottom": float(support_zone_bottom),
                        "SupportZoneTop": float(support_zone_top),
                        "ResistanceZoneBottom": float(resistance_zone_bottom),
                        "ResistanceZoneTop": float(resistance_zone_top),
                        "RetracePct": float(retrace_pct),
                        "DowntrendPct": float(downtrend_pct),
                        "PatternWidth": int(b2_idx - b1_idx),
                        "PatternHeight": float(pattern_height),
                        "SymmetryRatio": float(symmetry_ratio),
                        "PatternR2": float(pattern_r2),
                        "PriorLHLL": bool(prior_lh_ll),
                        "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                        "SecondPeakInvalidation": bool(second_peak_invalidation),
                        "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                        "BullishStructureBreak": bool(bullish_structure_break),
                        "VolumeRatio": entry_volume_ratio,
                        "VolumeConfirmed": bool(volume_confirmed),
                    }
                )
                break

    return pd.DataFrame(patterns, columns=columns)


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _find_last_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> pd.Series | None:
    if structure_events is None or structure_events.empty:
        return None

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    if subset.empty:
        return None

    return subset.sort_values("EventBarIdx").iloc[-1]


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if price > support_zone_top:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
    resistance_zone_bottom: float | None = None,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if (
            current_idx > entry_idx
            and resistance_zone_bottom is not None
            and close_price < resistance_zone_bottom
        ):
            return None
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _find_double_bottom_start(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    resistance_level: float,
    fallback_idx: int,
    fallback_price: float,
    lookback_bars: int,
) -> dict[str, Any]:
    search_start = max(0, int(b1_idx) - int(lookback_bars))
    closes = data["close"].to_numpy(dtype=float)
    highs = data["high"].to_numpy(dtype=float)

    for current_idx in range(int(b1_idx) - 1, search_start - 1, -1):
        if closes[current_idx] >= resistance_level or highs[current_idx] >= resistance_level:
            return {
                "BarIdx": current_idx,
                "Date": data.index[current_idx],
                "Price": float(closes[current_idx]),
            }

    return {
        "BarIdx": int(fallback_idx),
        "Date": data.index[int(fallback_idx)],
        "Price": float(fallback_price),
    }


def _compute_double_bottom_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    b1_idx: int,
    b1_price: float,
    resistance_idx: int,
    resistance_price: float,
    b2_idx: int,
    b2_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, b1_idx, resistance_idx, b2_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [start_price, b1_price, resistance_price, b2_price, entry_price],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for _, pattern in double_bottoms.iterrows():
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")
            point_indexes.extend(
                [
                    int(pattern["Bottom1BarIdx"]),
                    int(pattern["ResistanceBarIdx"]),
                    int(pattern["Bottom2BarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["Bottom1Price"]),
                    float(pattern["ResistancePrice"]),
                    float(pattern["Bottom2Price"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["B1", "Resistance", "B2", "Breakout"])
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforeB1Direction', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "AAPL"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_breakout_hold_resistance=True,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    max_symmetry_ratio: float = 2.5,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_resistance: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "LastCHoCHBeforeB1BarIdx",
        "LastCHoCHBeforeB1Date",
        "LastCHoCHBeforeB1Direction",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch_before_b1 = _find_last_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        bearish_choch_before_b1 = (
            last_choch_before_b1 is not None
            and str(last_choch_before_b1["Direction"]) == "Bearish"
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        resistance_candidates = ordered_swings[
            (ordered_swings["SwingType"] == "High")
            & (ordered_swings["BarIdx"].astype(int) > b1_idx)
            & (
                ordered_swings["BarIdx"].astype(int)
                < b1_idx + max_bars_between_bottoms
            )
        ]

        for _, resistance in resistance_candidates.iterrows():
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= b1_price:
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            earliest_breakout_idx = max(
                int(resistance["ConfirmBarIdx"]) + 1,
                resistance_idx + 2,
            )
            latest_breakout_idx = min(
                len(data) - 1,
                earliest_breakout_idx + max_entry_wait_bars,
            )

            for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                if closes[entry_idx] < support_zone_bottom:
                    break

                if closes[entry_idx] <= breakout_level:
                    continue

                right_lows = lows[resistance_idx + 1 : entry_idx]
                if len(right_lows) == 0:
                    continue

                b2_idx = resistance_idx + 1 + int(np.argmin(right_lows))
                b2_price = float(lows[b2_idx])
                bars_between = b2_idx - b1_idx
                if bars_between < min_bars_between_bottoms:
                    continue
                if bars_between > max_bars_between_bottoms:
                    break

                bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
                if bottom_difference_pct > bottom_tolerance_pct:
                    continue
                if not support_zone_bottom <= b2_price <= support_zone_top:
                    continue
                if resistance_price <= max(b1_price, b2_price):
                    continue

                left_width = resistance_idx - b1_idx
                right_width = b2_idx - resistance_idx
                if left_width < 1 or right_width < 1:
                    continue
                symmetry_ratio = max(left_width / right_width, right_width / left_width)
                if symmetry_ratio > max_symmetry_ratio:
                    continue

                left_midpoint = 0.5 * (b1_price + resistance_price)
                right_midpoint = 0.5 * (b2_price + resistance_price)
                if b1_price > right_midpoint or b2_price > left_midpoint:
                    continue

                highs_between = ordered_swings[
                    (ordered_swings["SwingType"] == "High")
                    & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                    & (ordered_swings["BarIdx"].astype(int) < b2_idx)
                ]
                if highs_between.empty:
                    continue
                highest_between = highs_between.sort_values(
                    ["Price", "BarIdx"],
                    ascending=[False, False],
                ).iloc[0]
                if int(highest_between["BarIdx"]) != resistance_idx:
                    continue

                second_peak_invalidation = _has_second_peak_between_zones(
                    ordered_swings,
                    start_idx=resistance_idx,
                    end_idx=b2_idx,
                    support_zone_top=support_zone_top,
                    resistance_zone_bottom=resistance_zone_bottom,
                )
                if invalidate_second_peak_between_zones and second_peak_invalidation:
                    continue

                support_fail_level = min(b1_price, b2_price) * (
                    1.0 - fail_buffer_pct / 100.0
                )
                if np.min(closes[resistance_idx + 1 : entry_idx + 1]) < support_fail_level:
                    continue

                post_b2_swing_invalidation = _has_any_swing_between(
                    ordered_swings,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                    continue

                bullish_structure_break = _has_bullish_structure_break(
                    structure_events,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_bullish_structure_break and not bullish_structure_break:
                    continue

                pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
                target_price = resistance_zone_top + pattern_height
                follow_through_level = resistance_zone_top + (
                    pattern_height * follow_through_pattern_ratio
                )
                follow_through_idx = _find_breakout_follow_through(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                    follow_through_level=follow_through_level,
                    support_zone_bottom=support_zone_bottom,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                )
                if require_breakout_follow_through and follow_through_idx is None:
                    continue

                pattern_start = _find_double_bottom_start(
                    data,
                    b1_idx=b1_idx,
                    resistance_level=resistance_zone_bottom,
                    fallback_idx=context_idx,
                    fallback_price=context_price,
                    lookback_bars=downtrend_lookback_bars,
                )
                pattern_r2 = _compute_double_bottom_r2(
                    closes=closes,
                    start_idx=int(pattern_start["BarIdx"]),
                    start_price=float(pattern_start["Price"]),
                    b1_idx=b1_idx,
                    b1_price=b1_price,
                    resistance_idx=resistance_idx,
                    resistance_price=resistance_price,
                    b2_idx=b2_idx,
                    b2_price=b2_price,
                    entry_idx=entry_idx,
                    entry_price=float(closes[entry_idx]),
                )
                if (
                    min_pattern_r2 is not None
                    and (
                        not np.isfinite(pattern_r2)
                        or float(pattern_r2) < float(min_pattern_r2)
                    )
                ):
                    continue

                entry_volume_ratio = (
                    float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                )
                volume_confirmed = (
                    bool(
                        np.isfinite(entry_volume_ratio)
                        and entry_volume_ratio >= volume_multiplier
                    )
                    if has_volume
                    else False
                )
                if require_volume_confirmation and not volume_confirmed:
                    continue

                pattern_key = (b1_idx, resistance_idx, b2_idx)
                if pattern_key in seen_patterns:
                    continue
                seen_patterns.add(pattern_key)

                stop_price = min(b1_price, b2_price)
                risk = closes[entry_idx] - stop_price
                reward = target_price - closes[entry_idx]

                patterns.append(
                    {
                        "PatternId": len(patterns) + 1,
                        "ContextHighBarIdx": context_idx,
                        "ContextHighDate": context["Date"],
                        "ContextHighPrice": context_price,
                        "PreviousHighBarIdx": previous_high_idx,
                        "PreviousHighDate": previous_high["Date"],
                        "PreviousHighPrice": previous_high_price,
                        "LastCHoCHBeforeB1BarIdx": int(
                            last_choch_before_b1["EventBarIdx"]
                        )
                        if last_choch_before_b1 is not None
                        else np.nan,
                        "LastCHoCHBeforeB1Date": last_choch_before_b1["EventDate"]
                        if last_choch_before_b1 is not None
                        else pd.NaT,
                        "LastCHoCHBeforeB1Direction": str(
                            last_choch_before_b1["Direction"]
                        )
                        if last_choch_before_b1 is not None
                        else None,
                        "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                        "PatternStartDate": pattern_start["Date"],
                        "PatternStartPrice": float(pattern_start["Price"]),
                        "Bottom1BarIdx": b1_idx,
                        "Bottom1Date": bottom1["Date"],
                        "Bottom1Price": b1_price,
                        "ResistanceBarIdx": resistance_idx,
                        "ResistanceDate": resistance["Date"],
                        "ResistancePrice": resistance_price,
                        "Bottom2BarIdx": b2_idx,
                        "Bottom2Date": data.index[b2_idx],
                        "Bottom2Price": b2_price,
                        "EntryBarIdx": entry_idx,
                        "EntryDate": data.index[entry_idx],
                        "EntryPrice": float(closes[entry_idx]),
                        "FollowThroughBarIdx": int(follow_through_idx)
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughDate": data.index[int(follow_through_idx)]
                        if follow_through_idx is not None
                        else pd.NaT,
                        "FollowThroughPrice": float(closes[int(follow_through_idx)])
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughLevel": float(follow_through_level),
                        "TargetPrice": float(target_price),
                        "StopPrice": float(stop_price),
                        "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                        "BottomDifferencePct": float(bottom_difference_pct),
                        "SupportZoneBottom": float(support_zone_bottom),
                        "SupportZoneTop": float(support_zone_top),
                        "ResistanceZoneBottom": float(resistance_zone_bottom),
                        "ResistanceZoneTop": float(resistance_zone_top),
                        "RetracePct": float(retrace_pct),
                        "DowntrendPct": float(downtrend_pct),
                        "PatternWidth": int(b2_idx - b1_idx),
                        "PatternHeight": float(pattern_height),
                        "SymmetryRatio": float(symmetry_ratio),
                        "PatternR2": float(pattern_r2),
                        "PriorLHLL": bool(prior_lh_ll),
                        "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                        "SecondPeakInvalidation": bool(second_peak_invalidation),
                        "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                        "BullishStructureBreak": bool(bullish_structure_break),
                        "VolumeRatio": entry_volume_ratio,
                        "VolumeConfirmed": bool(volume_confirmed),
                    }
                )
                break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_pattern_per_choch(result, columns=columns)
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _select_best_pattern_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforeB1BarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforeB1BarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "ResistancePrice",
                "Bottom1BarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _score_double_bottom_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    bottom1_price = max(abs(_finite_or_zero(row.get("Bottom1Price"))), 1e-12)
    height_pct = pattern_height / bottom1_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 80.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    bottom_diff_penalty = _finite_or_zero(row.get("BottomDifferencePct"))
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("Bottom2BarIdx")),
        0.0,
    )

    return (
        height_pct * 2.5
        + width_score
        + downtrend_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - bottom_diff_penalty * 2.0
        - entry_delay * 0.02
    )


def _finite_or_zero(value: Any, *, default: float = 0.0) -> float:
    try:
        number = float(value)
    except (TypeError, ValueError):
        return default
    if not np.isfinite(number):
        return default
    return number


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _find_last_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> pd.Series | None:
    if structure_events is None or structure_events.empty:
        return None

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    if subset.empty:
        return None

    return subset.sort_values("EventBarIdx").iloc[-1]


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if price > support_zone_top:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
    resistance_zone_bottom: float | None = None,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if (
            current_idx > entry_idx
            and resistance_zone_bottom is not None
            and close_price < resistance_zone_bottom
        ):
            return None
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _find_double_bottom_start(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    resistance_level: float,
    fallback_idx: int,
    fallback_price: float,
    lookback_bars: int,
) -> dict[str, Any]:
    search_start = max(0, int(b1_idx) - int(lookback_bars))
    closes = data["close"].to_numpy(dtype=float)
    highs = data["high"].to_numpy(dtype=float)

    for current_idx in range(int(b1_idx) - 1, search_start - 1, -1):
        if closes[current_idx] >= resistance_level or highs[current_idx] >= resistance_level:
            return {
                "BarIdx": current_idx,
                "Date": data.index[current_idx],
                "Price": float(closes[current_idx]),
            }

    return {
        "BarIdx": int(fallback_idx),
        "Date": data.index[int(fallback_idx)],
        "Price": float(fallback_price),
    }


def _compute_double_bottom_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    b1_idx: int,
    b1_price: float,
    resistance_idx: int,
    resistance_price: float,
    b2_idx: int,
    b2_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, b1_idx, resistance_idx, b2_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [start_price, b1_price, resistance_price, b2_price, entry_price],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for pattern_number, (_, pattern) in enumerate(double_bottoms.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")
            point_indexes.extend(
                [
                    int(pattern["Bottom1BarIdx"]),
                    int(pattern["ResistanceBarIdx"]),
                    int(pattern["Bottom2BarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["Bottom1Price"]),
                    float(pattern["ResistancePrice"]),
                    float(pattern["Bottom2Price"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["B1", "Resistance", "B2", "Breakout"])
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    showlegend=pattern_number == 0,
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforeB1Direction', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "AAPL"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_breakout_hold_resistance=True,
        one_pattern_per_bearish_choch=True,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    max_symmetry_ratio: float = 2.5,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_resistance: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "LastCHoCHBeforeB1BarIdx",
        "LastCHoCHBeforeB1Date",
        "LastCHoCHBeforeB1Direction",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch_before_b1 = _find_last_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        bearish_choch_before_b1 = (
            last_choch_before_b1 is not None
            and str(last_choch_before_b1["Direction"]) == "Bearish"
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        resistance_candidates = ordered_swings[
            (ordered_swings["SwingType"] == "High")
            & (ordered_swings["BarIdx"].astype(int) > b1_idx)
            & (
                ordered_swings["BarIdx"].astype(int)
                < b1_idx + max_bars_between_bottoms
            )
        ]

        for _, resistance in resistance_candidates.iterrows():
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= b1_price:
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            earliest_breakout_idx = max(
                int(resistance["ConfirmBarIdx"]) + 1,
                resistance_idx + 2,
            )
            latest_breakout_idx = min(
                len(data) - 1,
                earliest_breakout_idx + max_entry_wait_bars,
            )

            for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                if closes[entry_idx] < support_zone_bottom:
                    break

                if closes[entry_idx] <= breakout_level:
                    continue

                right_lows = lows[resistance_idx + 1 : entry_idx]
                if len(right_lows) == 0:
                    continue

                b2_idx = resistance_idx + 1 + int(np.argmin(right_lows))
                b2_price = float(lows[b2_idx])
                bars_between = b2_idx - b1_idx
                if bars_between < min_bars_between_bottoms:
                    continue
                if bars_between > max_bars_between_bottoms:
                    break

                bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
                if bottom_difference_pct > bottom_tolerance_pct:
                    continue
                if not support_zone_bottom <= b2_price <= support_zone_top:
                    continue
                if resistance_price <= max(b1_price, b2_price):
                    continue

                left_width = resistance_idx - b1_idx
                right_width = b2_idx - resistance_idx
                if left_width < 1 or right_width < 1:
                    continue
                symmetry_ratio = max(left_width / right_width, right_width / left_width)
                if symmetry_ratio > max_symmetry_ratio:
                    continue

                left_midpoint = 0.5 * (b1_price + resistance_price)
                right_midpoint = 0.5 * (b2_price + resistance_price)
                if b1_price > right_midpoint or b2_price > left_midpoint:
                    continue

                highs_between = ordered_swings[
                    (ordered_swings["SwingType"] == "High")
                    & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                    & (ordered_swings["BarIdx"].astype(int) < b2_idx)
                ]
                if highs_between.empty:
                    continue
                highest_between = highs_between.sort_values(
                    ["Price", "BarIdx"],
                    ascending=[False, False],
                ).iloc[0]
                if int(highest_between["BarIdx"]) != resistance_idx:
                    continue

                second_peak_invalidation = _has_second_peak_between_zones(
                    ordered_swings,
                    start_idx=resistance_idx,
                    end_idx=b2_idx,
                    support_zone_top=support_zone_top,
                    resistance_zone_bottom=resistance_zone_bottom,
                )
                if invalidate_second_peak_between_zones and second_peak_invalidation:
                    continue

                support_fail_level = min(b1_price, b2_price) * (
                    1.0 - fail_buffer_pct / 100.0
                )
                if np.min(closes[resistance_idx + 1 : entry_idx + 1]) < support_fail_level:
                    continue

                post_b2_swing_invalidation = _has_any_swing_between(
                    ordered_swings,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                    continue

                bullish_structure_break = _has_bullish_structure_break(
                    structure_events,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_bullish_structure_break and not bullish_structure_break:
                    continue

                pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
                target_price = resistance_zone_top + pattern_height
                follow_through_level = resistance_zone_top + (
                    pattern_height * follow_through_pattern_ratio
                )
                follow_through_idx = _find_breakout_follow_through(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                    follow_through_level=follow_through_level,
                    support_zone_bottom=support_zone_bottom,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                )
                if require_breakout_follow_through and follow_through_idx is None:
                    continue

                pattern_start = _find_double_bottom_start(
                    data,
                    b1_idx=b1_idx,
                    resistance_level=resistance_zone_bottom,
                    fallback_idx=context_idx,
                    fallback_price=context_price,
                    lookback_bars=downtrend_lookback_bars,
                )
                pattern_r2 = _compute_double_bottom_r2(
                    closes=closes,
                    start_idx=int(pattern_start["BarIdx"]),
                    start_price=float(pattern_start["Price"]),
                    b1_idx=b1_idx,
                    b1_price=b1_price,
                    resistance_idx=resistance_idx,
                    resistance_price=resistance_price,
                    b2_idx=b2_idx,
                    b2_price=b2_price,
                    entry_idx=entry_idx,
                    entry_price=float(closes[entry_idx]),
                )
                if (
                    min_pattern_r2 is not None
                    and (
                        not np.isfinite(pattern_r2)
                        or float(pattern_r2) < float(min_pattern_r2)
                    )
                ):
                    continue

                entry_volume_ratio = (
                    float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                )
                volume_confirmed = (
                    bool(
                        np.isfinite(entry_volume_ratio)
                        and entry_volume_ratio >= volume_multiplier
                    )
                    if has_volume
                    else False
                )
                if require_volume_confirmation and not volume_confirmed:
                    continue

                pattern_key = (b1_idx, resistance_idx, b2_idx)
                if pattern_key in seen_patterns:
                    continue
                seen_patterns.add(pattern_key)

                stop_price = min(b1_price, b2_price)
                risk = closes[entry_idx] - stop_price
                reward = target_price - closes[entry_idx]

                patterns.append(
                    {
                        "PatternId": len(patterns) + 1,
                        "ContextHighBarIdx": context_idx,
                        "ContextHighDate": context["Date"],
                        "ContextHighPrice": context_price,
                        "PreviousHighBarIdx": previous_high_idx,
                        "PreviousHighDate": previous_high["Date"],
                        "PreviousHighPrice": previous_high_price,
                        "LastCHoCHBeforeB1BarIdx": int(
                            last_choch_before_b1["EventBarIdx"]
                        )
                        if last_choch_before_b1 is not None
                        else np.nan,
                        "LastCHoCHBeforeB1Date": last_choch_before_b1["EventDate"]
                        if last_choch_before_b1 is not None
                        else pd.NaT,
                        "LastCHoCHBeforeB1Direction": str(
                            last_choch_before_b1["Direction"]
                        )
                        if last_choch_before_b1 is not None
                        else None,
                        "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                        "PatternStartDate": pattern_start["Date"],
                        "PatternStartPrice": float(pattern_start["Price"]),
                        "Bottom1BarIdx": b1_idx,
                        "Bottom1Date": bottom1["Date"],
                        "Bottom1Price": b1_price,
                        "ResistanceBarIdx": resistance_idx,
                        "ResistanceDate": resistance["Date"],
                        "ResistancePrice": resistance_price,
                        "Bottom2BarIdx": b2_idx,
                        "Bottom2Date": data.index[b2_idx],
                        "Bottom2Price": b2_price,
                        "EntryBarIdx": entry_idx,
                        "EntryDate": data.index[entry_idx],
                        "EntryPrice": float(closes[entry_idx]),
                        "FollowThroughBarIdx": int(follow_through_idx)
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughDate": data.index[int(follow_through_idx)]
                        if follow_through_idx is not None
                        else pd.NaT,
                        "FollowThroughPrice": float(closes[int(follow_through_idx)])
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughLevel": float(follow_through_level),
                        "TargetPrice": float(target_price),
                        "StopPrice": float(stop_price),
                        "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                        "BottomDifferencePct": float(bottom_difference_pct),
                        "SupportZoneBottom": float(support_zone_bottom),
                        "SupportZoneTop": float(support_zone_top),
                        "ResistanceZoneBottom": float(resistance_zone_bottom),
                        "ResistanceZoneTop": float(resistance_zone_top),
                        "RetracePct": float(retrace_pct),
                        "DowntrendPct": float(downtrend_pct),
                        "PatternWidth": int(b2_idx - b1_idx),
                        "PatternHeight": float(pattern_height),
                        "SymmetryRatio": float(symmetry_ratio),
                        "PatternR2": float(pattern_r2),
                        "PriorLHLL": bool(prior_lh_ll),
                        "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                        "SecondPeakInvalidation": bool(second_peak_invalidation),
                        "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                        "BullishStructureBreak": bool(bullish_structure_break),
                        "VolumeRatio": entry_volume_ratio,
                        "VolumeConfirmed": bool(volume_confirmed),
                    }
                )
                break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_pattern_per_choch(result, columns=columns)
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def detect_inverse_head_shoulders(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    shoulder_tolerance_pct: float = 3.0,
    min_head_depth_pct: float = 0.75,
    head_lowest_tolerance_pct: float = 0.15,
    neckline_zone_pct: float = 0.25,
    max_neckline_tilt_pct: float = 4.0,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_points: int = 3,
    max_pattern_bars: int = 260,
    max_symmetry_ratio: float = 2.5,
    neckline_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_prior_lh_ll: bool = True,
    require_last_choch_bearish: bool = True,
    require_head_lowest: bool = True,
    require_bullish_structure_break: bool = True,
    require_direct_breakout_after_right_shoulder: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_neckline: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed inverse head and shoulders patterns.

    Sequence:
    - prior bearish CHoCH/downtrend
    - left shoulder low
    - neckline high 1
    - deeper head low
    - neckline high 2
    - right shoulder low
    - close above the projected neckline
    """
    if shoulder_tolerance_pct < 0:
        raise ValueError("shoulder_tolerance_pct cannot be negative.")
    if min_head_depth_pct < 0:
        raise ValueError("min_head_depth_pct cannot be negative.")
    if head_lowest_tolerance_pct < 0:
        raise ValueError("head_lowest_tolerance_pct cannot be negative.")
    if neckline_zone_pct < 0:
        raise ValueError("neckline_zone_pct cannot be negative.")
    if max_neckline_tilt_pct < 0:
        raise ValueError("max_neckline_tilt_pct cannot be negative.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_points < 1:
        raise ValueError("min_bars_between_points must be at least 1.")
    if max_pattern_bars < min_bars_between_points * 4:
        raise ValueError("max_pattern_bars is too small for an IHS pattern.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if neckline_break_buffer_pct < 0:
        raise ValueError("neckline_break_buffer_pct cannot be negative.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "LastCHoCHBeforePatternBarIdx",
        "LastCHoCHBeforePatternDate",
        "LastCHoCHBeforePatternDirection",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "LeftShoulderBarIdx",
        "LeftShoulderDate",
        "LeftShoulderPrice",
        "Neckline1BarIdx",
        "Neckline1Date",
        "Neckline1Price",
        "HeadBarIdx",
        "HeadDate",
        "HeadPrice",
        "Neckline2BarIdx",
        "Neckline2Date",
        "Neckline2Price",
        "RightShoulderBarIdx",
        "RightShoulderDate",
        "RightShoulderPrice",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "NecklineEntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "ShoulderDifferencePct",
        "HeadDepthPct",
        "NecklineTiltPct",
        "NecklineZoneBottom",
        "NecklineZoneTop",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "LastCHoCHBearish",
        "HeadLowestInvalidation",
        "PostRightShoulderSwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for left_pos in range(len(lows_only) - 1):
        left_shoulder = lows_only.iloc[left_pos]
        left_idx = int(left_shoulder["BarIdx"])
        left_price = float(left_shoulder["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=left_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch = _find_last_choch_before(
            structure_events,
            before_idx=left_idx,
        )
        last_choch_bearish = (
            last_choch is not None and str(last_choch["Direction"]) == "Bearish"
        )
        if require_last_choch_bearish and not last_choch_bearish:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=left_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=left_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        if context_price <= left_price:
            continue

        head_candidates = lows_only[
            (lows_only["BarIdx"].astype(int) > left_idx + min_bars_between_points)
            & (lows_only["BarIdx"].astype(int) < left_idx + max_pattern_bars)
        ]

        for _, head in head_candidates.iterrows():
            head_idx = int(head["BarIdx"])
            head_price = float(head["Price"])
            if head_price >= left_price * (1.0 - min_head_depth_pct / 100.0):
                continue

            downtrend_pct = (context_price - head_price) / context_price * 100.0
            if downtrend_pct < min_downtrend_pct:
                continue

            leg_lows = ordered_swings[
                (ordered_swings["SwingType"] == "Low")
                & (ordered_swings["BarIdx"].astype(int) >= context_idx)
                & (ordered_swings["BarIdx"].astype(int) <= head_idx)
            ]
            if (
                not leg_lows.empty
                and head_price > float(leg_lows["Price"].min()) * 1.001
            ):
                continue

            left_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > left_idx)
                & (ordered_swings["BarIdx"].astype(int) < head_idx)
            ]
            if left_neck_candidates.empty:
                continue

            neck1 = left_neck_candidates.sort_values(
                ["Price", "BarIdx"],
                ascending=[False, False],
            ).iloc[0]
            neck1_idx = int(neck1["BarIdx"])
            neck1_price = float(neck1["Price"])
            if neck1_price <= left_price or neck1_price <= head_price:
                continue

            right_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > head_idx)
                & (ordered_swings["BarIdx"].astype(int) < left_idx + max_pattern_bars)
            ]
            if right_neck_candidates.empty:
                continue

            for _, neck2 in right_neck_candidates.iterrows():
                neck2_idx = int(neck2["BarIdx"])
                neck2_price = float(neck2["Price"])
                if neck2_idx - head_idx < min_bars_between_points:
                    continue
                if neck2_price <= head_price:
                    continue

                neckline_average = (neck1_price + neck2_price) / 2.0
                neckline_tilt_pct = (
                    abs(neck2_price - neck1_price)
                    / max(abs(neckline_average), 1e-12)
                    * 100.0
                )
                if neckline_tilt_pct > max_neckline_tilt_pct:
                    continue

                earliest_breakout_idx = max(
                    int(neck2["ConfirmBarIdx"]) + 1,
                    neck2_idx + 2,
                )
                latest_breakout_idx = min(
                    len(data) - 1,
                    earliest_breakout_idx + max_entry_wait_bars,
                )
                head_fail_level = head_price * (1.0 - fail_buffer_pct / 100.0)

                for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                    if closes[entry_idx] < head_fail_level:
                        break

                    neckline_entry_price = _line_value(
                        neck1_idx,
                        neck1_price,
                        neck2_idx,
                        neck2_price,
                        entry_idx,
                    )
                    breakout_level = neckline_entry_price * (
                        1.0 + neckline_break_buffer_pct / 100.0
                    )
                    if closes[entry_idx] <= breakout_level:
                        continue

                    right_lows = lows[neck2_idx + 1 : entry_idx]
                    if len(right_lows) == 0:
                        continue

                    right_idx = neck2_idx + 1 + int(np.argmin(right_lows))
                    right_price = float(lows[right_idx])
                    if right_idx - neck2_idx < min_bars_between_points:
                        continue
                    if right_price <= head_price * (1.0 + min_head_depth_pct / 100.0):
                        continue

                    pattern_low = float(np.min(lows[left_idx : entry_idx + 1]))
                    head_lowest_invalidation = pattern_low < head_price * (
                        1.0 - head_lowest_tolerance_pct / 100.0
                    )
                    if require_head_lowest and head_lowest_invalidation:
                        continue

                    shoulder_average = (left_price + right_price) / 2.0
                    shoulder_difference_pct = (
                        abs(right_price - left_price)
                        / max(abs(shoulder_average), 1e-12)
                        * 100.0
                    )
                    if shoulder_difference_pct > shoulder_tolerance_pct:
                        continue

                    head_depth_pct = (
                        (min(left_price, right_price) - head_price)
                        / max(abs(min(left_price, right_price)), 1e-12)
                        * 100.0
                    )
                    if head_depth_pct < min_head_depth_pct:
                        continue

                    if neck2_price <= right_price:
                        continue

                    left_midpoint = (head_price + neck1_price) / 2.0
                    right_midpoint = (head_price + neck2_price) / 2.0
                    if left_price >= left_midpoint or right_price >= right_midpoint:
                        continue

                    left_width = head_idx - left_idx
                    right_width = right_idx - head_idx
                    if left_width < 1 or right_width < 1:
                        continue
                    symmetry_ratio = max(left_width / right_width, right_width / left_width)
                    if symmetry_ratio > max_symmetry_ratio:
                        continue

                    post_right_swing_invalidation = _has_any_swing_between(
                        ordered_swings,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        require_direct_breakout_after_right_shoulder
                        and post_right_swing_invalidation
                    ):
                        continue

                    bullish_structure_break = _has_bullish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if require_bullish_structure_break and not bullish_structure_break:
                        continue

                    neckline_zone_bottom = neckline_entry_price * (
                        1.0 - neckline_zone_pct / 100.0
                    )
                    neckline_zone_top = neckline_entry_price * (
                        1.0 + neckline_zone_pct / 100.0
                    )
                    pattern_height = neckline_entry_price - head_price
                    target_price = neckline_entry_price + pattern_height
                    follow_through_level = neckline_entry_price + (
                        pattern_height * follow_through_pattern_ratio
                    )
                    follow_through_idx = _find_inverse_head_shoulders_follow_through(
                        closes=closes,
                        entry_idx=entry_idx,
                        max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                        follow_through_level=follow_through_level,
                        neckline_start_idx=neck1_idx,
                        neckline_start_price=neck1_price,
                        neckline_end_idx=neck2_idx,
                        neckline_end_price=neck2_price,
                        neckline_zone_pct=neckline_zone_pct,
                        require_hold_neckline=require_breakout_hold_neckline,
                        fail_level=head_fail_level,
                    )
                    if require_breakout_follow_through and follow_through_idx is None:
                        continue

                    pattern_start = _find_double_bottom_start(
                        data,
                        b1_idx=left_idx,
                        resistance_level=neck1_price,
                        fallback_idx=context_idx,
                        fallback_price=context_price,
                        lookback_bars=downtrend_lookback_bars,
                    )
                    pattern_r2 = _compute_inverse_head_shoulders_r2(
                        closes=closes,
                        start_idx=int(pattern_start["BarIdx"]),
                        start_price=float(pattern_start["Price"]),
                        left_idx=left_idx,
                        left_price=left_price,
                        neck1_idx=neck1_idx,
                        neck1_price=neck1_price,
                        head_idx=head_idx,
                        head_price=head_price,
                        neck2_idx=neck2_idx,
                        neck2_price=neck2_price,
                        right_idx=right_idx,
                        right_price=right_price,
                        entry_idx=entry_idx,
                        entry_price=float(closes[entry_idx]),
                    )
                    if (
                        min_pattern_r2 is not None
                        and (
                            not np.isfinite(pattern_r2)
                            or float(pattern_r2) < float(min_pattern_r2)
                        )
                    ):
                        continue

                    entry_volume_ratio = (
                        float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                    )
                    volume_confirmed = (
                        bool(
                            np.isfinite(entry_volume_ratio)
                            and entry_volume_ratio >= volume_multiplier
                        )
                        if has_volume
                        else False
                    )
                    if require_volume_confirmation and not volume_confirmed:
                        continue

                    pattern_key = (left_idx, neck1_idx, head_idx, neck2_idx, right_idx)
                    if pattern_key in seen_patterns:
                        continue
                    seen_patterns.add(pattern_key)

                    stop_price = head_price
                    risk = closes[entry_idx] - stop_price
                    reward = target_price - closes[entry_idx]

                    patterns.append(
                        {
                            "PatternId": len(patterns) + 1,
                            "ContextHighBarIdx": context_idx,
                            "ContextHighDate": context["Date"],
                            "ContextHighPrice": context_price,
                            "LastCHoCHBeforePatternBarIdx": int(last_choch["EventBarIdx"])
                            if last_choch is not None
                            else np.nan,
                            "LastCHoCHBeforePatternDate": last_choch["EventDate"]
                            if last_choch is not None
                            else pd.NaT,
                            "LastCHoCHBeforePatternDirection": str(last_choch["Direction"])
                            if last_choch is not None
                            else None,
                            "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                            "PatternStartDate": pattern_start["Date"],
                            "PatternStartPrice": float(pattern_start["Price"]),
                            "LeftShoulderBarIdx": left_idx,
                            "LeftShoulderDate": left_shoulder["Date"],
                            "LeftShoulderPrice": left_price,
                            "Neckline1BarIdx": neck1_idx,
                            "Neckline1Date": neck1["Date"],
                            "Neckline1Price": neck1_price,
                            "HeadBarIdx": head_idx,
                            "HeadDate": head["Date"],
                            "HeadPrice": head_price,
                            "Neckline2BarIdx": neck2_idx,
                            "Neckline2Date": neck2["Date"],
                            "Neckline2Price": neck2_price,
                            "RightShoulderBarIdx": right_idx,
                            "RightShoulderDate": data.index[right_idx],
                            "RightShoulderPrice": right_price,
                            "EntryBarIdx": entry_idx,
                            "EntryDate": data.index[entry_idx],
                            "EntryPrice": float(closes[entry_idx]),
                            "NecklineEntryPrice": float(neckline_entry_price),
                            "FollowThroughBarIdx": int(follow_through_idx)
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughDate": data.index[int(follow_through_idx)]
                            if follow_through_idx is not None
                            else pd.NaT,
                            "FollowThroughPrice": float(closes[int(follow_through_idx)])
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughLevel": float(follow_through_level),
                            "TargetPrice": float(target_price),
                            "StopPrice": float(stop_price),
                            "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                            "ShoulderDifferencePct": float(shoulder_difference_pct),
                            "HeadDepthPct": float(head_depth_pct),
                            "NecklineTiltPct": float(neckline_tilt_pct),
                            "NecklineZoneBottom": float(neckline_zone_bottom),
                            "NecklineZoneTop": float(neckline_zone_top),
                            "DowntrendPct": float(downtrend_pct),
                            "PatternWidth": int(right_idx - left_idx),
                            "PatternHeight": float(pattern_height),
                            "SymmetryRatio": float(symmetry_ratio),
                            "PatternR2": float(pattern_r2),
                            "PriorLHLL": bool(prior_lh_ll),
                            "LastCHoCHBearish": bool(last_choch_bearish),
                            "HeadLowestInvalidation": bool(head_lowest_invalidation),
                            "PostRightShoulderSwingInvalidation": bool(
                                post_right_swing_invalidation
                            ),
                            "BullishStructureBreak": bool(bullish_structure_break),
                            "VolumeRatio": entry_volume_ratio,
                            "VolumeConfirmed": bool(volume_confirmed),
                        }
                    )
                    break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_inverse_head_shoulders_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_inverse_head_shoulders(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _select_best_pattern_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforeB1BarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforeB1BarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "ResistancePrice",
                "Bottom1BarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _score_double_bottom_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    bottom1_price = max(abs(_finite_or_zero(row.get("Bottom1Price"))), 1e-12)
    height_pct = pattern_height / bottom1_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 80.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    bottom_diff_penalty = _finite_or_zero(row.get("BottomDifferencePct"))
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("Bottom2BarIdx")),
        0.0,
    )

    return (
        height_pct * 2.5
        + width_score
        + downtrend_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - bottom_diff_penalty * 2.0
        - entry_delay * 0.02
    )


def _select_best_inverse_head_shoulders_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforePatternBarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforePatternBarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "PatternHeight",
                "LeftShoulderBarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_non_overlapping_inverse_head_shoulders(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["LeftShoulderBarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _interval_overlap_pct(
    first: tuple[int, int],
    second: tuple[int, int],
) -> float:
    first_start, first_end = first
    second_start, second_end = second
    overlap = max(0, min(first_end, second_end) - max(first_start, second_start))
    if overlap <= 0:
        return 0.0

    first_width = max(first_end - first_start, 1)
    second_width = max(second_end - second_start, 1)
    return overlap / min(first_width, second_width)


def _score_inverse_head_shoulders_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    head_price = max(abs(_finite_or_zero(row.get("HeadPrice"))), 1e-12)
    height_pct = pattern_height / head_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 100.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    head_depth_score = _finite_or_zero(row.get("HeadDepthPct")) / 5.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    shoulder_penalty = _finite_or_zero(row.get("ShoulderDifferencePct")) / 2.0
    neckline_penalty = _finite_or_zero(row.get("NecklineTiltPct")) / 4.0
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("RightShoulderBarIdx")),
        0.0,
    )

    return (
        height_pct * 2.0
        + width_score
        + downtrend_score
        + head_depth_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - shoulder_penalty
        - neckline_penalty
        - entry_delay * 0.02
    )


def _finite_or_zero(value: Any, *, default: float = 0.0) -> float:
    try:
        number = float(value)
    except (TypeError, ValueError):
        return default
    if not np.isfinite(number):
        return default
    return number


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _find_last_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> pd.Series | None:
    if structure_events is None or structure_events.empty:
        return None

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    if subset.empty:
        return None

    return subset.sort_values("EventBarIdx").iloc[-1]


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if price > support_zone_top:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
    resistance_zone_bottom: float | None = None,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if (
            current_idx > entry_idx
            and resistance_zone_bottom is not None
            and close_price < resistance_zone_bottom
        ):
            return None
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _line_value(
    start_idx: int,
    start_price: float,
    end_idx: int,
    end_price: float,
    target_idx: int,
) -> float:
    if int(end_idx) == int(start_idx):
        return float(end_price)

    slope = (float(end_price) - float(start_price)) / (int(end_idx) - int(start_idx))
    return float(start_price) + slope * (int(target_idx) - int(start_idx))


def _find_inverse_head_shoulders_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    neckline_start_idx: int,
    neckline_start_price: float,
    neckline_end_idx: int,
    neckline_end_price: float,
    neckline_zone_pct: float,
    require_hold_neckline: bool,
    fail_level: float,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx

        if current_idx <= entry_idx:
            continue

        neckline_price = _line_value(
            neckline_start_idx,
            neckline_start_price,
            neckline_end_idx,
            neckline_end_price,
            current_idx,
        )
        neckline_bottom = neckline_price * (1.0 - neckline_zone_pct / 100.0)
        if require_hold_neckline and close_price < neckline_bottom:
            return None
        if close_price < fail_level:
            return None

    return None


def _find_double_bottom_start(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    resistance_level: float,
    fallback_idx: int,
    fallback_price: float,
    lookback_bars: int,
) -> dict[str, Any]:
    search_start = max(0, int(b1_idx) - int(lookback_bars))
    closes = data["close"].to_numpy(dtype=float)
    highs = data["high"].to_numpy(dtype=float)

    for current_idx in range(int(b1_idx) - 1, search_start - 1, -1):
        if closes[current_idx] >= resistance_level or highs[current_idx] >= resistance_level:
            return {
                "BarIdx": current_idx,
                "Date": data.index[current_idx],
                "Price": float(closes[current_idx]),
            }

    return {
        "BarIdx": int(fallback_idx),
        "Date": data.index[int(fallback_idx)],
        "Price": float(fallback_price),
    }


def _compute_double_bottom_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    b1_idx: int,
    b1_price: float,
    resistance_idx: int,
    resistance_price: float,
    b2_idx: int,
    b2_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, b1_idx, resistance_idx, b2_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [start_price, b1_price, resistance_price, b2_price, entry_price],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _compute_inverse_head_shoulders_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    left_idx: int,
    left_price: float,
    neck1_idx: int,
    neck1_price: float,
    head_idx: int,
    head_price: float,
    neck2_idx: int,
    neck2_price: float,
    right_idx: int,
    right_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, left_idx, neck1_idx, head_idx, neck2_idx, right_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [
            start_price,
            left_price,
            neck1_price,
            head_price,
            neck2_price,
            right_price,
            entry_price,
        ],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
    inverse_head_shoulders: pd.DataFrame | None = None,
    show_inverse_head_shoulders: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for pattern_number, (_, pattern) in enumerate(double_bottoms.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")
            point_indexes.extend(
                [
                    int(pattern["Bottom1BarIdx"]),
                    int(pattern["ResistanceBarIdx"]),
                    int(pattern["Bottom2BarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["Bottom1Price"]),
                    float(pattern["ResistancePrice"]),
                    float(pattern["Bottom2Price"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["B1", "Resistance", "B2", "Breakout"])
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    showlegend=pattern_number == 0,
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforeB1Direction', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    if (
        show_inverse_head_shoulders
        and inverse_head_shoulders is not None
        and not inverse_head_shoulders.empty
    ):
        ihs_color = "#38BDF8"
        neckline_color = "#E0F2FE"

        for pattern_number, (_, pattern) in enumerate(inverse_head_shoulders.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in inverse_head_shoulders.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")

            point_indexes.extend(
                [
                    int(pattern["LeftShoulderBarIdx"]),
                    int(pattern["Neckline1BarIdx"]),
                    int(pattern["HeadBarIdx"]),
                    int(pattern["Neckline2BarIdx"]),
                    int(pattern["RightShoulderBarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["LeftShoulderPrice"]),
                    float(pattern["Neckline1Price"]),
                    float(pattern["HeadPrice"]),
                    float(pattern["Neckline2Price"]),
                    float(pattern["RightShoulderPrice"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["LS", "N1", "Head", "N2", "RS", "Breakout"])

            if (
                "FollowThroughBarIdx" in inverse_head_shoulders.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")

            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [data.index[idx] for idx in point_indexes]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Inverse H&S",
                    showlegend=pattern_number == 0,
                    line=dict(color=ihs_color, width=4),
                    marker=dict(
                        color=ihs_color,
                        size=10,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color=ihs_color, size=11),
                    hovertext=(
                        f"Inverse H&S #{int(pattern['PatternId'])}<br>"
                        f"Shoulder diff: {float(pattern['ShoulderDifferencePct']):.2f}%<br>"
                        f"Head depth: {float(pattern['HeadDepthPct']):.2f}%<br>"
                        f"Neckline tilt: {float(pattern['NecklineTiltPct']):.2f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforePatternDirection', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            neckline1_idx = int(pattern["Neckline1BarIdx"])
            neckline2_idx = int(pattern["Neckline2BarIdx"])
            neckline1_price = float(pattern["Neckline1Price"])
            neckline2_price = float(pattern["Neckline2Price"])
            neckline_end_price = _line_value(
                neckline1_idx,
                neckline1_price,
                neckline2_idx,
                neckline2_price,
                end_idx,
            )
            neckline_x = (
                [neckline1_idx, end_idx]
                if remove_time_gaps
                else [pattern["Neckline1Date"], data.index[end_idx]]
            )
            fig.add_trace(
                go.Scatter(
                    x=neckline_x,
                    y=[neckline1_price, neckline_end_price],
                    mode="lines+text",
                    showlegend=False,
                    line=dict(color=neckline_color, width=2),
                    text=["Neckline Break", ""],
                    textposition="top center",
                    hoverinfo="x+y+text",
                )
            )

            zone_pct = (
                abs(float(pattern["NecklineZoneTop"]) - float(pattern["NecklineEntryPrice"]))
                / max(abs(float(pattern["NecklineEntryPrice"])), 1e-12)
                * 100.0
            )
            for multiplier in [1.0 - zone_pct / 100.0, 1.0 + zone_pct / 100.0]:
                fig.add_trace(
                    go.Scatter(
                        x=neckline_x,
                        y=[neckline1_price * multiplier, neckline_end_price * multiplier],
                        mode="lines",
                        showlegend=False,
                        line=dict(color=neckline_color, width=1, dash="dot"),
                        hoverinfo="skip",
                    )
                )

            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )
            for name, price, color, dash in [
                ("IHS Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("IHS Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                fig.add_trace(
                    go.Scatter(
                        x=projection_x,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "AAPL"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_breakout_hold_resistance=True,
        one_pattern_per_bearish_choch=True,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )
    inverse_head_shoulders_df = detect_inverse_head_shoulders(
        raw,
        swings_df,
        structure_events=structure_df,
        shoulder_tolerance_pct=3.0,
        min_head_depth_pct=0.75,
        head_lowest_tolerance_pct=0.15,
        neckline_zone_pct=0.25,
        max_neckline_tilt_pct=4.0,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        neckline_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_last_choch_bearish=True,
        require_head_lowest=True,
        require_bullish_structure_break=True,
        require_direct_breakout_after_right_shoulder=True,
        require_breakout_hold_neckline=True,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))
    print(inverse_head_shoulders_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
        inverse_head_shoulders=inverse_head_shoulders_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    max_symmetry_ratio: float = 2.5,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_resistance: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "LastCHoCHBeforeB1BarIdx",
        "LastCHoCHBeforeB1Date",
        "LastCHoCHBeforeB1Direction",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch_before_b1 = _find_last_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        bearish_choch_before_b1 = (
            last_choch_before_b1 is not None
            and str(last_choch_before_b1["Direction"]) == "Bearish"
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        resistance_candidates = ordered_swings[
            (ordered_swings["SwingType"] == "High")
            & (ordered_swings["BarIdx"].astype(int) > b1_idx)
            & (
                ordered_swings["BarIdx"].astype(int)
                < b1_idx + max_bars_between_bottoms
            )
        ]

        for _, resistance in resistance_candidates.iterrows():
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= b1_price:
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            earliest_breakout_idx = max(
                int(resistance["ConfirmBarIdx"]) + 1,
                resistance_idx + 2,
            )
            latest_breakout_idx = min(
                len(data) - 1,
                earliest_breakout_idx + max_entry_wait_bars,
            )

            for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                if closes[entry_idx] < support_zone_bottom:
                    break

                if closes[entry_idx] <= breakout_level:
                    continue

                right_lows = lows[resistance_idx + 1 : entry_idx]
                if len(right_lows) == 0:
                    continue

                b2_idx = resistance_idx + 1 + int(np.argmin(right_lows))
                b2_price = float(lows[b2_idx])
                bars_between = b2_idx - b1_idx
                if bars_between < min_bars_between_bottoms:
                    continue
                if bars_between > max_bars_between_bottoms:
                    break

                bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
                if bottom_difference_pct > bottom_tolerance_pct:
                    continue
                if not support_zone_bottom <= b2_price <= support_zone_top:
                    continue
                if resistance_price <= max(b1_price, b2_price):
                    continue

                left_width = resistance_idx - b1_idx
                right_width = b2_idx - resistance_idx
                if left_width < 1 or right_width < 1:
                    continue
                symmetry_ratio = max(left_width / right_width, right_width / left_width)
                if symmetry_ratio > max_symmetry_ratio:
                    continue

                left_midpoint = 0.5 * (b1_price + resistance_price)
                right_midpoint = 0.5 * (b2_price + resistance_price)
                if b1_price > right_midpoint or b2_price > left_midpoint:
                    continue

                highs_between = ordered_swings[
                    (ordered_swings["SwingType"] == "High")
                    & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                    & (ordered_swings["BarIdx"].astype(int) < b2_idx)
                ]
                if highs_between.empty:
                    continue
                highest_between = highs_between.sort_values(
                    ["Price", "BarIdx"],
                    ascending=[False, False],
                ).iloc[0]
                if int(highest_between["BarIdx"]) != resistance_idx:
                    continue

                second_peak_invalidation = _has_second_peak_between_zones(
                    ordered_swings,
                    start_idx=resistance_idx,
                    end_idx=b2_idx,
                    support_zone_top=support_zone_top,
                    resistance_zone_bottom=resistance_zone_bottom,
                )
                if invalidate_second_peak_between_zones and second_peak_invalidation:
                    continue

                support_fail_level = min(b1_price, b2_price) * (
                    1.0 - fail_buffer_pct / 100.0
                )
                if np.min(closes[resistance_idx + 1 : entry_idx + 1]) < support_fail_level:
                    continue

                post_b2_swing_invalidation = _has_any_swing_between(
                    ordered_swings,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                    continue

                bullish_structure_break = _has_bullish_structure_break(
                    structure_events,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_bullish_structure_break and not bullish_structure_break:
                    continue

                pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
                target_price = resistance_zone_top + pattern_height
                follow_through_level = resistance_zone_top + (
                    pattern_height * follow_through_pattern_ratio
                )
                follow_through_idx = _find_breakout_follow_through(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                    follow_through_level=follow_through_level,
                    support_zone_bottom=support_zone_bottom,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                )
                if require_breakout_follow_through and follow_through_idx is None:
                    continue

                pattern_start = _find_double_bottom_start(
                    data,
                    b1_idx=b1_idx,
                    resistance_level=resistance_zone_bottom,
                    fallback_idx=context_idx,
                    fallback_price=context_price,
                    lookback_bars=downtrend_lookback_bars,
                )
                pattern_r2 = _compute_double_bottom_r2(
                    closes=closes,
                    start_idx=int(pattern_start["BarIdx"]),
                    start_price=float(pattern_start["Price"]),
                    b1_idx=b1_idx,
                    b1_price=b1_price,
                    resistance_idx=resistance_idx,
                    resistance_price=resistance_price,
                    b2_idx=b2_idx,
                    b2_price=b2_price,
                    entry_idx=entry_idx,
                    entry_price=float(closes[entry_idx]),
                )
                if (
                    min_pattern_r2 is not None
                    and (
                        not np.isfinite(pattern_r2)
                        or float(pattern_r2) < float(min_pattern_r2)
                    )
                ):
                    continue

                entry_volume_ratio = (
                    float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                )
                volume_confirmed = (
                    bool(
                        np.isfinite(entry_volume_ratio)
                        and entry_volume_ratio >= volume_multiplier
                    )
                    if has_volume
                    else False
                )
                if require_volume_confirmation and not volume_confirmed:
                    continue

                pattern_key = (b1_idx, resistance_idx, b2_idx)
                if pattern_key in seen_patterns:
                    continue
                seen_patterns.add(pattern_key)

                stop_price = min(b1_price, b2_price)
                risk = closes[entry_idx] - stop_price
                reward = target_price - closes[entry_idx]

                patterns.append(
                    {
                        "PatternId": len(patterns) + 1,
                        "ContextHighBarIdx": context_idx,
                        "ContextHighDate": context["Date"],
                        "ContextHighPrice": context_price,
                        "PreviousHighBarIdx": previous_high_idx,
                        "PreviousHighDate": previous_high["Date"],
                        "PreviousHighPrice": previous_high_price,
                        "LastCHoCHBeforeB1BarIdx": int(
                            last_choch_before_b1["EventBarIdx"]
                        )
                        if last_choch_before_b1 is not None
                        else np.nan,
                        "LastCHoCHBeforeB1Date": last_choch_before_b1["EventDate"]
                        if last_choch_before_b1 is not None
                        else pd.NaT,
                        "LastCHoCHBeforeB1Direction": str(
                            last_choch_before_b1["Direction"]
                        )
                        if last_choch_before_b1 is not None
                        else None,
                        "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                        "PatternStartDate": pattern_start["Date"],
                        "PatternStartPrice": float(pattern_start["Price"]),
                        "Bottom1BarIdx": b1_idx,
                        "Bottom1Date": bottom1["Date"],
                        "Bottom1Price": b1_price,
                        "ResistanceBarIdx": resistance_idx,
                        "ResistanceDate": resistance["Date"],
                        "ResistancePrice": resistance_price,
                        "Bottom2BarIdx": b2_idx,
                        "Bottom2Date": data.index[b2_idx],
                        "Bottom2Price": b2_price,
                        "EntryBarIdx": entry_idx,
                        "EntryDate": data.index[entry_idx],
                        "EntryPrice": float(closes[entry_idx]),
                        "FollowThroughBarIdx": int(follow_through_idx)
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughDate": data.index[int(follow_through_idx)]
                        if follow_through_idx is not None
                        else pd.NaT,
                        "FollowThroughPrice": float(closes[int(follow_through_idx)])
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughLevel": float(follow_through_level),
                        "TargetPrice": float(target_price),
                        "StopPrice": float(stop_price),
                        "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                        "BottomDifferencePct": float(bottom_difference_pct),
                        "SupportZoneBottom": float(support_zone_bottom),
                        "SupportZoneTop": float(support_zone_top),
                        "ResistanceZoneBottom": float(resistance_zone_bottom),
                        "ResistanceZoneTop": float(resistance_zone_top),
                        "RetracePct": float(retrace_pct),
                        "DowntrendPct": float(downtrend_pct),
                        "PatternWidth": int(b2_idx - b1_idx),
                        "PatternHeight": float(pattern_height),
                        "SymmetryRatio": float(symmetry_ratio),
                        "PatternR2": float(pattern_r2),
                        "PriorLHLL": bool(prior_lh_ll),
                        "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                        "SecondPeakInvalidation": bool(second_peak_invalidation),
                        "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                        "BullishStructureBreak": bool(bullish_structure_break),
                        "VolumeRatio": entry_volume_ratio,
                        "VolumeConfirmed": bool(volume_confirmed),
                    }
                )
                break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_pattern_per_choch(result, columns=columns)
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def detect_inverse_head_shoulders(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    shoulder_tolerance_pct: float = 3.0,
    min_head_depth_pct: float = 0.75,
    head_lowest_tolerance_pct: float = 0.15,
    head_support_zone_pct: float = 0.25,
    neckline_zone_pct: float = 0.25,
    max_neckline_tilt_pct: float = 1.0,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_points: int = 3,
    max_pattern_bars: int = 260,
    max_symmetry_ratio: float = 2.5,
    neckline_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_prior_lh_ll: bool = True,
    require_last_choch_bearish: bool = True,
    require_head_lowest: bool = True,
    require_bullish_structure_break: bool = True,
    require_direct_breakout_after_right_shoulder: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_neckline: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed inverse head and shoulders patterns.

    Sequence:
    - prior bearish CHoCH/downtrend
    - left shoulder low
    - neckline high 1
    - deeper head low
    - neckline high 2
    - right shoulder low
    - close above the projected neckline
    """
    if shoulder_tolerance_pct < 0:
        raise ValueError("shoulder_tolerance_pct cannot be negative.")
    if min_head_depth_pct < 0:
        raise ValueError("min_head_depth_pct cannot be negative.")
    if head_lowest_tolerance_pct < 0:
        raise ValueError("head_lowest_tolerance_pct cannot be negative.")
    if head_support_zone_pct < 0:
        raise ValueError("head_support_zone_pct cannot be negative.")
    if neckline_zone_pct < 0:
        raise ValueError("neckline_zone_pct cannot be negative.")
    if max_neckline_tilt_pct < 0:
        raise ValueError("max_neckline_tilt_pct cannot be negative.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_points < 1:
        raise ValueError("min_bars_between_points must be at least 1.")
    if max_pattern_bars < min_bars_between_points * 4:
        raise ValueError("max_pattern_bars is too small for an IHS pattern.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if neckline_break_buffer_pct < 0:
        raise ValueError("neckline_break_buffer_pct cannot be negative.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "LastCHoCHBeforePatternBarIdx",
        "LastCHoCHBeforePatternDate",
        "LastCHoCHBeforePatternDirection",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "LeftShoulderBarIdx",
        "LeftShoulderDate",
        "LeftShoulderPrice",
        "Neckline1BarIdx",
        "Neckline1Date",
        "Neckline1Price",
        "HeadBarIdx",
        "HeadDate",
        "HeadPrice",
        "Neckline2BarIdx",
        "Neckline2Date",
        "Neckline2Price",
        "RightShoulderBarIdx",
        "RightShoulderDate",
        "RightShoulderPrice",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "NecklineEntryPrice",
        "ResistancePrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "ShoulderDifferencePct",
        "HeadDepthPct",
        "NecklineTiltPct",
        "HeadSupportZoneBottom",
        "HeadSupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "NecklineZoneBottom",
        "NecklineZoneTop",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "LastCHoCHBearish",
        "HeadLowestInvalidation",
        "PostRightShoulderSwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for left_pos in range(len(lows_only) - 1):
        left_shoulder = lows_only.iloc[left_pos]
        left_idx = int(left_shoulder["BarIdx"])
        left_price = float(left_shoulder["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=left_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch = _find_last_choch_before(
            structure_events,
            before_idx=left_idx,
        )
        last_choch_bearish = (
            last_choch is not None and str(last_choch["Direction"]) == "Bearish"
        )
        if require_last_choch_bearish and not last_choch_bearish:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=left_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=left_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        if context_price <= left_price:
            continue

        head_candidates = lows_only[
            (lows_only["BarIdx"].astype(int) > left_idx + min_bars_between_points)
            & (lows_only["BarIdx"].astype(int) < left_idx + max_pattern_bars)
        ]

        for _, head in head_candidates.iterrows():
            head_idx = int(head["BarIdx"])
            head_price = float(head["Price"])
            if head_price >= left_price * (1.0 - min_head_depth_pct / 100.0):
                continue

            downtrend_pct = (context_price - head_price) / context_price * 100.0
            if downtrend_pct < min_downtrend_pct:
                continue

            leg_lows = ordered_swings[
                (ordered_swings["SwingType"] == "Low")
                & (ordered_swings["BarIdx"].astype(int) >= context_idx)
                & (ordered_swings["BarIdx"].astype(int) <= head_idx)
            ]
            if (
                not leg_lows.empty
                and head_price > float(leg_lows["Price"].min()) * 1.001
            ):
                continue

            left_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > left_idx)
                & (ordered_swings["BarIdx"].astype(int) < head_idx)
            ]
            if left_neck_candidates.empty:
                continue

            neck1 = left_neck_candidates.sort_values(
                ["Price", "BarIdx"],
                ascending=[False, False],
            ).iloc[0]
            neck1_idx = int(neck1["BarIdx"])
            neck1_price = float(neck1["Price"])
            if neck1_price <= left_price or neck1_price <= head_price:
                continue

            right_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > head_idx)
                & (ordered_swings["BarIdx"].astype(int) < left_idx + max_pattern_bars)
            ]
            if right_neck_candidates.empty:
                continue

            for _, neck2 in right_neck_candidates.iterrows():
                neck2_idx = int(neck2["BarIdx"])
                neck2_price = float(neck2["Price"])
                if neck2_idx - head_idx < min_bars_between_points:
                    continue
                if neck2_price <= head_price:
                    continue

                neckline_average = (neck1_price + neck2_price) / 2.0
                neckline_tilt_pct = (
                    abs(neck2_price - neck1_price)
                    / max(abs(neckline_average), 1e-12)
                    * 100.0
                )
                resistance_price = neckline_average
                resistance_zone_bottom = resistance_price * (
                    1.0 - neckline_zone_pct / 100.0
                )
                resistance_zone_top = resistance_price * (
                    1.0 + neckline_zone_pct / 100.0
                )
                if not (
                    resistance_zone_bottom <= neck1_price <= resistance_zone_top
                    and resistance_zone_bottom <= neck2_price <= resistance_zone_top
                ):
                    continue
                if neckline_tilt_pct > max_neckline_tilt_pct:
                    continue

                head_support_zone_bottom = head_price * (
                    1.0 - head_support_zone_pct / 100.0
                )
                head_support_zone_top = head_price * (
                    1.0 + head_support_zone_pct / 100.0
                )
                earliest_breakout_idx = max(
                    int(neck2["ConfirmBarIdx"]) + 1,
                    neck2_idx + 2,
                )
                latest_breakout_idx = min(
                    len(data) - 1,
                    earliest_breakout_idx + max_entry_wait_bars,
                )
                head_fail_level = head_support_zone_bottom * (
                    1.0 - fail_buffer_pct / 100.0
                )

                for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                    if closes[entry_idx] < head_fail_level:
                        break

                    neckline_entry_price = resistance_price
                    breakout_level = resistance_zone_top * (
                        1.0 + neckline_break_buffer_pct / 100.0
                    )
                    if closes[entry_idx] <= breakout_level:
                        continue

                    right_lows = lows[neck2_idx + 1 : entry_idx]
                    if len(right_lows) == 0:
                        continue

                    right_idx = neck2_idx + 1 + int(np.argmin(right_lows))
                    right_price = float(lows[right_idx])
                    if right_idx - neck2_idx < min_bars_between_points:
                        continue
                    if right_price <= head_price * (1.0 + min_head_depth_pct / 100.0):
                        continue

                    pattern_low = float(np.min(lows[left_idx : entry_idx + 1]))
                    head_lowest_invalidation = pattern_low < head_price * (
                        1.0 - head_lowest_tolerance_pct / 100.0
                    )
                    if require_head_lowest and head_lowest_invalidation:
                        continue

                    if not (
                        head_support_zone_top < left_price < resistance_zone_bottom
                        and head_support_zone_top < right_price < resistance_zone_bottom
                    ):
                        continue

                    shoulder_average = (left_price + right_price) / 2.0
                    shoulder_difference_pct = (
                        abs(right_price - left_price)
                        / max(abs(shoulder_average), 1e-12)
                        * 100.0
                    )
                    if shoulder_difference_pct > shoulder_tolerance_pct:
                        continue

                    head_depth_pct = (
                        (min(left_price, right_price) - head_price)
                        / max(abs(min(left_price, right_price)), 1e-12)
                        * 100.0
                    )
                    if head_depth_pct < min_head_depth_pct:
                        continue

                    if neck2_price <= right_price:
                        continue

                    left_midpoint = (head_price + neck1_price) / 2.0
                    right_midpoint = (head_price + neck2_price) / 2.0
                    if left_price >= left_midpoint or right_price >= right_midpoint:
                        continue

                    left_width = head_idx - left_idx
                    right_width = right_idx - head_idx
                    if left_width < 1 or right_width < 1:
                        continue
                    symmetry_ratio = max(left_width / right_width, right_width / left_width)
                    if symmetry_ratio > max_symmetry_ratio:
                        continue

                    post_right_swing_invalidation = _has_any_swing_between(
                        ordered_swings,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        require_direct_breakout_after_right_shoulder
                        and post_right_swing_invalidation
                    ):
                        continue

                    bullish_structure_break = _has_bullish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if require_bullish_structure_break and not bullish_structure_break:
                        continue

                    neckline_zone_bottom = resistance_zone_bottom
                    neckline_zone_top = resistance_zone_top
                    pattern_height = resistance_zone_top - head_price
                    target_price = resistance_zone_top + pattern_height
                    follow_through_level = resistance_zone_top + (
                        pattern_height * follow_through_pattern_ratio
                    )
                    follow_through_idx = _find_breakout_follow_through(
                        closes=closes,
                        entry_idx=entry_idx,
                        max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                        follow_through_level=follow_through_level,
                        support_zone_bottom=head_support_zone_bottom,
                        resistance_zone_bottom=resistance_zone_bottom
                        if require_breakout_hold_neckline
                        else None,
                    )
                    if require_breakout_follow_through and follow_through_idx is None:
                        continue

                    pattern_start = _find_double_bottom_start(
                        data,
                        b1_idx=left_idx,
                        resistance_level=neck1_price,
                        fallback_idx=context_idx,
                        fallback_price=context_price,
                        lookback_bars=downtrend_lookback_bars,
                    )
                    pattern_r2 = _compute_inverse_head_shoulders_r2(
                        closes=closes,
                        start_idx=int(pattern_start["BarIdx"]),
                        start_price=float(pattern_start["Price"]),
                        left_idx=left_idx,
                        left_price=left_price,
                        neck1_idx=neck1_idx,
                        neck1_price=neck1_price,
                        head_idx=head_idx,
                        head_price=head_price,
                        neck2_idx=neck2_idx,
                        neck2_price=neck2_price,
                        right_idx=right_idx,
                        right_price=right_price,
                        entry_idx=entry_idx,
                        entry_price=float(closes[entry_idx]),
                    )
                    if (
                        min_pattern_r2 is not None
                        and (
                            not np.isfinite(pattern_r2)
                            or float(pattern_r2) < float(min_pattern_r2)
                        )
                    ):
                        continue

                    entry_volume_ratio = (
                        float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                    )
                    volume_confirmed = (
                        bool(
                            np.isfinite(entry_volume_ratio)
                            and entry_volume_ratio >= volume_multiplier
                        )
                        if has_volume
                        else False
                    )
                    if require_volume_confirmation and not volume_confirmed:
                        continue

                    pattern_key = (left_idx, neck1_idx, head_idx, neck2_idx, right_idx)
                    if pattern_key in seen_patterns:
                        continue
                    seen_patterns.add(pattern_key)

                    stop_price = head_price
                    risk = closes[entry_idx] - stop_price
                    reward = target_price - closes[entry_idx]

                    patterns.append(
                        {
                            "PatternId": len(patterns) + 1,
                            "ContextHighBarIdx": context_idx,
                            "ContextHighDate": context["Date"],
                            "ContextHighPrice": context_price,
                            "LastCHoCHBeforePatternBarIdx": int(last_choch["EventBarIdx"])
                            if last_choch is not None
                            else np.nan,
                            "LastCHoCHBeforePatternDate": last_choch["EventDate"]
                            if last_choch is not None
                            else pd.NaT,
                            "LastCHoCHBeforePatternDirection": str(last_choch["Direction"])
                            if last_choch is not None
                            else None,
                            "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                            "PatternStartDate": pattern_start["Date"],
                            "PatternStartPrice": float(pattern_start["Price"]),
                            "LeftShoulderBarIdx": left_idx,
                            "LeftShoulderDate": left_shoulder["Date"],
                            "LeftShoulderPrice": left_price,
                            "Neckline1BarIdx": neck1_idx,
                            "Neckline1Date": neck1["Date"],
                            "Neckline1Price": neck1_price,
                            "HeadBarIdx": head_idx,
                            "HeadDate": head["Date"],
                            "HeadPrice": head_price,
                            "Neckline2BarIdx": neck2_idx,
                            "Neckline2Date": neck2["Date"],
                            "Neckline2Price": neck2_price,
                            "RightShoulderBarIdx": right_idx,
                            "RightShoulderDate": data.index[right_idx],
                            "RightShoulderPrice": right_price,
                            "EntryBarIdx": entry_idx,
                            "EntryDate": data.index[entry_idx],
                            "EntryPrice": float(closes[entry_idx]),
                            "NecklineEntryPrice": float(neckline_entry_price),
                            "ResistancePrice": float(resistance_price),
                            "FollowThroughBarIdx": int(follow_through_idx)
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughDate": data.index[int(follow_through_idx)]
                            if follow_through_idx is not None
                            else pd.NaT,
                            "FollowThroughPrice": float(closes[int(follow_through_idx)])
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughLevel": float(follow_through_level),
                            "TargetPrice": float(target_price),
                            "StopPrice": float(stop_price),
                            "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                            "ShoulderDifferencePct": float(shoulder_difference_pct),
                            "HeadDepthPct": float(head_depth_pct),
                            "NecklineTiltPct": float(neckline_tilt_pct),
                            "HeadSupportZoneBottom": float(head_support_zone_bottom),
                            "HeadSupportZoneTop": float(head_support_zone_top),
                            "ResistanceZoneBottom": float(resistance_zone_bottom),
                            "ResistanceZoneTop": float(resistance_zone_top),
                            "NecklineZoneBottom": float(neckline_zone_bottom),
                            "NecklineZoneTop": float(neckline_zone_top),
                            "DowntrendPct": float(downtrend_pct),
                            "PatternWidth": int(right_idx - left_idx),
                            "PatternHeight": float(pattern_height),
                            "SymmetryRatio": float(symmetry_ratio),
                            "PatternR2": float(pattern_r2),
                            "PriorLHLL": bool(prior_lh_ll),
                            "LastCHoCHBearish": bool(last_choch_bearish),
                            "HeadLowestInvalidation": bool(head_lowest_invalidation),
                            "PostRightShoulderSwingInvalidation": bool(
                                post_right_swing_invalidation
                            ),
                            "BullishStructureBreak": bool(bullish_structure_break),
                            "VolumeRatio": entry_volume_ratio,
                            "VolumeConfirmed": bool(volume_confirmed),
                        }
                    )
                    break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_inverse_head_shoulders_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_inverse_head_shoulders(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _select_best_pattern_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforeB1BarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforeB1BarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "ResistancePrice",
                "Bottom1BarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _score_double_bottom_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    bottom1_price = max(abs(_finite_or_zero(row.get("Bottom1Price"))), 1e-12)
    height_pct = pattern_height / bottom1_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 80.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    bottom_diff_penalty = _finite_or_zero(row.get("BottomDifferencePct"))
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("Bottom2BarIdx")),
        0.0,
    )

    return (
        height_pct * 2.5
        + width_score
        + downtrend_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - bottom_diff_penalty * 2.0
        - entry_delay * 0.02
    )


def _select_best_inverse_head_shoulders_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforePatternBarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforePatternBarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "PatternHeight",
                "LeftShoulderBarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_non_overlapping_inverse_head_shoulders(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["LeftShoulderBarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _interval_overlap_pct(
    first: tuple[int, int],
    second: tuple[int, int],
) -> float:
    first_start, first_end = first
    second_start, second_end = second
    overlap = max(0, min(first_end, second_end) - max(first_start, second_start))
    if overlap <= 0:
        return 0.0

    first_width = max(first_end - first_start, 1)
    second_width = max(second_end - second_start, 1)
    return overlap / min(first_width, second_width)


def _score_inverse_head_shoulders_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    head_price = max(abs(_finite_or_zero(row.get("HeadPrice"))), 1e-12)
    height_pct = pattern_height / head_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 100.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    head_depth_score = _finite_or_zero(row.get("HeadDepthPct")) / 5.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    shoulder_penalty = _finite_or_zero(row.get("ShoulderDifferencePct")) / 2.0
    neckline_penalty = _finite_or_zero(row.get("NecklineTiltPct")) / 4.0
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("RightShoulderBarIdx")),
        0.0,
    )

    return (
        height_pct * 2.0
        + width_score
        + downtrend_score
        + head_depth_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - shoulder_penalty
        - neckline_penalty
        - entry_delay * 0.02
    )


def _finite_or_zero(value: Any, *, default: float = 0.0) -> float:
    try:
        number = float(value)
    except (TypeError, ValueError):
        return default
    if not np.isfinite(number):
        return default
    return number


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _find_last_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> pd.Series | None:
    if structure_events is None or structure_events.empty:
        return None

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    if subset.empty:
        return None

    return subset.sort_values("EventBarIdx").iloc[-1]


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if price > support_zone_top:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
    resistance_zone_bottom: float | None = None,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if (
            current_idx > entry_idx
            and resistance_zone_bottom is not None
            and close_price < resistance_zone_bottom
        ):
            return None
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _line_value(
    start_idx: int,
    start_price: float,
    end_idx: int,
    end_price: float,
    target_idx: int,
) -> float:
    if int(end_idx) == int(start_idx):
        return float(end_price)

    slope = (float(end_price) - float(start_price)) / (int(end_idx) - int(start_idx))
    return float(start_price) + slope * (int(target_idx) - int(start_idx))


def _find_inverse_head_shoulders_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    neckline_start_idx: int,
    neckline_start_price: float,
    neckline_end_idx: int,
    neckline_end_price: float,
    neckline_zone_pct: float,
    require_hold_neckline: bool,
    fail_level: float,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx

        if current_idx <= entry_idx:
            continue

        neckline_price = _line_value(
            neckline_start_idx,
            neckline_start_price,
            neckline_end_idx,
            neckline_end_price,
            current_idx,
        )
        neckline_bottom = neckline_price * (1.0 - neckline_zone_pct / 100.0)
        if require_hold_neckline and close_price < neckline_bottom:
            return None
        if close_price < fail_level:
            return None

    return None


def _find_double_bottom_start(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    resistance_level: float,
    fallback_idx: int,
    fallback_price: float,
    lookback_bars: int,
) -> dict[str, Any]:
    search_start = max(0, int(b1_idx) - int(lookback_bars))
    closes = data["close"].to_numpy(dtype=float)
    highs = data["high"].to_numpy(dtype=float)

    for current_idx in range(int(b1_idx) - 1, search_start - 1, -1):
        if closes[current_idx] >= resistance_level or highs[current_idx] >= resistance_level:
            return {
                "BarIdx": current_idx,
                "Date": data.index[current_idx],
                "Price": float(closes[current_idx]),
            }

    return {
        "BarIdx": int(fallback_idx),
        "Date": data.index[int(fallback_idx)],
        "Price": float(fallback_price),
    }


def _compute_double_bottom_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    b1_idx: int,
    b1_price: float,
    resistance_idx: int,
    resistance_price: float,
    b2_idx: int,
    b2_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, b1_idx, resistance_idx, b2_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [start_price, b1_price, resistance_price, b2_price, entry_price],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _compute_inverse_head_shoulders_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    left_idx: int,
    left_price: float,
    neck1_idx: int,
    neck1_price: float,
    head_idx: int,
    head_price: float,
    neck2_idx: int,
    neck2_price: float,
    right_idx: int,
    right_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, left_idx, neck1_idx, head_idx, neck2_idx, right_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [
            start_price,
            left_price,
            neck1_price,
            head_price,
            neck2_price,
            right_price,
            entry_price,
        ],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
    inverse_head_shoulders: pd.DataFrame | None = None,
    show_inverse_head_shoulders: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for pattern_number, (_, pattern) in enumerate(double_bottoms.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")
            point_indexes.extend(
                [
                    int(pattern["Bottom1BarIdx"]),
                    int(pattern["ResistanceBarIdx"]),
                    int(pattern["Bottom2BarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["Bottom1Price"]),
                    float(pattern["ResistancePrice"]),
                    float(pattern["Bottom2Price"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["B1", "Resistance", "B2", "Breakout"])
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    showlegend=pattern_number == 0,
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforeB1Direction', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    if (
        show_inverse_head_shoulders
        and inverse_head_shoulders is not None
        and not inverse_head_shoulders.empty
    ):
        ihs_color = "#38BDF8"
        neckline_color = "#E0F2FE"

        for pattern_number, (_, pattern) in enumerate(inverse_head_shoulders.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in inverse_head_shoulders.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")

            point_indexes.extend(
                [
                    int(pattern["LeftShoulderBarIdx"]),
                    int(pattern["Neckline1BarIdx"]),
                    int(pattern["HeadBarIdx"]),
                    int(pattern["Neckline2BarIdx"]),
                    int(pattern["RightShoulderBarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["LeftShoulderPrice"]),
                    float(pattern["Neckline1Price"]),
                    float(pattern["HeadPrice"]),
                    float(pattern["Neckline2Price"]),
                    float(pattern["RightShoulderPrice"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["LS", "N1", "Head", "N2", "RS", "Breakout"])

            if (
                "FollowThroughBarIdx" in inverse_head_shoulders.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")

            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [data.index[idx] for idx in point_indexes]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Inverse H&S",
                    showlegend=pattern_number == 0,
                    line=dict(color=ihs_color, width=4),
                    marker=dict(
                        color=ihs_color,
                        size=10,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color=ihs_color, size=11),
                    hovertext=(
                        f"Inverse H&S #{int(pattern['PatternId'])}<br>"
                        f"Shoulder diff: {float(pattern['ShoulderDifferencePct']):.2f}%<br>"
                        f"Head depth: {float(pattern['HeadDepthPct']):.2f}%<br>"
                        f"Neckline tilt: {float(pattern['NecklineTiltPct']):.2f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforePatternDirection', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["LeftShoulderBarIdx"])
                if remove_time_gaps
                else pattern["LeftShoulderDate"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "HeadSupportZoneBottom",
                "HeadSupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(inverse_head_shoulders.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["HeadSupportZoneBottom"]),
                    y1=float(pattern["HeadSupportZoneTop"]),
                    fillcolor="rgba(56,189,248,0.14)",
                    line=dict(color=ihs_color, width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(224,242,254,0.12)",
                    line=dict(color=neckline_color, width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["Neckline1BarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["Neckline1Date"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )
            for name, price, color, dash in [
                (
                    "IHS Resistance Break",
                    float(pattern["ResistanceZoneTop"]),
                    neckline_color,
                    "solid",
                ),
                ("IHS Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("IHS Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "IHS Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "AAPL"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_breakout_hold_resistance=True,
        one_pattern_per_bearish_choch=True,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )
    inverse_head_shoulders_df = detect_inverse_head_shoulders(
        raw,
        swings_df,
        structure_events=structure_df,
        shoulder_tolerance_pct=3.0,
        min_head_depth_pct=0.75,
        head_lowest_tolerance_pct=0.15,
        head_support_zone_pct=0.25,
        neckline_zone_pct=0.25,
        max_neckline_tilt_pct=1.0,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        neckline_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_last_choch_bearish=True,
        require_head_lowest=True,
        require_bullish_structure_break=True,
        require_direct_breakout_after_right_shoulder=True,
        require_breakout_hold_neckline=True,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))
    print(inverse_head_shoulders_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
        inverse_head_shoulders=inverse_head_shoulders_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    max_symmetry_ratio: float = 2.5,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_resistance: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "LastCHoCHBeforeB1BarIdx",
        "LastCHoCHBeforeB1Date",
        "LastCHoCHBeforeB1Direction",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch_before_b1 = _find_last_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        bearish_choch_before_b1 = (
            last_choch_before_b1 is not None
            and str(last_choch_before_b1["Direction"]) == "Bearish"
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        resistance_candidates = ordered_swings[
            (ordered_swings["SwingType"] == "High")
            & (ordered_swings["BarIdx"].astype(int) > b1_idx)
            & (
                ordered_swings["BarIdx"].astype(int)
                < b1_idx + max_bars_between_bottoms
            )
        ]

        for _, resistance in resistance_candidates.iterrows():
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= b1_price:
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            earliest_breakout_idx = max(
                int(resistance["ConfirmBarIdx"]) + 1,
                resistance_idx + 2,
            )
            latest_breakout_idx = min(
                len(data) - 1,
                earliest_breakout_idx + max_entry_wait_bars,
            )

            for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                if closes[entry_idx] < support_zone_bottom:
                    break

                if closes[entry_idx] <= breakout_level:
                    continue

                right_lows = lows[resistance_idx + 1 : entry_idx]
                if len(right_lows) == 0:
                    continue

                b2_idx = resistance_idx + 1 + int(np.argmin(right_lows))
                b2_price = float(lows[b2_idx])
                bars_between = b2_idx - b1_idx
                if bars_between < min_bars_between_bottoms:
                    continue
                if bars_between > max_bars_between_bottoms:
                    break

                bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
                if bottom_difference_pct > bottom_tolerance_pct:
                    continue
                if not support_zone_bottom <= b2_price <= support_zone_top:
                    continue
                if resistance_price <= max(b1_price, b2_price):
                    continue

                left_width = resistance_idx - b1_idx
                right_width = b2_idx - resistance_idx
                if left_width < 1 or right_width < 1:
                    continue
                symmetry_ratio = max(left_width / right_width, right_width / left_width)
                if symmetry_ratio > max_symmetry_ratio:
                    continue

                left_midpoint = 0.5 * (b1_price + resistance_price)
                right_midpoint = 0.5 * (b2_price + resistance_price)
                if b1_price > right_midpoint or b2_price > left_midpoint:
                    continue

                highs_between = ordered_swings[
                    (ordered_swings["SwingType"] == "High")
                    & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                    & (ordered_swings["BarIdx"].astype(int) < b2_idx)
                ]
                if highs_between.empty:
                    continue
                highest_between = highs_between.sort_values(
                    ["Price", "BarIdx"],
                    ascending=[False, False],
                ).iloc[0]
                if int(highest_between["BarIdx"]) != resistance_idx:
                    continue

                second_peak_invalidation = _has_second_peak_between_zones(
                    ordered_swings,
                    start_idx=resistance_idx,
                    end_idx=b2_idx,
                    support_zone_top=support_zone_top,
                    resistance_zone_bottom=resistance_zone_bottom,
                )
                if invalidate_second_peak_between_zones and second_peak_invalidation:
                    continue

                support_fail_level = min(b1_price, b2_price) * (
                    1.0 - fail_buffer_pct / 100.0
                )
                if np.min(closes[resistance_idx + 1 : entry_idx + 1]) < support_fail_level:
                    continue

                post_b2_swing_invalidation = _has_any_swing_between(
                    ordered_swings,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                    continue

                bullish_structure_break = _has_bullish_structure_break(
                    structure_events,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_bullish_structure_break and not bullish_structure_break:
                    continue

                pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
                target_price = resistance_zone_top + pattern_height
                follow_through_level = resistance_zone_top + (
                    pattern_height * follow_through_pattern_ratio
                )
                follow_through_idx = _find_breakout_follow_through(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                    follow_through_level=follow_through_level,
                    support_zone_bottom=support_zone_bottom,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                )
                if require_breakout_follow_through and follow_through_idx is None:
                    continue

                pattern_start = _find_double_bottom_start(
                    data,
                    b1_idx=b1_idx,
                    resistance_level=resistance_zone_bottom,
                    fallback_idx=context_idx,
                    fallback_price=context_price,
                    lookback_bars=downtrend_lookback_bars,
                )
                pattern_r2 = _compute_double_bottom_r2(
                    closes=closes,
                    start_idx=int(pattern_start["BarIdx"]),
                    start_price=float(pattern_start["Price"]),
                    b1_idx=b1_idx,
                    b1_price=b1_price,
                    resistance_idx=resistance_idx,
                    resistance_price=resistance_price,
                    b2_idx=b2_idx,
                    b2_price=b2_price,
                    entry_idx=entry_idx,
                    entry_price=float(closes[entry_idx]),
                )
                if (
                    min_pattern_r2 is not None
                    and (
                        not np.isfinite(pattern_r2)
                        or float(pattern_r2) < float(min_pattern_r2)
                    )
                ):
                    continue

                entry_volume_ratio = (
                    float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                )
                volume_confirmed = (
                    bool(
                        np.isfinite(entry_volume_ratio)
                        and entry_volume_ratio >= volume_multiplier
                    )
                    if has_volume
                    else False
                )
                if require_volume_confirmation and not volume_confirmed:
                    continue

                pattern_key = (b1_idx, resistance_idx, b2_idx)
                if pattern_key in seen_patterns:
                    continue
                seen_patterns.add(pattern_key)

                stop_price = min(b1_price, b2_price)
                risk = closes[entry_idx] - stop_price
                reward = target_price - closes[entry_idx]

                patterns.append(
                    {
                        "PatternId": len(patterns) + 1,
                        "ContextHighBarIdx": context_idx,
                        "ContextHighDate": context["Date"],
                        "ContextHighPrice": context_price,
                        "PreviousHighBarIdx": previous_high_idx,
                        "PreviousHighDate": previous_high["Date"],
                        "PreviousHighPrice": previous_high_price,
                        "LastCHoCHBeforeB1BarIdx": int(
                            last_choch_before_b1["EventBarIdx"]
                        )
                        if last_choch_before_b1 is not None
                        else np.nan,
                        "LastCHoCHBeforeB1Date": last_choch_before_b1["EventDate"]
                        if last_choch_before_b1 is not None
                        else pd.NaT,
                        "LastCHoCHBeforeB1Direction": str(
                            last_choch_before_b1["Direction"]
                        )
                        if last_choch_before_b1 is not None
                        else None,
                        "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                        "PatternStartDate": pattern_start["Date"],
                        "PatternStartPrice": float(pattern_start["Price"]),
                        "Bottom1BarIdx": b1_idx,
                        "Bottom1Date": bottom1["Date"],
                        "Bottom1Price": b1_price,
                        "ResistanceBarIdx": resistance_idx,
                        "ResistanceDate": resistance["Date"],
                        "ResistancePrice": resistance_price,
                        "Bottom2BarIdx": b2_idx,
                        "Bottom2Date": data.index[b2_idx],
                        "Bottom2Price": b2_price,
                        "EntryBarIdx": entry_idx,
                        "EntryDate": data.index[entry_idx],
                        "EntryPrice": float(closes[entry_idx]),
                        "FollowThroughBarIdx": int(follow_through_idx)
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughDate": data.index[int(follow_through_idx)]
                        if follow_through_idx is not None
                        else pd.NaT,
                        "FollowThroughPrice": float(closes[int(follow_through_idx)])
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughLevel": float(follow_through_level),
                        "TargetPrice": float(target_price),
                        "StopPrice": float(stop_price),
                        "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                        "BottomDifferencePct": float(bottom_difference_pct),
                        "SupportZoneBottom": float(support_zone_bottom),
                        "SupportZoneTop": float(support_zone_top),
                        "ResistanceZoneBottom": float(resistance_zone_bottom),
                        "ResistanceZoneTop": float(resistance_zone_top),
                        "RetracePct": float(retrace_pct),
                        "DowntrendPct": float(downtrend_pct),
                        "PatternWidth": int(b2_idx - b1_idx),
                        "PatternHeight": float(pattern_height),
                        "SymmetryRatio": float(symmetry_ratio),
                        "PatternR2": float(pattern_r2),
                        "PriorLHLL": bool(prior_lh_ll),
                        "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                        "SecondPeakInvalidation": bool(second_peak_invalidation),
                        "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                        "BullishStructureBreak": bool(bullish_structure_break),
                        "VolumeRatio": entry_volume_ratio,
                        "VolumeConfirmed": bool(volume_confirmed),
                    }
                )
                break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_pattern_per_choch(result, columns=columns)
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def detect_inverse_head_shoulders(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    shoulder_tolerance_pct: float = 3.0,
    min_head_depth_pct: float = 0.75,
    head_lowest_tolerance_pct: float = 0.15,
    head_support_zone_pct: float = 0.25,
    neckline_zone_pct: float = 0.25,
    max_neckline_tilt_pct: float = 1.0,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_points: int = 3,
    max_pattern_bars: int = 260,
    max_symmetry_ratio: float = 2.5,
    neckline_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_prior_lh_ll: bool = True,
    require_last_choch_bearish: bool = True,
    require_alternating_extrema_sequence: bool = True,
    require_head_lowest: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_bearish_structure_after_right_shoulder: bool = True,
    require_direct_breakout_after_right_shoulder: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_neckline: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed inverse head and shoulders patterns.

    Sequence:
    - prior bearish CHoCH/downtrend
    - left shoulder low
    - neckline high 1
    - deeper head low
    - neckline high 2
    - right shoulder low
    - close above the resistance zone
    """
    if shoulder_tolerance_pct < 0:
        raise ValueError("shoulder_tolerance_pct cannot be negative.")
    if min_head_depth_pct < 0:
        raise ValueError("min_head_depth_pct cannot be negative.")
    if head_lowest_tolerance_pct < 0:
        raise ValueError("head_lowest_tolerance_pct cannot be negative.")
    if head_support_zone_pct < 0:
        raise ValueError("head_support_zone_pct cannot be negative.")
    if neckline_zone_pct < 0:
        raise ValueError("neckline_zone_pct cannot be negative.")
    if max_neckline_tilt_pct < 0:
        raise ValueError("max_neckline_tilt_pct cannot be negative.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_points < 1:
        raise ValueError("min_bars_between_points must be at least 1.")
    if max_pattern_bars < min_bars_between_points * 4:
        raise ValueError("max_pattern_bars is too small for an IHS pattern.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if neckline_break_buffer_pct < 0:
        raise ValueError("neckline_break_buffer_pct cannot be negative.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "LastCHoCHBeforePatternBarIdx",
        "LastCHoCHBeforePatternDate",
        "LastCHoCHBeforePatternDirection",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "LeftShoulderBarIdx",
        "LeftShoulderDate",
        "LeftShoulderPrice",
        "Neckline1BarIdx",
        "Neckline1Date",
        "Neckline1Price",
        "HeadBarIdx",
        "HeadDate",
        "HeadPrice",
        "Neckline2BarIdx",
        "Neckline2Date",
        "Neckline2Price",
        "RightShoulderBarIdx",
        "RightShoulderDate",
        "RightShoulderPrice",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "NecklineEntryPrice",
        "ResistancePrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "ShoulderDifferencePct",
        "HeadDepthPct",
        "NecklineTiltPct",
        "HeadSupportZoneBottom",
        "HeadSupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "NecklineZoneBottom",
        "NecklineZoneTop",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "LastCHoCHBearish",
        "AlternatingExtremaSequence",
        "HeadLowestInvalidation",
        "BearishStructureAfterRightShoulderInvalidation",
        "PostRightShoulderSwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for left_pos in range(len(lows_only) - 1):
        left_shoulder = lows_only.iloc[left_pos]
        left_idx = int(left_shoulder["BarIdx"])
        left_price = float(left_shoulder["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=left_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch = _find_last_choch_before(
            structure_events,
            before_idx=left_idx,
        )
        last_choch_bearish = (
            last_choch is not None and str(last_choch["Direction"]) == "Bearish"
        )
        if require_last_choch_bearish and not last_choch_bearish:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=left_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=left_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        if context_price <= left_price:
            continue

        head_candidates = lows_only[
            (lows_only["BarIdx"].astype(int) > left_idx + min_bars_between_points)
            & (lows_only["BarIdx"].astype(int) < left_idx + max_pattern_bars)
        ]

        for _, head in head_candidates.iterrows():
            head_idx = int(head["BarIdx"])
            head_price = float(head["Price"])
            if head_price >= left_price * (1.0 - min_head_depth_pct / 100.0):
                continue

            downtrend_pct = (context_price - head_price) / context_price * 100.0
            if downtrend_pct < min_downtrend_pct:
                continue

            leg_lows = ordered_swings[
                (ordered_swings["SwingType"] == "Low")
                & (ordered_swings["BarIdx"].astype(int) >= context_idx)
                & (ordered_swings["BarIdx"].astype(int) <= head_idx)
            ]
            if (
                not leg_lows.empty
                and head_price > float(leg_lows["Price"].min()) * 1.001
            ):
                continue

            left_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > left_idx)
                & (ordered_swings["BarIdx"].astype(int) < head_idx)
            ]
            if left_neck_candidates.empty:
                continue

            neck1 = left_neck_candidates.sort_values(
                ["Price", "BarIdx"],
                ascending=[False, False],
            ).iloc[0]
            neck1_idx = int(neck1["BarIdx"])
            neck1_price = float(neck1["Price"])
            if neck1_price <= left_price or neck1_price <= head_price:
                continue

            right_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > head_idx)
                & (ordered_swings["BarIdx"].astype(int) < left_idx + max_pattern_bars)
            ]
            if right_neck_candidates.empty:
                continue

            for _, neck2 in right_neck_candidates.iterrows():
                neck2_idx = int(neck2["BarIdx"])
                neck2_price = float(neck2["Price"])
                if neck2_idx - head_idx < min_bars_between_points:
                    continue
                if neck2_price <= head_price:
                    continue

                neckline_average = (neck1_price + neck2_price) / 2.0
                neckline_tilt_pct = (
                    abs(neck2_price - neck1_price)
                    / max(abs(neckline_average), 1e-12)
                    * 100.0
                )
                resistance_price = neckline_average
                resistance_zone_bottom = resistance_price * (
                    1.0 - neckline_zone_pct / 100.0
                )
                resistance_zone_top = resistance_price * (
                    1.0 + neckline_zone_pct / 100.0
                )
                if not (
                    resistance_zone_bottom <= neck1_price <= resistance_zone_top
                    and resistance_zone_bottom <= neck2_price <= resistance_zone_top
                ):
                    continue
                if neckline_tilt_pct > max_neckline_tilt_pct:
                    continue

                alternating_extrema_sequence = _is_contiguous_extrema_sequence(
                    ordered_swings,
                    [
                        (left_idx, "Low"),
                        (neck1_idx, "High"),
                        (head_idx, "Low"),
                        (neck2_idx, "High"),
                    ],
                )
                if require_alternating_extrema_sequence and not alternating_extrema_sequence:
                    continue

                head_support_zone_bottom = head_price * (
                    1.0 - head_support_zone_pct / 100.0
                )
                head_support_zone_top = head_price * (
                    1.0 + head_support_zone_pct / 100.0
                )
                earliest_breakout_idx = max(
                    int(neck2["ConfirmBarIdx"]) + 1,
                    neck2_idx + 2,
                )
                latest_breakout_idx = min(
                    len(data) - 1,
                    earliest_breakout_idx + max_entry_wait_bars,
                )
                head_fail_level = head_support_zone_bottom * (
                    1.0 - fail_buffer_pct / 100.0
                )

                for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                    if closes[entry_idx] < head_fail_level:
                        break

                    neckline_entry_price = resistance_price
                    breakout_level = resistance_zone_top * (
                        1.0 + neckline_break_buffer_pct / 100.0
                    )
                    if closes[entry_idx] <= breakout_level:
                        continue

                    right_lows = lows[neck2_idx + 1 : entry_idx]
                    if len(right_lows) == 0:
                        continue

                    right_idx = neck2_idx + 1 + int(np.argmin(right_lows))
                    right_price = float(lows[right_idx])
                    if right_idx - neck2_idx < min_bars_between_points:
                        continue
                    if right_price <= head_price * (1.0 + min_head_depth_pct / 100.0):
                        continue

                    pattern_low = float(np.min(lows[left_idx : entry_idx + 1]))
                    head_lowest_invalidation = pattern_low < head_price * (
                        1.0 - head_lowest_tolerance_pct / 100.0
                    )
                    if require_head_lowest and head_lowest_invalidation:
                        continue

                    if not (
                        head_support_zone_top < left_price < resistance_zone_bottom
                        and head_support_zone_top < right_price < resistance_zone_bottom
                    ):
                        continue

                    shoulder_average = (left_price + right_price) / 2.0
                    shoulder_difference_pct = (
                        abs(right_price - left_price)
                        / max(abs(shoulder_average), 1e-12)
                        * 100.0
                    )
                    if shoulder_difference_pct > shoulder_tolerance_pct:
                        continue

                    head_depth_pct = (
                        (min(left_price, right_price) - head_price)
                        / max(abs(min(left_price, right_price)), 1e-12)
                        * 100.0
                    )
                    if head_depth_pct < min_head_depth_pct:
                        continue

                    if neck2_price <= right_price:
                        continue

                    left_balance_midpoint = (right_price + neck2_price) / 2.0
                    right_balance_midpoint = (left_price + neck1_price) / 2.0
                    if (
                        left_price > left_balance_midpoint
                        or right_price > right_balance_midpoint
                    ):
                        continue

                    left_width = head_idx - left_idx
                    right_width = right_idx - head_idx
                    if left_width < 1 or right_width < 1:
                        continue
                    symmetry_ratio = max(left_width / right_width, right_width / left_width)
                    if symmetry_ratio > max_symmetry_ratio:
                        continue

                    post_right_swing_invalidation = _has_any_swing_between(
                        ordered_swings,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        require_direct_breakout_after_right_shoulder
                        and post_right_swing_invalidation
                    ):
                        continue

                    bearish_structure_after_right_shoulder = _has_bearish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        invalidate_bearish_structure_after_right_shoulder
                        and bearish_structure_after_right_shoulder
                    ):
                        continue

                    bullish_structure_break = _has_bullish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if require_bullish_structure_break and not bullish_structure_break:
                        continue

                    neckline_zone_bottom = resistance_zone_bottom
                    neckline_zone_top = resistance_zone_top
                    pattern_height = resistance_zone_top - head_price
                    target_price = resistance_zone_top + pattern_height
                    follow_through_level = resistance_zone_top + (
                        pattern_height * follow_through_pattern_ratio
                    )
                    follow_through_idx = _find_breakout_follow_through(
                        closes=closes,
                        entry_idx=entry_idx,
                        max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                        follow_through_level=follow_through_level,
                        support_zone_bottom=head_support_zone_bottom,
                        resistance_zone_bottom=resistance_zone_bottom
                        if require_breakout_hold_neckline
                        else None,
                    )
                    if require_breakout_follow_through and follow_through_idx is None:
                        continue

                    pattern_start = _find_double_bottom_start(
                        data,
                        b1_idx=left_idx,
                        resistance_level=neck1_price,
                        fallback_idx=context_idx,
                        fallback_price=context_price,
                        lookback_bars=downtrend_lookback_bars,
                    )
                    pattern_r2 = _compute_inverse_head_shoulders_r2(
                        closes=closes,
                        start_idx=int(pattern_start["BarIdx"]),
                        start_price=float(pattern_start["Price"]),
                        left_idx=left_idx,
                        left_price=left_price,
                        neck1_idx=neck1_idx,
                        neck1_price=neck1_price,
                        head_idx=head_idx,
                        head_price=head_price,
                        neck2_idx=neck2_idx,
                        neck2_price=neck2_price,
                        right_idx=right_idx,
                        right_price=right_price,
                        entry_idx=entry_idx,
                        entry_price=float(closes[entry_idx]),
                    )
                    if (
                        min_pattern_r2 is not None
                        and (
                            not np.isfinite(pattern_r2)
                            or float(pattern_r2) < float(min_pattern_r2)
                        )
                    ):
                        continue

                    entry_volume_ratio = (
                        float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                    )
                    volume_confirmed = (
                        bool(
                            np.isfinite(entry_volume_ratio)
                            and entry_volume_ratio >= volume_multiplier
                        )
                        if has_volume
                        else False
                    )
                    if require_volume_confirmation and not volume_confirmed:
                        continue

                    pattern_key = (left_idx, neck1_idx, head_idx, neck2_idx, right_idx)
                    if pattern_key in seen_patterns:
                        continue
                    seen_patterns.add(pattern_key)

                    stop_price = head_price
                    risk = closes[entry_idx] - stop_price
                    reward = target_price - closes[entry_idx]

                    patterns.append(
                        {
                            "PatternId": len(patterns) + 1,
                            "ContextHighBarIdx": context_idx,
                            "ContextHighDate": context["Date"],
                            "ContextHighPrice": context_price,
                            "LastCHoCHBeforePatternBarIdx": int(last_choch["EventBarIdx"])
                            if last_choch is not None
                            else np.nan,
                            "LastCHoCHBeforePatternDate": last_choch["EventDate"]
                            if last_choch is not None
                            else pd.NaT,
                            "LastCHoCHBeforePatternDirection": str(last_choch["Direction"])
                            if last_choch is not None
                            else None,
                            "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                            "PatternStartDate": pattern_start["Date"],
                            "PatternStartPrice": float(pattern_start["Price"]),
                            "LeftShoulderBarIdx": left_idx,
                            "LeftShoulderDate": left_shoulder["Date"],
                            "LeftShoulderPrice": left_price,
                            "Neckline1BarIdx": neck1_idx,
                            "Neckline1Date": neck1["Date"],
                            "Neckline1Price": neck1_price,
                            "HeadBarIdx": head_idx,
                            "HeadDate": head["Date"],
                            "HeadPrice": head_price,
                            "Neckline2BarIdx": neck2_idx,
                            "Neckline2Date": neck2["Date"],
                            "Neckline2Price": neck2_price,
                            "RightShoulderBarIdx": right_idx,
                            "RightShoulderDate": data.index[right_idx],
                            "RightShoulderPrice": right_price,
                            "EntryBarIdx": entry_idx,
                            "EntryDate": data.index[entry_idx],
                            "EntryPrice": float(closes[entry_idx]),
                            "NecklineEntryPrice": float(neckline_entry_price),
                            "ResistancePrice": float(resistance_price),
                            "FollowThroughBarIdx": int(follow_through_idx)
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughDate": data.index[int(follow_through_idx)]
                            if follow_through_idx is not None
                            else pd.NaT,
                            "FollowThroughPrice": float(closes[int(follow_through_idx)])
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughLevel": float(follow_through_level),
                            "TargetPrice": float(target_price),
                            "StopPrice": float(stop_price),
                            "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                            "ShoulderDifferencePct": float(shoulder_difference_pct),
                            "HeadDepthPct": float(head_depth_pct),
                            "NecklineTiltPct": float(neckline_tilt_pct),
                            "HeadSupportZoneBottom": float(head_support_zone_bottom),
                            "HeadSupportZoneTop": float(head_support_zone_top),
                            "ResistanceZoneBottom": float(resistance_zone_bottom),
                            "ResistanceZoneTop": float(resistance_zone_top),
                            "NecklineZoneBottom": float(neckline_zone_bottom),
                            "NecklineZoneTop": float(neckline_zone_top),
                            "DowntrendPct": float(downtrend_pct),
                            "PatternWidth": int(right_idx - left_idx),
                            "PatternHeight": float(pattern_height),
                            "SymmetryRatio": float(symmetry_ratio),
                            "PatternR2": float(pattern_r2),
                            "PriorLHLL": bool(prior_lh_ll),
                            "LastCHoCHBearish": bool(last_choch_bearish),
                            "AlternatingExtremaSequence": bool(
                                alternating_extrema_sequence
                            ),
                            "HeadLowestInvalidation": bool(head_lowest_invalidation),
                            "BearishStructureAfterRightShoulderInvalidation": bool(
                                bearish_structure_after_right_shoulder
                            ),
                            "PostRightShoulderSwingInvalidation": bool(
                                post_right_swing_invalidation
                            ),
                            "BullishStructureBreak": bool(bullish_structure_break),
                            "VolumeRatio": entry_volume_ratio,
                            "VolumeConfirmed": bool(volume_confirmed),
                        }
                    )
                    break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_inverse_head_shoulders_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_inverse_head_shoulders(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _select_best_pattern_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforeB1BarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforeB1BarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "ResistancePrice",
                "Bottom1BarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _score_double_bottom_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    bottom1_price = max(abs(_finite_or_zero(row.get("Bottom1Price"))), 1e-12)
    height_pct = pattern_height / bottom1_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 80.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    bottom_diff_penalty = _finite_or_zero(row.get("BottomDifferencePct"))
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("Bottom2BarIdx")),
        0.0,
    )

    return (
        height_pct * 2.5
        + width_score
        + downtrend_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - bottom_diff_penalty * 2.0
        - entry_delay * 0.02
    )


def _select_best_inverse_head_shoulders_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforePatternBarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforePatternBarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "PatternHeight",
                "LeftShoulderBarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_non_overlapping_inverse_head_shoulders(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["LeftShoulderBarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _interval_overlap_pct(
    first: tuple[int, int],
    second: tuple[int, int],
) -> float:
    first_start, first_end = first
    second_start, second_end = second
    overlap = max(0, min(first_end, second_end) - max(first_start, second_start))
    if overlap <= 0:
        return 0.0

    first_width = max(first_end - first_start, 1)
    second_width = max(second_end - second_start, 1)
    return overlap / min(first_width, second_width)


def _score_inverse_head_shoulders_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    head_price = max(abs(_finite_or_zero(row.get("HeadPrice"))), 1e-12)
    height_pct = pattern_height / head_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 100.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    head_depth_score = _finite_or_zero(row.get("HeadDepthPct")) / 5.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    shoulder_penalty = _finite_or_zero(row.get("ShoulderDifferencePct")) / 2.0
    neckline_penalty = _finite_or_zero(row.get("NecklineTiltPct")) / 4.0
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("RightShoulderBarIdx")),
        0.0,
    )

    return (
        height_pct * 2.0
        + width_score
        + downtrend_score
        + head_depth_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - shoulder_penalty
        - neckline_penalty
        - entry_delay * 0.02
    )


def _finite_or_zero(value: Any, *, default: float = 0.0) -> float:
    try:
        number = float(value)
    except (TypeError, ValueError):
        return default
    if not np.isfinite(number):
        return default
    return number


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _find_last_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> pd.Series | None:
    if structure_events is None or structure_events.empty:
        return None

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    if subset.empty:
        return None

    return subset.sort_values("EventBarIdx").iloc[-1]


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if price > support_zone_top:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _is_contiguous_extrema_sequence(
    ordered_swings: pd.DataFrame,
    expected: list[tuple[int, str]],
) -> bool:
    if ordered_swings is None or ordered_swings.empty or not expected:
        return False

    first_idx = int(expected[0][0])
    last_idx = int(expected[-1][0])
    window = (
        ordered_swings[
            (ordered_swings["BarIdx"].astype(int) >= first_idx)
            & (ordered_swings["BarIdx"].astype(int) <= last_idx)
        ]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )
    if len(window) != len(expected):
        return False

    for row_position, (expected_idx, expected_type) in enumerate(expected):
        row = window.iloc[row_position]
        if int(row["BarIdx"]) != int(expected_idx):
            return False
        if str(row["SwingType"]) != str(expected_type):
            return False

    return True


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
    resistance_zone_bottom: float | None = None,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if (
            current_idx > entry_idx
            and resistance_zone_bottom is not None
            and close_price < resistance_zone_bottom
        ):
            return None
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _line_value(
    start_idx: int,
    start_price: float,
    end_idx: int,
    end_price: float,
    target_idx: int,
) -> float:
    if int(end_idx) == int(start_idx):
        return float(end_price)

    slope = (float(end_price) - float(start_price)) / (int(end_idx) - int(start_idx))
    return float(start_price) + slope * (int(target_idx) - int(start_idx))


def _find_inverse_head_shoulders_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    neckline_start_idx: int,
    neckline_start_price: float,
    neckline_end_idx: int,
    neckline_end_price: float,
    neckline_zone_pct: float,
    require_hold_neckline: bool,
    fail_level: float,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx

        if current_idx <= entry_idx:
            continue

        neckline_price = _line_value(
            neckline_start_idx,
            neckline_start_price,
            neckline_end_idx,
            neckline_end_price,
            current_idx,
        )
        neckline_bottom = neckline_price * (1.0 - neckline_zone_pct / 100.0)
        if require_hold_neckline and close_price < neckline_bottom:
            return None
        if close_price < fail_level:
            return None

    return None


def _find_double_bottom_start(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    resistance_level: float,
    fallback_idx: int,
    fallback_price: float,
    lookback_bars: int,
) -> dict[str, Any]:
    search_start = max(0, int(b1_idx) - int(lookback_bars))
    closes = data["close"].to_numpy(dtype=float)
    highs = data["high"].to_numpy(dtype=float)

    for current_idx in range(int(b1_idx) - 1, search_start - 1, -1):
        if closes[current_idx] >= resistance_level or highs[current_idx] >= resistance_level:
            return {
                "BarIdx": current_idx,
                "Date": data.index[current_idx],
                "Price": float(closes[current_idx]),
            }

    return {
        "BarIdx": int(fallback_idx),
        "Date": data.index[int(fallback_idx)],
        "Price": float(fallback_price),
    }


def _compute_double_bottom_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    b1_idx: int,
    b1_price: float,
    resistance_idx: int,
    resistance_price: float,
    b2_idx: int,
    b2_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, b1_idx, resistance_idx, b2_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [start_price, b1_price, resistance_price, b2_price, entry_price],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _compute_inverse_head_shoulders_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    left_idx: int,
    left_price: float,
    neck1_idx: int,
    neck1_price: float,
    head_idx: int,
    head_price: float,
    neck2_idx: int,
    neck2_price: float,
    right_idx: int,
    right_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, left_idx, neck1_idx, head_idx, neck2_idx, right_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [
            start_price,
            left_price,
            neck1_price,
            head_price,
            neck2_price,
            right_price,
            entry_price,
        ],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def _has_bearish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) > int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) < int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
    inverse_head_shoulders: pd.DataFrame | None = None,
    show_inverse_head_shoulders: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for pattern_number, (_, pattern) in enumerate(double_bottoms.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")
            point_indexes.extend(
                [
                    int(pattern["Bottom1BarIdx"]),
                    int(pattern["ResistanceBarIdx"]),
                    int(pattern["Bottom2BarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["Bottom1Price"]),
                    float(pattern["ResistancePrice"]),
                    float(pattern["Bottom2Price"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["B1", "Resistance", "B2", "Breakout"])
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    showlegend=pattern_number == 0,
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforeB1Direction', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    if (
        show_inverse_head_shoulders
        and inverse_head_shoulders is not None
        and not inverse_head_shoulders.empty
    ):
        ihs_color = "#38BDF8"
        neckline_color = "#E0F2FE"

        for pattern_number, (_, pattern) in enumerate(inverse_head_shoulders.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []

            point_indexes.extend(
                [
                    int(pattern["LeftShoulderBarIdx"]),
                    int(pattern["Neckline1BarIdx"]),
                    int(pattern["HeadBarIdx"]),
                    int(pattern["Neckline2BarIdx"]),
                    int(pattern["RightShoulderBarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["LeftShoulderPrice"]),
                    float(pattern["Neckline1Price"]),
                    float(pattern["HeadPrice"]),
                    float(pattern["Neckline2Price"]),
                    float(pattern["RightShoulderPrice"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["LS", "N1", "Head", "N2", "RS", "Breakout"])

            if (
                "FollowThroughBarIdx" in inverse_head_shoulders.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")

            point_positions = [
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [data.index[idx] for idx in point_indexes]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Inverse H&S",
                    showlegend=pattern_number == 0,
                    line=dict(color=ihs_color, width=4),
                    marker=dict(
                        color=ihs_color,
                        size=10,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color=ihs_color, size=11),
                    hovertext=(
                        f"Inverse H&S #{int(pattern['PatternId'])}<br>"
                        f"Shoulder diff: {float(pattern['ShoulderDifferencePct']):.2f}%<br>"
                        f"Head depth: {float(pattern['HeadDepthPct']):.2f}%<br>"
                        f"Neckline tilt: {float(pattern['NecklineTiltPct']):.2f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforePatternDirection', 'n/a')}<br>"
                        f"Start: {pattern.get('PatternStartDate', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["LeftShoulderBarIdx"])
                if remove_time_gaps
                else pattern["LeftShoulderDate"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "HeadSupportZoneBottom",
                "HeadSupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(inverse_head_shoulders.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["HeadSupportZoneBottom"]),
                    y1=float(pattern["HeadSupportZoneTop"]),
                    fillcolor="rgba(56,189,248,0.14)",
                    line=dict(color=ihs_color, width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(224,242,254,0.12)",
                    line=dict(color=neckline_color, width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["Neckline1BarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["Neckline1Date"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )
            for name, price, color, dash in [
                (
                    "IHS Resistance Break",
                    float(pattern["ResistanceZoneTop"]),
                    neckline_color,
                    "solid",
                ),
                ("IHS Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("IHS Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "IHS Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "AAPL"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    structure_df = detect_bos_choch(
        raw,
        swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_breakout_hold_resistance=True,
        one_pattern_per_bearish_choch=True,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )
    inverse_head_shoulders_df = detect_inverse_head_shoulders(
        raw,
        swings_df,
        structure_events=structure_df,
        shoulder_tolerance_pct=3.0,
        min_head_depth_pct=0.75,
        head_lowest_tolerance_pct=0.15,
        head_support_zone_pct=0.25,
        neckline_zone_pct=0.25,
        max_neckline_tilt_pct=1.0,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        neckline_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_last_choch_bearish=True,
        require_alternating_extrema_sequence=True,
        require_head_lowest=True,
        require_bullish_structure_break=True,
        invalidate_bearish_structure_after_right_shoulder=True,
        require_direct_breakout_after_right_shoulder=True,
        require_breakout_hold_neckline=True,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )

    print(swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))
    print(inverse_head_shoulders_df.tail(20))

    figure = plot_swing_points(
        raw,
        swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=True,
        double_bottoms=double_bottoms_df,
        inverse_head_shoulders=inverse_head_shoulders_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def build_pattern_swings(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    atr_period: int = 14,
    atr_multiplier: float = 1.25,
    min_move_pct: float = 0.80,
    preserve_structure_swings: bool = True,
    structure_min_move_pct: float | None = None,
) -> pd.DataFrame:
    """
    Convert local swings into a cleaner pattern-detection swing set.

    This works like an object-detection proposal stage: local pivots are the raw
    candidates, then ATR-aware zigzag distance and optional BOS/CHoCH
    participation decide which pivots are worth sending to pattern detection.
    """
    if atr_period < 1:
        raise ValueError("atr_period must be at least 1.")
    if atr_multiplier < 0:
        raise ValueError("atr_multiplier cannot be negative.")
    if min_move_pct < 0:
        raise ValueError("min_move_pct cannot be negative.")
    if structure_min_move_pct is not None and structure_min_move_pct < 0:
        raise ValueError("structure_min_move_pct cannot be negative.")

    base_columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]
    output_columns = base_columns + [
        "AtrPct",
        "RequiredMovePct",
        "MovePct",
        "PatternSwingReason",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=output_columns)

    data = prepare_ohlcv(df)
    atr_pct = _calculate_atr_pct(data, period=atr_period)
    ordered = swings.sort_values("BarIdx").reset_index(drop=True)
    forced_indexes = (
        _structure_swing_indexes(structure_events)
        if preserve_structure_swings and structure_events is not None
        else set()
    )
    structure_floor = (
        float(structure_min_move_pct)
        if structure_min_move_pct is not None
        else float(min_move_pct) * 0.50
    )

    rows: list[dict[str, Any]] = []
    for _, swing in ordered.iterrows():
        row = {column: swing[column] for column in base_columns if column in swing}
        bar_idx = int(row["BarIdx"])
        row["AtrPct"] = float(atr_pct[bar_idx]) if 0 <= bar_idx < len(atr_pct) else np.nan
        atr_required = (
            row["AtrPct"] * float(atr_multiplier)
            if np.isfinite(row["AtrPct"])
            else 0.0
        )
        row["RequiredMovePct"] = max(float(min_move_pct), float(atr_required))
        row["MovePct"] = np.nan
        row["PatternSwingReason"] = ""
        rows.append(row)

    filtered = _filter_pattern_candidate_swings(
        rows,
        forced_indexes=forced_indexes,
        structure_min_move_pct=structure_floor,
    )
    _add_structure_labels(filtered)
    return pd.DataFrame(filtered, columns=output_columns).reset_index(drop=True)


def _calculate_atr_pct(data: pd.DataFrame, *, period: int) -> np.ndarray:
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    previous_close = np.roll(close, 1)
    previous_close[0] = close[0]

    true_range = np.maximum.reduce(
        [
            high - low,
            np.abs(high - previous_close),
            np.abs(low - previous_close),
        ]
    )
    atr = pd.Series(true_range).rolling(period, min_periods=1).mean().to_numpy()
    return atr / np.maximum(np.abs(close), 1e-12) * 100.0


def _structure_swing_indexes(structure_events: pd.DataFrame | None) -> set[int]:
    if structure_events is None or structure_events.empty:
        return set()

    indexes: set[int] = set()
    for column in ["BrokenSwingBarIdx", "ProtectedSwingBarIdx"]:
        if column not in structure_events.columns:
            continue
        indexes.update(structure_events[column].dropna().astype(int).tolist())
    return indexes


def _filter_pattern_candidate_swings(
    swings: list[dict[str, Any]],
    *,
    forced_indexes: set[int],
    structure_min_move_pct: float,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            swing["PatternSwingReason"] = "start"
            selected.append(swing)
            continue

        previous = selected[-1]
        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                if len(selected) >= 2:
                    prior = selected[-2]
                    replacement_move_pct = abs(
                        float(swing["Price"]) - float(prior["Price"])
                    )
                    replacement_move_pct = (
                        replacement_move_pct
                        / max(abs(float(prior["Price"])), 1e-12)
                        * 100.0
                    )
                    swing["MovePct"] = replacement_move_pct
                swing["PatternSwingReason"] = previous.get(
                    "PatternSwingReason",
                    "replacement",
                )
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        swing["MovePct"] = move_pct
        required_move_pct = float(swing.get("RequiredMovePct", 0.0))
        is_structure_swing = int(swing["BarIdx"]) in forced_indexes

        if move_pct >= required_move_pct:
            swing["PatternSwingReason"] = "atr_zigzag"
            selected.append(swing)
            continue

        if is_structure_swing and move_pct >= structure_min_move_pct:
            swing["PatternSwingReason"] = "structure"
            selected.append(swing)

    return selected


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    max_symmetry_ratio: float = 2.5,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_resistance: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "LastCHoCHBeforeB1BarIdx",
        "LastCHoCHBeforeB1Date",
        "LastCHoCHBeforeB1Direction",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch_before_b1 = _find_last_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        bearish_choch_before_b1 = (
            last_choch_before_b1 is not None
            and str(last_choch_before_b1["Direction"]) == "Bearish"
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        resistance_candidates = ordered_swings[
            (ordered_swings["SwingType"] == "High")
            & (ordered_swings["BarIdx"].astype(int) > b1_idx)
            & (
                ordered_swings["BarIdx"].astype(int)
                < b1_idx + max_bars_between_bottoms
            )
        ]

        for _, resistance in resistance_candidates.iterrows():
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= b1_price:
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            earliest_breakout_idx = max(
                int(resistance["ConfirmBarIdx"]) + 1,
                resistance_idx + 2,
            )
            latest_breakout_idx = min(
                len(data) - 1,
                earliest_breakout_idx + max_entry_wait_bars,
            )

            for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                if closes[entry_idx] < support_zone_bottom:
                    break

                if closes[entry_idx] <= breakout_level:
                    continue

                right_lows = lows[resistance_idx + 1 : entry_idx]
                if len(right_lows) == 0:
                    continue

                b2_idx = resistance_idx + 1 + int(np.argmin(right_lows))
                b2_price = float(lows[b2_idx])
                bars_between = b2_idx - b1_idx
                if bars_between < min_bars_between_bottoms:
                    continue
                if bars_between > max_bars_between_bottoms:
                    break

                bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
                if bottom_difference_pct > bottom_tolerance_pct:
                    continue
                if not support_zone_bottom <= b2_price <= support_zone_top:
                    continue
                if resistance_price <= max(b1_price, b2_price):
                    continue

                left_width = resistance_idx - b1_idx
                right_width = b2_idx - resistance_idx
                if left_width < 1 or right_width < 1:
                    continue
                symmetry_ratio = max(left_width / right_width, right_width / left_width)
                if symmetry_ratio > max_symmetry_ratio:
                    continue

                left_midpoint = 0.5 * (b1_price + resistance_price)
                right_midpoint = 0.5 * (b2_price + resistance_price)
                if b1_price > right_midpoint or b2_price > left_midpoint:
                    continue

                highs_between = ordered_swings[
                    (ordered_swings["SwingType"] == "High")
                    & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                    & (ordered_swings["BarIdx"].astype(int) < b2_idx)
                ]
                if highs_between.empty:
                    continue
                highest_between = highs_between.sort_values(
                    ["Price", "BarIdx"],
                    ascending=[False, False],
                ).iloc[0]
                if int(highest_between["BarIdx"]) != resistance_idx:
                    continue

                second_peak_invalidation = _has_second_peak_between_zones(
                    ordered_swings,
                    start_idx=resistance_idx,
                    end_idx=b2_idx,
                    support_zone_top=support_zone_top,
                    resistance_zone_bottom=resistance_zone_bottom,
                )
                if invalidate_second_peak_between_zones and second_peak_invalidation:
                    continue

                support_fail_level = min(b1_price, b2_price) * (
                    1.0 - fail_buffer_pct / 100.0
                )
                if np.min(closes[resistance_idx + 1 : entry_idx + 1]) < support_fail_level:
                    continue

                post_b2_swing_invalidation = _has_any_swing_between(
                    ordered_swings,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                    continue

                bullish_structure_break = _has_bullish_structure_break(
                    structure_events,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_bullish_structure_break and not bullish_structure_break:
                    continue

                pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
                target_price = resistance_zone_top + pattern_height
                follow_through_level = resistance_zone_top + (
                    pattern_height * follow_through_pattern_ratio
                )
                follow_through_idx = _find_breakout_follow_through(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                    follow_through_level=follow_through_level,
                    support_zone_bottom=support_zone_bottom,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                )
                if require_breakout_follow_through and follow_through_idx is None:
                    continue

                pattern_start = _find_double_bottom_start(
                    data,
                    b1_idx=b1_idx,
                    resistance_level=resistance_zone_bottom,
                    fallback_idx=context_idx,
                    fallback_price=context_price,
                    lookback_bars=downtrend_lookback_bars,
                )
                pattern_r2 = _compute_double_bottom_r2(
                    closes=closes,
                    start_idx=int(pattern_start["BarIdx"]),
                    start_price=float(pattern_start["Price"]),
                    b1_idx=b1_idx,
                    b1_price=b1_price,
                    resistance_idx=resistance_idx,
                    resistance_price=resistance_price,
                    b2_idx=b2_idx,
                    b2_price=b2_price,
                    entry_idx=entry_idx,
                    entry_price=float(closes[entry_idx]),
                )
                if (
                    min_pattern_r2 is not None
                    and (
                        not np.isfinite(pattern_r2)
                        or float(pattern_r2) < float(min_pattern_r2)
                    )
                ):
                    continue

                entry_volume_ratio = (
                    float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                )
                volume_confirmed = (
                    bool(
                        np.isfinite(entry_volume_ratio)
                        and entry_volume_ratio >= volume_multiplier
                    )
                    if has_volume
                    else False
                )
                if require_volume_confirmation and not volume_confirmed:
                    continue

                pattern_key = (b1_idx, resistance_idx, b2_idx)
                if pattern_key in seen_patterns:
                    continue
                seen_patterns.add(pattern_key)

                stop_price = min(b1_price, b2_price)
                risk = closes[entry_idx] - stop_price
                reward = target_price - closes[entry_idx]

                patterns.append(
                    {
                        "PatternId": len(patterns) + 1,
                        "ContextHighBarIdx": context_idx,
                        "ContextHighDate": context["Date"],
                        "ContextHighPrice": context_price,
                        "PreviousHighBarIdx": previous_high_idx,
                        "PreviousHighDate": previous_high["Date"],
                        "PreviousHighPrice": previous_high_price,
                        "LastCHoCHBeforeB1BarIdx": int(
                            last_choch_before_b1["EventBarIdx"]
                        )
                        if last_choch_before_b1 is not None
                        else np.nan,
                        "LastCHoCHBeforeB1Date": last_choch_before_b1["EventDate"]
                        if last_choch_before_b1 is not None
                        else pd.NaT,
                        "LastCHoCHBeforeB1Direction": str(
                            last_choch_before_b1["Direction"]
                        )
                        if last_choch_before_b1 is not None
                        else None,
                        "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                        "PatternStartDate": pattern_start["Date"],
                        "PatternStartPrice": float(pattern_start["Price"]),
                        "Bottom1BarIdx": b1_idx,
                        "Bottom1Date": bottom1["Date"],
                        "Bottom1Price": b1_price,
                        "ResistanceBarIdx": resistance_idx,
                        "ResistanceDate": resistance["Date"],
                        "ResistancePrice": resistance_price,
                        "Bottom2BarIdx": b2_idx,
                        "Bottom2Date": data.index[b2_idx],
                        "Bottom2Price": b2_price,
                        "EntryBarIdx": entry_idx,
                        "EntryDate": data.index[entry_idx],
                        "EntryPrice": float(closes[entry_idx]),
                        "FollowThroughBarIdx": int(follow_through_idx)
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughDate": data.index[int(follow_through_idx)]
                        if follow_through_idx is not None
                        else pd.NaT,
                        "FollowThroughPrice": float(closes[int(follow_through_idx)])
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughLevel": float(follow_through_level),
                        "TargetPrice": float(target_price),
                        "StopPrice": float(stop_price),
                        "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                        "BottomDifferencePct": float(bottom_difference_pct),
                        "SupportZoneBottom": float(support_zone_bottom),
                        "SupportZoneTop": float(support_zone_top),
                        "ResistanceZoneBottom": float(resistance_zone_bottom),
                        "ResistanceZoneTop": float(resistance_zone_top),
                        "RetracePct": float(retrace_pct),
                        "DowntrendPct": float(downtrend_pct),
                        "PatternWidth": int(b2_idx - b1_idx),
                        "PatternHeight": float(pattern_height),
                        "SymmetryRatio": float(symmetry_ratio),
                        "PatternR2": float(pattern_r2),
                        "PriorLHLL": bool(prior_lh_ll),
                        "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                        "SecondPeakInvalidation": bool(second_peak_invalidation),
                        "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                        "BullishStructureBreak": bool(bullish_structure_break),
                        "VolumeRatio": entry_volume_ratio,
                        "VolumeConfirmed": bool(volume_confirmed),
                    }
                )
                break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_pattern_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_double_bottoms(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def detect_inverse_head_shoulders(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    shoulder_tolerance_pct: float = 3.0,
    min_head_depth_pct: float = 0.75,
    head_lowest_tolerance_pct: float = 0.15,
    head_support_zone_pct: float = 0.25,
    neckline_zone_pct: float = 0.25,
    max_neckline_tilt_pct: float = 1.0,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_points: int = 3,
    max_pattern_bars: int = 260,
    max_symmetry_ratio: float = 2.5,
    neckline_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_prior_lh_ll: bool = True,
    require_last_choch_bearish: bool = True,
    require_alternating_extrema_sequence: bool = True,
    require_head_lowest: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_bearish_structure_after_right_shoulder: bool = True,
    require_direct_breakout_after_right_shoulder: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_neckline: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed inverse head and shoulders patterns.

    Sequence:
    - prior bearish CHoCH/downtrend
    - left shoulder low
    - neckline high 1
    - deeper head low
    - neckline high 2
    - right shoulder low
    - close above the resistance zone
    """
    if shoulder_tolerance_pct < 0:
        raise ValueError("shoulder_tolerance_pct cannot be negative.")
    if min_head_depth_pct < 0:
        raise ValueError("min_head_depth_pct cannot be negative.")
    if head_lowest_tolerance_pct < 0:
        raise ValueError("head_lowest_tolerance_pct cannot be negative.")
    if head_support_zone_pct < 0:
        raise ValueError("head_support_zone_pct cannot be negative.")
    if neckline_zone_pct < 0:
        raise ValueError("neckline_zone_pct cannot be negative.")
    if max_neckline_tilt_pct < 0:
        raise ValueError("max_neckline_tilt_pct cannot be negative.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_points < 1:
        raise ValueError("min_bars_between_points must be at least 1.")
    if max_pattern_bars < min_bars_between_points * 4:
        raise ValueError("max_pattern_bars is too small for an IHS pattern.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if neckline_break_buffer_pct < 0:
        raise ValueError("neckline_break_buffer_pct cannot be negative.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "LastCHoCHBeforePatternBarIdx",
        "LastCHoCHBeforePatternDate",
        "LastCHoCHBeforePatternDirection",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "LeftShoulderBarIdx",
        "LeftShoulderDate",
        "LeftShoulderPrice",
        "Neckline1BarIdx",
        "Neckline1Date",
        "Neckline1Price",
        "HeadBarIdx",
        "HeadDate",
        "HeadPrice",
        "Neckline2BarIdx",
        "Neckline2Date",
        "Neckline2Price",
        "RightShoulderBarIdx",
        "RightShoulderDate",
        "RightShoulderPrice",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "NecklineEntryPrice",
        "ResistancePrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "ShoulderDifferencePct",
        "HeadDepthPct",
        "NecklineTiltPct",
        "HeadSupportZoneBottom",
        "HeadSupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "NecklineZoneBottom",
        "NecklineZoneTop",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "LastCHoCHBearish",
        "AlternatingExtremaSequence",
        "HeadLowestInvalidation",
        "BearishStructureAfterRightShoulderInvalidation",
        "PostRightShoulderSwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for left_pos in range(len(lows_only) - 1):
        left_shoulder = lows_only.iloc[left_pos]
        left_idx = int(left_shoulder["BarIdx"])
        left_price = float(left_shoulder["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=left_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch = _find_last_choch_before(
            structure_events,
            before_idx=left_idx,
        )
        last_choch_bearish = (
            last_choch is not None and str(last_choch["Direction"]) == "Bearish"
        )
        if require_last_choch_bearish and not last_choch_bearish:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=left_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=left_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        if context_price <= left_price:
            continue

        head_candidates = lows_only[
            (lows_only["BarIdx"].astype(int) > left_idx + min_bars_between_points)
            & (lows_only["BarIdx"].astype(int) < left_idx + max_pattern_bars)
        ]

        for _, head in head_candidates.iterrows():
            head_idx = int(head["BarIdx"])
            head_price = float(head["Price"])
            if head_price >= left_price * (1.0 - min_head_depth_pct / 100.0):
                continue

            downtrend_pct = (context_price - head_price) / context_price * 100.0
            if downtrend_pct < min_downtrend_pct:
                continue

            leg_lows = ordered_swings[
                (ordered_swings["SwingType"] == "Low")
                & (ordered_swings["BarIdx"].astype(int) >= context_idx)
                & (ordered_swings["BarIdx"].astype(int) <= head_idx)
            ]
            if (
                not leg_lows.empty
                and head_price > float(leg_lows["Price"].min()) * 1.001
            ):
                continue

            left_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > left_idx)
                & (ordered_swings["BarIdx"].astype(int) < head_idx)
            ]
            if left_neck_candidates.empty:
                continue

            neck1 = left_neck_candidates.sort_values(
                ["Price", "BarIdx"],
                ascending=[False, False],
            ).iloc[0]
            neck1_idx = int(neck1["BarIdx"])
            neck1_price = float(neck1["Price"])
            if neck1_price <= left_price or neck1_price <= head_price:
                continue

            right_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > head_idx)
                & (ordered_swings["BarIdx"].astype(int) < left_idx + max_pattern_bars)
            ]
            if right_neck_candidates.empty:
                continue

            for _, neck2 in right_neck_candidates.iterrows():
                neck2_idx = int(neck2["BarIdx"])
                neck2_price = float(neck2["Price"])
                if neck2_idx - head_idx < min_bars_between_points:
                    continue
                if neck2_price <= head_price:
                    continue

                neckline_average = (neck1_price + neck2_price) / 2.0
                neckline_tilt_pct = (
                    abs(neck2_price - neck1_price)
                    / max(abs(neckline_average), 1e-12)
                    * 100.0
                )
                resistance_price = neckline_average
                resistance_zone_bottom = resistance_price * (
                    1.0 - neckline_zone_pct / 100.0
                )
                resistance_zone_top = resistance_price * (
                    1.0 + neckline_zone_pct / 100.0
                )
                if not (
                    resistance_zone_bottom <= neck1_price <= resistance_zone_top
                    and resistance_zone_bottom <= neck2_price <= resistance_zone_top
                ):
                    continue
                if neckline_tilt_pct > max_neckline_tilt_pct:
                    continue

                alternating_extrema_sequence = _is_contiguous_extrema_sequence(
                    ordered_swings,
                    [
                        (left_idx, "Low"),
                        (neck1_idx, "High"),
                        (head_idx, "Low"),
                        (neck2_idx, "High"),
                    ],
                )
                if require_alternating_extrema_sequence and not alternating_extrema_sequence:
                    continue

                head_support_zone_bottom = head_price * (
                    1.0 - head_support_zone_pct / 100.0
                )
                head_support_zone_top = head_price * (
                    1.0 + head_support_zone_pct / 100.0
                )
                earliest_breakout_idx = max(
                    int(neck2["ConfirmBarIdx"]) + 1,
                    neck2_idx + 2,
                )
                latest_breakout_idx = min(
                    len(data) - 1,
                    earliest_breakout_idx + max_entry_wait_bars,
                )
                head_fail_level = head_support_zone_bottom * (
                    1.0 - fail_buffer_pct / 100.0
                )

                for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                    if closes[entry_idx] < head_fail_level:
                        break

                    neckline_entry_price = resistance_price
                    breakout_level = resistance_zone_top * (
                        1.0 + neckline_break_buffer_pct / 100.0
                    )
                    if closes[entry_idx] <= breakout_level:
                        continue

                    right_lows = lows[neck2_idx + 1 : entry_idx]
                    if len(right_lows) == 0:
                        continue

                    right_idx = neck2_idx + 1 + int(np.argmin(right_lows))
                    right_price = float(lows[right_idx])
                    if right_idx - neck2_idx < min_bars_between_points:
                        continue
                    if right_price <= head_price * (1.0 + min_head_depth_pct / 100.0):
                        continue

                    pattern_low = float(np.min(lows[left_idx : entry_idx + 1]))
                    head_lowest_invalidation = pattern_low < head_price * (
                        1.0 - head_lowest_tolerance_pct / 100.0
                    )
                    if require_head_lowest and head_lowest_invalidation:
                        continue

                    if not (
                        head_support_zone_top < left_price < resistance_zone_bottom
                        and head_support_zone_top < right_price < resistance_zone_bottom
                    ):
                        continue

                    shoulder_average = (left_price + right_price) / 2.0
                    shoulder_difference_pct = (
                        abs(right_price - left_price)
                        / max(abs(shoulder_average), 1e-12)
                        * 100.0
                    )
                    if shoulder_difference_pct > shoulder_tolerance_pct:
                        continue

                    head_depth_pct = (
                        (min(left_price, right_price) - head_price)
                        / max(abs(min(left_price, right_price)), 1e-12)
                        * 100.0
                    )
                    if head_depth_pct < min_head_depth_pct:
                        continue

                    if neck2_price <= right_price:
                        continue

                    left_balance_midpoint = (right_price + neck2_price) / 2.0
                    right_balance_midpoint = (left_price + neck1_price) / 2.0
                    if (
                        left_price > left_balance_midpoint
                        or right_price > right_balance_midpoint
                    ):
                        continue

                    left_width = head_idx - left_idx
                    right_width = right_idx - head_idx
                    if left_width < 1 or right_width < 1:
                        continue
                    symmetry_ratio = max(left_width / right_width, right_width / left_width)
                    if symmetry_ratio > max_symmetry_ratio:
                        continue

                    post_right_swing_invalidation = _has_any_swing_between(
                        ordered_swings,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        require_direct_breakout_after_right_shoulder
                        and post_right_swing_invalidation
                    ):
                        continue

                    bearish_structure_after_right_shoulder = _has_bearish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        invalidate_bearish_structure_after_right_shoulder
                        and bearish_structure_after_right_shoulder
                    ):
                        continue

                    bullish_structure_break = _has_bullish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if require_bullish_structure_break and not bullish_structure_break:
                        continue

                    neckline_zone_bottom = resistance_zone_bottom
                    neckline_zone_top = resistance_zone_top
                    pattern_height = resistance_zone_top - head_price
                    target_price = resistance_zone_top + pattern_height
                    follow_through_level = resistance_zone_top + (
                        pattern_height * follow_through_pattern_ratio
                    )
                    follow_through_idx = _find_breakout_follow_through(
                        closes=closes,
                        entry_idx=entry_idx,
                        max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                        follow_through_level=follow_through_level,
                        support_zone_bottom=head_support_zone_bottom,
                        resistance_zone_bottom=resistance_zone_bottom
                        if require_breakout_hold_neckline
                        else None,
                    )
                    if require_breakout_follow_through and follow_through_idx is None:
                        continue

                    pattern_start = _find_double_bottom_start(
                        data,
                        b1_idx=left_idx,
                        resistance_level=neck1_price,
                        fallback_idx=context_idx,
                        fallback_price=context_price,
                        lookback_bars=downtrend_lookback_bars,
                    )
                    pattern_r2 = _compute_inverse_head_shoulders_r2(
                        closes=closes,
                        start_idx=int(pattern_start["BarIdx"]),
                        start_price=float(pattern_start["Price"]),
                        left_idx=left_idx,
                        left_price=left_price,
                        neck1_idx=neck1_idx,
                        neck1_price=neck1_price,
                        head_idx=head_idx,
                        head_price=head_price,
                        neck2_idx=neck2_idx,
                        neck2_price=neck2_price,
                        right_idx=right_idx,
                        right_price=right_price,
                        entry_idx=entry_idx,
                        entry_price=float(closes[entry_idx]),
                    )
                    if (
                        min_pattern_r2 is not None
                        and (
                            not np.isfinite(pattern_r2)
                            or float(pattern_r2) < float(min_pattern_r2)
                        )
                    ):
                        continue

                    entry_volume_ratio = (
                        float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                    )
                    volume_confirmed = (
                        bool(
                            np.isfinite(entry_volume_ratio)
                            and entry_volume_ratio >= volume_multiplier
                        )
                        if has_volume
                        else False
                    )
                    if require_volume_confirmation and not volume_confirmed:
                        continue

                    pattern_key = (left_idx, neck1_idx, head_idx, neck2_idx, right_idx)
                    if pattern_key in seen_patterns:
                        continue
                    seen_patterns.add(pattern_key)

                    stop_price = head_price
                    risk = closes[entry_idx] - stop_price
                    reward = target_price - closes[entry_idx]

                    patterns.append(
                        {
                            "PatternId": len(patterns) + 1,
                            "ContextHighBarIdx": context_idx,
                            "ContextHighDate": context["Date"],
                            "ContextHighPrice": context_price,
                            "LastCHoCHBeforePatternBarIdx": int(last_choch["EventBarIdx"])
                            if last_choch is not None
                            else np.nan,
                            "LastCHoCHBeforePatternDate": last_choch["EventDate"]
                            if last_choch is not None
                            else pd.NaT,
                            "LastCHoCHBeforePatternDirection": str(last_choch["Direction"])
                            if last_choch is not None
                            else None,
                            "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                            "PatternStartDate": pattern_start["Date"],
                            "PatternStartPrice": float(pattern_start["Price"]),
                            "LeftShoulderBarIdx": left_idx,
                            "LeftShoulderDate": left_shoulder["Date"],
                            "LeftShoulderPrice": left_price,
                            "Neckline1BarIdx": neck1_idx,
                            "Neckline1Date": neck1["Date"],
                            "Neckline1Price": neck1_price,
                            "HeadBarIdx": head_idx,
                            "HeadDate": head["Date"],
                            "HeadPrice": head_price,
                            "Neckline2BarIdx": neck2_idx,
                            "Neckline2Date": neck2["Date"],
                            "Neckline2Price": neck2_price,
                            "RightShoulderBarIdx": right_idx,
                            "RightShoulderDate": data.index[right_idx],
                            "RightShoulderPrice": right_price,
                            "EntryBarIdx": entry_idx,
                            "EntryDate": data.index[entry_idx],
                            "EntryPrice": float(closes[entry_idx]),
                            "NecklineEntryPrice": float(neckline_entry_price),
                            "ResistancePrice": float(resistance_price),
                            "FollowThroughBarIdx": int(follow_through_idx)
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughDate": data.index[int(follow_through_idx)]
                            if follow_through_idx is not None
                            else pd.NaT,
                            "FollowThroughPrice": float(closes[int(follow_through_idx)])
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughLevel": float(follow_through_level),
                            "TargetPrice": float(target_price),
                            "StopPrice": float(stop_price),
                            "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                            "ShoulderDifferencePct": float(shoulder_difference_pct),
                            "HeadDepthPct": float(head_depth_pct),
                            "NecklineTiltPct": float(neckline_tilt_pct),
                            "HeadSupportZoneBottom": float(head_support_zone_bottom),
                            "HeadSupportZoneTop": float(head_support_zone_top),
                            "ResistanceZoneBottom": float(resistance_zone_bottom),
                            "ResistanceZoneTop": float(resistance_zone_top),
                            "NecklineZoneBottom": float(neckline_zone_bottom),
                            "NecklineZoneTop": float(neckline_zone_top),
                            "DowntrendPct": float(downtrend_pct),
                            "PatternWidth": int(right_idx - left_idx),
                            "PatternHeight": float(pattern_height),
                            "SymmetryRatio": float(symmetry_ratio),
                            "PatternR2": float(pattern_r2),
                            "PriorLHLL": bool(prior_lh_ll),
                            "LastCHoCHBearish": bool(last_choch_bearish),
                            "AlternatingExtremaSequence": bool(
                                alternating_extrema_sequence
                            ),
                            "HeadLowestInvalidation": bool(head_lowest_invalidation),
                            "BearishStructureAfterRightShoulderInvalidation": bool(
                                bearish_structure_after_right_shoulder
                            ),
                            "PostRightShoulderSwingInvalidation": bool(
                                post_right_swing_invalidation
                            ),
                            "BullishStructureBreak": bool(bullish_structure_break),
                            "VolumeRatio": entry_volume_ratio,
                            "VolumeConfirmed": bool(volume_confirmed),
                        }
                    )
                    break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_inverse_head_shoulders_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_inverse_head_shoulders(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _select_best_pattern_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforeB1BarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforeB1BarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "ResistancePrice",
                "Bottom1BarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _score_double_bottom_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    bottom1_price = max(abs(_finite_or_zero(row.get("Bottom1Price"))), 1e-12)
    height_pct = pattern_height / bottom1_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 80.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    bottom_diff_penalty = _finite_or_zero(row.get("BottomDifferencePct"))
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("Bottom2BarIdx")),
        0.0,
    )

    return (
        height_pct * 2.5
        + width_score
        + downtrend_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - bottom_diff_penalty * 2.0
        - entry_delay * 0.02
    )


def _select_non_overlapping_double_bottoms(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["Bottom1BarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_best_inverse_head_shoulders_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforePatternBarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforePatternBarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "PatternHeight",
                "LeftShoulderBarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_non_overlapping_inverse_head_shoulders(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["LeftShoulderBarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _interval_overlap_pct(
    first: tuple[int, int],
    second: tuple[int, int],
) -> float:
    first_start, first_end = first
    second_start, second_end = second
    overlap = max(0, min(first_end, second_end) - max(first_start, second_start))
    if overlap <= 0:
        return 0.0

    first_width = max(first_end - first_start, 1)
    second_width = max(second_end - second_start, 1)
    return overlap / min(first_width, second_width)


def _score_inverse_head_shoulders_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    head_price = max(abs(_finite_or_zero(row.get("HeadPrice"))), 1e-12)
    height_pct = pattern_height / head_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 100.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    head_depth_score = _finite_or_zero(row.get("HeadDepthPct")) / 5.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    shoulder_penalty = _finite_or_zero(row.get("ShoulderDifferencePct")) / 2.0
    neckline_penalty = _finite_or_zero(row.get("NecklineTiltPct")) / 4.0
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("RightShoulderBarIdx")),
        0.0,
    )

    return (
        height_pct * 2.0
        + width_score
        + downtrend_score
        + head_depth_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - shoulder_penalty
        - neckline_penalty
        - entry_delay * 0.02
    )


def _finite_or_zero(value: Any, *, default: float = 0.0) -> float:
    try:
        number = float(value)
    except (TypeError, ValueError):
        return default
    if not np.isfinite(number):
        return default
    return number


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _find_last_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> pd.Series | None:
    if structure_events is None or structure_events.empty:
        return None

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    if subset.empty:
        return None

    return subset.sort_values("EventBarIdx").iloc[-1]


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if price > support_zone_top:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _is_contiguous_extrema_sequence(
    ordered_swings: pd.DataFrame,
    expected: list[tuple[int, str]],
) -> bool:
    if ordered_swings is None or ordered_swings.empty or not expected:
        return False

    first_idx = int(expected[0][0])
    last_idx = int(expected[-1][0])
    window = (
        ordered_swings[
            (ordered_swings["BarIdx"].astype(int) >= first_idx)
            & (ordered_swings["BarIdx"].astype(int) <= last_idx)
        ]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )
    if len(window) != len(expected):
        return False

    for row_position, (expected_idx, expected_type) in enumerate(expected):
        row = window.iloc[row_position]
        if int(row["BarIdx"]) != int(expected_idx):
            return False
        if str(row["SwingType"]) != str(expected_type):
            return False

    return True


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
    resistance_zone_bottom: float | None = None,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if (
            current_idx > entry_idx
            and resistance_zone_bottom is not None
            and close_price < resistance_zone_bottom
        ):
            return None
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _line_value(
    start_idx: int,
    start_price: float,
    end_idx: int,
    end_price: float,
    target_idx: int,
) -> float:
    if int(end_idx) == int(start_idx):
        return float(end_price)

    slope = (float(end_price) - float(start_price)) / (int(end_idx) - int(start_idx))
    return float(start_price) + slope * (int(target_idx) - int(start_idx))


def _find_inverse_head_shoulders_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    neckline_start_idx: int,
    neckline_start_price: float,
    neckline_end_idx: int,
    neckline_end_price: float,
    neckline_zone_pct: float,
    require_hold_neckline: bool,
    fail_level: float,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx

        if current_idx <= entry_idx:
            continue

        neckline_price = _line_value(
            neckline_start_idx,
            neckline_start_price,
            neckline_end_idx,
            neckline_end_price,
            current_idx,
        )
        neckline_bottom = neckline_price * (1.0 - neckline_zone_pct / 100.0)
        if require_hold_neckline and close_price < neckline_bottom:
            return None
        if close_price < fail_level:
            return None

    return None


def _find_double_bottom_start(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    resistance_level: float,
    fallback_idx: int,
    fallback_price: float,
    lookback_bars: int,
) -> dict[str, Any]:
    search_start = max(0, int(b1_idx) - int(lookback_bars))
    closes = data["close"].to_numpy(dtype=float)
    highs = data["high"].to_numpy(dtype=float)

    for current_idx in range(int(b1_idx) - 1, search_start - 1, -1):
        if closes[current_idx] >= resistance_level or highs[current_idx] >= resistance_level:
            return {
                "BarIdx": current_idx,
                "Date": data.index[current_idx],
                "Price": float(closes[current_idx]),
            }

    return {
        "BarIdx": int(fallback_idx),
        "Date": data.index[int(fallback_idx)],
        "Price": float(fallback_price),
    }


def _compute_double_bottom_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    b1_idx: int,
    b1_price: float,
    resistance_idx: int,
    resistance_price: float,
    b2_idx: int,
    b2_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, b1_idx, resistance_idx, b2_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [start_price, b1_price, resistance_price, b2_price, entry_price],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _compute_inverse_head_shoulders_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    left_idx: int,
    left_price: float,
    neck1_idx: int,
    neck1_price: float,
    head_idx: int,
    head_price: float,
    neck2_idx: int,
    neck2_price: float,
    right_idx: int,
    right_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, left_idx, neck1_idx, head_idx, neck2_idx, right_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [
            start_price,
            left_price,
            neck1_price,
            head_price,
            neck2_price,
            right_price,
            entry_price,
        ],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def _has_bearish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) > int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) < int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def _format_pattern_swing_hover(row: pd.Series) -> str:
    parts: list[str] = []
    if "PatternSwingReason" in row.index and str(row["PatternSwingReason"]):
        parts.append(f"Reason: {row['PatternSwingReason']}")
    if "MovePct" in row.index and np.isfinite(float(row["MovePct"])):
        parts.append(f"Move: {float(row['MovePct']):.2f}%")
    if "RequiredMovePct" in row.index and np.isfinite(float(row["RequiredMovePct"])):
        parts.append(f"Required: {float(row['RequiredMovePct']):.2f}%")
    if "AtrPct" in row.index and np.isfinite(float(row["AtrPct"])):
        parts.append(f"ATR: {float(row['AtrPct']):.2f}%")
    if not parts:
        return ""
    return "<br>" + "<br>".join(parts)


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
    inverse_head_shoulders: pd.DataFrame | None = None,
    show_inverse_head_shoulders: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                        + _format_pattern_swing_hover(row)
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for pattern_number, (_, pattern) in enumerate(double_bottoms.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")
            point_indexes.extend(
                [
                    int(pattern["Bottom1BarIdx"]),
                    int(pattern["ResistanceBarIdx"]),
                    int(pattern["Bottom2BarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["Bottom1Price"]),
                    float(pattern["ResistancePrice"]),
                    float(pattern["Bottom2Price"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["B1", "Resistance", "B2", "Breakout"])
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    showlegend=pattern_number == 0,
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforeB1Direction', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    if (
        show_inverse_head_shoulders
        and inverse_head_shoulders is not None
        and not inverse_head_shoulders.empty
    ):
        ihs_color = "#38BDF8"
        neckline_color = "#E0F2FE"

        for pattern_number, (_, pattern) in enumerate(inverse_head_shoulders.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []

            point_indexes.extend(
                [
                    int(pattern["LeftShoulderBarIdx"]),
                    int(pattern["Neckline1BarIdx"]),
                    int(pattern["HeadBarIdx"]),
                    int(pattern["Neckline2BarIdx"]),
                    int(pattern["RightShoulderBarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["LeftShoulderPrice"]),
                    float(pattern["Neckline1Price"]),
                    float(pattern["HeadPrice"]),
                    float(pattern["Neckline2Price"]),
                    float(pattern["RightShoulderPrice"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["LS", "N1", "Head", "N2", "RS", "Breakout"])

            if (
                "FollowThroughBarIdx" in inverse_head_shoulders.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")

            point_positions = [
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [data.index[idx] for idx in point_indexes]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Inverse H&S",
                    showlegend=pattern_number == 0,
                    line=dict(color=ihs_color, width=4),
                    marker=dict(
                        color=ihs_color,
                        size=10,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color=ihs_color, size=11),
                    hovertext=(
                        f"Inverse H&S #{int(pattern['PatternId'])}<br>"
                        f"Shoulder diff: {float(pattern['ShoulderDifferencePct']):.2f}%<br>"
                        f"Head depth: {float(pattern['HeadDepthPct']):.2f}%<br>"
                        f"Neckline tilt: {float(pattern['NecklineTiltPct']):.2f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforePatternDirection', 'n/a')}<br>"
                        f"Start: {pattern.get('PatternStartDate', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["LeftShoulderBarIdx"])
                if remove_time_gaps
                else pattern["LeftShoulderDate"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "HeadSupportZoneBottom",
                "HeadSupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(inverse_head_shoulders.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["HeadSupportZoneBottom"]),
                    y1=float(pattern["HeadSupportZoneTop"]),
                    fillcolor="rgba(56,189,248,0.14)",
                    line=dict(color=ihs_color, width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(224,242,254,0.12)",
                    line=dict(color=neckline_color, width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["Neckline1BarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["Neckline1Date"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )
            for name, price, color, dash in [
                (
                    "IHS Resistance Break",
                    float(pattern["ResistanceZoneTop"]),
                    neckline_color,
                    "solid",
                ),
                ("IHS Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("IHS Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "IHS Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "MSFT"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    raw_swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    raw_structure_df = detect_bos_choch(
        raw,
        raw_swings_df,
        break_buffer_pct=0.03,
    )
    pattern_swings_df = build_pattern_swings(
        raw,
        raw_swings_df,
        structure_events=raw_structure_df,
        atr_period=14,
        atr_multiplier=1.25,
        min_move_pct=0.80,
        preserve_structure_swings=True,
        structure_min_move_pct=0.40,
    )
    structure_df = detect_bos_choch(
        raw,
        pattern_swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        pattern_swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_breakout_hold_resistance=True,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )
    inverse_head_shoulders_df = detect_inverse_head_shoulders(
        raw,
        pattern_swings_df,
        structure_events=structure_df,
        shoulder_tolerance_pct=3.0,
        min_head_depth_pct=0.75,
        head_lowest_tolerance_pct=0.15,
        head_support_zone_pct=0.25,
        neckline_zone_pct=0.25,
        max_neckline_tilt_pct=1.0,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        neckline_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_last_choch_bearish=True,
        require_alternating_extrema_sequence=True,
        require_head_lowest=True,
        require_bullish_structure_break=True,
        invalidate_bearish_structure_after_right_shoulder=True,
        require_direct_breakout_after_right_shoulder=True,
        require_breakout_hold_neckline=True,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )

    print(f"Raw swings: {len(raw_swings_df)}")
    print(f"Pattern swings: {len(pattern_swings_df)}")
    print(pattern_swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))
    print(inverse_head_shoulders_df.tail(20))

    figure = plot_swing_points(
        raw,
        pattern_swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=False,
        double_bottoms=double_bottoms_df,
        inverse_head_shoulders=inverse_head_shoulders_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def build_pattern_swings(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    atr_period: int = 14,
    atr_multiplier: float = 1.25,
    min_move_pct: float = 0.80,
    preserve_structure_swings: bool = True,
    structure_min_move_pct: float | None = None,
) -> pd.DataFrame:
    """
    Convert local swings into a cleaner pattern-detection swing set.

    This works like an object-detection proposal stage: local pivots are the raw
    candidates, then ATR-aware zigzag distance and optional BOS/CHoCH
    participation decide which pivots are worth sending to pattern detection.
    """
    if atr_period < 1:
        raise ValueError("atr_period must be at least 1.")
    if atr_multiplier < 0:
        raise ValueError("atr_multiplier cannot be negative.")
    if min_move_pct < 0:
        raise ValueError("min_move_pct cannot be negative.")
    if structure_min_move_pct is not None and structure_min_move_pct < 0:
        raise ValueError("structure_min_move_pct cannot be negative.")

    base_columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]
    computed_columns = [
        "AtrPct",
        "RequiredMovePct",
        "MovePct",
        "PatternSwingReason",
    ]
    extra_columns = [
        column
        for column in swings.columns
        if column not in base_columns and column not in computed_columns
    ]
    output_columns = base_columns + extra_columns + computed_columns

    if swings is None or swings.empty:
        return pd.DataFrame(columns=output_columns)

    data = prepare_ohlcv(df)
    atr_pct = _calculate_atr_pct(data, period=atr_period)
    ordered = swings.sort_values("BarIdx").reset_index(drop=True)
    forced_indexes = (
        _structure_swing_indexes(structure_events)
        if preserve_structure_swings and structure_events is not None
        else set()
    )
    structure_floor = (
        float(structure_min_move_pct)
        if structure_min_move_pct is not None
        else float(min_move_pct) * 0.50
    )

    rows: list[dict[str, Any]] = []
    for _, swing in ordered.iterrows():
        row = {column: swing[column] for column in base_columns if column in swing}
        for column in extra_columns:
            row[column] = swing[column]
        bar_idx = int(row["BarIdx"])
        row["AtrPct"] = float(atr_pct[bar_idx]) if 0 <= bar_idx < len(atr_pct) else np.nan
        atr_required = (
            row["AtrPct"] * float(atr_multiplier)
            if np.isfinite(row["AtrPct"])
            else 0.0
        )
        row["RequiredMovePct"] = max(float(min_move_pct), float(atr_required))
        row["MovePct"] = np.nan
        row["PatternSwingReason"] = ""
        rows.append(row)

    filtered = _filter_pattern_candidate_swings(
        rows,
        forced_indexes=forced_indexes,
        structure_min_move_pct=structure_floor,
    )
    _add_structure_labels(filtered)
    return pd.DataFrame(filtered, columns=output_columns).reset_index(drop=True)


def build_breakout_swings(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    break_buffer_pct: float = 0.0,
    include_broken_swings: bool = True,
    include_protected_swings: bool = True,
) -> pd.DataFrame:
    """
    Keep only swings that became real market-structure levels.

    Local pivots still provide raw candidates, but this layer promotes only the
    swings that were actually broken by BOS/CHoCH or acted as protected swing
    points during a break.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")
    if not include_broken_swings and not include_protected_swings:
        raise ValueError("At least one breakout swing source must be enabled.")

    base_columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]
    output_columns = base_columns + [
        "StructureRole",
        "BreakoutBarIdx",
        "BreakoutDate",
        "BreakoutPrice",
        "BreakoutLevel",
        "BreakoutEventType",
        "BreakoutDirection",
        "PatternSwingReason",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=output_columns)

    data = prepare_ohlcv(df)
    ordered = swings.sort_values("BarIdx").reset_index(drop=True)
    if structure_events is None:
        structure_events = detect_bos_choch(
            data,
            ordered,
            break_buffer_pct=break_buffer_pct,
        )
    if structure_events is None or structure_events.empty:
        return pd.DataFrame(columns=output_columns)

    swing_by_idx = {
        int(row["BarIdx"]): row
        for _, row in ordered.iterrows()
        if not pd.isna(row["BarIdx"])
    }
    candidates: dict[tuple[int, str], dict[str, Any]] = {}

    for _, event in structure_events.sort_values("EventBarIdx").iterrows():
        if include_broken_swings:
            _add_breakout_swing_candidate(
                candidates,
                swing_by_idx=swing_by_idx,
                event=event,
                swing_idx=event.get("BrokenSwingBarIdx"),
                role="broken",
                base_columns=base_columns,
            )
        if include_protected_swings:
            _add_breakout_swing_candidate(
                candidates,
                swing_by_idx=swing_by_idx,
                event=event,
                swing_idx=event.get("ProtectedSwingBarIdx"),
                role="protected",
                base_columns=base_columns,
            )

    rows = sorted(candidates.values(), key=lambda row: int(row["BarIdx"]))
    rows = _filter_consecutive_same_type(rows)
    _add_structure_labels(rows)
    return pd.DataFrame(rows, columns=output_columns).reset_index(drop=True)


def _add_breakout_swing_candidate(
    candidates: dict[tuple[int, str], dict[str, Any]],
    *,
    swing_by_idx: dict[int, pd.Series],
    event: pd.Series,
    swing_idx: Any,
    role: str,
    base_columns: list[str],
) -> None:
    if pd.isna(swing_idx):
        return

    source = swing_by_idx.get(int(swing_idx))
    if source is None:
        return

    row = {column: source[column] for column in base_columns}
    row["StructureRole"] = role
    row["BreakoutBarIdx"] = int(event["EventBarIdx"])
    row["BreakoutDate"] = event["EventDate"]
    row["BreakoutPrice"] = float(event["BreakPrice"])
    row["BreakoutLevel"] = float(event["Level"])
    row["BreakoutEventType"] = str(event["EventType"])
    row["BreakoutDirection"] = str(event["Direction"])
    row["PatternSwingReason"] = f"breakout_{role}"

    key = (int(row["BarIdx"]), str(row["SwingType"]))
    if key not in candidates:
        candidates[key] = row
        return

    existing = candidates[key]
    roles = set(str(existing["StructureRole"]).split("+"))
    roles.add(role)
    existing["StructureRole"] = "+".join(sorted(roles))
    existing["PatternSwingReason"] = f"breakout_{existing['StructureRole']}"
    if int(row["BreakoutBarIdx"]) < int(existing["BreakoutBarIdx"]):
        for column in [
            "BreakoutBarIdx",
            "BreakoutDate",
            "BreakoutPrice",
            "BreakoutLevel",
            "BreakoutEventType",
            "BreakoutDirection",
        ]:
            existing[column] = row[column]


def _calculate_atr_pct(data: pd.DataFrame, *, period: int) -> np.ndarray:
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    previous_close = np.roll(close, 1)
    previous_close[0] = close[0]

    true_range = np.maximum.reduce(
        [
            high - low,
            np.abs(high - previous_close),
            np.abs(low - previous_close),
        ]
    )
    atr = pd.Series(true_range).rolling(period, min_periods=1).mean().to_numpy()
    return atr / np.maximum(np.abs(close), 1e-12) * 100.0


def _structure_swing_indexes(structure_events: pd.DataFrame | None) -> set[int]:
    if structure_events is None or structure_events.empty:
        return set()

    indexes: set[int] = set()
    for column in ["BrokenSwingBarIdx", "ProtectedSwingBarIdx"]:
        if column not in structure_events.columns:
            continue
        indexes.update(structure_events[column].dropna().astype(int).tolist())
    return indexes


def _filter_pattern_candidate_swings(
    swings: list[dict[str, Any]],
    *,
    forced_indexes: set[int],
    structure_min_move_pct: float,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            swing["PatternSwingReason"] = "start"
            selected.append(swing)
            continue

        previous = selected[-1]
        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                if len(selected) >= 2:
                    prior = selected[-2]
                    replacement_move_pct = abs(
                        float(swing["Price"]) - float(prior["Price"])
                    )
                    replacement_move_pct = (
                        replacement_move_pct
                        / max(abs(float(prior["Price"])), 1e-12)
                        * 100.0
                    )
                    swing["MovePct"] = replacement_move_pct
                swing["PatternSwingReason"] = previous.get(
                    "PatternSwingReason",
                    "replacement",
                )
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        swing["MovePct"] = move_pct
        required_move_pct = float(swing.get("RequiredMovePct", 0.0))
        is_structure_swing = int(swing["BarIdx"]) in forced_indexes

        if move_pct >= required_move_pct:
            swing["PatternSwingReason"] = "atr_zigzag"
            selected.append(swing)
            continue

        if is_structure_swing and move_pct >= structure_min_move_pct:
            swing["PatternSwingReason"] = "structure"
            selected.append(swing)

    return selected


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    max_symmetry_ratio: float = 2.5,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_resistance: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "LastCHoCHBeforeB1BarIdx",
        "LastCHoCHBeforeB1Date",
        "LastCHoCHBeforeB1Direction",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch_before_b1 = _find_last_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        bearish_choch_before_b1 = (
            last_choch_before_b1 is not None
            and str(last_choch_before_b1["Direction"]) == "Bearish"
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        resistance_candidates = ordered_swings[
            (ordered_swings["SwingType"] == "High")
            & (ordered_swings["BarIdx"].astype(int) > b1_idx)
            & (
                ordered_swings["BarIdx"].astype(int)
                < b1_idx + max_bars_between_bottoms
            )
        ]

        for _, resistance in resistance_candidates.iterrows():
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= b1_price:
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            earliest_breakout_idx = max(
                int(resistance["ConfirmBarIdx"]) + 1,
                resistance_idx + 2,
            )
            latest_breakout_idx = min(
                len(data) - 1,
                earliest_breakout_idx + max_entry_wait_bars,
            )

            for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                if closes[entry_idx] < support_zone_bottom:
                    break

                if closes[entry_idx] <= breakout_level:
                    continue

                right_lows = lows[resistance_idx + 1 : entry_idx]
                if len(right_lows) == 0:
                    continue

                b2_idx = resistance_idx + 1 + int(np.argmin(right_lows))
                b2_price = float(lows[b2_idx])
                bars_between = b2_idx - b1_idx
                if bars_between < min_bars_between_bottoms:
                    continue
                if bars_between > max_bars_between_bottoms:
                    break

                bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
                if bottom_difference_pct > bottom_tolerance_pct:
                    continue
                if not support_zone_bottom <= b2_price <= support_zone_top:
                    continue
                if resistance_price <= max(b1_price, b2_price):
                    continue

                left_width = resistance_idx - b1_idx
                right_width = b2_idx - resistance_idx
                if left_width < 1 or right_width < 1:
                    continue
                symmetry_ratio = max(left_width / right_width, right_width / left_width)
                if symmetry_ratio > max_symmetry_ratio:
                    continue

                left_midpoint = 0.5 * (b1_price + resistance_price)
                right_midpoint = 0.5 * (b2_price + resistance_price)
                if b1_price > right_midpoint or b2_price > left_midpoint:
                    continue

                highs_between = ordered_swings[
                    (ordered_swings["SwingType"] == "High")
                    & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                    & (ordered_swings["BarIdx"].astype(int) < b2_idx)
                ]
                if highs_between.empty:
                    continue
                highest_between = highs_between.sort_values(
                    ["Price", "BarIdx"],
                    ascending=[False, False],
                ).iloc[0]
                if int(highest_between["BarIdx"]) != resistance_idx:
                    continue

                second_peak_invalidation = _has_second_peak_between_zones(
                    ordered_swings,
                    start_idx=resistance_idx,
                    end_idx=b2_idx,
                    support_zone_top=support_zone_top,
                    resistance_zone_bottom=resistance_zone_bottom,
                )
                if invalidate_second_peak_between_zones and second_peak_invalidation:
                    continue

                support_fail_level = min(b1_price, b2_price) * (
                    1.0 - fail_buffer_pct / 100.0
                )
                if np.min(closes[resistance_idx + 1 : entry_idx + 1]) < support_fail_level:
                    continue

                post_b2_swing_invalidation = _has_any_swing_between(
                    ordered_swings,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                    continue

                bullish_structure_break = _has_bullish_structure_break(
                    structure_events,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_bullish_structure_break and not bullish_structure_break:
                    continue

                pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
                target_price = resistance_zone_top + pattern_height
                follow_through_level = resistance_zone_top + (
                    pattern_height * follow_through_pattern_ratio
                )
                follow_through_idx = _find_breakout_follow_through(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                    follow_through_level=follow_through_level,
                    support_zone_bottom=support_zone_bottom,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                )
                if require_breakout_follow_through and follow_through_idx is None:
                    continue

                pattern_start = _find_double_bottom_start(
                    data,
                    b1_idx=b1_idx,
                    resistance_level=resistance_zone_bottom,
                    fallback_idx=context_idx,
                    fallback_price=context_price,
                    lookback_bars=downtrend_lookback_bars,
                )
                pattern_r2 = _compute_double_bottom_r2(
                    closes=closes,
                    start_idx=int(pattern_start["BarIdx"]),
                    start_price=float(pattern_start["Price"]),
                    b1_idx=b1_idx,
                    b1_price=b1_price,
                    resistance_idx=resistance_idx,
                    resistance_price=resistance_price,
                    b2_idx=b2_idx,
                    b2_price=b2_price,
                    entry_idx=entry_idx,
                    entry_price=float(closes[entry_idx]),
                )
                if (
                    min_pattern_r2 is not None
                    and (
                        not np.isfinite(pattern_r2)
                        or float(pattern_r2) < float(min_pattern_r2)
                    )
                ):
                    continue

                entry_volume_ratio = (
                    float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                )
                volume_confirmed = (
                    bool(
                        np.isfinite(entry_volume_ratio)
                        and entry_volume_ratio >= volume_multiplier
                    )
                    if has_volume
                    else False
                )
                if require_volume_confirmation and not volume_confirmed:
                    continue

                pattern_key = (b1_idx, resistance_idx, b2_idx)
                if pattern_key in seen_patterns:
                    continue
                seen_patterns.add(pattern_key)

                stop_price = min(b1_price, b2_price)
                risk = closes[entry_idx] - stop_price
                reward = target_price - closes[entry_idx]

                patterns.append(
                    {
                        "PatternId": len(patterns) + 1,
                        "ContextHighBarIdx": context_idx,
                        "ContextHighDate": context["Date"],
                        "ContextHighPrice": context_price,
                        "PreviousHighBarIdx": previous_high_idx,
                        "PreviousHighDate": previous_high["Date"],
                        "PreviousHighPrice": previous_high_price,
                        "LastCHoCHBeforeB1BarIdx": int(
                            last_choch_before_b1["EventBarIdx"]
                        )
                        if last_choch_before_b1 is not None
                        else np.nan,
                        "LastCHoCHBeforeB1Date": last_choch_before_b1["EventDate"]
                        if last_choch_before_b1 is not None
                        else pd.NaT,
                        "LastCHoCHBeforeB1Direction": str(
                            last_choch_before_b1["Direction"]
                        )
                        if last_choch_before_b1 is not None
                        else None,
                        "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                        "PatternStartDate": pattern_start["Date"],
                        "PatternStartPrice": float(pattern_start["Price"]),
                        "Bottom1BarIdx": b1_idx,
                        "Bottom1Date": bottom1["Date"],
                        "Bottom1Price": b1_price,
                        "ResistanceBarIdx": resistance_idx,
                        "ResistanceDate": resistance["Date"],
                        "ResistancePrice": resistance_price,
                        "Bottom2BarIdx": b2_idx,
                        "Bottom2Date": data.index[b2_idx],
                        "Bottom2Price": b2_price,
                        "EntryBarIdx": entry_idx,
                        "EntryDate": data.index[entry_idx],
                        "EntryPrice": float(closes[entry_idx]),
                        "FollowThroughBarIdx": int(follow_through_idx)
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughDate": data.index[int(follow_through_idx)]
                        if follow_through_idx is not None
                        else pd.NaT,
                        "FollowThroughPrice": float(closes[int(follow_through_idx)])
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughLevel": float(follow_through_level),
                        "TargetPrice": float(target_price),
                        "StopPrice": float(stop_price),
                        "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                        "BottomDifferencePct": float(bottom_difference_pct),
                        "SupportZoneBottom": float(support_zone_bottom),
                        "SupportZoneTop": float(support_zone_top),
                        "ResistanceZoneBottom": float(resistance_zone_bottom),
                        "ResistanceZoneTop": float(resistance_zone_top),
                        "RetracePct": float(retrace_pct),
                        "DowntrendPct": float(downtrend_pct),
                        "PatternWidth": int(b2_idx - b1_idx),
                        "PatternHeight": float(pattern_height),
                        "SymmetryRatio": float(symmetry_ratio),
                        "PatternR2": float(pattern_r2),
                        "PriorLHLL": bool(prior_lh_ll),
                        "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                        "SecondPeakInvalidation": bool(second_peak_invalidation),
                        "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                        "BullishStructureBreak": bool(bullish_structure_break),
                        "VolumeRatio": entry_volume_ratio,
                        "VolumeConfirmed": bool(volume_confirmed),
                    }
                )
                break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_pattern_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_double_bottoms(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def detect_inverse_head_shoulders(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    shoulder_tolerance_pct: float = 3.0,
    min_head_depth_pct: float = 0.75,
    head_lowest_tolerance_pct: float = 0.15,
    head_support_zone_pct: float = 0.25,
    neckline_zone_pct: float = 0.25,
    max_neckline_tilt_pct: float = 1.0,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_points: int = 3,
    max_pattern_bars: int = 260,
    max_symmetry_ratio: float = 2.5,
    neckline_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_prior_lh_ll: bool = True,
    require_last_choch_bearish: bool = True,
    require_alternating_extrema_sequence: bool = True,
    require_head_lowest: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_bearish_structure_after_right_shoulder: bool = True,
    require_direct_breakout_after_right_shoulder: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_neckline: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed inverse head and shoulders patterns.

    Sequence:
    - prior bearish CHoCH/downtrend
    - left shoulder low
    - neckline high 1
    - deeper head low
    - neckline high 2
    - right shoulder low
    - close above the resistance zone
    """
    if shoulder_tolerance_pct < 0:
        raise ValueError("shoulder_tolerance_pct cannot be negative.")
    if min_head_depth_pct < 0:
        raise ValueError("min_head_depth_pct cannot be negative.")
    if head_lowest_tolerance_pct < 0:
        raise ValueError("head_lowest_tolerance_pct cannot be negative.")
    if head_support_zone_pct < 0:
        raise ValueError("head_support_zone_pct cannot be negative.")
    if neckline_zone_pct < 0:
        raise ValueError("neckline_zone_pct cannot be negative.")
    if max_neckline_tilt_pct < 0:
        raise ValueError("max_neckline_tilt_pct cannot be negative.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_points < 1:
        raise ValueError("min_bars_between_points must be at least 1.")
    if max_pattern_bars < min_bars_between_points * 4:
        raise ValueError("max_pattern_bars is too small for an IHS pattern.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if neckline_break_buffer_pct < 0:
        raise ValueError("neckline_break_buffer_pct cannot be negative.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "LastCHoCHBeforePatternBarIdx",
        "LastCHoCHBeforePatternDate",
        "LastCHoCHBeforePatternDirection",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "LeftShoulderBarIdx",
        "LeftShoulderDate",
        "LeftShoulderPrice",
        "Neckline1BarIdx",
        "Neckline1Date",
        "Neckline1Price",
        "HeadBarIdx",
        "HeadDate",
        "HeadPrice",
        "Neckline2BarIdx",
        "Neckline2Date",
        "Neckline2Price",
        "RightShoulderBarIdx",
        "RightShoulderDate",
        "RightShoulderPrice",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "NecklineEntryPrice",
        "ResistancePrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "ShoulderDifferencePct",
        "HeadDepthPct",
        "NecklineTiltPct",
        "HeadSupportZoneBottom",
        "HeadSupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "NecklineZoneBottom",
        "NecklineZoneTop",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "LastCHoCHBearish",
        "AlternatingExtremaSequence",
        "HeadLowestInvalidation",
        "BearishStructureAfterRightShoulderInvalidation",
        "PostRightShoulderSwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for left_pos in range(len(lows_only) - 1):
        left_shoulder = lows_only.iloc[left_pos]
        left_idx = int(left_shoulder["BarIdx"])
        left_price = float(left_shoulder["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=left_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch = _find_last_choch_before(
            structure_events,
            before_idx=left_idx,
        )
        last_choch_bearish = (
            last_choch is not None and str(last_choch["Direction"]) == "Bearish"
        )
        if require_last_choch_bearish and not last_choch_bearish:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=left_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=left_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        if context_price <= left_price:
            continue

        head_candidates = lows_only[
            (lows_only["BarIdx"].astype(int) > left_idx + min_bars_between_points)
            & (lows_only["BarIdx"].astype(int) < left_idx + max_pattern_bars)
        ]

        for _, head in head_candidates.iterrows():
            head_idx = int(head["BarIdx"])
            head_price = float(head["Price"])
            if head_price >= left_price * (1.0 - min_head_depth_pct / 100.0):
                continue

            downtrend_pct = (context_price - head_price) / context_price * 100.0
            if downtrend_pct < min_downtrend_pct:
                continue

            leg_lows = ordered_swings[
                (ordered_swings["SwingType"] == "Low")
                & (ordered_swings["BarIdx"].astype(int) >= context_idx)
                & (ordered_swings["BarIdx"].astype(int) <= head_idx)
            ]
            if (
                not leg_lows.empty
                and head_price > float(leg_lows["Price"].min()) * 1.001
            ):
                continue

            left_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > left_idx)
                & (ordered_swings["BarIdx"].astype(int) < head_idx)
            ]
            if left_neck_candidates.empty:
                continue

            neck1 = left_neck_candidates.sort_values(
                ["Price", "BarIdx"],
                ascending=[False, False],
            ).iloc[0]
            neck1_idx = int(neck1["BarIdx"])
            neck1_price = float(neck1["Price"])
            if neck1_price <= left_price or neck1_price <= head_price:
                continue

            right_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > head_idx)
                & (ordered_swings["BarIdx"].astype(int) < left_idx + max_pattern_bars)
            ]
            if right_neck_candidates.empty:
                continue

            for _, neck2 in right_neck_candidates.iterrows():
                neck2_idx = int(neck2["BarIdx"])
                neck2_price = float(neck2["Price"])
                if neck2_idx - head_idx < min_bars_between_points:
                    continue
                if neck2_price <= head_price:
                    continue

                neckline_average = (neck1_price + neck2_price) / 2.0
                neckline_tilt_pct = (
                    abs(neck2_price - neck1_price)
                    / max(abs(neckline_average), 1e-12)
                    * 100.0
                )
                resistance_price = neckline_average
                resistance_zone_bottom = resistance_price * (
                    1.0 - neckline_zone_pct / 100.0
                )
                resistance_zone_top = resistance_price * (
                    1.0 + neckline_zone_pct / 100.0
                )
                if not (
                    resistance_zone_bottom <= neck1_price <= resistance_zone_top
                    and resistance_zone_bottom <= neck2_price <= resistance_zone_top
                ):
                    continue
                if neckline_tilt_pct > max_neckline_tilt_pct:
                    continue

                alternating_extrema_sequence = _is_contiguous_extrema_sequence(
                    ordered_swings,
                    [
                        (left_idx, "Low"),
                        (neck1_idx, "High"),
                        (head_idx, "Low"),
                        (neck2_idx, "High"),
                    ],
                )
                if require_alternating_extrema_sequence and not alternating_extrema_sequence:
                    continue

                head_support_zone_bottom = head_price * (
                    1.0 - head_support_zone_pct / 100.0
                )
                head_support_zone_top = head_price * (
                    1.0 + head_support_zone_pct / 100.0
                )
                earliest_breakout_idx = max(
                    int(neck2["ConfirmBarIdx"]) + 1,
                    neck2_idx + 2,
                )
                latest_breakout_idx = min(
                    len(data) - 1,
                    earliest_breakout_idx + max_entry_wait_bars,
                )
                head_fail_level = head_support_zone_bottom * (
                    1.0 - fail_buffer_pct / 100.0
                )

                for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                    if closes[entry_idx] < head_fail_level:
                        break

                    neckline_entry_price = resistance_price
                    breakout_level = resistance_zone_top * (
                        1.0 + neckline_break_buffer_pct / 100.0
                    )
                    if closes[entry_idx] <= breakout_level:
                        continue

                    right_lows = lows[neck2_idx + 1 : entry_idx]
                    if len(right_lows) == 0:
                        continue

                    right_idx = neck2_idx + 1 + int(np.argmin(right_lows))
                    right_price = float(lows[right_idx])
                    if right_idx - neck2_idx < min_bars_between_points:
                        continue
                    if right_price <= head_price * (1.0 + min_head_depth_pct / 100.0):
                        continue

                    pattern_low = float(np.min(lows[left_idx : entry_idx + 1]))
                    head_lowest_invalidation = pattern_low < head_price * (
                        1.0 - head_lowest_tolerance_pct / 100.0
                    )
                    if require_head_lowest and head_lowest_invalidation:
                        continue

                    if not (
                        head_support_zone_top < left_price < resistance_zone_bottom
                        and head_support_zone_top < right_price < resistance_zone_bottom
                    ):
                        continue

                    shoulder_average = (left_price + right_price) / 2.0
                    shoulder_difference_pct = (
                        abs(right_price - left_price)
                        / max(abs(shoulder_average), 1e-12)
                        * 100.0
                    )
                    if shoulder_difference_pct > shoulder_tolerance_pct:
                        continue

                    head_depth_pct = (
                        (min(left_price, right_price) - head_price)
                        / max(abs(min(left_price, right_price)), 1e-12)
                        * 100.0
                    )
                    if head_depth_pct < min_head_depth_pct:
                        continue

                    if neck2_price <= right_price:
                        continue

                    left_balance_midpoint = (right_price + neck2_price) / 2.0
                    right_balance_midpoint = (left_price + neck1_price) / 2.0
                    if (
                        left_price > left_balance_midpoint
                        or right_price > right_balance_midpoint
                    ):
                        continue

                    left_width = head_idx - left_idx
                    right_width = right_idx - head_idx
                    if left_width < 1 or right_width < 1:
                        continue
                    symmetry_ratio = max(left_width / right_width, right_width / left_width)
                    if symmetry_ratio > max_symmetry_ratio:
                        continue

                    post_right_swing_invalidation = _has_any_swing_between(
                        ordered_swings,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        require_direct_breakout_after_right_shoulder
                        and post_right_swing_invalidation
                    ):
                        continue

                    bearish_structure_after_right_shoulder = _has_bearish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        invalidate_bearish_structure_after_right_shoulder
                        and bearish_structure_after_right_shoulder
                    ):
                        continue

                    bullish_structure_break = _has_bullish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if require_bullish_structure_break and not bullish_structure_break:
                        continue

                    neckline_zone_bottom = resistance_zone_bottom
                    neckline_zone_top = resistance_zone_top
                    pattern_height = resistance_zone_top - head_price
                    target_price = resistance_zone_top + pattern_height
                    follow_through_level = resistance_zone_top + (
                        pattern_height * follow_through_pattern_ratio
                    )
                    follow_through_idx = _find_breakout_follow_through(
                        closes=closes,
                        entry_idx=entry_idx,
                        max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                        follow_through_level=follow_through_level,
                        support_zone_bottom=head_support_zone_bottom,
                        resistance_zone_bottom=resistance_zone_bottom
                        if require_breakout_hold_neckline
                        else None,
                    )
                    if require_breakout_follow_through and follow_through_idx is None:
                        continue

                    pattern_start = _find_double_bottom_start(
                        data,
                        b1_idx=left_idx,
                        resistance_level=neck1_price,
                        fallback_idx=context_idx,
                        fallback_price=context_price,
                        lookback_bars=downtrend_lookback_bars,
                    )
                    pattern_r2 = _compute_inverse_head_shoulders_r2(
                        closes=closes,
                        start_idx=int(pattern_start["BarIdx"]),
                        start_price=float(pattern_start["Price"]),
                        left_idx=left_idx,
                        left_price=left_price,
                        neck1_idx=neck1_idx,
                        neck1_price=neck1_price,
                        head_idx=head_idx,
                        head_price=head_price,
                        neck2_idx=neck2_idx,
                        neck2_price=neck2_price,
                        right_idx=right_idx,
                        right_price=right_price,
                        entry_idx=entry_idx,
                        entry_price=float(closes[entry_idx]),
                    )
                    if (
                        min_pattern_r2 is not None
                        and (
                            not np.isfinite(pattern_r2)
                            or float(pattern_r2) < float(min_pattern_r2)
                        )
                    ):
                        continue

                    entry_volume_ratio = (
                        float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                    )
                    volume_confirmed = (
                        bool(
                            np.isfinite(entry_volume_ratio)
                            and entry_volume_ratio >= volume_multiplier
                        )
                        if has_volume
                        else False
                    )
                    if require_volume_confirmation and not volume_confirmed:
                        continue

                    pattern_key = (left_idx, neck1_idx, head_idx, neck2_idx, right_idx)
                    if pattern_key in seen_patterns:
                        continue
                    seen_patterns.add(pattern_key)

                    stop_price = head_price
                    risk = closes[entry_idx] - stop_price
                    reward = target_price - closes[entry_idx]

                    patterns.append(
                        {
                            "PatternId": len(patterns) + 1,
                            "ContextHighBarIdx": context_idx,
                            "ContextHighDate": context["Date"],
                            "ContextHighPrice": context_price,
                            "LastCHoCHBeforePatternBarIdx": int(last_choch["EventBarIdx"])
                            if last_choch is not None
                            else np.nan,
                            "LastCHoCHBeforePatternDate": last_choch["EventDate"]
                            if last_choch is not None
                            else pd.NaT,
                            "LastCHoCHBeforePatternDirection": str(last_choch["Direction"])
                            if last_choch is not None
                            else None,
                            "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                            "PatternStartDate": pattern_start["Date"],
                            "PatternStartPrice": float(pattern_start["Price"]),
                            "LeftShoulderBarIdx": left_idx,
                            "LeftShoulderDate": left_shoulder["Date"],
                            "LeftShoulderPrice": left_price,
                            "Neckline1BarIdx": neck1_idx,
                            "Neckline1Date": neck1["Date"],
                            "Neckline1Price": neck1_price,
                            "HeadBarIdx": head_idx,
                            "HeadDate": head["Date"],
                            "HeadPrice": head_price,
                            "Neckline2BarIdx": neck2_idx,
                            "Neckline2Date": neck2["Date"],
                            "Neckline2Price": neck2_price,
                            "RightShoulderBarIdx": right_idx,
                            "RightShoulderDate": data.index[right_idx],
                            "RightShoulderPrice": right_price,
                            "EntryBarIdx": entry_idx,
                            "EntryDate": data.index[entry_idx],
                            "EntryPrice": float(closes[entry_idx]),
                            "NecklineEntryPrice": float(neckline_entry_price),
                            "ResistancePrice": float(resistance_price),
                            "FollowThroughBarIdx": int(follow_through_idx)
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughDate": data.index[int(follow_through_idx)]
                            if follow_through_idx is not None
                            else pd.NaT,
                            "FollowThroughPrice": float(closes[int(follow_through_idx)])
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughLevel": float(follow_through_level),
                            "TargetPrice": float(target_price),
                            "StopPrice": float(stop_price),
                            "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                            "ShoulderDifferencePct": float(shoulder_difference_pct),
                            "HeadDepthPct": float(head_depth_pct),
                            "NecklineTiltPct": float(neckline_tilt_pct),
                            "HeadSupportZoneBottom": float(head_support_zone_bottom),
                            "HeadSupportZoneTop": float(head_support_zone_top),
                            "ResistanceZoneBottom": float(resistance_zone_bottom),
                            "ResistanceZoneTop": float(resistance_zone_top),
                            "NecklineZoneBottom": float(neckline_zone_bottom),
                            "NecklineZoneTop": float(neckline_zone_top),
                            "DowntrendPct": float(downtrend_pct),
                            "PatternWidth": int(right_idx - left_idx),
                            "PatternHeight": float(pattern_height),
                            "SymmetryRatio": float(symmetry_ratio),
                            "PatternR2": float(pattern_r2),
                            "PriorLHLL": bool(prior_lh_ll),
                            "LastCHoCHBearish": bool(last_choch_bearish),
                            "AlternatingExtremaSequence": bool(
                                alternating_extrema_sequence
                            ),
                            "HeadLowestInvalidation": bool(head_lowest_invalidation),
                            "BearishStructureAfterRightShoulderInvalidation": bool(
                                bearish_structure_after_right_shoulder
                            ),
                            "PostRightShoulderSwingInvalidation": bool(
                                post_right_swing_invalidation
                            ),
                            "BullishStructureBreak": bool(bullish_structure_break),
                            "VolumeRatio": entry_volume_ratio,
                            "VolumeConfirmed": bool(volume_confirmed),
                        }
                    )
                    break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_inverse_head_shoulders_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_inverse_head_shoulders(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _select_best_pattern_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforeB1BarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforeB1BarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "ResistancePrice",
                "Bottom1BarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _score_double_bottom_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    bottom1_price = max(abs(_finite_or_zero(row.get("Bottom1Price"))), 1e-12)
    height_pct = pattern_height / bottom1_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 80.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    bottom_diff_penalty = _finite_or_zero(row.get("BottomDifferencePct"))
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("Bottom2BarIdx")),
        0.0,
    )

    return (
        height_pct * 2.5
        + width_score
        + downtrend_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - bottom_diff_penalty * 2.0
        - entry_delay * 0.02
    )


def _select_non_overlapping_double_bottoms(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["Bottom1BarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_best_inverse_head_shoulders_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforePatternBarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforePatternBarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "PatternHeight",
                "LeftShoulderBarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_non_overlapping_inverse_head_shoulders(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["LeftShoulderBarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _interval_overlap_pct(
    first: tuple[int, int],
    second: tuple[int, int],
) -> float:
    first_start, first_end = first
    second_start, second_end = second
    overlap = max(0, min(first_end, second_end) - max(first_start, second_start))
    if overlap <= 0:
        return 0.0

    first_width = max(first_end - first_start, 1)
    second_width = max(second_end - second_start, 1)
    return overlap / min(first_width, second_width)


def _score_inverse_head_shoulders_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    head_price = max(abs(_finite_or_zero(row.get("HeadPrice"))), 1e-12)
    height_pct = pattern_height / head_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 100.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    head_depth_score = _finite_or_zero(row.get("HeadDepthPct")) / 5.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    shoulder_penalty = _finite_or_zero(row.get("ShoulderDifferencePct")) / 2.0
    neckline_penalty = _finite_or_zero(row.get("NecklineTiltPct")) / 4.0
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("RightShoulderBarIdx")),
        0.0,
    )

    return (
        height_pct * 2.0
        + width_score
        + downtrend_score
        + head_depth_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - shoulder_penalty
        - neckline_penalty
        - entry_delay * 0.02
    )


def _finite_or_zero(value: Any, *, default: float = 0.0) -> float:
    try:
        number = float(value)
    except (TypeError, ValueError):
        return default
    if not np.isfinite(number):
        return default
    return number


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _find_last_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> pd.Series | None:
    if structure_events is None or structure_events.empty:
        return None

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    if subset.empty:
        return None

    return subset.sort_values("EventBarIdx").iloc[-1]


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if price > support_zone_top:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _is_contiguous_extrema_sequence(
    ordered_swings: pd.DataFrame,
    expected: list[tuple[int, str]],
) -> bool:
    if ordered_swings is None or ordered_swings.empty or not expected:
        return False

    first_idx = int(expected[0][0])
    last_idx = int(expected[-1][0])
    window = (
        ordered_swings[
            (ordered_swings["BarIdx"].astype(int) >= first_idx)
            & (ordered_swings["BarIdx"].astype(int) <= last_idx)
        ]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )
    if len(window) != len(expected):
        return False

    for row_position, (expected_idx, expected_type) in enumerate(expected):
        row = window.iloc[row_position]
        if int(row["BarIdx"]) != int(expected_idx):
            return False
        if str(row["SwingType"]) != str(expected_type):
            return False

    return True


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
    resistance_zone_bottom: float | None = None,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if (
            current_idx > entry_idx
            and resistance_zone_bottom is not None
            and close_price < resistance_zone_bottom
        ):
            return None
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _line_value(
    start_idx: int,
    start_price: float,
    end_idx: int,
    end_price: float,
    target_idx: int,
) -> float:
    if int(end_idx) == int(start_idx):
        return float(end_price)

    slope = (float(end_price) - float(start_price)) / (int(end_idx) - int(start_idx))
    return float(start_price) + slope * (int(target_idx) - int(start_idx))


def _find_inverse_head_shoulders_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    neckline_start_idx: int,
    neckline_start_price: float,
    neckline_end_idx: int,
    neckline_end_price: float,
    neckline_zone_pct: float,
    require_hold_neckline: bool,
    fail_level: float,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx

        if current_idx <= entry_idx:
            continue

        neckline_price = _line_value(
            neckline_start_idx,
            neckline_start_price,
            neckline_end_idx,
            neckline_end_price,
            current_idx,
        )
        neckline_bottom = neckline_price * (1.0 - neckline_zone_pct / 100.0)
        if require_hold_neckline and close_price < neckline_bottom:
            return None
        if close_price < fail_level:
            return None

    return None


def _find_double_bottom_start(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    resistance_level: float,
    fallback_idx: int,
    fallback_price: float,
    lookback_bars: int,
) -> dict[str, Any]:
    search_start = max(0, int(b1_idx) - int(lookback_bars))
    closes = data["close"].to_numpy(dtype=float)
    highs = data["high"].to_numpy(dtype=float)

    for current_idx in range(int(b1_idx) - 1, search_start - 1, -1):
        if closes[current_idx] >= resistance_level or highs[current_idx] >= resistance_level:
            return {
                "BarIdx": current_idx,
                "Date": data.index[current_idx],
                "Price": float(closes[current_idx]),
            }

    return {
        "BarIdx": int(fallback_idx),
        "Date": data.index[int(fallback_idx)],
        "Price": float(fallback_price),
    }


def _compute_double_bottom_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    b1_idx: int,
    b1_price: float,
    resistance_idx: int,
    resistance_price: float,
    b2_idx: int,
    b2_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, b1_idx, resistance_idx, b2_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [start_price, b1_price, resistance_price, b2_price, entry_price],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _compute_inverse_head_shoulders_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    left_idx: int,
    left_price: float,
    neck1_idx: int,
    neck1_price: float,
    head_idx: int,
    head_price: float,
    neck2_idx: int,
    neck2_price: float,
    right_idx: int,
    right_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, left_idx, neck1_idx, head_idx, neck2_idx, right_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [
            start_price,
            left_price,
            neck1_price,
            head_price,
            neck2_price,
            right_price,
            entry_price,
        ],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def _has_bearish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) > int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) < int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def _format_pattern_swing_hover(row: pd.Series) -> str:
    parts: list[str] = []
    if _row_has_value(row, "PatternSwingReason"):
        parts.append(f"Reason: {row['PatternSwingReason']}")
    if _row_has_value(row, "StructureRole"):
        parts.append(f"Role: {row['StructureRole']}")
    if _row_has_value(row, "BreakoutEventType") and _row_has_value(
        row,
        "BreakoutDirection",
    ):
        parts.append(f"Break: {row['BreakoutDirection']} {row['BreakoutEventType']}")
    if _row_has_value(row, "BreakoutDate"):
        parts.append(f"Break date: {row['BreakoutDate']}")
    if "BreakoutPrice" in row.index and np.isfinite(float(row["BreakoutPrice"])):
        parts.append(f"Break close: {float(row['BreakoutPrice']):.4f}")
    if "MovePct" in row.index and np.isfinite(float(row["MovePct"])):
        parts.append(f"Move: {float(row['MovePct']):.2f}%")
    if "RequiredMovePct" in row.index and np.isfinite(float(row["RequiredMovePct"])):
        parts.append(f"Required: {float(row['RequiredMovePct']):.2f}%")
    if "AtrPct" in row.index and np.isfinite(float(row["AtrPct"])):
        parts.append(f"ATR: {float(row['AtrPct']):.2f}%")
    if not parts:
        return ""
    return "<br>" + "<br>".join(parts)


def _row_has_value(row: pd.Series, column: str) -> bool:
    if column not in row.index:
        return False
    value = row[column]
    if pd.isna(value):
        return False
    return str(value) != ""


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
    inverse_head_shoulders: pd.DataFrame | None = None,
    show_inverse_head_shoulders: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                        + _format_pattern_swing_hover(row)
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for pattern_number, (_, pattern) in enumerate(double_bottoms.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")
            point_indexes.extend(
                [
                    int(pattern["Bottom1BarIdx"]),
                    int(pattern["ResistanceBarIdx"]),
                    int(pattern["Bottom2BarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["Bottom1Price"]),
                    float(pattern["ResistancePrice"]),
                    float(pattern["Bottom2Price"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["B1", "Resistance", "B2", "Breakout"])
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    showlegend=pattern_number == 0,
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforeB1Direction', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    if (
        show_inverse_head_shoulders
        and inverse_head_shoulders is not None
        and not inverse_head_shoulders.empty
    ):
        ihs_color = "#38BDF8"
        neckline_color = "#E0F2FE"

        for pattern_number, (_, pattern) in enumerate(inverse_head_shoulders.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []

            point_indexes.extend(
                [
                    int(pattern["LeftShoulderBarIdx"]),
                    int(pattern["Neckline1BarIdx"]),
                    int(pattern["HeadBarIdx"]),
                    int(pattern["Neckline2BarIdx"]),
                    int(pattern["RightShoulderBarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["LeftShoulderPrice"]),
                    float(pattern["Neckline1Price"]),
                    float(pattern["HeadPrice"]),
                    float(pattern["Neckline2Price"]),
                    float(pattern["RightShoulderPrice"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["LS", "N1", "Head", "N2", "RS", "Breakout"])

            if (
                "FollowThroughBarIdx" in inverse_head_shoulders.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")

            point_positions = [
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [data.index[idx] for idx in point_indexes]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Inverse H&S",
                    showlegend=pattern_number == 0,
                    line=dict(color=ihs_color, width=4),
                    marker=dict(
                        color=ihs_color,
                        size=10,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color=ihs_color, size=11),
                    hovertext=(
                        f"Inverse H&S #{int(pattern['PatternId'])}<br>"
                        f"Shoulder diff: {float(pattern['ShoulderDifferencePct']):.2f}%<br>"
                        f"Head depth: {float(pattern['HeadDepthPct']):.2f}%<br>"
                        f"Neckline tilt: {float(pattern['NecklineTiltPct']):.2f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforePatternDirection', 'n/a')}<br>"
                        f"Start: {pattern.get('PatternStartDate', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["LeftShoulderBarIdx"])
                if remove_time_gaps
                else pattern["LeftShoulderDate"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "HeadSupportZoneBottom",
                "HeadSupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(inverse_head_shoulders.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["HeadSupportZoneBottom"]),
                    y1=float(pattern["HeadSupportZoneTop"]),
                    fillcolor="rgba(56,189,248,0.14)",
                    line=dict(color=ihs_color, width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(224,242,254,0.12)",
                    line=dict(color=neckline_color, width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["Neckline1BarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["Neckline1Date"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )
            for name, price, color, dash in [
                (
                    "IHS Resistance Break",
                    float(pattern["ResistanceZoneTop"]),
                    neckline_color,
                    "solid",
                ),
                ("IHS Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("IHS Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "IHS Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "TUPRS.IS"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    raw_swings_df = detect_swing_points(
        raw,
        lookback=10,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    raw_structure_df = detect_bos_choch(
        raw,
        raw_swings_df,
        break_buffer_pct=0.03,
    )
    breakout_swings_df = build_breakout_swings(
        raw,
        raw_swings_df,
        structure_events=raw_structure_df,
        include_broken_swings=True,
        include_protected_swings=True,
    )
    pattern_swings_df = build_pattern_swings(
        raw,
        breakout_swings_df,
        structure_events=raw_structure_df,
        atr_period=14,
        atr_multiplier=0.90,
        min_move_pct=0.65,
        preserve_structure_swings=True,
        structure_min_move_pct=0.30,
    )
    structure_df = detect_bos_choch(
        raw,
        pattern_swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        pattern_swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_breakout_hold_resistance=True,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )
    inverse_head_shoulders_df = detect_inverse_head_shoulders(
        raw,
        pattern_swings_df,
        structure_events=structure_df,
        shoulder_tolerance_pct=3.0,
        min_head_depth_pct=0.75,
        head_lowest_tolerance_pct=0.15,
        head_support_zone_pct=0.25,
        neckline_zone_pct=0.25,
        max_neckline_tilt_pct=1.0,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        neckline_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_last_choch_bearish=True,
        require_alternating_extrema_sequence=True,
        require_head_lowest=True,
        require_bullish_structure_break=True,
        invalidate_bearish_structure_after_right_shoulder=True,
        require_direct_breakout_after_right_shoulder=True,
        require_breakout_hold_neckline=True,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )

    print(f"Raw swings: {len(raw_swings_df)}")
    print(f"Breakout swings: {len(breakout_swings_df)}")
    print(f"Pattern swings: {len(pattern_swings_df)}")
    print(pattern_swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))
    print(inverse_head_shoulders_df.tail(20))

    figure = plot_swing_points(
        raw,
        pattern_swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=False,
        double_bottoms=double_bottoms_df,
        inverse_head_shoulders=inverse_head_shoulders_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def build_pattern_swings(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    atr_period: int = 14,
    atr_multiplier: float = 1.25,
    min_move_pct: float = 0.80,
    preserve_structure_swings: bool = True,
    always_keep_structure_swings: bool = True,
    structure_min_move_pct: float | None = None,
) -> pd.DataFrame:
    """
    Convert local swings into a cleaner pattern-detection swing set.

    This works like an object-detection proposal stage: local pivots are the raw
    candidates, then ATR-aware zigzag distance and optional BOS/CHoCH
    participation decide which pivots are worth sending to pattern detection.
    """
    if atr_period < 1:
        raise ValueError("atr_period must be at least 1.")
    if atr_multiplier < 0:
        raise ValueError("atr_multiplier cannot be negative.")
    if min_move_pct < 0:
        raise ValueError("min_move_pct cannot be negative.")
    if structure_min_move_pct is not None and structure_min_move_pct < 0:
        raise ValueError("structure_min_move_pct cannot be negative.")

    base_columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]
    computed_columns = [
        "AtrPct",
        "RequiredMovePct",
        "MovePct",
        "PatternSwingReason",
    ]
    extra_columns = [
        column
        for column in swings.columns
        if column not in base_columns and column not in computed_columns
    ]
    output_columns = base_columns + extra_columns + computed_columns

    if swings is None or swings.empty:
        return pd.DataFrame(columns=output_columns)

    data = prepare_ohlcv(df)
    atr_pct = _calculate_atr_pct(data, period=atr_period)
    ordered = swings.sort_values("BarIdx").reset_index(drop=True)
    forced_indexes = (
        _structure_swing_indexes(structure_events)
        if preserve_structure_swings and structure_events is not None
        else set()
    )
    structure_floor = (
        float(structure_min_move_pct)
        if structure_min_move_pct is not None
        else float(min_move_pct) * 0.50
    )

    rows: list[dict[str, Any]] = []
    for _, swing in ordered.iterrows():
        row = {column: swing[column] for column in base_columns if column in swing}
        for column in extra_columns:
            row[column] = swing[column]
        bar_idx = int(row["BarIdx"])
        row["AtrPct"] = float(atr_pct[bar_idx]) if 0 <= bar_idx < len(atr_pct) else np.nan
        atr_required = (
            row["AtrPct"] * float(atr_multiplier)
            if np.isfinite(row["AtrPct"])
            else 0.0
        )
        row["RequiredMovePct"] = max(float(min_move_pct), float(atr_required))
        row["MovePct"] = np.nan
        existing_reason = (
            swing["PatternSwingReason"]
            if "PatternSwingReason" in swing.index
            and not pd.isna(swing["PatternSwingReason"])
            else ""
        )
        row["PatternSwingReason"] = str(existing_reason)
        rows.append(row)

    filtered = _filter_pattern_candidate_swings(
        rows,
        forced_indexes=forced_indexes,
        always_keep_structure_swings=always_keep_structure_swings,
        structure_min_move_pct=structure_floor,
    )
    _add_structure_labels(filtered)
    return pd.DataFrame(filtered, columns=output_columns).reset_index(drop=True)


def build_breakout_swings(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    break_buffer_pct: float = 0.0,
    include_broken_swings: bool = True,
    include_protected_swings: bool = True,
    collapse_consecutive_same_type: bool = False,
) -> pd.DataFrame:
    """
    Keep only swings that became real market-structure levels.

    Local pivots still provide raw candidates, but this layer promotes only the
    swings that were actually broken by BOS/CHoCH or acted as protected swing
    points during a break.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")
    if not include_broken_swings and not include_protected_swings:
        raise ValueError("At least one breakout swing source must be enabled.")

    base_columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]
    output_columns = base_columns + [
        "StructureRole",
        "BreakoutBarIdx",
        "BreakoutDate",
        "BreakoutPrice",
        "BreakoutLevel",
        "BreakoutEventType",
        "BreakoutDirection",
        "PatternSwingReason",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=output_columns)

    data = prepare_ohlcv(df)
    ordered = swings.sort_values("BarIdx").reset_index(drop=True)
    if structure_events is None:
        structure_events = detect_bos_choch(
            data,
            ordered,
            break_buffer_pct=break_buffer_pct,
        )
    if structure_events is None or structure_events.empty:
        return pd.DataFrame(columns=output_columns)

    swing_by_idx = {
        int(row["BarIdx"]): row
        for _, row in ordered.iterrows()
        if not pd.isna(row["BarIdx"])
    }
    candidates: dict[tuple[int, str], dict[str, Any]] = {}

    for _, event in structure_events.sort_values("EventBarIdx").iterrows():
        if include_broken_swings:
            _add_breakout_swing_candidate(
                candidates,
                swing_by_idx=swing_by_idx,
                event=event,
                swing_idx=event.get("BrokenSwingBarIdx"),
                role="broken",
                base_columns=base_columns,
            )
        if include_protected_swings:
            _add_breakout_swing_candidate(
                candidates,
                swing_by_idx=swing_by_idx,
                event=event,
                swing_idx=event.get("ProtectedSwingBarIdx"),
                role="protected",
                base_columns=base_columns,
            )

    rows = sorted(candidates.values(), key=lambda row: int(row["BarIdx"]))
    if collapse_consecutive_same_type:
        rows = _filter_consecutive_same_type(rows)
    _add_structure_labels(rows)
    return pd.DataFrame(rows, columns=output_columns).reset_index(drop=True)


def _add_breakout_swing_candidate(
    candidates: dict[tuple[int, str], dict[str, Any]],
    *,
    swing_by_idx: dict[int, pd.Series],
    event: pd.Series,
    swing_idx: Any,
    role: str,
    base_columns: list[str],
) -> None:
    if pd.isna(swing_idx):
        return

    source = swing_by_idx.get(int(swing_idx))
    if source is None:
        return

    row = {column: source[column] for column in base_columns}
    row["StructureRole"] = role
    row["BreakoutBarIdx"] = int(event["EventBarIdx"])
    row["BreakoutDate"] = event["EventDate"]
    row["BreakoutPrice"] = float(event["BreakPrice"])
    row["BreakoutLevel"] = float(event["Level"])
    row["BreakoutEventType"] = str(event["EventType"])
    row["BreakoutDirection"] = str(event["Direction"])
    row["PatternSwingReason"] = f"breakout_{role}"

    key = (int(row["BarIdx"]), str(row["SwingType"]))
    if key not in candidates:
        candidates[key] = row
        return

    existing = candidates[key]
    roles = set(str(existing["StructureRole"]).split("+"))
    roles.add(role)
    existing["StructureRole"] = "+".join(sorted(roles))
    existing["PatternSwingReason"] = f"breakout_{existing['StructureRole']}"
    if int(row["BreakoutBarIdx"]) < int(existing["BreakoutBarIdx"]):
        for column in [
            "BreakoutBarIdx",
            "BreakoutDate",
            "BreakoutPrice",
            "BreakoutLevel",
            "BreakoutEventType",
            "BreakoutDirection",
        ]:
            existing[column] = row[column]


def _calculate_atr_pct(data: pd.DataFrame, *, period: int) -> np.ndarray:
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    previous_close = np.roll(close, 1)
    previous_close[0] = close[0]

    true_range = np.maximum.reduce(
        [
            high - low,
            np.abs(high - previous_close),
            np.abs(low - previous_close),
        ]
    )
    atr = pd.Series(true_range).rolling(period, min_periods=1).mean().to_numpy()
    return atr / np.maximum(np.abs(close), 1e-12) * 100.0


def _structure_swing_indexes(structure_events: pd.DataFrame | None) -> set[int]:
    if structure_events is None or structure_events.empty:
        return set()

    indexes: set[int] = set()
    for column in ["BrokenSwingBarIdx", "ProtectedSwingBarIdx"]:
        if column not in structure_events.columns:
            continue
        indexes.update(structure_events[column].dropna().astype(int).tolist())
    return indexes


def _filter_pattern_candidate_swings(
    swings: list[dict[str, Any]],
    *,
    forced_indexes: set[int],
    always_keep_structure_swings: bool,
    structure_min_move_pct: float,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            if not str(swing.get("PatternSwingReason", "")):
                swing["PatternSwingReason"] = "start"
            selected.append(swing)
            continue

        previous = selected[-1]
        move_pct = _swing_move_pct(previous, swing)
        swing["MovePct"] = move_pct
        is_structure_swing = _is_forced_pattern_swing(
            swing,
            forced_indexes=forced_indexes,
        )

        if always_keep_structure_swings and is_structure_swing:
            if not str(swing.get("PatternSwingReason", "")):
                swing["PatternSwingReason"] = "structure"
            selected.append(swing)
            continue

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                previous_is_structure = _is_forced_pattern_swing(
                    previous,
                    forced_indexes=forced_indexes,
                )
                if previous_is_structure:
                    continue
                if len(selected) >= 2:
                    prior = selected[-2]
                    swing["MovePct"] = _swing_move_pct(prior, swing)
                swing["PatternSwingReason"] = previous.get(
                    "PatternSwingReason",
                    "replacement",
                )
                selected[-1] = swing
            continue

        required_move_pct = float(swing.get("RequiredMovePct", 0.0))

        if move_pct >= required_move_pct:
            if not str(swing.get("PatternSwingReason", "")):
                swing["PatternSwingReason"] = "atr_zigzag"
            selected.append(swing)
            continue

        if is_structure_swing and move_pct >= structure_min_move_pct:
            if not str(swing.get("PatternSwingReason", "")):
                swing["PatternSwingReason"] = "structure"
            selected.append(swing)

    return selected


def _is_forced_pattern_swing(
    swing: dict[str, Any],
    *,
    forced_indexes: set[int],
) -> bool:
    if int(swing["BarIdx"]) in forced_indexes:
        return True
    return str(swing.get("PatternSwingReason", "")).startswith("breakout_")


def _swing_move_pct(previous: dict[str, Any], current: dict[str, Any]) -> float:
    move_pct = abs(float(current["Price"]) - float(previous["Price"]))
    return move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    max_symmetry_ratio: float = 2.5,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_resistance: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "LastCHoCHBeforeB1BarIdx",
        "LastCHoCHBeforeB1Date",
        "LastCHoCHBeforeB1Direction",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch_before_b1 = _find_last_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        bearish_choch_before_b1 = (
            last_choch_before_b1 is not None
            and str(last_choch_before_b1["Direction"]) == "Bearish"
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        resistance_candidates = ordered_swings[
            (ordered_swings["SwingType"] == "High")
            & (ordered_swings["BarIdx"].astype(int) > b1_idx)
            & (
                ordered_swings["BarIdx"].astype(int)
                < b1_idx + max_bars_between_bottoms
            )
        ]

        for _, resistance in resistance_candidates.iterrows():
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= b1_price:
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            earliest_breakout_idx = max(
                int(resistance["ConfirmBarIdx"]) + 1,
                resistance_idx + 2,
            )
            latest_breakout_idx = min(
                len(data) - 1,
                earliest_breakout_idx + max_entry_wait_bars,
            )

            for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                if closes[entry_idx] < support_zone_bottom:
                    break

                if closes[entry_idx] <= breakout_level:
                    continue

                right_lows = lows[resistance_idx + 1 : entry_idx]
                if len(right_lows) == 0:
                    continue

                b2_idx = resistance_idx + 1 + int(np.argmin(right_lows))
                b2_price = float(lows[b2_idx])
                bars_between = b2_idx - b1_idx
                if bars_between < min_bars_between_bottoms:
                    continue
                if bars_between > max_bars_between_bottoms:
                    break

                bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
                if bottom_difference_pct > bottom_tolerance_pct:
                    continue
                if not support_zone_bottom <= b2_price <= support_zone_top:
                    continue
                if resistance_price <= max(b1_price, b2_price):
                    continue

                left_width = resistance_idx - b1_idx
                right_width = b2_idx - resistance_idx
                if left_width < 1 or right_width < 1:
                    continue
                symmetry_ratio = max(left_width / right_width, right_width / left_width)
                if symmetry_ratio > max_symmetry_ratio:
                    continue

                left_midpoint = 0.5 * (b1_price + resistance_price)
                right_midpoint = 0.5 * (b2_price + resistance_price)
                if b1_price > right_midpoint or b2_price > left_midpoint:
                    continue

                highs_between = ordered_swings[
                    (ordered_swings["SwingType"] == "High")
                    & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                    & (ordered_swings["BarIdx"].astype(int) < b2_idx)
                ]
                if highs_between.empty:
                    continue
                highest_between = highs_between.sort_values(
                    ["Price", "BarIdx"],
                    ascending=[False, False],
                ).iloc[0]
                if int(highest_between["BarIdx"]) != resistance_idx:
                    continue

                second_peak_invalidation = _has_second_peak_between_zones(
                    ordered_swings,
                    start_idx=resistance_idx,
                    end_idx=b2_idx,
                    support_zone_top=support_zone_top,
                    resistance_zone_bottom=resistance_zone_bottom,
                )
                if invalidate_second_peak_between_zones and second_peak_invalidation:
                    continue

                support_fail_level = min(b1_price, b2_price) * (
                    1.0 - fail_buffer_pct / 100.0
                )
                if np.min(closes[resistance_idx + 1 : entry_idx + 1]) < support_fail_level:
                    continue

                post_b2_swing_invalidation = _has_any_swing_between(
                    ordered_swings,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                    continue

                bullish_structure_break = _has_bullish_structure_break(
                    structure_events,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_bullish_structure_break and not bullish_structure_break:
                    continue

                pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
                target_price = resistance_zone_top + pattern_height
                follow_through_level = resistance_zone_top + (
                    pattern_height * follow_through_pattern_ratio
                )
                follow_through_idx = _find_breakout_follow_through(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                    follow_through_level=follow_through_level,
                    support_zone_bottom=support_zone_bottom,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                )
                if require_breakout_follow_through and follow_through_idx is None:
                    continue

                pattern_start = _find_double_bottom_start(
                    data,
                    b1_idx=b1_idx,
                    resistance_level=resistance_zone_bottom,
                    fallback_idx=context_idx,
                    fallback_price=context_price,
                    lookback_bars=downtrend_lookback_bars,
                )
                pattern_r2 = _compute_double_bottom_r2(
                    closes=closes,
                    start_idx=int(pattern_start["BarIdx"]),
                    start_price=float(pattern_start["Price"]),
                    b1_idx=b1_idx,
                    b1_price=b1_price,
                    resistance_idx=resistance_idx,
                    resistance_price=resistance_price,
                    b2_idx=b2_idx,
                    b2_price=b2_price,
                    entry_idx=entry_idx,
                    entry_price=float(closes[entry_idx]),
                )
                if (
                    min_pattern_r2 is not None
                    and (
                        not np.isfinite(pattern_r2)
                        or float(pattern_r2) < float(min_pattern_r2)
                    )
                ):
                    continue

                entry_volume_ratio = (
                    float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                )
                volume_confirmed = (
                    bool(
                        np.isfinite(entry_volume_ratio)
                        and entry_volume_ratio >= volume_multiplier
                    )
                    if has_volume
                    else False
                )
                if require_volume_confirmation and not volume_confirmed:
                    continue

                pattern_key = (b1_idx, resistance_idx, b2_idx)
                if pattern_key in seen_patterns:
                    continue
                seen_patterns.add(pattern_key)

                stop_price = min(b1_price, b2_price)
                risk = closes[entry_idx] - stop_price
                reward = target_price - closes[entry_idx]

                patterns.append(
                    {
                        "PatternId": len(patterns) + 1,
                        "ContextHighBarIdx": context_idx,
                        "ContextHighDate": context["Date"],
                        "ContextHighPrice": context_price,
                        "PreviousHighBarIdx": previous_high_idx,
                        "PreviousHighDate": previous_high["Date"],
                        "PreviousHighPrice": previous_high_price,
                        "LastCHoCHBeforeB1BarIdx": int(
                            last_choch_before_b1["EventBarIdx"]
                        )
                        if last_choch_before_b1 is not None
                        else np.nan,
                        "LastCHoCHBeforeB1Date": last_choch_before_b1["EventDate"]
                        if last_choch_before_b1 is not None
                        else pd.NaT,
                        "LastCHoCHBeforeB1Direction": str(
                            last_choch_before_b1["Direction"]
                        )
                        if last_choch_before_b1 is not None
                        else None,
                        "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                        "PatternStartDate": pattern_start["Date"],
                        "PatternStartPrice": float(pattern_start["Price"]),
                        "Bottom1BarIdx": b1_idx,
                        "Bottom1Date": bottom1["Date"],
                        "Bottom1Price": b1_price,
                        "ResistanceBarIdx": resistance_idx,
                        "ResistanceDate": resistance["Date"],
                        "ResistancePrice": resistance_price,
                        "Bottom2BarIdx": b2_idx,
                        "Bottom2Date": data.index[b2_idx],
                        "Bottom2Price": b2_price,
                        "EntryBarIdx": entry_idx,
                        "EntryDate": data.index[entry_idx],
                        "EntryPrice": float(closes[entry_idx]),
                        "FollowThroughBarIdx": int(follow_through_idx)
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughDate": data.index[int(follow_through_idx)]
                        if follow_through_idx is not None
                        else pd.NaT,
                        "FollowThroughPrice": float(closes[int(follow_through_idx)])
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughLevel": float(follow_through_level),
                        "TargetPrice": float(target_price),
                        "StopPrice": float(stop_price),
                        "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                        "BottomDifferencePct": float(bottom_difference_pct),
                        "SupportZoneBottom": float(support_zone_bottom),
                        "SupportZoneTop": float(support_zone_top),
                        "ResistanceZoneBottom": float(resistance_zone_bottom),
                        "ResistanceZoneTop": float(resistance_zone_top),
                        "RetracePct": float(retrace_pct),
                        "DowntrendPct": float(downtrend_pct),
                        "PatternWidth": int(b2_idx - b1_idx),
                        "PatternHeight": float(pattern_height),
                        "SymmetryRatio": float(symmetry_ratio),
                        "PatternR2": float(pattern_r2),
                        "PriorLHLL": bool(prior_lh_ll),
                        "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                        "SecondPeakInvalidation": bool(second_peak_invalidation),
                        "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                        "BullishStructureBreak": bool(bullish_structure_break),
                        "VolumeRatio": entry_volume_ratio,
                        "VolumeConfirmed": bool(volume_confirmed),
                    }
                )
                break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_pattern_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_double_bottoms(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def detect_inverse_head_shoulders(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    shoulder_tolerance_pct: float = 3.0,
    min_head_depth_pct: float = 0.75,
    head_lowest_tolerance_pct: float = 0.15,
    head_support_zone_pct: float = 0.25,
    neckline_zone_pct: float = 0.25,
    max_neckline_tilt_pct: float = 1.0,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_points: int = 3,
    max_pattern_bars: int = 260,
    max_symmetry_ratio: float = 2.5,
    neckline_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_prior_lh_ll: bool = True,
    require_last_choch_bearish: bool = True,
    require_alternating_extrema_sequence: bool = True,
    require_head_lowest: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_bearish_structure_after_right_shoulder: bool = True,
    require_direct_breakout_after_right_shoulder: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_neckline: bool = True,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed inverse head and shoulders patterns.

    Sequence:
    - prior bearish CHoCH/downtrend
    - left shoulder low
    - neckline high 1
    - deeper head low
    - neckline high 2
    - right shoulder low
    - close above the resistance zone
    """
    if shoulder_tolerance_pct < 0:
        raise ValueError("shoulder_tolerance_pct cannot be negative.")
    if min_head_depth_pct < 0:
        raise ValueError("min_head_depth_pct cannot be negative.")
    if head_lowest_tolerance_pct < 0:
        raise ValueError("head_lowest_tolerance_pct cannot be negative.")
    if head_support_zone_pct < 0:
        raise ValueError("head_support_zone_pct cannot be negative.")
    if neckline_zone_pct < 0:
        raise ValueError("neckline_zone_pct cannot be negative.")
    if max_neckline_tilt_pct < 0:
        raise ValueError("max_neckline_tilt_pct cannot be negative.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_points < 1:
        raise ValueError("min_bars_between_points must be at least 1.")
    if max_pattern_bars < min_bars_between_points * 4:
        raise ValueError("max_pattern_bars is too small for an IHS pattern.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if neckline_break_buffer_pct < 0:
        raise ValueError("neckline_break_buffer_pct cannot be negative.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "LastCHoCHBeforePatternBarIdx",
        "LastCHoCHBeforePatternDate",
        "LastCHoCHBeforePatternDirection",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "LeftShoulderBarIdx",
        "LeftShoulderDate",
        "LeftShoulderPrice",
        "Neckline1BarIdx",
        "Neckline1Date",
        "Neckline1Price",
        "HeadBarIdx",
        "HeadDate",
        "HeadPrice",
        "Neckline2BarIdx",
        "Neckline2Date",
        "Neckline2Price",
        "RightShoulderBarIdx",
        "RightShoulderDate",
        "RightShoulderPrice",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "NecklineEntryPrice",
        "ResistancePrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "ShoulderDifferencePct",
        "HeadDepthPct",
        "NecklineTiltPct",
        "HeadSupportZoneBottom",
        "HeadSupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "NecklineZoneBottom",
        "NecklineZoneTop",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "LastCHoCHBearish",
        "AlternatingExtremaSequence",
        "HeadLowestInvalidation",
        "BearishStructureAfterRightShoulderInvalidation",
        "PostRightShoulderSwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for left_pos in range(len(lows_only) - 1):
        left_shoulder = lows_only.iloc[left_pos]
        left_idx = int(left_shoulder["BarIdx"])
        left_price = float(left_shoulder["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=left_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch = _find_last_choch_before(
            structure_events,
            before_idx=left_idx,
        )
        last_choch_bearish = (
            last_choch is not None and str(last_choch["Direction"]) == "Bearish"
        )
        if require_last_choch_bearish and not last_choch_bearish:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=left_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=left_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        if context_price <= left_price:
            continue

        head_candidates = lows_only[
            (lows_only["BarIdx"].astype(int) > left_idx + min_bars_between_points)
            & (lows_only["BarIdx"].astype(int) < left_idx + max_pattern_bars)
        ]

        for _, head in head_candidates.iterrows():
            head_idx = int(head["BarIdx"])
            head_price = float(head["Price"])
            if head_price >= left_price * (1.0 - min_head_depth_pct / 100.0):
                continue

            downtrend_pct = (context_price - head_price) / context_price * 100.0
            if downtrend_pct < min_downtrend_pct:
                continue

            leg_lows = ordered_swings[
                (ordered_swings["SwingType"] == "Low")
                & (ordered_swings["BarIdx"].astype(int) >= context_idx)
                & (ordered_swings["BarIdx"].astype(int) <= head_idx)
            ]
            if (
                not leg_lows.empty
                and head_price > float(leg_lows["Price"].min()) * 1.001
            ):
                continue

            left_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > left_idx)
                & (ordered_swings["BarIdx"].astype(int) < head_idx)
            ]
            if left_neck_candidates.empty:
                continue

            neck1 = left_neck_candidates.sort_values(
                ["Price", "BarIdx"],
                ascending=[False, False],
            ).iloc[0]
            neck1_idx = int(neck1["BarIdx"])
            neck1_price = float(neck1["Price"])
            if neck1_price <= left_price or neck1_price <= head_price:
                continue

            right_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > head_idx)
                & (ordered_swings["BarIdx"].astype(int) < left_idx + max_pattern_bars)
            ]
            if right_neck_candidates.empty:
                continue

            for _, neck2 in right_neck_candidates.iterrows():
                neck2_idx = int(neck2["BarIdx"])
                neck2_price = float(neck2["Price"])
                if neck2_idx - head_idx < min_bars_between_points:
                    continue
                if neck2_price <= head_price:
                    continue

                neckline_average = (neck1_price + neck2_price) / 2.0
                neckline_tilt_pct = (
                    abs(neck2_price - neck1_price)
                    / max(abs(neckline_average), 1e-12)
                    * 100.0
                )
                resistance_price = neckline_average
                resistance_zone_bottom = resistance_price * (
                    1.0 - neckline_zone_pct / 100.0
                )
                resistance_zone_top = resistance_price * (
                    1.0 + neckline_zone_pct / 100.0
                )
                if not (
                    resistance_zone_bottom <= neck1_price <= resistance_zone_top
                    and resistance_zone_bottom <= neck2_price <= resistance_zone_top
                ):
                    continue
                if neckline_tilt_pct > max_neckline_tilt_pct:
                    continue

                alternating_extrema_sequence = _is_contiguous_extrema_sequence(
                    ordered_swings,
                    [
                        (left_idx, "Low"),
                        (neck1_idx, "High"),
                        (head_idx, "Low"),
                        (neck2_idx, "High"),
                    ],
                )
                if require_alternating_extrema_sequence and not alternating_extrema_sequence:
                    continue

                head_support_zone_bottom = head_price * (
                    1.0 - head_support_zone_pct / 100.0
                )
                head_support_zone_top = head_price * (
                    1.0 + head_support_zone_pct / 100.0
                )
                earliest_breakout_idx = max(
                    int(neck2["ConfirmBarIdx"]) + 1,
                    neck2_idx + 2,
                )
                latest_breakout_idx = min(
                    len(data) - 1,
                    earliest_breakout_idx + max_entry_wait_bars,
                )
                head_fail_level = head_support_zone_bottom * (
                    1.0 - fail_buffer_pct / 100.0
                )

                for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                    if closes[entry_idx] < head_fail_level:
                        break

                    neckline_entry_price = resistance_price
                    breakout_level = resistance_zone_top * (
                        1.0 + neckline_break_buffer_pct / 100.0
                    )
                    if closes[entry_idx] <= breakout_level:
                        continue

                    right_lows = lows[neck2_idx + 1 : entry_idx]
                    if len(right_lows) == 0:
                        continue

                    right_idx = neck2_idx + 1 + int(np.argmin(right_lows))
                    right_price = float(lows[right_idx])
                    if right_idx - neck2_idx < min_bars_between_points:
                        continue
                    if right_price <= head_price * (1.0 + min_head_depth_pct / 100.0):
                        continue

                    pattern_low = float(np.min(lows[left_idx : entry_idx + 1]))
                    head_lowest_invalidation = pattern_low < head_price * (
                        1.0 - head_lowest_tolerance_pct / 100.0
                    )
                    if require_head_lowest and head_lowest_invalidation:
                        continue

                    if not (
                        head_support_zone_top < left_price < resistance_zone_bottom
                        and head_support_zone_top < right_price < resistance_zone_bottom
                    ):
                        continue

                    shoulder_average = (left_price + right_price) / 2.0
                    shoulder_difference_pct = (
                        abs(right_price - left_price)
                        / max(abs(shoulder_average), 1e-12)
                        * 100.0
                    )
                    if shoulder_difference_pct > shoulder_tolerance_pct:
                        continue

                    head_depth_pct = (
                        (min(left_price, right_price) - head_price)
                        / max(abs(min(left_price, right_price)), 1e-12)
                        * 100.0
                    )
                    if head_depth_pct < min_head_depth_pct:
                        continue

                    if neck2_price <= right_price:
                        continue

                    left_balance_midpoint = (right_price + neck2_price) / 2.0
                    right_balance_midpoint = (left_price + neck1_price) / 2.0
                    if (
                        left_price > left_balance_midpoint
                        or right_price > right_balance_midpoint
                    ):
                        continue

                    left_width = head_idx - left_idx
                    right_width = right_idx - head_idx
                    if left_width < 1 or right_width < 1:
                        continue
                    symmetry_ratio = max(left_width / right_width, right_width / left_width)
                    if symmetry_ratio > max_symmetry_ratio:
                        continue

                    post_right_swing_invalidation = _has_any_swing_between(
                        ordered_swings,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        require_direct_breakout_after_right_shoulder
                        and post_right_swing_invalidation
                    ):
                        continue

                    bearish_structure_after_right_shoulder = _has_bearish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        invalidate_bearish_structure_after_right_shoulder
                        and bearish_structure_after_right_shoulder
                    ):
                        continue

                    bullish_structure_break = _has_bullish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if require_bullish_structure_break and not bullish_structure_break:
                        continue

                    neckline_zone_bottom = resistance_zone_bottom
                    neckline_zone_top = resistance_zone_top
                    pattern_height = resistance_zone_top - head_price
                    target_price = resistance_zone_top + pattern_height
                    follow_through_level = resistance_zone_top + (
                        pattern_height * follow_through_pattern_ratio
                    )
                    follow_through_idx = _find_breakout_follow_through(
                        closes=closes,
                        entry_idx=entry_idx,
                        max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                        follow_through_level=follow_through_level,
                        support_zone_bottom=head_support_zone_bottom,
                        resistance_zone_bottom=resistance_zone_bottom
                        if require_breakout_hold_neckline
                        else None,
                    )
                    if require_breakout_follow_through and follow_through_idx is None:
                        continue

                    pattern_start = _find_double_bottom_start(
                        data,
                        b1_idx=left_idx,
                        resistance_level=neck1_price,
                        fallback_idx=context_idx,
                        fallback_price=context_price,
                        lookback_bars=downtrend_lookback_bars,
                    )
                    pattern_r2 = _compute_inverse_head_shoulders_r2(
                        closes=closes,
                        start_idx=int(pattern_start["BarIdx"]),
                        start_price=float(pattern_start["Price"]),
                        left_idx=left_idx,
                        left_price=left_price,
                        neck1_idx=neck1_idx,
                        neck1_price=neck1_price,
                        head_idx=head_idx,
                        head_price=head_price,
                        neck2_idx=neck2_idx,
                        neck2_price=neck2_price,
                        right_idx=right_idx,
                        right_price=right_price,
                        entry_idx=entry_idx,
                        entry_price=float(closes[entry_idx]),
                    )
                    if (
                        min_pattern_r2 is not None
                        and (
                            not np.isfinite(pattern_r2)
                            or float(pattern_r2) < float(min_pattern_r2)
                        )
                    ):
                        continue

                    entry_volume_ratio = (
                        float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                    )
                    volume_confirmed = (
                        bool(
                            np.isfinite(entry_volume_ratio)
                            and entry_volume_ratio >= volume_multiplier
                        )
                        if has_volume
                        else False
                    )
                    if require_volume_confirmation and not volume_confirmed:
                        continue

                    pattern_key = (left_idx, neck1_idx, head_idx, neck2_idx, right_idx)
                    if pattern_key in seen_patterns:
                        continue
                    seen_patterns.add(pattern_key)

                    stop_price = head_price
                    risk = closes[entry_idx] - stop_price
                    reward = target_price - closes[entry_idx]

                    patterns.append(
                        {
                            "PatternId": len(patterns) + 1,
                            "ContextHighBarIdx": context_idx,
                            "ContextHighDate": context["Date"],
                            "ContextHighPrice": context_price,
                            "LastCHoCHBeforePatternBarIdx": int(last_choch["EventBarIdx"])
                            if last_choch is not None
                            else np.nan,
                            "LastCHoCHBeforePatternDate": last_choch["EventDate"]
                            if last_choch is not None
                            else pd.NaT,
                            "LastCHoCHBeforePatternDirection": str(last_choch["Direction"])
                            if last_choch is not None
                            else None,
                            "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                            "PatternStartDate": pattern_start["Date"],
                            "PatternStartPrice": float(pattern_start["Price"]),
                            "LeftShoulderBarIdx": left_idx,
                            "LeftShoulderDate": left_shoulder["Date"],
                            "LeftShoulderPrice": left_price,
                            "Neckline1BarIdx": neck1_idx,
                            "Neckline1Date": neck1["Date"],
                            "Neckline1Price": neck1_price,
                            "HeadBarIdx": head_idx,
                            "HeadDate": head["Date"],
                            "HeadPrice": head_price,
                            "Neckline2BarIdx": neck2_idx,
                            "Neckline2Date": neck2["Date"],
                            "Neckline2Price": neck2_price,
                            "RightShoulderBarIdx": right_idx,
                            "RightShoulderDate": data.index[right_idx],
                            "RightShoulderPrice": right_price,
                            "EntryBarIdx": entry_idx,
                            "EntryDate": data.index[entry_idx],
                            "EntryPrice": float(closes[entry_idx]),
                            "NecklineEntryPrice": float(neckline_entry_price),
                            "ResistancePrice": float(resistance_price),
                            "FollowThroughBarIdx": int(follow_through_idx)
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughDate": data.index[int(follow_through_idx)]
                            if follow_through_idx is not None
                            else pd.NaT,
                            "FollowThroughPrice": float(closes[int(follow_through_idx)])
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughLevel": float(follow_through_level),
                            "TargetPrice": float(target_price),
                            "StopPrice": float(stop_price),
                            "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                            "ShoulderDifferencePct": float(shoulder_difference_pct),
                            "HeadDepthPct": float(head_depth_pct),
                            "NecklineTiltPct": float(neckline_tilt_pct),
                            "HeadSupportZoneBottom": float(head_support_zone_bottom),
                            "HeadSupportZoneTop": float(head_support_zone_top),
                            "ResistanceZoneBottom": float(resistance_zone_bottom),
                            "ResistanceZoneTop": float(resistance_zone_top),
                            "NecklineZoneBottom": float(neckline_zone_bottom),
                            "NecklineZoneTop": float(neckline_zone_top),
                            "DowntrendPct": float(downtrend_pct),
                            "PatternWidth": int(right_idx - left_idx),
                            "PatternHeight": float(pattern_height),
                            "SymmetryRatio": float(symmetry_ratio),
                            "PatternR2": float(pattern_r2),
                            "PriorLHLL": bool(prior_lh_ll),
                            "LastCHoCHBearish": bool(last_choch_bearish),
                            "AlternatingExtremaSequence": bool(
                                alternating_extrema_sequence
                            ),
                            "HeadLowestInvalidation": bool(head_lowest_invalidation),
                            "BearishStructureAfterRightShoulderInvalidation": bool(
                                bearish_structure_after_right_shoulder
                            ),
                            "PostRightShoulderSwingInvalidation": bool(
                                post_right_swing_invalidation
                            ),
                            "BullishStructureBreak": bool(bullish_structure_break),
                            "VolumeRatio": entry_volume_ratio,
                            "VolumeConfirmed": bool(volume_confirmed),
                        }
                    )
                    break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_inverse_head_shoulders_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_inverse_head_shoulders(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _select_best_pattern_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforeB1BarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforeB1BarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "ResistancePrice",
                "Bottom1BarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _score_double_bottom_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    bottom1_price = max(abs(_finite_or_zero(row.get("Bottom1Price"))), 1e-12)
    height_pct = pattern_height / bottom1_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 80.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    bottom_diff_penalty = _finite_or_zero(row.get("BottomDifferencePct"))
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("Bottom2BarIdx")),
        0.0,
    )

    return (
        height_pct * 2.5
        + width_score
        + downtrend_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - bottom_diff_penalty * 2.0
        - entry_delay * 0.02
    )


def _select_non_overlapping_double_bottoms(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["Bottom1BarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_best_inverse_head_shoulders_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforePatternBarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforePatternBarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "PatternHeight",
                "LeftShoulderBarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_non_overlapping_inverse_head_shoulders(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["LeftShoulderBarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _interval_overlap_pct(
    first: tuple[int, int],
    second: tuple[int, int],
) -> float:
    first_start, first_end = first
    second_start, second_end = second
    overlap = max(0, min(first_end, second_end) - max(first_start, second_start))
    if overlap <= 0:
        return 0.0

    first_width = max(first_end - first_start, 1)
    second_width = max(second_end - second_start, 1)
    return overlap / min(first_width, second_width)


def _score_inverse_head_shoulders_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    head_price = max(abs(_finite_or_zero(row.get("HeadPrice"))), 1e-12)
    height_pct = pattern_height / head_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 100.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    head_depth_score = _finite_or_zero(row.get("HeadDepthPct")) / 5.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    shoulder_penalty = _finite_or_zero(row.get("ShoulderDifferencePct")) / 2.0
    neckline_penalty = _finite_or_zero(row.get("NecklineTiltPct")) / 4.0
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("RightShoulderBarIdx")),
        0.0,
    )

    return (
        height_pct * 2.0
        + width_score
        + downtrend_score
        + head_depth_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - shoulder_penalty
        - neckline_penalty
        - entry_delay * 0.02
    )


def _finite_or_zero(value: Any, *, default: float = 0.0) -> float:
    try:
        number = float(value)
    except (TypeError, ValueError):
        return default
    if not np.isfinite(number):
        return default
    return number


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _find_last_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> pd.Series | None:
    if structure_events is None or structure_events.empty:
        return None

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    if subset.empty:
        return None

    return subset.sort_values("EventBarIdx").iloc[-1]


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if price > support_zone_top:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _is_contiguous_extrema_sequence(
    ordered_swings: pd.DataFrame,
    expected: list[tuple[int, str]],
) -> bool:
    if ordered_swings is None or ordered_swings.empty or not expected:
        return False

    first_idx = int(expected[0][0])
    last_idx = int(expected[-1][0])
    window = (
        ordered_swings[
            (ordered_swings["BarIdx"].astype(int) >= first_idx)
            & (ordered_swings["BarIdx"].astype(int) <= last_idx)
        ]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )
    if len(window) != len(expected):
        return False

    for row_position, (expected_idx, expected_type) in enumerate(expected):
        row = window.iloc[row_position]
        if int(row["BarIdx"]) != int(expected_idx):
            return False
        if str(row["SwingType"]) != str(expected_type):
            return False

    return True


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
    resistance_zone_bottom: float | None = None,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if (
            current_idx > entry_idx
            and resistance_zone_bottom is not None
            and close_price < resistance_zone_bottom
        ):
            return None
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _line_value(
    start_idx: int,
    start_price: float,
    end_idx: int,
    end_price: float,
    target_idx: int,
) -> float:
    if int(end_idx) == int(start_idx):
        return float(end_price)

    slope = (float(end_price) - float(start_price)) / (int(end_idx) - int(start_idx))
    return float(start_price) + slope * (int(target_idx) - int(start_idx))


def _find_inverse_head_shoulders_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    neckline_start_idx: int,
    neckline_start_price: float,
    neckline_end_idx: int,
    neckline_end_price: float,
    neckline_zone_pct: float,
    require_hold_neckline: bool,
    fail_level: float,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx

        if current_idx <= entry_idx:
            continue

        neckline_price = _line_value(
            neckline_start_idx,
            neckline_start_price,
            neckline_end_idx,
            neckline_end_price,
            current_idx,
        )
        neckline_bottom = neckline_price * (1.0 - neckline_zone_pct / 100.0)
        if require_hold_neckline and close_price < neckline_bottom:
            return None
        if close_price < fail_level:
            return None

    return None


def _find_double_bottom_start(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    resistance_level: float,
    fallback_idx: int,
    fallback_price: float,
    lookback_bars: int,
) -> dict[str, Any]:
    search_start = max(0, int(b1_idx) - int(lookback_bars))
    closes = data["close"].to_numpy(dtype=float)
    highs = data["high"].to_numpy(dtype=float)

    for current_idx in range(int(b1_idx) - 1, search_start - 1, -1):
        if closes[current_idx] >= resistance_level or highs[current_idx] >= resistance_level:
            return {
                "BarIdx": current_idx,
                "Date": data.index[current_idx],
                "Price": float(closes[current_idx]),
            }

    return {
        "BarIdx": int(fallback_idx),
        "Date": data.index[int(fallback_idx)],
        "Price": float(fallback_price),
    }


def _compute_double_bottom_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    b1_idx: int,
    b1_price: float,
    resistance_idx: int,
    resistance_price: float,
    b2_idx: int,
    b2_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, b1_idx, resistance_idx, b2_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [start_price, b1_price, resistance_price, b2_price, entry_price],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _compute_inverse_head_shoulders_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    left_idx: int,
    left_price: float,
    neck1_idx: int,
    neck1_price: float,
    head_idx: int,
    head_price: float,
    neck2_idx: int,
    neck2_price: float,
    right_idx: int,
    right_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, left_idx, neck1_idx, head_idx, neck2_idx, right_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [
            start_price,
            left_price,
            neck1_price,
            head_price,
            neck2_price,
            right_price,
            entry_price,
        ],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def _has_bearish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) > int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) < int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def _format_pattern_swing_hover(row: pd.Series) -> str:
    parts: list[str] = []
    if _row_has_value(row, "PatternSwingReason"):
        parts.append(f"Reason: {row['PatternSwingReason']}")
    if _row_has_value(row, "StructureRole"):
        parts.append(f"Role: {row['StructureRole']}")
    if _row_has_value(row, "BreakoutEventType") and _row_has_value(
        row,
        "BreakoutDirection",
    ):
        parts.append(f"Break: {row['BreakoutDirection']} {row['BreakoutEventType']}")
    if _row_has_value(row, "BreakoutDate"):
        parts.append(f"Break date: {row['BreakoutDate']}")
    if "BreakoutPrice" in row.index and np.isfinite(float(row["BreakoutPrice"])):
        parts.append(f"Break close: {float(row['BreakoutPrice']):.4f}")
    if "MovePct" in row.index and np.isfinite(float(row["MovePct"])):
        parts.append(f"Move: {float(row['MovePct']):.2f}%")
    if "RequiredMovePct" in row.index and np.isfinite(float(row["RequiredMovePct"])):
        parts.append(f"Required: {float(row['RequiredMovePct']):.2f}%")
    if "AtrPct" in row.index and np.isfinite(float(row["AtrPct"])):
        parts.append(f"ATR: {float(row['AtrPct']):.2f}%")
    if not parts:
        return ""
    return "<br>" + "<br>".join(parts)


def _row_has_value(row: pd.Series, column: str) -> bool:
    if column not in row.index:
        return False
    value = row[column]
    if pd.isna(value):
        return False
    return str(value) != ""


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
    inverse_head_shoulders: pd.DataFrame | None = None,
    show_inverse_head_shoulders: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                        + _format_pattern_swing_hover(row)
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for pattern_number, (_, pattern) in enumerate(double_bottoms.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")
            point_indexes.extend(
                [
                    int(pattern["Bottom1BarIdx"]),
                    int(pattern["ResistanceBarIdx"]),
                    int(pattern["Bottom2BarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["Bottom1Price"]),
                    float(pattern["ResistancePrice"]),
                    float(pattern["Bottom2Price"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["B1", "Resistance", "B2", "Breakout"])
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    showlegend=pattern_number == 0,
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforeB1Direction', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    if (
        show_inverse_head_shoulders
        and inverse_head_shoulders is not None
        and not inverse_head_shoulders.empty
    ):
        ihs_color = "#38BDF8"
        neckline_color = "#E0F2FE"

        for pattern_number, (_, pattern) in enumerate(inverse_head_shoulders.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []

            point_indexes.extend(
                [
                    int(pattern["LeftShoulderBarIdx"]),
                    int(pattern["Neckline1BarIdx"]),
                    int(pattern["HeadBarIdx"]),
                    int(pattern["Neckline2BarIdx"]),
                    int(pattern["RightShoulderBarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["LeftShoulderPrice"]),
                    float(pattern["Neckline1Price"]),
                    float(pattern["HeadPrice"]),
                    float(pattern["Neckline2Price"]),
                    float(pattern["RightShoulderPrice"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["LS", "N1", "Head", "N2", "RS", "Breakout"])

            if (
                "FollowThroughBarIdx" in inverse_head_shoulders.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")

            point_positions = [
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [data.index[idx] for idx in point_indexes]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Inverse H&S",
                    showlegend=pattern_number == 0,
                    line=dict(color=ihs_color, width=4),
                    marker=dict(
                        color=ihs_color,
                        size=10,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color=ihs_color, size=11),
                    hovertext=(
                        f"Inverse H&S #{int(pattern['PatternId'])}<br>"
                        f"Shoulder diff: {float(pattern['ShoulderDifferencePct']):.2f}%<br>"
                        f"Head depth: {float(pattern['HeadDepthPct']):.2f}%<br>"
                        f"Neckline tilt: {float(pattern['NecklineTiltPct']):.2f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforePatternDirection', 'n/a')}<br>"
                        f"Start: {pattern.get('PatternStartDate', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["LeftShoulderBarIdx"])
                if remove_time_gaps
                else pattern["LeftShoulderDate"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "HeadSupportZoneBottom",
                "HeadSupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(inverse_head_shoulders.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["HeadSupportZoneBottom"]),
                    y1=float(pattern["HeadSupportZoneTop"]),
                    fillcolor="rgba(56,189,248,0.14)",
                    line=dict(color=ihs_color, width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(224,242,254,0.12)",
                    line=dict(color=neckline_color, width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["Neckline1BarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["Neckline1Date"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )
            for name, price, color, dash in [
                (
                    "IHS Resistance Break",
                    float(pattern["ResistanceZoneTop"]),
                    neckline_color,
                    "solid",
                ),
                ("IHS Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("IHS Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "IHS Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "TUPRS.IS"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    raw_swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    raw_structure_df = detect_bos_choch(
        raw,
        raw_swings_df,
        break_buffer_pct=0.03,
    )
    breakout_swings_df = build_breakout_swings(
        raw,
        raw_swings_df,
        structure_events=raw_structure_df,
        include_broken_swings=True,
        include_protected_swings=True,
        collapse_consecutive_same_type=False,
    )
    pattern_swings_df = build_pattern_swings(
        raw,
        breakout_swings_df,
        structure_events=raw_structure_df,
        atr_period=14,
        atr_multiplier=0.90,
        min_move_pct=0.65,
        preserve_structure_swings=True,
        always_keep_structure_swings=True,
        structure_min_move_pct=0.30,
    )
    structure_df = detect_bos_choch(
        raw,
        pattern_swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        pattern_swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_breakout_hold_resistance=True,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )
    inverse_head_shoulders_df = detect_inverse_head_shoulders(
        raw,
        pattern_swings_df,
        structure_events=structure_df,
        shoulder_tolerance_pct=3.0,
        min_head_depth_pct=0.75,
        head_lowest_tolerance_pct=0.15,
        head_support_zone_pct=0.25,
        neckline_zone_pct=0.25,
        max_neckline_tilt_pct=1.0,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        neckline_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_last_choch_bearish=True,
        require_alternating_extrema_sequence=True,
        require_head_lowest=True,
        require_bullish_structure_break=True,
        invalidate_bearish_structure_after_right_shoulder=True,
        require_direct_breakout_after_right_shoulder=True,
        require_breakout_hold_neckline=True,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )

    print(f"Raw swings: {len(raw_swings_df)}")
    print(f"Breakout swings: {len(breakout_swings_df)}")
    print(f"Pattern swings: {len(pattern_swings_df)}")
    print(pattern_swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))
    print(inverse_head_shoulders_df.tail(20))

    figure = plot_swing_points(
        raw,
        pattern_swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=False,
        double_bottoms=double_bottoms_df,
        inverse_head_shoulders=inverse_head_shoulders_df,
    )
    figure.show()
```

```python
from __future__ import annotations

from typing import Any

import numpy as np
import pandas as pd

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    """Clean and validate OHLCV data."""
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")

    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")

    return out


def _is_pivot_high(
    highs: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_high = highs[pivot_idx]
    left_highs = highs[pivot_idx - lookback : pivot_idx]
    right_highs = highs[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_highs) == lookback
        and len(right_highs) == confirm_candles
        and pivot_high > np.max(left_highs)
        and pivot_high >= np.max(right_highs)
    )


def _is_pivot_low(
    lows: np.ndarray,
    pivot_idx: int,
    lookback: int,
    confirm_candles: int,
) -> bool:
    pivot_low = lows[pivot_idx]
    left_lows = lows[pivot_idx - lookback : pivot_idx]
    right_lows = lows[pivot_idx + 1 : pivot_idx + confirm_candles + 1]

    return (
        len(left_lows) == lookback
        and len(right_lows) == confirm_candles
        and pivot_low < np.min(left_lows)
        and pivot_low <= np.min(right_lows)
    )


def detect_swing_points(
    df: pd.DataFrame,
    *,
    lookback: int = 5,
    confirm_candles: int = 3,
    remove_consecutive_same_type: bool = True,
    min_swing_move_pct: float = 0.50,
) -> pd.DataFrame:
    """
    Detect confirmed swing points.

    Rules:
    - A swing high looks back `lookback` candles and waits `confirm_candles`.
    - A swing low uses the same logic.
    - The swing is plotted on the pivot candle.
    - `ConfirmBarIdx` is the first candle where the swing is actually known.
    - Final points are filtered with a zigzag-style minimum move to reduce noise.
    """
    if lookback < 1:
        raise ValueError("lookback must be at least 1.")
    if confirm_candles < 1:
        raise ValueError("confirm_candles must be at least 1.")
    if min_swing_move_pct < 0:
        raise ValueError("min_swing_move_pct cannot be negative.")

    data = prepare_ohlcv(df)
    highs = data["high"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    n = len(data)

    columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]

    if n < lookback + confirm_candles + 1:
        return pd.DataFrame(columns=columns)

    raw_swings: list[dict[str, Any]] = []

    for pivot_idx in range(lookback, n - confirm_candles):
        confirm_idx = pivot_idx + confirm_candles

        if _is_pivot_high(highs, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "High",
                    "Price": float(highs[pivot_idx]),
                }
            )

        if _is_pivot_low(lows, pivot_idx, lookback, confirm_candles):
            raw_swings.append(
                {
                    "BarIdx": pivot_idx,
                    "ConfirmBarIdx": confirm_idx,
                    "Date": data.index[pivot_idx],
                    "ConfirmDate": data.index[confirm_idx],
                    "SwingType": "Low",
                    "Price": float(lows[pivot_idx]),
                }
            )

    if not raw_swings:
        return pd.DataFrame(columns=columns)

    raw_swings.sort(key=lambda row: (row["BarIdx"], row["SwingType"]))

    if remove_consecutive_same_type:
        swings = _filter_consecutive_same_type(raw_swings)
    else:
        swings = raw_swings

    swings = _filter_zigzag_moves(swings, min_move_pct=min_swing_move_pct)
    _add_structure_labels(swings)
    return pd.DataFrame(swings, columns=columns).reset_index(drop=True)


def _filter_consecutive_same_type(swings: list[dict[str, Any]]) -> list[dict[str, Any]]:
    """Keep only the more extreme point when same-type swings come back to back."""
    filtered: list[dict[str, Any]] = []

    for swing in swings:
        if not filtered or filtered[-1]["SwingType"] != swing["SwingType"]:
            filtered.append(swing)
            continue

        previous = filtered[-1]
        replace_previous = (
            swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
        ) or (
            swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
        )

        if replace_previous:
            filtered[-1] = swing

    return filtered


def _filter_zigzag_moves(
    swings: list[dict[str, Any]],
    *,
    min_move_pct: float,
) -> list[dict[str, Any]]:
    """Keep alternating pivots only after a minimum opposite move."""
    if min_move_pct <= 0 or len(swings) <= 1:
        return swings

    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            selected.append(swing)
            continue

        previous = selected[-1]

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                selected[-1] = swing
            continue

        move_pct = abs(float(swing["Price"]) - float(previous["Price"]))
        move_pct = move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0
        if move_pct >= min_move_pct:
            selected.append(swing)

    return selected


def _add_structure_labels(swings: list[dict[str, Any]]) -> None:
    """Add HH/LH/HL/LL labels."""
    last_high: float | None = None
    last_low: float | None = None

    for swing in swings:
        price = float(swing["Price"])

        if swing["SwingType"] == "High":
            swing["Label"] = "HH" if last_high is None or price > last_high else "LH"
            last_high = price
        else:
            swing["Label"] = "LL" if last_low is None or price < last_low else "HL"
            last_low = price


def detect_bos_choch(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    break_buffer_pct: float = 0.0,
) -> pd.DataFrame:
    """
    Detect BOS and CHoCH from confirmed swing breaks.

    Rules:
    - Only confirmed swings are used.
    - A bullish break needs a close above the latest confirmed swing high.
    - A bearish break needs a close below the latest confirmed swing low.
    - Same-direction breaks are BOS.
    - Opposite-direction breaks are CHoCH.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")

    columns = [
        "EventBarIdx",
        "EventDate",
        "EventType",
        "Direction",
        "PreviousBias",
        "NewBias",
        "BrokenSwingBarIdx",
        "BrokenSwingDate",
        "BrokenSwingType",
        "BrokenSwingLabel",
        "ProtectedSwingBarIdx",
        "ProtectedSwingDate",
        "ProtectedSwingType",
        "ProtectedSwingLabel",
        "Level",
        "BreakPrice",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    closes = data["close"].to_numpy(dtype=float)

    events_by_confirm: dict[int, list[dict[str, Any]]] = {}
    for _, row in swings.iterrows():
        confirm_idx = int(row["ConfirmBarIdx"])
        events_by_confirm.setdefault(confirm_idx, []).append(row.to_dict())

    latest_high: dict[str, Any] | None = None
    latest_low: dict[str, Any] | None = None
    broken_highs: set[int] = set()
    broken_lows: set[int] = set()
    bias = "Unknown"
    events: list[dict[str, Any]] = []

    for current_idx in range(len(data)):
        confirmed = sorted(
            events_by_confirm.get(current_idx, []),
            key=lambda item: (int(item["BarIdx"]), item["SwingType"]),
        )
        for swing in confirmed:
            if swing["SwingType"] == "High":
                latest_high = swing
            else:
                latest_low = swing

        close_price = float(closes[current_idx])

        if latest_high is not None:
            high_idx = int(latest_high["BarIdx"])
            high_level = float(latest_high["Price"])
            high_break_level = high_level * (1.0 + break_buffer_pct / 100.0)

            if high_idx not in broken_highs and close_price > high_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bullish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bullish",
                        previous_bias=previous_bias,
                        broken_swing=latest_high,
                        protected_swing=latest_low,
                        level=high_level,
                        break_price=close_price,
                    )
                )
                bias = "Bullish"
                broken_highs.add(high_idx)

        if latest_low is not None:
            low_idx = int(latest_low["BarIdx"])
            low_level = float(latest_low["Price"])
            low_break_level = low_level * (1.0 - break_buffer_pct / 100.0)

            if low_idx not in broken_lows and close_price < low_break_level:
                previous_bias = bias
                event_type = "BOS" if bias in {"Unknown", "Bearish"} else "CHoCH"
                events.append(
                    _make_structure_event(
                        data=data,
                        current_idx=current_idx,
                        event_type=event_type,
                        direction="Bearish",
                        previous_bias=previous_bias,
                        broken_swing=latest_low,
                        protected_swing=latest_high,
                        level=low_level,
                        break_price=close_price,
                    )
                )
                bias = "Bearish"
                broken_lows.add(low_idx)

    return pd.DataFrame(events, columns=columns)


def _make_structure_event(
    *,
    data: pd.DataFrame,
    current_idx: int,
    event_type: str,
    direction: str,
    previous_bias: str,
    broken_swing: dict[str, Any],
    protected_swing: dict[str, Any] | None,
    level: float,
    break_price: float,
) -> dict[str, Any]:
    return {
        "EventBarIdx": int(current_idx),
        "EventDate": data.index[current_idx],
        "EventType": event_type,
        "Direction": direction,
        "PreviousBias": previous_bias,
        "NewBias": direction,
        "BrokenSwingBarIdx": int(broken_swing["BarIdx"]),
        "BrokenSwingDate": broken_swing["Date"],
        "BrokenSwingType": broken_swing["SwingType"],
        "BrokenSwingLabel": broken_swing["Label"],
        "ProtectedSwingBarIdx": int(protected_swing["BarIdx"])
        if protected_swing is not None
        else np.nan,
        "ProtectedSwingDate": protected_swing["Date"]
        if protected_swing is not None
        else pd.NaT,
        "ProtectedSwingType": protected_swing["SwingType"]
        if protected_swing is not None
        else None,
        "ProtectedSwingLabel": protected_swing["Label"]
        if protected_swing is not None
        else None,
        "Level": float(level),
        "BreakPrice": float(break_price),
    }


def filter_swings_by_structure(
    swings: pd.DataFrame,
    structure_events: pd.DataFrame,
) -> pd.DataFrame:
    """Keep only swings that are used by a BOS/CHoCH event."""
    if swings is None or swings.empty:
        return swings
    if structure_events is None or structure_events.empty:
        return swings.iloc[0:0].copy()

    indexes: set[int] = set(
        structure_events["BrokenSwingBarIdx"].dropna().astype(int).tolist()
    )
    indexes.update(
        structure_events["ProtectedSwingBarIdx"].dropna().astype(int).tolist()
    )

    return (
        swings[swings["BarIdx"].astype(int).isin(indexes)]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )


def build_pattern_swings(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    atr_period: int = 14,
    atr_multiplier: float = 1.25,
    min_move_pct: float = 0.80,
    preserve_structure_swings: bool = True,
    always_keep_structure_swings: bool = True,
    structure_min_move_pct: float | None = None,
) -> pd.DataFrame:
    """
    Convert local swings into a cleaner pattern-detection swing set.

    This works like an object-detection proposal stage: local pivots are the raw
    candidates, then ATR-aware zigzag distance and optional BOS/CHoCH
    participation decide which pivots are worth sending to pattern detection.
    """
    if atr_period < 1:
        raise ValueError("atr_period must be at least 1.")
    if atr_multiplier < 0:
        raise ValueError("atr_multiplier cannot be negative.")
    if min_move_pct < 0:
        raise ValueError("min_move_pct cannot be negative.")
    if structure_min_move_pct is not None and structure_min_move_pct < 0:
        raise ValueError("structure_min_move_pct cannot be negative.")

    base_columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]
    computed_columns = [
        "AtrPct",
        "RequiredMovePct",
        "MovePct",
        "PatternSwingReason",
    ]
    extra_columns = [
        column
        for column in swings.columns
        if column not in base_columns and column not in computed_columns
    ]
    output_columns = base_columns + extra_columns + computed_columns

    if swings is None or swings.empty:
        return pd.DataFrame(columns=output_columns)

    data = prepare_ohlcv(df)
    atr_pct = _calculate_atr_pct(data, period=atr_period)
    ordered = swings.sort_values("BarIdx").reset_index(drop=True)
    forced_indexes = (
        _structure_swing_indexes(structure_events)
        if preserve_structure_swings and structure_events is not None
        else set()
    )
    structure_floor = (
        float(structure_min_move_pct)
        if structure_min_move_pct is not None
        else float(min_move_pct) * 0.50
    )

    rows: list[dict[str, Any]] = []
    for _, swing in ordered.iterrows():
        row = {column: swing[column] for column in base_columns if column in swing}
        for column in extra_columns:
            row[column] = swing[column]
        bar_idx = int(row["BarIdx"])
        row["AtrPct"] = float(atr_pct[bar_idx]) if 0 <= bar_idx < len(atr_pct) else np.nan
        atr_required = (
            row["AtrPct"] * float(atr_multiplier)
            if np.isfinite(row["AtrPct"])
            else 0.0
        )
        row["RequiredMovePct"] = max(float(min_move_pct), float(atr_required))
        row["MovePct"] = np.nan
        existing_reason = (
            swing["PatternSwingReason"]
            if "PatternSwingReason" in swing.index
            and not pd.isna(swing["PatternSwingReason"])
            else ""
        )
        row["PatternSwingReason"] = str(existing_reason)
        rows.append(row)

    filtered = _filter_pattern_candidate_swings(
        rows,
        forced_indexes=forced_indexes,
        always_keep_structure_swings=always_keep_structure_swings,
        structure_min_move_pct=structure_floor,
    )
    _add_structure_labels(filtered)
    return pd.DataFrame(filtered, columns=output_columns).reset_index(drop=True)


def build_breakout_swings(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    break_buffer_pct: float = 0.0,
    include_broken_swings: bool = True,
    include_protected_swings: bool = True,
    collapse_consecutive_same_type: bool = False,
) -> pd.DataFrame:
    """
    Keep only swings that became real market-structure levels.

    Local pivots still provide raw candidates, but this layer promotes only the
    swings that were actually broken by BOS/CHoCH or acted as protected swing
    points during a break.
    """
    if break_buffer_pct < 0:
        raise ValueError("break_buffer_pct cannot be negative.")
    if not include_broken_swings and not include_protected_swings:
        raise ValueError("At least one breakout swing source must be enabled.")

    base_columns = [
        "BarIdx",
        "ConfirmBarIdx",
        "Date",
        "ConfirmDate",
        "SwingType",
        "Label",
        "Price",
    ]
    output_columns = base_columns + [
        "StructureRole",
        "BreakoutBarIdx",
        "BreakoutDate",
        "BreakoutPrice",
        "BreakoutLevel",
        "BreakoutEventType",
        "BreakoutDirection",
        "PatternSwingReason",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=output_columns)

    data = prepare_ohlcv(df)
    ordered = swings.sort_values("BarIdx").reset_index(drop=True)
    if structure_events is None:
        structure_events = detect_bos_choch(
            data,
            ordered,
            break_buffer_pct=break_buffer_pct,
        )
    if structure_events is None or structure_events.empty:
        return pd.DataFrame(columns=output_columns)

    swing_by_idx = {
        int(row["BarIdx"]): row
        for _, row in ordered.iterrows()
        if not pd.isna(row["BarIdx"])
    }
    candidates: dict[tuple[int, str], dict[str, Any]] = {}

    for _, event in structure_events.sort_values("EventBarIdx").iterrows():
        if include_broken_swings:
            _add_breakout_swing_candidate(
                candidates,
                swing_by_idx=swing_by_idx,
                event=event,
                swing_idx=event.get("BrokenSwingBarIdx"),
                role="broken",
                base_columns=base_columns,
            )
        if include_protected_swings:
            _add_breakout_swing_candidate(
                candidates,
                swing_by_idx=swing_by_idx,
                event=event,
                swing_idx=event.get("ProtectedSwingBarIdx"),
                role="protected",
                base_columns=base_columns,
            )

    rows = sorted(candidates.values(), key=lambda row: int(row["BarIdx"]))
    if collapse_consecutive_same_type:
        rows = _filter_consecutive_same_type(rows)
    _add_structure_labels(rows)
    return pd.DataFrame(rows, columns=output_columns).reset_index(drop=True)


def _add_breakout_swing_candidate(
    candidates: dict[tuple[int, str], dict[str, Any]],
    *,
    swing_by_idx: dict[int, pd.Series],
    event: pd.Series,
    swing_idx: Any,
    role: str,
    base_columns: list[str],
) -> None:
    if pd.isna(swing_idx):
        return

    source = swing_by_idx.get(int(swing_idx))
    if source is None:
        return

    row = {column: source[column] for column in base_columns}
    row["StructureRole"] = role
    row["BreakoutBarIdx"] = int(event["EventBarIdx"])
    row["BreakoutDate"] = event["EventDate"]
    row["BreakoutPrice"] = float(event["BreakPrice"])
    row["BreakoutLevel"] = float(event["Level"])
    row["BreakoutEventType"] = str(event["EventType"])
    row["BreakoutDirection"] = str(event["Direction"])
    row["PatternSwingReason"] = f"breakout_{role}"

    key = (int(row["BarIdx"]), str(row["SwingType"]))
    if key not in candidates:
        candidates[key] = row
        return

    existing = candidates[key]
    roles = set(str(existing["StructureRole"]).split("+"))
    roles.add(role)
    existing["StructureRole"] = "+".join(sorted(roles))
    existing["PatternSwingReason"] = f"breakout_{existing['StructureRole']}"
    if int(row["BreakoutBarIdx"]) < int(existing["BreakoutBarIdx"]):
        for column in [
            "BreakoutBarIdx",
            "BreakoutDate",
            "BreakoutPrice",
            "BreakoutLevel",
            "BreakoutEventType",
            "BreakoutDirection",
        ]:
            existing[column] = row[column]


def _calculate_atr_pct(data: pd.DataFrame, *, period: int) -> np.ndarray:
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    previous_close = np.roll(close, 1)
    previous_close[0] = close[0]

    true_range = np.maximum.reduce(
        [
            high - low,
            np.abs(high - previous_close),
            np.abs(low - previous_close),
        ]
    )
    atr = pd.Series(true_range).rolling(period, min_periods=1).mean().to_numpy()
    return atr / np.maximum(np.abs(close), 1e-12) * 100.0


def _structure_swing_indexes(structure_events: pd.DataFrame | None) -> set[int]:
    if structure_events is None or structure_events.empty:
        return set()

    indexes: set[int] = set()
    for column in ["BrokenSwingBarIdx", "ProtectedSwingBarIdx"]:
        if column not in structure_events.columns:
            continue
        indexes.update(structure_events[column].dropna().astype(int).tolist())
    return indexes


def _filter_pattern_candidate_swings(
    swings: list[dict[str, Any]],
    *,
    forced_indexes: set[int],
    always_keep_structure_swings: bool,
    structure_min_move_pct: float,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []

    for swing in swings:
        if not selected:
            if not str(swing.get("PatternSwingReason", "")):
                swing["PatternSwingReason"] = "start"
            selected.append(swing)
            continue

        previous = selected[-1]
        move_pct = _swing_move_pct(previous, swing)
        swing["MovePct"] = move_pct
        is_structure_swing = _is_forced_pattern_swing(
            swing,
            forced_indexes=forced_indexes,
        )

        if always_keep_structure_swings and is_structure_swing:
            if not str(swing.get("PatternSwingReason", "")):
                swing["PatternSwingReason"] = "structure"
            selected.append(swing)
            continue

        if swing["SwingType"] == previous["SwingType"]:
            more_extreme = (
                swing["SwingType"] == "High" and swing["Price"] >= previous["Price"]
            ) or (
                swing["SwingType"] == "Low" and swing["Price"] <= previous["Price"]
            )
            if more_extreme:
                previous_is_structure = _is_forced_pattern_swing(
                    previous,
                    forced_indexes=forced_indexes,
                )
                if previous_is_structure:
                    continue
                if len(selected) >= 2:
                    prior = selected[-2]
                    swing["MovePct"] = _swing_move_pct(prior, swing)
                swing["PatternSwingReason"] = previous.get(
                    "PatternSwingReason",
                    "replacement",
                )
                selected[-1] = swing
            continue

        required_move_pct = float(swing.get("RequiredMovePct", 0.0))

        if move_pct >= required_move_pct:
            if not str(swing.get("PatternSwingReason", "")):
                swing["PatternSwingReason"] = "atr_zigzag"
            selected.append(swing)
            continue

        if is_structure_swing and move_pct >= structure_min_move_pct:
            if not str(swing.get("PatternSwingReason", "")):
                swing["PatternSwingReason"] = "structure"
            selected.append(swing)

    return selected


def _is_forced_pattern_swing(
    swing: dict[str, Any],
    *,
    forced_indexes: set[int],
) -> bool:
    if int(swing["BarIdx"]) in forced_indexes:
        return True
    return str(swing.get("PatternSwingReason", "")).startswith("breakout_")


def _swing_move_pct(previous: dict[str, Any], current: dict[str, Any]) -> float:
    move_pct = abs(float(current["Price"]) - float(previous["Price"]))
    return move_pct / max(abs(float(previous["Price"])), 1e-12) * 100.0


def detect_double_bottoms(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    bottom_tolerance_pct: float = 0.35,
    support_zone_pct: float = 0.20,
    resistance_zone_pct: float = 0.25,
    min_retrace_pct: float = 0.38,
    max_retrace_pct: float = 0.90,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_bottoms: int = 4,
    max_bars_between_bottoms: int = 220,
    max_symmetry_ratio: float = 2.5,
    resistance_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_b1_ll: bool = False,
    require_prior_lh_ll: bool = True,
    require_bearish_choch_before_b1: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_second_peak_between_zones: bool = True,
    require_direct_breakout_after_b2: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_resistance: bool = True,
    require_target_before_failure: bool = False,
    target_wait_bars: int | None = None,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed double bottoms from clean swing structure.

    Uses the article-style rules:
    - prior downward move
    - B1 is the first bottom
    - resistance is the swing high after B1
    - B2 is close to B1, default within 0.35%
    - B1 support zone and resistance zone stay fixed
    - pattern completes only after a close above resistance zone
    - optional volume confirmation on breakout
    """
    if bottom_tolerance_pct < 0:
        raise ValueError("bottom_tolerance_pct cannot be negative.")
    if support_zone_pct < 0:
        raise ValueError("support_zone_pct cannot be negative.")
    if resistance_zone_pct < 0:
        raise ValueError("resistance_zone_pct cannot be negative.")
    if resistance_break_buffer_pct < 0:
        raise ValueError("resistance_break_buffer_pct cannot be negative.")
    if not 0 <= min_retrace_pct <= max_retrace_pct:
        raise ValueError("Use 0 <= min_retrace_pct <= max_retrace_pct.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_bottoms < 1:
        raise ValueError("min_bars_between_bottoms must be at least 1.")
    if max_bars_between_bottoms < min_bars_between_bottoms:
        raise ValueError("max_bars_between_bottoms must be >= min_bars_between_bottoms.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if target_wait_bars is not None and target_wait_bars < 1:
        raise ValueError("target_wait_bars must be at least 1 when provided.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "PreviousHighBarIdx",
        "PreviousHighDate",
        "PreviousHighPrice",
        "LastCHoCHBeforeB1BarIdx",
        "LastCHoCHBeforeB1Date",
        "LastCHoCHBeforeB1Direction",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "Bottom1BarIdx",
        "Bottom1Date",
        "Bottom1Price",
        "ResistanceBarIdx",
        "ResistanceDate",
        "ResistancePrice",
        "Bottom2BarIdx",
        "Bottom2Date",
        "Bottom2Price",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetHitBarIdx",
        "TargetHitDate",
        "TargetHitPrice",
        "TargetBeforeFailure",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "BottomDifferencePct",
        "SupportZoneBottom",
        "SupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "RetracePct",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "BearishCHoCHBeforeB1",
        "SecondPeakInvalidation",
        "PostB2SwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for b1_pos in range(len(lows_only) - 1):
        bottom1 = lows_only.iloc[b1_pos]

        if require_b1_ll and bottom1["Label"] != "LL":
            continue

        b1_idx = int(bottom1["BarIdx"])
        b1_price = float(bottom1["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=b1_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch_before_b1 = _find_last_choch_before(
            structure_events,
            before_idx=b1_idx,
        )
        bearish_choch_before_b1 = (
            last_choch_before_b1 is not None
            and str(last_choch_before_b1["Direction"]) == "Bearish"
        )
        if require_bearish_choch_before_b1 and not bearish_choch_before_b1:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=b1_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=b1_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        previous_high = _find_previous_high(ordered_swings, b1_idx=b1_idx)
        if previous_high is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        previous_high_idx = int(previous_high["BarIdx"])
        previous_high_price = float(previous_high["Price"])
        if context_price <= b1_price:
            continue

        downtrend_pct = (context_price - b1_price) / context_price * 100.0
        if downtrend_pct < min_downtrend_pct:
            continue

        leg_swing_lows = ordered_swings[
            (ordered_swings["SwingType"] == "Low")
            & (ordered_swings["BarIdx"].astype(int) >= context_idx)
            & (ordered_swings["BarIdx"].astype(int) <= b1_idx)
        ]
        if (
            not leg_swing_lows.empty
            and b1_price > float(leg_swing_lows["Price"].min()) * 1.001
        ):
            continue

        support_zone_bottom = b1_price * (1.0 - support_zone_pct / 100.0)
        support_zone_top = b1_price * (1.0 + support_zone_pct / 100.0)

        resistance_candidates = ordered_swings[
            (ordered_swings["SwingType"] == "High")
            & (ordered_swings["BarIdx"].astype(int) > b1_idx)
            & (
                ordered_swings["BarIdx"].astype(int)
                < b1_idx + max_bars_between_bottoms
            )
        ]

        for _, resistance in resistance_candidates.iterrows():
            resistance_idx = int(resistance["BarIdx"])
            resistance_price = float(resistance["Price"])

            if resistance_price > context_price:
                continue
            if resistance_price <= b1_price:
                continue

            retrace_pct = (resistance_price - b1_price) / (context_price - b1_price)
            if retrace_pct < min_retrace_pct or retrace_pct > max_retrace_pct:
                continue

            resistance_zone_bottom = resistance_price * (1.0 - resistance_zone_pct / 100.0)
            resistance_zone_top = resistance_price * (1.0 + resistance_zone_pct / 100.0)
            breakout_level = resistance_zone_top * (
                1.0 + resistance_break_buffer_pct / 100.0
            )
            earliest_breakout_idx = max(
                int(resistance["ConfirmBarIdx"]) + 1,
                resistance_idx + 2,
            )
            latest_breakout_idx = min(
                len(data) - 1,
                earliest_breakout_idx + max_entry_wait_bars,
            )

            for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                if closes[entry_idx] < support_zone_bottom:
                    break

                if closes[entry_idx] <= breakout_level:
                    continue

                right_lows = lows[resistance_idx + 1 : entry_idx]
                if len(right_lows) == 0:
                    continue

                b2_idx = resistance_idx + 1 + int(np.argmin(right_lows))
                b2_price = float(lows[b2_idx])
                bars_between = b2_idx - b1_idx
                if bars_between < min_bars_between_bottoms:
                    continue
                if bars_between > max_bars_between_bottoms:
                    break

                bottom_difference_pct = abs(b2_price - b1_price) / b1_price * 100.0
                if bottom_difference_pct > bottom_tolerance_pct:
                    continue
                if not support_zone_bottom <= b2_price <= support_zone_top:
                    continue
                if resistance_price <= max(b1_price, b2_price):
                    continue

                left_width = resistance_idx - b1_idx
                right_width = b2_idx - resistance_idx
                if left_width < 1 or right_width < 1:
                    continue
                symmetry_ratio = max(left_width / right_width, right_width / left_width)
                if symmetry_ratio > max_symmetry_ratio:
                    continue

                left_midpoint = 0.5 * (b1_price + resistance_price)
                right_midpoint = 0.5 * (b2_price + resistance_price)
                if b1_price > right_midpoint or b2_price > left_midpoint:
                    continue

                highs_between = ordered_swings[
                    (ordered_swings["SwingType"] == "High")
                    & (ordered_swings["BarIdx"].astype(int) > b1_idx)
                    & (ordered_swings["BarIdx"].astype(int) < b2_idx)
                ]
                if highs_between.empty:
                    continue
                highest_between = highs_between.sort_values(
                    ["Price", "BarIdx"],
                    ascending=[False, False],
                ).iloc[0]
                if int(highest_between["BarIdx"]) != resistance_idx:
                    continue

                second_peak_invalidation = _has_second_peak_between_zones(
                    ordered_swings,
                    start_idx=resistance_idx,
                    end_idx=b2_idx,
                    support_zone_top=support_zone_top,
                    resistance_zone_bottom=resistance_zone_bottom,
                )
                if invalidate_second_peak_between_zones and second_peak_invalidation:
                    continue

                support_fail_level = min(b1_price, b2_price) * (
                    1.0 - fail_buffer_pct / 100.0
                )
                if np.min(closes[resistance_idx + 1 : entry_idx + 1]) < support_fail_level:
                    continue

                post_b2_swing_invalidation = _has_any_swing_between(
                    ordered_swings,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_direct_breakout_after_b2 and post_b2_swing_invalidation:
                    continue

                bullish_structure_break = _has_bullish_structure_break(
                    structure_events,
                    start_idx=b2_idx,
                    end_idx=entry_idx,
                )
                if require_bullish_structure_break and not bullish_structure_break:
                    continue

                pattern_height = resistance_zone_top - ((b1_price + b2_price) / 2.0)
                target_price = resistance_zone_top + pattern_height
                follow_through_level = resistance_zone_top + (
                    pattern_height * follow_through_pattern_ratio
                )
                follow_through_idx = _find_breakout_follow_through(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                    follow_through_level=follow_through_level,
                    support_zone_bottom=support_zone_bottom,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                )
                if require_breakout_follow_through and follow_through_idx is None:
                    continue

                stop_price = min(b1_price, b2_price)
                target_deadline_idx = min(
                    len(data) - 1,
                    entry_idx
                    + (
                        int(target_wait_bars)
                        if target_wait_bars is not None
                        else max(follow_through_bars, b2_idx - b1_idx)
                    ),
                )
                target_hit_idx = _find_bullish_target_before_failure(
                    closes=closes,
                    entry_idx=entry_idx,
                    max_idx=target_deadline_idx,
                    target_price=target_price,
                    stop_price=stop_price,
                    resistance_zone_bottom=resistance_zone_bottom
                    if require_breakout_hold_resistance
                    else None,
                    structure_events=structure_events,
                )
                target_before_failure = target_hit_idx is not None
                if require_target_before_failure and not target_before_failure:
                    continue

                pattern_start = _find_double_bottom_start(
                    data,
                    b1_idx=b1_idx,
                    resistance_level=resistance_zone_bottom,
                    fallback_idx=context_idx,
                    fallback_price=context_price,
                    lookback_bars=downtrend_lookback_bars,
                )
                pattern_r2 = _compute_double_bottom_r2(
                    closes=closes,
                    start_idx=int(pattern_start["BarIdx"]),
                    start_price=float(pattern_start["Price"]),
                    b1_idx=b1_idx,
                    b1_price=b1_price,
                    resistance_idx=resistance_idx,
                    resistance_price=resistance_price,
                    b2_idx=b2_idx,
                    b2_price=b2_price,
                    entry_idx=entry_idx,
                    entry_price=float(closes[entry_idx]),
                )
                if (
                    min_pattern_r2 is not None
                    and (
                        not np.isfinite(pattern_r2)
                        or float(pattern_r2) < float(min_pattern_r2)
                    )
                ):
                    continue

                entry_volume_ratio = (
                    float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                )
                volume_confirmed = (
                    bool(
                        np.isfinite(entry_volume_ratio)
                        and entry_volume_ratio >= volume_multiplier
                    )
                    if has_volume
                    else False
                )
                if require_volume_confirmation and not volume_confirmed:
                    continue

                pattern_key = (b1_idx, resistance_idx, b2_idx)
                if pattern_key in seen_patterns:
                    continue
                seen_patterns.add(pattern_key)

                risk = closes[entry_idx] - stop_price
                reward = target_price - closes[entry_idx]

                patterns.append(
                    {
                        "PatternId": len(patterns) + 1,
                        "ContextHighBarIdx": context_idx,
                        "ContextHighDate": context["Date"],
                        "ContextHighPrice": context_price,
                        "PreviousHighBarIdx": previous_high_idx,
                        "PreviousHighDate": previous_high["Date"],
                        "PreviousHighPrice": previous_high_price,
                        "LastCHoCHBeforeB1BarIdx": int(
                            last_choch_before_b1["EventBarIdx"]
                        )
                        if last_choch_before_b1 is not None
                        else np.nan,
                        "LastCHoCHBeforeB1Date": last_choch_before_b1["EventDate"]
                        if last_choch_before_b1 is not None
                        else pd.NaT,
                        "LastCHoCHBeforeB1Direction": str(
                            last_choch_before_b1["Direction"]
                        )
                        if last_choch_before_b1 is not None
                        else None,
                        "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                        "PatternStartDate": pattern_start["Date"],
                        "PatternStartPrice": float(pattern_start["Price"]),
                        "Bottom1BarIdx": b1_idx,
                        "Bottom1Date": bottom1["Date"],
                        "Bottom1Price": b1_price,
                        "ResistanceBarIdx": resistance_idx,
                        "ResistanceDate": resistance["Date"],
                        "ResistancePrice": resistance_price,
                        "Bottom2BarIdx": b2_idx,
                        "Bottom2Date": data.index[b2_idx],
                        "Bottom2Price": b2_price,
                        "EntryBarIdx": entry_idx,
                        "EntryDate": data.index[entry_idx],
                        "EntryPrice": float(closes[entry_idx]),
                        "FollowThroughBarIdx": int(follow_through_idx)
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughDate": data.index[int(follow_through_idx)]
                        if follow_through_idx is not None
                        else pd.NaT,
                        "FollowThroughPrice": float(closes[int(follow_through_idx)])
                        if follow_through_idx is not None
                        else np.nan,
                        "FollowThroughLevel": float(follow_through_level),
                        "TargetHitBarIdx": int(target_hit_idx)
                        if target_hit_idx is not None
                        else np.nan,
                        "TargetHitDate": data.index[int(target_hit_idx)]
                        if target_hit_idx is not None
                        else pd.NaT,
                        "TargetHitPrice": float(closes[int(target_hit_idx)])
                        if target_hit_idx is not None
                        else np.nan,
                        "TargetBeforeFailure": bool(target_before_failure),
                        "TargetPrice": float(target_price),
                        "StopPrice": float(stop_price),
                        "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                        "BottomDifferencePct": float(bottom_difference_pct),
                        "SupportZoneBottom": float(support_zone_bottom),
                        "SupportZoneTop": float(support_zone_top),
                        "ResistanceZoneBottom": float(resistance_zone_bottom),
                        "ResistanceZoneTop": float(resistance_zone_top),
                        "RetracePct": float(retrace_pct),
                        "DowntrendPct": float(downtrend_pct),
                        "PatternWidth": int(b2_idx - b1_idx),
                        "PatternHeight": float(pattern_height),
                        "SymmetryRatio": float(symmetry_ratio),
                        "PatternR2": float(pattern_r2),
                        "PriorLHLL": bool(prior_lh_ll),
                        "BearishCHoCHBeforeB1": bool(bearish_choch_before_b1),
                        "SecondPeakInvalidation": bool(second_peak_invalidation),
                        "PostB2SwingInvalidation": bool(post_b2_swing_invalidation),
                        "BullishStructureBreak": bool(bullish_structure_break),
                        "VolumeRatio": entry_volume_ratio,
                        "VolumeConfirmed": bool(volume_confirmed),
                    }
                )
                break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_pattern_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_double_bottoms(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def detect_inverse_head_shoulders(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    structure_events: pd.DataFrame | None = None,
    shoulder_tolerance_pct: float = 3.0,
    min_head_depth_pct: float = 0.75,
    head_lowest_tolerance_pct: float = 0.15,
    head_support_zone_pct: float = 0.25,
    neckline_zone_pct: float = 0.25,
    max_neckline_tilt_pct: float = 1.0,
    min_downtrend_pct: float = 2.0,
    downtrend_lookback_bars: int = 250,
    min_bars_between_points: int = 3,
    max_pattern_bars: int = 260,
    max_symmetry_ratio: float = 2.5,
    neckline_break_buffer_pct: float = 0.0,
    max_entry_wait_bars: int = 120,
    fail_buffer_pct: float = 0.0,
    require_prior_lh_ll: bool = True,
    require_last_choch_bearish: bool = True,
    require_alternating_extrema_sequence: bool = True,
    require_head_lowest: bool = True,
    require_bullish_structure_break: bool = True,
    invalidate_bearish_structure_after_right_shoulder: bool = True,
    require_direct_breakout_after_right_shoulder: bool = True,
    require_breakout_follow_through: bool = True,
    require_breakout_hold_neckline: bool = True,
    require_target_before_failure: bool = False,
    target_wait_bars: int | None = None,
    one_pattern_per_bearish_choch: bool = True,
    max_pattern_overlap_pct: float | None = 0.35,
    follow_through_bars: int = 30,
    follow_through_pattern_ratio: float = 0.35,
    min_pattern_r2: float | None = None,
    require_volume_confirmation: bool = False,
    volume_lookback: int = 20,
    volume_multiplier: float = 1.20,
) -> pd.DataFrame:
    """
    Detect completed inverse head and shoulders patterns.

    Sequence:
    - prior bearish CHoCH/downtrend
    - left shoulder low
    - neckline high 1
    - deeper head low
    - neckline high 2
    - right shoulder low
    - close above the resistance zone
    """
    if shoulder_tolerance_pct < 0:
        raise ValueError("shoulder_tolerance_pct cannot be negative.")
    if min_head_depth_pct < 0:
        raise ValueError("min_head_depth_pct cannot be negative.")
    if head_lowest_tolerance_pct < 0:
        raise ValueError("head_lowest_tolerance_pct cannot be negative.")
    if head_support_zone_pct < 0:
        raise ValueError("head_support_zone_pct cannot be negative.")
    if neckline_zone_pct < 0:
        raise ValueError("neckline_zone_pct cannot be negative.")
    if max_neckline_tilt_pct < 0:
        raise ValueError("max_neckline_tilt_pct cannot be negative.")
    if min_downtrend_pct < 0:
        raise ValueError("min_downtrend_pct cannot be negative.")
    if downtrend_lookback_bars < 1:
        raise ValueError("downtrend_lookback_bars must be at least 1.")
    if min_bars_between_points < 1:
        raise ValueError("min_bars_between_points must be at least 1.")
    if max_pattern_bars < min_bars_between_points * 4:
        raise ValueError("max_pattern_bars is too small for an IHS pattern.")
    if max_symmetry_ratio < 1:
        raise ValueError("max_symmetry_ratio must be at least 1.")
    if max_pattern_overlap_pct is not None and not 0 <= max_pattern_overlap_pct <= 1:
        raise ValueError("max_pattern_overlap_pct must be between 0 and 1.")
    if neckline_break_buffer_pct < 0:
        raise ValueError("neckline_break_buffer_pct cannot be negative.")
    if max_entry_wait_bars < 1:
        raise ValueError("max_entry_wait_bars must be at least 1.")
    if target_wait_bars is not None and target_wait_bars < 1:
        raise ValueError("target_wait_bars must be at least 1 when provided.")
    if follow_through_bars < 1:
        raise ValueError("follow_through_bars must be at least 1.")
    if not 0 <= follow_through_pattern_ratio <= 1:
        raise ValueError("follow_through_pattern_ratio must be between 0 and 1.")
    if volume_lookback < 1:
        raise ValueError("volume_lookback must be at least 1.")
    if volume_multiplier <= 0:
        raise ValueError("volume_multiplier must be positive.")

    columns = [
        "PatternId",
        "ContextHighBarIdx",
        "ContextHighDate",
        "ContextHighPrice",
        "LastCHoCHBeforePatternBarIdx",
        "LastCHoCHBeforePatternDate",
        "LastCHoCHBeforePatternDirection",
        "PatternStartBarIdx",
        "PatternStartDate",
        "PatternStartPrice",
        "LeftShoulderBarIdx",
        "LeftShoulderDate",
        "LeftShoulderPrice",
        "Neckline1BarIdx",
        "Neckline1Date",
        "Neckline1Price",
        "HeadBarIdx",
        "HeadDate",
        "HeadPrice",
        "Neckline2BarIdx",
        "Neckline2Date",
        "Neckline2Price",
        "RightShoulderBarIdx",
        "RightShoulderDate",
        "RightShoulderPrice",
        "EntryBarIdx",
        "EntryDate",
        "EntryPrice",
        "NecklineEntryPrice",
        "ResistancePrice",
        "FollowThroughBarIdx",
        "FollowThroughDate",
        "FollowThroughPrice",
        "FollowThroughLevel",
        "TargetHitBarIdx",
        "TargetHitDate",
        "TargetHitPrice",
        "TargetBeforeFailure",
        "TargetPrice",
        "StopPrice",
        "RiskReward",
        "ShoulderDifferencePct",
        "HeadDepthPct",
        "NecklineTiltPct",
        "HeadSupportZoneBottom",
        "HeadSupportZoneTop",
        "ResistanceZoneBottom",
        "ResistanceZoneTop",
        "NecklineZoneBottom",
        "NecklineZoneTop",
        "DowntrendPct",
        "PatternWidth",
        "PatternHeight",
        "SymmetryRatio",
        "PatternR2",
        "PriorLHLL",
        "LastCHoCHBearish",
        "AlternatingExtremaSequence",
        "HeadLowestInvalidation",
        "BearishStructureAfterRightShoulderInvalidation",
        "PostRightShoulderSwingInvalidation",
        "BullishStructureBreak",
        "VolumeRatio",
        "VolumeConfirmed",
    ]

    if swings is None or swings.empty:
        return pd.DataFrame(columns=columns)

    data = prepare_ohlcv(df)
    ordered_swings = swings.sort_values("BarIdx").reset_index(drop=True)

    if structure_events is None:
        structure_events = detect_bos_choch(data, ordered_swings)

    closes = data["close"].to_numpy(dtype=float)
    lows = data["low"].to_numpy(dtype=float)
    has_volume = "volume" in data.columns
    volume_ratio = pd.Series(np.nan, index=data.index)
    if has_volume:
        volume = pd.to_numeric(data["volume"], errors="coerce")
        volume_average = volume.rolling(volume_lookback, min_periods=1).mean().shift(1)
        volume_ratio = volume / volume_average

    patterns: list[dict[str, Any]] = []
    seen_patterns: set[tuple[int, int, int, int, int]] = set()
    lows_only = ordered_swings[ordered_swings["SwingType"] == "Low"].reset_index(drop=True)

    for left_pos in range(len(lows_only) - 1):
        left_shoulder = lows_only.iloc[left_pos]
        left_idx = int(left_shoulder["BarIdx"])
        left_price = float(left_shoulder["Price"])

        prior_lh_ll = _has_prior_lh_and_ll(ordered_swings, b1_idx=left_idx)
        if require_prior_lh_ll and not prior_lh_ll:
            continue

        last_choch = _find_last_choch_before(
            structure_events,
            before_idx=left_idx,
        )
        last_choch_bearish = (
            last_choch is not None and str(last_choch["Direction"]) == "Bearish"
        )
        if require_last_choch_bearish and not last_choch_bearish:
            continue

        context = _find_context_high(
            ordered_swings,
            b1_idx=left_idx,
            lookback_bars=downtrend_lookback_bars,
        )
        if context is None:
            context = _find_context_high_from_price(
                data,
                b1_idx=left_idx,
                lookback_bars=downtrend_lookback_bars,
            )
        if context is None:
            continue

        context_idx = int(context["BarIdx"])
        context_price = float(context["Price"])
        if context_price <= left_price:
            continue

        head_candidates = lows_only[
            (lows_only["BarIdx"].astype(int) > left_idx + min_bars_between_points)
            & (lows_only["BarIdx"].astype(int) < left_idx + max_pattern_bars)
        ]

        for _, head in head_candidates.iterrows():
            head_idx = int(head["BarIdx"])
            head_price = float(head["Price"])
            if head_price >= left_price * (1.0 - min_head_depth_pct / 100.0):
                continue

            downtrend_pct = (context_price - head_price) / context_price * 100.0
            if downtrend_pct < min_downtrend_pct:
                continue

            leg_lows = ordered_swings[
                (ordered_swings["SwingType"] == "Low")
                & (ordered_swings["BarIdx"].astype(int) >= context_idx)
                & (ordered_swings["BarIdx"].astype(int) <= head_idx)
            ]
            if (
                not leg_lows.empty
                and head_price > float(leg_lows["Price"].min()) * 1.001
            ):
                continue

            left_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > left_idx)
                & (ordered_swings["BarIdx"].astype(int) < head_idx)
            ]
            if left_neck_candidates.empty:
                continue

            neck1 = left_neck_candidates.sort_values(
                ["Price", "BarIdx"],
                ascending=[False, False],
            ).iloc[0]
            neck1_idx = int(neck1["BarIdx"])
            neck1_price = float(neck1["Price"])
            if neck1_price <= left_price or neck1_price <= head_price:
                continue

            right_neck_candidates = ordered_swings[
                (ordered_swings["SwingType"] == "High")
                & (ordered_swings["BarIdx"].astype(int) > head_idx)
                & (ordered_swings["BarIdx"].astype(int) < left_idx + max_pattern_bars)
            ]
            if right_neck_candidates.empty:
                continue

            for _, neck2 in right_neck_candidates.iterrows():
                neck2_idx = int(neck2["BarIdx"])
                neck2_price = float(neck2["Price"])
                if neck2_idx - head_idx < min_bars_between_points:
                    continue
                if neck2_price <= head_price:
                    continue

                neckline_average = (neck1_price + neck2_price) / 2.0
                neckline_tilt_pct = (
                    abs(neck2_price - neck1_price)
                    / max(abs(neckline_average), 1e-12)
                    * 100.0
                )
                resistance_price = neckline_average
                resistance_zone_bottom = resistance_price * (
                    1.0 - neckline_zone_pct / 100.0
                )
                resistance_zone_top = resistance_price * (
                    1.0 + neckline_zone_pct / 100.0
                )
                if not (
                    resistance_zone_bottom <= neck1_price <= resistance_zone_top
                    and resistance_zone_bottom <= neck2_price <= resistance_zone_top
                ):
                    continue
                if neckline_tilt_pct > max_neckline_tilt_pct:
                    continue

                alternating_extrema_sequence = _is_contiguous_extrema_sequence(
                    ordered_swings,
                    [
                        (left_idx, "Low"),
                        (neck1_idx, "High"),
                        (head_idx, "Low"),
                        (neck2_idx, "High"),
                    ],
                )
                if require_alternating_extrema_sequence and not alternating_extrema_sequence:
                    continue

                head_support_zone_bottom = head_price * (
                    1.0 - head_support_zone_pct / 100.0
                )
                head_support_zone_top = head_price * (
                    1.0 + head_support_zone_pct / 100.0
                )
                earliest_breakout_idx = max(
                    int(neck2["ConfirmBarIdx"]) + 1,
                    neck2_idx + 2,
                )
                latest_breakout_idx = min(
                    len(data) - 1,
                    earliest_breakout_idx + max_entry_wait_bars,
                )
                head_fail_level = head_support_zone_bottom * (
                    1.0 - fail_buffer_pct / 100.0
                )

                for entry_idx in range(earliest_breakout_idx, latest_breakout_idx + 1):
                    if closes[entry_idx] < head_fail_level:
                        break

                    neckline_entry_price = resistance_price
                    breakout_level = resistance_zone_top * (
                        1.0 + neckline_break_buffer_pct / 100.0
                    )
                    if closes[entry_idx] <= breakout_level:
                        continue

                    right_lows = lows[neck2_idx + 1 : entry_idx]
                    if len(right_lows) == 0:
                        continue

                    right_idx = neck2_idx + 1 + int(np.argmin(right_lows))
                    right_price = float(lows[right_idx])
                    if right_idx - neck2_idx < min_bars_between_points:
                        continue
                    if right_price <= head_price * (1.0 + min_head_depth_pct / 100.0):
                        continue

                    pattern_low = float(np.min(lows[left_idx : entry_idx + 1]))
                    head_lowest_invalidation = pattern_low < head_price * (
                        1.0 - head_lowest_tolerance_pct / 100.0
                    )
                    if require_head_lowest and head_lowest_invalidation:
                        continue

                    if not (
                        head_support_zone_top < left_price < resistance_zone_bottom
                        and head_support_zone_top < right_price < resistance_zone_bottom
                    ):
                        continue

                    shoulder_average = (left_price + right_price) / 2.0
                    shoulder_difference_pct = (
                        abs(right_price - left_price)
                        / max(abs(shoulder_average), 1e-12)
                        * 100.0
                    )
                    if shoulder_difference_pct > shoulder_tolerance_pct:
                        continue

                    head_depth_pct = (
                        (min(left_price, right_price) - head_price)
                        / max(abs(min(left_price, right_price)), 1e-12)
                        * 100.0
                    )
                    if head_depth_pct < min_head_depth_pct:
                        continue

                    if neck2_price <= right_price:
                        continue

                    left_balance_midpoint = (right_price + neck2_price) / 2.0
                    right_balance_midpoint = (left_price + neck1_price) / 2.0
                    if (
                        left_price > left_balance_midpoint
                        or right_price > right_balance_midpoint
                    ):
                        continue

                    left_width = head_idx - left_idx
                    right_width = right_idx - head_idx
                    if left_width < 1 or right_width < 1:
                        continue
                    symmetry_ratio = max(left_width / right_width, right_width / left_width)
                    if symmetry_ratio > max_symmetry_ratio:
                        continue

                    post_right_swing_invalidation = _has_any_swing_between(
                        ordered_swings,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        require_direct_breakout_after_right_shoulder
                        and post_right_swing_invalidation
                    ):
                        continue

                    bearish_structure_after_right_shoulder = _has_bearish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if (
                        invalidate_bearish_structure_after_right_shoulder
                        and bearish_structure_after_right_shoulder
                    ):
                        continue

                    bullish_structure_break = _has_bullish_structure_break(
                        structure_events,
                        start_idx=right_idx,
                        end_idx=entry_idx,
                    )
                    if require_bullish_structure_break and not bullish_structure_break:
                        continue

                    neckline_zone_bottom = resistance_zone_bottom
                    neckline_zone_top = resistance_zone_top
                    pattern_height = resistance_zone_top - head_price
                    target_price = resistance_zone_top + pattern_height
                    follow_through_level = resistance_zone_top + (
                        pattern_height * follow_through_pattern_ratio
                    )
                    follow_through_idx = _find_breakout_follow_through(
                        closes=closes,
                        entry_idx=entry_idx,
                        max_idx=min(len(data) - 1, entry_idx + follow_through_bars),
                        follow_through_level=follow_through_level,
                        support_zone_bottom=head_support_zone_bottom,
                        resistance_zone_bottom=resistance_zone_bottom
                        if require_breakout_hold_neckline
                        else None,
                    )
                    if require_breakout_follow_through and follow_through_idx is None:
                        continue

                    stop_price = head_price
                    target_deadline_idx = min(
                        len(data) - 1,
                        entry_idx
                        + (
                            int(target_wait_bars)
                            if target_wait_bars is not None
                            else max(follow_through_bars, right_idx - left_idx)
                        ),
                    )
                    target_hit_idx = _find_bullish_target_before_failure(
                        closes=closes,
                        entry_idx=entry_idx,
                        max_idx=target_deadline_idx,
                        target_price=target_price,
                        stop_price=stop_price,
                        resistance_zone_bottom=resistance_zone_bottom
                        if require_breakout_hold_neckline
                        else None,
                        structure_events=structure_events,
                    )
                    target_before_failure = target_hit_idx is not None
                    if require_target_before_failure and not target_before_failure:
                        continue

                    pattern_start = _find_double_bottom_start(
                        data,
                        b1_idx=left_idx,
                        resistance_level=neck1_price,
                        fallback_idx=context_idx,
                        fallback_price=context_price,
                        lookback_bars=downtrend_lookback_bars,
                    )
                    pattern_r2 = _compute_inverse_head_shoulders_r2(
                        closes=closes,
                        start_idx=int(pattern_start["BarIdx"]),
                        start_price=float(pattern_start["Price"]),
                        left_idx=left_idx,
                        left_price=left_price,
                        neck1_idx=neck1_idx,
                        neck1_price=neck1_price,
                        head_idx=head_idx,
                        head_price=head_price,
                        neck2_idx=neck2_idx,
                        neck2_price=neck2_price,
                        right_idx=right_idx,
                        right_price=right_price,
                        entry_idx=entry_idx,
                        entry_price=float(closes[entry_idx]),
                    )
                    if (
                        min_pattern_r2 is not None
                        and (
                            not np.isfinite(pattern_r2)
                            or float(pattern_r2) < float(min_pattern_r2)
                        )
                    ):
                        continue

                    entry_volume_ratio = (
                        float(volume_ratio.iloc[entry_idx]) if has_volume else np.nan
                    )
                    volume_confirmed = (
                        bool(
                            np.isfinite(entry_volume_ratio)
                            and entry_volume_ratio >= volume_multiplier
                        )
                        if has_volume
                        else False
                    )
                    if require_volume_confirmation and not volume_confirmed:
                        continue

                    pattern_key = (left_idx, neck1_idx, head_idx, neck2_idx, right_idx)
                    if pattern_key in seen_patterns:
                        continue
                    seen_patterns.add(pattern_key)

                    risk = closes[entry_idx] - stop_price
                    reward = target_price - closes[entry_idx]

                    patterns.append(
                        {
                            "PatternId": len(patterns) + 1,
                            "ContextHighBarIdx": context_idx,
                            "ContextHighDate": context["Date"],
                            "ContextHighPrice": context_price,
                            "LastCHoCHBeforePatternBarIdx": int(last_choch["EventBarIdx"])
                            if last_choch is not None
                            else np.nan,
                            "LastCHoCHBeforePatternDate": last_choch["EventDate"]
                            if last_choch is not None
                            else pd.NaT,
                            "LastCHoCHBeforePatternDirection": str(last_choch["Direction"])
                            if last_choch is not None
                            else None,
                            "PatternStartBarIdx": int(pattern_start["BarIdx"]),
                            "PatternStartDate": pattern_start["Date"],
                            "PatternStartPrice": float(pattern_start["Price"]),
                            "LeftShoulderBarIdx": left_idx,
                            "LeftShoulderDate": left_shoulder["Date"],
                            "LeftShoulderPrice": left_price,
                            "Neckline1BarIdx": neck1_idx,
                            "Neckline1Date": neck1["Date"],
                            "Neckline1Price": neck1_price,
                            "HeadBarIdx": head_idx,
                            "HeadDate": head["Date"],
                            "HeadPrice": head_price,
                            "Neckline2BarIdx": neck2_idx,
                            "Neckline2Date": neck2["Date"],
                            "Neckline2Price": neck2_price,
                            "RightShoulderBarIdx": right_idx,
                            "RightShoulderDate": data.index[right_idx],
                            "RightShoulderPrice": right_price,
                            "EntryBarIdx": entry_idx,
                            "EntryDate": data.index[entry_idx],
                            "EntryPrice": float(closes[entry_idx]),
                            "NecklineEntryPrice": float(neckline_entry_price),
                            "ResistancePrice": float(resistance_price),
                            "FollowThroughBarIdx": int(follow_through_idx)
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughDate": data.index[int(follow_through_idx)]
                            if follow_through_idx is not None
                            else pd.NaT,
                            "FollowThroughPrice": float(closes[int(follow_through_idx)])
                            if follow_through_idx is not None
                            else np.nan,
                            "FollowThroughLevel": float(follow_through_level),
                            "TargetHitBarIdx": int(target_hit_idx)
                            if target_hit_idx is not None
                            else np.nan,
                            "TargetHitDate": data.index[int(target_hit_idx)]
                            if target_hit_idx is not None
                            else pd.NaT,
                            "TargetHitPrice": float(closes[int(target_hit_idx)])
                            if target_hit_idx is not None
                            else np.nan,
                            "TargetBeforeFailure": bool(target_before_failure),
                            "TargetPrice": float(target_price),
                            "StopPrice": float(stop_price),
                            "RiskReward": float(reward / risk) if risk > 0 else np.nan,
                            "ShoulderDifferencePct": float(shoulder_difference_pct),
                            "HeadDepthPct": float(head_depth_pct),
                            "NecklineTiltPct": float(neckline_tilt_pct),
                            "HeadSupportZoneBottom": float(head_support_zone_bottom),
                            "HeadSupportZoneTop": float(head_support_zone_top),
                            "ResistanceZoneBottom": float(resistance_zone_bottom),
                            "ResistanceZoneTop": float(resistance_zone_top),
                            "NecklineZoneBottom": float(neckline_zone_bottom),
                            "NecklineZoneTop": float(neckline_zone_top),
                            "DowntrendPct": float(downtrend_pct),
                            "PatternWidth": int(right_idx - left_idx),
                            "PatternHeight": float(pattern_height),
                            "SymmetryRatio": float(symmetry_ratio),
                            "PatternR2": float(pattern_r2),
                            "PriorLHLL": bool(prior_lh_ll),
                            "LastCHoCHBearish": bool(last_choch_bearish),
                            "AlternatingExtremaSequence": bool(
                                alternating_extrema_sequence
                            ),
                            "HeadLowestInvalidation": bool(head_lowest_invalidation),
                            "BearishStructureAfterRightShoulderInvalidation": bool(
                                bearish_structure_after_right_shoulder
                            ),
                            "PostRightShoulderSwingInvalidation": bool(
                                post_right_swing_invalidation
                            ),
                            "BullishStructureBreak": bool(bullish_structure_break),
                            "VolumeRatio": entry_volume_ratio,
                            "VolumeConfirmed": bool(volume_confirmed),
                        }
                    )
                    break

    result = pd.DataFrame(patterns, columns=columns)
    if one_pattern_per_bearish_choch:
        result = _select_best_inverse_head_shoulders_per_choch(result, columns=columns)
    if max_pattern_overlap_pct is not None:
        result = _select_non_overlapping_inverse_head_shoulders(
            result,
            columns=columns,
            max_overlap_pct=max_pattern_overlap_pct,
        )
    if not result.empty:
        result = result.sort_values("EntryBarIdx").reset_index(drop=True)
        result["PatternId"] = np.arange(1, len(result) + 1)
    return result


def _find_context_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < b1_idx)
        & (ordered_swings["BarIdx"].astype(int) >= b1_idx - lookback_bars)
    ]
    if candidates.empty:
        return None
    return candidates.sort_values(["Price", "BarIdx"], ascending=[False, False]).iloc[0]


def _select_best_pattern_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforeB1BarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforeB1BarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "ResistancePrice",
                "Bottom1BarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _score_double_bottom_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    bottom1_price = max(abs(_finite_or_zero(row.get("Bottom1Price"))), 1e-12)
    height_pct = pattern_height / bottom1_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 80.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    bottom_diff_penalty = _finite_or_zero(row.get("BottomDifferencePct"))
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("Bottom2BarIdx")),
        0.0,
    )

    return (
        height_pct * 2.5
        + width_score
        + downtrend_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - bottom_diff_penalty * 2.0
        - entry_delay * 0.02
    )


def _select_non_overlapping_double_bottoms(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(_score_double_bottom_candidate, axis=1)
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["Bottom1BarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_best_inverse_head_shoulders_per_choch(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
) -> pd.DataFrame:
    if patterns.empty or "LastCHoCHBeforePatternBarIdx" not in patterns.columns:
        return patterns

    scored = patterns.copy()
    choch_key = scored["LastCHoCHBeforePatternBarIdx"].fillna(-1).astype(int)
    scored["_SelectionGroup"] = choch_key
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )

    selected_indexes: list[int] = []
    for _, group in scored.groupby("_SelectionGroup", sort=True):
        if len(group) == 1:
            selected_indexes.append(int(group.index[0]))
            continue

        best = group.sort_values(
            [
                "_SelectionScore",
                "PatternWidth",
                "PatternHeight",
                "LeftShoulderBarIdx",
            ],
            ascending=[False, False, False, True],
        ).iloc[0]
        selected_indexes.append(int(best.name))

    return (
        scored.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _select_non_overlapping_inverse_head_shoulders(
    patterns: pd.DataFrame,
    *,
    columns: list[str],
    max_overlap_pct: float,
) -> pd.DataFrame:
    if patterns.empty:
        return patterns

    scored = patterns.copy()
    scored["_SelectionScore"] = scored.apply(
        _score_inverse_head_shoulders_candidate,
        axis=1,
    )
    scored = scored.sort_values(
        [
            "_SelectionScore",
            "PatternWidth",
            "PatternHeight",
            "EntryBarIdx",
        ],
        ascending=[False, False, False, True],
    )

    selected_indexes: list[int] = []
    selected_intervals: list[tuple[int, int]] = []

    for index, row in scored.iterrows():
        start_idx = int(row["LeftShoulderBarIdx"])
        end_idx = int(row["EntryBarIdx"])
        candidate_interval = (start_idx, end_idx)

        if any(
            _interval_overlap_pct(candidate_interval, selected) > max_overlap_pct
            for selected in selected_intervals
        ):
            continue

        selected_indexes.append(int(index))
        selected_intervals.append(candidate_interval)

    return (
        patterns.loc[selected_indexes, columns]
        .sort_values("EntryBarIdx")
        .reset_index(drop=True)
    )


def _interval_overlap_pct(
    first: tuple[int, int],
    second: tuple[int, int],
) -> float:
    first_start, first_end = first
    second_start, second_end = second
    overlap = max(0, min(first_end, second_end) - max(first_start, second_start))
    if overlap <= 0:
        return 0.0

    first_width = max(first_end - first_start, 1)
    second_width = max(second_end - second_start, 1)
    return overlap / min(first_width, second_width)


def _score_inverse_head_shoulders_candidate(row: pd.Series) -> float:
    pattern_height = _finite_or_zero(row.get("PatternHeight"))
    head_price = max(abs(_finite_or_zero(row.get("HeadPrice"))), 1e-12)
    height_pct = pattern_height / head_price * 100.0
    width_score = min(_finite_or_zero(row.get("PatternWidth")) / 100.0, 1.5)
    downtrend_score = _finite_or_zero(row.get("DowntrendPct")) / 10.0
    head_depth_score = _finite_or_zero(row.get("HeadDepthPct")) / 5.0
    risk_reward_score = min(max(_finite_or_zero(row.get("RiskReward")), 0.0), 5.0) / 5.0
    r2 = _finite_or_zero(row.get("PatternR2"))
    r2_score = max(min(r2, 1.0), -1.0)

    symmetry_penalty = abs(_finite_or_zero(row.get("SymmetryRatio"), default=1.0) - 1.0)
    shoulder_penalty = _finite_or_zero(row.get("ShoulderDifferencePct")) / 2.0
    neckline_penalty = _finite_or_zero(row.get("NecklineTiltPct")) / 4.0
    entry_delay = max(
        _finite_or_zero(row.get("EntryBarIdx"))
        - _finite_or_zero(row.get("RightShoulderBarIdx")),
        0.0,
    )

    return (
        height_pct * 2.0
        + width_score
        + downtrend_score
        + head_depth_score
        + risk_reward_score
        + r2_score
        - symmetry_penalty * 1.5
        - shoulder_penalty
        - neckline_penalty
        - entry_delay * 0.02
    )


def _finite_or_zero(value: Any, *, default: float = 0.0) -> float:
    try:
        number = float(value)
    except (TypeError, ValueError):
        return default
    if not np.isfinite(number):
        return default
    return number


def _find_previous_high(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> pd.Series | None:
    candidates = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) < int(b1_idx))
    ]
    if candidates.empty:
        return None
    return candidates.sort_values("BarIdx").iloc[-1]


def _has_prior_lh_and_ll(
    ordered_swings: pd.DataFrame,
    *,
    b1_idx: int,
) -> bool:
    previous = ordered_swings[ordered_swings["BarIdx"].astype(int) < int(b1_idx)]
    if previous.empty:
        return False

    labels = set(previous["Label"].dropna().astype(str).tolist())
    return "LH" in labels and "LL" in labels


def _find_last_choch_before(
    structure_events: pd.DataFrame,
    *,
    before_idx: int,
) -> pd.Series | None:
    if structure_events is None or structure_events.empty:
        return None

    subset = structure_events[
        (structure_events["EventType"] == "CHoCH")
        & (structure_events["EventBarIdx"].astype(int) < int(before_idx))
    ]
    if subset.empty:
        return None

    return subset.sort_values("EventBarIdx").iloc[-1]


def _has_second_peak_between_zones(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
    support_zone_top: float,
    resistance_zone_bottom: float,
) -> bool:
    peaks = ordered_swings[
        (ordered_swings["SwingType"] == "High")
        & (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    if peaks.empty:
        return False

    for _, peak in peaks.iterrows():
        price = float(peak["Price"])
        if price > support_zone_top:
            return True

    return False


def _has_any_swing_between(
    ordered_swings: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    swings_between = ordered_swings[
        (ordered_swings["BarIdx"].astype(int) > int(start_idx))
        & (ordered_swings["BarIdx"].astype(int) < int(end_idx))
    ]
    return not swings_between.empty


def _is_contiguous_extrema_sequence(
    ordered_swings: pd.DataFrame,
    expected: list[tuple[int, str]],
) -> bool:
    if ordered_swings is None or ordered_swings.empty or not expected:
        return False

    first_idx = int(expected[0][0])
    last_idx = int(expected[-1][0])
    window = (
        ordered_swings[
            (ordered_swings["BarIdx"].astype(int) >= first_idx)
            & (ordered_swings["BarIdx"].astype(int) <= last_idx)
        ]
        .sort_values("BarIdx")
        .reset_index(drop=True)
    )
    if len(window) != len(expected):
        return False

    for row_position, (expected_idx, expected_type) in enumerate(expected):
        row = window.iloc[row_position]
        if int(row["BarIdx"]) != int(expected_idx):
            return False
        if str(row["SwingType"]) != str(expected_type):
            return False

    return True


def _find_breakout_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    support_zone_bottom: float,
    resistance_zone_bottom: float | None = None,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx
        if (
            current_idx > entry_idx
            and resistance_zone_bottom is not None
            and close_price < resistance_zone_bottom
        ):
            return None
        if current_idx > entry_idx and close_price < support_zone_bottom:
            return None
    return None


def _find_bullish_target_before_failure(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    target_price: float,
    stop_price: float,
    resistance_zone_bottom: float | None = None,
    structure_events: pd.DataFrame | None = None,
) -> int | None:
    bearish_event_indexes = _structure_event_indexes_between(
        structure_events,
        direction="Bearish",
        start_idx=entry_idx,
        end_idx=max_idx,
    )

    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= target_price:
            return current_idx

        if current_idx <= entry_idx:
            continue
        if current_idx in bearish_event_indexes:
            return None
        if resistance_zone_bottom is not None and close_price < resistance_zone_bottom:
            return None
        if close_price < stop_price:
            return None

    return None


def _structure_event_indexes_between(
    structure_events: pd.DataFrame | None,
    *,
    direction: str,
    start_idx: int,
    end_idx: int,
) -> set[int]:
    required_columns = {"Direction", "EventBarIdx", "EventType"}
    if (
        structure_events is None
        or structure_events.empty
        or not required_columns.issubset(structure_events.columns)
    ):
        return set()

    subset = structure_events[
        (structure_events["Direction"] == direction)
        & (structure_events["EventBarIdx"].astype(int) > int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return set(subset["EventBarIdx"].astype(int).tolist())


def _line_value(
    start_idx: int,
    start_price: float,
    end_idx: int,
    end_price: float,
    target_idx: int,
) -> float:
    if int(end_idx) == int(start_idx):
        return float(end_price)

    slope = (float(end_price) - float(start_price)) / (int(end_idx) - int(start_idx))
    return float(start_price) + slope * (int(target_idx) - int(start_idx))


def _find_inverse_head_shoulders_follow_through(
    *,
    closes: np.ndarray,
    entry_idx: int,
    max_idx: int,
    follow_through_level: float,
    neckline_start_idx: int,
    neckline_start_price: float,
    neckline_end_idx: int,
    neckline_end_price: float,
    neckline_zone_pct: float,
    require_hold_neckline: bool,
    fail_level: float,
) -> int | None:
    for current_idx in range(int(entry_idx), int(max_idx) + 1):
        close_price = float(closes[current_idx])
        if close_price >= follow_through_level:
            return current_idx

        if current_idx <= entry_idx:
            continue

        neckline_price = _line_value(
            neckline_start_idx,
            neckline_start_price,
            neckline_end_idx,
            neckline_end_price,
            current_idx,
        )
        neckline_bottom = neckline_price * (1.0 - neckline_zone_pct / 100.0)
        if require_hold_neckline and close_price < neckline_bottom:
            return None
        if close_price < fail_level:
            return None

    return None


def _find_double_bottom_start(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    resistance_level: float,
    fallback_idx: int,
    fallback_price: float,
    lookback_bars: int,
) -> dict[str, Any]:
    search_start = max(0, int(b1_idx) - int(lookback_bars))
    closes = data["close"].to_numpy(dtype=float)
    highs = data["high"].to_numpy(dtype=float)

    for current_idx in range(int(b1_idx) - 1, search_start - 1, -1):
        if closes[current_idx] >= resistance_level or highs[current_idx] >= resistance_level:
            return {
                "BarIdx": current_idx,
                "Date": data.index[current_idx],
                "Price": float(closes[current_idx]),
            }

    return {
        "BarIdx": int(fallback_idx),
        "Date": data.index[int(fallback_idx)],
        "Price": float(fallback_price),
    }


def _compute_double_bottom_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    b1_idx: int,
    b1_price: float,
    resistance_idx: int,
    resistance_price: float,
    b2_idx: int,
    b2_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, b1_idx, resistance_idx, b2_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [start_price, b1_price, resistance_price, b2_price, entry_price],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _compute_inverse_head_shoulders_r2(
    *,
    closes: np.ndarray,
    start_idx: int,
    start_price: float,
    left_idx: int,
    left_price: float,
    neck1_idx: int,
    neck1_price: float,
    head_idx: int,
    head_price: float,
    neck2_idx: int,
    neck2_price: float,
    right_idx: int,
    right_price: float,
    entry_idx: int,
    entry_price: float,
) -> float:
    points_x = np.array(
        [start_idx, left_idx, neck1_idx, head_idx, neck2_idx, right_idx, entry_idx],
        dtype=float,
    )
    points_y = np.array(
        [
            start_price,
            left_price,
            neck1_price,
            head_price,
            neck2_price,
            right_price,
            entry_price,
        ],
        dtype=float,
    )

    if np.any(np.diff(points_x) <= 0):
        return np.nan

    model_x = np.arange(int(start_idx), int(entry_idx) + 1, dtype=float)
    raw = closes[int(start_idx) : int(entry_idx) + 1]
    if len(raw) != len(model_x) or len(raw) < 2:
        return np.nan

    model = np.interp(model_x, points_x, points_y)
    mean = float(np.mean(raw))
    ss_res = float(np.sum((raw - model) ** 2.0))
    ss_tot = float(np.sum((raw - mean) ** 2.0))
    if ss_tot <= 0:
        return np.nan

    return 1.0 - ss_res / ss_tot


def _find_context_high_from_price(
    data: pd.DataFrame,
    *,
    b1_idx: int,
    lookback_bars: int,
) -> dict[str, Any] | None:
    start_idx = max(0, int(b1_idx) - int(lookback_bars))
    if start_idx >= int(b1_idx):
        return None

    highs = data["high"].iloc[start_idx:b1_idx]
    if highs.empty:
        return None

    relative_idx = int(np.argmax(highs.to_numpy(dtype=float)))
    bar_idx = start_idx + relative_idx
    return {
        "BarIdx": bar_idx,
        "Date": data.index[bar_idx],
        "Price": float(data["high"].iloc[bar_idx]),
    }


def _has_bullish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bullish")
        & (structure_events["EventBarIdx"].astype(int) >= int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) <= int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def _has_bearish_structure_break(
    structure_events: pd.DataFrame,
    *,
    start_idx: int,
    end_idx: int,
) -> bool:
    if structure_events is None or structure_events.empty:
        return False

    subset = structure_events[
        (structure_events["Direction"] == "Bearish")
        & (structure_events["EventBarIdx"].astype(int) > int(start_idx))
        & (structure_events["EventBarIdx"].astype(int) < int(end_idx))
        & (structure_events["EventType"].isin(["BOS", "CHoCH"]))
    ]
    return not subset.empty


def _format_pattern_swing_hover(row: pd.Series) -> str:
    parts: list[str] = []
    if _row_has_value(row, "PatternSwingReason"):
        parts.append(f"Reason: {row['PatternSwingReason']}")
    if _row_has_value(row, "StructureRole"):
        parts.append(f"Role: {row['StructureRole']}")
    if _row_has_value(row, "BreakoutEventType") and _row_has_value(
        row,
        "BreakoutDirection",
    ):
        parts.append(f"Break: {row['BreakoutDirection']} {row['BreakoutEventType']}")
    if _row_has_value(row, "BreakoutDate"):
        parts.append(f"Break date: {row['BreakoutDate']}")
    if "BreakoutPrice" in row.index and np.isfinite(float(row["BreakoutPrice"])):
        parts.append(f"Break close: {float(row['BreakoutPrice']):.4f}")
    if "MovePct" in row.index and np.isfinite(float(row["MovePct"])):
        parts.append(f"Move: {float(row['MovePct']):.2f}%")
    if "RequiredMovePct" in row.index and np.isfinite(float(row["RequiredMovePct"])):
        parts.append(f"Required: {float(row['RequiredMovePct']):.2f}%")
    if "AtrPct" in row.index and np.isfinite(float(row["AtrPct"])):
        parts.append(f"ATR: {float(row['AtrPct']):.2f}%")
    if not parts:
        return ""
    return "<br>" + "<br>".join(parts)


def _row_has_value(row: pd.Series, column: str) -> bool:
    if column not in row.index:
        return False
    value = row[column]
    if pd.isna(value):
        return False
    return str(value) != ""


def plot_swing_points(
    df: pd.DataFrame,
    swings: pd.DataFrame,
    *,
    title: str = "Swing Points",
    remove_time_gaps: bool = True,
    structure_events: pd.DataFrame | None = None,
    show_structure_events: bool = True,
    show_only_structure_swings: bool = True,
    structure_break_buffer_pct: float = 0.0,
    double_bottoms: pd.DataFrame | None = None,
    show_double_bottoms: bool = True,
    inverse_head_shoulders: pd.DataFrame | None = None,
    show_inverse_head_shoulders: bool = True,
) -> go.Figure:
    """Plot candles and swing points."""
    if go is None:
        raise ImportError("Plotly is required for plotting. Install it with: pip install plotly")

    data = prepare_ohlcv(df)
    plot_x = np.arange(len(data)) if remove_time_gaps else data.index
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    if structure_events is None and (show_structure_events or show_only_structure_swings):
        structure_events = detect_bos_choch(
            data,
            swings,
            break_buffer_pct=structure_break_buffer_pct,
        )

    plotted_swings = swings
    if show_only_structure_swings and structure_events is not None:
        plotted_swings = filter_swings_by_structure(swings, structure_events)

    fig = go.Figure()

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    colors = {
        "HH": "#22C55E",
        "LH": "#F97316",
        "HL": "#3B82F6",
        "LL": "#EF4444",
    }

    for label in ["HH", "LH", "HL", "LL"]:
        subset = plotted_swings[plotted_swings["Label"] == label]
        if subset.empty:
            continue

        is_high = label in {"HH", "LH"}
        swing_x = (
            subset["BarIdx"].astype(int).to_numpy()
            if remove_time_gaps
            else subset["Date"]
        )
        fig.add_trace(
            go.Scatter(
                x=swing_x,
                y=subset["Price"],
                mode="markers+text",
                name=label,
                marker=dict(
                    symbol="triangle-down" if is_high else "triangle-up",
                    size=10,
                    color=colors[label],
                ),
                text=label,
                textposition="top center" if is_high else "bottom center",
                textfont=dict(size=10, color=colors[label]),
                hovertext=[
                    (
                        f"{row['Label']}<br>"
                        f"Pivot: {row['Date']}<br>"
                        f"Confirm: {row['ConfirmDate']}<br>"
                        f"Price: {float(row['Price']):.4f}"
                        + _format_pattern_swing_hover(row)
                    )
                    for _, row in subset.iterrows()
                ],
                hoverinfo="text",
            )
        )

    if (
        show_structure_events
        and structure_events is not None
        and not structure_events.empty
    ):
        for _, event in structure_events.iterrows():
            direction = str(event["Direction"])
            event_type = str(event["EventType"])
            color = "#22C55E" if direction == "Bullish" else "#EF4444"
            broken_idx = int(event["BrokenSwingBarIdx"])
            event_idx = int(event["EventBarIdx"])
            level = float(event["Level"])
            break_price = float(event["BreakPrice"])

            if remove_time_gaps:
                line_x = [broken_idx, event_idx]
                marker_x = [event_idx]
            else:
                line_x = [event["BrokenSwingDate"], event["EventDate"]]
                marker_x = [event["EventDate"]]

            fig.add_trace(
                go.Scatter(
                    x=line_x,
                    y=[level, level],
                    mode="lines",
                    name=f"{direction} {event_type}",
                    showlegend=False,
                    line=dict(color=color, width=2, dash="dot"),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Broken: {event['BrokenSwingLabel']} @ {level:.4f}<br>"
                        f"Close: {break_price:.4f}<br>"
                        f"Date: {event['EventDate']}"
                    ),
                    hoverinfo="text",
                )
            )
            fig.add_trace(
                go.Scatter(
                    x=marker_x,
                    y=[break_price],
                    mode="markers+text",
                    name=event_type,
                    showlegend=False,
                    marker=dict(
                        symbol="triangle-up"
                        if direction == "Bullish"
                        else "triangle-down",
                        size=12,
                        color=color,
                        line=dict(width=1, color="white"),
                    ),
                    text=event_type,
                    textposition="top center"
                    if direction == "Bullish"
                    else "bottom center",
                    textfont=dict(size=11, color=color),
                    hovertext=(
                        f"{direction} {event_type}<br>"
                        f"Level: {level:.4f}<br>"
                        f"Close: {break_price:.4f}"
                    ),
                    hoverinfo="text",
                )
            )

    if show_double_bottoms and double_bottoms is not None and not double_bottoms.empty:
        for pattern_number, (_, pattern) in enumerate(double_bottoms.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []
            if (
                "PatternStartBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["PatternStartBarIdx"])
            ):
                point_indexes.append(int(pattern["PatternStartBarIdx"]))
                point_prices.append(float(pattern["PatternStartPrice"]))
                point_labels.append("Start")
            point_indexes.extend(
                [
                    int(pattern["Bottom1BarIdx"]),
                    int(pattern["ResistanceBarIdx"]),
                    int(pattern["Bottom2BarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["Bottom1Price"]),
                    float(pattern["ResistancePrice"]),
                    float(pattern["Bottom2Price"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["B1", "Resistance", "B2", "Breakout"])
            if (
                "FollowThroughBarIdx" in double_bottoms.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")
            point_positions = [
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [
                    data.index[idx]
                    for idx in point_indexes
                ]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Double Bottom",
                    showlegend=pattern_number == 0,
                    line=dict(color="#FACC15", width=4),
                    marker=dict(
                        color="#FACC15",
                        size=11,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color="#FACC15", size=11),
                    hovertext=(
                        f"Double Bottom #{int(pattern['PatternId'])}<br>"
                        f"Bottom diff: {float(pattern['BottomDifferencePct']):.2f}%<br>"
                        f"Retrace: {float(pattern['RetracePct']) * 100.0:.1f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforeB1Direction', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["Bottom1BarIdx"])
                if remove_time_gaps
                else pattern["Bottom1Date"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "SupportZoneBottom",
                "SupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(double_bottoms.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["SupportZoneBottom"]),
                    y1=float(pattern["SupportZoneTop"]),
                    fillcolor="rgba(34,197,94,0.16)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(248,250,252,0.12)",
                    line=dict(color="#F8FAFC", width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["ResistanceBarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["ResistanceDate"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )

            for name, price, color, dash in [
                ("Resistance Break", float(pattern["ResistanceZoneTop"]), "#F8FAFC", "solid"),
                ("Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    if (
        show_inverse_head_shoulders
        and inverse_head_shoulders is not None
        and not inverse_head_shoulders.empty
    ):
        ihs_color = "#38BDF8"
        neckline_color = "#E0F2FE"

        for pattern_number, (_, pattern) in enumerate(inverse_head_shoulders.iterrows()):
            point_indexes = []
            point_prices = []
            point_labels = []

            point_indexes.extend(
                [
                    int(pattern["LeftShoulderBarIdx"]),
                    int(pattern["Neckline1BarIdx"]),
                    int(pattern["HeadBarIdx"]),
                    int(pattern["Neckline2BarIdx"]),
                    int(pattern["RightShoulderBarIdx"]),
                    int(pattern["EntryBarIdx"]),
                ]
            )
            point_prices.extend(
                [
                    float(pattern["LeftShoulderPrice"]),
                    float(pattern["Neckline1Price"]),
                    float(pattern["HeadPrice"]),
                    float(pattern["Neckline2Price"]),
                    float(pattern["RightShoulderPrice"]),
                    float(pattern["EntryPrice"]),
                ]
            )
            point_labels.extend(["LS", "N1", "Head", "N2", "RS", "Breakout"])

            if (
                "FollowThroughBarIdx" in inverse_head_shoulders.columns
                and not pd.isna(pattern["FollowThroughBarIdx"])
                and int(pattern["FollowThroughBarIdx"]) != int(pattern["EntryBarIdx"])
            ):
                point_indexes.append(int(pattern["FollowThroughBarIdx"]))
                point_prices.append(float(pattern["FollowThroughPrice"]))
                point_labels.append("Follow")

            point_positions = [
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "bottom center",
                "top center",
                "top center",
            ][: len(point_labels)]
            point_x = (
                point_indexes
                if remove_time_gaps
                else [data.index[idx] for idx in point_indexes]
            )

            fig.add_trace(
                go.Scatter(
                    x=point_x,
                    y=point_prices,
                    mode="lines+markers+text",
                    name="Inverse H&S",
                    showlegend=pattern_number == 0,
                    line=dict(color=ihs_color, width=4),
                    marker=dict(
                        color=ihs_color,
                        size=10,
                        line=dict(width=1, color="white"),
                    ),
                    text=point_labels,
                    textposition=point_positions,
                    textfont=dict(color=ihs_color, size=11),
                    hovertext=(
                        f"Inverse H&S #{int(pattern['PatternId'])}<br>"
                        f"Shoulder diff: {float(pattern['ShoulderDifferencePct']):.2f}%<br>"
                        f"Head depth: {float(pattern['HeadDepthPct']):.2f}%<br>"
                        f"Neckline tilt: {float(pattern['NecklineTiltPct']):.2f}%<br>"
                        f"Downtrend: {float(pattern['DowntrendPct']):.2f}%<br>"
                        f"Last CHoCH: {pattern.get('LastCHoCHBeforePatternDirection', 'n/a')}<br>"
                        f"Start: {pattern.get('PatternStartDate', 'n/a')}<br>"
                        f"Symmetry: {float(pattern['SymmetryRatio']):.2f}<br>"
                        f"R2: {float(pattern['PatternR2']):.2f}<br>"
                        f"R/R: {float(pattern['RiskReward']):.2f}"
                    ),
                    hoverinfo="text",
                )
            )

            entry_idx = int(pattern["EntryBarIdx"])
            end_idx = min(len(data) - 1, entry_idx + 30)
            zone_x0 = (
                int(pattern["LeftShoulderBarIdx"])
                if remove_time_gaps
                else pattern["LeftShoulderDate"]
            )
            zone_x1 = entry_idx if remove_time_gaps else pattern["EntryDate"]
            if {
                "HeadSupportZoneBottom",
                "HeadSupportZoneTop",
                "ResistanceZoneBottom",
                "ResistanceZoneTop",
            }.issubset(inverse_head_shoulders.columns):
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["HeadSupportZoneBottom"]),
                    y1=float(pattern["HeadSupportZoneTop"]),
                    fillcolor="rgba(56,189,248,0.14)",
                    line=dict(color=ihs_color, width=1),
                    layer="below",
                )
                fig.add_shape(
                    type="rect",
                    x0=zone_x0,
                    x1=zone_x1,
                    y0=float(pattern["ResistanceZoneBottom"]),
                    y1=float(pattern["ResistanceZoneTop"]),
                    fillcolor="rgba(224,242,254,0.12)",
                    line=dict(color=neckline_color, width=1),
                    layer="below",
                )

            line_x = (
                [int(pattern["Neckline1BarIdx"]), end_idx]
                if remove_time_gaps
                else [pattern["Neckline1Date"], data.index[end_idx]]
            )
            projection_x = (
                [entry_idx, end_idx]
                if remove_time_gaps
                else [pattern["EntryDate"], data.index[end_idx]]
            )
            for name, price, color, dash in [
                (
                    "IHS Resistance Break",
                    float(pattern["ResistanceZoneTop"]),
                    neckline_color,
                    "solid",
                ),
                ("IHS Target", float(pattern["TargetPrice"]), "#14B8A6", "dash"),
                ("IHS Stop", float(pattern["StopPrice"]), "#EF4444", "dash"),
            ]:
                x_values = line_x if name == "IHS Resistance Break" else projection_x
                fig.add_trace(
                    go.Scatter(
                        x=x_values,
                        y=[price, price],
                        mode="lines+text",
                        showlegend=False,
                        line=dict(color=color, width=2, dash=dash),
                        text=[name, ""],
                        textposition="top center",
                        hoverinfo="x+y+text",
                    )
                )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=800,
        hovermode="closest",
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=55, t=80, b=55),
    )
    if remove_time_gaps:
        tick_count = min(12, len(data))
        tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
        fig.update_xaxes(
            title_text="Bars",
            tickmode="array",
            tickvals=tick_indexes,
            ticktext=date_labels[tick_indexes],
            tickangle=-30,
        )
    else:
        fig.update_xaxes(title_text="Date")
    return fig


if __name__ == "__main__":
    # Example:
    # pip install yahooquery pandas numpy plotly
    from yahooquery import Ticker

    symbol = "TUPRS.IS"
    raw = Ticker(symbol).history(period="2y", interval="1h")

    if isinstance(raw.index, pd.MultiIndex):
        raw = raw.reset_index(level="symbol", drop=True)

    raw_swings_df = detect_swing_points(
        raw,
        lookback=5,
        confirm_candles=3,
        min_swing_move_pct=0.50,
    )
    raw_structure_df = detect_bos_choch(
        raw,
        raw_swings_df,
        break_buffer_pct=0.03,
    )
    breakout_swings_df = build_breakout_swings(
        raw,
        raw_swings_df,
        structure_events=raw_structure_df,
        include_broken_swings=True,
        include_protected_swings=True,
        collapse_consecutive_same_type=False,
    )
    pattern_swings_df = build_pattern_swings(
        raw,
        breakout_swings_df,
        structure_events=raw_structure_df,
        atr_period=14,
        atr_multiplier=0.90,
        min_move_pct=0.65,
        preserve_structure_swings=True,
        always_keep_structure_swings=True,
        structure_min_move_pct=0.30,
    )
    structure_df = detect_bos_choch(
        raw,
        pattern_swings_df,
        break_buffer_pct=0.03,
    )
    double_bottoms_df = detect_double_bottoms(
        raw,
        pattern_swings_df,
        structure_events=structure_df,
        bottom_tolerance_pct=0.35,
        support_zone_pct=0.20,
        resistance_zone_pct=0.25,
        min_retrace_pct=0.38,
        max_retrace_pct=0.90,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        resistance_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_bearish_choch_before_b1=True,
        invalidate_second_peak_between_zones=True,
        require_direct_breakout_after_b2=True,
        require_breakout_hold_resistance=True,
        require_target_before_failure=False,
        target_wait_bars=None,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )
    inverse_head_shoulders_df = detect_inverse_head_shoulders(
        raw,
        pattern_swings_df,
        structure_events=structure_df,
        shoulder_tolerance_pct=3.0,
        min_head_depth_pct=0.75,
        head_lowest_tolerance_pct=0.15,
        head_support_zone_pct=0.25,
        neckline_zone_pct=0.25,
        max_neckline_tilt_pct=1.0,
        min_downtrend_pct=2.0,
        max_symmetry_ratio=2.5,
        neckline_break_buffer_pct=0.03,
        require_prior_lh_ll=True,
        require_last_choch_bearish=True,
        require_alternating_extrema_sequence=True,
        require_head_lowest=True,
        require_bullish_structure_break=True,
        invalidate_bearish_structure_after_right_shoulder=True,
        require_direct_breakout_after_right_shoulder=True,
        require_breakout_hold_neckline=True,
        require_target_before_failure=False,
        target_wait_bars=None,
        one_pattern_per_bearish_choch=True,
        max_pattern_overlap_pct=0.35,
        min_pattern_r2=None,
        require_volume_confirmation=False,
    )

    print(f"Raw swings: {len(raw_swings_df)}")
    print(f"Breakout swings: {len(breakout_swings_df)}")
    print(f"Pattern swings: {len(pattern_swings_df)}")
    print(pattern_swings_df.tail(20))
    print(structure_df.tail(20))
    print(double_bottoms_df.tail(20))
    print(inverse_head_shoulders_df.tail(20))

    figure = plot_swing_points(
        raw,
        pattern_swings_df,
        title=f"{symbol} Swing Points",
        structure_events=structure_df,
        show_only_structure_swings=False,
        double_bottoms=double_bottoms_df,
        inverse_head_shoulders=inverse_head_shoulders_df,
    )
    figure.show()
```
