```python
from __future__ import annotations

import argparse
import json
import math
import warnings
from dataclasses import asdict, dataclass
from datetime import datetime
from pathlib import Path
from typing import Any

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    from sklearn.base import clone
    from sklearn.decomposition import PCA
    from sklearn.ensemble import HistGradientBoostingClassifier, RandomForestClassifier
    from sklearn.impute import SimpleImputer
    from sklearn.inspection import permutation_importance
    from sklearn.linear_model import LogisticRegression
    from sklearn.metrics import (
        accuracy_score,
        average_precision_score,
        balanced_accuracy_score,
        brier_score_loss,
        precision_recall_curve,
        precision_score,
        recall_score,
        roc_auc_score,
    )
    from sklearn.model_selection import TimeSeriesSplit
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
    from sklearn.utils.class_weight import compute_sample_weight
except ModuleNotFoundError:
    clone = None
    PCA = None
    HistGradientBoostingClassifier = None
    RandomForestClassifier = None
    SimpleImputer = None
    permutation_importance = None
    LogisticRegression = None
    TimeSeriesSplit = None
    Pipeline = None
    StandardScaler = None
    compute_sample_weight = None
    accuracy_score = None
    average_precision_score = None
    balanced_accuracy_score = None
    brier_score_loss = None
    precision_recall_curve = None
    precision_score = None
    recall_score = None
    roc_auc_score = None

try:
    from hmmlearn.hmm import GaussianHMM
except ModuleNotFoundError:
    GaussianHMM = None

try:
    from statsmodels.tsa.arima.model import ARIMA
except ModuleNotFoundError:
    ARIMA = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


@dataclass
class ReversalMLConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    csv_path: str | None = None
    output_dir: str = "outputs/ml_reversal_runs"
    random_state: int = 42

    label_lookback: int = 80
    label_horizon: int = 40
    min_prior_drawdown_pct: float = 2.0
    max_distance_to_low_atr: float = 2.5
    target_atr: float = 2.0
    stop_atr: float = 1.2

    test_size_pct: float = 0.25
    n_splits: int = 5
    probability_threshold: float | None = None
    target_precision: float = 0.55
    cooldown_bars: int = 30

    model_names: tuple[str, ...] = ("boosting", "logistic", "random_forest")
    use_hmm: bool = True
    hmm_states: int = 4
    use_arima: bool = False
    arima_order: tuple[int, int, int] = (1, 0, 1)
    arima_window: int = 160
    use_pca: bool = False
    pca_variance: float = 0.95
    permutation_repeats: int = 10


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")
    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if "volume" not in out.columns:
        out["volume"] = 0.0
    out["volume"] = out["volume"].fillna(0.0).clip(lower=0.0)

    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")
    return out


INTRADAY_INTERVALS = {"1m", "2m", "5m", "15m", "30m", "60m", "90m", "1h"}


def fetch_ohlcv(symbol: str, period: str, interval: str) -> pd.DataFrame:
    if Ticker is None:
        raise ImportError(
            "yahooquery is required for online data fetch. "
            "Install with: pip install yahooquery"
        )

    errors: list[str] = []
    for candidate_period in data_period_attempts(period=period, interval=interval):
        try:
            raw = Ticker(symbol).history(period=candidate_period, interval=interval)
            frame = normalize_yahoo_history(raw, symbol=symbol)
            data = prepare_ohlcv(frame)
            if candidate_period != period:
                warnings.warn(
                    f"No usable data for period={period!r}, interval={interval!r}; "
                    f"using period={candidate_period!r} instead."
                )
            return data
        except Exception as exc:
            errors.append(f"{candidate_period}: {type(exc).__name__}: {exc}")

    raise ValueError(
        f"No data returned for {symbol} with interval={interval}. "
        "For Yahoo intraday data, use a shorter period such as 60d/30d/7d "
        "or a wider interval such as 1d. Attempts: "
        + " | ".join(errors)
    )


def data_period_attempts(*, period: str, interval: str) -> list[str]:
    attempts = [str(period)]
    if str(interval).lower() in INTRADAY_INTERVALS:
        attempts.extend(["60d", "30d", "7d", "5d", "1mo"])
    else:
        attempts.extend(["2y", "1y", "6mo", "3mo"])
    return list(dict.fromkeys(attempts))


def normalize_yahoo_history(raw: Any, *, symbol: str) -> pd.DataFrame:
    if raw is None:
        raise ValueError("empty response")
    if isinstance(raw, dict):
        message = raw.get(symbol, raw)
        raise ValueError(f"Yahoo returned an error response: {message}")
    if isinstance(raw, pd.Series):
        raise ValueError(f"Yahoo returned a Series instead of OHLCV data: {raw.to_dict()}")
    if not isinstance(raw, pd.DataFrame):
        raise ValueError(f"Unexpected Yahoo response type: {type(raw).__name__}")
    if raw.empty:
        raise ValueError("empty response")

    frame = raw.copy()
    if isinstance(frame.index, pd.MultiIndex):
        if "symbol" in frame.index.names:
            frame = frame.reset_index(level="symbol", drop=True)
        else:
            frame = frame.reset_index()
    return frame


def load_ohlcv(config: ReversalMLConfig) -> pd.DataFrame:
    if config.csv_path:
        return prepare_ohlcv(pd.read_csv(config.csv_path))
    return fetch_ohlcv(config.symbol, config.period, config.interval)


def make_feature_frame(data: pd.DataFrame, config: ReversalMLConfig) -> pd.DataFrame:
    close = data["close"].astype(float)
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    open_ = data["open"].astype(float)
    volume = data["volume"].astype(float)

    features = pd.DataFrame(index=data.index)
    log_close = np.log(close)
    log_return = log_close.diff()

    features["log_return_1"] = log_return
    for window in [2, 3, 5, 10, 20]:
        features[f"log_return_sum_{window}"] = log_return.rolling(window).sum()
        features[f"realized_vol_{window}"] = log_return.rolling(window).std()

    atr = average_true_range(data, period=14)
    features["atr"] = atr
    features["atr_pct"] = atr / close.replace(0, np.nan) * 100.0
    features["atr_pct_change_5"] = features["atr_pct"].pct_change(5)

    features["rsi_7"] = rsi(close, period=7)
    features["rsi_14"] = rsi(close, period=14)
    macd_line, macd_signal, macd_hist = macd(close)
    features["macd_line"] = macd_line
    features["macd_signal"] = macd_signal
    features["macd_hist"] = macd_hist
    features["stoch_k_14"] = stochastic_k(high, low, close, period=14)
    features["stoch_d_3"] = features["stoch_k_14"].rolling(3).mean()

    feature_windows = sorted({20, 50, 80, int(config.label_lookback)})
    for window in feature_windows:
        roll_high = high.rolling(window).max()
        roll_low = low.rolling(window).min()
        roll_mid = close.rolling(window).mean()
        roll_std = close.rolling(window).std()

        features[f"drawdown_from_high_{window}"] = (roll_high - close) / roll_high * 100.0
        features[f"rebound_from_low_{window}"] = (close - roll_low) / roll_low * 100.0
        features[f"dist_to_low_atr_{window}"] = (close - roll_low) / atr.replace(0, np.nan)
        features[f"dist_to_high_atr_{window}"] = (roll_high - close) / atr.replace(0, np.nan)
        features[f"bollinger_z_{window}"] = (close - roll_mid) / roll_std.replace(0, np.nan)
        features[f"range_pct_{window}"] = (roll_high - roll_low) / roll_mid.replace(0, np.nan) * 100.0
        features[f"slope_pct_{window}"] = rolling_slope_pct(close, window=window)

    candle_range = (high - low).replace(0, np.nan)
    features["body_pct"] = (close - open_) / open_.replace(0, np.nan) * 100.0
    features["body_to_range"] = (close - open_).abs() / candle_range
    features["upper_wick_to_range"] = (high - np.maximum(open_, close)) / candle_range
    features["lower_wick_to_range"] = (np.minimum(open_, close) - low) / candle_range
    features["close_position_in_range"] = (close - low) / candle_range
    features["gap_pct"] = (open_ - close.shift(1)) / close.shift(1).replace(0, np.nan) * 100.0
    features["green_streak"] = signed_streak(close >= open_)

    features["log_volume"] = np.log1p(volume)
    features["volume_z_20"] = rolling_zscore(features["log_volume"], window=20)
    features["volume_ratio_20"] = volume / volume.rolling(20).mean().replace(0, np.nan)
    obv = on_balance_volume(close, volume)
    features["obv_slope_20"] = rolling_slope_pct(obv, window=20)

    features["prior_down_context"] = (
        (features[f"drawdown_from_high_{config.label_lookback}"] >= config.min_prior_drawdown_pct)
        & (features[f"dist_to_low_atr_{config.label_lookback}"] <= config.max_distance_to_low_atr)
    ).astype(float)

    if config.use_arima:
        arima_features = walk_forward_arima_features(
            log_return,
            order=config.arima_order,
            window=config.arima_window,
        )
        features = features.join(arima_features)

    features = features.replace([np.inf, -np.inf], np.nan)
    return features


def make_reversal_labels(
    data: pd.DataFrame,
    features: pd.DataFrame,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    close = data["close"].to_numpy(dtype=float)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    atr = features["atr"].to_numpy(dtype=float)
    prior_drawdown = features[f"drawdown_from_high_{config.label_lookback}"].to_numpy(dtype=float)
    dist_to_low = features[f"dist_to_low_atr_{config.label_lookback}"].to_numpy(dtype=float)

    labels = pd.DataFrame(index=data.index)
    labels["reversal_label"] = np.nan
    labels["label_target_price"] = np.nan
    labels["label_stop_price"] = np.nan
    labels["label_hit_bar_idx"] = np.nan
    labels["label_hit_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["label_hit_type"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["prior_drawdown_pct"] = prior_drawdown
    labels["distance_to_low_atr"] = dist_to_low

    n = len(data)
    for bar_idx in range(n):
        if bar_idx + config.label_horizon >= n:
            continue
        if not np.isfinite(atr[bar_idx]) or atr[bar_idx] <= 0:
            continue

        target_price = close[bar_idx] + config.target_atr * atr[bar_idx]
        stop_price = close[bar_idx] - config.stop_atr * atr[bar_idx]
        labels.iloc[bar_idx, labels.columns.get_loc("label_target_price")] = target_price
        labels.iloc[bar_idx, labels.columns.get_loc("label_stop_price")] = stop_price

        in_reversal_context = (
            np.isfinite(prior_drawdown[bar_idx])
            and np.isfinite(dist_to_low[bar_idx])
            and prior_drawdown[bar_idx] >= config.min_prior_drawdown_pct
            and dist_to_low[bar_idx] <= config.max_distance_to_low_atr
        )
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=bar_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        label = int(in_reversal_context and hit_type == "target")
        labels.iloc[bar_idx, labels.columns.get_loc("reversal_label")] = label
        if hit_idx is not None:
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_bar_idx")] = int(hit_idx)
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_date")] = data.index[int(hit_idx)]
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_type")] = hit_type

    return labels


def first_long_barrier_hit(
    *,
    high: np.ndarray,
    low: np.ndarray,
    start_idx: int,
    horizon: int,
    target_price: float,
    stop_price: float,
) -> tuple[str, int | None]:
    max_idx = min(len(high) - 1, int(start_idx) + int(horizon))
    for idx in range(int(start_idx) + 1, max_idx + 1):
        stop_hit = low[idx] <= stop_price
        target_hit = high[idx] >= target_price
        if stop_hit and target_hit:
            return "stop", idx
        if stop_hit:
            return "stop", idx
        if target_hit:
            return "target", idx
    return "timeout", None


def add_hmm_features(
    features: pd.DataFrame,
    train_index: pd.Index,
    config: ReversalMLConfig,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    state_summary_columns = ["state", "train_count", "mean_return", "mean_vol", "mean_drawdown"]
    if not config.use_hmm:
        return features, pd.DataFrame(columns=state_summary_columns)
    if GaussianHMM is None or SimpleImputer is None or StandardScaler is None:
        warnings.warn("hmmlearn/sklearn unavailable; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    hmm_columns = [
        column
        for column in [
            "log_return_1",
            "realized_vol_10",
            "atr_pct",
            "volume_z_20",
            f"drawdown_from_high_{config.label_lookback}",
            f"dist_to_low_atr_{config.label_lookback}",
        ]
        if column in features.columns
    ]
    if len(hmm_columns) < 3:
        return features, pd.DataFrame(columns=state_summary_columns)

    train = features.loc[train_index, hmm_columns]
    if train.dropna(how="all").shape[0] < config.hmm_states * 20:
        warnings.warn("Not enough training rows for HMM; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    imputer = SimpleImputer(strategy="median")
    scaler = StandardScaler()
    train_values = scaler.fit_transform(imputer.fit_transform(train))
    all_values = scaler.transform(imputer.transform(features[hmm_columns]))

    hmm = GaussianHMM(
        n_components=config.hmm_states,
        covariance_type="diag",
        n_iter=200,
        tol=1e-4,
        random_state=config.random_state,
    )
    hmm.fit(train_values)
    probabilities = hmm.predict_proba(all_values)
    states = probabilities.argmax(axis=1)

    out = features.copy()
    out["hmm_state"] = states.astype(float)
    for state in range(config.hmm_states):
        out[f"hmm_state_prob_{state}"] = probabilities[:, state]

    state_rows: list[dict[str, Any]] = []
    train_augmented = out.loc[train_index].copy()
    for state in range(config.hmm_states):
        subset = train_augmented[train_augmented["hmm_state"] == state]
        state_rows.append(
            {
                "state": state,
                "train_count": int(len(subset)),
                "mean_return": float(subset["log_return_1"].mean()),
                "mean_vol": float(subset["realized_vol_10"].mean()),
                "mean_drawdown": float(
                    subset[f"drawdown_from_high_{config.label_lookback}"].mean()
                ),
            }
        )
    return out, pd.DataFrame(state_rows, columns=state_summary_columns)


def train_reversal_models(
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    config: ReversalMLConfig,
) -> dict[str, Any]:
    require_sklearn()

    labeled = features.join(labels[["reversal_label"]], how="inner")
    labeled = labeled.dropna(subset=["reversal_label"])
    if len(labeled) < 200:
        raise ValueError("Not enough labeled rows. Increase period or use a higher timeframe.")

    y = labeled["reversal_label"].astype(int)
    feature_columns = [
        column
        for column in labeled.columns
        if column != "reversal_label" and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]

    split_idx = int(len(labeled) * (1.0 - config.test_size_pct))
    split_idx = min(max(split_idx, 50), len(labeled) - 20)
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]

    features_with_hmm, hmm_summary = add_hmm_features(features, train_index, config)
    labeled = features_with_hmm.join(labels[["reversal_label"]], how="inner")
    labeled = labeled.dropna(subset=["reversal_label"])
    y = labeled["reversal_label"].astype(int)
    feature_columns = [
        column
        for column in labeled.columns
        if column != "reversal_label" and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]
    X_train = X.loc[train_index]
    y_train = y.loc[train_index]
    X_test = X.loc[test_index]
    y_test = y.loc[test_index]

    X_train_clip, clip_bounds = fit_winsorize(X_train)
    X_test_clip = apply_winsorize(X_test, clip_bounds)
    X_all_clip = apply_winsorize(features_with_hmm[feature_columns], clip_bounds)

    models = build_models(config)
    results: list[dict[str, Any]] = []
    fitted_models: dict[str, Pipeline] = {}

    for name, model in models.items():
        fitted = fit_pipeline_with_weights(model, X_train_clip, y_train)
        fitted_models[name] = fitted
        test_probability = positive_probability(fitted, X_test_clip)
        metrics = evaluate_probabilities(y_test, test_probability)
        cv_metrics = time_series_cv_score(
            model,
            X_train_clip,
            y_train,
            n_splits=config.n_splits,
        )
        result = {"model": name, **metrics, **cv_metrics}
        results.append(result)

    model_report = pd.DataFrame(results)
    best_name = choose_best_model(model_report)
    best_model = fitted_models[best_name]
    all_probabilities = positive_probability(best_model, X_all_clip)

    threshold = config.probability_threshold
    if threshold is None:
        threshold = choose_probability_threshold(
            y_test,
            positive_probability(best_model, X_test_clip),
            target_precision=config.target_precision,
        )

    importance = permutation_xai(
        best_model,
        X_test_clip,
        y_test,
        feature_columns,
        repeats=config.permutation_repeats,
        random_state=config.random_state,
    )
    pca_loadings = extract_pca_loadings(best_model, feature_columns)
    signals = build_signal_table(
        data=data,
        features=features_with_hmm,
        labels=labels,
        probabilities=pd.Series(all_probabilities, index=features_with_hmm.index),
        feature_columns=feature_columns,
        feature_importance=importance,
        train_features=X_train_clip,
        threshold=threshold,
        config=config,
    )

    return {
        "features": features_with_hmm,
        "labels": labels,
        "feature_columns": feature_columns,
        "model_report": model_report,
        "best_model_name": best_name,
        "best_model": best_model,
        "probabilities": pd.Series(all_probabilities, index=features_with_hmm.index, name="reversal_probability"),
        "threshold": threshold,
        "signals": signals,
        "feature_importance": importance,
        "pca_loadings": pca_loadings,
        "hmm_state_summary": hmm_summary,
    }


def require_sklearn() -> None:
    if Pipeline is None:
        raise ImportError(
            "scikit-learn is required for ML training. Install with: pip install scikit-learn"
        )


def build_models(config: ReversalMLConfig) -> dict[str, Pipeline]:
    requested = {name.strip().lower() for name in config.model_names}
    models: dict[str, Pipeline] = {}

    if "boosting" in requested:
        models["boosting"] = make_pipeline(
            HistGradientBoostingClassifier(
                learning_rate=0.05,
                max_iter=350,
                max_leaf_nodes=15,
                l2_regularization=0.05,
                early_stopping=True,
                random_state=config.random_state,
            ),
            config,
            scale=False,
        )
    if "logistic" in requested:
        models["logistic"] = make_pipeline(
            LogisticRegression(
                max_iter=2500,
                class_weight="balanced",
                solver="lbfgs",
                random_state=config.random_state,
            ),
            config,
            scale=True,
        )
    if "random_forest" in requested:
        models["random_forest"] = make_pipeline(
            RandomForestClassifier(
                n_estimators=400,
                max_depth=7,
                min_samples_leaf=20,
                class_weight="balanced_subsample",
                random_state=config.random_state,
                n_jobs=-1,
            ),
            config,
            scale=False,
        )
    if not models:
        raise ValueError("No valid model names requested.")
    return models


def make_pipeline(estimator: Any, config: ReversalMLConfig, *, scale: bool) -> Pipeline:
    steps: list[tuple[str, Any]] = [("imputer", SimpleImputer(strategy="median"))]
    if scale or config.use_pca:
        steps.append(("scaler", StandardScaler()))
    if config.use_pca:
        steps.append(("pca", PCA(n_components=config.pca_variance, svd_solver="full")))
    steps.append(("model", estimator))
    return Pipeline(steps)


def fit_pipeline_with_weights(model: Pipeline, X: pd.DataFrame, y: pd.Series) -> Pipeline:
    sample_weight = compute_sample_weight(class_weight="balanced", y=y)
    try:
        return model.fit(X, y, model__sample_weight=sample_weight)
    except TypeError:
        return model.fit(X, y)


def positive_probability(model: Pipeline, X: pd.DataFrame) -> np.ndarray:
    if hasattr(model, "predict_proba"):
        probability = model.predict_proba(X)
        if probability.shape[1] == 1:
            return np.zeros(len(X), dtype=float)
        return probability[:, 1]
    decision = model.decision_function(X)
    return 1.0 / (1.0 + np.exp(-decision))


def evaluate_probabilities(y_true: pd.Series, probability: np.ndarray) -> dict[str, float]:
    out: dict[str, float] = {
        "test_average_precision": np.nan,
        "test_roc_auc": np.nan,
        "test_brier": np.nan,
        "test_balanced_accuracy_050": np.nan,
        "test_precision_050": np.nan,
        "test_recall_050": np.nan,
        "test_positive_rate": float(np.mean(y_true)),
    }
    if len(np.unique(y_true)) < 2:
        return out

    prediction = (probability >= 0.5).astype(int)
    out["test_average_precision"] = float(average_precision_score(y_true, probability))
    out["test_roc_auc"] = float(roc_auc_score(y_true, probability))
    out["test_brier"] = float(brier_score_loss(y_true, probability))
    out["test_balanced_accuracy_050"] = float(balanced_accuracy_score(y_true, prediction))
    out["test_precision_050"] = float(precision_score(y_true, prediction, zero_division=0))
    out["test_recall_050"] = float(recall_score(y_true, prediction, zero_division=0))
    out["test_accuracy_050"] = float(accuracy_score(y_true, prediction))
    return out


def time_series_cv_score(
    model: Pipeline,
    X: pd.DataFrame,
    y: pd.Series,
    *,
    n_splits: int,
) -> dict[str, float]:
    out = {"cv_average_precision": np.nan, "cv_roc_auc": np.nan}
    if len(X) < 200 or y.nunique() < 2:
        return out

    split_count = min(n_splits, max(2, len(X) // 200))
    splitter = TimeSeriesSplit(n_splits=split_count)
    ap_scores: list[float] = []
    auc_scores: list[float] = []
    for train_idx, test_idx in splitter.split(X):
        y_train = y.iloc[train_idx]
        y_valid = y.iloc[test_idx]
        if y_train.nunique() < 2 or y_valid.nunique() < 2:
            continue
        fitted = fit_pipeline_with_weights(clone(model), X.iloc[train_idx], y_train)
        prob = positive_probability(fitted, X.iloc[test_idx])
        ap_scores.append(float(average_precision_score(y_valid, prob)))
        auc_scores.append(float(roc_auc_score(y_valid, prob)))

    if ap_scores:
        out["cv_average_precision"] = float(np.mean(ap_scores))
    if auc_scores:
        out["cv_roc_auc"] = float(np.mean(auc_scores))
    return out


def choose_best_model(model_report: pd.DataFrame) -> str:
    ranking = model_report.copy()
    ranking["_rank_score"] = ranking["test_average_precision"].fillna(-1.0)
    ranking = ranking.sort_values(["_rank_score", "test_roc_auc"], ascending=False)
    return str(ranking.iloc[0]["model"])


def choose_probability_threshold(
    y_true: pd.Series,
    probability: np.ndarray,
    *,
    target_precision: float,
) -> float:
    if len(np.unique(y_true)) < 2:
        return 0.65

    precision, recall, thresholds = precision_recall_curve(y_true, probability)
    candidates: list[tuple[float, float, float]] = []
    for threshold, prec, rec in zip(thresholds, precision[:-1], recall[:-1]):
        if prec >= target_precision and rec > 0:
            candidates.append((float(threshold), float(prec), float(rec)))
    if not candidates:
        return float(np.quantile(probability, 0.90))
    candidates.sort(key=lambda item: (item[2], item[1]), reverse=True)
    return float(candidates[0][0])


def permutation_xai(
    model: Pipeline,
    X_test: pd.DataFrame,
    y_test: pd.Series,
    feature_columns: list[str],
    *,
    repeats: int,
    random_state: int,
) -> pd.DataFrame:
    columns = ["feature", "importance_mean", "importance_std", "rank"]
    if y_test.nunique() < 2 or permutation_importance is None:
        return pd.DataFrame(columns=columns)

    result = permutation_importance(
        model,
        X_test,
        y_test,
        n_repeats=repeats,
        random_state=random_state,
        scoring="average_precision",
        n_jobs=-1,
    )
    importance = pd.DataFrame(
        {
            "feature": feature_columns,
            "importance_mean": result.importances_mean,
            "importance_std": result.importances_std,
        }
    )
    importance = importance.sort_values("importance_mean", ascending=False).reset_index(drop=True)
    importance["rank"] = np.arange(1, len(importance) + 1)
    return importance[columns]


def extract_pca_loadings(model: Pipeline, feature_columns: list[str]) -> pd.DataFrame:
    if "pca" not in model.named_steps:
        return pd.DataFrame(columns=["component", "feature", "loading", "abs_loading", "rank"])

    pca = model.named_steps["pca"]
    rows: list[dict[str, Any]] = []
    for component_idx, component in enumerate(pca.components_):
        for feature, loading in zip(feature_columns, component):
            rows.append(
                {
                    "component": f"PC{component_idx + 1}",
                    "feature": feature,
                    "loading": float(loading),
                    "abs_loading": float(abs(loading)),
                }
            )
    out = pd.DataFrame(rows)
    out["rank"] = out.groupby("component")["abs_loading"].rank(ascending=False, method="first")
    return out.sort_values(["component", "rank"]).reset_index(drop=True)


def build_signal_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    candidate_probability = probabilities.dropna()
    candidate_probability = candidate_probability[candidate_probability >= threshold]
    if candidate_probability.empty:
        return empty_signal_frame()

    selected_indexes = non_max_suppression_1d(
        candidate_probability,
        cooldown_bars=config.cooldown_bars,
        all_index=features.index,
    )

    train_stats = make_train_feature_stats(train_features)
    rows: list[dict[str, Any]] = []
    atr = features["atr"]
    for timestamp in selected_indexes:
        bar_idx = int(data.index.get_loc(timestamp))
        entry_price = float(data.loc[timestamp, "close"])
        target_price = entry_price + config.target_atr * float(atr.loc[timestamp])
        stop_price = entry_price - config.stop_atr * float(atr.loc[timestamp])
        label_row = labels.loc[timestamp] if timestamp in labels.index else pd.Series(dtype=object)
        hit_idx = label_row.get("label_hit_bar_idx", np.nan)
        hit_date = label_row.get("label_hit_date", pd.NaT)
        hit_type = label_row.get("label_hit_type", None)
        realized_label = label_row.get("reversal_label", np.nan)
        top_drivers = local_driver_summary(
            features.loc[timestamp, feature_columns],
            feature_importance,
            train_stats,
            top_n=6,
        )
        rows.append(
            {
                "EntryBarIdx": bar_idx,
                "EntryDate": timestamp,
                "EntryPrice": entry_price,
                "ReversalProbability": float(probabilities.loc[timestamp]),
                "SignalThreshold": float(threshold),
                "TargetPrice": target_price,
                "StopPrice": stop_price,
                "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                "OutcomeKnown": bool(np.isfinite(realized_label)),
                "OutcomeLabel": realized_label,
                "HitBarIdx": int(hit_idx) if np.isfinite(hit_idx) else np.nan,
                "HitDate": hit_date,
                "HitType": hit_type,
                "ShapeHint": classify_shape_hint(data, bar_idx),
                "TopDrivers": top_drivers,
            }
        )
    return pd.DataFrame(rows).sort_values("EntryBarIdx").reset_index(drop=True)


def empty_signal_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "ReversalProbability",
            "SignalThreshold",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def non_max_suppression_1d(
    probability: pd.Series,
    *,
    cooldown_bars: int,
    all_index: pd.Index,
) -> list[pd.Timestamp]:
    ranked = probability.sort_values(ascending=False)
    selected_positions: list[int] = []
    selected_timestamps: list[pd.Timestamp] = []

    for timestamp, _ in ranked.items():
        position = int(all_index.get_loc(timestamp))
        if any(abs(position - selected) <= cooldown_bars for selected in selected_positions):
            continue
        selected_positions.append(position)
        selected_timestamps.append(timestamp)

    selected_timestamps.sort(key=lambda ts: int(all_index.get_loc(ts)))
    return selected_timestamps


def make_train_feature_stats(train_features: pd.DataFrame) -> pd.DataFrame:
    stats = pd.DataFrame(index=train_features.columns)
    stats["median"] = train_features.median(numeric_only=True)
    q25 = train_features.quantile(0.25, numeric_only=True)
    q75 = train_features.quantile(0.75, numeric_only=True)
    stats["iqr"] = (q75 - q25).replace(0, np.nan)
    return stats


def local_driver_summary(
    row: pd.Series,
    feature_importance: pd.DataFrame,
    train_stats: pd.DataFrame,
    *,
    top_n: int,
) -> str:
    if feature_importance.empty:
        return ""

    importance_map = feature_importance.set_index("feature")["importance_mean"].to_dict()
    scored: list[tuple[float, str]] = []
    for feature, value in row.items():
        if feature not in importance_map or feature not in train_stats.index:
            continue
        if not np.isfinite(value):
            continue
        iqr = train_stats.loc[feature, "iqr"]
        median = train_stats.loc[feature, "median"]
        if not np.isfinite(iqr) or iqr == 0:
            continue
        robust_z = (float(value) - float(median)) / float(iqr)
        score = abs(float(importance_map[feature]) * robust_z)
        if score <= 0:
            continue
        scored.append((score, f"{feature}={float(value):.4g} z={robust_z:.2f}"))

    scored.sort(reverse=True)
    return " | ".join(text for _, text in scored[:top_n])


def classify_shape_hint(data: pd.DataFrame, bar_idx: int, *, lookback: int = 80) -> str:
    start = max(0, int(bar_idx) - lookback)
    if int(bar_idx) - start < 20:
        return "ML Bullish Reversal"

    window = data.iloc[start : int(bar_idx) + 1]
    lows = window["low"].to_numpy(dtype=float)
    highs = window["high"].to_numpy(dtype=float)
    close = window["close"].to_numpy(dtype=float)
    atr = float(average_true_range(window, period=14).iloc[-1])
    if not np.isfinite(atr) or atr <= 0:
        atr = max(float(np.nanstd(close)), 1e-12)

    pivot_lows = simple_pivots(lows, mode="low", order=3)
    pivot_highs = simple_pivots(highs, mode="high", order=3)
    if len(pivot_lows) >= 2:
        l1, l2 = pivot_lows[-2], pivot_lows[-1]
        low_diff_atr = abs(lows[l2] - lows[l1]) / atr
        if l2 > l1 and low_diff_atr <= 1.0:
            between_high = np.max(highs[l1:l2 + 1])
            if between_high - max(lows[l1], lows[l2]) >= atr:
                return "ML Double Bottom-like"

    if len(pivot_lows) >= 3 and len(pivot_highs) >= 2:
        left, head, right = pivot_lows[-3], pivot_lows[-2], pivot_lows[-1]
        if left < head < right and lows[head] < min(lows[left], lows[right]):
            shoulder_diff_atr = abs(lows[left] - lows[right]) / atr
            if shoulder_diff_atr <= 1.5:
                return "ML Inverse H&S-like"

    return "ML Bullish Reversal"


def simple_pivots(values: np.ndarray, *, mode: str, order: int) -> list[int]:
    pivots: list[int] = []
    for idx in range(order, len(values) - order):
        left = values[idx - order : idx]
        right = values[idx + 1 : idx + order + 1]
        if mode == "low" and values[idx] <= np.min(left) and values[idx] <= np.min(right):
            pivots.append(idx)
        if mode == "high" and values[idx] >= np.max(left) and values[idx] >= np.max(right):
            pivots.append(idx)
    return pivots


def fit_winsorize(
    X_train: pd.DataFrame,
    *,
    lower: float = 0.01,
    upper: float = 0.99,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    bounds = pd.DataFrame(index=X_train.columns)
    bounds["lower"] = X_train.quantile(lower, numeric_only=True)
    bounds["upper"] = X_train.quantile(upper, numeric_only=True)
    return apply_winsorize(X_train, bounds), bounds


def apply_winsorize(X: pd.DataFrame, bounds: pd.DataFrame) -> pd.DataFrame:
    out = X.copy()
    for column in out.columns:
        if column not in bounds.index:
            continue
        lower = bounds.loc[column, "lower"]
        upper = bounds.loc[column, "upper"]
        if np.isfinite(lower) and np.isfinite(upper) and lower < upper:
            out[column] = out[column].clip(lower=lower, upper=upper)
    return out


def average_true_range(data: pd.DataFrame, *, period: int = 14) -> pd.Series:
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    close = data["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [
            high - low,
            (high - previous_close).abs(),
            (low - previous_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    return true_range.rolling(period, min_periods=1).mean()


def rsi(close: pd.Series, *, period: int = 14) -> pd.Series:
    delta = close.diff()
    gains = delta.clip(lower=0.0)
    losses = -delta.clip(upper=0.0)
    avg_gain = gains.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    avg_loss = losses.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    rs = avg_gain / avg_loss.replace(0, np.nan)
    return 100.0 - (100.0 / (1.0 + rs))


def macd(close: pd.Series) -> tuple[pd.Series, pd.Series, pd.Series]:
    fast = close.ewm(span=12, adjust=False).mean()
    slow = close.ewm(span=26, adjust=False).mean()
    line = fast - slow
    signal = line.ewm(span=9, adjust=False).mean()
    return line, signal, line - signal


def stochastic_k(
    high: pd.Series,
    low: pd.Series,
    close: pd.Series,
    *,
    period: int,
) -> pd.Series:
    lowest = low.rolling(period).min()
    highest = high.rolling(period).max()
    return (close - lowest) / (highest - lowest).replace(0, np.nan) * 100.0


def rolling_zscore(series: pd.Series, *, window: int) -> pd.Series:
    mean = series.rolling(window).mean()
    std = series.rolling(window).std()
    return (series - mean) / std.replace(0, np.nan)


def rolling_slope_pct(series: pd.Series, *, window: int) -> pd.Series:
    x = np.arange(window, dtype=float)
    x = x - x.mean()

    def slope(values: np.ndarray) -> float:
        if np.isnan(values).any():
            return np.nan
        y = values.astype(float)
        y = y - y.mean()
        denom = float(np.dot(x, x))
        if denom == 0:
            return np.nan
        raw_slope = float(np.dot(x, y) / denom)
        return raw_slope / max(abs(float(values[-1])), 1e-12) * 100.0

    return series.rolling(window).apply(slope, raw=True)


def signed_streak(condition: pd.Series) -> pd.Series:
    values = condition.fillna(False).to_numpy(dtype=bool)
    streaks = np.zeros(len(values), dtype=float)
    current = 0
    previous = None
    for idx, value in enumerate(values):
        if previous is None or value != previous:
            current = 1
        else:
            current += 1
        streaks[idx] = current if value else -current
        previous = value
    return pd.Series(streaks, index=condition.index)


def on_balance_volume(close: pd.Series, volume: pd.Series) -> pd.Series:
    direction = np.sign(close.diff()).fillna(0.0)
    return (direction * volume).cumsum()


def walk_forward_arima_features(
    returns: pd.Series,
    *,
    order: tuple[int, int, int],
    window: int,
) -> pd.DataFrame:
    columns = ["arima_forecast_return", "arima_residual", "arima_residual_z_40"]
    out = pd.DataFrame(np.nan, index=returns.index, columns=columns)
    if ARIMA is None:
        warnings.warn("statsmodels unavailable; ARIMA features skipped.")
        return out

    values = returns.astype(float)
    for idx in range(window, len(values)):
        history = values.iloc[idx - window : idx].dropna()
        if len(history) < max(30, sum(order) * 10):
            continue
        try:
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                result = ARIMA(history, order=order).fit()
            forecast = float(result.forecast(1).iloc[0])
            residual = float(values.iloc[idx] - forecast)
            out.iloc[idx, out.columns.get_loc("arima_forecast_return")] = forecast
            out.iloc[idx, out.columns.get_loc("arima_residual")] = residual
        except Exception:
            continue
    out["arima_residual_z_40"] = rolling_zscore(out["arima_residual"], window=40)
    return out


def save_outputs(
    data: pd.DataFrame,
    result: dict[str, Any],
    config: ReversalMLConfig,
) -> Path:
    run_name = f"{config.symbol.replace('.', '_')}_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
    run_dir = Path(config.output_dir) / run_name
    run_dir.mkdir(parents=True, exist_ok=True)

    result["signals"].to_csv(run_dir / "signals.csv", index=False)
    result["model_report"].to_csv(run_dir / "model_report.csv", index=False)
    result["feature_importance"].to_csv(run_dir / "feature_importance.csv", index=False)
    result["hmm_state_summary"].to_csv(run_dir / "hmm_state_summary.csv", index=False)
    result["pca_loadings"].to_csv(run_dir / "pca_loadings.csv", index=False)

    probability_frame = pd.DataFrame(
        {
            "Date": result["probabilities"].index,
            "ReversalProbability": result["probabilities"].to_numpy(dtype=float),
        }
    )
    probability_frame.to_csv(run_dir / "probabilities.csv", index=False)

    with open(run_dir / "config.json", "w", encoding="utf-8") as handle:
        json.dump(
            {
                **asdict(config),
                "best_model_name": result["best_model_name"],
                "threshold": result["threshold"],
            },
            handle,
            indent=2,
            default=str,
        )

    plot_path = run_dir / "reversal_ml_plot.html"
    make_plot(
        data=data,
        probabilities=result["probabilities"],
        signals=result["signals"],
        title=f"{config.symbol} ML Reversal Signals",
        output_path=plot_path,
    )
    return run_dir


def make_plot(
    *,
    data: pd.DataFrame,
    probabilities: pd.Series,
    signals: pd.DataFrame,
    title: str,
    output_path: Path,
) -> None:
    if go is None:
        return

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probabilities.reindex(data.index).to_numpy(dtype=float),
            name="Reversal probability",
            mode="lines",
            yaxis="y2",
            line=dict(color="#38BDF8", width=2),
        )
    )

    if not signals.empty:
        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers+text",
                name="ML Buy Signal",
                marker=dict(color="#FACC15", size=12, symbol="circle"),
                text=[f"{prob:.2f}" for prob in signals["ReversalProbability"]],
                textposition="top center",
                hovertext=[
                    (
                        f"ML Buy<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Probability: {row['ReversalProbability']:.3f}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Probability",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=65, t=80, b=55),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def run(config: ReversalMLConfig) -> Path:
    data = load_ohlcv(config)
    features = make_feature_frame(data, config)
    labels = make_reversal_labels(data, features, config)
    result = train_reversal_models(data, features, labels, config)
    run_dir = save_outputs(data, result, config)

    print(f"Saved run: {run_dir}")
    print(f"Best model: {result['best_model_name']}")
    print(f"Threshold: {result['threshold']:.4f}")
    print(f"Signals: {len(result['signals'])}")
    if not result["model_report"].empty:
        print(result["model_report"].to_string(index=False))
    if not result["signals"].empty:
        print(result["signals"].tail(10).to_string(index=False))
    return run_dir


def parse_args(argv: list[str] | None = None) -> ReversalMLConfig:
    parser = argparse.ArgumentParser(description="ML bullish reversal detector with XAI outputs.")
    parser.add_argument("--symbol", default="THYAO.IS")
    parser.add_argument("--period", default="60d")
    parser.add_argument("--interval", default="5m")
    parser.add_argument("--csv", dest="csv_path", default=None)
    parser.add_argument("--output-dir", default="outputs/ml_reversal_runs")
    parser.add_argument("--label-lookback", type=int, default=80)
    parser.add_argument("--label-horizon", type=int, default=40)
    parser.add_argument("--min-prior-drawdown-pct", type=float, default=2.0)
    parser.add_argument("--max-distance-to-low-atr", type=float, default=2.5)
    parser.add_argument("--target-atr", type=float, default=2.0)
    parser.add_argument("--stop-atr", type=float, default=1.2)
    parser.add_argument("--probability-threshold", type=float, default=None)
    parser.add_argument("--target-precision", type=float, default=0.55)
    parser.add_argument("--cooldown-bars", type=int, default=30)
    parser.add_argument("--models", default="boosting,logistic,random_forest")
    parser.add_argument("--use-hmm", action=argparse.BooleanOptionalAction, default=True)
    parser.add_argument("--hmm-states", type=int, default=4)
    parser.add_argument("--use-arima", action=argparse.BooleanOptionalAction, default=False)
    parser.add_argument("--arima-order", default="1,0,1")
    parser.add_argument("--arima-window", type=int, default=160)
    parser.add_argument("--use-pca", action=argparse.BooleanOptionalAction, default=False)
    parser.add_argument("--pca-variance", type=float, default=0.95)
    parser.add_argument("--random-state", type=int, default=42)
    args, unknown_args = parser.parse_known_args(argv)
    if unknown_args:
        warnings.warn(
            "Ignoring unknown CLI arguments, usually injected by notebooks: "
            + " ".join(unknown_args)
        )

    arima_order = tuple(int(part.strip()) for part in args.arima_order.split(","))
    if len(arima_order) != 3:
        raise ValueError("--arima-order must contain three comma-separated integers, e.g. 1,0,1")

    return ReversalMLConfig(
        symbol=args.symbol,
        period=args.period,
        interval=args.interval,
        csv_path=args.csv_path,
        output_dir=args.output_dir,
        random_state=args.random_state,
        label_lookback=args.label_lookback,
        label_horizon=args.label_horizon,
        min_prior_drawdown_pct=args.min_prior_drawdown_pct,
        max_distance_to_low_atr=args.max_distance_to_low_atr,
        target_atr=args.target_atr,
        stop_atr=args.stop_atr,
        probability_threshold=args.probability_threshold,
        target_precision=args.target_precision,
        cooldown_bars=args.cooldown_bars,
        model_names=tuple(part.strip() for part in args.models.split(",") if part.strip()),
        use_hmm=args.use_hmm,
        hmm_states=args.hmm_states,
        use_arima=args.use_arima,
        arima_order=arima_order,
        arima_window=args.arima_window,
        use_pca=args.use_pca,
        pca_variance=args.pca_variance,
    )


if __name__ == "__main__":
    run(parse_args())
```

```python
from __future__ import annotations

import argparse
import json
import math
import warnings
from dataclasses import asdict, dataclass
from datetime import datetime
from pathlib import Path
from typing import Any

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    from sklearn.base import clone
    from sklearn.decomposition import PCA
    from sklearn.ensemble import HistGradientBoostingClassifier, RandomForestClassifier
    from sklearn.impute import SimpleImputer
    from sklearn.inspection import permutation_importance
    from sklearn.linear_model import LogisticRegression
    from sklearn.metrics import (
        accuracy_score,
        average_precision_score,
        balanced_accuracy_score,
        brier_score_loss,
        precision_recall_curve,
        precision_score,
        recall_score,
        roc_auc_score,
    )
    from sklearn.model_selection import TimeSeriesSplit
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
    from sklearn.utils.class_weight import compute_sample_weight
except ModuleNotFoundError:
    clone = None
    PCA = None
    HistGradientBoostingClassifier = None
    RandomForestClassifier = None
    SimpleImputer = None
    permutation_importance = None
    LogisticRegression = None
    TimeSeriesSplit = None
    Pipeline = None
    StandardScaler = None
    compute_sample_weight = None
    accuracy_score = None
    average_precision_score = None
    balanced_accuracy_score = None
    brier_score_loss = None
    precision_recall_curve = None
    precision_score = None
    recall_score = None
    roc_auc_score = None

try:
    from hmmlearn.hmm import GaussianHMM
except ModuleNotFoundError:
    GaussianHMM = None

try:
    from statsmodels.tsa.arima.model import ARIMA
except ModuleNotFoundError:
    ARIMA = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


@dataclass
class ReversalMLConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    csv_path: str | None = None
    output_dir: str = "outputs/ml_reversal_runs"
    random_state: int = 42

    label_lookback: int = 80
    label_horizon: int = 40
    min_prior_drawdown_pct: float = 2.0
    max_distance_to_low_atr: float = 2.5
    target_atr: float = 2.0
    stop_atr: float = 1.2

    test_size_pct: float = 0.25
    n_splits: int = 5
    probability_threshold: float | None = None
    target_precision: float = 0.55
    cooldown_bars: int = 30

    model_names: tuple[str, ...] = ("boosting", "logistic", "random_forest")
    use_hmm: bool = True
    hmm_states: int = 4
    use_arima: bool = False
    arima_order: tuple[int, int, int] = (1, 0, 1)
    arima_window: int = 160
    use_pca: bool = False
    pca_variance: float = 0.95
    permutation_repeats: int = 10


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")
    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if "volume" not in out.columns:
        out["volume"] = 0.0
    out["volume"] = out["volume"].fillna(0.0).clip(lower=0.0)

    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")
    return out


INTRADAY_INTERVALS = {"1m", "2m", "5m", "15m", "30m", "60m", "90m", "1h"}


def fetch_ohlcv(symbol: str, period: str, interval: str) -> pd.DataFrame:
    if Ticker is None:
        raise ImportError(
            "yahooquery is required for online data fetch. "
            "Install with: pip install yahooquery"
        )

    errors: list[str] = []
    for candidate_period in data_period_attempts(period=period, interval=interval):
        try:
            raw = Ticker(symbol).history(period=candidate_period, interval=interval)
            frame = normalize_yahoo_history(raw, symbol=symbol)
            data = prepare_ohlcv(frame)
            if candidate_period != period:
                warnings.warn(
                    f"No usable data for period={period!r}, interval={interval!r}; "
                    f"using period={candidate_period!r} instead."
                )
            return data
        except Exception as exc:
            errors.append(f"{candidate_period}: {type(exc).__name__}: {exc}")

    raise ValueError(
        f"No data returned for {symbol} with interval={interval}. "
        "For Yahoo intraday data, use a shorter period such as 60d/30d/7d "
        "or a wider interval such as 1d. Attempts: "
        + " | ".join(errors)
    )


def data_period_attempts(*, period: str, interval: str) -> list[str]:
    attempts = [str(period)]
    if str(interval).lower() in INTRADAY_INTERVALS:
        attempts.extend(["60d", "30d", "7d", "5d", "1mo"])
    else:
        attempts.extend(["2y", "1y", "6mo", "3mo"])
    return list(dict.fromkeys(attempts))


def normalize_yahoo_history(raw: Any, *, symbol: str) -> pd.DataFrame:
    if raw is None:
        raise ValueError("empty response")
    if isinstance(raw, dict):
        message = raw.get(symbol, raw)
        raise ValueError(f"Yahoo returned an error response: {message}")
    if isinstance(raw, pd.Series):
        raise ValueError(f"Yahoo returned a Series instead of OHLCV data: {raw.to_dict()}")
    if not isinstance(raw, pd.DataFrame):
        raise ValueError(f"Unexpected Yahoo response type: {type(raw).__name__}")
    if raw.empty:
        raise ValueError("empty response")

    frame = raw.copy()
    if isinstance(frame.index, pd.MultiIndex):
        if "symbol" in frame.index.names:
            frame = frame.reset_index(level="symbol", drop=True)
        else:
            frame = frame.reset_index()
    return frame


def load_ohlcv(config: ReversalMLConfig) -> pd.DataFrame:
    if config.csv_path:
        return prepare_ohlcv(pd.read_csv(config.csv_path))
    return fetch_ohlcv(config.symbol, config.period, config.interval)


def make_feature_frame(data: pd.DataFrame, config: ReversalMLConfig) -> pd.DataFrame:
    close = data["close"].astype(float)
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    open_ = data["open"].astype(float)
    volume = data["volume"].astype(float)

    features = pd.DataFrame(index=data.index)
    log_close = np.log(close)
    log_return = log_close.diff()

    features["log_return_1"] = log_return
    for window in [2, 3, 5, 10, 20]:
        features[f"log_return_sum_{window}"] = log_return.rolling(window).sum()
        features[f"realized_vol_{window}"] = log_return.rolling(window).std()

    atr = average_true_range(data, period=14)
    features["atr"] = atr
    features["atr_pct"] = atr / close.replace(0, np.nan) * 100.0
    features["atr_pct_change_5"] = features["atr_pct"].pct_change(5)

    features["rsi_7"] = rsi(close, period=7)
    features["rsi_14"] = rsi(close, period=14)
    macd_line, macd_signal, macd_hist = macd(close)
    features["macd_line"] = macd_line
    features["macd_signal"] = macd_signal
    features["macd_hist"] = macd_hist
    features["stoch_k_14"] = stochastic_k(high, low, close, period=14)
    features["stoch_d_3"] = features["stoch_k_14"].rolling(3).mean()

    feature_windows = sorted({20, 50, 80, int(config.label_lookback)})
    for window in feature_windows:
        roll_high = high.rolling(window).max()
        roll_low = low.rolling(window).min()
        roll_mid = close.rolling(window).mean()
        roll_std = close.rolling(window).std()

        features[f"drawdown_from_high_{window}"] = (roll_high - close) / roll_high * 100.0
        features[f"rebound_from_low_{window}"] = (close - roll_low) / roll_low * 100.0
        features[f"dist_to_low_atr_{window}"] = (close - roll_low) / atr.replace(0, np.nan)
        features[f"dist_to_high_atr_{window}"] = (roll_high - close) / atr.replace(0, np.nan)
        features[f"bollinger_z_{window}"] = (close - roll_mid) / roll_std.replace(0, np.nan)
        features[f"range_pct_{window}"] = (roll_high - roll_low) / roll_mid.replace(0, np.nan) * 100.0
        features[f"slope_pct_{window}"] = rolling_slope_pct(close, window=window)

    candle_range = (high - low).replace(0, np.nan)
    features["body_pct"] = (close - open_) / open_.replace(0, np.nan) * 100.0
    features["body_to_range"] = (close - open_).abs() / candle_range
    features["upper_wick_to_range"] = (high - np.maximum(open_, close)) / candle_range
    features["lower_wick_to_range"] = (np.minimum(open_, close) - low) / candle_range
    features["close_position_in_range"] = (close - low) / candle_range
    features["gap_pct"] = (open_ - close.shift(1)) / close.shift(1).replace(0, np.nan) * 100.0
    features["green_streak"] = signed_streak(close >= open_)

    features["log_volume"] = np.log1p(volume)
    features["volume_z_20"] = rolling_zscore(features["log_volume"], window=20)
    features["volume_ratio_20"] = volume / volume.rolling(20).mean().replace(0, np.nan)
    obv = on_balance_volume(close, volume)
    features["obv_slope_20"] = rolling_slope_pct(obv, window=20)

    features["prior_down_context"] = (
        (features[f"drawdown_from_high_{config.label_lookback}"] >= config.min_prior_drawdown_pct)
        & (features[f"dist_to_low_atr_{config.label_lookback}"] <= config.max_distance_to_low_atr)
    ).astype(float)

    if config.use_arima:
        arima_features = walk_forward_arima_features(
            log_return,
            order=config.arima_order,
            window=config.arima_window,
        )
        features = features.join(arima_features)

    features = features.replace([np.inf, -np.inf], np.nan)
    return features


def make_reversal_labels(
    data: pd.DataFrame,
    features: pd.DataFrame,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    close = data["close"].to_numpy(dtype=float)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    atr = features["atr"].to_numpy(dtype=float)
    prior_drawdown = features[f"drawdown_from_high_{config.label_lookback}"].to_numpy(dtype=float)
    dist_to_low = features[f"dist_to_low_atr_{config.label_lookback}"].to_numpy(dtype=float)

    labels = pd.DataFrame(index=data.index)
    labels["reversal_label"] = np.nan
    labels["label_target_price"] = np.nan
    labels["label_stop_price"] = np.nan
    labels["label_hit_bar_idx"] = np.nan
    labels["label_hit_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["label_hit_type"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["prior_drawdown_pct"] = prior_drawdown
    labels["distance_to_low_atr"] = dist_to_low

    n = len(data)
    for bar_idx in range(n):
        if bar_idx + config.label_horizon >= n:
            continue
        if not np.isfinite(atr[bar_idx]) or atr[bar_idx] <= 0:
            continue

        target_price = close[bar_idx] + config.target_atr * atr[bar_idx]
        stop_price = close[bar_idx] - config.stop_atr * atr[bar_idx]
        labels.iloc[bar_idx, labels.columns.get_loc("label_target_price")] = target_price
        labels.iloc[bar_idx, labels.columns.get_loc("label_stop_price")] = stop_price

        in_reversal_context = (
            np.isfinite(prior_drawdown[bar_idx])
            and np.isfinite(dist_to_low[bar_idx])
            and prior_drawdown[bar_idx] >= config.min_prior_drawdown_pct
            and dist_to_low[bar_idx] <= config.max_distance_to_low_atr
        )
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=bar_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        label = int(in_reversal_context and hit_type == "target")
        labels.iloc[bar_idx, labels.columns.get_loc("reversal_label")] = label
        if hit_idx is not None:
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_bar_idx")] = int(hit_idx)
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_date")] = data.index[int(hit_idx)]
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_type")] = hit_type

    return labels


def first_long_barrier_hit(
    *,
    high: np.ndarray,
    low: np.ndarray,
    start_idx: int,
    horizon: int,
    target_price: float,
    stop_price: float,
) -> tuple[str, int | None]:
    max_idx = min(len(high) - 1, int(start_idx) + int(horizon))
    for idx in range(int(start_idx) + 1, max_idx + 1):
        stop_hit = low[idx] <= stop_price
        target_hit = high[idx] >= target_price
        if stop_hit and target_hit:
            return "stop", idx
        if stop_hit:
            return "stop", idx
        if target_hit:
            return "target", idx
    return "timeout", None


def add_hmm_features(
    features: pd.DataFrame,
    train_index: pd.Index,
    config: ReversalMLConfig,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    state_summary_columns = ["state", "train_count", "mean_return", "mean_vol", "mean_drawdown"]
    if not config.use_hmm:
        return features, pd.DataFrame(columns=state_summary_columns)
    if GaussianHMM is None or SimpleImputer is None or StandardScaler is None:
        warnings.warn("hmmlearn/sklearn unavailable; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    hmm_columns = [
        column
        for column in [
            "log_return_1",
            "realized_vol_10",
            "atr_pct",
            "volume_z_20",
            f"drawdown_from_high_{config.label_lookback}",
            f"dist_to_low_atr_{config.label_lookback}",
        ]
        if column in features.columns
    ]
    if len(hmm_columns) < 3:
        return features, pd.DataFrame(columns=state_summary_columns)

    train = features.loc[train_index, hmm_columns]
    if train.dropna(how="all").shape[0] < config.hmm_states * 20:
        warnings.warn("Not enough training rows for HMM; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    imputer = SimpleImputer(strategy="median")
    scaler = StandardScaler()
    train_values = scaler.fit_transform(imputer.fit_transform(train))
    all_values = scaler.transform(imputer.transform(features[hmm_columns]))

    hmm = GaussianHMM(
        n_components=config.hmm_states,
        covariance_type="diag",
        n_iter=200,
        tol=1e-4,
        random_state=config.random_state,
    )
    hmm.fit(train_values)
    probabilities = hmm.predict_proba(all_values)
    states = probabilities.argmax(axis=1)

    out = features.copy()
    out["hmm_state"] = states.astype(float)
    for state in range(config.hmm_states):
        out[f"hmm_state_prob_{state}"] = probabilities[:, state]

    state_rows: list[dict[str, Any]] = []
    train_augmented = out.loc[train_index].copy()
    for state in range(config.hmm_states):
        subset = train_augmented[train_augmented["hmm_state"] == state]
        state_rows.append(
            {
                "state": state,
                "train_count": int(len(subset)),
                "mean_return": float(subset["log_return_1"].mean()),
                "mean_vol": float(subset["realized_vol_10"].mean()),
                "mean_drawdown": float(
                    subset[f"drawdown_from_high_{config.label_lookback}"].mean()
                ),
            }
        )
    return out, pd.DataFrame(state_rows, columns=state_summary_columns)


def train_reversal_models(
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    config: ReversalMLConfig,
) -> dict[str, Any]:
    require_sklearn()

    labeled = features.join(labels[["reversal_label"]], how="inner")
    labeled = labeled.dropna(subset=["reversal_label"])
    if len(labeled) < 200:
        raise ValueError("Not enough labeled rows. Increase period or use a higher timeframe.")

    y = labeled["reversal_label"].astype(int)
    feature_columns = [
        column
        for column in labeled.columns
        if column != "reversal_label" and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]

    split_idx = int(len(labeled) * (1.0 - config.test_size_pct))
    split_idx = min(max(split_idx, 50), len(labeled) - 20)
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]

    features_with_hmm, hmm_summary = add_hmm_features(features, train_index, config)
    labeled = features_with_hmm.join(labels[["reversal_label"]], how="inner")
    labeled = labeled.dropna(subset=["reversal_label"])
    y = labeled["reversal_label"].astype(int)
    feature_columns = [
        column
        for column in labeled.columns
        if column != "reversal_label" and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]
    X_train = X.loc[train_index]
    y_train = y.loc[train_index]
    X_test = X.loc[test_index]
    y_test = y.loc[test_index]

    X_train_clip, clip_bounds = fit_winsorize(X_train)
    X_test_clip = apply_winsorize(X_test, clip_bounds)
    X_all_clip = apply_winsorize(features_with_hmm[feature_columns], clip_bounds)

    models = build_models(config)
    results: list[dict[str, Any]] = []
    fitted_models: dict[str, Pipeline] = {}

    for name, model in models.items():
        fitted = fit_pipeline_with_weights(model, X_train_clip, y_train)
        fitted_models[name] = fitted
        test_probability = positive_probability(fitted, X_test_clip)
        metrics = evaluate_probabilities(y_test, test_probability)
        cv_metrics = time_series_cv_score(
            model,
            X_train_clip,
            y_train,
            n_splits=config.n_splits,
        )
        result = {"model": name, **metrics, **cv_metrics}
        results.append(result)

    model_report = pd.DataFrame(results)
    best_name = choose_best_model(model_report)
    per_model_probabilities: dict[str, pd.Series] = {}
    per_model_thresholds: dict[str, float] = {}
    per_model_feature_importance: dict[str, pd.DataFrame] = {}
    per_model_signals: dict[str, pd.DataFrame] = {}

    for name, fitted in fitted_models.items():
        all_probability = positive_probability(fitted, X_all_clip)
        test_probability = positive_probability(fitted, X_test_clip)
        model_threshold = config.probability_threshold
        if model_threshold is None:
            model_threshold = choose_probability_threshold(
                y_test,
                test_probability,
                target_precision=config.target_precision,
            )

        probability_series = pd.Series(
            all_probability,
            index=features_with_hmm.index,
            name=f"{name}_reversal_probability",
        )
        importance = permutation_xai(
            fitted,
            X_test_clip,
            y_test,
            feature_columns,
            repeats=config.permutation_repeats,
            random_state=config.random_state,
        )
        signals = build_signal_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=probability_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not signals.empty:
            signals.insert(0, "Model", name)

        per_model_probabilities[name] = probability_series
        per_model_thresholds[name] = float(model_threshold)
        per_model_feature_importance[name] = importance
        per_model_signals[name] = signals

    best_model = fitted_models[best_name]
    threshold = per_model_thresholds[best_name]
    probabilities = per_model_probabilities[best_name].rename("reversal_probability")
    signals = per_model_signals[best_name].copy()
    if "Model" in signals.columns:
        signals = signals.drop(columns=["Model"])
    feature_importance = per_model_feature_importance[best_name]
    pca_loadings = extract_pca_loadings(best_model, feature_columns)

    return {
        "features": features_with_hmm,
        "labels": labels,
        "feature_columns": feature_columns,
        "model_report": model_report,
        "best_model_name": best_name,
        "best_model": best_model,
        "probabilities": probabilities,
        "threshold": threshold,
        "signals": signals,
        "feature_importance": feature_importance,
        "per_model_probabilities": per_model_probabilities,
        "per_model_thresholds": per_model_thresholds,
        "per_model_signals": per_model_signals,
        "per_model_feature_importance": per_model_feature_importance,
        "pca_loadings": pca_loadings,
        "hmm_state_summary": hmm_summary,
    }


def require_sklearn() -> None:
    if Pipeline is None:
        raise ImportError(
            "scikit-learn is required for ML training. Install with: pip install scikit-learn"
        )


def build_models(config: ReversalMLConfig) -> dict[str, Pipeline]:
    requested = {name.strip().lower() for name in config.model_names}
    models: dict[str, Pipeline] = {}

    if "boosting" in requested:
        models["boosting"] = make_pipeline(
            HistGradientBoostingClassifier(
                learning_rate=0.05,
                max_iter=350,
                max_leaf_nodes=15,
                l2_regularization=0.05,
                early_stopping=True,
                random_state=config.random_state,
            ),
            config,
            scale=False,
        )
    if "logistic" in requested:
        models["logistic"] = make_pipeline(
            LogisticRegression(
                max_iter=2500,
                class_weight="balanced",
                solver="lbfgs",
                random_state=config.random_state,
            ),
            config,
            scale=True,
        )
    if "random_forest" in requested:
        models["random_forest"] = make_pipeline(
            RandomForestClassifier(
                n_estimators=400,
                max_depth=7,
                min_samples_leaf=20,
                class_weight="balanced_subsample",
                random_state=config.random_state,
                n_jobs=-1,
            ),
            config,
            scale=False,
        )
    if not models:
        raise ValueError("No valid model names requested.")
    return models


def make_pipeline(estimator: Any, config: ReversalMLConfig, *, scale: bool) -> Pipeline:
    steps: list[tuple[str, Any]] = [("imputer", SimpleImputer(strategy="median"))]
    if scale or config.use_pca:
        steps.append(("scaler", StandardScaler()))
    if config.use_pca:
        steps.append(("pca", PCA(n_components=config.pca_variance, svd_solver="full")))
    steps.append(("model", estimator))
    return Pipeline(steps)


def fit_pipeline_with_weights(model: Pipeline, X: pd.DataFrame, y: pd.Series) -> Pipeline:
    sample_weight = compute_sample_weight(class_weight="balanced", y=y)
    try:
        return model.fit(X, y, model__sample_weight=sample_weight)
    except TypeError:
        return model.fit(X, y)


def positive_probability(model: Pipeline, X: pd.DataFrame) -> np.ndarray:
    if hasattr(model, "predict_proba"):
        probability = model.predict_proba(X)
        if probability.shape[1] == 1:
            return np.zeros(len(X), dtype=float)
        return probability[:, 1]
    decision = model.decision_function(X)
    return 1.0 / (1.0 + np.exp(-decision))


def evaluate_probabilities(y_true: pd.Series, probability: np.ndarray) -> dict[str, float]:
    out: dict[str, float] = {
        "test_average_precision": np.nan,
        "test_roc_auc": np.nan,
        "test_brier": np.nan,
        "test_balanced_accuracy_050": np.nan,
        "test_precision_050": np.nan,
        "test_recall_050": np.nan,
        "test_positive_rate": float(np.mean(y_true)),
    }
    if len(np.unique(y_true)) < 2:
        return out

    prediction = (probability >= 0.5).astype(int)
    out["test_average_precision"] = float(average_precision_score(y_true, probability))
    out["test_roc_auc"] = float(roc_auc_score(y_true, probability))
    out["test_brier"] = float(brier_score_loss(y_true, probability))
    out["test_balanced_accuracy_050"] = float(balanced_accuracy_score(y_true, prediction))
    out["test_precision_050"] = float(precision_score(y_true, prediction, zero_division=0))
    out["test_recall_050"] = float(recall_score(y_true, prediction, zero_division=0))
    out["test_accuracy_050"] = float(accuracy_score(y_true, prediction))
    return out


def time_series_cv_score(
    model: Pipeline,
    X: pd.DataFrame,
    y: pd.Series,
    *,
    n_splits: int,
) -> dict[str, float]:
    out = {"cv_average_precision": np.nan, "cv_roc_auc": np.nan}
    if len(X) < 200 or y.nunique() < 2:
        return out

    split_count = min(n_splits, max(2, len(X) // 200))
    splitter = TimeSeriesSplit(n_splits=split_count)
    ap_scores: list[float] = []
    auc_scores: list[float] = []
    for train_idx, test_idx in splitter.split(X):
        y_train = y.iloc[train_idx]
        y_valid = y.iloc[test_idx]
        if y_train.nunique() < 2 or y_valid.nunique() < 2:
            continue
        fitted = fit_pipeline_with_weights(clone(model), X.iloc[train_idx], y_train)
        prob = positive_probability(fitted, X.iloc[test_idx])
        ap_scores.append(float(average_precision_score(y_valid, prob)))
        auc_scores.append(float(roc_auc_score(y_valid, prob)))

    if ap_scores:
        out["cv_average_precision"] = float(np.mean(ap_scores))
    if auc_scores:
        out["cv_roc_auc"] = float(np.mean(auc_scores))
    return out


def choose_best_model(model_report: pd.DataFrame) -> str:
    ranking = model_report.copy()
    ranking["_rank_score"] = ranking["test_average_precision"].fillna(-1.0)
    ranking = ranking.sort_values(["_rank_score", "test_roc_auc"], ascending=False)
    return str(ranking.iloc[0]["model"])


def choose_probability_threshold(
    y_true: pd.Series,
    probability: np.ndarray,
    *,
    target_precision: float,
) -> float:
    if len(np.unique(y_true)) < 2:
        return 0.65

    precision, recall, thresholds = precision_recall_curve(y_true, probability)
    candidates: list[tuple[float, float, float]] = []
    for threshold, prec, rec in zip(thresholds, precision[:-1], recall[:-1]):
        if prec >= target_precision and rec > 0:
            candidates.append((float(threshold), float(prec), float(rec)))
    if not candidates:
        return float(np.quantile(probability, 0.90))
    candidates.sort(key=lambda item: (item[2], item[1]), reverse=True)
    return float(candidates[0][0])


def permutation_xai(
    model: Pipeline,
    X_test: pd.DataFrame,
    y_test: pd.Series,
    feature_columns: list[str],
    *,
    repeats: int,
    random_state: int,
) -> pd.DataFrame:
    columns = ["feature", "importance_mean", "importance_std", "rank"]
    if y_test.nunique() < 2 or permutation_importance is None:
        return pd.DataFrame(columns=columns)

    result = permutation_importance(
        model,
        X_test,
        y_test,
        n_repeats=repeats,
        random_state=random_state,
        scoring="average_precision",
        n_jobs=-1,
    )
    importance = pd.DataFrame(
        {
            "feature": feature_columns,
            "importance_mean": result.importances_mean,
            "importance_std": result.importances_std,
        }
    )
    importance = importance.sort_values("importance_mean", ascending=False).reset_index(drop=True)
    importance["rank"] = np.arange(1, len(importance) + 1)
    return importance[columns]


def extract_pca_loadings(model: Pipeline, feature_columns: list[str]) -> pd.DataFrame:
    if "pca" not in model.named_steps:
        return pd.DataFrame(columns=["component", "feature", "loading", "abs_loading", "rank"])

    pca = model.named_steps["pca"]
    rows: list[dict[str, Any]] = []
    for component_idx, component in enumerate(pca.components_):
        for feature, loading in zip(feature_columns, component):
            rows.append(
                {
                    "component": f"PC{component_idx + 1}",
                    "feature": feature,
                    "loading": float(loading),
                    "abs_loading": float(abs(loading)),
                }
            )
    out = pd.DataFrame(rows)
    out["rank"] = out.groupby("component")["abs_loading"].rank(ascending=False, method="first")
    return out.sort_values(["component", "rank"]).reset_index(drop=True)


def build_signal_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    candidate_probability = probabilities.dropna()
    candidate_probability = candidate_probability[candidate_probability >= threshold]
    if candidate_probability.empty:
        return empty_signal_frame()

    selected_indexes = non_max_suppression_1d(
        candidate_probability,
        cooldown_bars=config.cooldown_bars,
        all_index=features.index,
    )

    train_stats = make_train_feature_stats(train_features)
    rows: list[dict[str, Any]] = []
    atr = features["atr"]
    for timestamp in selected_indexes:
        bar_idx = int(data.index.get_loc(timestamp))
        entry_price = float(data.loc[timestamp, "close"])
        target_price = entry_price + config.target_atr * float(atr.loc[timestamp])
        stop_price = entry_price - config.stop_atr * float(atr.loc[timestamp])
        label_row = labels.loc[timestamp] if timestamp in labels.index else pd.Series(dtype=object)
        hit_idx = label_row.get("label_hit_bar_idx", np.nan)
        hit_date = label_row.get("label_hit_date", pd.NaT)
        hit_type = label_row.get("label_hit_type", None)
        realized_label = label_row.get("reversal_label", np.nan)
        top_drivers = local_driver_summary(
            features.loc[timestamp, feature_columns],
            feature_importance,
            train_stats,
            top_n=6,
        )
        rows.append(
            {
                "EntryBarIdx": bar_idx,
                "EntryDate": timestamp,
                "EntryPrice": entry_price,
                "ReversalProbability": float(probabilities.loc[timestamp]),
                "SignalThreshold": float(threshold),
                "TargetPrice": target_price,
                "StopPrice": stop_price,
                "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                "OutcomeKnown": bool(np.isfinite(realized_label)),
                "OutcomeLabel": realized_label,
                "HitBarIdx": int(hit_idx) if np.isfinite(hit_idx) else np.nan,
                "HitDate": hit_date,
                "HitType": hit_type,
                "ShapeHint": classify_shape_hint(data, bar_idx),
                "TopDrivers": top_drivers,
            }
        )
    return pd.DataFrame(rows).sort_values("EntryBarIdx").reset_index(drop=True)


def empty_signal_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "ReversalProbability",
            "SignalThreshold",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def non_max_suppression_1d(
    probability: pd.Series,
    *,
    cooldown_bars: int,
    all_index: pd.Index,
) -> list[pd.Timestamp]:
    ranked = probability.sort_values(ascending=False)
    selected_positions: list[int] = []
    selected_timestamps: list[pd.Timestamp] = []

    for timestamp, _ in ranked.items():
        position = int(all_index.get_loc(timestamp))
        if any(abs(position - selected) <= cooldown_bars for selected in selected_positions):
            continue
        selected_positions.append(position)
        selected_timestamps.append(timestamp)

    selected_timestamps.sort(key=lambda ts: int(all_index.get_loc(ts)))
    return selected_timestamps


def make_train_feature_stats(train_features: pd.DataFrame) -> pd.DataFrame:
    stats = pd.DataFrame(index=train_features.columns)
    stats["median"] = train_features.median(numeric_only=True)
    q25 = train_features.quantile(0.25, numeric_only=True)
    q75 = train_features.quantile(0.75, numeric_only=True)
    stats["iqr"] = (q75 - q25).replace(0, np.nan)
    return stats


def local_driver_summary(
    row: pd.Series,
    feature_importance: pd.DataFrame,
    train_stats: pd.DataFrame,
    *,
    top_n: int,
) -> str:
    if feature_importance.empty:
        return ""

    importance_map = feature_importance.set_index("feature")["importance_mean"].to_dict()
    scored: list[tuple[float, str]] = []
    for feature, value in row.items():
        if feature not in importance_map or feature not in train_stats.index:
            continue
        if not np.isfinite(value):
            continue
        iqr = train_stats.loc[feature, "iqr"]
        median = train_stats.loc[feature, "median"]
        if not np.isfinite(iqr) or iqr == 0:
            continue
        robust_z = (float(value) - float(median)) / float(iqr)
        score = abs(float(importance_map[feature]) * robust_z)
        if score <= 0:
            continue
        scored.append((score, f"{feature}={float(value):.4g} z={robust_z:.2f}"))

    scored.sort(reverse=True)
    return " | ".join(text for _, text in scored[:top_n])


def classify_shape_hint(data: pd.DataFrame, bar_idx: int, *, lookback: int = 80) -> str:
    start = max(0, int(bar_idx) - lookback)
    if int(bar_idx) - start < 20:
        return "ML Bullish Reversal"

    window = data.iloc[start : int(bar_idx) + 1]
    lows = window["low"].to_numpy(dtype=float)
    highs = window["high"].to_numpy(dtype=float)
    close = window["close"].to_numpy(dtype=float)
    atr = float(average_true_range(window, period=14).iloc[-1])
    if not np.isfinite(atr) or atr <= 0:
        atr = max(float(np.nanstd(close)), 1e-12)

    pivot_lows = simple_pivots(lows, mode="low", order=3)
    pivot_highs = simple_pivots(highs, mode="high", order=3)
    if len(pivot_lows) >= 2:
        l1, l2 = pivot_lows[-2], pivot_lows[-1]
        low_diff_atr = abs(lows[l2] - lows[l1]) / atr
        if l2 > l1 and low_diff_atr <= 1.0:
            between_high = np.max(highs[l1:l2 + 1])
            if between_high - max(lows[l1], lows[l2]) >= atr:
                return "ML Double Bottom-like"

    if len(pivot_lows) >= 3 and len(pivot_highs) >= 2:
        left, head, right = pivot_lows[-3], pivot_lows[-2], pivot_lows[-1]
        if left < head < right and lows[head] < min(lows[left], lows[right]):
            shoulder_diff_atr = abs(lows[left] - lows[right]) / atr
            if shoulder_diff_atr <= 1.5:
                return "ML Inverse H&S-like"

    return "ML Bullish Reversal"


def simple_pivots(values: np.ndarray, *, mode: str, order: int) -> list[int]:
    pivots: list[int] = []
    for idx in range(order, len(values) - order):
        left = values[idx - order : idx]
        right = values[idx + 1 : idx + order + 1]
        if mode == "low" and values[idx] <= np.min(left) and values[idx] <= np.min(right):
            pivots.append(idx)
        if mode == "high" and values[idx] >= np.max(left) and values[idx] >= np.max(right):
            pivots.append(idx)
    return pivots


def fit_winsorize(
    X_train: pd.DataFrame,
    *,
    lower: float = 0.01,
    upper: float = 0.99,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    bounds = pd.DataFrame(index=X_train.columns)
    bounds["lower"] = X_train.quantile(lower, numeric_only=True)
    bounds["upper"] = X_train.quantile(upper, numeric_only=True)
    return apply_winsorize(X_train, bounds), bounds


def apply_winsorize(X: pd.DataFrame, bounds: pd.DataFrame) -> pd.DataFrame:
    out = X.copy()
    for column in out.columns:
        if column not in bounds.index:
            continue
        lower = bounds.loc[column, "lower"]
        upper = bounds.loc[column, "upper"]
        if np.isfinite(lower) and np.isfinite(upper) and lower < upper:
            out[column] = out[column].clip(lower=lower, upper=upper)
    return out


def average_true_range(data: pd.DataFrame, *, period: int = 14) -> pd.Series:
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    close = data["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [
            high - low,
            (high - previous_close).abs(),
            (low - previous_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    return true_range.rolling(period, min_periods=1).mean()


def rsi(close: pd.Series, *, period: int = 14) -> pd.Series:
    delta = close.diff()
    gains = delta.clip(lower=0.0)
    losses = -delta.clip(upper=0.0)
    avg_gain = gains.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    avg_loss = losses.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    rs = avg_gain / avg_loss.replace(0, np.nan)
    return 100.0 - (100.0 / (1.0 + rs))


def macd(close: pd.Series) -> tuple[pd.Series, pd.Series, pd.Series]:
    fast = close.ewm(span=12, adjust=False).mean()
    slow = close.ewm(span=26, adjust=False).mean()
    line = fast - slow
    signal = line.ewm(span=9, adjust=False).mean()
    return line, signal, line - signal


def stochastic_k(
    high: pd.Series,
    low: pd.Series,
    close: pd.Series,
    *,
    period: int,
) -> pd.Series:
    lowest = low.rolling(period).min()
    highest = high.rolling(period).max()
    return (close - lowest) / (highest - lowest).replace(0, np.nan) * 100.0


def rolling_zscore(series: pd.Series, *, window: int) -> pd.Series:
    mean = series.rolling(window).mean()
    std = series.rolling(window).std()
    return (series - mean) / std.replace(0, np.nan)


def rolling_slope_pct(series: pd.Series, *, window: int) -> pd.Series:
    x = np.arange(window, dtype=float)
    x = x - x.mean()

    def slope(values: np.ndarray) -> float:
        if np.isnan(values).any():
            return np.nan
        y = values.astype(float)
        y = y - y.mean()
        denom = float(np.dot(x, x))
        if denom == 0:
            return np.nan
        raw_slope = float(np.dot(x, y) / denom)
        return raw_slope / max(abs(float(values[-1])), 1e-12) * 100.0

    return series.rolling(window).apply(slope, raw=True)


def signed_streak(condition: pd.Series) -> pd.Series:
    values = condition.fillna(False).to_numpy(dtype=bool)
    streaks = np.zeros(len(values), dtype=float)
    current = 0
    previous = None
    for idx, value in enumerate(values):
        if previous is None or value != previous:
            current = 1
        else:
            current += 1
        streaks[idx] = current if value else -current
        previous = value
    return pd.Series(streaks, index=condition.index)


def on_balance_volume(close: pd.Series, volume: pd.Series) -> pd.Series:
    direction = np.sign(close.diff()).fillna(0.0)
    return (direction * volume).cumsum()


def walk_forward_arima_features(
    returns: pd.Series,
    *,
    order: tuple[int, int, int],
    window: int,
) -> pd.DataFrame:
    columns = ["arima_forecast_return", "arima_residual", "arima_residual_z_40"]
    out = pd.DataFrame(np.nan, index=returns.index, columns=columns)
    if ARIMA is None:
        warnings.warn("statsmodels unavailable; ARIMA features skipped.")
        return out

    values = returns.astype(float)
    for idx in range(window, len(values)):
        history = values.iloc[idx - window : idx].dropna()
        if len(history) < max(30, sum(order) * 10):
            continue
        try:
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                result = ARIMA(history, order=order).fit()
            forecast = float(result.forecast(1).iloc[0])
            residual = float(values.iloc[idx] - forecast)
            out.iloc[idx, out.columns.get_loc("arima_forecast_return")] = forecast
            out.iloc[idx, out.columns.get_loc("arima_residual")] = residual
        except Exception:
            continue
    out["arima_residual_z_40"] = rolling_zscore(out["arima_residual"], window=40)
    return out


def save_outputs(
    data: pd.DataFrame,
    result: dict[str, Any],
    config: ReversalMLConfig,
) -> Path:
    run_name = f"{config.symbol.replace('.', '_')}_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
    run_dir = Path(config.output_dir) / run_name
    run_dir.mkdir(parents=True, exist_ok=True)
    if go is None:
        warnings.warn("plotly unavailable; HTML plot files skipped. Install with: pip install plotly")

    result["signals"].to_csv(run_dir / "signals.csv", index=False)
    result["model_report"].to_csv(run_dir / "model_report.csv", index=False)
    result["feature_importance"].to_csv(run_dir / "feature_importance.csv", index=False)
    result["hmm_state_summary"].to_csv(run_dir / "hmm_state_summary.csv", index=False)
    result["pca_loadings"].to_csv(run_dir / "pca_loadings.csv", index=False)

    probability_frame = pd.DataFrame(
        {
            "Date": result["probabilities"].index,
            "ReversalProbability": result["probabilities"].to_numpy(dtype=float),
        }
    )
    probability_frame.to_csv(run_dir / "probabilities.csv", index=False)

    all_probability_frame = pd.DataFrame({"Date": data.index})
    for model_name, probabilities in result["per_model_probabilities"].items():
        all_probability_frame[f"{model_name}_probability"] = probabilities.reindex(data.index).to_numpy(dtype=float)
    all_probability_frame.to_csv(run_dir / "all_model_probabilities.csv", index=False)

    models_dir = run_dir / "models"
    models_dir.mkdir(exist_ok=True)
    for model_name, probabilities in result["per_model_probabilities"].items():
        safe_model_name = safe_filename(model_name)
        signals = result["per_model_signals"].get(model_name, empty_signal_frame())
        importance = result["per_model_feature_importance"].get(
            model_name,
            pd.DataFrame(columns=["feature", "importance_mean", "importance_std", "rank"]),
        )
        threshold = result["per_model_thresholds"].get(model_name, np.nan)
        signals.to_csv(models_dir / f"{safe_model_name}_signals.csv", index=False)
        importance.to_csv(models_dir / f"{safe_model_name}_feature_importance.csv", index=False)
        make_plot(
            data=data,
            probabilities=probabilities,
            signals=signals,
            title=f"{config.symbol} {model_name} ML Reversal Signals",
            output_path=models_dir / f"{safe_model_name}_plot.html",
            model_name=model_name,
            threshold=threshold,
        )
        make_feature_importance_plot(
            importance=importance,
            title=f"{config.symbol} {model_name} XAI Feature Importance",
            output_path=models_dir / f"{safe_model_name}_feature_importance.html",
        )

    with open(run_dir / "config.json", "w", encoding="utf-8") as handle:
        json.dump(
            {
                **asdict(config),
                "best_model_name": result["best_model_name"],
                "threshold": result["threshold"],
            },
            handle,
            indent=2,
            default=str,
        )

    plot_path = run_dir / "reversal_ml_plot.html"
    make_plot(
        data=data,
        probabilities=result["probabilities"],
        signals=result["signals"],
        title=f"{config.symbol} ML Reversal Signals",
        output_path=plot_path,
        model_name=f"best: {result['best_model_name']}",
        threshold=result["threshold"],
    )
    make_model_comparison_plot(
        data=data,
        per_model_probabilities=result["per_model_probabilities"],
        per_model_signals=result["per_model_signals"],
        per_model_thresholds=result["per_model_thresholds"],
        title=f"{config.symbol} ML Reversal Model Comparison",
        output_path=run_dir / "all_models_comparison_plot.html",
    )
    return run_dir


def safe_filename(value: str) -> str:
    safe = "".join(character if character.isalnum() or character in {"-", "_"} else "_" for character in value)
    return safe.strip("_") or "model"


def make_plot(
    *,
    data: pd.DataFrame,
    probabilities: pd.Series,
    signals: pd.DataFrame,
    title: str,
    output_path: Path,
    model_name: str | None = None,
    threshold: float | None = None,
) -> None:
    if go is None:
        return

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probabilities.reindex(data.index).to_numpy(dtype=float),
            name=f"{model_name or 'Model'} probability",
            mode="lines",
            yaxis="y2",
            line=dict(color="#38BDF8", width=2),
        )
    )
    if threshold is not None and np.isfinite(float(threshold)):
        fig.add_trace(
            go.Scatter(
                x=[plot_x[0], plot_x[-1]],
                y=[float(threshold), float(threshold)],
                name="Signal threshold",
                mode="lines",
                yaxis="y2",
                line=dict(color="#FACC15", width=1, dash="dash"),
                hoverinfo="skip",
            )
        )

    if not signals.empty:
        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers+text",
                name=f"{model_name or 'ML'} Buy Signal",
                marker=dict(color="#FACC15", size=12, symbol="circle"),
                text=[f"{prob:.2f}" for prob in signals["ReversalProbability"]],
                textposition="top center",
                hovertext=[
                    (
                        f"ML Buy<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Probability: {row['ReversalProbability']:.3f}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Probability",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=65, t=80, b=55),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def make_feature_importance_plot(
    *,
    importance: pd.DataFrame,
    title: str,
    output_path: Path,
    top_n: int = 25,
) -> None:
    if go is None or importance.empty:
        return

    plot_frame = importance.head(top_n).copy()
    plot_frame = plot_frame.sort_values("importance_mean", ascending=True)
    fig = go.Figure()
    fig.add_trace(
        go.Bar(
            x=plot_frame["importance_mean"],
            y=plot_frame["feature"],
            orientation="h",
            error_x=dict(
                type="data",
                array=plot_frame["importance_std"].clip(lower=0).to_numpy(dtype=float),
                visible=True,
                color="#94A3B8",
            ),
            marker=dict(color="#38BDF8"),
            hovertext=[
                (
                    f"Feature: {row['feature']}<br>"
                    f"Importance: {row['importance_mean']:.6f}<br>"
                    f"Std: {row['importance_std']:.6f}<br>"
                    f"Rank: {int(row['rank'])}"
                )
                for _, row in plot_frame.iterrows()
            ],
            hoverinfo="text",
            name="Permutation importance",
        )
    )
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=max(520, 24 * len(plot_frame) + 180),
        xaxis=dict(title="Average precision decrease"),
        yaxis=dict(title="Feature"),
        margin=dict(l=180, r=50, t=80, b=60),
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def make_model_comparison_plot(
    *,
    data: pd.DataFrame,
    per_model_probabilities: dict[str, pd.Series],
    per_model_signals: dict[str, pd.DataFrame],
    per_model_thresholds: dict[str, float],
    title: str,
    output_path: Path,
) -> None:
    if go is None or not per_model_probabilities:
        return

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    colors = ["#38BDF8", "#FACC15", "#A78BFA", "#34D399", "#FB7185", "#F97316"]

    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    for model_idx, (model_name, probabilities) in enumerate(per_model_probabilities.items()):
        color = colors[model_idx % len(colors)]
        threshold = per_model_thresholds.get(model_name, np.nan)
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=probabilities.reindex(data.index).to_numpy(dtype=float),
                name=f"{model_name} probability",
                mode="lines",
                yaxis="y2",
                line=dict(color=color, width=2),
            )
        )
        if threshold is not None and np.isfinite(float(threshold)):
            fig.add_trace(
                go.Scatter(
                    x=[plot_x[0], plot_x[-1]],
                    y=[float(threshold), float(threshold)],
                    name=f"{model_name} threshold",
                    mode="lines",
                    yaxis="y2",
                    line=dict(color=color, width=1, dash="dot"),
                    hoverinfo="skip",
                    showlegend=False,
                )
            )

        signals = per_model_signals.get(model_name, empty_signal_frame())
        if signals is None or signals.empty:
            continue

        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers",
                name=f"{model_name} signals",
                marker=dict(
                    color=color,
                    size=10,
                    symbol="circle-open",
                    line=dict(width=2),
                ),
                hovertext=[
                    (
                        f"Model: {model_name}<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Probability: {row['ReversalProbability']:.3f}<br>"
                        f"Threshold: {row['SignalThreshold']:.3f}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=860,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Probability",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=65, t=90, b=60),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def run(config: ReversalMLConfig) -> Path:
    data = load_ohlcv(config)
    features = make_feature_frame(data, config)
    labels = make_reversal_labels(data, features, config)
    result = train_reversal_models(data, features, labels, config)
    run_dir = save_outputs(data, result, config)

    print(f"Saved run: {run_dir}")
    print(f"Best model: {result['best_model_name']}")
    print(f"Threshold: {result['threshold']:.4f}")
    print(f"Signals: {len(result['signals'])}")
    if not result["model_report"].empty:
        print(result["model_report"].to_string(index=False))
    if not result["signals"].empty:
        print(result["signals"].tail(10).to_string(index=False))
    return run_dir


def parse_args(argv: list[str] | None = None) -> ReversalMLConfig:
    parser = argparse.ArgumentParser(description="ML bullish reversal detector with XAI outputs.")
    parser.add_argument("--symbol", default="THYAO.IS")
    parser.add_argument("--period", default="60d")
    parser.add_argument("--interval", default="5m")
    parser.add_argument("--csv", dest="csv_path", default=None)
    parser.add_argument("--output-dir", default="outputs/ml_reversal_runs")
    parser.add_argument("--label-lookback", type=int, default=80)
    parser.add_argument("--label-horizon", type=int, default=40)
    parser.add_argument("--min-prior-drawdown-pct", type=float, default=2.0)
    parser.add_argument("--max-distance-to-low-atr", type=float, default=2.5)
    parser.add_argument("--target-atr", type=float, default=2.0)
    parser.add_argument("--stop-atr", type=float, default=1.2)
    parser.add_argument("--probability-threshold", type=float, default=None)
    parser.add_argument("--target-precision", type=float, default=0.55)
    parser.add_argument("--cooldown-bars", type=int, default=30)
    parser.add_argument("--models", default="boosting,logistic,random_forest")
    parser.add_argument("--use-hmm", action=argparse.BooleanOptionalAction, default=True)
    parser.add_argument("--hmm-states", type=int, default=4)
    parser.add_argument("--use-arima", action=argparse.BooleanOptionalAction, default=False)
    parser.add_argument("--arima-order", default="1,0,1")
    parser.add_argument("--arima-window", type=int, default=160)
    parser.add_argument("--use-pca", action=argparse.BooleanOptionalAction, default=False)
    parser.add_argument("--pca-variance", type=float, default=0.95)
    parser.add_argument("--random-state", type=int, default=42)
    args, unknown_args = parser.parse_known_args(argv)
    if unknown_args:
        warnings.warn(
            "Ignoring unknown CLI arguments, usually injected by notebooks: "
            + " ".join(unknown_args)
        )

    arima_order = tuple(int(part.strip()) for part in args.arima_order.split(","))
    if len(arima_order) != 3:
        raise ValueError("--arima-order must contain three comma-separated integers, e.g. 1,0,1")

    return ReversalMLConfig(
        symbol=args.symbol,
        period=args.period,
        interval=args.interval,
        csv_path=args.csv_path,
        output_dir=args.output_dir,
        random_state=args.random_state,
        label_lookback=args.label_lookback,
        label_horizon=args.label_horizon,
        min_prior_drawdown_pct=args.min_prior_drawdown_pct,
        max_distance_to_low_atr=args.max_distance_to_low_atr,
        target_atr=args.target_atr,
        stop_atr=args.stop_atr,
        probability_threshold=args.probability_threshold,
        target_precision=args.target_precision,
        cooldown_bars=args.cooldown_bars,
        model_names=tuple(part.strip() for part in args.models.split(",") if part.strip()),
        use_hmm=args.use_hmm,
        hmm_states=args.hmm_states,
        use_arima=args.use_arima,
        arima_order=arima_order,
        arima_window=args.arima_window,
        use_pca=args.use_pca,
        pca_variance=args.pca_variance,
    )


if __name__ == "__main__":
    run(parse_args())
```

```python
"""
Notebook usage:

from outputs.ml_reversal_pattern_detector import run_notebook

analysis = run_notebook(
    symbol="THYAO.IS",
    period="60d",
    interval="5m",
    show_plots=True,
)

analysis["result"]["intervals"]
analysis["csv_paths"]
analysis["plot_paths"]
"""

from __future__ import annotations

import json
import math
import warnings
from dataclasses import asdict, dataclass
from datetime import datetime
from pathlib import Path
from typing import Any

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    from sklearn.base import clone
    from sklearn.decomposition import PCA
    from sklearn.ensemble import HistGradientBoostingClassifier, RandomForestClassifier
    from sklearn.impute import SimpleImputer
    from sklearn.inspection import permutation_importance
    from sklearn.linear_model import LogisticRegression
    from sklearn.metrics import (
        accuracy_score,
        average_precision_score,
        balanced_accuracy_score,
        brier_score_loss,
        precision_recall_curve,
        precision_score,
        recall_score,
        roc_auc_score,
    )
    from sklearn.model_selection import TimeSeriesSplit
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
    from sklearn.utils.class_weight import compute_sample_weight
except ModuleNotFoundError:
    clone = None
    PCA = None
    HistGradientBoostingClassifier = None
    RandomForestClassifier = None
    SimpleImputer = None
    permutation_importance = None
    LogisticRegression = None
    TimeSeriesSplit = None
    Pipeline = None
    StandardScaler = None
    compute_sample_weight = None
    accuracy_score = None
    average_precision_score = None
    balanced_accuracy_score = None
    brier_score_loss = None
    precision_recall_curve = None
    precision_score = None
    recall_score = None
    roc_auc_score = None

try:
    from hmmlearn.hmm import GaussianHMM
except ModuleNotFoundError:
    GaussianHMM = None

try:
    from statsmodels.tsa.arima.model import ARIMA
except ModuleNotFoundError:
    ARIMA = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


@dataclass
class ReversalMLConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    csv_path: str | None = None
    output_dir: str = "outputs/ml_reversal_runs"
    random_state: int = 42

    label_lookback: int = 80
    label_horizon: int = 40
    min_prior_drawdown_pct: float = 2.0
    max_distance_to_low_atr: float = 2.5
    target_atr: float = 2.0
    stop_atr: float = 1.2

    test_size_pct: float = 0.25
    n_splits: int = 5
    probability_threshold: float | None = None
    target_precision: float = 0.55
    cooldown_bars: int = 30
    interval_probability_floor: float | None = None
    interval_merge_gap_bars: int = 6
    min_interval_bars: int = 5
    max_interval_bars: int = 90
    interval_expand_atr: float = 0.35
    confidence_window: int = 20
    confidence_z: float = 1.0

    model_names: tuple[str, ...] = ("boosting", "logistic", "random_forest")
    use_hmm: bool = True
    hmm_states: int = 4
    use_arima: bool = False
    arima_order: tuple[int, int, int] = (1, 0, 1)
    arima_window: int = 160
    use_pca: bool = False
    pca_variance: float = 0.95
    permutation_repeats: int = 10


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")
    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if "volume" not in out.columns:
        out["volume"] = 0.0
    out["volume"] = out["volume"].fillna(0.0).clip(lower=0.0)

    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")
    return out


INTRADAY_INTERVALS = {"1m", "2m", "5m", "15m", "30m", "60m", "90m", "1h"}


def fetch_ohlcv(symbol: str, period: str, interval: str) -> pd.DataFrame:
    if Ticker is None:
        raise ImportError(
            "yahooquery is required for online data fetch. "
            "Install with: pip install yahooquery"
        )

    errors: list[str] = []
    for candidate_period in data_period_attempts(period=period, interval=interval):
        try:
            raw = Ticker(symbol).history(period=candidate_period, interval=interval)
            frame = normalize_yahoo_history(raw, symbol=symbol)
            data = prepare_ohlcv(frame)
            if candidate_period != period:
                warnings.warn(
                    f"No usable data for period={period!r}, interval={interval!r}; "
                    f"using period={candidate_period!r} instead."
                )
            return data
        except Exception as exc:
            errors.append(f"{candidate_period}: {type(exc).__name__}: {exc}")

    raise ValueError(
        f"No data returned for {symbol} with interval={interval}. "
        "For Yahoo intraday data, use a shorter period such as 60d/30d/7d "
        "or a wider interval such as 1d. Attempts: "
        + " | ".join(errors)
    )


def data_period_attempts(*, period: str, interval: str) -> list[str]:
    attempts = [str(period)]
    if str(interval).lower() in INTRADAY_INTERVALS:
        attempts.extend(["60d", "30d", "7d", "5d", "1mo"])
    else:
        attempts.extend(["2y", "1y", "6mo", "3mo"])
    return list(dict.fromkeys(attempts))


def normalize_yahoo_history(raw: Any, *, symbol: str) -> pd.DataFrame:
    if raw is None:
        raise ValueError("empty response")
    if isinstance(raw, dict):
        message = raw.get(symbol, raw)
        raise ValueError(f"Yahoo returned an error response: {message}")
    if isinstance(raw, pd.Series):
        raise ValueError(f"Yahoo returned a Series instead of OHLCV data: {raw.to_dict()}")
    if not isinstance(raw, pd.DataFrame):
        raise ValueError(f"Unexpected Yahoo response type: {type(raw).__name__}")
    if raw.empty:
        raise ValueError("empty response")

    frame = raw.copy()
    if isinstance(frame.index, pd.MultiIndex):
        if "symbol" in frame.index.names:
            frame = frame.reset_index(level="symbol", drop=True)
        else:
            frame = frame.reset_index()
    return frame


def load_ohlcv(config: ReversalMLConfig) -> pd.DataFrame:
    if config.csv_path:
        return prepare_ohlcv(pd.read_csv(config.csv_path))
    return fetch_ohlcv(config.symbol, config.period, config.interval)


def make_feature_frame(data: pd.DataFrame, config: ReversalMLConfig) -> pd.DataFrame:
    close = data["close"].astype(float)
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    open_ = data["open"].astype(float)
    volume = data["volume"].astype(float)

    features = pd.DataFrame(index=data.index)
    log_close = np.log(close)
    log_return = log_close.diff()

    features["log_return_1"] = log_return
    for window in [2, 3, 5, 10, 20, 40, 80]:
        features[f"log_return_sum_{window}"] = log_return.rolling(window).sum()
        features[f"realized_vol_{window}"] = log_return.rolling(window).std()

    atr = average_true_range(data, period=14)
    features["atr"] = atr
    features["atr_pct"] = atr / close.replace(0, np.nan) * 100.0
    features["atr_pct_change_5"] = features["atr_pct"].pct_change(5)

    features["rsi_7"] = rsi(close, period=7)
    features["rsi_14"] = rsi(close, period=14)
    macd_line, macd_signal, macd_hist = macd(close)
    features["macd_line"] = macd_line
    features["macd_signal"] = macd_signal
    features["macd_hist"] = macd_hist
    features["stoch_k_14"] = stochastic_k(high, low, close, period=14)
    features["stoch_d_3"] = features["stoch_k_14"].rolling(3).mean()

    feature_windows = sorted({20, 50, 80, int(config.label_lookback)})
    for window in feature_windows:
        roll_high = high.rolling(window).max()
        roll_low = low.rolling(window).min()
        roll_mid = close.rolling(window).mean()
        roll_std = close.rolling(window).std()

        features[f"drawdown_from_high_{window}"] = (roll_high - close) / roll_high * 100.0
        features[f"rebound_from_low_{window}"] = (close - roll_low) / roll_low * 100.0
        features[f"dist_to_low_atr_{window}"] = (close - roll_low) / atr.replace(0, np.nan)
        features[f"dist_to_high_atr_{window}"] = (roll_high - close) / atr.replace(0, np.nan)
        features[f"bollinger_z_{window}"] = (close - roll_mid) / roll_std.replace(0, np.nan)
        features[f"range_pct_{window}"] = (roll_high - roll_low) / roll_mid.replace(0, np.nan) * 100.0
        features[f"slope_pct_{window}"] = rolling_slope_pct(close, window=window)

    features["volatility_expansion_5_20"] = (
        features["realized_vol_5"] / features["realized_vol_20"].replace(0, np.nan)
    )
    features["volatility_compression_20_80"] = (
        features["realized_vol_20"] / features["realized_vol_80"].replace(0, np.nan)
    )
    features["range_compression_20_80"] = (
        features["range_pct_20"] / features["range_pct_80"].replace(0, np.nan)
    )

    candle_range = (high - low).replace(0, np.nan)
    features["body_pct"] = (close - open_) / open_.replace(0, np.nan) * 100.0
    features["body_to_range"] = (close - open_).abs() / candle_range
    features["upper_wick_to_range"] = (high - np.maximum(open_, close)) / candle_range
    features["lower_wick_to_range"] = (np.minimum(open_, close) - low) / candle_range
    features["close_position_in_range"] = (close - low) / candle_range
    features["gap_pct"] = (open_ - close.shift(1)) / close.shift(1).replace(0, np.nan) * 100.0
    features["green_streak"] = signed_streak(close >= open_)

    features["log_volume"] = np.log1p(volume)
    features["volume_z_20"] = rolling_zscore(features["log_volume"], window=20)
    features["volume_ratio_20"] = volume / volume.rolling(20).mean().replace(0, np.nan)
    up_volume = volume.where(close >= open_, 0.0)
    down_volume = volume.where(close < open_, 0.0)
    rolling_volume_20 = volume.rolling(20).sum().replace(0, np.nan)
    features["up_volume_share_20"] = up_volume.rolling(20).sum() / rolling_volume_20
    features["down_volume_share_20"] = down_volume.rolling(20).sum() / rolling_volume_20
    features["volume_expansion_5_20"] = (
        volume.rolling(5).mean() / volume.rolling(20).mean().replace(0, np.nan)
    )
    obv = on_balance_volume(close, volume)
    features["obv_slope_20"] = rolling_slope_pct(obv, window=20)

    features["prior_down_context"] = (
        (features[f"drawdown_from_high_{config.label_lookback}"] >= config.min_prior_drawdown_pct)
        & (features[f"dist_to_low_atr_{config.label_lookback}"] <= config.max_distance_to_low_atr)
    ).astype(float)

    if config.use_arima:
        arima_features = walk_forward_arima_features(
            log_return,
            order=config.arima_order,
            window=config.arima_window,
        )
        features = features.join(arima_features)

    features = features.replace([np.inf, -np.inf], np.nan)
    return features


def make_reversal_labels(
    data: pd.DataFrame,
    features: pd.DataFrame,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    close = data["close"].to_numpy(dtype=float)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    atr = features["atr"].to_numpy(dtype=float)
    prior_drawdown = features[f"drawdown_from_high_{config.label_lookback}"].to_numpy(dtype=float)
    dist_to_low = features[f"dist_to_low_atr_{config.label_lookback}"].to_numpy(dtype=float)

    labels = pd.DataFrame(index=data.index)
    labels["reversal_label"] = np.nan
    labels["label_target_price"] = np.nan
    labels["label_stop_price"] = np.nan
    labels["label_hit_bar_idx"] = np.nan
    labels["label_hit_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["label_hit_type"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["prior_drawdown_pct"] = prior_drawdown
    labels["distance_to_low_atr"] = dist_to_low

    n = len(data)
    for bar_idx in range(n):
        if bar_idx + config.label_horizon >= n:
            continue
        if not np.isfinite(atr[bar_idx]) or atr[bar_idx] <= 0:
            continue

        target_price = close[bar_idx] + config.target_atr * atr[bar_idx]
        stop_price = close[bar_idx] - config.stop_atr * atr[bar_idx]
        labels.iloc[bar_idx, labels.columns.get_loc("label_target_price")] = target_price
        labels.iloc[bar_idx, labels.columns.get_loc("label_stop_price")] = stop_price

        in_reversal_context = (
            np.isfinite(prior_drawdown[bar_idx])
            and np.isfinite(dist_to_low[bar_idx])
            and prior_drawdown[bar_idx] >= config.min_prior_drawdown_pct
            and dist_to_low[bar_idx] <= config.max_distance_to_low_atr
        )
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=bar_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        label = int(in_reversal_context and hit_type == "target")
        labels.iloc[bar_idx, labels.columns.get_loc("reversal_label")] = label
        if hit_idx is not None:
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_bar_idx")] = int(hit_idx)
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_date")] = data.index[int(hit_idx)]
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_type")] = hit_type

    return labels


def first_long_barrier_hit(
    *,
    high: np.ndarray,
    low: np.ndarray,
    start_idx: int,
    horizon: int,
    target_price: float,
    stop_price: float,
) -> tuple[str, int | None]:
    max_idx = min(len(high) - 1, int(start_idx) + int(horizon))
    for idx in range(int(start_idx) + 1, max_idx + 1):
        stop_hit = low[idx] <= stop_price
        target_hit = high[idx] >= target_price
        if stop_hit and target_hit:
            return "stop", idx
        if stop_hit:
            return "stop", idx
        if target_hit:
            return "target", idx
    return "timeout", None


def add_hmm_features(
    features: pd.DataFrame,
    train_index: pd.Index,
    config: ReversalMLConfig,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    state_summary_columns = ["state", "train_count", "mean_return", "mean_vol", "mean_drawdown"]
    if not config.use_hmm:
        return features, pd.DataFrame(columns=state_summary_columns)
    if GaussianHMM is None or SimpleImputer is None or StandardScaler is None:
        warnings.warn("hmmlearn/sklearn unavailable; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    hmm_columns = [
        column
        for column in [
            "log_return_1",
            "realized_vol_10",
            "atr_pct",
            "volume_z_20",
            f"drawdown_from_high_{config.label_lookback}",
            f"dist_to_low_atr_{config.label_lookback}",
        ]
        if column in features.columns
    ]
    if len(hmm_columns) < 3:
        return features, pd.DataFrame(columns=state_summary_columns)

    train = features.loc[train_index, hmm_columns]
    if train.dropna(how="all").shape[0] < config.hmm_states * 20:
        warnings.warn("Not enough training rows for HMM; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    imputer = SimpleImputer(strategy="median")
    scaler = StandardScaler()
    train_values = scaler.fit_transform(imputer.fit_transform(train))
    all_values = scaler.transform(imputer.transform(features[hmm_columns]))

    hmm = GaussianHMM(
        n_components=config.hmm_states,
        covariance_type="diag",
        n_iter=200,
        tol=1e-4,
        random_state=config.random_state,
    )
    try:
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            hmm.fit(train_values)
            probabilities = hmm.predict_proba(all_values)
    except Exception as exc:
        warnings.warn(f"HMM regime features skipped after fit failure: {type(exc).__name__}: {exc}")
        return features, pd.DataFrame(columns=state_summary_columns)
    states = probabilities.argmax(axis=1)

    out = features.copy()
    out["hmm_state"] = states.astype(float)
    for state in range(config.hmm_states):
        out[f"hmm_state_prob_{state}"] = probabilities[:, state]

    state_rows: list[dict[str, Any]] = []
    train_augmented = out.loc[train_index].copy()
    for state in range(config.hmm_states):
        subset = train_augmented[train_augmented["hmm_state"] == state]
        state_rows.append(
            {
                "state": state,
                "train_count": int(len(subset)),
                "mean_return": float(subset["log_return_1"].mean()),
                "mean_vol": float(subset["realized_vol_10"].mean()),
                "mean_drawdown": float(
                    subset[f"drawdown_from_high_{config.label_lookback}"].mean()
                ),
            }
        )
    return out, pd.DataFrame(state_rows, columns=state_summary_columns)


def train_reversal_models(
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    config: ReversalMLConfig,
) -> dict[str, Any]:
    require_sklearn()

    labeled = features.join(labels[["reversal_label"]], how="inner")
    labeled = labeled.dropna(subset=["reversal_label"])
    if len(labeled) < 200:
        raise ValueError("Not enough labeled rows. Increase period or use a higher timeframe.")

    y = labeled["reversal_label"].astype(int)
    feature_columns = [
        column
        for column in labeled.columns
        if column != "reversal_label" and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]

    split_idx = int(len(labeled) * (1.0 - config.test_size_pct))
    split_idx = min(max(split_idx, 50), len(labeled) - 20)
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]

    features_with_hmm, hmm_summary = add_hmm_features(features, train_index, config)
    labeled = features_with_hmm.join(labels[["reversal_label"]], how="inner")
    labeled = labeled.dropna(subset=["reversal_label"])
    y = labeled["reversal_label"].astype(int)
    feature_columns = [
        column
        for column in labeled.columns
        if column != "reversal_label" and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]
    X_train = X.loc[train_index]
    y_train = y.loc[train_index]
    X_test = X.loc[test_index]
    y_test = y.loc[test_index]

    X_train_clip, clip_bounds = fit_winsorize(X_train)
    X_test_clip = apply_winsorize(X_test, clip_bounds)
    X_all_clip = apply_winsorize(features_with_hmm[feature_columns], clip_bounds)

    models = build_models(config)
    results: list[dict[str, Any]] = []
    fitted_models: dict[str, Pipeline] = {}

    for name, model in models.items():
        fitted = fit_pipeline_with_weights(model, X_train_clip, y_train)
        fitted_models[name] = fitted
        test_probability = positive_probability(fitted, X_test_clip)
        metrics = evaluate_probabilities(y_test, test_probability)
        cv_metrics = time_series_cv_score(
            model,
            X_train_clip,
            y_train,
            n_splits=config.n_splits,
        )
        result = {"model": name, **metrics, **cv_metrics}
        results.append(result)

    model_report = pd.DataFrame(results)
    best_name = choose_best_model(model_report)
    per_model_probabilities: dict[str, pd.Series] = {}
    per_model_thresholds: dict[str, float] = {}
    per_model_feature_importance: dict[str, pd.DataFrame] = {}
    per_model_signals: dict[str, pd.DataFrame] = {}
    per_model_intervals: dict[str, pd.DataFrame] = {}

    for name, fitted in fitted_models.items():
        all_probability = positive_probability(fitted, X_all_clip)
        test_probability = positive_probability(fitted, X_test_clip)
        model_threshold = config.probability_threshold
        if model_threshold is None:
            model_threshold = choose_probability_threshold(
                y_test,
                test_probability,
                target_precision=config.target_precision,
            )

        probability_series = pd.Series(
            all_probability,
            index=features_with_hmm.index,
            name=f"{name}_reversal_probability",
        )
        importance = permutation_xai(
            fitted,
            X_test_clip,
            y_test,
            feature_columns,
            repeats=config.permutation_repeats,
            random_state=config.random_state,
        )
        signals = build_signal_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=probability_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not signals.empty:
            signals.insert(0, "Model", name)
        intervals = build_interval_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=probability_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not intervals.empty:
            intervals.insert(0, "Model", name)

        per_model_probabilities[name] = probability_series
        per_model_thresholds[name] = float(model_threshold)
        per_model_feature_importance[name] = importance
        per_model_signals[name] = signals
        per_model_intervals[name] = intervals

    best_model = fitted_models[best_name]
    threshold = per_model_thresholds[best_name]
    probabilities = per_model_probabilities[best_name].rename("reversal_probability")
    signals = per_model_signals[best_name].copy()
    if "Model" in signals.columns:
        signals = signals.drop(columns=["Model"])
    intervals = per_model_intervals[best_name].copy()
    if "Model" in intervals.columns:
        intervals = intervals.drop(columns=["Model"])
    feature_importance = per_model_feature_importance[best_name]
    pca_loadings = extract_pca_loadings(best_model, feature_columns)

    return {
        "features": features_with_hmm,
        "labels": labels,
        "feature_columns": feature_columns,
        "model_report": model_report,
        "best_model_name": best_name,
        "best_model": best_model,
        "probabilities": probabilities,
        "threshold": threshold,
        "signals": signals,
        "intervals": intervals,
        "feature_importance": feature_importance,
        "per_model_probabilities": per_model_probabilities,
        "per_model_thresholds": per_model_thresholds,
        "per_model_signals": per_model_signals,
        "per_model_intervals": per_model_intervals,
        "per_model_feature_importance": per_model_feature_importance,
        "pca_loadings": pca_loadings,
        "hmm_state_summary": hmm_summary,
    }


def require_sklearn() -> None:
    if Pipeline is None:
        raise ImportError(
            "scikit-learn is required for ML training. Install with: pip install scikit-learn"
        )


def build_models(config: ReversalMLConfig) -> dict[str, Pipeline]:
    requested = {name.strip().lower() for name in config.model_names}
    models: dict[str, Pipeline] = {}

    if "boosting" in requested:
        models["boosting"] = make_pipeline(
            HistGradientBoostingClassifier(
                learning_rate=0.05,
                max_iter=350,
                max_leaf_nodes=15,
                l2_regularization=0.05,
                early_stopping=True,
                random_state=config.random_state,
            ),
            config,
            scale=False,
        )
    if "logistic" in requested:
        models["logistic"] = make_pipeline(
            LogisticRegression(
                max_iter=2500,
                class_weight="balanced",
                solver="lbfgs",
                random_state=config.random_state,
            ),
            config,
            scale=True,
        )
    if "random_forest" in requested:
        models["random_forest"] = make_pipeline(
            RandomForestClassifier(
                n_estimators=400,
                max_depth=7,
                min_samples_leaf=20,
                class_weight="balanced_subsample",
                random_state=config.random_state,
                n_jobs=-1,
            ),
            config,
            scale=False,
        )
    if not models:
        raise ValueError("No valid model names requested.")
    return models


def make_pipeline(estimator: Any, config: ReversalMLConfig, *, scale: bool) -> Pipeline:
    steps: list[tuple[str, Any]] = [("imputer", SimpleImputer(strategy="median"))]
    if scale or config.use_pca:
        steps.append(("scaler", StandardScaler()))
    if config.use_pca:
        steps.append(("pca", PCA(n_components=config.pca_variance, svd_solver="full")))
    steps.append(("model", estimator))
    return Pipeline(steps)


def fit_pipeline_with_weights(model: Pipeline, X: pd.DataFrame, y: pd.Series) -> Pipeline:
    sample_weight = compute_sample_weight(class_weight="balanced", y=y)
    try:
        return model.fit(X, y, model__sample_weight=sample_weight)
    except TypeError:
        return model.fit(X, y)


def positive_probability(model: Pipeline, X: pd.DataFrame) -> np.ndarray:
    if hasattr(model, "predict_proba"):
        probability = model.predict_proba(X)
        if probability.shape[1] == 1:
            return np.zeros(len(X), dtype=float)
        return probability[:, 1]
    decision = model.decision_function(X)
    return 1.0 / (1.0 + np.exp(-decision))


def evaluate_probabilities(y_true: pd.Series, probability: np.ndarray) -> dict[str, float]:
    out: dict[str, float] = {
        "test_average_precision": np.nan,
        "test_roc_auc": np.nan,
        "test_brier": np.nan,
        "test_balanced_accuracy_050": np.nan,
        "test_precision_050": np.nan,
        "test_recall_050": np.nan,
        "test_positive_rate": float(np.mean(y_true)),
    }
    if len(np.unique(y_true)) < 2:
        return out

    prediction = (probability >= 0.5).astype(int)
    out["test_average_precision"] = float(average_precision_score(y_true, probability))
    out["test_roc_auc"] = float(roc_auc_score(y_true, probability))
    out["test_brier"] = float(brier_score_loss(y_true, probability))
    out["test_balanced_accuracy_050"] = float(balanced_accuracy_score(y_true, prediction))
    out["test_precision_050"] = float(precision_score(y_true, prediction, zero_division=0))
    out["test_recall_050"] = float(recall_score(y_true, prediction, zero_division=0))
    out["test_accuracy_050"] = float(accuracy_score(y_true, prediction))
    return out


def time_series_cv_score(
    model: Pipeline,
    X: pd.DataFrame,
    y: pd.Series,
    *,
    n_splits: int,
) -> dict[str, float]:
    out = {"cv_average_precision": np.nan, "cv_roc_auc": np.nan}
    if len(X) < 200 or y.nunique() < 2:
        return out

    split_count = min(n_splits, max(2, len(X) // 200))
    splitter = TimeSeriesSplit(n_splits=split_count)
    ap_scores: list[float] = []
    auc_scores: list[float] = []
    for train_idx, test_idx in splitter.split(X):
        y_train = y.iloc[train_idx]
        y_valid = y.iloc[test_idx]
        if y_train.nunique() < 2 or y_valid.nunique() < 2:
            continue
        fitted = fit_pipeline_with_weights(clone(model), X.iloc[train_idx], y_train)
        prob = positive_probability(fitted, X.iloc[test_idx])
        ap_scores.append(float(average_precision_score(y_valid, prob)))
        auc_scores.append(float(roc_auc_score(y_valid, prob)))

    if ap_scores:
        out["cv_average_precision"] = float(np.mean(ap_scores))
    if auc_scores:
        out["cv_roc_auc"] = float(np.mean(auc_scores))
    return out


def choose_best_model(model_report: pd.DataFrame) -> str:
    ranking = model_report.copy()
    ranking["_rank_score"] = ranking["test_average_precision"].fillna(-1.0)
    ranking = ranking.sort_values(["_rank_score", "test_roc_auc"], ascending=False)
    return str(ranking.iloc[0]["model"])


def choose_probability_threshold(
    y_true: pd.Series,
    probability: np.ndarray,
    *,
    target_precision: float,
) -> float:
    if len(np.unique(y_true)) < 2:
        return 0.65

    precision, recall, thresholds = precision_recall_curve(y_true, probability)
    candidates: list[tuple[float, float, float]] = []
    for threshold, prec, rec in zip(thresholds, precision[:-1], recall[:-1]):
        if prec >= target_precision and rec > 0:
            candidates.append((float(threshold), float(prec), float(rec)))
    if not candidates:
        return float(np.quantile(probability, 0.90))
    candidates.sort(key=lambda item: (item[2], item[1]), reverse=True)
    return float(candidates[0][0])


def permutation_xai(
    model: Pipeline,
    X_test: pd.DataFrame,
    y_test: pd.Series,
    feature_columns: list[str],
    *,
    repeats: int,
    random_state: int,
) -> pd.DataFrame:
    columns = ["feature", "importance_mean", "importance_std", "rank"]
    if y_test.nunique() < 2 or permutation_importance is None:
        return pd.DataFrame(columns=columns)

    result = permutation_importance(
        model,
        X_test,
        y_test,
        n_repeats=repeats,
        random_state=random_state,
        scoring="average_precision",
        n_jobs=-1,
    )
    importance = pd.DataFrame(
        {
            "feature": feature_columns,
            "importance_mean": result.importances_mean,
            "importance_std": result.importances_std,
        }
    )
    importance = importance.sort_values("importance_mean", ascending=False).reset_index(drop=True)
    importance["rank"] = np.arange(1, len(importance) + 1)
    return importance[columns]


def extract_pca_loadings(model: Pipeline, feature_columns: list[str]) -> pd.DataFrame:
    if "pca" not in model.named_steps:
        return pd.DataFrame(columns=["component", "feature", "loading", "abs_loading", "rank"])

    pca = model.named_steps["pca"]
    rows: list[dict[str, Any]] = []
    for component_idx, component in enumerate(pca.components_):
        for feature, loading in zip(feature_columns, component):
            rows.append(
                {
                    "component": f"PC{component_idx + 1}",
                    "feature": feature,
                    "loading": float(loading),
                    "abs_loading": float(abs(loading)),
                }
            )
    out = pd.DataFrame(rows)
    out["rank"] = out.groupby("component")["abs_loading"].rank(ascending=False, method="first")
    return out.sort_values(["component", "rank"]).reset_index(drop=True)


def build_signal_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    candidate_probability = probabilities.dropna()
    candidate_probability = candidate_probability[candidate_probability >= threshold]
    if candidate_probability.empty:
        return empty_signal_frame()

    selected_indexes = non_max_suppression_1d(
        candidate_probability,
        cooldown_bars=config.cooldown_bars,
        all_index=features.index,
    )

    train_stats = make_train_feature_stats(train_features)
    rows: list[dict[str, Any]] = []
    atr = features["atr"]
    for timestamp in selected_indexes:
        bar_idx = int(data.index.get_loc(timestamp))
        entry_price = float(data.loc[timestamp, "close"])
        target_price = entry_price + config.target_atr * float(atr.loc[timestamp])
        stop_price = entry_price - config.stop_atr * float(atr.loc[timestamp])
        label_row = labels.loc[timestamp] if timestamp in labels.index else pd.Series(dtype=object)
        hit_idx = label_row.get("label_hit_bar_idx", np.nan)
        hit_date = label_row.get("label_hit_date", pd.NaT)
        hit_type = label_row.get("label_hit_type", None)
        realized_label = label_row.get("reversal_label", np.nan)
        top_drivers = local_driver_summary(
            features.loc[timestamp, feature_columns],
            feature_importance,
            train_stats,
            top_n=6,
        )
        rows.append(
            {
                "EntryBarIdx": bar_idx,
                "EntryDate": timestamp,
                "EntryPrice": entry_price,
                "ReversalProbability": float(probabilities.loc[timestamp]),
                "SignalThreshold": float(threshold),
                "TargetPrice": target_price,
                "StopPrice": stop_price,
                "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                "OutcomeKnown": is_finite_number(realized_label),
                "OutcomeLabel": realized_label,
                "HitBarIdx": int(hit_idx) if is_finite_number(hit_idx) else np.nan,
                "HitDate": hit_date,
                "HitType": hit_type,
                "ShapeHint": classify_shape_hint(data, bar_idx),
                "TopDrivers": top_drivers,
            }
        )
    return pd.DataFrame(rows).sort_values("EntryBarIdx").reset_index(drop=True)


def build_interval_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    aligned_probability = probabilities.reindex(features.index).astype(float)
    confidence = probability_confidence_frame(
        aligned_probability,
        window=config.confidence_window,
        z=config.confidence_z,
    )
    floor = config.interval_probability_floor
    if floor is None:
        floor = max(0.05, min(0.95, float(threshold) * 0.80))

    prior_context = features.get("prior_down_context", pd.Series(0.0, index=features.index))
    context_seen = prior_context.fillna(0.0).rolling(
        config.label_lookback,
        min_periods=1,
    ).max()
    candidate_mask = (
        (aligned_probability >= floor)
        | (
            (confidence["upper"] >= float(threshold))
            & (confidence["mean"] >= floor * 0.85)
        )
    )
    candidate_mask = (candidate_mask.fillna(False) & (context_seen > 0)).to_numpy(dtype=bool)
    segments = merge_boolean_segments(
        boolean_segments(candidate_mask),
        max_gap=max(0, int(config.interval_merge_gap_bars)),
    )
    if not segments:
        return empty_interval_frame()

    rows: list[dict[str, Any]] = []
    train_stats = make_train_feature_stats(train_features)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    atr = features["atr"].reindex(data.index).to_numpy(dtype=float)
    min_bars = max(1, int(config.min_interval_bars))
    max_bars = max(min_bars, int(config.max_interval_bars))

    for raw_start, raw_end in segments:
        for seg_start, seg_end in split_segment(raw_start, raw_end, max_bars=max_bars, min_bars=min_bars):
            segment_probability = aligned_probability.iloc[seg_start : seg_end + 1].dropna()
            if segment_probability.empty:
                continue
            segment_confidence = confidence.iloc[seg_start : seg_end + 1]
            if (
                float(segment_probability.max()) < float(threshold)
                and float(segment_confidence["upper"].max()) < float(threshold)
            ):
                continue

            peak_timestamp = segment_probability.idxmax()
            peak_idx = int(features.index.get_loc(peak_timestamp))
            search_start = max(0, peak_idx - max_bars + 1)
            search_start = max(search_start, seg_start - config.interval_merge_gap_bars)
            search_end = max(seg_end, peak_idx)
            bottom_idx = search_start + int(np.nanargmin(low[search_start : search_end + 1]))
            start_idx = min(seg_start, bottom_idx)
            end_idx = max(seg_end, peak_idx)

            if end_idx - start_idx + 1 < min_bars:
                start_idx = max(0, peak_idx - min_bars // 2)
                end_idx = min(len(data) - 1, start_idx + min_bars - 1)
            if end_idx - start_idx + 1 > max_bars:
                start_idx = max(0, end_idx - max_bars + 1)

            context_start = max(0, start_idx - config.label_lookback // 2)
            context_slice = prior_context.iloc[context_start : peak_idx + 1].fillna(0.0)
            if context_slice.max() <= 0:
                continue

            interval_probability = aligned_probability.iloc[start_idx : end_idx + 1].dropna()
            if interval_probability.empty:
                continue
            probability_mean = float(interval_probability.mean())
            probability_max = float(interval_probability.max())
            interval_confidence_low, interval_confidence_high = probability_interval_bounds(
                interval_probability,
                z=config.confidence_z,
            )
            if probability_max < float(threshold) and interval_confidence_high < float(threshold):
                continue

            support_low = finite_min(low[start_idx : end_idx + 1])
            resistance_high = finite_max(high[start_idx : end_idx + 1])
            if not np.isfinite(support_low) or not np.isfinite(resistance_high):
                continue

            entry_price = float(close[end_idx])
            interval_atr = finite_mean(atr[start_idx : end_idx + 1])
            if not np.isfinite(interval_atr) or interval_atr <= 0:
                interval_atr = max((resistance_high - support_low) * 0.10, abs(entry_price) * 0.001, 1e-12)
            price_pad = max(0.0, float(config.interval_expand_atr)) * interval_atr
            target_price = entry_price + config.target_atr * interval_atr
            stop_price = entry_price - config.stop_atr * interval_atr
            peak_label = labels.loc[peak_timestamp] if peak_timestamp in labels.index else pd.Series(dtype=object)
            hit_idx = peak_label.get("label_hit_bar_idx", np.nan)
            realized_label = peak_label.get("reversal_label", np.nan)
            top_drivers = local_driver_summary(
                features.loc[peak_timestamp, feature_columns],
                feature_importance,
                train_stats,
                top_n=6,
            )

            volume_z_max = finite_max(series_values(features, "volume_z_20", start_idx, end_idx))
            volume_ratio_max = finite_max(series_values(features, "volume_ratio_20", start_idx, end_idx))
            volatility_expansion = finite_mean(
                series_values(features, "volatility_expansion_5_20", start_idx, end_idx)
            )
            volatility_compression = finite_mean(
                series_values(features, "volatility_compression_20_80", start_idx, end_idx)
            )
            range_compression = finite_mean(
                series_values(features, "range_compression_20_80", start_idx, end_idx)
            )
            context_ratio = float(prior_context.iloc[start_idx : end_idx + 1].fillna(0.0).mean())
            interval_score = score_interval(
                probability_mean=probability_mean,
                probability_max=probability_max,
                context_ratio=context_ratio,
                volume_z_max=volume_z_max,
                volatility_expansion=volatility_expansion,
                volatility_compression=volatility_compression,
                confidence_low=interval_confidence_low,
                confidence_high=interval_confidence_high,
            )

            rows.append(
                {
                    "StartBarIdx": int(start_idx),
                    "EndBarIdx": int(end_idx),
                    "PeakBarIdx": int(peak_idx),
                    "StartDate": data.index[start_idx],
                    "EndDate": data.index[end_idx],
                    "PeakDate": peak_timestamp,
                    "Bars": int(end_idx - start_idx + 1),
                    "SupportLow": support_low,
                    "ResistanceHigh": resistance_high,
                    "BoxLow": support_low - price_pad,
                    "BoxHigh": resistance_high + price_pad,
                    "EntryBarIdx": int(end_idx),
                    "EntryDate": data.index[end_idx],
                    "EntryPrice": entry_price,
                    "TargetPrice": target_price,
                    "StopPrice": stop_price,
                    "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                    "ProbabilityMean": probability_mean,
                    "ProbabilityMax": probability_max,
                    "ProbabilityConfidenceLow": interval_confidence_low,
                    "ProbabilityConfidenceHigh": interval_confidence_high,
                    "SignalThreshold": float(threshold),
                    "VolumeZMax": volume_z_max,
                    "VolumeRatioMax": volume_ratio_max,
                    "VolatilityExpansionMean": volatility_expansion,
                    "VolatilityCompressionMean": volatility_compression,
                    "RangeCompressionMean": range_compression,
                    "PriorDownContextRatio": context_ratio,
                    "IntervalScore": interval_score,
                    "OutcomeKnown": is_finite_number(realized_label),
                    "OutcomeLabel": realized_label,
                    "HitBarIdx": int(hit_idx) if is_finite_number(hit_idx) else np.nan,
                    "HitDate": peak_label.get("label_hit_date", pd.NaT),
                    "HitType": peak_label.get("label_hit_type", None),
                    "ShapeHint": classify_shape_hint(data, peak_idx),
                    "TopDrivers": top_drivers,
                }
            )

    if not rows:
        return empty_interval_frame()
    rows = suppress_overlapping_interval_rows(rows)
    return pd.DataFrame(rows).sort_values("StartBarIdx").reset_index(drop=True)


def empty_signal_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "ReversalProbability",
            "SignalThreshold",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def empty_interval_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "StartBarIdx",
            "EndBarIdx",
            "PeakBarIdx",
            "StartDate",
            "EndDate",
            "PeakDate",
            "Bars",
            "SupportLow",
            "ResistanceHigh",
            "BoxLow",
            "BoxHigh",
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "ProbabilityMean",
            "ProbabilityMax",
            "ProbabilityConfidenceLow",
            "ProbabilityConfidenceHigh",
            "SignalThreshold",
            "VolumeZMax",
            "VolumeRatioMax",
            "VolatilityExpansionMean",
            "VolatilityCompressionMean",
            "RangeCompressionMean",
            "PriorDownContextRatio",
            "IntervalScore",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def non_max_suppression_1d(
    probability: pd.Series,
    *,
    cooldown_bars: int,
    all_index: pd.Index,
) -> list[pd.Timestamp]:
    ranked = probability.sort_values(ascending=False)
    selected_positions: list[int] = []
    selected_timestamps: list[pd.Timestamp] = []

    for timestamp, _ in ranked.items():
        position = int(all_index.get_loc(timestamp))
        if any(abs(position - selected) <= cooldown_bars for selected in selected_positions):
            continue
        selected_positions.append(position)
        selected_timestamps.append(timestamp)

    selected_timestamps.sort(key=lambda ts: int(all_index.get_loc(ts)))
    return selected_timestamps


def probability_confidence_frame(probability: pd.Series, *, window: int, z: float) -> pd.DataFrame:
    values = probability.astype(float)
    window = max(2, int(window))
    mean = values.rolling(window, min_periods=max(2, window // 4)).mean()
    std = values.rolling(window, min_periods=max(2, window // 4)).std().fillna(0.0)
    count = values.rolling(window, min_periods=max(2, window // 4)).count().replace(0, np.nan)
    stderr = std / np.sqrt(count)
    return pd.DataFrame(
        {
            "mean": mean,
            "lower": (mean - float(z) * stderr).clip(0.0, 1.0),
            "upper": (mean + float(z) * stderr).clip(0.0, 1.0),
        },
        index=probability.index,
    )


def probability_interval_bounds(probability: pd.Series, *, z: float) -> tuple[float, float]:
    values = probability.dropna().to_numpy(dtype=float)
    if len(values) == 0:
        return np.nan, np.nan
    mean = float(np.mean(values))
    if len(values) == 1:
        return mean, mean
    stderr = float(np.std(values, ddof=1) / math.sqrt(len(values)))
    return max(0.0, mean - float(z) * stderr), min(1.0, mean + float(z) * stderr)


def boolean_segments(mask: np.ndarray) -> list[tuple[int, int]]:
    segments: list[tuple[int, int]] = []
    start: int | None = None
    for idx, active in enumerate(mask):
        if active and start is None:
            start = idx
        elif not active and start is not None:
            segments.append((start, idx - 1))
            start = None
    if start is not None:
        segments.append((start, len(mask) - 1))
    return segments


def merge_boolean_segments(
    segments: list[tuple[int, int]],
    *,
    max_gap: int,
) -> list[tuple[int, int]]:
    if not segments:
        return []
    merged: list[tuple[int, int]] = [segments[0]]
    for start, end in segments[1:]:
        previous_start, previous_end = merged[-1]
        if start - previous_end - 1 <= max_gap:
            merged[-1] = (previous_start, end)
        else:
            merged.append((start, end))
    return merged


def split_segment(
    start: int,
    end: int,
    *,
    max_bars: int,
    min_bars: int,
) -> list[tuple[int, int]]:
    length = end - start + 1
    if length <= max_bars:
        return [(start, end)]

    chunks: list[tuple[int, int]] = []
    overlap = max(min_bars, max_bars // 4)
    chunk_start = start
    while chunk_start <= end:
        chunk_end = min(end, chunk_start + max_bars - 1)
        chunks.append((chunk_start, chunk_end))
        if chunk_end >= end:
            break
        chunk_start = max(chunk_start + 1, chunk_end - overlap + 1)
    return chunks


def series_values(features: pd.DataFrame, column: str, start_idx: int, end_idx: int) -> np.ndarray:
    if column not in features.columns:
        return np.array([], dtype=float)
    return features[column].iloc[start_idx : end_idx + 1].to_numpy(dtype=float)


def is_finite_number(value: Any) -> bool:
    try:
        return bool(np.isfinite(float(value)))
    except (TypeError, ValueError):
        return False


def finite_mean(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.mean(array))


def finite_min(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.min(array))


def finite_max(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.max(array))


def score_interval(
    *,
    probability_mean: float,
    probability_max: float,
    context_ratio: float,
    volume_z_max: float,
    volatility_expansion: float,
    volatility_compression: float,
    confidence_low: float,
    confidence_high: float,
) -> float:
    volume_score = np.tanh(max(0.0, 0.0 if not np.isfinite(volume_z_max) else volume_z_max) / 3.0)
    expansion_score = np.tanh(
        max(0.0, (0.0 if not np.isfinite(volatility_expansion) else volatility_expansion) - 1.0)
    )
    compression_score = 1.0 - min(
        1.0,
        max(0.0, 0.0 if not np.isfinite(volatility_compression) else volatility_compression),
    )
    confidence_width = max(
        0.0,
        (0.0 if not np.isfinite(confidence_high) else confidence_high)
        - (0.0 if not np.isfinite(confidence_low) else confidence_low),
    )
    stability_score = 1.0 - min(1.0, confidence_width)
    score = (
        0.38 * probability_max
        + 0.27 * probability_mean
        + 0.15 * min(1.0, max(0.0, context_ratio))
        + 0.07 * volume_score
        + 0.06 * expansion_score
        + 0.04 * compression_score
        + 0.03 * stability_score
    )
    return float(score)


def suppress_overlapping_interval_rows(
    rows: list[dict[str, Any]],
    *,
    max_overlap_ratio: float = 0.55,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []
    for row in sorted(rows, key=lambda item: float(item["IntervalScore"]), reverse=True):
        start = int(row["StartBarIdx"])
        end = int(row["EndBarIdx"])
        length = max(1, end - start + 1)
        overlaps = False
        for existing in selected:
            existing_start = int(existing["StartBarIdx"])
            existing_end = int(existing["EndBarIdx"])
            existing_length = max(1, existing_end - existing_start + 1)
            overlap = max(0, min(end, existing_end) - max(start, existing_start) + 1)
            if overlap / min(length, existing_length) >= max_overlap_ratio:
                overlaps = True
                break
        if not overlaps:
            selected.append(row)
    return selected


def make_train_feature_stats(train_features: pd.DataFrame) -> pd.DataFrame:
    stats = pd.DataFrame(index=train_features.columns)
    stats["median"] = train_features.median(numeric_only=True)
    q25 = train_features.quantile(0.25, numeric_only=True)
    q75 = train_features.quantile(0.75, numeric_only=True)
    stats["iqr"] = (q75 - q25).replace(0, np.nan)
    return stats


def local_driver_summary(
    row: pd.Series,
    feature_importance: pd.DataFrame,
    train_stats: pd.DataFrame,
    *,
    top_n: int,
) -> str:
    if feature_importance.empty:
        return ""

    importance_map = feature_importance.set_index("feature")["importance_mean"].to_dict()
    scored: list[tuple[float, str]] = []
    for feature, value in row.items():
        if feature not in importance_map or feature not in train_stats.index:
            continue
        if not np.isfinite(value):
            continue
        iqr = train_stats.loc[feature, "iqr"]
        median = train_stats.loc[feature, "median"]
        if not np.isfinite(iqr) or iqr == 0:
            continue
        robust_z = (float(value) - float(median)) / float(iqr)
        score = abs(float(importance_map[feature]) * robust_z)
        if score <= 0:
            continue
        scored.append((score, f"{feature}={float(value):.4g} z={robust_z:.2f}"))

    scored.sort(reverse=True)
    return " | ".join(text for _, text in scored[:top_n])


def classify_shape_hint(data: pd.DataFrame, bar_idx: int, *, lookback: int = 80) -> str:
    start = max(0, int(bar_idx) - lookback)
    if int(bar_idx) - start < 20:
        return "ML Bullish Reversal"

    window = data.iloc[start : int(bar_idx) + 1]
    lows = window["low"].to_numpy(dtype=float)
    highs = window["high"].to_numpy(dtype=float)
    close = window["close"].to_numpy(dtype=float)
    atr = float(average_true_range(window, period=14).iloc[-1])
    if not np.isfinite(atr) or atr <= 0:
        atr = max(float(np.nanstd(close)), 1e-12)

    pivot_lows = simple_pivots(lows, mode="low", order=3)
    pivot_highs = simple_pivots(highs, mode="high", order=3)
    if len(pivot_lows) >= 2:
        l1, l2 = pivot_lows[-2], pivot_lows[-1]
        low_diff_atr = abs(lows[l2] - lows[l1]) / atr
        if l2 > l1 and low_diff_atr <= 1.0:
            between_high = np.max(highs[l1:l2 + 1])
            if between_high - max(lows[l1], lows[l2]) >= atr:
                return "ML Double Bottom-like"

    if len(pivot_lows) >= 3 and len(pivot_highs) >= 2:
        left, head, right = pivot_lows[-3], pivot_lows[-2], pivot_lows[-1]
        if left < head < right and lows[head] < min(lows[left], lows[right]):
            shoulder_diff_atr = abs(lows[left] - lows[right]) / atr
            if shoulder_diff_atr <= 1.5:
                return "ML Inverse H&S-like"

    return "ML Bullish Reversal"


def simple_pivots(values: np.ndarray, *, mode: str, order: int) -> list[int]:
    pivots: list[int] = []
    for idx in range(order, len(values) - order):
        left = values[idx - order : idx]
        right = values[idx + 1 : idx + order + 1]
        if mode == "low" and values[idx] <= np.min(left) and values[idx] <= np.min(right):
            pivots.append(idx)
        if mode == "high" and values[idx] >= np.max(left) and values[idx] >= np.max(right):
            pivots.append(idx)
    return pivots


def fit_winsorize(
    X_train: pd.DataFrame,
    *,
    lower: float = 0.01,
    upper: float = 0.99,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    bounds = pd.DataFrame(index=X_train.columns)
    bounds["lower"] = X_train.quantile(lower, numeric_only=True)
    bounds["upper"] = X_train.quantile(upper, numeric_only=True)
    return apply_winsorize(X_train, bounds), bounds


def apply_winsorize(X: pd.DataFrame, bounds: pd.DataFrame) -> pd.DataFrame:
    out = X.copy()
    for column in out.columns:
        if column not in bounds.index:
            continue
        lower = bounds.loc[column, "lower"]
        upper = bounds.loc[column, "upper"]
        if np.isfinite(lower) and np.isfinite(upper) and lower < upper:
            out[column] = out[column].clip(lower=lower, upper=upper)
    return out


def average_true_range(data: pd.DataFrame, *, period: int = 14) -> pd.Series:
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    close = data["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [
            high - low,
            (high - previous_close).abs(),
            (low - previous_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    return true_range.rolling(period, min_periods=1).mean()


def rsi(close: pd.Series, *, period: int = 14) -> pd.Series:
    delta = close.diff()
    gains = delta.clip(lower=0.0)
    losses = -delta.clip(upper=0.0)
    avg_gain = gains.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    avg_loss = losses.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    rs = avg_gain / avg_loss.replace(0, np.nan)
    return 100.0 - (100.0 / (1.0 + rs))


def macd(close: pd.Series) -> tuple[pd.Series, pd.Series, pd.Series]:
    fast = close.ewm(span=12, adjust=False).mean()
    slow = close.ewm(span=26, adjust=False).mean()
    line = fast - slow
    signal = line.ewm(span=9, adjust=False).mean()
    return line, signal, line - signal


def stochastic_k(
    high: pd.Series,
    low: pd.Series,
    close: pd.Series,
    *,
    period: int,
) -> pd.Series:
    lowest = low.rolling(period).min()
    highest = high.rolling(period).max()
    return (close - lowest) / (highest - lowest).replace(0, np.nan) * 100.0


def rolling_zscore(series: pd.Series, *, window: int) -> pd.Series:
    mean = series.rolling(window).mean()
    std = series.rolling(window).std()
    return (series - mean) / std.replace(0, np.nan)


def rolling_slope_pct(series: pd.Series, *, window: int) -> pd.Series:
    x = np.arange(window, dtype=float)
    x = x - x.mean()

    def slope(values: np.ndarray) -> float:
        if np.isnan(values).any():
            return np.nan
        y = values.astype(float)
        y = y - y.mean()
        denom = float(np.dot(x, x))
        if denom == 0:
            return np.nan
        raw_slope = float(np.dot(x, y) / denom)
        return raw_slope / max(abs(float(values[-1])), 1e-12) * 100.0

    return series.rolling(window).apply(slope, raw=True)


def signed_streak(condition: pd.Series) -> pd.Series:
    values = condition.fillna(False).to_numpy(dtype=bool)
    streaks = np.zeros(len(values), dtype=float)
    current = 0
    previous = None
    for idx, value in enumerate(values):
        if previous is None or value != previous:
            current = 1
        else:
            current += 1
        streaks[idx] = current if value else -current
        previous = value
    return pd.Series(streaks, index=condition.index)


def on_balance_volume(close: pd.Series, volume: pd.Series) -> pd.Series:
    direction = np.sign(close.diff()).fillna(0.0)
    return (direction * volume).cumsum()


def walk_forward_arima_features(
    returns: pd.Series,
    *,
    order: tuple[int, int, int],
    window: int,
) -> pd.DataFrame:
    columns = ["arima_forecast_return", "arima_residual", "arima_residual_z_40"]
    out = pd.DataFrame(np.nan, index=returns.index, columns=columns)
    if ARIMA is None:
        warnings.warn("statsmodels unavailable; ARIMA features skipped.")
        return out

    values = returns.astype(float)
    for idx in range(window, len(values)):
        history = values.iloc[idx - window : idx].dropna()
        if len(history) < max(30, sum(order) * 10):
            continue
        try:
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                result = ARIMA(history, order=order).fit()
            forecast = float(result.forecast(1).iloc[0])
            residual = float(values.iloc[idx] - forecast)
            out.iloc[idx, out.columns.get_loc("arima_forecast_return")] = forecast
            out.iloc[idx, out.columns.get_loc("arima_residual")] = residual
        except Exception:
            continue
    out["arima_residual_z_40"] = rolling_zscore(out["arima_residual"], window=40)
    return out


def save_outputs(
    data: pd.DataFrame,
    result: dict[str, Any],
    config: ReversalMLConfig,
) -> Path:
    run_name = f"{config.symbol.replace('.', '_')}_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
    run_dir = Path(config.output_dir) / run_name
    run_dir.mkdir(parents=True, exist_ok=True)
    if go is None:
        warnings.warn("plotly unavailable; HTML plot files skipped. Install with: pip install plotly")

    result["signals"].to_csv(run_dir / "signals.csv", index=False)
    result["intervals"].to_csv(run_dir / "pattern_intervals.csv", index=False)
    result["model_report"].to_csv(run_dir / "model_report.csv", index=False)
    result["feature_importance"].to_csv(run_dir / "feature_importance.csv", index=False)
    result["hmm_state_summary"].to_csv(run_dir / "hmm_state_summary.csv", index=False)
    result["pca_loadings"].to_csv(run_dir / "pca_loadings.csv", index=False)

    probability_frame = pd.DataFrame(
        {
            "Date": result["probabilities"].index,
            "ReversalProbability": result["probabilities"].to_numpy(dtype=float),
        }
    )
    probability_frame.to_csv(run_dir / "probabilities.csv", index=False)

    all_probability_frame = pd.DataFrame({"Date": data.index})
    for model_name, probabilities in result["per_model_probabilities"].items():
        all_probability_frame[f"{model_name}_probability"] = probabilities.reindex(data.index).to_numpy(dtype=float)
    all_probability_frame.to_csv(run_dir / "all_model_probabilities.csv", index=False)

    all_interval_frames = [
        intervals
        for intervals in result["per_model_intervals"].values()
        if intervals is not None and not intervals.empty
    ]
    if all_interval_frames:
        pd.concat(all_interval_frames, ignore_index=True).to_csv(
            run_dir / "all_model_pattern_intervals.csv",
            index=False,
        )
    else:
        empty_interval_frame().to_csv(run_dir / "all_model_pattern_intervals.csv", index=False)

    models_dir = run_dir / "models"
    models_dir.mkdir(exist_ok=True)
    for model_name, probabilities in result["per_model_probabilities"].items():
        safe_model_name = safe_filename(model_name)
        signals = result["per_model_signals"].get(model_name, empty_signal_frame())
        importance = result["per_model_feature_importance"].get(
            model_name,
            pd.DataFrame(columns=["feature", "importance_mean", "importance_std", "rank"]),
        )
        threshold = result["per_model_thresholds"].get(model_name, np.nan)
        intervals = result["per_model_intervals"].get(model_name, empty_interval_frame())
        signals.to_csv(models_dir / f"{safe_model_name}_signals.csv", index=False)
        intervals.to_csv(models_dir / f"{safe_model_name}_pattern_intervals.csv", index=False)
        importance.to_csv(models_dir / f"{safe_model_name}_feature_importance.csv", index=False)
        make_plot(
            data=data,
            probabilities=probabilities,
            signals=signals,
            intervals=intervals,
            title=f"{config.symbol} {model_name} ML Reversal Signals",
            output_path=models_dir / f"{safe_model_name}_plot.html",
            model_name=model_name,
            threshold=threshold,
            confidence_window=config.confidence_window,
            confidence_z=config.confidence_z,
        )
        make_feature_importance_plot(
            importance=importance,
            title=f"{config.symbol} {model_name} XAI Feature Importance",
            output_path=models_dir / f"{safe_model_name}_feature_importance.html",
        )

    with open(run_dir / "config.json", "w", encoding="utf-8") as handle:
        json.dump(
            {
                **asdict(config),
                "best_model_name": result["best_model_name"],
                "threshold": result["threshold"],
            },
            handle,
            indent=2,
            default=str,
        )

    plot_path = run_dir / "reversal_ml_plot.html"
    make_plot(
        data=data,
        probabilities=result["probabilities"],
        signals=result["signals"],
        intervals=result["intervals"],
        title=f"{config.symbol} ML Reversal Signals",
        output_path=plot_path,
        model_name=f"best: {result['best_model_name']}",
        threshold=result["threshold"],
        confidence_window=config.confidence_window,
        confidence_z=config.confidence_z,
    )
    make_model_comparison_plot(
        data=data,
        per_model_probabilities=result["per_model_probabilities"],
        per_model_signals=result["per_model_signals"],
        per_model_intervals=result["per_model_intervals"],
        per_model_thresholds=result["per_model_thresholds"],
        title=f"{config.symbol} ML Reversal Model Comparison",
        output_path=run_dir / "all_models_comparison_plot.html",
    )
    return run_dir


def safe_filename(value: str) -> str:
    safe = "".join(character if character.isalnum() or character in {"-", "_"} else "_" for character in value)
    return safe.strip("_") or "model"


def hex_to_rgba(hex_color: str, alpha: float) -> str:
    value = hex_color.lstrip("#")
    if len(value) != 6:
        return f"rgba(56, 189, 248, {alpha})"
    red, green, blue = (int(value[idx : idx + 2], 16) for idx in (0, 2, 4))
    return f"rgba({red}, {green}, {blue}, {alpha})"


def add_interval_boxes(
    fig: Any,
    intervals: pd.DataFrame | None,
    *,
    color: str,
    label: str,
    opacity: float,
    show_text: bool,
) -> None:
    if intervals is None or intervals.empty:
        return

    center_x: list[float] = []
    center_y: list[float] = []
    hover_text: list[str] = []
    visible_text: list[str] = []
    for interval_number, (_, row) in enumerate(intervals.iterrows(), start=1):
        try:
            start_idx = int(row["StartBarIdx"])
            end_idx = int(row["EndBarIdx"])
            box_low = float(row["BoxLow"])
            box_high = float(row["BoxHigh"])
        except (KeyError, TypeError, ValueError):
            continue
        if not all(np.isfinite(value) for value in [start_idx, end_idx, box_low, box_high]):
            continue

        fig.add_shape(
            type="rect",
            xref="x",
            yref="y",
            x0=start_idx - 0.5,
            x1=end_idx + 0.5,
            y0=box_low,
            y1=box_high,
            fillcolor=hex_to_rgba(color, opacity),
            line=dict(color=color, width=2),
            layer="below",
        )
        center_x.append((start_idx + end_idx) / 2)
        center_y.append(box_high)
        probability_max = float(row.get("ProbabilityMax", np.nan))
        probability_mean = float(row.get("ProbabilityMean", np.nan))
        visible_text.append(f"Box {interval_number} {probability_max:.2f}")
        hover_text.append(
            (
                f"{label} interval<br>"
                f"Start: {row.get('StartDate')}<br>"
                f"End: {row.get('EndDate')}<br>"
                f"Peak: {row.get('PeakDate')}<br>"
                f"Mean probability: {probability_mean:.3f}<br>"
                f"Max probability: {probability_max:.3f}<br>"
                f"CI: {float(row.get('ProbabilityConfidenceLow', np.nan)):.3f}"
                f" - {float(row.get('ProbabilityConfidenceHigh', np.nan)):.3f}<br>"
                f"Volume z max: {float(row.get('VolumeZMax', np.nan)):.3f}<br>"
                f"Vol expansion: {float(row.get('VolatilityExpansionMean', np.nan)):.3f}<br>"
                f"Score: {float(row.get('IntervalScore', np.nan)):.3f}<br>"
                f"Shape: {row.get('ShapeHint')}<br>"
                f"Drivers: {row.get('TopDrivers')}"
            )
        )

    if not center_x:
        return
    mode = "markers+text" if show_text else "markers"
    fig.add_trace(
        go.Scatter(
            x=center_x,
            y=center_y,
            mode=mode,
            name=f"{label} pattern interval",
            marker=dict(
                color=color,
                size=11,
                symbol="square-open",
                line=dict(width=2),
            ),
            text=visible_text if show_text else None,
            textposition="top center",
            hovertext=hover_text,
            hoverinfo="text",
        )
    )


def make_plot(
    *,
    data: pd.DataFrame,
    probabilities: pd.Series,
    signals: pd.DataFrame,
    intervals: pd.DataFrame | None = None,
    title: str,
    output_path: Path,
    model_name: str | None = None,
    threshold: float | None = None,
    confidence_window: int = 20,
    confidence_z: float = 1.0,
) -> None:
    if go is None:
        return

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )
    add_interval_boxes(
        fig,
        intervals,
        color="#FACC15",
        label=model_name or "ML",
        opacity=0.16,
        show_text=True,
    )
    aligned_probability = probabilities.reindex(data.index).astype(float)
    probability_ci = probability_confidence_frame(
        aligned_probability,
        window=confidence_window,
        z=confidence_z,
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probability_ci["lower"].to_numpy(dtype=float),
            name="Probability CI lower",
            mode="lines",
            yaxis="y2",
            line=dict(width=0),
            hoverinfo="skip",
            showlegend=False,
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probability_ci["upper"].to_numpy(dtype=float),
            name="Probability confidence band",
            mode="lines",
            yaxis="y2",
            fill="tonexty",
            fillcolor="rgba(56, 189, 248, 0.14)",
            line=dict(width=0),
            hoverinfo="skip",
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=aligned_probability.to_numpy(dtype=float),
            name=f"{model_name or 'Model'} probability",
            mode="lines",
            yaxis="y2",
            line=dict(color="#38BDF8", width=2),
        )
    )
    if threshold is not None and np.isfinite(float(threshold)):
        fig.add_trace(
            go.Scatter(
                x=[plot_x[0], plot_x[-1]],
                y=[float(threshold), float(threshold)],
                name="Signal threshold",
                mode="lines",
                yaxis="y2",
                line=dict(color="#FACC15", width=1, dash="dash"),
                hoverinfo="skip",
            )
        )

    if not signals.empty:
        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers+text",
                name=f"{model_name or 'ML'} Buy Signal",
                marker=dict(color="#FACC15", size=12, symbol="circle"),
                text=[f"{prob:.2f}" for prob in signals["ReversalProbability"]],
                textposition="top center",
                hovertext=[
                    (
                        f"ML Buy<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Probability: {row['ReversalProbability']:.3f}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Probability",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=65, t=80, b=55),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def make_feature_importance_plot(
    *,
    importance: pd.DataFrame,
    title: str,
    output_path: Path,
    top_n: int = 25,
) -> None:
    if go is None or importance.empty:
        return

    plot_frame = importance.head(top_n).copy()
    plot_frame = plot_frame.sort_values("importance_mean", ascending=True)
    fig = go.Figure()
    fig.add_trace(
        go.Bar(
            x=plot_frame["importance_mean"],
            y=plot_frame["feature"],
            orientation="h",
            error_x=dict(
                type="data",
                array=plot_frame["importance_std"].clip(lower=0).to_numpy(dtype=float),
                visible=True,
                color="#94A3B8",
            ),
            marker=dict(color="#38BDF8"),
            hovertext=[
                (
                    f"Feature: {row['feature']}<br>"
                    f"Importance: {row['importance_mean']:.6f}<br>"
                    f"Std: {row['importance_std']:.6f}<br>"
                    f"Rank: {int(row['rank'])}"
                )
                for _, row in plot_frame.iterrows()
            ],
            hoverinfo="text",
            name="Permutation importance",
        )
    )
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=max(520, 24 * len(plot_frame) + 180),
        xaxis=dict(title="Average precision decrease"),
        yaxis=dict(title="Feature"),
        margin=dict(l=180, r=50, t=80, b=60),
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def make_model_comparison_plot(
    *,
    data: pd.DataFrame,
    per_model_probabilities: dict[str, pd.Series],
    per_model_signals: dict[str, pd.DataFrame],
    per_model_intervals: dict[str, pd.DataFrame],
    per_model_thresholds: dict[str, float],
    title: str,
    output_path: Path,
) -> None:
    if go is None or not per_model_probabilities:
        return

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    colors = ["#38BDF8", "#FACC15", "#A78BFA", "#34D399", "#FB7185", "#F97316"]

    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    for model_idx, (model_name, probabilities) in enumerate(per_model_probabilities.items()):
        color = colors[model_idx % len(colors)]
        threshold = per_model_thresholds.get(model_name, np.nan)
        add_interval_boxes(
            fig,
            per_model_intervals.get(model_name, empty_interval_frame()),
            color=color,
            label=model_name,
            opacity=0.08,
            show_text=False,
        )
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=probabilities.reindex(data.index).to_numpy(dtype=float),
                name=f"{model_name} probability",
                mode="lines",
                yaxis="y2",
                line=dict(color=color, width=2),
            )
        )
        if threshold is not None and np.isfinite(float(threshold)):
            fig.add_trace(
                go.Scatter(
                    x=[plot_x[0], plot_x[-1]],
                    y=[float(threshold), float(threshold)],
                    name=f"{model_name} threshold",
                    mode="lines",
                    yaxis="y2",
                    line=dict(color=color, width=1, dash="dot"),
                    hoverinfo="skip",
                    showlegend=False,
                )
            )

        signals = per_model_signals.get(model_name, empty_signal_frame())
        if signals is None or signals.empty:
            continue

        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers",
                name=f"{model_name} signals",
                marker=dict(
                    color=color,
                    size=10,
                    symbol="circle-open",
                    line=dict(width=2),
                ),
                hovertext=[
                    (
                        f"Model: {model_name}<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Probability: {row['ReversalProbability']:.3f}<br>"
                        f"Threshold: {row['SignalThreshold']:.3f}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=860,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Probability",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=65, t=90, b=60),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def run(config: ReversalMLConfig) -> Path:
    analysis = run_analysis(config, save=True)
    print_run_summary(analysis)
    return analysis["run_dir"]


def run_analysis(config: ReversalMLConfig, *, save: bool = True) -> dict[str, Any]:
    data = load_ohlcv(config)
    features = make_feature_frame(data, config)
    labels = make_reversal_labels(data, features, config)
    result = train_reversal_models(data, features, labels, config)
    run_dir = save_outputs(data, result, config) if save else None
    return {
        "config": config,
        "data": data,
        "features": features,
        "labels": labels,
        "result": result,
        "run_dir": run_dir,
        "csv_paths": collect_csv_paths(run_dir, result) if run_dir is not None else {},
        "plot_paths": collect_plot_paths(run_dir, result) if run_dir is not None else {},
    }


def run_notebook(
    *,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    csv_path: str | None = None,
    output_dir: str = "outputs/ml_reversal_runs",
    show_plots: bool = True,
    print_outputs: bool = True,
    save: bool = True,
    **overrides: Any,
) -> dict[str, Any]:
    config_values = {
        "symbol": symbol,
        "period": period,
        "interval": interval,
        "csv_path": csv_path,
        "output_dir": output_dir,
    }
    config_values.update(overrides)
    config = ReversalMLConfig(**config_values)
    analysis = run_analysis(config, save=save)
    if print_outputs:
        print_run_summary(analysis)
    if show_plots and analysis["plot_paths"]:
        display_notebook_plot_links(analysis["plot_paths"])
    return analysis


def collect_csv_paths(run_dir: Path | None, result: dict[str, Any]) -> dict[str, Path]:
    if run_dir is None:
        return {}
    paths: dict[str, Path] = {
        "signals": run_dir / "signals.csv",
        "pattern_intervals": run_dir / "pattern_intervals.csv",
        "model_report": run_dir / "model_report.csv",
        "feature_importance": run_dir / "feature_importance.csv",
        "hmm_state_summary": run_dir / "hmm_state_summary.csv",
        "pca_loadings": run_dir / "pca_loadings.csv",
        "probabilities": run_dir / "probabilities.csv",
        "all_model_probabilities": run_dir / "all_model_probabilities.csv",
        "all_model_pattern_intervals": run_dir / "all_model_pattern_intervals.csv",
    }
    models_dir = run_dir / "models"
    for model_name in result.get("per_model_probabilities", {}):
        safe_model_name = safe_filename(str(model_name))
        paths[f"{model_name}_signals"] = models_dir / f"{safe_model_name}_signals.csv"
        paths[f"{model_name}_pattern_intervals"] = (
            models_dir / f"{safe_model_name}_pattern_intervals.csv"
        )
        paths[f"{model_name}_feature_importance"] = (
            models_dir / f"{safe_model_name}_feature_importance.csv"
        )
    return paths


def collect_plot_paths(run_dir: Path | None, result: dict[str, Any]) -> dict[str, Path]:
    if run_dir is None:
        return {}
    paths: dict[str, Path] = {
        "best_model": run_dir / "reversal_ml_plot.html",
        "model_comparison": run_dir / "all_models_comparison_plot.html",
    }
    models_dir = run_dir / "models"
    for model_name in result.get("per_model_probabilities", {}):
        safe_model_name = safe_filename(str(model_name))
        paths[f"{model_name}_plot"] = models_dir / f"{safe_model_name}_plot.html"
        paths[f"{model_name}_feature_importance"] = (
            models_dir / f"{safe_model_name}_feature_importance.html"
        )
    return paths


def print_run_summary(analysis: dict[str, Any]) -> None:
    config = analysis["config"]
    result = analysis["result"]
    run_dir = analysis.get("run_dir")

    print("ML reversal run")
    print(f"Run dir: {run_dir if run_dir is not None else 'not saved'}")
    print(
        f"Symbol: {config.symbol} | Period: {config.period} | Interval: {config.interval} | "
        f"HMM: {'on' if config.use_hmm else 'off'}"
    )
    print(f"Best model: {result['best_model_name']} | Threshold: {result['threshold']:.4f}")
    print(f"Pattern intervals: {len(result['intervals'])} | Point signals: {len(result['signals'])}")

    if not result["model_report"].empty:
        print("\nModel report:")
        print(result["model_report"].to_string(index=False))

    if not result["intervals"].empty:
        preview_columns = [
            column
            for column in [
                "StartDate",
                "EndDate",
                "PeakDate",
                "Bars",
                "ProbabilityMax",
                "ProbabilityMean",
                "IntervalScore",
                "BoxLow",
                "BoxHigh",
                "ShapeHint",
            ]
            if column in result["intervals"].columns
        ]
        print("\nTop pattern intervals:")
        print(
            result["intervals"]
            .sort_values("IntervalScore", ascending=False)
            .head(10)[preview_columns]
            .to_string(index=False)
        )

    csv_paths = analysis.get("csv_paths", {})
    if csv_paths:
        print("\nCSV files:")
        for name, path in csv_paths.items():
            status = "created" if path.exists() else "expected"
            print(f"  {name}: {path} [{status}]")

    plot_paths = analysis.get("plot_paths", {})
    if plot_paths:
        print("\nPlot files:")
        if go is None:
            print("  plotly is not installed, so HTML plot files were skipped.")
        for name, path in plot_paths.items():
            status = "created" if path.exists() else "not created"
            print(f"  {name}: {path} [{status}]")


def display_notebook_plot_links(plot_paths: dict[str, Path]) -> None:
    try:
        from IPython.display import HTML, display
    except ModuleNotFoundError:
        warnings.warn("IPython unavailable; returning plot paths without notebook display.")
        return
    links = "".join(
        f'<li><a href="{path}" target="_blank">{name}</a></li>'
        for name, path in plot_paths.items()
        if path.exists()
    )
    if links:
        display(HTML(f"<b>ML reversal plots</b><ul>{links}</ul>"))


if __name__ == "__main__":
    run_notebook()
```

```python
"""
Notebook usage:

from outputs.ml_reversal_pattern_detector import run_notebook

analysis = run_notebook(
    symbol="THYAO.IS",
    period="60d",
    interval="5m",
    show_plots=True,
)

analysis["result"]["intervals"]
analysis["csv_paths"]
analysis["plot_paths"]
"""

from __future__ import annotations

import json
import math
import warnings
from dataclasses import asdict, dataclass
from datetime import datetime
from pathlib import Path
from typing import Any

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    from sklearn.base import clone
    from sklearn.decomposition import PCA
    from sklearn.ensemble import HistGradientBoostingClassifier, RandomForestClassifier
    from sklearn.impute import SimpleImputer
    from sklearn.inspection import permutation_importance
    from sklearn.linear_model import LogisticRegression
    from sklearn.metrics import (
        accuracy_score,
        average_precision_score,
        balanced_accuracy_score,
        brier_score_loss,
        precision_recall_curve,
        precision_score,
        recall_score,
        roc_auc_score,
    )
    from sklearn.model_selection import TimeSeriesSplit
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
    from sklearn.utils.class_weight import compute_sample_weight
except ModuleNotFoundError:
    clone = None
    PCA = None
    HistGradientBoostingClassifier = None
    RandomForestClassifier = None
    SimpleImputer = None
    permutation_importance = None
    LogisticRegression = None
    TimeSeriesSplit = None
    Pipeline = None
    StandardScaler = None
    compute_sample_weight = None
    accuracy_score = None
    average_precision_score = None
    balanced_accuracy_score = None
    brier_score_loss = None
    precision_recall_curve = None
    precision_score = None
    recall_score = None
    roc_auc_score = None

try:
    from hmmlearn.hmm import GaussianHMM
except ModuleNotFoundError:
    GaussianHMM = None

try:
    from statsmodels.tsa.arima.model import ARIMA
except ModuleNotFoundError:
    ARIMA = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


@dataclass
class ReversalMLConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    csv_path: str | None = None
    output_dir: str = "outputs/ml_reversal_runs"
    random_state: int = 42

    label_lookback: int = 80
    label_horizon: int = 40
    min_prior_drawdown_pct: float = 2.0
    max_distance_to_low_atr: float = 2.5
    target_atr: float = 2.0
    stop_atr: float = 1.2
    switch_left_bars: int = 12
    switch_right_bars: int = 5
    switch_confirmation_bars: int = 8
    switch_label_window_bars: int = 1
    switch_min_rebound_atr: float = 0.55
    switch_max_rebound_atr: float = 2.4
    switch_box_left_bars: int = 5
    switch_box_right_bars: int = 10

    test_size_pct: float = 0.25
    n_splits: int = 5
    probability_threshold: float | None = None
    target_precision: float = 0.55
    cooldown_bars: int = 30
    interval_probability_floor: float | None = None
    interval_merge_gap_bars: int = 6
    min_interval_bars: int = 5
    max_interval_bars: int = 90
    interval_expand_atr: float = 0.35
    confidence_window: int = 20
    confidence_z: float = 1.0

    model_names: tuple[str, ...] = ("boosting", "logistic", "random_forest")
    use_hmm: bool = True
    hmm_states: int = 4
    use_arima: bool = False
    arima_order: tuple[int, int, int] = (1, 0, 1)
    arima_window: int = 160
    use_pca: bool = False
    pca_variance: float = 0.95
    permutation_repeats: int = 10


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")
    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if "volume" not in out.columns:
        out["volume"] = 0.0
    out["volume"] = out["volume"].fillna(0.0).clip(lower=0.0)

    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")
    return out


INTRADAY_INTERVALS = {"1m", "2m", "5m", "15m", "30m", "60m", "90m", "1h"}


def fetch_ohlcv(symbol: str, period: str, interval: str) -> pd.DataFrame:
    if Ticker is None:
        raise ImportError(
            "yahooquery is required for online data fetch. "
            "Install with: pip install yahooquery"
        )

    errors: list[str] = []
    for candidate_period in data_period_attempts(period=period, interval=interval):
        try:
            raw = Ticker(symbol).history(period=candidate_period, interval=interval)
            frame = normalize_yahoo_history(raw, symbol=symbol)
            data = prepare_ohlcv(frame)
            if candidate_period != period:
                warnings.warn(
                    f"No usable data for period={period!r}, interval={interval!r}; "
                    f"using period={candidate_period!r} instead."
                )
            return data
        except Exception as exc:
            errors.append(f"{candidate_period}: {type(exc).__name__}: {exc}")

    raise ValueError(
        f"No data returned for {symbol} with interval={interval}. "
        "For Yahoo intraday data, use a shorter period such as 60d/30d/7d "
        "or a wider interval such as 1d. Attempts: "
        + " | ".join(errors)
    )


def data_period_attempts(*, period: str, interval: str) -> list[str]:
    attempts = [str(period)]
    if str(interval).lower() in INTRADAY_INTERVALS:
        attempts.extend(["60d", "30d", "7d", "5d", "1mo"])
    else:
        attempts.extend(["2y", "1y", "6mo", "3mo"])
    return list(dict.fromkeys(attempts))


def normalize_yahoo_history(raw: Any, *, symbol: str) -> pd.DataFrame:
    if raw is None:
        raise ValueError("empty response")
    if isinstance(raw, dict):
        message = raw.get(symbol, raw)
        raise ValueError(f"Yahoo returned an error response: {message}")
    if isinstance(raw, pd.Series):
        raise ValueError(f"Yahoo returned a Series instead of OHLCV data: {raw.to_dict()}")
    if not isinstance(raw, pd.DataFrame):
        raise ValueError(f"Unexpected Yahoo response type: {type(raw).__name__}")
    if raw.empty:
        raise ValueError("empty response")

    frame = raw.copy()
    if isinstance(frame.index, pd.MultiIndex):
        if "symbol" in frame.index.names:
            frame = frame.reset_index(level="symbol", drop=True)
        else:
            frame = frame.reset_index()
    return frame


def load_ohlcv(config: ReversalMLConfig) -> pd.DataFrame:
    if config.csv_path:
        return prepare_ohlcv(pd.read_csv(config.csv_path))
    return fetch_ohlcv(config.symbol, config.period, config.interval)


def make_feature_frame(data: pd.DataFrame, config: ReversalMLConfig) -> pd.DataFrame:
    close = data["close"].astype(float)
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    open_ = data["open"].astype(float)
    volume = data["volume"].astype(float)

    features = pd.DataFrame(index=data.index)
    log_close = np.log(close)
    log_return = log_close.diff()

    features["log_return_1"] = log_return
    for window in [2, 3, 5, 10, 20, 40, 80]:
        features[f"log_return_sum_{window}"] = log_return.rolling(window).sum()
        features[f"realized_vol_{window}"] = log_return.rolling(window).std()

    atr = average_true_range(data, period=14)
    features["atr"] = atr
    features["atr_pct"] = atr / close.replace(0, np.nan) * 100.0
    features["atr_pct_change_5"] = features["atr_pct"].pct_change(5)

    features["rsi_7"] = rsi(close, period=7)
    features["rsi_14"] = rsi(close, period=14)
    macd_line, macd_signal, macd_hist = macd(close)
    features["macd_line"] = macd_line
    features["macd_signal"] = macd_signal
    features["macd_hist"] = macd_hist
    features["stoch_k_14"] = stochastic_k(high, low, close, period=14)
    features["stoch_d_3"] = features["stoch_k_14"].rolling(3).mean()

    feature_windows = sorted({20, 50, 80, int(config.label_lookback)})
    for window in feature_windows:
        roll_high = high.rolling(window).max()
        roll_low = low.rolling(window).min()
        roll_mid = close.rolling(window).mean()
        roll_std = close.rolling(window).std()

        features[f"drawdown_from_high_{window}"] = (roll_high - close) / roll_high * 100.0
        features[f"rebound_from_low_{window}"] = (close - roll_low) / roll_low * 100.0
        features[f"dist_to_low_atr_{window}"] = (close - roll_low) / atr.replace(0, np.nan)
        features[f"dist_to_high_atr_{window}"] = (roll_high - close) / atr.replace(0, np.nan)
        features[f"bollinger_z_{window}"] = (close - roll_mid) / roll_std.replace(0, np.nan)
        features[f"range_pct_{window}"] = (roll_high - roll_low) / roll_mid.replace(0, np.nan) * 100.0
        features[f"slope_pct_{window}"] = rolling_slope_pct(close, window=window)

    for window in [5, 8, 13, 21]:
        prior_high = high.rolling(window).max().shift(1)
        recent_low = low.rolling(window).min()
        recent_high = high.rolling(window).max()
        recent_range = (recent_high - recent_low).replace(0, np.nan)
        features[f"reclaim_from_low_atr_{window}"] = (close - recent_low) / atr.replace(0, np.nan)
        features[f"break_above_prior_high_atr_{window}"] = (close - prior_high) / atr.replace(0, np.nan)
        features[f"low_range_position_{window}"] = (close - recent_low) / recent_range
        features[f"short_turn_score_{window}"] = (
            features[f"reclaim_from_low_atr_{window}"]
            + features[f"break_above_prior_high_atr_{window}"].clip(lower=-2.0, upper=2.0)
        )

    features["volatility_expansion_5_20"] = (
        features["realized_vol_5"] / features["realized_vol_20"].replace(0, np.nan)
    )
    features["volatility_compression_20_80"] = (
        features["realized_vol_20"] / features["realized_vol_80"].replace(0, np.nan)
    )
    features["range_compression_20_80"] = (
        features["range_pct_20"] / features["range_pct_80"].replace(0, np.nan)
    )

    candle_range = (high - low).replace(0, np.nan)
    features["body_pct"] = (close - open_) / open_.replace(0, np.nan) * 100.0
    features["body_to_range"] = (close - open_).abs() / candle_range
    features["upper_wick_to_range"] = (high - np.maximum(open_, close)) / candle_range
    features["lower_wick_to_range"] = (np.minimum(open_, close) - low) / candle_range
    features["close_position_in_range"] = (close - low) / candle_range
    features["gap_pct"] = (open_ - close.shift(1)) / close.shift(1).replace(0, np.nan) * 100.0
    features["green_streak"] = signed_streak(close >= open_)

    features["log_volume"] = np.log1p(volume)
    features["volume_z_20"] = rolling_zscore(features["log_volume"], window=20)
    features["volume_ratio_20"] = volume / volume.rolling(20).mean().replace(0, np.nan)
    up_volume = volume.where(close >= open_, 0.0)
    down_volume = volume.where(close < open_, 0.0)
    rolling_volume_20 = volume.rolling(20).sum().replace(0, np.nan)
    features["up_volume_share_20"] = up_volume.rolling(20).sum() / rolling_volume_20
    features["down_volume_share_20"] = down_volume.rolling(20).sum() / rolling_volume_20
    features["volume_expansion_5_20"] = (
        volume.rolling(5).mean() / volume.rolling(20).mean().replace(0, np.nan)
    )
    obv = on_balance_volume(close, volume)
    features["obv_slope_20"] = rolling_slope_pct(obv, window=20)

    features["prior_down_context"] = (
        (features[f"drawdown_from_high_{config.label_lookback}"] >= config.min_prior_drawdown_pct)
        & (features[f"dist_to_low_atr_{config.label_lookback}"] <= config.max_distance_to_low_atr)
    ).astype(float)

    if config.use_arima:
        arima_features = walk_forward_arima_features(
            log_return,
            order=config.arima_order,
            window=config.arima_window,
        )
        features = features.join(arima_features)

    features = features.replace([np.inf, -np.inf], np.nan)
    return features


def make_reversal_labels(
    data: pd.DataFrame,
    features: pd.DataFrame,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    open_ = data["open"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    atr = features["atr"].to_numpy(dtype=float)
    prior_drawdown = features[f"drawdown_from_high_{config.label_lookback}"].to_numpy(dtype=float)
    dist_to_low = features[f"dist_to_low_atr_{config.label_lookback}"].to_numpy(dtype=float)

    labels = pd.DataFrame(index=data.index)
    labels["reversal_label"] = np.nan
    labels["label_target_price"] = np.nan
    labels["label_stop_price"] = np.nan
    labels["label_hit_bar_idx"] = np.nan
    labels["label_hit_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["label_hit_type"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["prior_drawdown_pct"] = prior_drawdown
    labels["distance_to_low_atr"] = dist_to_low
    labels["switch_kind"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["switch_turn_bar_idx"] = np.nan
    labels["switch_turn_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["switch_confirm_bar_idx"] = np.nan
    labels["switch_confirm_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["bars_from_switch_turn"] = np.nan
    labels["switch_rebound_atr"] = np.nan

    n = len(data)
    switch_events = detect_bullish_switch_events(
        data=data,
        features=features,
        open_=open_,
        high=high,
        low=low,
        close=close,
        atr=atr,
        prior_drawdown=prior_drawdown,
        dist_to_low=dist_to_low,
        config=config,
    )
    switch_lookup = make_switch_label_lookup(switch_events, n=n, config=config)

    for bar_idx in range(n):
        if bar_idx + config.label_horizon >= n:
            continue
        if not np.isfinite(atr[bar_idx]) or atr[bar_idx] <= 0:
            continue

        target_price = close[bar_idx] + config.target_atr * atr[bar_idx]
        stop_price = close[bar_idx] - config.stop_atr * atr[bar_idx]
        labels.iloc[bar_idx, labels.columns.get_loc("label_target_price")] = target_price
        labels.iloc[bar_idx, labels.columns.get_loc("label_stop_price")] = stop_price

        in_reversal_context = (
            np.isfinite(prior_drawdown[bar_idx])
            and np.isfinite(dist_to_low[bar_idx])
            and prior_drawdown[bar_idx] >= config.min_prior_drawdown_pct
            and dist_to_low[bar_idx] <= config.max_distance_to_low_atr
        )
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=bar_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        switch_event = switch_lookup.get(bar_idx)
        label = int(
            switch_event is not None
            and in_reversal_context
            and hit_type == "target"
        )
        labels.iloc[bar_idx, labels.columns.get_loc("reversal_label")] = label
        if hit_idx is not None:
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_bar_idx")] = int(hit_idx)
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_date")] = data.index[int(hit_idx)]
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_type")] = hit_type
        if switch_event is not None:
            turn_idx = int(switch_event["turn_idx"])
            confirm_idx = int(switch_event["confirm_idx"])
            labels.iloc[bar_idx, labels.columns.get_loc("switch_kind")] = switch_event["kind"]
            labels.iloc[bar_idx, labels.columns.get_loc("switch_turn_bar_idx")] = turn_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_turn_date")] = data.index[turn_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("switch_confirm_bar_idx")] = confirm_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_confirm_date")] = data.index[confirm_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("bars_from_switch_turn")] = bar_idx - turn_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_rebound_atr")] = switch_event["rebound_atr"]

    return labels


def detect_bullish_switch_events(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr: np.ndarray,
    prior_drawdown: np.ndarray,
    dist_to_low: np.ndarray,
    config: ReversalMLConfig,
) -> list[dict[str, Any]]:
    prior_context = features.get("prior_down_context", pd.Series(0.0, index=features.index)).to_numpy(
        dtype=float
    )
    left_bars = max(3, int(config.switch_left_bars))
    right_bars = max(1, int(config.switch_right_bars))
    confirmation_bars = max(right_bars, int(config.switch_confirmation_bars))
    n = len(data)
    max_future_need = max(right_bars, confirmation_bars, int(config.label_horizon))
    events: list[dict[str, Any]] = []

    for turn_idx in range(left_bars, n - max_future_need):
        if not np.isfinite(atr[turn_idx]) or atr[turn_idx] <= 0:
            continue
        if not is_bullish_switch_context(
            turn_idx=turn_idx,
            close=close,
            high=high,
            prior_context=prior_context,
            prior_drawdown=prior_drawdown,
            dist_to_low=dist_to_low,
            config=config,
        ):
            continue
        if not is_local_valley(low, turn_idx=turn_idx, left_bars=left_bars, right_bars=right_bars):
            continue

        confirm_idx = find_bullish_switch_confirmation(
            turn_idx=turn_idx,
            open_=open_,
            high=high,
            low=low,
            close=close,
            atr=atr,
            config=config,
        )
        if confirm_idx is None or confirm_idx + config.label_horizon >= n:
            continue

        target_price = close[confirm_idx] + config.target_atr * atr[confirm_idx]
        stop_price = close[confirm_idx] - config.stop_atr * atr[confirm_idx]
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=confirm_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        if hit_type != "target":
            continue

        events.append(
            {
                "kind": "valley",
                "turn_idx": int(turn_idx),
                "confirm_idx": int(confirm_idx),
                "hit_idx": int(hit_idx) if hit_idx is not None else None,
                "rebound_atr": float((close[confirm_idx] - low[turn_idx]) / atr[turn_idx]),
                "turn_price": float(low[turn_idx]),
                "confirm_price": float(close[confirm_idx]),
            }
        )

    return suppress_overlapping_switch_events(
        events,
        min_gap=max(2, int(config.switch_confirmation_bars)),
    )


def is_bullish_switch_context(
    *,
    turn_idx: int,
    close: np.ndarray,
    high: np.ndarray,
    prior_context: np.ndarray,
    prior_drawdown: np.ndarray,
    dist_to_low: np.ndarray,
    config: ReversalMLConfig,
) -> bool:
    context_start = max(0, int(turn_idx) - int(config.label_lookback))
    context_seen = finite_max(prior_context[context_start : turn_idx + 1])
    if not np.isfinite(context_seen) or context_seen <= 0:
        return False
    if not np.isfinite(prior_drawdown[turn_idx]) or not np.isfinite(dist_to_low[turn_idx]):
        return False
    if prior_drawdown[turn_idx] < config.min_prior_drawdown_pct:
        return False
    if dist_to_low[turn_idx] > config.max_distance_to_low_atr:
        return False

    approach_start = max(0, int(turn_idx) - max(5, int(config.switch_left_bars)))
    if turn_idx - approach_start < 4:
        return False
    approach_high = finite_max(high[approach_start : turn_idx + 1])
    if not np.isfinite(approach_high) or approach_high <= 0:
        return False
    approach_drawdown = (approach_high - close[turn_idx]) / approach_high * 100.0
    if approach_drawdown < config.min_prior_drawdown_pct * 0.35:
        return False
    return close[turn_idx] <= np.nanmedian(close[approach_start : turn_idx + 1])


def is_local_valley(
    low: np.ndarray,
    *,
    turn_idx: int,
    left_bars: int,
    right_bars: int,
) -> bool:
    start = max(0, int(turn_idx) - int(left_bars))
    end = min(len(low) - 1, int(turn_idx) + int(right_bars))
    window = low[start : end + 1]
    if len(window) == 0 or not np.isfinite(low[turn_idx]):
        return False
    valley = finite_min(window)
    if not np.isfinite(valley):
        return False
    if low[turn_idx] > valley + max(abs(valley) * 1e-8, 1e-10):
        return False
    right_window = low[turn_idx + 1 : end + 1]
    return len(right_window) == 0 or finite_min(right_window) > low[turn_idx] - max(abs(low[turn_idx]) * 1e-8, 1e-10)


def find_bullish_switch_confirmation(
    *,
    turn_idx: int,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr: np.ndarray,
    config: ReversalMLConfig,
) -> int | None:
    atr_at_turn = float(atr[turn_idx])
    if not np.isfinite(atr_at_turn) or atr_at_turn <= 0:
        return None
    end_idx = min(len(close) - 1, int(turn_idx) + int(config.switch_confirmation_bars))
    rebound_level = low[turn_idx] + config.switch_min_rebound_atr * atr_at_turn
    micro_start = max(0, int(turn_idx) - max(3, int(config.switch_left_bars) // 2))
    micro_break_level = finite_max(high[micro_start:turn_idx])

    for idx in range(int(turn_idx) + 1, end_idx + 1):
        rebound_atr = (close[idx] - low[turn_idx]) / atr_at_turn
        if rebound_atr < config.switch_min_rebound_atr:
            continue
        if rebound_atr > config.switch_max_rebound_atr:
            continue
        closes_green = close[idx] > open_[idx]
        breaks_previous_high = idx > 0 and close[idx] > high[idx - 1]
        breaks_micro_high = np.isfinite(micro_break_level) and close[idx] > micro_break_level
        reclaims_valley = close[idx] >= rebound_level
        if reclaims_valley and (closes_green or breaks_previous_high or breaks_micro_high):
            return int(idx)
    return None


def make_switch_label_lookup(
    events: list[dict[str, Any]],
    *,
    n: int,
    config: ReversalMLConfig,
) -> dict[int, dict[str, Any]]:
    lookup: dict[int, dict[str, Any]] = {}
    radius = max(0, int(config.switch_label_window_bars))
    for event in events:
        anchor_idx = int(event["confirm_idx"])
        turn_idx = int(event["turn_idx"])
        for label_idx in range(anchor_idx - radius, anchor_idx + radius + 1):
            if label_idx < turn_idx or label_idx < 0 or label_idx >= n:
                continue
            current = lookup.get(label_idx)
            if current is None:
                lookup[label_idx] = event
                continue
            current_distance = abs(label_idx - int(current["confirm_idx"]))
            new_distance = abs(label_idx - anchor_idx)
            if new_distance < current_distance:
                lookup[label_idx] = event
    return lookup


def suppress_overlapping_switch_events(
    events: list[dict[str, Any]],
    *,
    min_gap: int,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []
    for event in sorted(events, key=lambda item: int(item["confirm_idx"])):
        if not selected:
            selected.append(event)
            continue
        previous = selected[-1]
        if int(event["confirm_idx"]) - int(previous["confirm_idx"]) > min_gap:
            selected.append(event)
            continue
        if float(event["turn_price"]) < float(previous["turn_price"]):
            selected[-1] = event
    return selected


def first_long_barrier_hit(
    *,
    high: np.ndarray,
    low: np.ndarray,
    start_idx: int,
    horizon: int,
    target_price: float,
    stop_price: float,
) -> tuple[str, int | None]:
    max_idx = min(len(high) - 1, int(start_idx) + int(horizon))
    for idx in range(int(start_idx) + 1, max_idx + 1):
        stop_hit = low[idx] <= stop_price
        target_hit = high[idx] >= target_price
        if stop_hit and target_hit:
            return "stop", idx
        if stop_hit:
            return "stop", idx
        if target_hit:
            return "target", idx
    return "timeout", None


def add_hmm_features(
    features: pd.DataFrame,
    train_index: pd.Index,
    config: ReversalMLConfig,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    state_summary_columns = ["state", "train_count", "mean_return", "mean_vol", "mean_drawdown"]
    if not config.use_hmm:
        return features, pd.DataFrame(columns=state_summary_columns)
    if GaussianHMM is None or SimpleImputer is None or StandardScaler is None:
        warnings.warn("hmmlearn/sklearn unavailable; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    hmm_columns = [
        column
        for column in [
            "log_return_1",
            "realized_vol_10",
            "atr_pct",
            "volume_z_20",
            f"drawdown_from_high_{config.label_lookback}",
            f"dist_to_low_atr_{config.label_lookback}",
        ]
        if column in features.columns
    ]
    if len(hmm_columns) < 3:
        return features, pd.DataFrame(columns=state_summary_columns)

    train = features.loc[train_index, hmm_columns]
    if train.dropna(how="all").shape[0] < config.hmm_states * 20:
        warnings.warn("Not enough training rows for HMM; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    imputer = SimpleImputer(strategy="median")
    scaler = StandardScaler()
    train_values = scaler.fit_transform(imputer.fit_transform(train))
    all_values = scaler.transform(imputer.transform(features[hmm_columns]))

    hmm = GaussianHMM(
        n_components=config.hmm_states,
        covariance_type="diag",
        n_iter=200,
        tol=1e-4,
        random_state=config.random_state,
    )
    try:
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            hmm.fit(train_values)
            probabilities = hmm.predict_proba(all_values)
    except Exception as exc:
        warnings.warn(f"HMM regime features skipped after fit failure: {type(exc).__name__}: {exc}")
        return features, pd.DataFrame(columns=state_summary_columns)
    states = probabilities.argmax(axis=1)

    out = features.copy()
    out["hmm_state"] = states.astype(float)
    for state in range(config.hmm_states):
        out[f"hmm_state_prob_{state}"] = probabilities[:, state]

    state_rows: list[dict[str, Any]] = []
    train_augmented = out.loc[train_index].copy()
    for state in range(config.hmm_states):
        subset = train_augmented[train_augmented["hmm_state"] == state]
        state_rows.append(
            {
                "state": state,
                "train_count": int(len(subset)),
                "mean_return": float(subset["log_return_1"].mean()),
                "mean_vol": float(subset["realized_vol_10"].mean()),
                "mean_drawdown": float(
                    subset[f"drawdown_from_high_{config.label_lookback}"].mean()
                ),
            }
        )
    return out, pd.DataFrame(state_rows, columns=state_summary_columns)


def train_reversal_models(
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    config: ReversalMLConfig,
) -> dict[str, Any]:
    require_sklearn()

    labeled = features.join(labels[["reversal_label"]], how="inner")
    labeled = labeled.dropna(subset=["reversal_label"])
    if len(labeled) < 200:
        raise ValueError("Not enough labeled rows. Increase period or use a higher timeframe.")

    y = labeled["reversal_label"].astype(int)
    feature_columns = [
        column
        for column in labeled.columns
        if column != "reversal_label" and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]

    split_idx = int(len(labeled) * (1.0 - config.test_size_pct))
    split_idx = min(max(split_idx, 50), len(labeled) - 20)
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]

    features_with_hmm, hmm_summary = add_hmm_features(features, train_index, config)
    labeled = features_with_hmm.join(labels[["reversal_label"]], how="inner")
    labeled = labeled.dropna(subset=["reversal_label"])
    y = labeled["reversal_label"].astype(int)
    if y.nunique() < 2:
        positives = int(y.sum())
        raise ValueError(
            "Switch labeling produced only one class "
            f"(positive labels={positives}). Increase period/history or loosen "
            "switch_min_rebound_atr, switch_confirmation_bars, min_prior_drawdown_pct, "
            "or max_distance_to_low_atr."
        )
    feature_columns = [
        column
        for column in labeled.columns
        if column != "reversal_label" and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]
    X_train = X.loc[train_index]
    y_train = y.loc[train_index]
    X_test = X.loc[test_index]
    y_test = y.loc[test_index]
    if y_train.nunique() < 2:
        raise ValueError(
            "Training split has only one class after switch labeling. "
            "Use more history, lower test_size_pct, or loosen the switch label settings."
        )

    X_train_clip, clip_bounds = fit_winsorize(X_train)
    X_test_clip = apply_winsorize(X_test, clip_bounds)
    X_all_clip = apply_winsorize(features_with_hmm[feature_columns], clip_bounds)

    models = build_models(config)
    results: list[dict[str, Any]] = []
    fitted_models: dict[str, Pipeline] = {}

    for name, model in models.items():
        fitted = fit_pipeline_with_weights(model, X_train_clip, y_train)
        fitted_models[name] = fitted
        test_probability = positive_probability(fitted, X_test_clip)
        metrics = evaluate_probabilities(y_test, test_probability)
        cv_metrics = time_series_cv_score(
            model,
            X_train_clip,
            y_train,
            n_splits=config.n_splits,
        )
        result = {"model": name, **metrics, **cv_metrics}
        results.append(result)

    model_report = pd.DataFrame(results)
    best_name = choose_best_model(model_report)
    per_model_probabilities: dict[str, pd.Series] = {}
    per_model_thresholds: dict[str, float] = {}
    per_model_feature_importance: dict[str, pd.DataFrame] = {}
    per_model_signals: dict[str, pd.DataFrame] = {}
    per_model_intervals: dict[str, pd.DataFrame] = {}

    for name, fitted in fitted_models.items():
        all_probability = positive_probability(fitted, X_all_clip)
        test_probability = positive_probability(fitted, X_test_clip)
        model_threshold = config.probability_threshold
        if model_threshold is None:
            model_threshold = choose_probability_threshold(
                y_test,
                test_probability,
                target_precision=config.target_precision,
            )

        probability_series = pd.Series(
            all_probability,
            index=features_with_hmm.index,
            name=f"{name}_reversal_probability",
        )
        importance = permutation_xai(
            fitted,
            X_test_clip,
            y_test,
            feature_columns,
            repeats=config.permutation_repeats,
            random_state=config.random_state,
        )
        signals = build_signal_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=probability_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not signals.empty:
            signals.insert(0, "Model", name)
        intervals = build_interval_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=probability_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not intervals.empty:
            intervals.insert(0, "Model", name)

        per_model_probabilities[name] = probability_series
        per_model_thresholds[name] = float(model_threshold)
        per_model_feature_importance[name] = importance
        per_model_signals[name] = signals
        per_model_intervals[name] = intervals

    best_model = fitted_models[best_name]
    threshold = per_model_thresholds[best_name]
    probabilities = per_model_probabilities[best_name].rename("reversal_probability")
    signals = per_model_signals[best_name].copy()
    if "Model" in signals.columns:
        signals = signals.drop(columns=["Model"])
    intervals = per_model_intervals[best_name].copy()
    if "Model" in intervals.columns:
        intervals = intervals.drop(columns=["Model"])
    feature_importance = per_model_feature_importance[best_name]
    pca_loadings = extract_pca_loadings(best_model, feature_columns)

    return {
        "features": features_with_hmm,
        "labels": labels,
        "feature_columns": feature_columns,
        "model_report": model_report,
        "best_model_name": best_name,
        "best_model": best_model,
        "probabilities": probabilities,
        "threshold": threshold,
        "signals": signals,
        "intervals": intervals,
        "feature_importance": feature_importance,
        "per_model_probabilities": per_model_probabilities,
        "per_model_thresholds": per_model_thresholds,
        "per_model_signals": per_model_signals,
        "per_model_intervals": per_model_intervals,
        "per_model_feature_importance": per_model_feature_importance,
        "pca_loadings": pca_loadings,
        "hmm_state_summary": hmm_summary,
    }


def require_sklearn() -> None:
    if Pipeline is None:
        raise ImportError(
            "scikit-learn is required for ML training. Install with: pip install scikit-learn"
        )


def build_models(config: ReversalMLConfig) -> dict[str, Pipeline]:
    requested = {name.strip().lower() for name in config.model_names}
    models: dict[str, Pipeline] = {}

    if "boosting" in requested:
        models["boosting"] = make_pipeline(
            HistGradientBoostingClassifier(
                learning_rate=0.05,
                max_iter=350,
                max_leaf_nodes=15,
                l2_regularization=0.05,
                early_stopping=True,
                random_state=config.random_state,
            ),
            config,
            scale=False,
        )
    if "logistic" in requested:
        models["logistic"] = make_pipeline(
            LogisticRegression(
                max_iter=2500,
                class_weight="balanced",
                solver="lbfgs",
                random_state=config.random_state,
            ),
            config,
            scale=True,
        )
    if "random_forest" in requested:
        models["random_forest"] = make_pipeline(
            RandomForestClassifier(
                n_estimators=400,
                max_depth=7,
                min_samples_leaf=20,
                class_weight="balanced_subsample",
                random_state=config.random_state,
                n_jobs=-1,
            ),
            config,
            scale=False,
        )
    if not models:
        raise ValueError("No valid model names requested.")
    return models


def make_pipeline(estimator: Any, config: ReversalMLConfig, *, scale: bool) -> Pipeline:
    steps: list[tuple[str, Any]] = [("imputer", SimpleImputer(strategy="median"))]
    if scale or config.use_pca:
        steps.append(("scaler", StandardScaler()))
    if config.use_pca:
        steps.append(("pca", PCA(n_components=config.pca_variance, svd_solver="full")))
    steps.append(("model", estimator))
    return Pipeline(steps)


def fit_pipeline_with_weights(model: Pipeline, X: pd.DataFrame, y: pd.Series) -> Pipeline:
    sample_weight = compute_sample_weight(class_weight="balanced", y=y)
    try:
        return model.fit(X, y, model__sample_weight=sample_weight)
    except TypeError:
        return model.fit(X, y)


def positive_probability(model: Pipeline, X: pd.DataFrame) -> np.ndarray:
    if hasattr(model, "predict_proba"):
        probability = model.predict_proba(X)
        if probability.shape[1] == 1:
            return np.zeros(len(X), dtype=float)
        return probability[:, 1]
    decision = model.decision_function(X)
    return 1.0 / (1.0 + np.exp(-decision))


def evaluate_probabilities(y_true: pd.Series, probability: np.ndarray) -> dict[str, float]:
    out: dict[str, float] = {
        "test_average_precision": np.nan,
        "test_roc_auc": np.nan,
        "test_brier": np.nan,
        "test_balanced_accuracy_050": np.nan,
        "test_precision_050": np.nan,
        "test_recall_050": np.nan,
        "test_positive_rate": float(np.mean(y_true)),
    }
    if len(np.unique(y_true)) < 2:
        return out

    prediction = (probability >= 0.5).astype(int)
    out["test_average_precision"] = float(average_precision_score(y_true, probability))
    out["test_roc_auc"] = float(roc_auc_score(y_true, probability))
    out["test_brier"] = float(brier_score_loss(y_true, probability))
    out["test_balanced_accuracy_050"] = float(balanced_accuracy_score(y_true, prediction))
    out["test_precision_050"] = float(precision_score(y_true, prediction, zero_division=0))
    out["test_recall_050"] = float(recall_score(y_true, prediction, zero_division=0))
    out["test_accuracy_050"] = float(accuracy_score(y_true, prediction))
    return out


def time_series_cv_score(
    model: Pipeline,
    X: pd.DataFrame,
    y: pd.Series,
    *,
    n_splits: int,
) -> dict[str, float]:
    out = {"cv_average_precision": np.nan, "cv_roc_auc": np.nan}
    if len(X) < 200 or y.nunique() < 2:
        return out

    split_count = min(n_splits, max(2, len(X) // 200))
    splitter = TimeSeriesSplit(n_splits=split_count)
    ap_scores: list[float] = []
    auc_scores: list[float] = []
    for train_idx, test_idx in splitter.split(X):
        y_train = y.iloc[train_idx]
        y_valid = y.iloc[test_idx]
        if y_train.nunique() < 2 or y_valid.nunique() < 2:
            continue
        fitted = fit_pipeline_with_weights(clone(model), X.iloc[train_idx], y_train)
        prob = positive_probability(fitted, X.iloc[test_idx])
        ap_scores.append(float(average_precision_score(y_valid, prob)))
        auc_scores.append(float(roc_auc_score(y_valid, prob)))

    if ap_scores:
        out["cv_average_precision"] = float(np.mean(ap_scores))
    if auc_scores:
        out["cv_roc_auc"] = float(np.mean(auc_scores))
    return out


def choose_best_model(model_report: pd.DataFrame) -> str:
    ranking = model_report.copy()
    ranking["_rank_score"] = ranking["test_average_precision"].fillna(-1.0)
    ranking = ranking.sort_values(["_rank_score", "test_roc_auc"], ascending=False)
    return str(ranking.iloc[0]["model"])


def choose_probability_threshold(
    y_true: pd.Series,
    probability: np.ndarray,
    *,
    target_precision: float,
) -> float:
    if len(np.unique(y_true)) < 2:
        return 0.65

    precision, recall, thresholds = precision_recall_curve(y_true, probability)
    candidates: list[tuple[float, float, float]] = []
    for threshold, prec, rec in zip(thresholds, precision[:-1], recall[:-1]):
        if prec >= target_precision and rec > 0:
            candidates.append((float(threshold), float(prec), float(rec)))
    if not candidates:
        return float(np.quantile(probability, 0.90))
    candidates.sort(key=lambda item: (item[2], item[1]), reverse=True)
    return float(candidates[0][0])


def permutation_xai(
    model: Pipeline,
    X_test: pd.DataFrame,
    y_test: pd.Series,
    feature_columns: list[str],
    *,
    repeats: int,
    random_state: int,
) -> pd.DataFrame:
    columns = ["feature", "importance_mean", "importance_std", "rank"]
    if y_test.nunique() < 2 or permutation_importance is None:
        return pd.DataFrame(columns=columns)

    result = permutation_importance(
        model,
        X_test,
        y_test,
        n_repeats=repeats,
        random_state=random_state,
        scoring="average_precision",
        n_jobs=-1,
    )
    importance = pd.DataFrame(
        {
            "feature": feature_columns,
            "importance_mean": result.importances_mean,
            "importance_std": result.importances_std,
        }
    )
    importance = importance.sort_values("importance_mean", ascending=False).reset_index(drop=True)
    importance["rank"] = np.arange(1, len(importance) + 1)
    return importance[columns]


def extract_pca_loadings(model: Pipeline, feature_columns: list[str]) -> pd.DataFrame:
    if "pca" not in model.named_steps:
        return pd.DataFrame(columns=["component", "feature", "loading", "abs_loading", "rank"])

    pca = model.named_steps["pca"]
    rows: list[dict[str, Any]] = []
    for component_idx, component in enumerate(pca.components_):
        for feature, loading in zip(feature_columns, component):
            rows.append(
                {
                    "component": f"PC{component_idx + 1}",
                    "feature": feature,
                    "loading": float(loading),
                    "abs_loading": float(abs(loading)),
                }
            )
    out = pd.DataFrame(rows)
    out["rank"] = out.groupby("component")["abs_loading"].rank(ascending=False, method="first")
    return out.sort_values(["component", "rank"]).reset_index(drop=True)


def build_signal_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    candidate_probability = probabilities.dropna()
    candidate_probability = candidate_probability[candidate_probability >= threshold]
    if candidate_probability.empty:
        return empty_signal_frame()

    selected_indexes = non_max_suppression_1d(
        candidate_probability,
        cooldown_bars=config.cooldown_bars,
        all_index=features.index,
    )

    train_stats = make_train_feature_stats(train_features)
    rows: list[dict[str, Any]] = []
    atr = features["atr"]
    for timestamp in selected_indexes:
        bar_idx = int(data.index.get_loc(timestamp))
        entry_price = float(data.loc[timestamp, "close"])
        target_price = entry_price + config.target_atr * float(atr.loc[timestamp])
        stop_price = entry_price - config.stop_atr * float(atr.loc[timestamp])
        label_row = labels.loc[timestamp] if timestamp in labels.index else pd.Series(dtype=object)
        hit_idx = label_row.get("label_hit_bar_idx", np.nan)
        hit_date = label_row.get("label_hit_date", pd.NaT)
        hit_type = label_row.get("label_hit_type", None)
        realized_label = label_row.get("reversal_label", np.nan)
        switch_kind = label_row.get("switch_kind", None)
        switch_turn_idx = label_row.get("switch_turn_bar_idx", np.nan)
        switch_turn_date = label_row.get("switch_turn_date", pd.NaT)
        switch_confirm_idx = label_row.get("switch_confirm_bar_idx", np.nan)
        switch_confirm_date = label_row.get("switch_confirm_date", pd.NaT)
        bars_from_switch_turn = label_row.get("bars_from_switch_turn", np.nan)
        top_drivers = local_driver_summary(
            features.loc[timestamp, feature_columns],
            feature_importance,
            train_stats,
            top_n=6,
        )
        rows.append(
            {
                "EntryBarIdx": bar_idx,
                "EntryDate": timestamp,
                "EntryPrice": entry_price,
                "ReversalProbability": float(probabilities.loc[timestamp]),
                "SignalThreshold": float(threshold),
                "TargetPrice": target_price,
                "StopPrice": stop_price,
                "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                "OutcomeKnown": is_finite_number(realized_label),
                "OutcomeLabel": realized_label,
                "HitBarIdx": int(hit_idx) if is_finite_number(hit_idx) else np.nan,
                "HitDate": hit_date,
                "HitType": hit_type,
                "SwitchKind": switch_kind,
                "SwitchTurnBarIdx": int(switch_turn_idx) if is_finite_number(switch_turn_idx) else np.nan,
                "SwitchTurnDate": switch_turn_date,
                "SwitchConfirmBarIdx": int(switch_confirm_idx) if is_finite_number(switch_confirm_idx) else np.nan,
                "SwitchConfirmDate": switch_confirm_date,
                "BarsFromSwitchTurn": bars_from_switch_turn,
                "ShapeHint": classify_shape_hint(data, bar_idx),
                "TopDrivers": top_drivers,
            }
        )
    return pd.DataFrame(rows).sort_values("EntryBarIdx").reset_index(drop=True)


def build_interval_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    aligned_probability = probabilities.reindex(features.index).astype(float)
    confidence = probability_confidence_frame(
        aligned_probability,
        window=config.confidence_window,
        z=config.confidence_z,
    )
    floor = config.interval_probability_floor
    if floor is None:
        floor = max(0.05, min(0.95, float(threshold) * 0.80))

    prior_context = features.get("prior_down_context", pd.Series(0.0, index=features.index))
    context_seen = prior_context.fillna(0.0).rolling(
        config.label_lookback,
        min_periods=1,
    ).max()
    candidate_mask = (
        (aligned_probability >= floor)
        | (
            (confidence["upper"] >= float(threshold))
            & (confidence["mean"] >= floor * 0.85)
        )
    )
    candidate_mask = (candidate_mask.fillna(False) & (context_seen > 0)).to_numpy(dtype=bool)
    segments = merge_boolean_segments(
        boolean_segments(candidate_mask),
        max_gap=max(0, int(config.interval_merge_gap_bars)),
    )
    if not segments:
        return empty_interval_frame()

    rows: list[dict[str, Any]] = []
    train_stats = make_train_feature_stats(train_features)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    atr = features["atr"].reindex(data.index).to_numpy(dtype=float)
    min_bars = max(1, int(config.min_interval_bars))
    max_bars = max(min_bars, int(config.max_interval_bars))

    for raw_start, raw_end in segments:
        for seg_start, seg_end in split_segment(raw_start, raw_end, max_bars=max_bars, min_bars=min_bars):
            segment_probability = aligned_probability.iloc[seg_start : seg_end + 1].dropna()
            if segment_probability.empty:
                continue
            segment_confidence = confidence.iloc[seg_start : seg_end + 1]
            if (
                float(segment_probability.max()) < float(threshold)
                and float(segment_confidence["upper"].max()) < float(threshold)
            ):
                continue

            peak_timestamp = segment_probability.idxmax()
            peak_idx = int(features.index.get_loc(peak_timestamp))
            search_start = max(0, peak_idx - max(config.switch_left_bars, config.switch_box_left_bars))
            search_end = min(len(data) - 1, peak_idx + max(config.switch_right_bars, 1))
            bottom_idx = search_start + int(np.nanargmin(low[search_start : search_end + 1]))
            start_idx = max(0, bottom_idx - int(config.switch_box_left_bars))
            end_idx = min(
                len(data) - 1,
                max(peak_idx, bottom_idx) + int(config.switch_box_right_bars),
            )

            if end_idx - start_idx + 1 < min_bars:
                start_idx = max(0, peak_idx - min_bars // 2)
                end_idx = min(len(data) - 1, start_idx + min_bars - 1)
            if end_idx - start_idx + 1 > max_bars:
                center_idx = int(round((peak_idx + bottom_idx) / 2))
                start_idx = max(0, center_idx - max_bars // 2)
                end_idx = min(len(data) - 1, start_idx + max_bars - 1)

            context_start = max(0, start_idx - config.label_lookback // 2)
            context_slice = prior_context.iloc[context_start : peak_idx + 1].fillna(0.0)
            if context_slice.max() <= 0:
                continue

            interval_probability = aligned_probability.iloc[start_idx : end_idx + 1].dropna()
            if interval_probability.empty:
                continue
            probability_mean = float(interval_probability.mean())
            probability_max = float(interval_probability.max())
            interval_confidence_low, interval_confidence_high = probability_interval_bounds(
                interval_probability,
                z=config.confidence_z,
            )
            if probability_max < float(threshold) and interval_confidence_high < float(threshold):
                continue

            support_low = finite_min(low[start_idx : end_idx + 1])
            resistance_high = finite_max(high[start_idx : end_idx + 1])
            if not np.isfinite(support_low) or not np.isfinite(resistance_high):
                continue

            entry_price = float(close[end_idx])
            interval_atr = finite_mean(atr[start_idx : end_idx + 1])
            if not np.isfinite(interval_atr) or interval_atr <= 0:
                interval_atr = max((resistance_high - support_low) * 0.10, abs(entry_price) * 0.001, 1e-12)
            price_pad = max(0.0, float(config.interval_expand_atr)) * interval_atr
            target_price = entry_price + config.target_atr * interval_atr
            stop_price = entry_price - config.stop_atr * interval_atr
            peak_label = labels.loc[peak_timestamp] if peak_timestamp in labels.index else pd.Series(dtype=object)
            hit_idx = peak_label.get("label_hit_bar_idx", np.nan)
            realized_label = peak_label.get("reversal_label", np.nan)
            top_drivers = local_driver_summary(
                features.loc[peak_timestamp, feature_columns],
                feature_importance,
                train_stats,
                top_n=6,
            )

            volume_z_max = finite_max(series_values(features, "volume_z_20", start_idx, end_idx))
            volume_ratio_max = finite_max(series_values(features, "volume_ratio_20", start_idx, end_idx))
            volatility_expansion = finite_mean(
                series_values(features, "volatility_expansion_5_20", start_idx, end_idx)
            )
            volatility_compression = finite_mean(
                series_values(features, "volatility_compression_20_80", start_idx, end_idx)
            )
            range_compression = finite_mean(
                series_values(features, "range_compression_20_80", start_idx, end_idx)
            )
            context_ratio = float(prior_context.iloc[start_idx : end_idx + 1].fillna(0.0).mean())
            interval_score = score_interval(
                probability_mean=probability_mean,
                probability_max=probability_max,
                context_ratio=context_ratio,
                volume_z_max=volume_z_max,
                volatility_expansion=volatility_expansion,
                volatility_compression=volatility_compression,
                confidence_low=interval_confidence_low,
                confidence_high=interval_confidence_high,
            )

            rows.append(
                {
                    "StartBarIdx": int(start_idx),
                    "EndBarIdx": int(end_idx),
                    "PeakBarIdx": int(peak_idx),
                    "ProbabilityPeakBarIdx": int(peak_idx),
                    "TurnBarIdx": int(bottom_idx),
                    "SwitchKind": "valley",
                    "StartDate": data.index[start_idx],
                    "EndDate": data.index[end_idx],
                    "PeakDate": peak_timestamp,
                    "ProbabilityPeakDate": peak_timestamp,
                    "TurnDate": data.index[bottom_idx],
                    "Bars": int(end_idx - start_idx + 1),
                    "SupportLow": support_low,
                    "ResistanceHigh": resistance_high,
                    "BoxLow": support_low - price_pad,
                    "BoxHigh": resistance_high + price_pad,
                    "EntryBarIdx": int(end_idx),
                    "EntryDate": data.index[end_idx],
                    "EntryPrice": entry_price,
                    "TargetPrice": target_price,
                    "StopPrice": stop_price,
                    "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                    "ProbabilityMean": probability_mean,
                    "ProbabilityMax": probability_max,
                    "ProbabilityConfidenceLow": interval_confidence_low,
                    "ProbabilityConfidenceHigh": interval_confidence_high,
                    "SignalThreshold": float(threshold),
                    "VolumeZMax": volume_z_max,
                    "VolumeRatioMax": volume_ratio_max,
                    "VolatilityExpansionMean": volatility_expansion,
                    "VolatilityCompressionMean": volatility_compression,
                    "RangeCompressionMean": range_compression,
                    "PriorDownContextRatio": context_ratio,
                    "IntervalScore": interval_score,
                    "OutcomeKnown": is_finite_number(realized_label),
                    "OutcomeLabel": realized_label,
                    "HitBarIdx": int(hit_idx) if is_finite_number(hit_idx) else np.nan,
                    "HitDate": peak_label.get("label_hit_date", pd.NaT),
                    "HitType": peak_label.get("label_hit_type", None),
                    "ShapeHint": classify_shape_hint(data, peak_idx),
                    "TopDrivers": top_drivers,
                }
            )

    if not rows:
        return empty_interval_frame()
    rows = suppress_overlapping_interval_rows(rows)
    return pd.DataFrame(rows).sort_values("StartBarIdx").reset_index(drop=True)


def empty_signal_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "ReversalProbability",
            "SignalThreshold",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "SwitchKind",
            "SwitchTurnBarIdx",
            "SwitchTurnDate",
            "SwitchConfirmBarIdx",
            "SwitchConfirmDate",
            "BarsFromSwitchTurn",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def empty_interval_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "StartBarIdx",
            "EndBarIdx",
            "PeakBarIdx",
            "ProbabilityPeakBarIdx",
            "TurnBarIdx",
            "SwitchKind",
            "StartDate",
            "EndDate",
            "PeakDate",
            "ProbabilityPeakDate",
            "TurnDate",
            "Bars",
            "SupportLow",
            "ResistanceHigh",
            "BoxLow",
            "BoxHigh",
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "ProbabilityMean",
            "ProbabilityMax",
            "ProbabilityConfidenceLow",
            "ProbabilityConfidenceHigh",
            "SignalThreshold",
            "VolumeZMax",
            "VolumeRatioMax",
            "VolatilityExpansionMean",
            "VolatilityCompressionMean",
            "RangeCompressionMean",
            "PriorDownContextRatio",
            "IntervalScore",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def non_max_suppression_1d(
    probability: pd.Series,
    *,
    cooldown_bars: int,
    all_index: pd.Index,
) -> list[pd.Timestamp]:
    ranked = probability.sort_values(ascending=False)
    selected_positions: list[int] = []
    selected_timestamps: list[pd.Timestamp] = []

    for timestamp, _ in ranked.items():
        position = int(all_index.get_loc(timestamp))
        if any(abs(position - selected) <= cooldown_bars for selected in selected_positions):
            continue
        selected_positions.append(position)
        selected_timestamps.append(timestamp)

    selected_timestamps.sort(key=lambda ts: int(all_index.get_loc(ts)))
    return selected_timestamps


def probability_confidence_frame(probability: pd.Series, *, window: int, z: float) -> pd.DataFrame:
    values = probability.astype(float)
    window = max(2, int(window))
    mean = values.rolling(window, min_periods=max(2, window // 4)).mean()
    std = values.rolling(window, min_periods=max(2, window // 4)).std().fillna(0.0)
    count = values.rolling(window, min_periods=max(2, window // 4)).count().replace(0, np.nan)
    stderr = std / np.sqrt(count)
    return pd.DataFrame(
        {
            "mean": mean,
            "lower": (mean - float(z) * stderr).clip(0.0, 1.0),
            "upper": (mean + float(z) * stderr).clip(0.0, 1.0),
        },
        index=probability.index,
    )


def probability_interval_bounds(probability: pd.Series, *, z: float) -> tuple[float, float]:
    values = probability.dropna().to_numpy(dtype=float)
    if len(values) == 0:
        return np.nan, np.nan
    mean = float(np.mean(values))
    if len(values) == 1:
        return mean, mean
    stderr = float(np.std(values, ddof=1) / math.sqrt(len(values)))
    return max(0.0, mean - float(z) * stderr), min(1.0, mean + float(z) * stderr)


def boolean_segments(mask: np.ndarray) -> list[tuple[int, int]]:
    segments: list[tuple[int, int]] = []
    start: int | None = None
    for idx, active in enumerate(mask):
        if active and start is None:
            start = idx
        elif not active and start is not None:
            segments.append((start, idx - 1))
            start = None
    if start is not None:
        segments.append((start, len(mask) - 1))
    return segments


def merge_boolean_segments(
    segments: list[tuple[int, int]],
    *,
    max_gap: int,
) -> list[tuple[int, int]]:
    if not segments:
        return []
    merged: list[tuple[int, int]] = [segments[0]]
    for start, end in segments[1:]:
        previous_start, previous_end = merged[-1]
        if start - previous_end - 1 <= max_gap:
            merged[-1] = (previous_start, end)
        else:
            merged.append((start, end))
    return merged


def split_segment(
    start: int,
    end: int,
    *,
    max_bars: int,
    min_bars: int,
) -> list[tuple[int, int]]:
    length = end - start + 1
    if length <= max_bars:
        return [(start, end)]

    chunks: list[tuple[int, int]] = []
    overlap = max(min_bars, max_bars // 4)
    chunk_start = start
    while chunk_start <= end:
        chunk_end = min(end, chunk_start + max_bars - 1)
        chunks.append((chunk_start, chunk_end))
        if chunk_end >= end:
            break
        chunk_start = max(chunk_start + 1, chunk_end - overlap + 1)
    return chunks


def series_values(features: pd.DataFrame, column: str, start_idx: int, end_idx: int) -> np.ndarray:
    if column not in features.columns:
        return np.array([], dtype=float)
    return features[column].iloc[start_idx : end_idx + 1].to_numpy(dtype=float)


def is_finite_number(value: Any) -> bool:
    try:
        return bool(np.isfinite(float(value)))
    except (TypeError, ValueError):
        return False


def finite_mean(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.mean(array))


def finite_min(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.min(array))


def finite_max(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.max(array))


def score_interval(
    *,
    probability_mean: float,
    probability_max: float,
    context_ratio: float,
    volume_z_max: float,
    volatility_expansion: float,
    volatility_compression: float,
    confidence_low: float,
    confidence_high: float,
) -> float:
    volume_score = np.tanh(max(0.0, 0.0 if not np.isfinite(volume_z_max) else volume_z_max) / 3.0)
    expansion_score = np.tanh(
        max(0.0, (0.0 if not np.isfinite(volatility_expansion) else volatility_expansion) - 1.0)
    )
    compression_score = 1.0 - min(
        1.0,
        max(0.0, 0.0 if not np.isfinite(volatility_compression) else volatility_compression),
    )
    confidence_width = max(
        0.0,
        (0.0 if not np.isfinite(confidence_high) else confidence_high)
        - (0.0 if not np.isfinite(confidence_low) else confidence_low),
    )
    stability_score = 1.0 - min(1.0, confidence_width)
    score = (
        0.38 * probability_max
        + 0.27 * probability_mean
        + 0.15 * min(1.0, max(0.0, context_ratio))
        + 0.07 * volume_score
        + 0.06 * expansion_score
        + 0.04 * compression_score
        + 0.03 * stability_score
    )
    return float(score)


def suppress_overlapping_interval_rows(
    rows: list[dict[str, Any]],
    *,
    max_overlap_ratio: float = 0.55,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []
    for row in sorted(rows, key=lambda item: float(item["IntervalScore"]), reverse=True):
        start = int(row["StartBarIdx"])
        end = int(row["EndBarIdx"])
        length = max(1, end - start + 1)
        overlaps = False
        for existing in selected:
            existing_start = int(existing["StartBarIdx"])
            existing_end = int(existing["EndBarIdx"])
            existing_length = max(1, existing_end - existing_start + 1)
            overlap = max(0, min(end, existing_end) - max(start, existing_start) + 1)
            if overlap / min(length, existing_length) >= max_overlap_ratio:
                overlaps = True
                break
        if not overlaps:
            selected.append(row)
    return selected


def make_train_feature_stats(train_features: pd.DataFrame) -> pd.DataFrame:
    stats = pd.DataFrame(index=train_features.columns)
    stats["median"] = train_features.median(numeric_only=True)
    q25 = train_features.quantile(0.25, numeric_only=True)
    q75 = train_features.quantile(0.75, numeric_only=True)
    stats["iqr"] = (q75 - q25).replace(0, np.nan)
    return stats


def local_driver_summary(
    row: pd.Series,
    feature_importance: pd.DataFrame,
    train_stats: pd.DataFrame,
    *,
    top_n: int,
) -> str:
    if feature_importance.empty:
        return ""

    importance_map = feature_importance.set_index("feature")["importance_mean"].to_dict()
    scored: list[tuple[float, str]] = []
    for feature, value in row.items():
        if feature not in importance_map or feature not in train_stats.index:
            continue
        if not np.isfinite(value):
            continue
        iqr = train_stats.loc[feature, "iqr"]
        median = train_stats.loc[feature, "median"]
        if not np.isfinite(iqr) or iqr == 0:
            continue
        robust_z = (float(value) - float(median)) / float(iqr)
        score = abs(float(importance_map[feature]) * robust_z)
        if score <= 0:
            continue
        scored.append((score, f"{feature}={float(value):.4g} z={robust_z:.2f}"))

    scored.sort(reverse=True)
    return " | ".join(text for _, text in scored[:top_n])


def classify_shape_hint(data: pd.DataFrame, bar_idx: int, *, lookback: int = 80) -> str:
    start = max(0, int(bar_idx) - lookback)
    if int(bar_idx) - start < 20:
        return "ML Bullish Reversal"

    window = data.iloc[start : int(bar_idx) + 1]
    lows = window["low"].to_numpy(dtype=float)
    highs = window["high"].to_numpy(dtype=float)
    close = window["close"].to_numpy(dtype=float)
    atr = float(average_true_range(window, period=14).iloc[-1])
    if not np.isfinite(atr) or atr <= 0:
        atr = max(float(np.nanstd(close)), 1e-12)

    pivot_lows = simple_pivots(lows, mode="low", order=3)
    pivot_highs = simple_pivots(highs, mode="high", order=3)
    if len(pivot_lows) >= 2:
        l1, l2 = pivot_lows[-2], pivot_lows[-1]
        low_diff_atr = abs(lows[l2] - lows[l1]) / atr
        if l2 > l1 and low_diff_atr <= 1.0:
            between_high = np.max(highs[l1:l2 + 1])
            if between_high - max(lows[l1], lows[l2]) >= atr:
                return "ML Double Bottom-like"

    if len(pivot_lows) >= 3 and len(pivot_highs) >= 2:
        left, head, right = pivot_lows[-3], pivot_lows[-2], pivot_lows[-1]
        if left < head < right and lows[head] < min(lows[left], lows[right]):
            shoulder_diff_atr = abs(lows[left] - lows[right]) / atr
            if shoulder_diff_atr <= 1.5:
                return "ML Inverse H&S-like"

    return "ML Bullish Reversal"


def simple_pivots(values: np.ndarray, *, mode: str, order: int) -> list[int]:
    pivots: list[int] = []
    for idx in range(order, len(values) - order):
        left = values[idx - order : idx]
        right = values[idx + 1 : idx + order + 1]
        if mode == "low" and values[idx] <= np.min(left) and values[idx] <= np.min(right):
            pivots.append(idx)
        if mode == "high" and values[idx] >= np.max(left) and values[idx] >= np.max(right):
            pivots.append(idx)
    return pivots


def fit_winsorize(
    X_train: pd.DataFrame,
    *,
    lower: float = 0.01,
    upper: float = 0.99,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    bounds = pd.DataFrame(index=X_train.columns)
    bounds["lower"] = X_train.quantile(lower, numeric_only=True)
    bounds["upper"] = X_train.quantile(upper, numeric_only=True)
    return apply_winsorize(X_train, bounds), bounds


def apply_winsorize(X: pd.DataFrame, bounds: pd.DataFrame) -> pd.DataFrame:
    out = X.copy()
    for column in out.columns:
        if column not in bounds.index:
            continue
        lower = bounds.loc[column, "lower"]
        upper = bounds.loc[column, "upper"]
        if np.isfinite(lower) and np.isfinite(upper) and lower < upper:
            out[column] = out[column].clip(lower=lower, upper=upper)
    return out


def average_true_range(data: pd.DataFrame, *, period: int = 14) -> pd.Series:
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    close = data["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [
            high - low,
            (high - previous_close).abs(),
            (low - previous_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    return true_range.rolling(period, min_periods=1).mean()


def rsi(close: pd.Series, *, period: int = 14) -> pd.Series:
    delta = close.diff()
    gains = delta.clip(lower=0.0)
    losses = -delta.clip(upper=0.0)
    avg_gain = gains.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    avg_loss = losses.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    rs = avg_gain / avg_loss.replace(0, np.nan)
    return 100.0 - (100.0 / (1.0 + rs))


def macd(close: pd.Series) -> tuple[pd.Series, pd.Series, pd.Series]:
    fast = close.ewm(span=12, adjust=False).mean()
    slow = close.ewm(span=26, adjust=False).mean()
    line = fast - slow
    signal = line.ewm(span=9, adjust=False).mean()
    return line, signal, line - signal


def stochastic_k(
    high: pd.Series,
    low: pd.Series,
    close: pd.Series,
    *,
    period: int,
) -> pd.Series:
    lowest = low.rolling(period).min()
    highest = high.rolling(period).max()
    return (close - lowest) / (highest - lowest).replace(0, np.nan) * 100.0


def rolling_zscore(series: pd.Series, *, window: int) -> pd.Series:
    mean = series.rolling(window).mean()
    std = series.rolling(window).std()
    return (series - mean) / std.replace(0, np.nan)


def rolling_slope_pct(series: pd.Series, *, window: int) -> pd.Series:
    x = np.arange(window, dtype=float)
    x = x - x.mean()

    def slope(values: np.ndarray) -> float:
        if np.isnan(values).any():
            return np.nan
        y = values.astype(float)
        y = y - y.mean()
        denom = float(np.dot(x, x))
        if denom == 0:
            return np.nan
        raw_slope = float(np.dot(x, y) / denom)
        return raw_slope / max(abs(float(values[-1])), 1e-12) * 100.0

    return series.rolling(window).apply(slope, raw=True)


def signed_streak(condition: pd.Series) -> pd.Series:
    values = condition.fillna(False).to_numpy(dtype=bool)
    streaks = np.zeros(len(values), dtype=float)
    current = 0
    previous = None
    for idx, value in enumerate(values):
        if previous is None or value != previous:
            current = 1
        else:
            current += 1
        streaks[idx] = current if value else -current
        previous = value
    return pd.Series(streaks, index=condition.index)


def on_balance_volume(close: pd.Series, volume: pd.Series) -> pd.Series:
    direction = np.sign(close.diff()).fillna(0.0)
    return (direction * volume).cumsum()


def walk_forward_arima_features(
    returns: pd.Series,
    *,
    order: tuple[int, int, int],
    window: int,
) -> pd.DataFrame:
    columns = ["arima_forecast_return", "arima_residual", "arima_residual_z_40"]
    out = pd.DataFrame(np.nan, index=returns.index, columns=columns)
    if ARIMA is None:
        warnings.warn("statsmodels unavailable; ARIMA features skipped.")
        return out

    values = returns.astype(float)
    for idx in range(window, len(values)):
        history = values.iloc[idx - window : idx].dropna()
        if len(history) < max(30, sum(order) * 10):
            continue
        try:
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                result = ARIMA(history, order=order).fit()
            forecast = float(result.forecast(1).iloc[0])
            residual = float(values.iloc[idx] - forecast)
            out.iloc[idx, out.columns.get_loc("arima_forecast_return")] = forecast
            out.iloc[idx, out.columns.get_loc("arima_residual")] = residual
        except Exception:
            continue
    out["arima_residual_z_40"] = rolling_zscore(out["arima_residual"], window=40)
    return out


def save_outputs(
    data: pd.DataFrame,
    result: dict[str, Any],
    config: ReversalMLConfig,
) -> Path:
    run_name = f"{config.symbol.replace('.', '_')}_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
    run_dir = Path(config.output_dir) / run_name
    run_dir.mkdir(parents=True, exist_ok=True)
    if go is None:
        warnings.warn("plotly unavailable; HTML plot files skipped. Install with: pip install plotly")

    result["signals"].to_csv(run_dir / "signals.csv", index=False)
    result["intervals"].to_csv(run_dir / "pattern_intervals.csv", index=False)
    result["model_report"].to_csv(run_dir / "model_report.csv", index=False)
    result["feature_importance"].to_csv(run_dir / "feature_importance.csv", index=False)
    result["hmm_state_summary"].to_csv(run_dir / "hmm_state_summary.csv", index=False)
    result["pca_loadings"].to_csv(run_dir / "pca_loadings.csv", index=False)

    probability_frame = pd.DataFrame(
        {
            "Date": result["probabilities"].index,
            "ReversalProbability": result["probabilities"].to_numpy(dtype=float),
        }
    )
    probability_frame.to_csv(run_dir / "probabilities.csv", index=False)

    all_probability_frame = pd.DataFrame({"Date": data.index})
    for model_name, probabilities in result["per_model_probabilities"].items():
        all_probability_frame[f"{model_name}_probability"] = probabilities.reindex(data.index).to_numpy(dtype=float)
    all_probability_frame.to_csv(run_dir / "all_model_probabilities.csv", index=False)

    all_interval_frames = [
        intervals
        for intervals in result["per_model_intervals"].values()
        if intervals is not None and not intervals.empty
    ]
    if all_interval_frames:
        pd.concat(all_interval_frames, ignore_index=True).to_csv(
            run_dir / "all_model_pattern_intervals.csv",
            index=False,
        )
    else:
        empty_interval_frame().to_csv(run_dir / "all_model_pattern_intervals.csv", index=False)

    models_dir = run_dir / "models"
    models_dir.mkdir(exist_ok=True)
    for model_name, probabilities in result["per_model_probabilities"].items():
        safe_model_name = safe_filename(model_name)
        signals = result["per_model_signals"].get(model_name, empty_signal_frame())
        importance = result["per_model_feature_importance"].get(
            model_name,
            pd.DataFrame(columns=["feature", "importance_mean", "importance_std", "rank"]),
        )
        threshold = result["per_model_thresholds"].get(model_name, np.nan)
        intervals = result["per_model_intervals"].get(model_name, empty_interval_frame())
        signals.to_csv(models_dir / f"{safe_model_name}_signals.csv", index=False)
        intervals.to_csv(models_dir / f"{safe_model_name}_pattern_intervals.csv", index=False)
        importance.to_csv(models_dir / f"{safe_model_name}_feature_importance.csv", index=False)
        make_plot(
            data=data,
            probabilities=probabilities,
            signals=signals,
            intervals=intervals,
            title=f"{config.symbol} {model_name} ML Reversal Signals",
            output_path=models_dir / f"{safe_model_name}_plot.html",
            model_name=model_name,
            threshold=threshold,
            confidence_window=config.confidence_window,
            confidence_z=config.confidence_z,
        )
        make_feature_importance_plot(
            importance=importance,
            title=f"{config.symbol} {model_name} XAI Feature Importance",
            output_path=models_dir / f"{safe_model_name}_feature_importance.html",
        )

    with open(run_dir / "config.json", "w", encoding="utf-8") as handle:
        json.dump(
            {
                **asdict(config),
                "best_model_name": result["best_model_name"],
                "threshold": result["threshold"],
            },
            handle,
            indent=2,
            default=str,
        )

    plot_path = run_dir / "reversal_ml_plot.html"
    make_plot(
        data=data,
        probabilities=result["probabilities"],
        signals=result["signals"],
        intervals=result["intervals"],
        title=f"{config.symbol} ML Reversal Signals",
        output_path=plot_path,
        model_name=f"best: {result['best_model_name']}",
        threshold=result["threshold"],
        confidence_window=config.confidence_window,
        confidence_z=config.confidence_z,
    )
    make_model_comparison_plot(
        data=data,
        per_model_probabilities=result["per_model_probabilities"],
        per_model_signals=result["per_model_signals"],
        per_model_intervals=result["per_model_intervals"],
        per_model_thresholds=result["per_model_thresholds"],
        title=f"{config.symbol} ML Reversal Model Comparison",
        output_path=run_dir / "all_models_comparison_plot.html",
    )
    return run_dir


def safe_filename(value: str) -> str:
    safe = "".join(character if character.isalnum() or character in {"-", "_"} else "_" for character in value)
    return safe.strip("_") or "model"


def hex_to_rgba(hex_color: str, alpha: float) -> str:
    value = hex_color.lstrip("#")
    if len(value) != 6:
        return f"rgba(56, 189, 248, {alpha})"
    red, green, blue = (int(value[idx : idx + 2], 16) for idx in (0, 2, 4))
    return f"rgba({red}, {green}, {blue}, {alpha})"


def add_interval_boxes(
    fig: Any,
    intervals: pd.DataFrame | None,
    *,
    color: str,
    label: str,
    opacity: float,
    show_text: bool,
) -> None:
    if intervals is None or intervals.empty:
        return

    center_x: list[float] = []
    center_y: list[float] = []
    hover_text: list[str] = []
    visible_text: list[str] = []
    for interval_number, (_, row) in enumerate(intervals.iterrows(), start=1):
        try:
            start_idx = int(row["StartBarIdx"])
            end_idx = int(row["EndBarIdx"])
            box_low = float(row["BoxLow"])
            box_high = float(row["BoxHigh"])
        except (KeyError, TypeError, ValueError):
            continue
        if not all(np.isfinite(value) for value in [start_idx, end_idx, box_low, box_high]):
            continue

        fig.add_shape(
            type="rect",
            xref="x",
            yref="y",
            x0=start_idx - 0.5,
            x1=end_idx + 0.5,
            y0=box_low,
            y1=box_high,
            fillcolor=hex_to_rgba(color, opacity),
            line=dict(color=color, width=2),
            layer="below",
        )
        center_x.append((start_idx + end_idx) / 2)
        center_y.append(box_high)
        probability_max = float(row.get("ProbabilityMax", np.nan))
        probability_mean = float(row.get("ProbabilityMean", np.nan))
        visible_text.append(f"Box {interval_number} {probability_max:.2f}")
        hover_text.append(
            (
                f"{label} interval<br>"
                f"Start: {row.get('StartDate')}<br>"
                f"End: {row.get('EndDate')}<br>"
                f"Turn: {row.get('TurnDate')}<br>"
                f"Probability peak: {row.get('ProbabilityPeakDate', row.get('PeakDate'))}<br>"
                f"Switch kind: {row.get('SwitchKind')}<br>"
                f"Mean probability: {probability_mean:.3f}<br>"
                f"Max probability: {probability_max:.3f}<br>"
                f"CI: {float(row.get('ProbabilityConfidenceLow', np.nan)):.3f}"
                f" - {float(row.get('ProbabilityConfidenceHigh', np.nan)):.3f}<br>"
                f"Volume z max: {float(row.get('VolumeZMax', np.nan)):.3f}<br>"
                f"Vol expansion: {float(row.get('VolatilityExpansionMean', np.nan)):.3f}<br>"
                f"Score: {float(row.get('IntervalScore', np.nan)):.3f}<br>"
                f"Shape: {row.get('ShapeHint')}<br>"
                f"Drivers: {row.get('TopDrivers')}"
            )
        )

    if not center_x:
        return
    mode = "markers+text" if show_text else "markers"
    fig.add_trace(
        go.Scatter(
            x=center_x,
            y=center_y,
            mode=mode,
            name=f"{label} pattern interval",
            marker=dict(
                color=color,
                size=11,
                symbol="square-open",
                line=dict(width=2),
            ),
            text=visible_text if show_text else None,
            textposition="top center",
            hovertext=hover_text,
            hoverinfo="text",
        )
    )


def make_plot(
    *,
    data: pd.DataFrame,
    probabilities: pd.Series,
    signals: pd.DataFrame,
    intervals: pd.DataFrame | None = None,
    title: str,
    output_path: Path,
    model_name: str | None = None,
    threshold: float | None = None,
    confidence_window: int = 20,
    confidence_z: float = 1.0,
) -> None:
    if go is None:
        return

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )
    add_interval_boxes(
        fig,
        intervals,
        color="#FACC15",
        label=model_name or "ML",
        opacity=0.16,
        show_text=True,
    )
    aligned_probability = probabilities.reindex(data.index).astype(float)
    probability_ci = probability_confidence_frame(
        aligned_probability,
        window=confidence_window,
        z=confidence_z,
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probability_ci["lower"].to_numpy(dtype=float),
            name="Probability CI lower",
            mode="lines",
            yaxis="y2",
            line=dict(width=0),
            hoverinfo="skip",
            showlegend=False,
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probability_ci["upper"].to_numpy(dtype=float),
            name="Probability confidence band",
            mode="lines",
            yaxis="y2",
            fill="tonexty",
            fillcolor="rgba(56, 189, 248, 0.14)",
            line=dict(width=0),
            hoverinfo="skip",
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=aligned_probability.to_numpy(dtype=float),
            name=f"{model_name or 'Model'} probability",
            mode="lines",
            yaxis="y2",
            line=dict(color="#38BDF8", width=2),
        )
    )
    if threshold is not None and np.isfinite(float(threshold)):
        fig.add_trace(
            go.Scatter(
                x=[plot_x[0], plot_x[-1]],
                y=[float(threshold), float(threshold)],
                name="Signal threshold",
                mode="lines",
                yaxis="y2",
                line=dict(color="#FACC15", width=1, dash="dash"),
                hoverinfo="skip",
            )
        )

    if not signals.empty:
        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers+text",
                name=f"{model_name or 'ML'} Buy Signal",
                marker=dict(color="#FACC15", size=12, symbol="circle"),
                text=[f"{prob:.2f}" for prob in signals["ReversalProbability"]],
                textposition="top center",
                hovertext=[
                    (
                        f"ML Buy<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Probability: {row['ReversalProbability']:.3f}<br>"
                        f"Switch: {row.get('SwitchKind')} at {row.get('SwitchTurnDate')}<br>"
                        f"Bars from turn: {row.get('BarsFromSwitchTurn')}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Probability",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=65, t=80, b=55),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def make_feature_importance_plot(
    *,
    importance: pd.DataFrame,
    title: str,
    output_path: Path,
    top_n: int = 25,
) -> None:
    if go is None or importance.empty:
        return

    plot_frame = importance.head(top_n).copy()
    plot_frame = plot_frame.sort_values("importance_mean", ascending=True)
    fig = go.Figure()
    fig.add_trace(
        go.Bar(
            x=plot_frame["importance_mean"],
            y=plot_frame["feature"],
            orientation="h",
            error_x=dict(
                type="data",
                array=plot_frame["importance_std"].clip(lower=0).to_numpy(dtype=float),
                visible=True,
                color="#94A3B8",
            ),
            marker=dict(color="#38BDF8"),
            hovertext=[
                (
                    f"Feature: {row['feature']}<br>"
                    f"Importance: {row['importance_mean']:.6f}<br>"
                    f"Std: {row['importance_std']:.6f}<br>"
                    f"Rank: {int(row['rank'])}"
                )
                for _, row in plot_frame.iterrows()
            ],
            hoverinfo="text",
            name="Permutation importance",
        )
    )
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=max(520, 24 * len(plot_frame) + 180),
        xaxis=dict(title="Average precision decrease"),
        yaxis=dict(title="Feature"),
        margin=dict(l=180, r=50, t=80, b=60),
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def make_model_comparison_plot(
    *,
    data: pd.DataFrame,
    per_model_probabilities: dict[str, pd.Series],
    per_model_signals: dict[str, pd.DataFrame],
    per_model_intervals: dict[str, pd.DataFrame],
    per_model_thresholds: dict[str, float],
    title: str,
    output_path: Path,
) -> None:
    if go is None or not per_model_probabilities:
        return

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    colors = ["#38BDF8", "#FACC15", "#A78BFA", "#34D399", "#FB7185", "#F97316"]

    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    for model_idx, (model_name, probabilities) in enumerate(per_model_probabilities.items()):
        color = colors[model_idx % len(colors)]
        threshold = per_model_thresholds.get(model_name, np.nan)
        add_interval_boxes(
            fig,
            per_model_intervals.get(model_name, empty_interval_frame()),
            color=color,
            label=model_name,
            opacity=0.08,
            show_text=False,
        )
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=probabilities.reindex(data.index).to_numpy(dtype=float),
                name=f"{model_name} probability",
                mode="lines",
                yaxis="y2",
                line=dict(color=color, width=2),
            )
        )
        if threshold is not None and np.isfinite(float(threshold)):
            fig.add_trace(
                go.Scatter(
                    x=[plot_x[0], plot_x[-1]],
                    y=[float(threshold), float(threshold)],
                    name=f"{model_name} threshold",
                    mode="lines",
                    yaxis="y2",
                    line=dict(color=color, width=1, dash="dot"),
                    hoverinfo="skip",
                    showlegend=False,
                )
            )

        signals = per_model_signals.get(model_name, empty_signal_frame())
        if signals is None or signals.empty:
            continue

        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers",
                name=f"{model_name} signals",
                marker=dict(
                    color=color,
                    size=10,
                    symbol="circle-open",
                    line=dict(width=2),
                ),
                hovertext=[
                    (
                        f"Model: {model_name}<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Probability: {row['ReversalProbability']:.3f}<br>"
                        f"Threshold: {row['SignalThreshold']:.3f}<br>"
                        f"Switch: {row.get('SwitchKind')} at {row.get('SwitchTurnDate')}<br>"
                        f"Bars from turn: {row.get('BarsFromSwitchTurn')}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=860,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Probability",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=65, t=90, b=60),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def run(config: ReversalMLConfig) -> Path:
    analysis = run_analysis(config, save=True)
    print_run_summary(analysis)
    return analysis["run_dir"]


def run_analysis(config: ReversalMLConfig, *, save: bool = True) -> dict[str, Any]:
    data = load_ohlcv(config)
    features = make_feature_frame(data, config)
    labels = make_reversal_labels(data, features, config)
    result = train_reversal_models(data, features, labels, config)
    run_dir = save_outputs(data, result, config) if save else None
    return {
        "config": config,
        "data": data,
        "features": features,
        "labels": labels,
        "result": result,
        "run_dir": run_dir,
        "csv_paths": collect_csv_paths(run_dir, result) if run_dir is not None else {},
        "plot_paths": collect_plot_paths(run_dir, result) if run_dir is not None else {},
    }


def run_notebook(
    *,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    csv_path: str | None = None,
    output_dir: str = "outputs/ml_reversal_runs",
    show_plots: bool = True,
    print_outputs: bool = True,
    save: bool = True,
    **overrides: Any,
) -> dict[str, Any]:
    config_values = {
        "symbol": symbol,
        "period": period,
        "interval": interval,
        "csv_path": csv_path,
        "output_dir": output_dir,
    }
    config_values.update(overrides)
    config = ReversalMLConfig(**config_values)
    analysis = run_analysis(config, save=save)
    if print_outputs:
        print_run_summary(analysis)
    if show_plots and analysis["plot_paths"]:
        display_notebook_plot_links(analysis["plot_paths"])
    return analysis


def collect_csv_paths(run_dir: Path | None, result: dict[str, Any]) -> dict[str, Path]:
    if run_dir is None:
        return {}
    paths: dict[str, Path] = {
        "signals": run_dir / "signals.csv",
        "pattern_intervals": run_dir / "pattern_intervals.csv",
        "model_report": run_dir / "model_report.csv",
        "feature_importance": run_dir / "feature_importance.csv",
        "hmm_state_summary": run_dir / "hmm_state_summary.csv",
        "pca_loadings": run_dir / "pca_loadings.csv",
        "probabilities": run_dir / "probabilities.csv",
        "all_model_probabilities": run_dir / "all_model_probabilities.csv",
        "all_model_pattern_intervals": run_dir / "all_model_pattern_intervals.csv",
    }
    models_dir = run_dir / "models"
    for model_name in result.get("per_model_probabilities", {}):
        safe_model_name = safe_filename(str(model_name))
        paths[f"{model_name}_signals"] = models_dir / f"{safe_model_name}_signals.csv"
        paths[f"{model_name}_pattern_intervals"] = (
            models_dir / f"{safe_model_name}_pattern_intervals.csv"
        )
        paths[f"{model_name}_feature_importance"] = (
            models_dir / f"{safe_model_name}_feature_importance.csv"
        )
    return paths


def collect_plot_paths(run_dir: Path | None, result: dict[str, Any]) -> dict[str, Path]:
    if run_dir is None:
        return {}
    paths: dict[str, Path] = {
        "best_model": run_dir / "reversal_ml_plot.html",
        "model_comparison": run_dir / "all_models_comparison_plot.html",
    }
    models_dir = run_dir / "models"
    for model_name in result.get("per_model_probabilities", {}):
        safe_model_name = safe_filename(str(model_name))
        paths[f"{model_name}_plot"] = models_dir / f"{safe_model_name}_plot.html"
        paths[f"{model_name}_feature_importance"] = (
            models_dir / f"{safe_model_name}_feature_importance.html"
        )
    return paths


def print_run_summary(analysis: dict[str, Any]) -> None:
    config = analysis["config"]
    result = analysis["result"]
    labels = analysis.get("labels")
    run_dir = analysis.get("run_dir")

    print("ML reversal run")
    print(f"Run dir: {run_dir if run_dir is not None else 'not saved'}")
    print(
        f"Symbol: {config.symbol} | Period: {config.period} | Interval: {config.interval} | "
        f"HMM: {'on' if config.use_hmm else 'off'}"
    )
    print(f"Best model: {result['best_model_name']} | Threshold: {result['threshold']:.4f}")
    print(f"Pattern intervals: {len(result['intervals'])} | Point signals: {len(result['signals'])}")
    if isinstance(labels, pd.DataFrame) and "reversal_label" in labels.columns:
        positive_labels = int(labels["reversal_label"].fillna(0).sum())
        labeled_rows = int(labels["reversal_label"].notna().sum())
        print(f"Switch labels: {positive_labels} positive / {labeled_rows} labeled rows")

    if not result["model_report"].empty:
        print("\nModel report:")
        print(result["model_report"].to_string(index=False))

    if not result["intervals"].empty:
        preview_columns = [
            column
            for column in [
                "StartDate",
                "EndDate",
                "TurnDate",
                "ProbabilityPeakDate",
                "SwitchKind",
                "Bars",
                "ProbabilityMax",
                "ProbabilityMean",
                "IntervalScore",
                "BoxLow",
                "BoxHigh",
                "ShapeHint",
            ]
            if column in result["intervals"].columns
        ]
        print("\nTop pattern intervals:")
        print(
            result["intervals"]
            .sort_values("IntervalScore", ascending=False)
            .head(10)[preview_columns]
            .to_string(index=False)
        )

    csv_paths = analysis.get("csv_paths", {})
    if csv_paths:
        print("\nCSV files:")
        for name, path in csv_paths.items():
            status = "created" if path.exists() else "expected"
            print(f"  {name}: {path} [{status}]")

    plot_paths = analysis.get("plot_paths", {})
    if plot_paths:
        print("\nPlot files:")
        if go is None:
            print("  plotly is not installed, so HTML plot files were skipped.")
        for name, path in plot_paths.items():
            status = "created" if path.exists() else "not created"
            print(f"  {name}: {path} [{status}]")


def display_notebook_plot_links(plot_paths: dict[str, Path]) -> None:
    try:
        from IPython.display import HTML, display
    except ModuleNotFoundError:
        warnings.warn("IPython unavailable; returning plot paths without notebook display.")
        return
    links = "".join(
        f'<li><a href="{path}" target="_blank">{name}</a></li>'
        for name, path in plot_paths.items()
        if path.exists()
    )
    if links:
        display(HTML(f"<b>ML reversal plots</b><ul>{links}</ul>"))


if __name__ == "__main__":
    run_notebook()
```

```python
"""
Notebook usage:

from outputs.ml_reversal_pattern_detector import run_notebook

analysis = run_notebook(
    symbol="THYAO.IS",
    period="60d",
    interval="5m",
    show_plots=True,
)

analysis["result"]["intervals"]
analysis["result"]["forecasts"]
analysis["csv_paths"]
analysis["plot_paths"]
"""

from __future__ import annotations

import json
import math
import warnings
from dataclasses import asdict, dataclass
from datetime import datetime
from pathlib import Path
from typing import Any

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    from sklearn.base import clone
    from sklearn.decomposition import PCA
    from sklearn.ensemble import HistGradientBoostingRegressor, RandomForestRegressor
    from sklearn.impute import SimpleImputer
    from sklearn.inspection import permutation_importance
    from sklearn.linear_model import Ridge
    from sklearn.metrics import (
        mean_absolute_error,
        mean_squared_error,
        precision_recall_curve,
        r2_score,
    )
    from sklearn.model_selection import TimeSeriesSplit
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
except ModuleNotFoundError:
    clone = None
    PCA = None
    HistGradientBoostingRegressor = None
    RandomForestRegressor = None
    SimpleImputer = None
    permutation_importance = None
    Ridge = None
    TimeSeriesSplit = None
    Pipeline = None
    StandardScaler = None
    mean_absolute_error = None
    mean_squared_error = None
    precision_recall_curve = None
    r2_score = None

try:
    from hmmlearn.hmm import GaussianHMM
except ModuleNotFoundError:
    GaussianHMM = None

try:
    from statsmodels.tsa.arima.model import ARIMA
except ModuleNotFoundError:
    ARIMA = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


@dataclass
class ReversalMLConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    csv_path: str | None = None
    output_dir: str = "outputs/ml_reversal_runs"
    random_state: int = 42

    label_lookback: int = 80
    label_horizon: int = 40
    min_prior_drawdown_pct: float = 2.0
    max_distance_to_low_atr: float = 2.5
    target_atr: float = 2.0
    stop_atr: float = 1.2
    forecast_horizon: int | None = None
    forecast_target_column: str = "forecast_forward_return_pct"
    forecast_return_threshold_pct: float | None = None
    forecast_score_threshold: float | None = None
    forecast_scale_floor_pct: float = 0.15
    breakout_lookback: int = 8
    breakout_buffer_atr: float = 0.05
    forecast_cross_lookback: int = 3
    forecast_min_score_change: float = 0.08
    switch_left_bars: int = 12
    switch_right_bars: int = 5
    switch_confirmation_bars: int = 8
    switch_label_window_bars: int = 1
    switch_min_rebound_atr: float = 0.55
    switch_max_rebound_atr: float = 2.4
    switch_box_left_bars: int = 5
    switch_box_right_bars: int = 10

    test_size_pct: float = 0.25
    n_splits: int = 5
    probability_threshold: float | None = None
    target_precision: float = 0.55
    cooldown_bars: int = 30
    interval_probability_floor: float | None = None
    interval_merge_gap_bars: int = 6
    min_interval_bars: int = 5
    max_interval_bars: int = 90
    interval_expand_atr: float = 0.35
    confidence_window: int = 20
    confidence_z: float = 1.0

    model_names: tuple[str, ...] = ("boosting", "ridge", "random_forest")
    use_hmm: bool = True
    hmm_states: int = 4
    use_arima: bool = False
    arima_order: tuple[int, int, int] = (1, 0, 1)
    arima_window: int = 160
    use_pca: bool = False
    pca_variance: float = 0.95
    permutation_repeats: int = 10


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")
    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if "volume" not in out.columns:
        out["volume"] = 0.0
    out["volume"] = out["volume"].fillna(0.0).clip(lower=0.0)

    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")
    return out


INTRADAY_INTERVALS = {"1m", "2m", "5m", "15m", "30m", "60m", "90m", "1h"}


def fetch_ohlcv(symbol: str, period: str, interval: str) -> pd.DataFrame:
    if Ticker is None:
        raise ImportError(
            "yahooquery is required for online data fetch. "
            "Install with: pip install yahooquery"
        )

    errors: list[str] = []
    for candidate_period in data_period_attempts(period=period, interval=interval):
        try:
            raw = Ticker(symbol).history(period=candidate_period, interval=interval)
            frame = normalize_yahoo_history(raw, symbol=symbol)
            data = prepare_ohlcv(frame)
            if candidate_period != period:
                warnings.warn(
                    f"No usable data for period={period!r}, interval={interval!r}; "
                    f"using period={candidate_period!r} instead."
                )
            return data
        except Exception as exc:
            errors.append(f"{candidate_period}: {type(exc).__name__}: {exc}")

    raise ValueError(
        f"No data returned for {symbol} with interval={interval}. "
        "For Yahoo intraday data, use a shorter period such as 60d/30d/7d "
        "or a wider interval such as 1d. Attempts: "
        + " | ".join(errors)
    )


def data_period_attempts(*, period: str, interval: str) -> list[str]:
    attempts = [str(period)]
    if str(interval).lower() in INTRADAY_INTERVALS:
        attempts.extend(["60d", "30d", "7d", "5d", "1mo"])
    else:
        attempts.extend(["2y", "1y", "6mo", "3mo"])
    return list(dict.fromkeys(attempts))


def normalize_yahoo_history(raw: Any, *, symbol: str) -> pd.DataFrame:
    if raw is None:
        raise ValueError("empty response")
    if isinstance(raw, dict):
        message = raw.get(symbol, raw)
        raise ValueError(f"Yahoo returned an error response: {message}")
    if isinstance(raw, pd.Series):
        raise ValueError(f"Yahoo returned a Series instead of OHLCV data: {raw.to_dict()}")
    if not isinstance(raw, pd.DataFrame):
        raise ValueError(f"Unexpected Yahoo response type: {type(raw).__name__}")
    if raw.empty:
        raise ValueError("empty response")

    frame = raw.copy()
    if isinstance(frame.index, pd.MultiIndex):
        if "symbol" in frame.index.names:
            frame = frame.reset_index(level="symbol", drop=True)
        else:
            frame = frame.reset_index()
    return frame


def load_ohlcv(config: ReversalMLConfig) -> pd.DataFrame:
    if config.csv_path:
        return prepare_ohlcv(pd.read_csv(config.csv_path))
    return fetch_ohlcv(config.symbol, config.period, config.interval)


def make_feature_frame(data: pd.DataFrame, config: ReversalMLConfig) -> pd.DataFrame:
    close = data["close"].astype(float)
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    open_ = data["open"].astype(float)
    volume = data["volume"].astype(float)

    features = pd.DataFrame(index=data.index)
    log_close = np.log(close)
    log_return = log_close.diff()

    features["log_return_1"] = log_return
    for window in [2, 3, 5, 10, 20, 40, 80]:
        features[f"log_return_sum_{window}"] = log_return.rolling(window).sum()
        features[f"realized_vol_{window}"] = log_return.rolling(window).std()

    atr = average_true_range(data, period=14)
    features["atr"] = atr
    features["atr_pct"] = atr / close.replace(0, np.nan) * 100.0
    features["atr_pct_change_5"] = features["atr_pct"].pct_change(5)

    features["rsi_7"] = rsi(close, period=7)
    features["rsi_14"] = rsi(close, period=14)
    macd_line, macd_signal, macd_hist = macd(close)
    features["macd_line"] = macd_line
    features["macd_signal"] = macd_signal
    features["macd_hist"] = macd_hist
    features["stoch_k_14"] = stochastic_k(high, low, close, period=14)
    features["stoch_d_3"] = features["stoch_k_14"].rolling(3).mean()

    feature_windows = sorted({20, 50, 80, int(config.label_lookback)})
    for window in feature_windows:
        roll_high = high.rolling(window).max()
        roll_low = low.rolling(window).min()
        roll_mid = close.rolling(window).mean()
        roll_std = close.rolling(window).std()

        features[f"drawdown_from_high_{window}"] = (roll_high - close) / roll_high * 100.0
        features[f"rebound_from_low_{window}"] = (close - roll_low) / roll_low * 100.0
        features[f"dist_to_low_atr_{window}"] = (close - roll_low) / atr.replace(0, np.nan)
        features[f"dist_to_high_atr_{window}"] = (roll_high - close) / atr.replace(0, np.nan)
        features[f"bollinger_z_{window}"] = (close - roll_mid) / roll_std.replace(0, np.nan)
        features[f"range_pct_{window}"] = (roll_high - roll_low) / roll_mid.replace(0, np.nan) * 100.0
        features[f"slope_pct_{window}"] = rolling_slope_pct(close, window=window)

    for window in [5, 8, 13, 21]:
        prior_high = high.rolling(window).max().shift(1)
        recent_low = low.rolling(window).min()
        recent_high = high.rolling(window).max()
        recent_range = (recent_high - recent_low).replace(0, np.nan)
        features[f"reclaim_from_low_atr_{window}"] = (close - recent_low) / atr.replace(0, np.nan)
        features[f"break_above_prior_high_atr_{window}"] = (close - prior_high) / atr.replace(0, np.nan)
        features[f"low_range_position_{window}"] = (close - recent_low) / recent_range
        features[f"short_turn_score_{window}"] = (
            features[f"reclaim_from_low_atr_{window}"]
            + features[f"break_above_prior_high_atr_{window}"].clip(lower=-2.0, upper=2.0)
        )

    features["volatility_expansion_5_20"] = (
        features["realized_vol_5"] / features["realized_vol_20"].replace(0, np.nan)
    )
    features["volatility_compression_20_80"] = (
        features["realized_vol_20"] / features["realized_vol_80"].replace(0, np.nan)
    )
    features["range_compression_20_80"] = (
        features["range_pct_20"] / features["range_pct_80"].replace(0, np.nan)
    )

    candle_range = (high - low).replace(0, np.nan)
    features["body_pct"] = (close - open_) / open_.replace(0, np.nan) * 100.0
    features["body_to_range"] = (close - open_).abs() / candle_range
    features["upper_wick_to_range"] = (high - np.maximum(open_, close)) / candle_range
    features["lower_wick_to_range"] = (np.minimum(open_, close) - low) / candle_range
    features["close_position_in_range"] = (close - low) / candle_range
    features["gap_pct"] = (open_ - close.shift(1)) / close.shift(1).replace(0, np.nan) * 100.0
    features["green_streak"] = signed_streak(close >= open_)

    features["log_volume"] = np.log1p(volume)
    features["volume_z_20"] = rolling_zscore(features["log_volume"], window=20)
    features["volume_ratio_20"] = volume / volume.rolling(20).mean().replace(0, np.nan)
    up_volume = volume.where(close >= open_, 0.0)
    down_volume = volume.where(close < open_, 0.0)
    rolling_volume_20 = volume.rolling(20).sum().replace(0, np.nan)
    features["up_volume_share_20"] = up_volume.rolling(20).sum() / rolling_volume_20
    features["down_volume_share_20"] = down_volume.rolling(20).sum() / rolling_volume_20
    features["volume_expansion_5_20"] = (
        volume.rolling(5).mean() / volume.rolling(20).mean().replace(0, np.nan)
    )
    obv = on_balance_volume(close, volume)
    features["obv_slope_20"] = rolling_slope_pct(obv, window=20)

    features["prior_down_context"] = (
        (features[f"drawdown_from_high_{config.label_lookback}"] >= config.min_prior_drawdown_pct)
        & (features[f"dist_to_low_atr_{config.label_lookback}"] <= config.max_distance_to_low_atr)
    ).astype(float)

    if config.use_arima:
        arima_features = walk_forward_arima_features(
            log_return,
            order=config.arima_order,
            window=config.arima_window,
        )
        features = features.join(arima_features)

    features = features.replace([np.inf, -np.inf], np.nan)
    return features


def make_reversal_labels(
    data: pd.DataFrame,
    features: pd.DataFrame,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    open_ = data["open"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    atr = features["atr"].to_numpy(dtype=float)
    prior_drawdown = features[f"drawdown_from_high_{config.label_lookback}"].to_numpy(dtype=float)
    dist_to_low = features[f"dist_to_low_atr_{config.label_lookback}"].to_numpy(dtype=float)

    labels = pd.DataFrame(index=data.index)
    labels["reversal_label"] = np.nan
    labels["label_target_price"] = np.nan
    labels["label_stop_price"] = np.nan
    labels["label_hit_bar_idx"] = np.nan
    labels["label_hit_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["label_hit_type"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["forecast_forward_return_pct"] = np.nan
    labels["forecast_max_return_pct"] = np.nan
    labels["forecast_min_return_pct"] = np.nan
    labels["forecast_return_atr"] = np.nan
    labels["forecast_horizon_bars"] = np.nan
    labels["prior_drawdown_pct"] = prior_drawdown
    labels["distance_to_low_atr"] = dist_to_low
    labels["switch_kind"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["switch_turn_bar_idx"] = np.nan
    labels["switch_turn_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["switch_confirm_bar_idx"] = np.nan
    labels["switch_confirm_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["bars_from_switch_turn"] = np.nan
    labels["switch_rebound_atr"] = np.nan

    n = len(data)
    switch_events = detect_bullish_switch_events(
        data=data,
        features=features,
        open_=open_,
        high=high,
        low=low,
        close=close,
        atr=atr,
        prior_drawdown=prior_drawdown,
        dist_to_low=dist_to_low,
        config=config,
    )
    switch_lookup = make_switch_label_lookup(switch_events, n=n, config=config)

    forecast_horizon = int(config.forecast_horizon or config.label_horizon)
    for bar_idx in range(n):
        if bar_idx + config.label_horizon >= n:
            continue
        if not np.isfinite(atr[bar_idx]) or atr[bar_idx] <= 0:
            continue

        forecast_end_idx = min(n - 1, bar_idx + forecast_horizon)
        if forecast_end_idx > bar_idx and close[bar_idx] != 0:
            future_slice = slice(bar_idx + 1, forecast_end_idx + 1)
            forward_return_pct = (close[forecast_end_idx] / close[bar_idx] - 1.0) * 100.0
            max_return_pct = (finite_max(high[future_slice]) / close[bar_idx] - 1.0) * 100.0
            min_return_pct = (finite_min(low[future_slice]) / close[bar_idx] - 1.0) * 100.0
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_forward_return_pct")] = forward_return_pct
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_max_return_pct")] = max_return_pct
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_min_return_pct")] = min_return_pct
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_return_atr")] = (
                close[forecast_end_idx] - close[bar_idx]
            ) / atr[bar_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_horizon_bars")] = forecast_end_idx - bar_idx

        target_price = close[bar_idx] + config.target_atr * atr[bar_idx]
        stop_price = close[bar_idx] - config.stop_atr * atr[bar_idx]
        labels.iloc[bar_idx, labels.columns.get_loc("label_target_price")] = target_price
        labels.iloc[bar_idx, labels.columns.get_loc("label_stop_price")] = stop_price

        in_reversal_context = (
            np.isfinite(prior_drawdown[bar_idx])
            and np.isfinite(dist_to_low[bar_idx])
            and prior_drawdown[bar_idx] >= config.min_prior_drawdown_pct
            and dist_to_low[bar_idx] <= config.max_distance_to_low_atr
        )
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=bar_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        switch_event = switch_lookup.get(bar_idx)
        label = int(
            switch_event is not None
            and in_reversal_context
            and hit_type == "target"
        )
        labels.iloc[bar_idx, labels.columns.get_loc("reversal_label")] = label
        if hit_idx is not None:
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_bar_idx")] = int(hit_idx)
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_date")] = data.index[int(hit_idx)]
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_type")] = hit_type
        if switch_event is not None:
            turn_idx = int(switch_event["turn_idx"])
            confirm_idx = int(switch_event["confirm_idx"])
            labels.iloc[bar_idx, labels.columns.get_loc("switch_kind")] = switch_event["kind"]
            labels.iloc[bar_idx, labels.columns.get_loc("switch_turn_bar_idx")] = turn_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_turn_date")] = data.index[turn_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("switch_confirm_bar_idx")] = confirm_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_confirm_date")] = data.index[confirm_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("bars_from_switch_turn")] = bar_idx - turn_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_rebound_atr")] = switch_event["rebound_atr"]

    return labels


def detect_bullish_switch_events(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr: np.ndarray,
    prior_drawdown: np.ndarray,
    dist_to_low: np.ndarray,
    config: ReversalMLConfig,
) -> list[dict[str, Any]]:
    prior_context = features.get("prior_down_context", pd.Series(0.0, index=features.index)).to_numpy(
        dtype=float
    )
    left_bars = max(3, int(config.switch_left_bars))
    right_bars = max(1, int(config.switch_right_bars))
    confirmation_bars = max(right_bars, int(config.switch_confirmation_bars))
    n = len(data)
    max_future_need = max(right_bars, confirmation_bars, int(config.label_horizon))
    events: list[dict[str, Any]] = []

    for turn_idx in range(left_bars, n - max_future_need):
        if not np.isfinite(atr[turn_idx]) or atr[turn_idx] <= 0:
            continue
        if not is_bullish_switch_context(
            turn_idx=turn_idx,
            close=close,
            high=high,
            prior_context=prior_context,
            prior_drawdown=prior_drawdown,
            dist_to_low=dist_to_low,
            config=config,
        ):
            continue
        if not is_local_valley(low, turn_idx=turn_idx, left_bars=left_bars, right_bars=right_bars):
            continue

        confirm_idx = find_bullish_switch_confirmation(
            turn_idx=turn_idx,
            open_=open_,
            high=high,
            low=low,
            close=close,
            atr=atr,
            config=config,
        )
        if confirm_idx is None or confirm_idx + config.label_horizon >= n:
            continue

        target_price = close[confirm_idx] + config.target_atr * atr[confirm_idx]
        stop_price = close[confirm_idx] - config.stop_atr * atr[confirm_idx]
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=confirm_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        if hit_type != "target":
            continue

        events.append(
            {
                "kind": "valley",
                "turn_idx": int(turn_idx),
                "confirm_idx": int(confirm_idx),
                "hit_idx": int(hit_idx) if hit_idx is not None else None,
                "rebound_atr": float((close[confirm_idx] - low[turn_idx]) / atr[turn_idx]),
                "turn_price": float(low[turn_idx]),
                "confirm_price": float(close[confirm_idx]),
            }
        )

    return suppress_overlapping_switch_events(
        events,
        min_gap=max(2, int(config.switch_confirmation_bars)),
    )


def is_bullish_switch_context(
    *,
    turn_idx: int,
    close: np.ndarray,
    high: np.ndarray,
    prior_context: np.ndarray,
    prior_drawdown: np.ndarray,
    dist_to_low: np.ndarray,
    config: ReversalMLConfig,
) -> bool:
    context_start = max(0, int(turn_idx) - int(config.label_lookback))
    context_seen = finite_max(prior_context[context_start : turn_idx + 1])
    if not np.isfinite(context_seen) or context_seen <= 0:
        return False
    if not np.isfinite(prior_drawdown[turn_idx]) or not np.isfinite(dist_to_low[turn_idx]):
        return False
    if prior_drawdown[turn_idx] < config.min_prior_drawdown_pct:
        return False
    if dist_to_low[turn_idx] > config.max_distance_to_low_atr:
        return False

    approach_start = max(0, int(turn_idx) - max(5, int(config.switch_left_bars)))
    if turn_idx - approach_start < 4:
        return False
    approach_high = finite_max(high[approach_start : turn_idx + 1])
    if not np.isfinite(approach_high) or approach_high <= 0:
        return False
    approach_drawdown = (approach_high - close[turn_idx]) / approach_high * 100.0
    if approach_drawdown < config.min_prior_drawdown_pct * 0.35:
        return False
    return close[turn_idx] <= np.nanmedian(close[approach_start : turn_idx + 1])


def is_local_valley(
    low: np.ndarray,
    *,
    turn_idx: int,
    left_bars: int,
    right_bars: int,
) -> bool:
    start = max(0, int(turn_idx) - int(left_bars))
    end = min(len(low) - 1, int(turn_idx) + int(right_bars))
    window = low[start : end + 1]
    if len(window) == 0 or not np.isfinite(low[turn_idx]):
        return False
    valley = finite_min(window)
    if not np.isfinite(valley):
        return False
    if low[turn_idx] > valley + max(abs(valley) * 1e-8, 1e-10):
        return False
    right_window = low[turn_idx + 1 : end + 1]
    return len(right_window) == 0 or finite_min(right_window) > low[turn_idx] - max(abs(low[turn_idx]) * 1e-8, 1e-10)


def find_bullish_switch_confirmation(
    *,
    turn_idx: int,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr: np.ndarray,
    config: ReversalMLConfig,
) -> int | None:
    atr_at_turn = float(atr[turn_idx])
    if not np.isfinite(atr_at_turn) or atr_at_turn <= 0:
        return None
    end_idx = min(len(close) - 1, int(turn_idx) + int(config.switch_confirmation_bars))
    rebound_level = low[turn_idx] + config.switch_min_rebound_atr * atr_at_turn
    micro_start = max(0, int(turn_idx) - max(3, int(config.switch_left_bars) // 2))
    micro_break_level = finite_max(high[micro_start:turn_idx])

    for idx in range(int(turn_idx) + 1, end_idx + 1):
        rebound_atr = (close[idx] - low[turn_idx]) / atr_at_turn
        if rebound_atr < config.switch_min_rebound_atr:
            continue
        if rebound_atr > config.switch_max_rebound_atr:
            continue
        closes_green = close[idx] > open_[idx]
        breaks_previous_high = idx > 0 and close[idx] > high[idx - 1]
        breaks_micro_high = np.isfinite(micro_break_level) and close[idx] > micro_break_level
        reclaims_valley = close[idx] >= rebound_level
        if reclaims_valley and (closes_green or breaks_previous_high or breaks_micro_high):
            return int(idx)
    return None


def make_switch_label_lookup(
    events: list[dict[str, Any]],
    *,
    n: int,
    config: ReversalMLConfig,
) -> dict[int, dict[str, Any]]:
    lookup: dict[int, dict[str, Any]] = {}
    radius = max(0, int(config.switch_label_window_bars))
    for event in events:
        anchor_idx = int(event["confirm_idx"])
        turn_idx = int(event["turn_idx"])
        for label_idx in range(anchor_idx - radius, anchor_idx + radius + 1):
            if label_idx < turn_idx or label_idx < 0 or label_idx >= n:
                continue
            current = lookup.get(label_idx)
            if current is None:
                lookup[label_idx] = event
                continue
            current_distance = abs(label_idx - int(current["confirm_idx"]))
            new_distance = abs(label_idx - anchor_idx)
            if new_distance < current_distance:
                lookup[label_idx] = event
    return lookup


def suppress_overlapping_switch_events(
    events: list[dict[str, Any]],
    *,
    min_gap: int,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []
    for event in sorted(events, key=lambda item: int(item["confirm_idx"])):
        if not selected:
            selected.append(event)
            continue
        previous = selected[-1]
        if int(event["confirm_idx"]) - int(previous["confirm_idx"]) > min_gap:
            selected.append(event)
            continue
        if float(event["turn_price"]) < float(previous["turn_price"]):
            selected[-1] = event
    return selected


def first_long_barrier_hit(
    *,
    high: np.ndarray,
    low: np.ndarray,
    start_idx: int,
    horizon: int,
    target_price: float,
    stop_price: float,
) -> tuple[str, int | None]:
    max_idx = min(len(high) - 1, int(start_idx) + int(horizon))
    for idx in range(int(start_idx) + 1, max_idx + 1):
        stop_hit = low[idx] <= stop_price
        target_hit = high[idx] >= target_price
        if stop_hit and target_hit:
            return "stop", idx
        if stop_hit:
            return "stop", idx
        if target_hit:
            return "target", idx
    return "timeout", None


def add_hmm_features(
    features: pd.DataFrame,
    train_index: pd.Index,
    config: ReversalMLConfig,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    state_summary_columns = ["state", "train_count", "mean_return", "mean_vol", "mean_drawdown"]
    if not config.use_hmm:
        return features, pd.DataFrame(columns=state_summary_columns)
    if GaussianHMM is None or SimpleImputer is None or StandardScaler is None:
        warnings.warn("hmmlearn/sklearn unavailable; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    hmm_columns = [
        column
        for column in [
            "log_return_1",
            "realized_vol_10",
            "atr_pct",
            "volume_z_20",
            f"drawdown_from_high_{config.label_lookback}",
            f"dist_to_low_atr_{config.label_lookback}",
        ]
        if column in features.columns
    ]
    if len(hmm_columns) < 3:
        return features, pd.DataFrame(columns=state_summary_columns)

    train = features.loc[train_index, hmm_columns]
    if train.dropna(how="all").shape[0] < config.hmm_states * 20:
        warnings.warn("Not enough training rows for HMM; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    imputer = SimpleImputer(strategy="median")
    scaler = StandardScaler()
    train_values = scaler.fit_transform(imputer.fit_transform(train))
    all_values = scaler.transform(imputer.transform(features[hmm_columns]))

    hmm = GaussianHMM(
        n_components=config.hmm_states,
        covariance_type="diag",
        n_iter=200,
        tol=1e-4,
        random_state=config.random_state,
    )
    try:
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            hmm.fit(train_values)
            probabilities = hmm.predict_proba(all_values)
    except Exception as exc:
        warnings.warn(f"HMM regime features skipped after fit failure: {type(exc).__name__}: {exc}")
        return features, pd.DataFrame(columns=state_summary_columns)
    states = probabilities.argmax(axis=1)

    out = features.copy()
    out["hmm_state"] = states.astype(float)
    for state in range(config.hmm_states):
        out[f"hmm_state_prob_{state}"] = probabilities[:, state]

    state_rows: list[dict[str, Any]] = []
    train_augmented = out.loc[train_index].copy()
    for state in range(config.hmm_states):
        subset = train_augmented[train_augmented["hmm_state"] == state]
        state_rows.append(
            {
                "state": state,
                "train_count": int(len(subset)),
                "mean_return": float(subset["log_return_1"].mean()),
                "mean_vol": float(subset["realized_vol_10"].mean()),
                "mean_drawdown": float(
                    subset[f"drawdown_from_high_{config.label_lookback}"].mean()
                ),
            }
        )
    return out, pd.DataFrame(state_rows, columns=state_summary_columns)


def train_reversal_models(
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    config: ReversalMLConfig,
) -> dict[str, Any]:
    require_sklearn()

    target_column = config.forecast_target_column
    if target_column not in labels.columns:
        raise ValueError(f"Unknown forecast target column: {target_column}")

    labeled = features.join(labels[[target_column]], how="inner")
    labeled = labeled.dropna(subset=[target_column])
    if len(labeled) < 200:
        raise ValueError("Not enough labeled rows. Increase period or use a higher timeframe.")

    split_idx = int(len(labeled) * (1.0 - config.test_size_pct))
    split_idx = min(max(split_idx, 50), len(labeled) - 20)
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]

    features_with_hmm, hmm_summary = add_hmm_features(features, train_index, config)
    labeled = features_with_hmm.join(labels[[target_column]], how="inner")
    labeled = labeled.dropna(subset=[target_column])
    y = labeled[target_column].astype(float)
    feature_columns = [
        column
        for column in labeled.columns
        if column != target_column and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]
    split_idx = int(len(labeled) * (1.0 - config.test_size_pct))
    split_idx = min(max(split_idx, 50), len(labeled) - 20)
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]
    X_train = X.loc[train_index]
    y_train = y.loc[train_index]
    X_test = X.loc[test_index]
    y_test = y.loc[test_index]
    if y_train.dropna().nunique() < 2:
        raise ValueError(
            "Training split has no return variation. Use more history or a higher timeframe."
        )

    X_train_clip, clip_bounds = fit_winsorize(X_train)
    X_test_clip = apply_winsorize(X_test, clip_bounds)
    X_all_clip = apply_winsorize(features_with_hmm[feature_columns], clip_bounds)
    return_threshold_pct = choose_return_threshold(y_train, config)
    forecast_score_scale = choose_forecast_score_scale(y_train, config)
    train_sample_weight = make_return_sample_weight(labels, y_train, train_index, config)

    models = build_models(config)
    results: list[dict[str, Any]] = []
    fitted_models: dict[str, Pipeline] = {}

    for name, model in models.items():
        fitted = fit_pipeline_with_weights(
            model,
            X_train_clip,
            y_train,
            sample_weight=train_sample_weight,
        )
        fitted_models[name] = fitted
        test_forecast = predict_return(fitted, X_test_clip)
        metrics = evaluate_return_forecasts(
            y_test,
            test_forecast,
            return_threshold_pct=return_threshold_pct,
        )
        cv_metrics = time_series_cv_score(
            model,
            X_train_clip,
            y_train,
            n_splits=config.n_splits,
            return_threshold_pct=return_threshold_pct,
        )
        result = {"model": name, **metrics, **cv_metrics}
        results.append(result)

    model_report = pd.DataFrame(results)
    best_name = choose_best_model(model_report)
    per_model_probabilities: dict[str, pd.Series] = {}
    per_model_forecasts: dict[str, pd.Series] = {}
    per_model_thresholds: dict[str, float] = {}
    per_model_feature_importance: dict[str, pd.DataFrame] = {}
    per_model_signals: dict[str, pd.DataFrame] = {}
    per_model_intervals: dict[str, pd.DataFrame] = {}

    for name, fitted in fitted_models.items():
        all_forecast = predict_return(fitted, X_all_clip)
        test_forecast = predict_return(fitted, X_test_clip)
        all_score = forecast_to_score(
            all_forecast,
            return_threshold_pct=return_threshold_pct,
            scale_pct=forecast_score_scale,
        )
        test_score = forecast_to_score(
            test_forecast,
            return_threshold_pct=return_threshold_pct,
            scale_pct=forecast_score_scale,
        )
        model_threshold = config.forecast_score_threshold
        if model_threshold is None:
            model_threshold = config.probability_threshold
        if model_threshold is None:
            model_threshold = choose_forecast_score_threshold(
                y_test,
                test_score,
                return_threshold_pct=return_threshold_pct,
                target_precision=config.target_precision,
            )

        forecast_series = pd.Series(
            all_forecast,
            index=features_with_hmm.index,
            name=f"{name}_return_forecast_pct",
        )
        score_series = pd.Series(
            all_score,
            index=features_with_hmm.index,
            name=f"{name}_forecast_score",
        )
        importance = permutation_xai(
            fitted,
            X_test_clip,
            y_test,
            feature_columns,
            repeats=config.permutation_repeats,
            random_state=config.random_state,
        )
        signals = build_signal_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=score_series,
            return_forecasts=forecast_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not signals.empty:
            signals.insert(0, "Model", name)
        intervals = build_interval_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=score_series,
            return_forecasts=forecast_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not intervals.empty:
            intervals.insert(0, "Model", name)

        per_model_probabilities[name] = score_series
        per_model_forecasts[name] = forecast_series
        per_model_thresholds[name] = float(model_threshold)
        per_model_feature_importance[name] = importance
        per_model_signals[name] = signals
        per_model_intervals[name] = intervals

    best_model = fitted_models[best_name]
    threshold = per_model_thresholds[best_name]
    probabilities = per_model_probabilities[best_name].rename("forecast_score")
    forecasts = per_model_forecasts[best_name].rename("return_forecast_pct")
    signals = per_model_signals[best_name].copy()
    if "Model" in signals.columns:
        signals = signals.drop(columns=["Model"])
    intervals = per_model_intervals[best_name].copy()
    if "Model" in intervals.columns:
        intervals = intervals.drop(columns=["Model"])
    feature_importance = per_model_feature_importance[best_name]
    pca_loadings = extract_pca_loadings(best_model, feature_columns)

    return {
        "features": features_with_hmm,
        "labels": labels,
        "feature_columns": feature_columns,
        "model_report": model_report,
        "best_model_name": best_name,
        "best_model": best_model,
        "probabilities": probabilities,
        "forecasts": forecasts,
        "threshold": threshold,
        "return_threshold_pct": return_threshold_pct,
        "forecast_score_scale": forecast_score_scale,
        "signals": signals,
        "intervals": intervals,
        "feature_importance": feature_importance,
        "per_model_probabilities": per_model_probabilities,
        "per_model_forecasts": per_model_forecasts,
        "per_model_thresholds": per_model_thresholds,
        "per_model_signals": per_model_signals,
        "per_model_intervals": per_model_intervals,
        "per_model_feature_importance": per_model_feature_importance,
        "pca_loadings": pca_loadings,
        "hmm_state_summary": hmm_summary,
    }


def require_sklearn() -> None:
    if Pipeline is None:
        raise ImportError(
            "scikit-learn is required for ML training. Install with: pip install scikit-learn"
        )


def build_models(config: ReversalMLConfig) -> dict[str, Pipeline]:
    requested = {name.strip().lower() for name in config.model_names}
    models: dict[str, Pipeline] = {}

    if "boosting" in requested:
        models["boosting"] = make_pipeline(
            HistGradientBoostingRegressor(
                learning_rate=0.05,
                max_iter=350,
                max_leaf_nodes=15,
                l2_regularization=0.05,
                early_stopping=True,
                random_state=config.random_state,
            ),
            config,
            scale=False,
        )
    if "ridge" in requested:
        models["ridge"] = make_pipeline(
            Ridge(
                alpha=5.0,
                random_state=config.random_state,
            ),
            config,
            scale=True,
        )
    if "random_forest" in requested:
        models["random_forest"] = make_pipeline(
            RandomForestRegressor(
                n_estimators=400,
                max_depth=7,
                min_samples_leaf=20,
                random_state=config.random_state,
                n_jobs=-1,
            ),
            config,
            scale=False,
        )
    if not models:
        raise ValueError("No valid model names requested.")
    return models


def make_pipeline(estimator: Any, config: ReversalMLConfig, *, scale: bool) -> Pipeline:
    steps: list[tuple[str, Any]] = [("imputer", SimpleImputer(strategy="median"))]
    if scale or config.use_pca:
        steps.append(("scaler", StandardScaler()))
    if config.use_pca:
        steps.append(("pca", PCA(n_components=config.pca_variance, svd_solver="full")))
    steps.append(("model", estimator))
    return Pipeline(steps)


def fit_pipeline_with_weights(
    model: Pipeline,
    X: pd.DataFrame,
    y: pd.Series,
    *,
    sample_weight: np.ndarray | pd.Series | None = None,
) -> Pipeline:
    try:
        return model.fit(X, y, model__sample_weight=sample_weight)
    except TypeError:
        return model.fit(X, y)


def predict_return(model: Pipeline, X: pd.DataFrame) -> np.ndarray:
    return np.asarray(model.predict(X), dtype=float)


def evaluate_return_forecasts(
    y_true: pd.Series,
    forecast: np.ndarray,
    *,
    return_threshold_pct: float,
) -> dict[str, float]:
    y_values = y_true.to_numpy(dtype=float)
    forecast_values = np.asarray(forecast, dtype=float)
    finite_mask = np.isfinite(y_values) & np.isfinite(forecast_values)
    out: dict[str, float] = {
        "test_mae": np.nan,
        "test_rmse": np.nan,
        "test_r2": np.nan,
        "test_correlation": np.nan,
        "test_directional_accuracy": np.nan,
        "test_actual_positive_rate": np.nan,
        "test_predicted_positive_rate": np.nan,
        "test_positive_precision": np.nan,
        "test_positive_recall": np.nan,
    }
    if finite_mask.sum() < 3:
        return out

    y_values = y_values[finite_mask]
    forecast_values = forecast_values[finite_mask]
    out["test_mae"] = float(mean_absolute_error(y_values, forecast_values))
    out["test_rmse"] = float(math.sqrt(mean_squared_error(y_values, forecast_values)))
    out["test_r2"] = float(r2_score(y_values, forecast_values))
    if np.std(y_values) > 0 and np.std(forecast_values) > 0:
        out["test_correlation"] = float(np.corrcoef(y_values, forecast_values)[0, 1])
    out["test_directional_accuracy"] = float(np.mean(np.sign(y_values) == np.sign(forecast_values)))

    actual_positive = y_values >= float(return_threshold_pct)
    predicted_positive = forecast_values >= float(return_threshold_pct)
    out["test_actual_positive_rate"] = float(np.mean(actual_positive))
    out["test_predicted_positive_rate"] = float(np.mean(predicted_positive))
    true_positive = float(np.sum(actual_positive & predicted_positive))
    out["test_positive_precision"] = true_positive / max(float(np.sum(predicted_positive)), 1.0)
    out["test_positive_recall"] = true_positive / max(float(np.sum(actual_positive)), 1.0)
    return out


def time_series_cv_score(
    model: Pipeline,
    X: pd.DataFrame,
    y: pd.Series,
    *,
    n_splits: int,
    return_threshold_pct: float,
) -> dict[str, float]:
    out = {"cv_mae": np.nan, "cv_rmse": np.nan, "cv_correlation": np.nan}
    if len(X) < 200 or y.dropna().nunique() < 2:
        return out

    split_count = min(n_splits, max(2, len(X) // 200))
    splitter = TimeSeriesSplit(n_splits=split_count)
    mae_scores: list[float] = []
    rmse_scores: list[float] = []
    corr_scores: list[float] = []
    for train_idx, test_idx in splitter.split(X):
        y_train = y.iloc[train_idx]
        y_valid = y.iloc[test_idx]
        if y_train.dropna().nunique() < 2:
            continue
        fitted = fit_pipeline_with_weights(clone(model), X.iloc[train_idx], y_train)
        forecast = predict_return(fitted, X.iloc[test_idx])
        metrics = evaluate_return_forecasts(
            y_valid,
            forecast,
            return_threshold_pct=return_threshold_pct,
        )
        if np.isfinite(metrics["test_mae"]):
            mae_scores.append(metrics["test_mae"])
        if np.isfinite(metrics["test_rmse"]):
            rmse_scores.append(metrics["test_rmse"])
        if np.isfinite(metrics["test_correlation"]):
            corr_scores.append(metrics["test_correlation"])

    if mae_scores:
        out["cv_mae"] = float(np.mean(mae_scores))
    if rmse_scores:
        out["cv_rmse"] = float(np.mean(rmse_scores))
    if corr_scores:
        out["cv_correlation"] = float(np.mean(corr_scores))
    return out


def choose_best_model(model_report: pd.DataFrame) -> str:
    ranking = model_report.copy()
    mae_rank = ranking["test_mae"].rank(ascending=True, pct=True).fillna(1.0)
    ranking["_rank_score"] = (
        ranking["test_correlation"].fillna(0.0)
        + ranking["test_directional_accuracy"].fillna(0.0)
        - mae_rank
    )
    ranking = ranking.sort_values(["_rank_score", "test_rmse"], ascending=[False, True])
    return str(ranking.iloc[0]["model"])


def choose_forecast_score_threshold(
    y_true: pd.Series,
    forecast_score: np.ndarray,
    *,
    return_threshold_pct: float,
    target_precision: float,
) -> float:
    y_values = y_true.to_numpy(dtype=float)
    score_values = np.asarray(forecast_score, dtype=float)
    finite_mask = np.isfinite(y_values) & np.isfinite(score_values)
    if finite_mask.sum() < 3:
        return 0.60
    actual_positive = y_values[finite_mask] >= float(return_threshold_pct)
    score_values = score_values[finite_mask]
    if len(np.unique(actual_positive)) < 2 or precision_recall_curve is None:
        return 0.60

    precision, recall, thresholds = precision_recall_curve(actual_positive.astype(int), score_values)
    candidates: list[tuple[float, float, float]] = []
    for threshold, prec, rec in zip(thresholds, precision[:-1], recall[:-1]):
        if prec >= target_precision and rec > 0:
            candidates.append((float(threshold), float(prec), float(rec)))
    if not candidates:
        return float(np.quantile(score_values, 0.80))
    candidates.sort(key=lambda item: (item[2], item[1]), reverse=True)
    return float(candidates[0][0])


def choose_return_threshold(y_train: pd.Series, config: ReversalMLConfig) -> float:
    if config.forecast_return_threshold_pct is not None:
        return float(config.forecast_return_threshold_pct)

    values = y_train.dropna().to_numpy(dtype=float)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return float(config.forecast_scale_floor_pct)
    positive_values = values[values > 0]
    if len(positive_values) >= 10:
        threshold = float(np.quantile(positive_values, 0.40))
    else:
        threshold = float(np.quantile(values, 0.65))
    return max(float(config.forecast_scale_floor_pct), threshold)


def choose_forecast_score_scale(y_train: pd.Series, config: ReversalMLConfig) -> float:
    values = y_train.dropna().to_numpy(dtype=float)
    values = values[np.isfinite(values)]
    if len(values) < 5:
        return float(config.forecast_scale_floor_pct)
    q25, q75 = np.quantile(values, [0.25, 0.75])
    robust_scale = float((q75 - q25) / 2.0)
    std_scale = float(np.std(values))
    scale = max(robust_scale, std_scale * 0.35, float(config.forecast_scale_floor_pct), 1e-6)
    return float(scale)


def forecast_to_score(
    forecast: np.ndarray | pd.Series,
    *,
    return_threshold_pct: float,
    scale_pct: float,
) -> np.ndarray:
    values = np.asarray(forecast, dtype=float)
    z = (values - float(return_threshold_pct)) / max(float(scale_pct), 1e-6)
    z = np.clip(z, -30.0, 30.0)
    return 1.0 / (1.0 + np.exp(-z))


def make_return_sample_weight(
    labels: pd.DataFrame,
    y_train: pd.Series,
    train_index: pd.Index,
    config: ReversalMLConfig,
) -> np.ndarray:
    weights = np.ones(len(y_train), dtype=float)
    aligned_labels = labels.reindex(train_index)
    if "switch_kind" in aligned_labels.columns:
        weights += aligned_labels["switch_kind"].notna().to_numpy(dtype=float) * 1.5
    if "prior_drawdown_pct" in aligned_labels.columns:
        drawdown = aligned_labels["prior_drawdown_pct"].fillna(0.0).to_numpy(dtype=float)
        weights += np.clip(drawdown / max(config.min_prior_drawdown_pct, 1e-6), 0.0, 2.0) * 0.35
    values = y_train.to_numpy(dtype=float)
    finite_values = values[np.isfinite(values)]
    if len(finite_values) > 5:
        scale = max(float(np.quantile(np.abs(finite_values), 0.75)), config.forecast_scale_floor_pct)
        weights += np.clip(np.abs(values) / scale, 0.0, 2.0) * 0.25
    return weights


def permutation_xai(
    model: Pipeline,
    X_test: pd.DataFrame,
    y_test: pd.Series,
    feature_columns: list[str],
    *,
    repeats: int,
    random_state: int,
) -> pd.DataFrame:
    columns = ["feature", "importance_mean", "importance_std", "rank"]
    if len(X_test) < 5 or permutation_importance is None:
        return pd.DataFrame(columns=columns)

    result = permutation_importance(
        model,
        X_test,
        y_test,
        n_repeats=repeats,
        random_state=random_state,
        scoring="neg_mean_absolute_error",
        n_jobs=-1,
    )
    importance = pd.DataFrame(
        {
            "feature": feature_columns,
            "importance_mean": result.importances_mean,
            "importance_std": result.importances_std,
        }
    )
    importance = importance.sort_values("importance_mean", ascending=False).reset_index(drop=True)
    importance["rank"] = np.arange(1, len(importance) + 1)
    return importance[columns]


def extract_pca_loadings(model: Pipeline, feature_columns: list[str]) -> pd.DataFrame:
    if "pca" not in model.named_steps:
        return pd.DataFrame(columns=["component", "feature", "loading", "abs_loading", "rank"])

    pca = model.named_steps["pca"]
    rows: list[dict[str, Any]] = []
    for component_idx, component in enumerate(pca.components_):
        for feature, loading in zip(feature_columns, component):
            rows.append(
                {
                    "component": f"PC{component_idx + 1}",
                    "feature": feature,
                    "loading": float(loading),
                    "abs_loading": float(abs(loading)),
                }
            )
    out = pd.DataFrame(rows)
    out["rank"] = out.groupby("component")["abs_loading"].rank(ascending=False, method="first")
    return out.sort_values(["component", "rank"]).reset_index(drop=True)


def build_signal_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    return_forecasts: pd.Series | None,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    candidate_probability = probabilities.dropna()
    candidate_probability = candidate_probability[candidate_probability >= threshold]
    if return_forecasts is not None and not candidate_probability.empty:
        floor = config.interval_probability_floor
        if floor is None:
            floor = max(0.05, min(0.95, float(threshold) * 0.80))
        breakout_filter = forecast_breakout_filter(
            data=data,
            features=features,
            forecast_score=probabilities.reindex(features.index).astype(float),
            return_forecast=return_forecasts.reindex(features.index).astype(float),
            threshold=float(threshold),
            floor=float(floor),
            config=config,
        )
        candidate_probability = candidate_probability[
            breakout_filter.reindex(candidate_probability.index).fillna(False)
        ]
    if candidate_probability.empty:
        return empty_signal_frame()

    selected_indexes = non_max_suppression_1d(
        candidate_probability,
        cooldown_bars=config.cooldown_bars,
        all_index=features.index,
    )

    train_stats = make_train_feature_stats(train_features)
    rows: list[dict[str, Any]] = []
    atr = features["atr"]
    for timestamp in selected_indexes:
        bar_idx = int(data.index.get_loc(timestamp))
        entry_price = float(data.loc[timestamp, "close"])
        return_forecast_pct = (
            float(return_forecasts.loc[timestamp])
            if return_forecasts is not None and timestamp in return_forecasts.index
            else np.nan
        )
        target_price = entry_price + config.target_atr * float(atr.loc[timestamp])
        stop_price = entry_price - config.stop_atr * float(atr.loc[timestamp])
        label_row = labels.loc[timestamp] if timestamp in labels.index else pd.Series(dtype=object)
        hit_idx = label_row.get("label_hit_bar_idx", np.nan)
        hit_date = label_row.get("label_hit_date", pd.NaT)
        hit_type = label_row.get("label_hit_type", None)
        realized_label = label_row.get("reversal_label", np.nan)
        switch_kind = label_row.get("switch_kind", None)
        switch_turn_idx = label_row.get("switch_turn_bar_idx", np.nan)
        switch_turn_date = label_row.get("switch_turn_date", pd.NaT)
        switch_confirm_idx = label_row.get("switch_confirm_bar_idx", np.nan)
        switch_confirm_date = label_row.get("switch_confirm_date", pd.NaT)
        bars_from_switch_turn = label_row.get("bars_from_switch_turn", np.nan)
        top_drivers = local_driver_summary(
            features.loc[timestamp, feature_columns],
            feature_importance,
            train_stats,
            top_n=6,
        )
        rows.append(
            {
                "EntryBarIdx": bar_idx,
                "EntryDate": timestamp,
                "EntryPrice": entry_price,
                "ReturnForecastPct": return_forecast_pct,
                "ForecastScore": float(probabilities.loc[timestamp]),
                "ReversalProbability": float(probabilities.loc[timestamp]),
                "SignalThreshold": float(threshold),
                "TargetPrice": target_price,
                "StopPrice": stop_price,
                "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                "OutcomeKnown": is_finite_number(realized_label),
                "OutcomeLabel": realized_label,
                "HitBarIdx": int(hit_idx) if is_finite_number(hit_idx) else np.nan,
                "HitDate": hit_date,
                "HitType": hit_type,
                "SwitchKind": switch_kind,
                "SwitchTurnBarIdx": int(switch_turn_idx) if is_finite_number(switch_turn_idx) else np.nan,
                "SwitchTurnDate": switch_turn_date,
                "SwitchConfirmBarIdx": int(switch_confirm_idx) if is_finite_number(switch_confirm_idx) else np.nan,
                "SwitchConfirmDate": switch_confirm_date,
                "BarsFromSwitchTurn": bars_from_switch_turn,
                "ShapeHint": classify_shape_hint(data, bar_idx),
                "TopDrivers": top_drivers,
            }
        )
    return pd.DataFrame(rows).sort_values("EntryBarIdx").reset_index(drop=True)


def build_interval_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    return_forecasts: pd.Series | None,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    aligned_probability = probabilities.reindex(features.index).astype(float)
    aligned_return_forecast = (
        return_forecasts.reindex(features.index).astype(float)
        if return_forecasts is not None
        else pd.Series(np.nan, index=features.index)
    )
    confidence = probability_confidence_frame(
        aligned_probability,
        window=config.confidence_window,
        z=config.confidence_z,
    )
    floor = config.interval_probability_floor
    if floor is None:
        floor = max(0.05, min(0.95, float(threshold) * 0.80))

    prior_context = features.get("prior_down_context", pd.Series(0.0, index=features.index))
    context_seen = prior_context.fillna(0.0).rolling(
        config.label_lookback,
        min_periods=1,
    ).max()
    candidate_signal = (
        (aligned_probability >= floor)
        | (
            (confidence["upper"] >= float(threshold))
            & (confidence["mean"] >= floor * 0.85)
        )
    )
    breakout_filter = forecast_breakout_filter(
        data=data,
        features=features,
        forecast_score=aligned_probability,
        return_forecast=aligned_return_forecast,
        threshold=float(threshold),
        floor=float(floor),
        config=config,
    )
    candidate_mask = (
        candidate_signal.fillna(False)
        & (context_seen > 0)
        & breakout_filter.fillna(False)
    ).to_numpy(dtype=bool)
    segments = merge_boolean_segments(
        boolean_segments(candidate_mask),
        max_gap=max(0, int(config.interval_merge_gap_bars)),
    )
    if not segments:
        return empty_interval_frame()

    rows: list[dict[str, Any]] = []
    train_stats = make_train_feature_stats(train_features)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    atr = features["atr"].reindex(data.index).to_numpy(dtype=float)
    min_bars = max(1, int(config.min_interval_bars))
    max_bars = max(min_bars, int(config.max_interval_bars))

    for raw_start, raw_end in segments:
        for seg_start, seg_end in split_segment(raw_start, raw_end, max_bars=max_bars, min_bars=min_bars):
            segment_probability = aligned_probability.iloc[seg_start : seg_end + 1].dropna()
            if segment_probability.empty:
                continue
            segment_confidence = confidence.iloc[seg_start : seg_end + 1]
            if (
                float(segment_probability.max()) < float(threshold)
                and float(segment_confidence["upper"].max()) < float(threshold)
            ):
                continue

            peak_timestamp = segment_probability.idxmax()
            peak_idx = int(features.index.get_loc(peak_timestamp))
            search_start = max(0, peak_idx - max(config.switch_left_bars, config.switch_box_left_bars))
            search_end = min(len(data) - 1, peak_idx + max(config.switch_right_bars, 1))
            bottom_idx = search_start + int(np.nanargmin(low[search_start : search_end + 1]))
            start_idx = max(0, bottom_idx - int(config.switch_box_left_bars))
            end_idx = min(
                len(data) - 1,
                max(peak_idx, bottom_idx) + int(config.switch_box_right_bars),
            )

            if end_idx - start_idx + 1 < min_bars:
                start_idx = max(0, peak_idx - min_bars // 2)
                end_idx = min(len(data) - 1, start_idx + min_bars - 1)
            if end_idx - start_idx + 1 > max_bars:
                center_idx = int(round((peak_idx + bottom_idx) / 2))
                start_idx = max(0, center_idx - max_bars // 2)
                end_idx = min(len(data) - 1, start_idx + max_bars - 1)

            context_start = max(0, start_idx - config.label_lookback // 2)
            context_slice = prior_context.iloc[context_start : peak_idx + 1].fillna(0.0)
            if context_slice.max() <= 0:
                continue

            interval_probability = aligned_probability.iloc[start_idx : end_idx + 1].dropna()
            if interval_probability.empty:
                continue
            interval_forecast = aligned_return_forecast.iloc[start_idx : end_idx + 1].dropna()
            probability_mean = float(interval_probability.mean())
            probability_max = float(interval_probability.max())
            forecast_mean = float(interval_forecast.mean()) if not interval_forecast.empty else np.nan
            forecast_max = float(interval_forecast.max()) if not interval_forecast.empty else np.nan
            forecast_peak = (
                float(aligned_return_forecast.iloc[peak_idx])
                if np.isfinite(aligned_return_forecast.iloc[peak_idx])
                else np.nan
            )
            interval_confidence_low, interval_confidence_high = probability_interval_bounds(
                interval_probability,
                z=config.confidence_z,
            )
            if probability_max < float(threshold) and interval_confidence_high < float(threshold):
                continue

            support_low = finite_min(low[start_idx : end_idx + 1])
            resistance_high = finite_max(high[start_idx : end_idx + 1])
            if not np.isfinite(support_low) or not np.isfinite(resistance_high):
                continue

            entry_price = float(close[end_idx])
            interval_atr = finite_mean(atr[start_idx : end_idx + 1])
            if not np.isfinite(interval_atr) or interval_atr <= 0:
                interval_atr = max((resistance_high - support_low) * 0.10, abs(entry_price) * 0.001, 1e-12)
            price_pad = max(0.0, float(config.interval_expand_atr)) * interval_atr
            target_price = entry_price + config.target_atr * interval_atr
            stop_price = entry_price - config.stop_atr * interval_atr
            peak_label = labels.loc[peak_timestamp] if peak_timestamp in labels.index else pd.Series(dtype=object)
            hit_idx = peak_label.get("label_hit_bar_idx", np.nan)
            realized_label = peak_label.get("reversal_label", np.nan)
            top_drivers = local_driver_summary(
                features.loc[peak_timestamp, feature_columns],
                feature_importance,
                train_stats,
                top_n=6,
            )

            volume_z_max = finite_max(series_values(features, "volume_z_20", start_idx, end_idx))
            volume_ratio_max = finite_max(series_values(features, "volume_ratio_20", start_idx, end_idx))
            volatility_expansion = finite_mean(
                series_values(features, "volatility_expansion_5_20", start_idx, end_idx)
            )
            volatility_compression = finite_mean(
                series_values(features, "volatility_compression_20_80", start_idx, end_idx)
            )
            range_compression = finite_mean(
                series_values(features, "range_compression_20_80", start_idx, end_idx)
            )
            context_ratio = float(prior_context.iloc[start_idx : end_idx + 1].fillna(0.0).mean())
            interval_score = score_interval(
                probability_mean=probability_mean,
                probability_max=probability_max,
                context_ratio=context_ratio,
                volume_z_max=volume_z_max,
                volatility_expansion=volatility_expansion,
                volatility_compression=volatility_compression,
                confidence_low=interval_confidence_low,
                confidence_high=interval_confidence_high,
            )

            rows.append(
                {
                    "StartBarIdx": int(start_idx),
                    "EndBarIdx": int(end_idx),
                    "PeakBarIdx": int(peak_idx),
                    "ProbabilityPeakBarIdx": int(peak_idx),
                    "TurnBarIdx": int(bottom_idx),
                    "SwitchKind": "valley",
                    "StartDate": data.index[start_idx],
                    "EndDate": data.index[end_idx],
                    "PeakDate": peak_timestamp,
                    "ProbabilityPeakDate": peak_timestamp,
                    "TurnDate": data.index[bottom_idx],
                    "Bars": int(end_idx - start_idx + 1),
                    "SupportLow": support_low,
                    "ResistanceHigh": resistance_high,
                    "BoxLow": support_low - price_pad,
                    "BoxHigh": resistance_high + price_pad,
                    "EntryBarIdx": int(end_idx),
                    "EntryDate": data.index[end_idx],
                    "EntryPrice": entry_price,
                    "TargetPrice": target_price,
                    "StopPrice": stop_price,
                    "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                    "ReturnForecastPctMean": forecast_mean,
                    "ReturnForecastPctMax": forecast_max,
                    "ReturnForecastPctPeak": forecast_peak,
                    "ForecastScoreMean": probability_mean,
                    "ForecastScoreMax": probability_max,
                    "ProbabilityMean": probability_mean,
                    "ProbabilityMax": probability_max,
                    "ProbabilityConfidenceLow": interval_confidence_low,
                    "ProbabilityConfidenceHigh": interval_confidence_high,
                    "SignalThreshold": float(threshold),
                    "VolumeZMax": volume_z_max,
                    "VolumeRatioMax": volume_ratio_max,
                    "VolatilityExpansionMean": volatility_expansion,
                    "VolatilityCompressionMean": volatility_compression,
                    "RangeCompressionMean": range_compression,
                    "PriorDownContextRatio": context_ratio,
                    "IntervalScore": interval_score,
                    "OutcomeKnown": is_finite_number(realized_label),
                    "OutcomeLabel": realized_label,
                    "HitBarIdx": int(hit_idx) if is_finite_number(hit_idx) else np.nan,
                    "HitDate": peak_label.get("label_hit_date", pd.NaT),
                    "HitType": peak_label.get("label_hit_type", None),
                    "ShapeHint": classify_shape_hint(data, peak_idx),
                    "TopDrivers": top_drivers,
                }
            )

    if not rows:
        return empty_interval_frame()
    rows = suppress_overlapping_interval_rows(rows)
    return pd.DataFrame(rows).sort_values("StartBarIdx").reset_index(drop=True)


def forecast_breakout_filter(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    forecast_score: pd.Series,
    return_forecast: pd.Series,
    threshold: float,
    floor: float,
    config: ReversalMLConfig,
) -> pd.Series:
    close = data["close"].astype(float).reindex(features.index)
    high = data["high"].astype(float).reindex(features.index)
    low = data["low"].astype(float).reindex(features.index)
    atr = features["atr"].astype(float).reindex(features.index)
    lookback = max(2, int(config.breakout_lookback))
    cross_lookback = max(1, int(config.forecast_cross_lookback))

    prior_resistance = high.rolling(lookback, min_periods=max(2, lookback // 2)).max().shift(1)
    reclaim_buffer = max(0.0, float(config.breakout_buffer_atr)) * atr
    resistance_reclaim = close >= (prior_resistance - reclaim_buffer)

    reclaim_column = f"reclaim_from_low_atr_{8 if 'reclaim_from_low_atr_8' in features.columns else 5}"
    low_position_column = f"low_range_position_{13 if 'low_range_position_13' in features.columns else 8}"
    valley_rebound = features.get(reclaim_column, pd.Series(np.nan, index=features.index)).astype(float) >= (
        config.switch_min_rebound_atr * 0.75
    )
    near_low = (
        features.get(
            f"dist_to_low_atr_{config.label_lookback}",
            pd.Series(np.nan, index=features.index),
        ).astype(float)
        <= config.max_distance_to_low_atr * 1.25
    ) | (
        features.get(low_position_column, pd.Series(np.nan, index=features.index)).astype(float) <= 0.65
    )

    score_cross = (forecast_score >= threshold) & (forecast_score.shift(cross_lookback) <= floor)
    score_rising = forecast_score.diff(cross_lookback) >= float(config.forecast_min_score_change)
    forecast_turning_positive = (
        (return_forecast > 0)
        | (return_forecast.diff(cross_lookback) > 0)
        | score_cross
    )
    return (
        (resistance_reclaim | valley_rebound | score_cross | score_rising)
        & near_low.fillna(False)
        & forecast_turning_positive.fillna(False)
    )


def empty_signal_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "ReturnForecastPct",
            "ForecastScore",
            "ReversalProbability",
            "SignalThreshold",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "SwitchKind",
            "SwitchTurnBarIdx",
            "SwitchTurnDate",
            "SwitchConfirmBarIdx",
            "SwitchConfirmDate",
            "BarsFromSwitchTurn",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def empty_interval_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "StartBarIdx",
            "EndBarIdx",
            "PeakBarIdx",
            "ProbabilityPeakBarIdx",
            "TurnBarIdx",
            "SwitchKind",
            "StartDate",
            "EndDate",
            "PeakDate",
            "ProbabilityPeakDate",
            "TurnDate",
            "Bars",
            "SupportLow",
            "ResistanceHigh",
            "BoxLow",
            "BoxHigh",
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "ReturnForecastPctMean",
            "ReturnForecastPctMax",
            "ReturnForecastPctPeak",
            "ForecastScoreMean",
            "ForecastScoreMax",
            "ProbabilityMean",
            "ProbabilityMax",
            "ProbabilityConfidenceLow",
            "ProbabilityConfidenceHigh",
            "SignalThreshold",
            "VolumeZMax",
            "VolumeRatioMax",
            "VolatilityExpansionMean",
            "VolatilityCompressionMean",
            "RangeCompressionMean",
            "PriorDownContextRatio",
            "IntervalScore",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def non_max_suppression_1d(
    probability: pd.Series,
    *,
    cooldown_bars: int,
    all_index: pd.Index,
) -> list[pd.Timestamp]:
    ranked = probability.sort_values(ascending=False)
    selected_positions: list[int] = []
    selected_timestamps: list[pd.Timestamp] = []

    for timestamp, _ in ranked.items():
        position = int(all_index.get_loc(timestamp))
        if any(abs(position - selected) <= cooldown_bars for selected in selected_positions):
            continue
        selected_positions.append(position)
        selected_timestamps.append(timestamp)

    selected_timestamps.sort(key=lambda ts: int(all_index.get_loc(ts)))
    return selected_timestamps


def probability_confidence_frame(probability: pd.Series, *, window: int, z: float) -> pd.DataFrame:
    values = probability.astype(float)
    window = max(2, int(window))
    mean = values.rolling(window, min_periods=max(2, window // 4)).mean()
    std = values.rolling(window, min_periods=max(2, window // 4)).std().fillna(0.0)
    count = values.rolling(window, min_periods=max(2, window // 4)).count().replace(0, np.nan)
    stderr = std / np.sqrt(count)
    return pd.DataFrame(
        {
            "mean": mean,
            "lower": (mean - float(z) * stderr).clip(0.0, 1.0),
            "upper": (mean + float(z) * stderr).clip(0.0, 1.0),
        },
        index=probability.index,
    )


def probability_interval_bounds(probability: pd.Series, *, z: float) -> tuple[float, float]:
    values = probability.dropna().to_numpy(dtype=float)
    if len(values) == 0:
        return np.nan, np.nan
    mean = float(np.mean(values))
    if len(values) == 1:
        return mean, mean
    stderr = float(np.std(values, ddof=1) / math.sqrt(len(values)))
    return max(0.0, mean - float(z) * stderr), min(1.0, mean + float(z) * stderr)


def boolean_segments(mask: np.ndarray) -> list[tuple[int, int]]:
    segments: list[tuple[int, int]] = []
    start: int | None = None
    for idx, active in enumerate(mask):
        if active and start is None:
            start = idx
        elif not active and start is not None:
            segments.append((start, idx - 1))
            start = None
    if start is not None:
        segments.append((start, len(mask) - 1))
    return segments


def merge_boolean_segments(
    segments: list[tuple[int, int]],
    *,
    max_gap: int,
) -> list[tuple[int, int]]:
    if not segments:
        return []
    merged: list[tuple[int, int]] = [segments[0]]
    for start, end in segments[1:]:
        previous_start, previous_end = merged[-1]
        if start - previous_end - 1 <= max_gap:
            merged[-1] = (previous_start, end)
        else:
            merged.append((start, end))
    return merged


def split_segment(
    start: int,
    end: int,
    *,
    max_bars: int,
    min_bars: int,
) -> list[tuple[int, int]]:
    length = end - start + 1
    if length <= max_bars:
        return [(start, end)]

    chunks: list[tuple[int, int]] = []
    overlap = max(min_bars, max_bars // 4)
    chunk_start = start
    while chunk_start <= end:
        chunk_end = min(end, chunk_start + max_bars - 1)
        chunks.append((chunk_start, chunk_end))
        if chunk_end >= end:
            break
        chunk_start = max(chunk_start + 1, chunk_end - overlap + 1)
    return chunks


def series_values(features: pd.DataFrame, column: str, start_idx: int, end_idx: int) -> np.ndarray:
    if column not in features.columns:
        return np.array([], dtype=float)
    return features[column].iloc[start_idx : end_idx + 1].to_numpy(dtype=float)


def is_finite_number(value: Any) -> bool:
    try:
        return bool(np.isfinite(float(value)))
    except (TypeError, ValueError):
        return False


def finite_mean(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.mean(array))


def finite_min(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.min(array))


def finite_max(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.max(array))


def score_interval(
    *,
    probability_mean: float,
    probability_max: float,
    context_ratio: float,
    volume_z_max: float,
    volatility_expansion: float,
    volatility_compression: float,
    confidence_low: float,
    confidence_high: float,
) -> float:
    volume_score = np.tanh(max(0.0, 0.0 if not np.isfinite(volume_z_max) else volume_z_max) / 3.0)
    expansion_score = np.tanh(
        max(0.0, (0.0 if not np.isfinite(volatility_expansion) else volatility_expansion) - 1.0)
    )
    compression_score = 1.0 - min(
        1.0,
        max(0.0, 0.0 if not np.isfinite(volatility_compression) else volatility_compression),
    )
    confidence_width = max(
        0.0,
        (0.0 if not np.isfinite(confidence_high) else confidence_high)
        - (0.0 if not np.isfinite(confidence_low) else confidence_low),
    )
    stability_score = 1.0 - min(1.0, confidence_width)
    score = (
        0.38 * probability_max
        + 0.27 * probability_mean
        + 0.15 * min(1.0, max(0.0, context_ratio))
        + 0.07 * volume_score
        + 0.06 * expansion_score
        + 0.04 * compression_score
        + 0.03 * stability_score
    )
    return float(score)


def suppress_overlapping_interval_rows(
    rows: list[dict[str, Any]],
    *,
    max_overlap_ratio: float = 0.55,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []
    for row in sorted(rows, key=lambda item: float(item["IntervalScore"]), reverse=True):
        start = int(row["StartBarIdx"])
        end = int(row["EndBarIdx"])
        length = max(1, end - start + 1)
        overlaps = False
        for existing in selected:
            existing_start = int(existing["StartBarIdx"])
            existing_end = int(existing["EndBarIdx"])
            existing_length = max(1, existing_end - existing_start + 1)
            overlap = max(0, min(end, existing_end) - max(start, existing_start) + 1)
            if overlap / min(length, existing_length) >= max_overlap_ratio:
                overlaps = True
                break
        if not overlaps:
            selected.append(row)
    return selected


def make_train_feature_stats(train_features: pd.DataFrame) -> pd.DataFrame:
    stats = pd.DataFrame(index=train_features.columns)
    stats["median"] = train_features.median(numeric_only=True)
    q25 = train_features.quantile(0.25, numeric_only=True)
    q75 = train_features.quantile(0.75, numeric_only=True)
    stats["iqr"] = (q75 - q25).replace(0, np.nan)
    return stats


def local_driver_summary(
    row: pd.Series,
    feature_importance: pd.DataFrame,
    train_stats: pd.DataFrame,
    *,
    top_n: int,
) -> str:
    if feature_importance.empty:
        return ""

    importance_map = feature_importance.set_index("feature")["importance_mean"].to_dict()
    scored: list[tuple[float, str]] = []
    for feature, value in row.items():
        if feature not in importance_map or feature not in train_stats.index:
            continue
        if not np.isfinite(value):
            continue
        iqr = train_stats.loc[feature, "iqr"]
        median = train_stats.loc[feature, "median"]
        if not np.isfinite(iqr) or iqr == 0:
            continue
        robust_z = (float(value) - float(median)) / float(iqr)
        score = abs(float(importance_map[feature]) * robust_z)
        if score <= 0:
            continue
        scored.append((score, f"{feature}={float(value):.4g} z={robust_z:.2f}"))

    scored.sort(reverse=True)
    return " | ".join(text for _, text in scored[:top_n])


def classify_shape_hint(data: pd.DataFrame, bar_idx: int, *, lookback: int = 80) -> str:
    start = max(0, int(bar_idx) - lookback)
    if int(bar_idx) - start < 20:
        return "ML Bullish Reversal"

    window = data.iloc[start : int(bar_idx) + 1]
    lows = window["low"].to_numpy(dtype=float)
    highs = window["high"].to_numpy(dtype=float)
    close = window["close"].to_numpy(dtype=float)
    atr = float(average_true_range(window, period=14).iloc[-1])
    if not np.isfinite(atr) or atr <= 0:
        atr = max(float(np.nanstd(close)), 1e-12)

    pivot_lows = simple_pivots(lows, mode="low", order=3)
    pivot_highs = simple_pivots(highs, mode="high", order=3)
    if len(pivot_lows) >= 2:
        l1, l2 = pivot_lows[-2], pivot_lows[-1]
        low_diff_atr = abs(lows[l2] - lows[l1]) / atr
        if l2 > l1 and low_diff_atr <= 1.0:
            between_high = np.max(highs[l1:l2 + 1])
            if between_high - max(lows[l1], lows[l2]) >= atr:
                return "ML Double Bottom-like"

    if len(pivot_lows) >= 3 and len(pivot_highs) >= 2:
        left, head, right = pivot_lows[-3], pivot_lows[-2], pivot_lows[-1]
        if left < head < right and lows[head] < min(lows[left], lows[right]):
            shoulder_diff_atr = abs(lows[left] - lows[right]) / atr
            if shoulder_diff_atr <= 1.5:
                return "ML Inverse H&S-like"

    return "ML Bullish Reversal"


def simple_pivots(values: np.ndarray, *, mode: str, order: int) -> list[int]:
    pivots: list[int] = []
    for idx in range(order, len(values) - order):
        left = values[idx - order : idx]
        right = values[idx + 1 : idx + order + 1]
        if mode == "low" and values[idx] <= np.min(left) and values[idx] <= np.min(right):
            pivots.append(idx)
        if mode == "high" and values[idx] >= np.max(left) and values[idx] >= np.max(right):
            pivots.append(idx)
    return pivots


def fit_winsorize(
    X_train: pd.DataFrame,
    *,
    lower: float = 0.01,
    upper: float = 0.99,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    bounds = pd.DataFrame(index=X_train.columns)
    bounds["lower"] = X_train.quantile(lower, numeric_only=True)
    bounds["upper"] = X_train.quantile(upper, numeric_only=True)
    return apply_winsorize(X_train, bounds), bounds


def apply_winsorize(X: pd.DataFrame, bounds: pd.DataFrame) -> pd.DataFrame:
    out = X.copy()
    for column in out.columns:
        if column not in bounds.index:
            continue
        lower = bounds.loc[column, "lower"]
        upper = bounds.loc[column, "upper"]
        if np.isfinite(lower) and np.isfinite(upper) and lower < upper:
            out[column] = out[column].clip(lower=lower, upper=upper)
    return out


def average_true_range(data: pd.DataFrame, *, period: int = 14) -> pd.Series:
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    close = data["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [
            high - low,
            (high - previous_close).abs(),
            (low - previous_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    return true_range.rolling(period, min_periods=1).mean()


def rsi(close: pd.Series, *, period: int = 14) -> pd.Series:
    delta = close.diff()
    gains = delta.clip(lower=0.0)
    losses = -delta.clip(upper=0.0)
    avg_gain = gains.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    avg_loss = losses.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    rs = avg_gain / avg_loss.replace(0, np.nan)
    return 100.0 - (100.0 / (1.0 + rs))


def macd(close: pd.Series) -> tuple[pd.Series, pd.Series, pd.Series]:
    fast = close.ewm(span=12, adjust=False).mean()
    slow = close.ewm(span=26, adjust=False).mean()
    line = fast - slow
    signal = line.ewm(span=9, adjust=False).mean()
    return line, signal, line - signal


def stochastic_k(
    high: pd.Series,
    low: pd.Series,
    close: pd.Series,
    *,
    period: int,
) -> pd.Series:
    lowest = low.rolling(period).min()
    highest = high.rolling(period).max()
    return (close - lowest) / (highest - lowest).replace(0, np.nan) * 100.0


def rolling_zscore(series: pd.Series, *, window: int) -> pd.Series:
    mean = series.rolling(window).mean()
    std = series.rolling(window).std()
    return (series - mean) / std.replace(0, np.nan)


def rolling_slope_pct(series: pd.Series, *, window: int) -> pd.Series:
    x = np.arange(window, dtype=float)
    x = x - x.mean()

    def slope(values: np.ndarray) -> float:
        if np.isnan(values).any():
            return np.nan
        y = values.astype(float)
        y = y - y.mean()
        denom = float(np.dot(x, x))
        if denom == 0:
            return np.nan
        raw_slope = float(np.dot(x, y) / denom)
        return raw_slope / max(abs(float(values[-1])), 1e-12) * 100.0

    return series.rolling(window).apply(slope, raw=True)


def signed_streak(condition: pd.Series) -> pd.Series:
    values = condition.fillna(False).to_numpy(dtype=bool)
    streaks = np.zeros(len(values), dtype=float)
    current = 0
    previous = None
    for idx, value in enumerate(values):
        if previous is None or value != previous:
            current = 1
        else:
            current += 1
        streaks[idx] = current if value else -current
        previous = value
    return pd.Series(streaks, index=condition.index)


def on_balance_volume(close: pd.Series, volume: pd.Series) -> pd.Series:
    direction = np.sign(close.diff()).fillna(0.0)
    return (direction * volume).cumsum()


def walk_forward_arima_features(
    returns: pd.Series,
    *,
    order: tuple[int, int, int],
    window: int,
) -> pd.DataFrame:
    columns = ["arima_forecast_return", "arima_residual", "arima_residual_z_40"]
    out = pd.DataFrame(np.nan, index=returns.index, columns=columns)
    if ARIMA is None:
        warnings.warn("statsmodels unavailable; ARIMA features skipped.")
        return out

    values = returns.astype(float)
    for idx in range(window, len(values)):
        history = values.iloc[idx - window : idx].dropna()
        if len(history) < max(30, sum(order) * 10):
            continue
        try:
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                result = ARIMA(history, order=order).fit()
            forecast = float(result.forecast(1).iloc[0])
            residual = float(values.iloc[idx] - forecast)
            out.iloc[idx, out.columns.get_loc("arima_forecast_return")] = forecast
            out.iloc[idx, out.columns.get_loc("arima_residual")] = residual
        except Exception:
            continue
    out["arima_residual_z_40"] = rolling_zscore(out["arima_residual"], window=40)
    return out


def save_outputs(
    data: pd.DataFrame,
    result: dict[str, Any],
    config: ReversalMLConfig,
) -> Path:
    run_name = f"{config.symbol.replace('.', '_')}_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
    run_dir = Path(config.output_dir) / run_name
    run_dir.mkdir(parents=True, exist_ok=True)
    if go is None:
        warnings.warn("plotly unavailable; HTML plot files skipped. Install with: pip install plotly")

    result["signals"].to_csv(run_dir / "signals.csv", index=False)
    result["intervals"].to_csv(run_dir / "pattern_intervals.csv", index=False)
    result["model_report"].to_csv(run_dir / "model_report.csv", index=False)
    result["feature_importance"].to_csv(run_dir / "feature_importance.csv", index=False)
    result["hmm_state_summary"].to_csv(run_dir / "hmm_state_summary.csv", index=False)
    result["pca_loadings"].to_csv(run_dir / "pca_loadings.csv", index=False)

    probability_frame = pd.DataFrame(
        {
            "Date": result["probabilities"].index,
            "ForecastScore": result["probabilities"].to_numpy(dtype=float),
            "ReversalProbability": result["probabilities"].to_numpy(dtype=float),
            "ReturnForecastPct": result["forecasts"].reindex(result["probabilities"].index).to_numpy(dtype=float),
        }
    )
    probability_frame.to_csv(run_dir / "probabilities.csv", index=False)
    probability_frame.to_csv(run_dir / "return_forecasts.csv", index=False)

    all_probability_frame = pd.DataFrame({"Date": data.index})
    for model_name, probabilities in result["per_model_probabilities"].items():
        all_probability_frame[f"{model_name}_forecast_score"] = probabilities.reindex(data.index).to_numpy(dtype=float)
        all_probability_frame[f"{model_name}_return_forecast_pct"] = (
            result["per_model_forecasts"][model_name].reindex(data.index).to_numpy(dtype=float)
        )
    all_probability_frame.to_csv(run_dir / "all_model_probabilities.csv", index=False)
    all_probability_frame.to_csv(run_dir / "all_model_return_forecasts.csv", index=False)

    hmm_regimes = make_hmm_regime_frame(result["features"])
    hmm_regimes.to_csv(run_dir / "hmm_regimes.csv", index=False)

    all_interval_frames = [
        intervals
        for intervals in result["per_model_intervals"].values()
        if intervals is not None and not intervals.empty
    ]
    if all_interval_frames:
        pd.concat(all_interval_frames, ignore_index=True).to_csv(
            run_dir / "all_model_pattern_intervals.csv",
            index=False,
        )
    else:
        empty_interval_frame().to_csv(run_dir / "all_model_pattern_intervals.csv", index=False)

    models_dir = run_dir / "models"
    models_dir.mkdir(exist_ok=True)
    for model_name, probabilities in result["per_model_probabilities"].items():
        safe_model_name = safe_filename(model_name)
        forecasts = result["per_model_forecasts"][model_name]
        signals = result["per_model_signals"].get(model_name, empty_signal_frame())
        importance = result["per_model_feature_importance"].get(
            model_name,
            pd.DataFrame(columns=["feature", "importance_mean", "importance_std", "rank"]),
        )
        threshold = result["per_model_thresholds"].get(model_name, np.nan)
        intervals = result["per_model_intervals"].get(model_name, empty_interval_frame())
        signals.to_csv(models_dir / f"{safe_model_name}_signals.csv", index=False)
        intervals.to_csv(models_dir / f"{safe_model_name}_pattern_intervals.csv", index=False)
        importance.to_csv(models_dir / f"{safe_model_name}_feature_importance.csv", index=False)
        pd.DataFrame(
            {
                "Date": data.index,
                "ForecastScore": probabilities.reindex(data.index).to_numpy(dtype=float),
                "ReturnForecastPct": forecasts.reindex(data.index).to_numpy(dtype=float),
            }
        ).to_csv(models_dir / f"{safe_model_name}_return_forecast.csv", index=False)
        make_plot(
            data=data,
            probabilities=probabilities,
            return_forecasts=forecasts,
            signals=signals,
            intervals=intervals,
            title=f"{config.symbol} {model_name} ML Reversal Signals",
            output_path=models_dir / f"{safe_model_name}_plot.html",
            model_name=model_name,
            threshold=threshold,
            confidence_window=config.confidence_window,
            confidence_z=config.confidence_z,
        )
        make_feature_importance_plot(
            importance=importance,
            title=f"{config.symbol} {model_name} XAI Feature Importance",
            output_path=models_dir / f"{safe_model_name}_feature_importance.html",
        )

    with open(run_dir / "config.json", "w", encoding="utf-8") as handle:
        json.dump(
            {
                **asdict(config),
                "best_model_name": result["best_model_name"],
                "threshold": result["threshold"],
                "return_threshold_pct": result["return_threshold_pct"],
                "forecast_score_scale": result["forecast_score_scale"],
            },
            handle,
            indent=2,
            default=str,
        )

    plot_path = run_dir / "reversal_ml_plot.html"
    make_plot(
        data=data,
        probabilities=result["probabilities"],
        return_forecasts=result["forecasts"],
        signals=result["signals"],
        intervals=result["intervals"],
        title=f"{config.symbol} ML Reversal Signals",
        output_path=plot_path,
        model_name=f"best: {result['best_model_name']}",
        threshold=result["threshold"],
        confidence_window=config.confidence_window,
        confidence_z=config.confidence_z,
    )
    make_model_comparison_plot(
        data=data,
        per_model_probabilities=result["per_model_probabilities"],
        per_model_forecasts=result["per_model_forecasts"],
        per_model_signals=result["per_model_signals"],
        per_model_intervals=result["per_model_intervals"],
        per_model_thresholds=result["per_model_thresholds"],
        title=f"{config.symbol} ML Reversal Model Comparison",
        output_path=run_dir / "all_models_comparison_plot.html",
    )
    make_hmm_regime_plot(
        data=data,
        features=result["features"],
        state_summary=result["hmm_state_summary"],
        title=f"{config.symbol} HMM Regime Map",
        output_path=run_dir / "hmm_regime_plot.html",
    )
    return run_dir


def safe_filename(value: str) -> str:
    safe = "".join(character if character.isalnum() or character in {"-", "_"} else "_" for character in value)
    return safe.strip("_") or "model"


def hex_to_rgba(hex_color: str, alpha: float) -> str:
    value = hex_color.lstrip("#")
    if len(value) != 6:
        return f"rgba(56, 189, 248, {alpha})"
    red, green, blue = (int(value[idx : idx + 2], 16) for idx in (0, 2, 4))
    return f"rgba({red}, {green}, {blue}, {alpha})"


def add_interval_boxes(
    fig: Any,
    intervals: pd.DataFrame | None,
    *,
    color: str,
    label: str,
    opacity: float,
    show_text: bool,
) -> None:
    if intervals is None or intervals.empty:
        return

    center_x: list[float] = []
    center_y: list[float] = []
    hover_text: list[str] = []
    visible_text: list[str] = []
    for interval_number, (_, row) in enumerate(intervals.iterrows(), start=1):
        try:
            start_idx = int(row["StartBarIdx"])
            end_idx = int(row["EndBarIdx"])
            box_low = float(row["BoxLow"])
            box_high = float(row["BoxHigh"])
        except (KeyError, TypeError, ValueError):
            continue
        if not all(np.isfinite(value) for value in [start_idx, end_idx, box_low, box_high]):
            continue

        fig.add_shape(
            type="rect",
            xref="x",
            yref="y",
            x0=start_idx - 0.5,
            x1=end_idx + 0.5,
            y0=box_low,
            y1=box_high,
            fillcolor=hex_to_rgba(color, opacity),
            line=dict(color=color, width=2),
            layer="below",
        )
        center_x.append((start_idx + end_idx) / 2)
        center_y.append(box_high)
        probability_max = float(row.get("ProbabilityMax", np.nan))
        probability_mean = float(row.get("ProbabilityMean", np.nan))
        visible_text.append(f"Box {interval_number} {probability_max:.2f}")
        hover_text.append(
            (
                f"{label} interval<br>"
                f"Start: {row.get('StartDate')}<br>"
                f"End: {row.get('EndDate')}<br>"
                f"Turn: {row.get('TurnDate')}<br>"
                f"Probability peak: {row.get('ProbabilityPeakDate', row.get('PeakDate'))}<br>"
                f"Switch kind: {row.get('SwitchKind')}<br>"
                f"Return forecast peak: {float(row.get('ReturnForecastPctPeak', np.nan)):.3f}%<br>"
                f"Return forecast mean: {float(row.get('ReturnForecastPctMean', np.nan)):.3f}%<br>"
                f"Mean forecast score: {probability_mean:.3f}<br>"
                f"Max forecast score: {probability_max:.3f}<br>"
                f"Score CI: {float(row.get('ProbabilityConfidenceLow', np.nan)):.3f}"
                f" - {float(row.get('ProbabilityConfidenceHigh', np.nan)):.3f}<br>"
                f"Volume z max: {float(row.get('VolumeZMax', np.nan)):.3f}<br>"
                f"Vol expansion: {float(row.get('VolatilityExpansionMean', np.nan)):.3f}<br>"
                f"Score: {float(row.get('IntervalScore', np.nan)):.3f}<br>"
                f"Shape: {row.get('ShapeHint')}<br>"
                f"Drivers: {row.get('TopDrivers')}"
            )
        )

    if not center_x:
        return
    mode = "markers+text" if show_text else "markers"
    fig.add_trace(
        go.Scatter(
            x=center_x,
            y=center_y,
            mode=mode,
            name=f"{label} pattern interval",
            marker=dict(
                color=color,
                size=11,
                symbol="square-open",
                line=dict(width=2),
            ),
            text=visible_text if show_text else None,
            textposition="top center",
            hovertext=hover_text,
            hoverinfo="text",
        )
    )


def make_plot(
    *,
    data: pd.DataFrame,
    probabilities: pd.Series,
    return_forecasts: pd.Series | None,
    signals: pd.DataFrame,
    intervals: pd.DataFrame | None = None,
    title: str,
    output_path: Path,
    model_name: str | None = None,
    threshold: float | None = None,
    confidence_window: int = 20,
    confidence_z: float = 1.0,
) -> None:
    if go is None:
        return

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )
    add_interval_boxes(
        fig,
        intervals,
        color="#FACC15",
        label=model_name or "ML",
        opacity=0.16,
        show_text=True,
    )
    aligned_probability = probabilities.reindex(data.index).astype(float)
    probability_ci = probability_confidence_frame(
        aligned_probability,
        window=confidence_window,
        z=confidence_z,
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probability_ci["lower"].to_numpy(dtype=float),
            name="Forecast score CI lower",
            mode="lines",
            yaxis="y2",
            line=dict(width=0),
            hoverinfo="skip",
            showlegend=False,
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probability_ci["upper"].to_numpy(dtype=float),
            name="Forecast score confidence band",
            mode="lines",
            yaxis="y2",
            fill="tonexty",
            fillcolor="rgba(56, 189, 248, 0.14)",
            line=dict(width=0),
            hoverinfo="skip",
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=aligned_probability.to_numpy(dtype=float),
            name=f"{model_name or 'Model'} forecast score",
            mode="lines",
            yaxis="y2",
            line=dict(color="#38BDF8", width=2),
        )
    )
    if return_forecasts is not None:
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=return_forecasts.reindex(data.index).to_numpy(dtype=float),
                name=f"{model_name or 'Model'} return forecast %",
                mode="lines",
                yaxis="y3",
                line=dict(color="#F97316", width=2),
            )
        )
    if threshold is not None and np.isfinite(float(threshold)):
        fig.add_trace(
            go.Scatter(
                x=[plot_x[0], plot_x[-1]],
                y=[float(threshold), float(threshold)],
                name="Signal threshold",
                mode="lines",
                yaxis="y2",
                line=dict(color="#FACC15", width=1, dash="dash"),
                hoverinfo="skip",
            )
        )

    if not signals.empty:
        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers+text",
                name=f"{model_name or 'ML'} Forecast Breakout",
                marker=dict(color="#FACC15", size=12, symbol="circle"),
                text=[f"{prob:.2f}" for prob in signals["ReversalProbability"]],
                textposition="top center",
                hovertext=[
                    (
                        f"ML Buy<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Forecast score: {row['ForecastScore']:.3f}<br>"
                        f"Return forecast: {row['ReturnForecastPct']:.3f}%<br>"
                        f"Switch: {row.get('SwitchKind')} at {row.get('SwitchTurnDate')}<br>"
                        f"Bars from turn: {row.get('BarsFromSwitchTurn')}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Forecast Score",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        yaxis3=dict(
            title="Return Forecast %",
            overlaying="y",
            side="right",
            anchor="free",
            position=0.98,
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=110, t=80, b=55),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def make_feature_importance_plot(
    *,
    importance: pd.DataFrame,
    title: str,
    output_path: Path,
    top_n: int = 25,
) -> None:
    if go is None or importance.empty:
        return

    plot_frame = importance.head(top_n).copy()
    plot_frame = plot_frame.sort_values("importance_mean", ascending=True)
    fig = go.Figure()
    fig.add_trace(
        go.Bar(
            x=plot_frame["importance_mean"],
            y=plot_frame["feature"],
            orientation="h",
            error_x=dict(
                type="data",
                array=plot_frame["importance_std"].clip(lower=0).to_numpy(dtype=float),
                visible=True,
                color="#94A3B8",
            ),
            marker=dict(color="#38BDF8"),
            hovertext=[
                (
                    f"Feature: {row['feature']}<br>"
                    f"Importance: {row['importance_mean']:.6f}<br>"
                    f"Std: {row['importance_std']:.6f}<br>"
                    f"Rank: {int(row['rank'])}"
                )
                for _, row in plot_frame.iterrows()
            ],
            hoverinfo="text",
            name="Permutation importance",
        )
    )
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=max(520, 24 * len(plot_frame) + 180),
        xaxis=dict(title="Neg-MAE permutation importance"),
        yaxis=dict(title="Feature"),
        margin=dict(l=180, r=50, t=80, b=60),
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def make_hmm_regime_frame(features: pd.DataFrame) -> pd.DataFrame:
    hmm_columns = [
        column
        for column in features.columns
        if column == "hmm_state" or column.startswith("hmm_state_prob_")
    ]
    if not hmm_columns:
        return pd.DataFrame(columns=["Date", "hmm_state"])
    frame = features[hmm_columns].copy()
    frame.insert(0, "Date", features.index)
    return frame


def make_hmm_regime_plot(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    state_summary: pd.DataFrame,
    title: str,
    output_path: Path,
) -> None:
    if go is None or "hmm_state" not in features.columns:
        return

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    states = features["hmm_state"].reindex(data.index)
    state_values = states.to_numpy(dtype=float)
    finite_states = state_values[np.isfinite(state_values)]
    if len(finite_states) == 0:
        return

    colors = ["#38BDF8", "#FACC15", "#A78BFA", "#34D399", "#FB7185", "#F97316", "#22D3EE"]
    fig = go.Figure()
    for start_idx, end_idx, state in hmm_state_segments(state_values):
        color = colors[int(state) % len(colors)]
        fig.add_shape(
            type="rect",
            xref="x",
            yref="paper",
            x0=start_idx - 0.5,
            x1=end_idx + 0.5,
            y0=0,
            y1=1,
            fillcolor=hex_to_rgba(color, 0.10),
            line=dict(width=0),
            layer="below",
        )

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=state_values,
            name="HMM state",
            mode="lines",
            yaxis="y2",
            line=dict(color="#FACC15", width=2, shape="hv"),
        )
    )

    probability_columns = [
        column for column in features.columns if column.startswith("hmm_state_prob_")
    ]
    if probability_columns:
        max_probability = features[probability_columns].max(axis=1).reindex(data.index)
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=max_probability.to_numpy(dtype=float),
                name="HMM max probability",
                mode="lines",
                yaxis="y3",
                line=dict(color="#38BDF8", width=2),
            )
        )

    if state_summary is not None and not state_summary.empty:
        annotations = []
        summary = state_summary.copy()
        for _, row in summary.iterrows():
            annotations.append(
                f"S{int(row['state'])}: n={int(row['train_count'])}, "
                f"ret={float(row['mean_return']):.4g}, "
                f"vol={float(row['mean_vol']):.4g}, "
                f"dd={float(row['mean_drawdown']):.3g}"
            )
        fig.add_annotation(
            text="<br>".join(annotations),
            xref="paper",
            yref="paper",
            x=0.01,
            y=0.98,
            align="left",
            showarrow=False,
            bgcolor="rgba(15, 23, 42, 0.70)",
            bordercolor="#334155",
            font=dict(size=11),
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="HMM State",
            overlaying="y",
            side="right",
            showgrid=False,
        ),
        yaxis3=dict(
            title="State Confidence",
            overlaying="y",
            side="right",
            anchor="free",
            position=0.98,
            range=[0, 1],
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=110, t=90, b=60),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def hmm_state_segments(state_values: np.ndarray) -> list[tuple[int, int, int]]:
    segments: list[tuple[int, int, int]] = []
    start_idx: int | None = None
    active_state: int | None = None
    for idx, state in enumerate(state_values):
        if not np.isfinite(state):
            if start_idx is not None and active_state is not None:
                segments.append((start_idx, idx - 1, active_state))
            start_idx = None
            active_state = None
            continue
        state_int = int(state)
        if active_state is None:
            start_idx = idx
            active_state = state_int
            continue
        if state_int != active_state:
            segments.append((int(start_idx), idx - 1, active_state))
            start_idx = idx
            active_state = state_int
    if start_idx is not None and active_state is not None:
        segments.append((start_idx, len(state_values) - 1, active_state))
    return segments


def make_model_comparison_plot(
    *,
    data: pd.DataFrame,
    per_model_probabilities: dict[str, pd.Series],
    per_model_forecasts: dict[str, pd.Series],
    per_model_signals: dict[str, pd.DataFrame],
    per_model_intervals: dict[str, pd.DataFrame],
    per_model_thresholds: dict[str, float],
    title: str,
    output_path: Path,
) -> None:
    if go is None or not per_model_probabilities:
        return

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    colors = ["#38BDF8", "#FACC15", "#A78BFA", "#34D399", "#FB7185", "#F97316"]

    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    for model_idx, (model_name, probabilities) in enumerate(per_model_probabilities.items()):
        color = colors[model_idx % len(colors)]
        threshold = per_model_thresholds.get(model_name, np.nan)
        add_interval_boxes(
            fig,
            per_model_intervals.get(model_name, empty_interval_frame()),
            color=color,
            label=model_name,
            opacity=0.08,
            show_text=False,
        )
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=probabilities.reindex(data.index).to_numpy(dtype=float),
                name=f"{model_name} forecast score",
                mode="lines",
                yaxis="y2",
                line=dict(color=color, width=2),
            )
        )
        if model_name in per_model_forecasts:
            fig.add_trace(
                go.Scatter(
                    x=plot_x,
                    y=per_model_forecasts[model_name].reindex(data.index).to_numpy(dtype=float),
                    name=f"{model_name} return forecast %",
                    mode="lines",
                    yaxis="y3",
                    line=dict(color=color, width=1, dash="dash"),
                    opacity=0.75,
                )
            )
        if threshold is not None and np.isfinite(float(threshold)):
            fig.add_trace(
                go.Scatter(
                    x=[plot_x[0], plot_x[-1]],
                    y=[float(threshold), float(threshold)],
                    name=f"{model_name} threshold",
                    mode="lines",
                    yaxis="y2",
                    line=dict(color=color, width=1, dash="dot"),
                    hoverinfo="skip",
                    showlegend=False,
                )
            )

        signals = per_model_signals.get(model_name, empty_signal_frame())
        if signals is None or signals.empty:
            continue

        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers",
                name=f"{model_name} signals",
                marker=dict(
                    color=color,
                    size=10,
                    symbol="circle-open",
                    line=dict(width=2),
                ),
                hovertext=[
                    (
                        f"Model: {model_name}<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Forecast score: {row['ForecastScore']:.3f}<br>"
                        f"Return forecast: {row['ReturnForecastPct']:.3f}%<br>"
                        f"Threshold: {row['SignalThreshold']:.3f}<br>"
                        f"Switch: {row.get('SwitchKind')} at {row.get('SwitchTurnDate')}<br>"
                        f"Bars from turn: {row.get('BarsFromSwitchTurn')}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=860,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Forecast Score",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        yaxis3=dict(
            title="Return Forecast %",
            overlaying="y",
            side="right",
            anchor="free",
            position=0.98,
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=110, t=90, b=60),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    fig.write_html(str(output_path), include_plotlyjs="cdn")


def run(config: ReversalMLConfig) -> Path:
    analysis = run_analysis(config, save=True)
    print_run_summary(analysis)
    return analysis["run_dir"]


def run_analysis(config: ReversalMLConfig, *, save: bool = True) -> dict[str, Any]:
    data = load_ohlcv(config)
    features = make_feature_frame(data, config)
    labels = make_reversal_labels(data, features, config)
    result = train_reversal_models(data, features, labels, config)
    run_dir = save_outputs(data, result, config) if save else None
    return {
        "config": config,
        "data": data,
        "features": features,
        "labels": labels,
        "result": result,
        "run_dir": run_dir,
        "csv_paths": collect_csv_paths(run_dir, result) if run_dir is not None else {},
        "plot_paths": collect_plot_paths(run_dir, result) if run_dir is not None else {},
    }


def run_notebook(
    *,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    csv_path: str | None = None,
    output_dir: str = "outputs/ml_reversal_runs",
    show_plots: bool = True,
    print_outputs: bool = True,
    save: bool = True,
    **overrides: Any,
) -> dict[str, Any]:
    config_values = {
        "symbol": symbol,
        "period": period,
        "interval": interval,
        "csv_path": csv_path,
        "output_dir": output_dir,
    }
    config_values.update(overrides)
    config = ReversalMLConfig(**config_values)
    analysis = run_analysis(config, save=save)
    if print_outputs:
        print_run_summary(analysis)
    if show_plots and analysis["plot_paths"]:
        display_notebook_plot_links(analysis["plot_paths"])
    return analysis


def collect_csv_paths(run_dir: Path | None, result: dict[str, Any]) -> dict[str, Path]:
    if run_dir is None:
        return {}
    paths: dict[str, Path] = {
        "signals": run_dir / "signals.csv",
        "pattern_intervals": run_dir / "pattern_intervals.csv",
        "model_report": run_dir / "model_report.csv",
        "feature_importance": run_dir / "feature_importance.csv",
        "hmm_state_summary": run_dir / "hmm_state_summary.csv",
        "hmm_regimes": run_dir / "hmm_regimes.csv",
        "pca_loadings": run_dir / "pca_loadings.csv",
        "probabilities": run_dir / "probabilities.csv",
        "return_forecasts": run_dir / "return_forecasts.csv",
        "all_model_probabilities": run_dir / "all_model_probabilities.csv",
        "all_model_return_forecasts": run_dir / "all_model_return_forecasts.csv",
        "all_model_pattern_intervals": run_dir / "all_model_pattern_intervals.csv",
    }
    models_dir = run_dir / "models"
    for model_name in result.get("per_model_probabilities", {}):
        safe_model_name = safe_filename(str(model_name))
        paths[f"{model_name}_signals"] = models_dir / f"{safe_model_name}_signals.csv"
        paths[f"{model_name}_pattern_intervals"] = (
            models_dir / f"{safe_model_name}_pattern_intervals.csv"
        )
        paths[f"{model_name}_feature_importance"] = (
            models_dir / f"{safe_model_name}_feature_importance.csv"
        )
        paths[f"{model_name}_return_forecast"] = models_dir / f"{safe_model_name}_return_forecast.csv"
    return paths


def collect_plot_paths(run_dir: Path | None, result: dict[str, Any]) -> dict[str, Path]:
    if run_dir is None:
        return {}
    paths: dict[str, Path] = {
        "best_model": run_dir / "reversal_ml_plot.html",
        "model_comparison": run_dir / "all_models_comparison_plot.html",
        "hmm_regime": run_dir / "hmm_regime_plot.html",
    }
    models_dir = run_dir / "models"
    for model_name in result.get("per_model_probabilities", {}):
        safe_model_name = safe_filename(str(model_name))
        paths[f"{model_name}_plot"] = models_dir / f"{safe_model_name}_plot.html"
        paths[f"{model_name}_feature_importance"] = (
            models_dir / f"{safe_model_name}_feature_importance.html"
        )
    return paths


def print_run_summary(analysis: dict[str, Any]) -> None:
    config = analysis["config"]
    result = analysis["result"]
    labels = analysis.get("labels")
    run_dir = analysis.get("run_dir")

    print("ML reversal run")
    print(f"Run dir: {run_dir if run_dir is not None else 'not saved'}")
    print(
        f"Symbol: {config.symbol} | Period: {config.period} | Interval: {config.interval} | "
        f"HMM: {'on' if config.use_hmm else 'off'}"
    )
    print(
        f"Best model: {result['best_model_name']} | "
        f"Return threshold: {result['return_threshold_pct']:.4f}% | "
        f"Forecast score threshold: {result['threshold']:.4f}"
    )
    print(f"Pattern intervals: {len(result['intervals'])} | Point signals: {len(result['signals'])}")
    if isinstance(labels, pd.DataFrame) and "reversal_label" in labels.columns:
        positive_labels = int(labels["reversal_label"].fillna(0).sum())
        labeled_rows = int(labels["reversal_label"].notna().sum())
        print(f"Switch labels: {positive_labels} positive / {labeled_rows} labeled rows")

    if not result["model_report"].empty:
        print("\nModel report:")
        print(result["model_report"].to_string(index=False))

    if not result["intervals"].empty:
        preview_columns = [
            column
            for column in [
                "StartDate",
                "EndDate",
                "TurnDate",
                "ProbabilityPeakDate",
                "SwitchKind",
                "Bars",
                "ReturnForecastPctPeak",
                "ReturnForecastPctMean",
                "ForecastScoreMax",
                "ForecastScoreMean",
                "IntervalScore",
                "BoxLow",
                "BoxHigh",
                "ShapeHint",
            ]
            if column in result["intervals"].columns
        ]
        print("\nTop pattern intervals:")
        print(
            result["intervals"]
            .sort_values("IntervalScore", ascending=False)
            .head(10)[preview_columns]
            .to_string(index=False)
        )

    csv_paths = analysis.get("csv_paths", {})
    if csv_paths:
        print("\nCSV files:")
        for name, path in csv_paths.items():
            status = "created" if path.exists() else "expected"
            print(f"  {name}: {path} [{status}]")

    plot_paths = analysis.get("plot_paths", {})
    if plot_paths:
        print("\nPlot files:")
        if go is None:
            print("  plotly is not installed, so HTML plot files were skipped.")
        for name, path in plot_paths.items():
            status = "created" if path.exists() else "not created"
            print(f"  {name}: {path} [{status}]")


def display_notebook_plot_links(plot_paths: dict[str, Path]) -> None:
    try:
        from IPython.display import HTML, display
    except ModuleNotFoundError:
        warnings.warn("IPython unavailable; returning plot paths without notebook display.")
        return
    links = "".join(
        f'<li><a href="{path}" target="_blank">{name}</a></li>'
        for name, path in plot_paths.items()
        if path.exists()
    )
    if links:
        display(HTML(f"<b>ML reversal plots</b><ul>{links}</ul>"))


if __name__ == "__main__":
    run_notebook()
```

```python
"""
Notebook usage:

from outputs.ml_reversal_pattern_detector import run_notebook

analysis = run_notebook(
    symbol="THYAO.IS",
    period="60d",
    interval="5m",
    show_plots=True,
)

analysis["result"]["intervals"]
analysis["result"]["forecasts"]
analysis["csv_paths"]
analysis["plot_paths"]
analysis["figures"]
"""

from __future__ import annotations

import json
import math
import warnings
from dataclasses import asdict, dataclass
from datetime import datetime
from pathlib import Path
from typing import Any

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    from sklearn.base import clone
    from sklearn.decomposition import PCA
    from sklearn.ensemble import HistGradientBoostingRegressor, RandomForestRegressor
    from sklearn.impute import SimpleImputer
    from sklearn.inspection import permutation_importance
    from sklearn.linear_model import Ridge
    from sklearn.metrics import (
        mean_absolute_error,
        mean_squared_error,
        precision_recall_curve,
        r2_score,
    )
    from sklearn.model_selection import TimeSeriesSplit
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
except ModuleNotFoundError:
    clone = None
    PCA = None
    HistGradientBoostingRegressor = None
    RandomForestRegressor = None
    SimpleImputer = None
    permutation_importance = None
    Ridge = None
    TimeSeriesSplit = None
    Pipeline = None
    StandardScaler = None
    mean_absolute_error = None
    mean_squared_error = None
    precision_recall_curve = None
    r2_score = None

try:
    from hmmlearn.hmm import GaussianHMM
except ModuleNotFoundError:
    GaussianHMM = None

try:
    from statsmodels.tsa.arima.model import ARIMA
except ModuleNotFoundError:
    ARIMA = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


@dataclass
class ReversalMLConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    csv_path: str | None = None
    output_dir: str = "outputs/ml_reversal_runs"
    random_state: int = 42

    label_lookback: int = 80
    label_horizon: int = 40
    min_prior_drawdown_pct: float = 2.0
    max_distance_to_low_atr: float = 2.5
    target_atr: float = 2.0
    stop_atr: float = 1.2
    forecast_horizon: int | None = None
    forecast_target_column: str = "forecast_forward_return_pct"
    forecast_return_threshold_pct: float | None = None
    forecast_score_threshold: float | None = None
    forecast_scale_floor_pct: float = 0.15
    breakout_lookback: int = 8
    breakout_buffer_atr: float = 0.05
    forecast_cross_lookback: int = 3
    forecast_min_score_change: float = 0.08
    switch_left_bars: int = 12
    switch_right_bars: int = 5
    switch_confirmation_bars: int = 8
    switch_label_window_bars: int = 1
    switch_min_rebound_atr: float = 0.55
    switch_max_rebound_atr: float = 2.4
    switch_box_left_bars: int = 5
    switch_box_right_bars: int = 10

    test_size_pct: float = 0.25
    n_splits: int = 5
    probability_threshold: float | None = None
    target_precision: float = 0.55
    cooldown_bars: int = 30
    interval_probability_floor: float | None = None
    interval_merge_gap_bars: int = 6
    min_interval_bars: int = 5
    max_interval_bars: int = 90
    interval_expand_atr: float = 0.35
    confidence_window: int = 20
    confidence_z: float = 1.0

    model_names: tuple[str, ...] = ("boosting", "ridge", "random_forest")
    use_hmm: bool = True
    hmm_states: int = 4
    use_arima: bool = False
    arima_order: tuple[int, int, int] = (1, 0, 1)
    arima_window: int = 160
    use_pca: bool = False
    pca_variance: float = 0.95
    permutation_repeats: int = 10


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")
    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if "volume" not in out.columns:
        out["volume"] = 0.0
    out["volume"] = out["volume"].fillna(0.0).clip(lower=0.0)

    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")
    return out


INTRADAY_INTERVALS = {"1m", "2m", "5m", "15m", "30m", "60m", "90m", "1h"}


def fetch_ohlcv(symbol: str, period: str, interval: str) -> pd.DataFrame:
    if Ticker is None:
        raise ImportError(
            "yahooquery is required for online data fetch. "
            "Install with: pip install yahooquery"
        )

    errors: list[str] = []
    for candidate_period in data_period_attempts(period=period, interval=interval):
        try:
            raw = Ticker(symbol).history(period=candidate_period, interval=interval)
            frame = normalize_yahoo_history(raw, symbol=symbol)
            data = prepare_ohlcv(frame)
            if candidate_period != period:
                warnings.warn(
                    f"No usable data for period={period!r}, interval={interval!r}; "
                    f"using period={candidate_period!r} instead."
                )
            return data
        except Exception as exc:
            errors.append(f"{candidate_period}: {type(exc).__name__}: {exc}")

    raise ValueError(
        f"No data returned for {symbol} with interval={interval}. "
        "For Yahoo intraday data, use a shorter period such as 60d/30d/7d "
        "or a wider interval such as 1d. Attempts: "
        + " | ".join(errors)
    )


def data_period_attempts(*, period: str, interval: str) -> list[str]:
    attempts = [str(period)]
    if str(interval).lower() in INTRADAY_INTERVALS:
        attempts.extend(["60d", "30d", "7d", "5d", "1mo"])
    else:
        attempts.extend(["2y", "1y", "6mo", "3mo"])
    return list(dict.fromkeys(attempts))


def normalize_yahoo_history(raw: Any, *, symbol: str) -> pd.DataFrame:
    if raw is None:
        raise ValueError("empty response")
    if isinstance(raw, dict):
        message = raw.get(symbol, raw)
        raise ValueError(f"Yahoo returned an error response: {message}")
    if isinstance(raw, pd.Series):
        raise ValueError(f"Yahoo returned a Series instead of OHLCV data: {raw.to_dict()}")
    if not isinstance(raw, pd.DataFrame):
        raise ValueError(f"Unexpected Yahoo response type: {type(raw).__name__}")
    if raw.empty:
        raise ValueError("empty response")

    frame = raw.copy()
    if isinstance(frame.index, pd.MultiIndex):
        if "symbol" in frame.index.names:
            frame = frame.reset_index(level="symbol", drop=True)
        else:
            frame = frame.reset_index()
    return frame


def load_ohlcv(config: ReversalMLConfig) -> pd.DataFrame:
    if config.csv_path:
        return prepare_ohlcv(pd.read_csv(config.csv_path))
    return fetch_ohlcv(config.symbol, config.period, config.interval)


def make_feature_frame(data: pd.DataFrame, config: ReversalMLConfig) -> pd.DataFrame:
    close = data["close"].astype(float)
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    open_ = data["open"].astype(float)
    volume = data["volume"].astype(float)

    features = pd.DataFrame(index=data.index)
    log_close = np.log(close)
    log_return = log_close.diff()

    features["log_return_1"] = log_return
    for window in [2, 3, 5, 10, 20, 40, 80]:
        features[f"log_return_sum_{window}"] = log_return.rolling(window).sum()
        features[f"realized_vol_{window}"] = log_return.rolling(window).std()

    atr = average_true_range(data, period=14)
    features["atr"] = atr
    features["atr_pct"] = atr / close.replace(0, np.nan) * 100.0
    features["atr_pct_change_5"] = features["atr_pct"].pct_change(5)

    features["rsi_7"] = rsi(close, period=7)
    features["rsi_14"] = rsi(close, period=14)
    macd_line, macd_signal, macd_hist = macd(close)
    features["macd_line"] = macd_line
    features["macd_signal"] = macd_signal
    features["macd_hist"] = macd_hist
    features["stoch_k_14"] = stochastic_k(high, low, close, period=14)
    features["stoch_d_3"] = features["stoch_k_14"].rolling(3).mean()

    feature_windows = sorted({20, 50, 80, int(config.label_lookback)})
    for window in feature_windows:
        roll_high = high.rolling(window).max()
        roll_low = low.rolling(window).min()
        roll_mid = close.rolling(window).mean()
        roll_std = close.rolling(window).std()

        features[f"drawdown_from_high_{window}"] = (roll_high - close) / roll_high * 100.0
        features[f"rebound_from_low_{window}"] = (close - roll_low) / roll_low * 100.0
        features[f"dist_to_low_atr_{window}"] = (close - roll_low) / atr.replace(0, np.nan)
        features[f"dist_to_high_atr_{window}"] = (roll_high - close) / atr.replace(0, np.nan)
        features[f"bollinger_z_{window}"] = (close - roll_mid) / roll_std.replace(0, np.nan)
        features[f"range_pct_{window}"] = (roll_high - roll_low) / roll_mid.replace(0, np.nan) * 100.0
        features[f"slope_pct_{window}"] = rolling_slope_pct(close, window=window)

    for window in [5, 8, 13, 21]:
        prior_high = high.rolling(window).max().shift(1)
        recent_low = low.rolling(window).min()
        recent_high = high.rolling(window).max()
        recent_range = (recent_high - recent_low).replace(0, np.nan)
        features[f"reclaim_from_low_atr_{window}"] = (close - recent_low) / atr.replace(0, np.nan)
        features[f"break_above_prior_high_atr_{window}"] = (close - prior_high) / atr.replace(0, np.nan)
        features[f"low_range_position_{window}"] = (close - recent_low) / recent_range
        features[f"short_turn_score_{window}"] = (
            features[f"reclaim_from_low_atr_{window}"]
            + features[f"break_above_prior_high_atr_{window}"].clip(lower=-2.0, upper=2.0)
        )

    features["volatility_expansion_5_20"] = (
        features["realized_vol_5"] / features["realized_vol_20"].replace(0, np.nan)
    )
    features["volatility_compression_20_80"] = (
        features["realized_vol_20"] / features["realized_vol_80"].replace(0, np.nan)
    )
    features["range_compression_20_80"] = (
        features["range_pct_20"] / features["range_pct_80"].replace(0, np.nan)
    )

    candle_range = (high - low).replace(0, np.nan)
    features["body_pct"] = (close - open_) / open_.replace(0, np.nan) * 100.0
    features["body_to_range"] = (close - open_).abs() / candle_range
    features["upper_wick_to_range"] = (high - np.maximum(open_, close)) / candle_range
    features["lower_wick_to_range"] = (np.minimum(open_, close) - low) / candle_range
    features["close_position_in_range"] = (close - low) / candle_range
    features["gap_pct"] = (open_ - close.shift(1)) / close.shift(1).replace(0, np.nan) * 100.0
    features["green_streak"] = signed_streak(close >= open_)

    features["log_volume"] = np.log1p(volume)
    features["volume_z_20"] = rolling_zscore(features["log_volume"], window=20)
    features["volume_ratio_20"] = volume / volume.rolling(20).mean().replace(0, np.nan)
    up_volume = volume.where(close >= open_, 0.0)
    down_volume = volume.where(close < open_, 0.0)
    rolling_volume_20 = volume.rolling(20).sum().replace(0, np.nan)
    features["up_volume_share_20"] = up_volume.rolling(20).sum() / rolling_volume_20
    features["down_volume_share_20"] = down_volume.rolling(20).sum() / rolling_volume_20
    features["volume_expansion_5_20"] = (
        volume.rolling(5).mean() / volume.rolling(20).mean().replace(0, np.nan)
    )
    obv = on_balance_volume(close, volume)
    features["obv_slope_20"] = rolling_slope_pct(obv, window=20)

    features["prior_down_context"] = (
        (features[f"drawdown_from_high_{config.label_lookback}"] >= config.min_prior_drawdown_pct)
        & (features[f"dist_to_low_atr_{config.label_lookback}"] <= config.max_distance_to_low_atr)
    ).astype(float)

    if config.use_arima:
        arima_features = walk_forward_arima_features(
            log_return,
            order=config.arima_order,
            window=config.arima_window,
        )
        features = features.join(arima_features)

    features = features.replace([np.inf, -np.inf], np.nan)
    return features


def make_reversal_labels(
    data: pd.DataFrame,
    features: pd.DataFrame,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    open_ = data["open"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    atr = features["atr"].to_numpy(dtype=float)
    prior_drawdown = features[f"drawdown_from_high_{config.label_lookback}"].to_numpy(dtype=float)
    dist_to_low = features[f"dist_to_low_atr_{config.label_lookback}"].to_numpy(dtype=float)

    labels = pd.DataFrame(index=data.index)
    labels["reversal_label"] = np.nan
    labels["label_target_price"] = np.nan
    labels["label_stop_price"] = np.nan
    labels["label_hit_bar_idx"] = np.nan
    labels["label_hit_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["label_hit_type"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["forecast_forward_return_pct"] = np.nan
    labels["forecast_max_return_pct"] = np.nan
    labels["forecast_min_return_pct"] = np.nan
    labels["forecast_return_atr"] = np.nan
    labels["forecast_horizon_bars"] = np.nan
    labels["prior_drawdown_pct"] = prior_drawdown
    labels["distance_to_low_atr"] = dist_to_low
    labels["switch_kind"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["switch_turn_bar_idx"] = np.nan
    labels["switch_turn_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["switch_confirm_bar_idx"] = np.nan
    labels["switch_confirm_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["bars_from_switch_turn"] = np.nan
    labels["switch_rebound_atr"] = np.nan

    n = len(data)
    switch_events = detect_bullish_switch_events(
        data=data,
        features=features,
        open_=open_,
        high=high,
        low=low,
        close=close,
        atr=atr,
        prior_drawdown=prior_drawdown,
        dist_to_low=dist_to_low,
        config=config,
    )
    switch_lookup = make_switch_label_lookup(switch_events, n=n, config=config)

    forecast_horizon = int(config.forecast_horizon or config.label_horizon)
    for bar_idx in range(n):
        if bar_idx + config.label_horizon >= n:
            continue
        if not np.isfinite(atr[bar_idx]) or atr[bar_idx] <= 0:
            continue

        forecast_end_idx = min(n - 1, bar_idx + forecast_horizon)
        if forecast_end_idx > bar_idx and close[bar_idx] != 0:
            future_slice = slice(bar_idx + 1, forecast_end_idx + 1)
            forward_return_pct = (close[forecast_end_idx] / close[bar_idx] - 1.0) * 100.0
            max_return_pct = (finite_max(high[future_slice]) / close[bar_idx] - 1.0) * 100.0
            min_return_pct = (finite_min(low[future_slice]) / close[bar_idx] - 1.0) * 100.0
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_forward_return_pct")] = forward_return_pct
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_max_return_pct")] = max_return_pct
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_min_return_pct")] = min_return_pct
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_return_atr")] = (
                close[forecast_end_idx] - close[bar_idx]
            ) / atr[bar_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_horizon_bars")] = forecast_end_idx - bar_idx

        target_price = close[bar_idx] + config.target_atr * atr[bar_idx]
        stop_price = close[bar_idx] - config.stop_atr * atr[bar_idx]
        labels.iloc[bar_idx, labels.columns.get_loc("label_target_price")] = target_price
        labels.iloc[bar_idx, labels.columns.get_loc("label_stop_price")] = stop_price

        in_reversal_context = (
            np.isfinite(prior_drawdown[bar_idx])
            and np.isfinite(dist_to_low[bar_idx])
            and prior_drawdown[bar_idx] >= config.min_prior_drawdown_pct
            and dist_to_low[bar_idx] <= config.max_distance_to_low_atr
        )
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=bar_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        switch_event = switch_lookup.get(bar_idx)
        label = int(
            switch_event is not None
            and in_reversal_context
            and hit_type == "target"
        )
        labels.iloc[bar_idx, labels.columns.get_loc("reversal_label")] = label
        if hit_idx is not None:
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_bar_idx")] = int(hit_idx)
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_date")] = data.index[int(hit_idx)]
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_type")] = hit_type
        if switch_event is not None:
            turn_idx = int(switch_event["turn_idx"])
            confirm_idx = int(switch_event["confirm_idx"])
            labels.iloc[bar_idx, labels.columns.get_loc("switch_kind")] = switch_event["kind"]
            labels.iloc[bar_idx, labels.columns.get_loc("switch_turn_bar_idx")] = turn_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_turn_date")] = data.index[turn_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("switch_confirm_bar_idx")] = confirm_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_confirm_date")] = data.index[confirm_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("bars_from_switch_turn")] = bar_idx - turn_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_rebound_atr")] = switch_event["rebound_atr"]

    return labels


def detect_bullish_switch_events(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr: np.ndarray,
    prior_drawdown: np.ndarray,
    dist_to_low: np.ndarray,
    config: ReversalMLConfig,
) -> list[dict[str, Any]]:
    prior_context = features.get("prior_down_context", pd.Series(0.0, index=features.index)).to_numpy(
        dtype=float
    )
    left_bars = max(3, int(config.switch_left_bars))
    right_bars = max(1, int(config.switch_right_bars))
    confirmation_bars = max(right_bars, int(config.switch_confirmation_bars))
    n = len(data)
    max_future_need = max(right_bars, confirmation_bars, int(config.label_horizon))
    events: list[dict[str, Any]] = []

    for turn_idx in range(left_bars, n - max_future_need):
        if not np.isfinite(atr[turn_idx]) or atr[turn_idx] <= 0:
            continue
        if not is_bullish_switch_context(
            turn_idx=turn_idx,
            close=close,
            high=high,
            prior_context=prior_context,
            prior_drawdown=prior_drawdown,
            dist_to_low=dist_to_low,
            config=config,
        ):
            continue
        if not is_local_valley(low, turn_idx=turn_idx, left_bars=left_bars, right_bars=right_bars):
            continue

        confirm_idx = find_bullish_switch_confirmation(
            turn_idx=turn_idx,
            open_=open_,
            high=high,
            low=low,
            close=close,
            atr=atr,
            config=config,
        )
        if confirm_idx is None or confirm_idx + config.label_horizon >= n:
            continue

        target_price = close[confirm_idx] + config.target_atr * atr[confirm_idx]
        stop_price = close[confirm_idx] - config.stop_atr * atr[confirm_idx]
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=confirm_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        if hit_type != "target":
            continue

        events.append(
            {
                "kind": "valley",
                "turn_idx": int(turn_idx),
                "confirm_idx": int(confirm_idx),
                "hit_idx": int(hit_idx) if hit_idx is not None else None,
                "rebound_atr": float((close[confirm_idx] - low[turn_idx]) / atr[turn_idx]),
                "turn_price": float(low[turn_idx]),
                "confirm_price": float(close[confirm_idx]),
            }
        )

    return suppress_overlapping_switch_events(
        events,
        min_gap=max(2, int(config.switch_confirmation_bars)),
    )


def is_bullish_switch_context(
    *,
    turn_idx: int,
    close: np.ndarray,
    high: np.ndarray,
    prior_context: np.ndarray,
    prior_drawdown: np.ndarray,
    dist_to_low: np.ndarray,
    config: ReversalMLConfig,
) -> bool:
    context_start = max(0, int(turn_idx) - int(config.label_lookback))
    context_seen = finite_max(prior_context[context_start : turn_idx + 1])
    if not np.isfinite(context_seen) or context_seen <= 0:
        return False
    if not np.isfinite(prior_drawdown[turn_idx]) or not np.isfinite(dist_to_low[turn_idx]):
        return False
    if prior_drawdown[turn_idx] < config.min_prior_drawdown_pct:
        return False
    if dist_to_low[turn_idx] > config.max_distance_to_low_atr:
        return False

    approach_start = max(0, int(turn_idx) - max(5, int(config.switch_left_bars)))
    if turn_idx - approach_start < 4:
        return False
    approach_high = finite_max(high[approach_start : turn_idx + 1])
    if not np.isfinite(approach_high) or approach_high <= 0:
        return False
    approach_drawdown = (approach_high - close[turn_idx]) / approach_high * 100.0
    if approach_drawdown < config.min_prior_drawdown_pct * 0.35:
        return False
    return close[turn_idx] <= np.nanmedian(close[approach_start : turn_idx + 1])


def is_local_valley(
    low: np.ndarray,
    *,
    turn_idx: int,
    left_bars: int,
    right_bars: int,
) -> bool:
    start = max(0, int(turn_idx) - int(left_bars))
    end = min(len(low) - 1, int(turn_idx) + int(right_bars))
    window = low[start : end + 1]
    if len(window) == 0 or not np.isfinite(low[turn_idx]):
        return False
    valley = finite_min(window)
    if not np.isfinite(valley):
        return False
    if low[turn_idx] > valley + max(abs(valley) * 1e-8, 1e-10):
        return False
    right_window = low[turn_idx + 1 : end + 1]
    return len(right_window) == 0 or finite_min(right_window) > low[turn_idx] - max(abs(low[turn_idx]) * 1e-8, 1e-10)


def find_bullish_switch_confirmation(
    *,
    turn_idx: int,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr: np.ndarray,
    config: ReversalMLConfig,
) -> int | None:
    atr_at_turn = float(atr[turn_idx])
    if not np.isfinite(atr_at_turn) or atr_at_turn <= 0:
        return None
    end_idx = min(len(close) - 1, int(turn_idx) + int(config.switch_confirmation_bars))
    rebound_level = low[turn_idx] + config.switch_min_rebound_atr * atr_at_turn
    micro_start = max(0, int(turn_idx) - max(3, int(config.switch_left_bars) // 2))
    micro_break_level = finite_max(high[micro_start:turn_idx])

    for idx in range(int(turn_idx) + 1, end_idx + 1):
        rebound_atr = (close[idx] - low[turn_idx]) / atr_at_turn
        if rebound_atr < config.switch_min_rebound_atr:
            continue
        if rebound_atr > config.switch_max_rebound_atr:
            continue
        closes_green = close[idx] > open_[idx]
        breaks_previous_high = idx > 0 and close[idx] > high[idx - 1]
        breaks_micro_high = np.isfinite(micro_break_level) and close[idx] > micro_break_level
        reclaims_valley = close[idx] >= rebound_level
        if reclaims_valley and (closes_green or breaks_previous_high or breaks_micro_high):
            return int(idx)
    return None


def make_switch_label_lookup(
    events: list[dict[str, Any]],
    *,
    n: int,
    config: ReversalMLConfig,
) -> dict[int, dict[str, Any]]:
    lookup: dict[int, dict[str, Any]] = {}
    radius = max(0, int(config.switch_label_window_bars))
    for event in events:
        anchor_idx = int(event["confirm_idx"])
        turn_idx = int(event["turn_idx"])
        for label_idx in range(anchor_idx - radius, anchor_idx + radius + 1):
            if label_idx < turn_idx or label_idx < 0 or label_idx >= n:
                continue
            current = lookup.get(label_idx)
            if current is None:
                lookup[label_idx] = event
                continue
            current_distance = abs(label_idx - int(current["confirm_idx"]))
            new_distance = abs(label_idx - anchor_idx)
            if new_distance < current_distance:
                lookup[label_idx] = event
    return lookup


def suppress_overlapping_switch_events(
    events: list[dict[str, Any]],
    *,
    min_gap: int,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []
    for event in sorted(events, key=lambda item: int(item["confirm_idx"])):
        if not selected:
            selected.append(event)
            continue
        previous = selected[-1]
        if int(event["confirm_idx"]) - int(previous["confirm_idx"]) > min_gap:
            selected.append(event)
            continue
        if float(event["turn_price"]) < float(previous["turn_price"]):
            selected[-1] = event
    return selected


def first_long_barrier_hit(
    *,
    high: np.ndarray,
    low: np.ndarray,
    start_idx: int,
    horizon: int,
    target_price: float,
    stop_price: float,
) -> tuple[str, int | None]:
    max_idx = min(len(high) - 1, int(start_idx) + int(horizon))
    for idx in range(int(start_idx) + 1, max_idx + 1):
        stop_hit = low[idx] <= stop_price
        target_hit = high[idx] >= target_price
        if stop_hit and target_hit:
            return "stop", idx
        if stop_hit:
            return "stop", idx
        if target_hit:
            return "target", idx
    return "timeout", None


def add_hmm_features(
    features: pd.DataFrame,
    train_index: pd.Index,
    config: ReversalMLConfig,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    state_summary_columns = ["state", "train_count", "mean_return", "mean_vol", "mean_drawdown"]
    if not config.use_hmm:
        return features, pd.DataFrame(columns=state_summary_columns)
    if GaussianHMM is None or SimpleImputer is None or StandardScaler is None:
        warnings.warn("hmmlearn/sklearn unavailable; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    hmm_columns = [
        column
        for column in [
            "log_return_1",
            "realized_vol_10",
            "atr_pct",
            "volume_z_20",
            f"drawdown_from_high_{config.label_lookback}",
            f"dist_to_low_atr_{config.label_lookback}",
        ]
        if column in features.columns
    ]
    if len(hmm_columns) < 3:
        return features, pd.DataFrame(columns=state_summary_columns)

    train = features.loc[train_index, hmm_columns]
    if train.dropna(how="all").shape[0] < config.hmm_states * 20:
        warnings.warn("Not enough training rows for HMM; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    imputer = SimpleImputer(strategy="median")
    scaler = StandardScaler()
    train_values = scaler.fit_transform(imputer.fit_transform(train))
    all_values = scaler.transform(imputer.transform(features[hmm_columns]))

    hmm = GaussianHMM(
        n_components=config.hmm_states,
        covariance_type="diag",
        n_iter=200,
        tol=1e-4,
        random_state=config.random_state,
    )
    try:
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            hmm.fit(train_values)
            probabilities = hmm.predict_proba(all_values)
    except Exception as exc:
        warnings.warn(f"HMM regime features skipped after fit failure: {type(exc).__name__}: {exc}")
        return features, pd.DataFrame(columns=state_summary_columns)
    states = probabilities.argmax(axis=1)

    out = features.copy()
    out["hmm_state"] = states.astype(float)
    for state in range(config.hmm_states):
        out[f"hmm_state_prob_{state}"] = probabilities[:, state]

    state_rows: list[dict[str, Any]] = []
    train_augmented = out.loc[train_index].copy()
    for state in range(config.hmm_states):
        subset = train_augmented[train_augmented["hmm_state"] == state]
        state_rows.append(
            {
                "state": state,
                "train_count": int(len(subset)),
                "mean_return": float(subset["log_return_1"].mean()),
                "mean_vol": float(subset["realized_vol_10"].mean()),
                "mean_drawdown": float(
                    subset[f"drawdown_from_high_{config.label_lookback}"].mean()
                ),
            }
        )
    return out, pd.DataFrame(state_rows, columns=state_summary_columns)


def train_reversal_models(
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    config: ReversalMLConfig,
) -> dict[str, Any]:
    require_sklearn()

    target_column = config.forecast_target_column
    if target_column not in labels.columns:
        raise ValueError(f"Unknown forecast target column: {target_column}")

    labeled = features.join(labels[[target_column]], how="inner")
    labeled = labeled.dropna(subset=[target_column])
    if len(labeled) < 200:
        raise ValueError("Not enough labeled rows. Increase period or use a higher timeframe.")

    split_idx = int(len(labeled) * (1.0 - config.test_size_pct))
    split_idx = min(max(split_idx, 50), len(labeled) - 20)
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]

    features_with_hmm, hmm_summary = add_hmm_features(features, train_index, config)
    labeled = features_with_hmm.join(labels[[target_column]], how="inner")
    labeled = labeled.dropna(subset=[target_column])
    y = labeled[target_column].astype(float)
    feature_columns = [
        column
        for column in labeled.columns
        if column != target_column and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]
    split_idx = int(len(labeled) * (1.0 - config.test_size_pct))
    split_idx = min(max(split_idx, 50), len(labeled) - 20)
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]
    X_train = X.loc[train_index]
    y_train = y.loc[train_index]
    X_test = X.loc[test_index]
    y_test = y.loc[test_index]
    if y_train.dropna().nunique() < 2:
        raise ValueError(
            "Training split has no return variation. Use more history or a higher timeframe."
        )

    X_train_clip, clip_bounds = fit_winsorize(X_train)
    X_test_clip = apply_winsorize(X_test, clip_bounds)
    X_all_clip = apply_winsorize(features_with_hmm[feature_columns], clip_bounds)
    return_threshold_pct = choose_return_threshold(y_train, config)
    forecast_score_scale = choose_forecast_score_scale(y_train, config)
    train_sample_weight = make_return_sample_weight(labels, y_train, train_index, config)

    models = build_models(config)
    results: list[dict[str, Any]] = []
    fitted_models: dict[str, Pipeline] = {}

    for name, model in models.items():
        fitted = fit_pipeline_with_weights(
            model,
            X_train_clip,
            y_train,
            sample_weight=train_sample_weight,
        )
        fitted_models[name] = fitted
        test_forecast = predict_return(fitted, X_test_clip)
        metrics = evaluate_return_forecasts(
            y_test,
            test_forecast,
            return_threshold_pct=return_threshold_pct,
        )
        cv_metrics = time_series_cv_score(
            model,
            X_train_clip,
            y_train,
            n_splits=config.n_splits,
            return_threshold_pct=return_threshold_pct,
        )
        result = {"model": name, **metrics, **cv_metrics}
        results.append(result)

    model_report = pd.DataFrame(results)
    best_name = choose_best_model(model_report)
    per_model_probabilities: dict[str, pd.Series] = {}
    per_model_forecasts: dict[str, pd.Series] = {}
    per_model_thresholds: dict[str, float] = {}
    per_model_feature_importance: dict[str, pd.DataFrame] = {}
    per_model_signals: dict[str, pd.DataFrame] = {}
    per_model_intervals: dict[str, pd.DataFrame] = {}

    for name, fitted in fitted_models.items():
        all_forecast = predict_return(fitted, X_all_clip)
        test_forecast = predict_return(fitted, X_test_clip)
        all_score = forecast_to_score(
            all_forecast,
            return_threshold_pct=return_threshold_pct,
            scale_pct=forecast_score_scale,
        )
        test_score = forecast_to_score(
            test_forecast,
            return_threshold_pct=return_threshold_pct,
            scale_pct=forecast_score_scale,
        )
        model_threshold = config.forecast_score_threshold
        if model_threshold is None:
            model_threshold = config.probability_threshold
        if model_threshold is None:
            model_threshold = choose_forecast_score_threshold(
                y_test,
                test_score,
                return_threshold_pct=return_threshold_pct,
                target_precision=config.target_precision,
            )

        forecast_series = pd.Series(
            all_forecast,
            index=features_with_hmm.index,
            name=f"{name}_return_forecast_pct",
        )
        score_series = pd.Series(
            all_score,
            index=features_with_hmm.index,
            name=f"{name}_forecast_score",
        )
        importance = permutation_xai(
            fitted,
            X_test_clip,
            y_test,
            feature_columns,
            repeats=config.permutation_repeats,
            random_state=config.random_state,
        )
        signals = build_signal_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=score_series,
            return_forecasts=forecast_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not signals.empty:
            signals.insert(0, "Model", name)
        intervals = build_interval_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=score_series,
            return_forecasts=forecast_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not intervals.empty:
            intervals.insert(0, "Model", name)

        per_model_probabilities[name] = score_series
        per_model_forecasts[name] = forecast_series
        per_model_thresholds[name] = float(model_threshold)
        per_model_feature_importance[name] = importance
        per_model_signals[name] = signals
        per_model_intervals[name] = intervals

    best_model = fitted_models[best_name]
    threshold = per_model_thresholds[best_name]
    probabilities = per_model_probabilities[best_name].rename("forecast_score")
    forecasts = per_model_forecasts[best_name].rename("return_forecast_pct")
    signals = per_model_signals[best_name].copy()
    if "Model" in signals.columns:
        signals = signals.drop(columns=["Model"])
    intervals = per_model_intervals[best_name].copy()
    if "Model" in intervals.columns:
        intervals = intervals.drop(columns=["Model"])
    feature_importance = per_model_feature_importance[best_name]
    pca_loadings = extract_pca_loadings(best_model, feature_columns)

    return {
        "features": features_with_hmm,
        "labels": labels,
        "feature_columns": feature_columns,
        "model_report": model_report,
        "best_model_name": best_name,
        "best_model": best_model,
        "probabilities": probabilities,
        "forecasts": forecasts,
        "threshold": threshold,
        "return_threshold_pct": return_threshold_pct,
        "forecast_score_scale": forecast_score_scale,
        "signals": signals,
        "intervals": intervals,
        "feature_importance": feature_importance,
        "per_model_probabilities": per_model_probabilities,
        "per_model_forecasts": per_model_forecasts,
        "per_model_thresholds": per_model_thresholds,
        "per_model_signals": per_model_signals,
        "per_model_intervals": per_model_intervals,
        "per_model_feature_importance": per_model_feature_importance,
        "pca_loadings": pca_loadings,
        "hmm_state_summary": hmm_summary,
    }


def require_sklearn() -> None:
    if Pipeline is None:
        raise ImportError(
            "scikit-learn is required for ML training. Install with: pip install scikit-learn"
        )


def build_models(config: ReversalMLConfig) -> dict[str, Pipeline]:
    requested = {name.strip().lower() for name in config.model_names}
    models: dict[str, Pipeline] = {}

    if "boosting" in requested:
        models["boosting"] = make_pipeline(
            HistGradientBoostingRegressor(
                learning_rate=0.05,
                max_iter=350,
                max_leaf_nodes=15,
                l2_regularization=0.05,
                early_stopping=True,
                random_state=config.random_state,
            ),
            config,
            scale=False,
        )
    if "ridge" in requested:
        models["ridge"] = make_pipeline(
            Ridge(
                alpha=5.0,
                random_state=config.random_state,
            ),
            config,
            scale=True,
        )
    if "random_forest" in requested:
        models["random_forest"] = make_pipeline(
            RandomForestRegressor(
                n_estimators=400,
                max_depth=7,
                min_samples_leaf=20,
                random_state=config.random_state,
                n_jobs=-1,
            ),
            config,
            scale=False,
        )
    if not models:
        raise ValueError("No valid model names requested.")
    return models


def make_pipeline(estimator: Any, config: ReversalMLConfig, *, scale: bool) -> Pipeline:
    steps: list[tuple[str, Any]] = [("imputer", SimpleImputer(strategy="median"))]
    if scale or config.use_pca:
        steps.append(("scaler", StandardScaler()))
    if config.use_pca:
        steps.append(("pca", PCA(n_components=config.pca_variance, svd_solver="full")))
    steps.append(("model", estimator))
    return Pipeline(steps)


def fit_pipeline_with_weights(
    model: Pipeline,
    X: pd.DataFrame,
    y: pd.Series,
    *,
    sample_weight: np.ndarray | pd.Series | None = None,
) -> Pipeline:
    try:
        return model.fit(X, y, model__sample_weight=sample_weight)
    except TypeError:
        return model.fit(X, y)


def predict_return(model: Pipeline, X: pd.DataFrame) -> np.ndarray:
    return np.asarray(model.predict(X), dtype=float)


def evaluate_return_forecasts(
    y_true: pd.Series,
    forecast: np.ndarray,
    *,
    return_threshold_pct: float,
) -> dict[str, float]:
    y_values = y_true.to_numpy(dtype=float)
    forecast_values = np.asarray(forecast, dtype=float)
    finite_mask = np.isfinite(y_values) & np.isfinite(forecast_values)
    out: dict[str, float] = {
        "test_mae": np.nan,
        "test_rmse": np.nan,
        "test_r2": np.nan,
        "test_correlation": np.nan,
        "test_directional_accuracy": np.nan,
        "test_actual_positive_rate": np.nan,
        "test_predicted_positive_rate": np.nan,
        "test_positive_precision": np.nan,
        "test_positive_recall": np.nan,
    }
    if finite_mask.sum() < 3:
        return out

    y_values = y_values[finite_mask]
    forecast_values = forecast_values[finite_mask]
    out["test_mae"] = float(mean_absolute_error(y_values, forecast_values))
    out["test_rmse"] = float(math.sqrt(mean_squared_error(y_values, forecast_values)))
    out["test_r2"] = float(r2_score(y_values, forecast_values))
    if np.std(y_values) > 0 and np.std(forecast_values) > 0:
        out["test_correlation"] = float(np.corrcoef(y_values, forecast_values)[0, 1])
    out["test_directional_accuracy"] = float(np.mean(np.sign(y_values) == np.sign(forecast_values)))

    actual_positive = y_values >= float(return_threshold_pct)
    predicted_positive = forecast_values >= float(return_threshold_pct)
    out["test_actual_positive_rate"] = float(np.mean(actual_positive))
    out["test_predicted_positive_rate"] = float(np.mean(predicted_positive))
    true_positive = float(np.sum(actual_positive & predicted_positive))
    out["test_positive_precision"] = true_positive / max(float(np.sum(predicted_positive)), 1.0)
    out["test_positive_recall"] = true_positive / max(float(np.sum(actual_positive)), 1.0)
    return out


def time_series_cv_score(
    model: Pipeline,
    X: pd.DataFrame,
    y: pd.Series,
    *,
    n_splits: int,
    return_threshold_pct: float,
) -> dict[str, float]:
    out = {"cv_mae": np.nan, "cv_rmse": np.nan, "cv_correlation": np.nan}
    if len(X) < 200 or y.dropna().nunique() < 2:
        return out

    split_count = min(n_splits, max(2, len(X) // 200))
    splitter = TimeSeriesSplit(n_splits=split_count)
    mae_scores: list[float] = []
    rmse_scores: list[float] = []
    corr_scores: list[float] = []
    for train_idx, test_idx in splitter.split(X):
        y_train = y.iloc[train_idx]
        y_valid = y.iloc[test_idx]
        if y_train.dropna().nunique() < 2:
            continue
        fitted = fit_pipeline_with_weights(clone(model), X.iloc[train_idx], y_train)
        forecast = predict_return(fitted, X.iloc[test_idx])
        metrics = evaluate_return_forecasts(
            y_valid,
            forecast,
            return_threshold_pct=return_threshold_pct,
        )
        if np.isfinite(metrics["test_mae"]):
            mae_scores.append(metrics["test_mae"])
        if np.isfinite(metrics["test_rmse"]):
            rmse_scores.append(metrics["test_rmse"])
        if np.isfinite(metrics["test_correlation"]):
            corr_scores.append(metrics["test_correlation"])

    if mae_scores:
        out["cv_mae"] = float(np.mean(mae_scores))
    if rmse_scores:
        out["cv_rmse"] = float(np.mean(rmse_scores))
    if corr_scores:
        out["cv_correlation"] = float(np.mean(corr_scores))
    return out


def choose_best_model(model_report: pd.DataFrame) -> str:
    ranking = model_report.copy()
    mae_rank = ranking["test_mae"].rank(ascending=True, pct=True).fillna(1.0)
    ranking["_rank_score"] = (
        ranking["test_correlation"].fillna(0.0)
        + ranking["test_directional_accuracy"].fillna(0.0)
        - mae_rank
    )
    ranking = ranking.sort_values(["_rank_score", "test_rmse"], ascending=[False, True])
    return str(ranking.iloc[0]["model"])


def choose_forecast_score_threshold(
    y_true: pd.Series,
    forecast_score: np.ndarray,
    *,
    return_threshold_pct: float,
    target_precision: float,
) -> float:
    y_values = y_true.to_numpy(dtype=float)
    score_values = np.asarray(forecast_score, dtype=float)
    finite_mask = np.isfinite(y_values) & np.isfinite(score_values)
    if finite_mask.sum() < 3:
        return 0.60
    actual_positive = y_values[finite_mask] >= float(return_threshold_pct)
    score_values = score_values[finite_mask]
    if len(np.unique(actual_positive)) < 2 or precision_recall_curve is None:
        return 0.60

    precision, recall, thresholds = precision_recall_curve(actual_positive.astype(int), score_values)
    candidates: list[tuple[float, float, float]] = []
    for threshold, prec, rec in zip(thresholds, precision[:-1], recall[:-1]):
        if prec >= target_precision and rec > 0:
            candidates.append((float(threshold), float(prec), float(rec)))
    if not candidates:
        return float(np.quantile(score_values, 0.80))
    candidates.sort(key=lambda item: (item[2], item[1]), reverse=True)
    return float(candidates[0][0])


def choose_return_threshold(y_train: pd.Series, config: ReversalMLConfig) -> float:
    if config.forecast_return_threshold_pct is not None:
        return float(config.forecast_return_threshold_pct)

    values = y_train.dropna().to_numpy(dtype=float)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return float(config.forecast_scale_floor_pct)
    positive_values = values[values > 0]
    if len(positive_values) >= 10:
        threshold = float(np.quantile(positive_values, 0.40))
    else:
        threshold = float(np.quantile(values, 0.65))
    return max(float(config.forecast_scale_floor_pct), threshold)


def choose_forecast_score_scale(y_train: pd.Series, config: ReversalMLConfig) -> float:
    values = y_train.dropna().to_numpy(dtype=float)
    values = values[np.isfinite(values)]
    if len(values) < 5:
        return float(config.forecast_scale_floor_pct)
    q25, q75 = np.quantile(values, [0.25, 0.75])
    robust_scale = float((q75 - q25) / 2.0)
    std_scale = float(np.std(values))
    scale = max(robust_scale, std_scale * 0.35, float(config.forecast_scale_floor_pct), 1e-6)
    return float(scale)


def forecast_to_score(
    forecast: np.ndarray | pd.Series,
    *,
    return_threshold_pct: float,
    scale_pct: float,
) -> np.ndarray:
    values = np.asarray(forecast, dtype=float)
    z = (values - float(return_threshold_pct)) / max(float(scale_pct), 1e-6)
    z = np.clip(z, -30.0, 30.0)
    return 1.0 / (1.0 + np.exp(-z))


def make_return_sample_weight(
    labels: pd.DataFrame,
    y_train: pd.Series,
    train_index: pd.Index,
    config: ReversalMLConfig,
) -> np.ndarray:
    weights = np.ones(len(y_train), dtype=float)
    aligned_labels = labels.reindex(train_index)
    if "switch_kind" in aligned_labels.columns:
        weights += aligned_labels["switch_kind"].notna().to_numpy(dtype=float) * 1.5
    if "prior_drawdown_pct" in aligned_labels.columns:
        drawdown = aligned_labels["prior_drawdown_pct"].fillna(0.0).to_numpy(dtype=float)
        weights += np.clip(drawdown / max(config.min_prior_drawdown_pct, 1e-6), 0.0, 2.0) * 0.35
    values = y_train.to_numpy(dtype=float)
    finite_values = values[np.isfinite(values)]
    if len(finite_values) > 5:
        scale = max(float(np.quantile(np.abs(finite_values), 0.75)), config.forecast_scale_floor_pct)
        weights += np.clip(np.abs(values) / scale, 0.0, 2.0) * 0.25
    return weights


def permutation_xai(
    model: Pipeline,
    X_test: pd.DataFrame,
    y_test: pd.Series,
    feature_columns: list[str],
    *,
    repeats: int,
    random_state: int,
) -> pd.DataFrame:
    columns = ["feature", "importance_mean", "importance_std", "rank"]
    if len(X_test) < 5 or permutation_importance is None:
        return pd.DataFrame(columns=columns)

    result = permutation_importance(
        model,
        X_test,
        y_test,
        n_repeats=repeats,
        random_state=random_state,
        scoring="neg_mean_absolute_error",
        n_jobs=-1,
    )
    importance = pd.DataFrame(
        {
            "feature": feature_columns,
            "importance_mean": result.importances_mean,
            "importance_std": result.importances_std,
        }
    )
    importance = importance.sort_values("importance_mean", ascending=False).reset_index(drop=True)
    importance["rank"] = np.arange(1, len(importance) + 1)
    return importance[columns]


def extract_pca_loadings(model: Pipeline, feature_columns: list[str]) -> pd.DataFrame:
    if "pca" not in model.named_steps:
        return pd.DataFrame(columns=["component", "feature", "loading", "abs_loading", "rank"])

    pca = model.named_steps["pca"]
    rows: list[dict[str, Any]] = []
    for component_idx, component in enumerate(pca.components_):
        for feature, loading in zip(feature_columns, component):
            rows.append(
                {
                    "component": f"PC{component_idx + 1}",
                    "feature": feature,
                    "loading": float(loading),
                    "abs_loading": float(abs(loading)),
                }
            )
    out = pd.DataFrame(rows)
    out["rank"] = out.groupby("component")["abs_loading"].rank(ascending=False, method="first")
    return out.sort_values(["component", "rank"]).reset_index(drop=True)


def build_signal_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    return_forecasts: pd.Series | None,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    candidate_probability = probabilities.dropna()
    candidate_probability = candidate_probability[candidate_probability >= threshold]
    if return_forecasts is not None and not candidate_probability.empty:
        floor = config.interval_probability_floor
        if floor is None:
            floor = max(0.05, min(0.95, float(threshold) * 0.80))
        breakout_filter = forecast_breakout_filter(
            data=data,
            features=features,
            forecast_score=probabilities.reindex(features.index).astype(float),
            return_forecast=return_forecasts.reindex(features.index).astype(float),
            threshold=float(threshold),
            floor=float(floor),
            config=config,
        )
        candidate_probability = candidate_probability[
            breakout_filter.reindex(candidate_probability.index).fillna(False)
        ]
    if candidate_probability.empty:
        return empty_signal_frame()

    selected_indexes = non_max_suppression_1d(
        candidate_probability,
        cooldown_bars=config.cooldown_bars,
        all_index=features.index,
    )

    train_stats = make_train_feature_stats(train_features)
    rows: list[dict[str, Any]] = []
    atr = features["atr"]
    for timestamp in selected_indexes:
        bar_idx = int(data.index.get_loc(timestamp))
        entry_price = float(data.loc[timestamp, "close"])
        return_forecast_pct = (
            float(return_forecasts.loc[timestamp])
            if return_forecasts is not None and timestamp in return_forecasts.index
            else np.nan
        )
        target_price = entry_price + config.target_atr * float(atr.loc[timestamp])
        stop_price = entry_price - config.stop_atr * float(atr.loc[timestamp])
        label_row = labels.loc[timestamp] if timestamp in labels.index else pd.Series(dtype=object)
        hit_idx = label_row.get("label_hit_bar_idx", np.nan)
        hit_date = label_row.get("label_hit_date", pd.NaT)
        hit_type = label_row.get("label_hit_type", None)
        realized_label = label_row.get("reversal_label", np.nan)
        switch_kind = label_row.get("switch_kind", None)
        switch_turn_idx = label_row.get("switch_turn_bar_idx", np.nan)
        switch_turn_date = label_row.get("switch_turn_date", pd.NaT)
        switch_confirm_idx = label_row.get("switch_confirm_bar_idx", np.nan)
        switch_confirm_date = label_row.get("switch_confirm_date", pd.NaT)
        bars_from_switch_turn = label_row.get("bars_from_switch_turn", np.nan)
        top_drivers = local_driver_summary(
            features.loc[timestamp, feature_columns],
            feature_importance,
            train_stats,
            top_n=6,
        )
        rows.append(
            {
                "EntryBarIdx": bar_idx,
                "EntryDate": timestamp,
                "EntryPrice": entry_price,
                "ReturnForecastPct": return_forecast_pct,
                "ForecastScore": float(probabilities.loc[timestamp]),
                "ReversalProbability": float(probabilities.loc[timestamp]),
                "SignalThreshold": float(threshold),
                "TargetPrice": target_price,
                "StopPrice": stop_price,
                "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                "OutcomeKnown": is_finite_number(realized_label),
                "OutcomeLabel": realized_label,
                "HitBarIdx": int(hit_idx) if is_finite_number(hit_idx) else np.nan,
                "HitDate": hit_date,
                "HitType": hit_type,
                "SwitchKind": switch_kind,
                "SwitchTurnBarIdx": int(switch_turn_idx) if is_finite_number(switch_turn_idx) else np.nan,
                "SwitchTurnDate": switch_turn_date,
                "SwitchConfirmBarIdx": int(switch_confirm_idx) if is_finite_number(switch_confirm_idx) else np.nan,
                "SwitchConfirmDate": switch_confirm_date,
                "BarsFromSwitchTurn": bars_from_switch_turn,
                "ShapeHint": classify_shape_hint(data, bar_idx),
                "TopDrivers": top_drivers,
            }
        )
    return pd.DataFrame(rows).sort_values("EntryBarIdx").reset_index(drop=True)


def build_interval_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    return_forecasts: pd.Series | None,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    aligned_probability = probabilities.reindex(features.index).astype(float)
    aligned_return_forecast = (
        return_forecasts.reindex(features.index).astype(float)
        if return_forecasts is not None
        else pd.Series(np.nan, index=features.index)
    )
    confidence = probability_confidence_frame(
        aligned_probability,
        window=config.confidence_window,
        z=config.confidence_z,
    )
    floor = config.interval_probability_floor
    if floor is None:
        floor = max(0.05, min(0.95, float(threshold) * 0.80))

    prior_context = features.get("prior_down_context", pd.Series(0.0, index=features.index))
    context_seen = prior_context.fillna(0.0).rolling(
        config.label_lookback,
        min_periods=1,
    ).max()
    candidate_signal = (
        (aligned_probability >= floor)
        | (
            (confidence["upper"] >= float(threshold))
            & (confidence["mean"] >= floor * 0.85)
        )
    )
    breakout_filter = forecast_breakout_filter(
        data=data,
        features=features,
        forecast_score=aligned_probability,
        return_forecast=aligned_return_forecast,
        threshold=float(threshold),
        floor=float(floor),
        config=config,
    )
    candidate_mask = (
        candidate_signal.fillna(False)
        & (context_seen > 0)
        & breakout_filter.fillna(False)
    ).to_numpy(dtype=bool)
    segments = merge_boolean_segments(
        boolean_segments(candidate_mask),
        max_gap=max(0, int(config.interval_merge_gap_bars)),
    )
    if not segments:
        return empty_interval_frame()

    rows: list[dict[str, Any]] = []
    train_stats = make_train_feature_stats(train_features)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    atr = features["atr"].reindex(data.index).to_numpy(dtype=float)
    min_bars = max(1, int(config.min_interval_bars))
    max_bars = max(min_bars, int(config.max_interval_bars))

    for raw_start, raw_end in segments:
        for seg_start, seg_end in split_segment(raw_start, raw_end, max_bars=max_bars, min_bars=min_bars):
            segment_probability = aligned_probability.iloc[seg_start : seg_end + 1].dropna()
            if segment_probability.empty:
                continue
            segment_confidence = confidence.iloc[seg_start : seg_end + 1]
            if (
                float(segment_probability.max()) < float(threshold)
                and float(segment_confidence["upper"].max()) < float(threshold)
            ):
                continue

            peak_timestamp = segment_probability.idxmax()
            peak_idx = int(features.index.get_loc(peak_timestamp))
            search_start = max(0, peak_idx - max(config.switch_left_bars, config.switch_box_left_bars))
            search_end = min(len(data) - 1, peak_idx + max(config.switch_right_bars, 1))
            bottom_idx = search_start + int(np.nanargmin(low[search_start : search_end + 1]))
            start_idx = max(0, bottom_idx - int(config.switch_box_left_bars))
            end_idx = min(
                len(data) - 1,
                max(peak_idx, bottom_idx) + int(config.switch_box_right_bars),
            )

            if end_idx - start_idx + 1 < min_bars:
                start_idx = max(0, peak_idx - min_bars // 2)
                end_idx = min(len(data) - 1, start_idx + min_bars - 1)
            if end_idx - start_idx + 1 > max_bars:
                center_idx = int(round((peak_idx + bottom_idx) / 2))
                start_idx = max(0, center_idx - max_bars // 2)
                end_idx = min(len(data) - 1, start_idx + max_bars - 1)

            context_start = max(0, start_idx - config.label_lookback // 2)
            context_slice = prior_context.iloc[context_start : peak_idx + 1].fillna(0.0)
            if context_slice.max() <= 0:
                continue

            interval_probability = aligned_probability.iloc[start_idx : end_idx + 1].dropna()
            if interval_probability.empty:
                continue
            interval_forecast = aligned_return_forecast.iloc[start_idx : end_idx + 1].dropna()
            probability_mean = float(interval_probability.mean())
            probability_max = float(interval_probability.max())
            forecast_mean = float(interval_forecast.mean()) if not interval_forecast.empty else np.nan
            forecast_max = float(interval_forecast.max()) if not interval_forecast.empty else np.nan
            forecast_peak = (
                float(aligned_return_forecast.iloc[peak_idx])
                if np.isfinite(aligned_return_forecast.iloc[peak_idx])
                else np.nan
            )
            interval_confidence_low, interval_confidence_high = probability_interval_bounds(
                interval_probability,
                z=config.confidence_z,
            )
            if probability_max < float(threshold) and interval_confidence_high < float(threshold):
                continue

            support_low = finite_min(low[start_idx : end_idx + 1])
            resistance_high = finite_max(high[start_idx : end_idx + 1])
            if not np.isfinite(support_low) or not np.isfinite(resistance_high):
                continue

            entry_price = float(close[end_idx])
            interval_atr = finite_mean(atr[start_idx : end_idx + 1])
            if not np.isfinite(interval_atr) or interval_atr <= 0:
                interval_atr = max((resistance_high - support_low) * 0.10, abs(entry_price) * 0.001, 1e-12)
            price_pad = max(0.0, float(config.interval_expand_atr)) * interval_atr
            target_price = entry_price + config.target_atr * interval_atr
            stop_price = entry_price - config.stop_atr * interval_atr
            peak_label = labels.loc[peak_timestamp] if peak_timestamp in labels.index else pd.Series(dtype=object)
            hit_idx = peak_label.get("label_hit_bar_idx", np.nan)
            realized_label = peak_label.get("reversal_label", np.nan)
            top_drivers = local_driver_summary(
                features.loc[peak_timestamp, feature_columns],
                feature_importance,
                train_stats,
                top_n=6,
            )

            volume_z_max = finite_max(series_values(features, "volume_z_20", start_idx, end_idx))
            volume_ratio_max = finite_max(series_values(features, "volume_ratio_20", start_idx, end_idx))
            volatility_expansion = finite_mean(
                series_values(features, "volatility_expansion_5_20", start_idx, end_idx)
            )
            volatility_compression = finite_mean(
                series_values(features, "volatility_compression_20_80", start_idx, end_idx)
            )
            range_compression = finite_mean(
                series_values(features, "range_compression_20_80", start_idx, end_idx)
            )
            context_ratio = float(prior_context.iloc[start_idx : end_idx + 1].fillna(0.0).mean())
            interval_score = score_interval(
                probability_mean=probability_mean,
                probability_max=probability_max,
                context_ratio=context_ratio,
                volume_z_max=volume_z_max,
                volatility_expansion=volatility_expansion,
                volatility_compression=volatility_compression,
                confidence_low=interval_confidence_low,
                confidence_high=interval_confidence_high,
            )

            rows.append(
                {
                    "StartBarIdx": int(start_idx),
                    "EndBarIdx": int(end_idx),
                    "PeakBarIdx": int(peak_idx),
                    "ProbabilityPeakBarIdx": int(peak_idx),
                    "TurnBarIdx": int(bottom_idx),
                    "SwitchKind": "valley",
                    "StartDate": data.index[start_idx],
                    "EndDate": data.index[end_idx],
                    "PeakDate": peak_timestamp,
                    "ProbabilityPeakDate": peak_timestamp,
                    "TurnDate": data.index[bottom_idx],
                    "Bars": int(end_idx - start_idx + 1),
                    "SupportLow": support_low,
                    "ResistanceHigh": resistance_high,
                    "BoxLow": support_low - price_pad,
                    "BoxHigh": resistance_high + price_pad,
                    "EntryBarIdx": int(end_idx),
                    "EntryDate": data.index[end_idx],
                    "EntryPrice": entry_price,
                    "TargetPrice": target_price,
                    "StopPrice": stop_price,
                    "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                    "ReturnForecastPctMean": forecast_mean,
                    "ReturnForecastPctMax": forecast_max,
                    "ReturnForecastPctPeak": forecast_peak,
                    "ForecastScoreMean": probability_mean,
                    "ForecastScoreMax": probability_max,
                    "ProbabilityMean": probability_mean,
                    "ProbabilityMax": probability_max,
                    "ProbabilityConfidenceLow": interval_confidence_low,
                    "ProbabilityConfidenceHigh": interval_confidence_high,
                    "SignalThreshold": float(threshold),
                    "VolumeZMax": volume_z_max,
                    "VolumeRatioMax": volume_ratio_max,
                    "VolatilityExpansionMean": volatility_expansion,
                    "VolatilityCompressionMean": volatility_compression,
                    "RangeCompressionMean": range_compression,
                    "PriorDownContextRatio": context_ratio,
                    "IntervalScore": interval_score,
                    "OutcomeKnown": is_finite_number(realized_label),
                    "OutcomeLabel": realized_label,
                    "HitBarIdx": int(hit_idx) if is_finite_number(hit_idx) else np.nan,
                    "HitDate": peak_label.get("label_hit_date", pd.NaT),
                    "HitType": peak_label.get("label_hit_type", None),
                    "ShapeHint": classify_shape_hint(data, peak_idx),
                    "TopDrivers": top_drivers,
                }
            )

    if not rows:
        return empty_interval_frame()
    rows = suppress_overlapping_interval_rows(rows)
    return pd.DataFrame(rows).sort_values("StartBarIdx").reset_index(drop=True)


def forecast_breakout_filter(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    forecast_score: pd.Series,
    return_forecast: pd.Series,
    threshold: float,
    floor: float,
    config: ReversalMLConfig,
) -> pd.Series:
    close = data["close"].astype(float).reindex(features.index)
    high = data["high"].astype(float).reindex(features.index)
    low = data["low"].astype(float).reindex(features.index)
    atr = features["atr"].astype(float).reindex(features.index)
    lookback = max(2, int(config.breakout_lookback))
    cross_lookback = max(1, int(config.forecast_cross_lookback))

    prior_resistance = high.rolling(lookback, min_periods=max(2, lookback // 2)).max().shift(1)
    reclaim_buffer = max(0.0, float(config.breakout_buffer_atr)) * atr
    resistance_reclaim = close >= (prior_resistance - reclaim_buffer)

    reclaim_column = f"reclaim_from_low_atr_{8 if 'reclaim_from_low_atr_8' in features.columns else 5}"
    low_position_column = f"low_range_position_{13 if 'low_range_position_13' in features.columns else 8}"
    valley_rebound = features.get(reclaim_column, pd.Series(np.nan, index=features.index)).astype(float) >= (
        config.switch_min_rebound_atr * 0.75
    )
    near_low = (
        features.get(
            f"dist_to_low_atr_{config.label_lookback}",
            pd.Series(np.nan, index=features.index),
        ).astype(float)
        <= config.max_distance_to_low_atr * 1.25
    ) | (
        features.get(low_position_column, pd.Series(np.nan, index=features.index)).astype(float) <= 0.65
    )

    score_cross = (forecast_score >= threshold) & (forecast_score.shift(cross_lookback) <= floor)
    score_rising = forecast_score.diff(cross_lookback) >= float(config.forecast_min_score_change)
    forecast_turning_positive = (
        (return_forecast > 0)
        | (return_forecast.diff(cross_lookback) > 0)
        | score_cross
    )
    return (
        (resistance_reclaim | valley_rebound | score_cross | score_rising)
        & near_low.fillna(False)
        & forecast_turning_positive.fillna(False)
    )


def empty_signal_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "ReturnForecastPct",
            "ForecastScore",
            "ReversalProbability",
            "SignalThreshold",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "SwitchKind",
            "SwitchTurnBarIdx",
            "SwitchTurnDate",
            "SwitchConfirmBarIdx",
            "SwitchConfirmDate",
            "BarsFromSwitchTurn",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def empty_interval_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "StartBarIdx",
            "EndBarIdx",
            "PeakBarIdx",
            "ProbabilityPeakBarIdx",
            "TurnBarIdx",
            "SwitchKind",
            "StartDate",
            "EndDate",
            "PeakDate",
            "ProbabilityPeakDate",
            "TurnDate",
            "Bars",
            "SupportLow",
            "ResistanceHigh",
            "BoxLow",
            "BoxHigh",
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "ReturnForecastPctMean",
            "ReturnForecastPctMax",
            "ReturnForecastPctPeak",
            "ForecastScoreMean",
            "ForecastScoreMax",
            "ProbabilityMean",
            "ProbabilityMax",
            "ProbabilityConfidenceLow",
            "ProbabilityConfidenceHigh",
            "SignalThreshold",
            "VolumeZMax",
            "VolumeRatioMax",
            "VolatilityExpansionMean",
            "VolatilityCompressionMean",
            "RangeCompressionMean",
            "PriorDownContextRatio",
            "IntervalScore",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def non_max_suppression_1d(
    probability: pd.Series,
    *,
    cooldown_bars: int,
    all_index: pd.Index,
) -> list[pd.Timestamp]:
    ranked = probability.sort_values(ascending=False)
    selected_positions: list[int] = []
    selected_timestamps: list[pd.Timestamp] = []

    for timestamp, _ in ranked.items():
        position = int(all_index.get_loc(timestamp))
        if any(abs(position - selected) <= cooldown_bars for selected in selected_positions):
            continue
        selected_positions.append(position)
        selected_timestamps.append(timestamp)

    selected_timestamps.sort(key=lambda ts: int(all_index.get_loc(ts)))
    return selected_timestamps


def probability_confidence_frame(probability: pd.Series, *, window: int, z: float) -> pd.DataFrame:
    values = probability.astype(float)
    window = max(2, int(window))
    mean = values.rolling(window, min_periods=max(2, window // 4)).mean()
    std = values.rolling(window, min_periods=max(2, window // 4)).std().fillna(0.0)
    count = values.rolling(window, min_periods=max(2, window // 4)).count().replace(0, np.nan)
    stderr = std / np.sqrt(count)
    return pd.DataFrame(
        {
            "mean": mean,
            "lower": (mean - float(z) * stderr).clip(0.0, 1.0),
            "upper": (mean + float(z) * stderr).clip(0.0, 1.0),
        },
        index=probability.index,
    )


def probability_interval_bounds(probability: pd.Series, *, z: float) -> tuple[float, float]:
    values = probability.dropna().to_numpy(dtype=float)
    if len(values) == 0:
        return np.nan, np.nan
    mean = float(np.mean(values))
    if len(values) == 1:
        return mean, mean
    stderr = float(np.std(values, ddof=1) / math.sqrt(len(values)))
    return max(0.0, mean - float(z) * stderr), min(1.0, mean + float(z) * stderr)


def boolean_segments(mask: np.ndarray) -> list[tuple[int, int]]:
    segments: list[tuple[int, int]] = []
    start: int | None = None
    for idx, active in enumerate(mask):
        if active and start is None:
            start = idx
        elif not active and start is not None:
            segments.append((start, idx - 1))
            start = None
    if start is not None:
        segments.append((start, len(mask) - 1))
    return segments


def merge_boolean_segments(
    segments: list[tuple[int, int]],
    *,
    max_gap: int,
) -> list[tuple[int, int]]:
    if not segments:
        return []
    merged: list[tuple[int, int]] = [segments[0]]
    for start, end in segments[1:]:
        previous_start, previous_end = merged[-1]
        if start - previous_end - 1 <= max_gap:
            merged[-1] = (previous_start, end)
        else:
            merged.append((start, end))
    return merged


def split_segment(
    start: int,
    end: int,
    *,
    max_bars: int,
    min_bars: int,
) -> list[tuple[int, int]]:
    length = end - start + 1
    if length <= max_bars:
        return [(start, end)]

    chunks: list[tuple[int, int]] = []
    overlap = max(min_bars, max_bars // 4)
    chunk_start = start
    while chunk_start <= end:
        chunk_end = min(end, chunk_start + max_bars - 1)
        chunks.append((chunk_start, chunk_end))
        if chunk_end >= end:
            break
        chunk_start = max(chunk_start + 1, chunk_end - overlap + 1)
    return chunks


def series_values(features: pd.DataFrame, column: str, start_idx: int, end_idx: int) -> np.ndarray:
    if column not in features.columns:
        return np.array([], dtype=float)
    return features[column].iloc[start_idx : end_idx + 1].to_numpy(dtype=float)


def is_finite_number(value: Any) -> bool:
    try:
        return bool(np.isfinite(float(value)))
    except (TypeError, ValueError):
        return False


def finite_mean(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.mean(array))


def finite_min(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.min(array))


def finite_max(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.max(array))


def score_interval(
    *,
    probability_mean: float,
    probability_max: float,
    context_ratio: float,
    volume_z_max: float,
    volatility_expansion: float,
    volatility_compression: float,
    confidence_low: float,
    confidence_high: float,
) -> float:
    volume_score = np.tanh(max(0.0, 0.0 if not np.isfinite(volume_z_max) else volume_z_max) / 3.0)
    expansion_score = np.tanh(
        max(0.0, (0.0 if not np.isfinite(volatility_expansion) else volatility_expansion) - 1.0)
    )
    compression_score = 1.0 - min(
        1.0,
        max(0.0, 0.0 if not np.isfinite(volatility_compression) else volatility_compression),
    )
    confidence_width = max(
        0.0,
        (0.0 if not np.isfinite(confidence_high) else confidence_high)
        - (0.0 if not np.isfinite(confidence_low) else confidence_low),
    )
    stability_score = 1.0 - min(1.0, confidence_width)
    score = (
        0.38 * probability_max
        + 0.27 * probability_mean
        + 0.15 * min(1.0, max(0.0, context_ratio))
        + 0.07 * volume_score
        + 0.06 * expansion_score
        + 0.04 * compression_score
        + 0.03 * stability_score
    )
    return float(score)


def suppress_overlapping_interval_rows(
    rows: list[dict[str, Any]],
    *,
    max_overlap_ratio: float = 0.55,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []
    for row in sorted(rows, key=lambda item: float(item["IntervalScore"]), reverse=True):
        start = int(row["StartBarIdx"])
        end = int(row["EndBarIdx"])
        length = max(1, end - start + 1)
        overlaps = False
        for existing in selected:
            existing_start = int(existing["StartBarIdx"])
            existing_end = int(existing["EndBarIdx"])
            existing_length = max(1, existing_end - existing_start + 1)
            overlap = max(0, min(end, existing_end) - max(start, existing_start) + 1)
            if overlap / min(length, existing_length) >= max_overlap_ratio:
                overlaps = True
                break
        if not overlaps:
            selected.append(row)
    return selected


def make_train_feature_stats(train_features: pd.DataFrame) -> pd.DataFrame:
    stats = pd.DataFrame(index=train_features.columns)
    stats["median"] = train_features.median(numeric_only=True)
    q25 = train_features.quantile(0.25, numeric_only=True)
    q75 = train_features.quantile(0.75, numeric_only=True)
    stats["iqr"] = (q75 - q25).replace(0, np.nan)
    return stats


def local_driver_summary(
    row: pd.Series,
    feature_importance: pd.DataFrame,
    train_stats: pd.DataFrame,
    *,
    top_n: int,
) -> str:
    if feature_importance.empty:
        return ""

    importance_map = feature_importance.set_index("feature")["importance_mean"].to_dict()
    scored: list[tuple[float, str]] = []
    for feature, value in row.items():
        if feature not in importance_map or feature not in train_stats.index:
            continue
        if not np.isfinite(value):
            continue
        iqr = train_stats.loc[feature, "iqr"]
        median = train_stats.loc[feature, "median"]
        if not np.isfinite(iqr) or iqr == 0:
            continue
        robust_z = (float(value) - float(median)) / float(iqr)
        score = abs(float(importance_map[feature]) * robust_z)
        if score <= 0:
            continue
        scored.append((score, f"{feature}={float(value):.4g} z={robust_z:.2f}"))

    scored.sort(reverse=True)
    return " | ".join(text for _, text in scored[:top_n])


def classify_shape_hint(data: pd.DataFrame, bar_idx: int, *, lookback: int = 80) -> str:
    start = max(0, int(bar_idx) - lookback)
    if int(bar_idx) - start < 20:
        return "ML Bullish Reversal"

    window = data.iloc[start : int(bar_idx) + 1]
    lows = window["low"].to_numpy(dtype=float)
    highs = window["high"].to_numpy(dtype=float)
    close = window["close"].to_numpy(dtype=float)
    atr = float(average_true_range(window, period=14).iloc[-1])
    if not np.isfinite(atr) or atr <= 0:
        atr = max(float(np.nanstd(close)), 1e-12)

    pivot_lows = simple_pivots(lows, mode="low", order=3)
    pivot_highs = simple_pivots(highs, mode="high", order=3)
    if len(pivot_lows) >= 2:
        l1, l2 = pivot_lows[-2], pivot_lows[-1]
        low_diff_atr = abs(lows[l2] - lows[l1]) / atr
        if l2 > l1 and low_diff_atr <= 1.0:
            between_high = np.max(highs[l1:l2 + 1])
            if between_high - max(lows[l1], lows[l2]) >= atr:
                return "ML Double Bottom-like"

    if len(pivot_lows) >= 3 and len(pivot_highs) >= 2:
        left, head, right = pivot_lows[-3], pivot_lows[-2], pivot_lows[-1]
        if left < head < right and lows[head] < min(lows[left], lows[right]):
            shoulder_diff_atr = abs(lows[left] - lows[right]) / atr
            if shoulder_diff_atr <= 1.5:
                return "ML Inverse H&S-like"

    return "ML Bullish Reversal"


def simple_pivots(values: np.ndarray, *, mode: str, order: int) -> list[int]:
    pivots: list[int] = []
    for idx in range(order, len(values) - order):
        left = values[idx - order : idx]
        right = values[idx + 1 : idx + order + 1]
        if mode == "low" and values[idx] <= np.min(left) and values[idx] <= np.min(right):
            pivots.append(idx)
        if mode == "high" and values[idx] >= np.max(left) and values[idx] >= np.max(right):
            pivots.append(idx)
    return pivots


def fit_winsorize(
    X_train: pd.DataFrame,
    *,
    lower: float = 0.01,
    upper: float = 0.99,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    bounds = pd.DataFrame(index=X_train.columns)
    bounds["lower"] = X_train.quantile(lower, numeric_only=True)
    bounds["upper"] = X_train.quantile(upper, numeric_only=True)
    return apply_winsorize(X_train, bounds), bounds


def apply_winsorize(X: pd.DataFrame, bounds: pd.DataFrame) -> pd.DataFrame:
    out = X.copy()
    for column in out.columns:
        if column not in bounds.index:
            continue
        lower = bounds.loc[column, "lower"]
        upper = bounds.loc[column, "upper"]
        if np.isfinite(lower) and np.isfinite(upper) and lower < upper:
            out[column] = out[column].clip(lower=lower, upper=upper)
    return out


def average_true_range(data: pd.DataFrame, *, period: int = 14) -> pd.Series:
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    close = data["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [
            high - low,
            (high - previous_close).abs(),
            (low - previous_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    return true_range.rolling(period, min_periods=1).mean()


def rsi(close: pd.Series, *, period: int = 14) -> pd.Series:
    delta = close.diff()
    gains = delta.clip(lower=0.0)
    losses = -delta.clip(upper=0.0)
    avg_gain = gains.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    avg_loss = losses.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    rs = avg_gain / avg_loss.replace(0, np.nan)
    return 100.0 - (100.0 / (1.0 + rs))


def macd(close: pd.Series) -> tuple[pd.Series, pd.Series, pd.Series]:
    fast = close.ewm(span=12, adjust=False).mean()
    slow = close.ewm(span=26, adjust=False).mean()
    line = fast - slow
    signal = line.ewm(span=9, adjust=False).mean()
    return line, signal, line - signal


def stochastic_k(
    high: pd.Series,
    low: pd.Series,
    close: pd.Series,
    *,
    period: int,
) -> pd.Series:
    lowest = low.rolling(period).min()
    highest = high.rolling(period).max()
    return (close - lowest) / (highest - lowest).replace(0, np.nan) * 100.0


def rolling_zscore(series: pd.Series, *, window: int) -> pd.Series:
    mean = series.rolling(window).mean()
    std = series.rolling(window).std()
    return (series - mean) / std.replace(0, np.nan)


def rolling_slope_pct(series: pd.Series, *, window: int) -> pd.Series:
    x = np.arange(window, dtype=float)
    x = x - x.mean()

    def slope(values: np.ndarray) -> float:
        if np.isnan(values).any():
            return np.nan
        y = values.astype(float)
        y = y - y.mean()
        denom = float(np.dot(x, x))
        if denom == 0:
            return np.nan
        raw_slope = float(np.dot(x, y) / denom)
        return raw_slope / max(abs(float(values[-1])), 1e-12) * 100.0

    return series.rolling(window).apply(slope, raw=True)


def signed_streak(condition: pd.Series) -> pd.Series:
    values = condition.fillna(False).to_numpy(dtype=bool)
    streaks = np.zeros(len(values), dtype=float)
    current = 0
    previous = None
    for idx, value in enumerate(values):
        if previous is None or value != previous:
            current = 1
        else:
            current += 1
        streaks[idx] = current if value else -current
        previous = value
    return pd.Series(streaks, index=condition.index)


def on_balance_volume(close: pd.Series, volume: pd.Series) -> pd.Series:
    direction = np.sign(close.diff()).fillna(0.0)
    return (direction * volume).cumsum()


def walk_forward_arima_features(
    returns: pd.Series,
    *,
    order: tuple[int, int, int],
    window: int,
) -> pd.DataFrame:
    columns = ["arima_forecast_return", "arima_residual", "arima_residual_z_40"]
    out = pd.DataFrame(np.nan, index=returns.index, columns=columns)
    if ARIMA is None:
        warnings.warn("statsmodels unavailable; ARIMA features skipped.")
        return out

    values = returns.astype(float)
    for idx in range(window, len(values)):
        history = values.iloc[idx - window : idx].dropna()
        if len(history) < max(30, sum(order) * 10):
            continue
        try:
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                result = ARIMA(history, order=order).fit()
            forecast = float(result.forecast(1).iloc[0])
            residual = float(values.iloc[idx] - forecast)
            out.iloc[idx, out.columns.get_loc("arima_forecast_return")] = forecast
            out.iloc[idx, out.columns.get_loc("arima_residual")] = residual
        except Exception:
            continue
    out["arima_residual_z_40"] = rolling_zscore(out["arima_residual"], window=40)
    return out


def save_outputs(
    data: pd.DataFrame,
    result: dict[str, Any],
    config: ReversalMLConfig,
) -> Path:
    run_name = f"{config.symbol.replace('.', '_')}_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
    run_dir = Path(config.output_dir) / run_name
    run_dir.mkdir(parents=True, exist_ok=True)
    if go is None:
        warnings.warn("plotly unavailable; HTML plot files skipped. Install with: pip install plotly")

    result["signals"].to_csv(run_dir / "signals.csv", index=False)
    result["intervals"].to_csv(run_dir / "pattern_intervals.csv", index=False)
    result["model_report"].to_csv(run_dir / "model_report.csv", index=False)
    result["feature_importance"].to_csv(run_dir / "feature_importance.csv", index=False)
    result["hmm_state_summary"].to_csv(run_dir / "hmm_state_summary.csv", index=False)
    result["pca_loadings"].to_csv(run_dir / "pca_loadings.csv", index=False)

    probability_frame = pd.DataFrame(
        {
            "Date": result["probabilities"].index,
            "ForecastScore": result["probabilities"].to_numpy(dtype=float),
            "ReversalProbability": result["probabilities"].to_numpy(dtype=float),
            "ReturnForecastPct": result["forecasts"].reindex(result["probabilities"].index).to_numpy(dtype=float),
        }
    )
    probability_frame.to_csv(run_dir / "probabilities.csv", index=False)
    probability_frame.to_csv(run_dir / "return_forecasts.csv", index=False)

    all_probability_frame = pd.DataFrame({"Date": data.index})
    for model_name, probabilities in result["per_model_probabilities"].items():
        all_probability_frame[f"{model_name}_forecast_score"] = probabilities.reindex(data.index).to_numpy(dtype=float)
        all_probability_frame[f"{model_name}_return_forecast_pct"] = (
            result["per_model_forecasts"][model_name].reindex(data.index).to_numpy(dtype=float)
        )
    all_probability_frame.to_csv(run_dir / "all_model_probabilities.csv", index=False)
    all_probability_frame.to_csv(run_dir / "all_model_return_forecasts.csv", index=False)

    hmm_regimes = make_hmm_regime_frame(result["features"])
    hmm_regimes.to_csv(run_dir / "hmm_regimes.csv", index=False)

    all_interval_frames = [
        intervals
        for intervals in result["per_model_intervals"].values()
        if intervals is not None and not intervals.empty
    ]
    if all_interval_frames:
        pd.concat(all_interval_frames, ignore_index=True).to_csv(
            run_dir / "all_model_pattern_intervals.csv",
            index=False,
        )
    else:
        empty_interval_frame().to_csv(run_dir / "all_model_pattern_intervals.csv", index=False)

    models_dir = run_dir / "models"
    models_dir.mkdir(exist_ok=True)
    for model_name, probabilities in result["per_model_probabilities"].items():
        safe_model_name = safe_filename(model_name)
        forecasts = result["per_model_forecasts"][model_name]
        signals = result["per_model_signals"].get(model_name, empty_signal_frame())
        importance = result["per_model_feature_importance"].get(
            model_name,
            pd.DataFrame(columns=["feature", "importance_mean", "importance_std", "rank"]),
        )
        threshold = result["per_model_thresholds"].get(model_name, np.nan)
        intervals = result["per_model_intervals"].get(model_name, empty_interval_frame())
        signals.to_csv(models_dir / f"{safe_model_name}_signals.csv", index=False)
        intervals.to_csv(models_dir / f"{safe_model_name}_pattern_intervals.csv", index=False)
        importance.to_csv(models_dir / f"{safe_model_name}_feature_importance.csv", index=False)
        pd.DataFrame(
            {
                "Date": data.index,
                "ForecastScore": probabilities.reindex(data.index).to_numpy(dtype=float),
                "ReturnForecastPct": forecasts.reindex(data.index).to_numpy(dtype=float),
            }
        ).to_csv(models_dir / f"{safe_model_name}_return_forecast.csv", index=False)
        make_plot(
            data=data,
            probabilities=probabilities,
            return_forecasts=forecasts,
            signals=signals,
            intervals=intervals,
            title=f"{config.symbol} {model_name} ML Reversal Signals",
            output_path=models_dir / f"{safe_model_name}_plot.html",
            model_name=model_name,
            threshold=threshold,
            confidence_window=config.confidence_window,
            confidence_z=config.confidence_z,
        )
        make_feature_importance_plot(
            importance=importance,
            title=f"{config.symbol} {model_name} XAI Feature Importance",
            output_path=models_dir / f"{safe_model_name}_feature_importance.html",
        )

    with open(run_dir / "config.json", "w", encoding="utf-8") as handle:
        json.dump(
            {
                **asdict(config),
                "best_model_name": result["best_model_name"],
                "threshold": result["threshold"],
                "return_threshold_pct": result["return_threshold_pct"],
                "forecast_score_scale": result["forecast_score_scale"],
            },
            handle,
            indent=2,
            default=str,
        )

    plot_path = run_dir / "reversal_ml_plot.html"
    make_plot(
        data=data,
        probabilities=result["probabilities"],
        return_forecasts=result["forecasts"],
        signals=result["signals"],
        intervals=result["intervals"],
        title=f"{config.symbol} ML Reversal Signals",
        output_path=plot_path,
        model_name=f"best: {result['best_model_name']}",
        threshold=result["threshold"],
        confidence_window=config.confidence_window,
        confidence_z=config.confidence_z,
    )
    make_model_comparison_plot(
        data=data,
        per_model_probabilities=result["per_model_probabilities"],
        per_model_forecasts=result["per_model_forecasts"],
        per_model_signals=result["per_model_signals"],
        per_model_intervals=result["per_model_intervals"],
        per_model_thresholds=result["per_model_thresholds"],
        title=f"{config.symbol} ML Reversal Model Comparison",
        output_path=run_dir / "all_models_comparison_plot.html",
    )
    make_hmm_regime_plot(
        data=data,
        features=result["features"],
        state_summary=result["hmm_state_summary"],
        title=f"{config.symbol} HMM Regime Map",
        output_path=run_dir / "hmm_regime_plot.html",
    )
    return run_dir


def safe_filename(value: str) -> str:
    safe = "".join(character if character.isalnum() or character in {"-", "_"} else "_" for character in value)
    return safe.strip("_") or "model"


def hex_to_rgba(hex_color: str, alpha: float) -> str:
    value = hex_color.lstrip("#")
    if len(value) != 6:
        return f"rgba(56, 189, 248, {alpha})"
    red, green, blue = (int(value[idx : idx + 2], 16) for idx in (0, 2, 4))
    return f"rgba({red}, {green}, {blue}, {alpha})"


def add_interval_boxes(
    fig: Any,
    intervals: pd.DataFrame | None,
    *,
    color: str,
    label: str,
    opacity: float,
    show_text: bool,
) -> None:
    if intervals is None or intervals.empty:
        return

    center_x: list[float] = []
    center_y: list[float] = []
    hover_text: list[str] = []
    visible_text: list[str] = []
    for interval_number, (_, row) in enumerate(intervals.iterrows(), start=1):
        try:
            start_idx = int(row["StartBarIdx"])
            end_idx = int(row["EndBarIdx"])
            box_low = float(row["BoxLow"])
            box_high = float(row["BoxHigh"])
        except (KeyError, TypeError, ValueError):
            continue
        if not all(np.isfinite(value) for value in [start_idx, end_idx, box_low, box_high]):
            continue

        fig.add_shape(
            type="rect",
            xref="x",
            yref="y",
            x0=start_idx - 0.5,
            x1=end_idx + 0.5,
            y0=box_low,
            y1=box_high,
            fillcolor=hex_to_rgba(color, opacity),
            line=dict(color=color, width=2),
            layer="below",
        )
        center_x.append((start_idx + end_idx) / 2)
        center_y.append(box_high)
        probability_max = float(row.get("ProbabilityMax", np.nan))
        probability_mean = float(row.get("ProbabilityMean", np.nan))
        visible_text.append(f"Box {interval_number} {probability_max:.2f}")
        hover_text.append(
            (
                f"{label} interval<br>"
                f"Start: {row.get('StartDate')}<br>"
                f"End: {row.get('EndDate')}<br>"
                f"Turn: {row.get('TurnDate')}<br>"
                f"Probability peak: {row.get('ProbabilityPeakDate', row.get('PeakDate'))}<br>"
                f"Switch kind: {row.get('SwitchKind')}<br>"
                f"Return forecast peak: {float(row.get('ReturnForecastPctPeak', np.nan)):.3f}%<br>"
                f"Return forecast mean: {float(row.get('ReturnForecastPctMean', np.nan)):.3f}%<br>"
                f"Mean forecast score: {probability_mean:.3f}<br>"
                f"Max forecast score: {probability_max:.3f}<br>"
                f"Score CI: {float(row.get('ProbabilityConfidenceLow', np.nan)):.3f}"
                f" - {float(row.get('ProbabilityConfidenceHigh', np.nan)):.3f}<br>"
                f"Volume z max: {float(row.get('VolumeZMax', np.nan)):.3f}<br>"
                f"Vol expansion: {float(row.get('VolatilityExpansionMean', np.nan)):.3f}<br>"
                f"Score: {float(row.get('IntervalScore', np.nan)):.3f}<br>"
                f"Shape: {row.get('ShapeHint')}<br>"
                f"Drivers: {row.get('TopDrivers')}"
            )
        )

    if not center_x:
        return
    mode = "markers+text" if show_text else "markers"
    fig.add_trace(
        go.Scatter(
            x=center_x,
            y=center_y,
            mode=mode,
            name=f"{label} pattern interval",
            marker=dict(
                color=color,
                size=11,
                symbol="square-open",
                line=dict(width=2),
            ),
            text=visible_text if show_text else None,
            textposition="top center",
            hovertext=hover_text,
            hoverinfo="text",
        )
    )


def make_plot(
    *,
    data: pd.DataFrame,
    probabilities: pd.Series,
    return_forecasts: pd.Series | None,
    signals: pd.DataFrame,
    intervals: pd.DataFrame | None = None,
    title: str,
    output_path: Path | None,
    model_name: str | None = None,
    threshold: float | None = None,
    confidence_window: int = 20,
    confidence_z: float = 1.0,
) -> Any | None:
    if go is None:
        return None

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )
    add_interval_boxes(
        fig,
        intervals,
        color="#FACC15",
        label=model_name or "ML",
        opacity=0.16,
        show_text=True,
    )
    aligned_probability = probabilities.reindex(data.index).astype(float)
    probability_ci = probability_confidence_frame(
        aligned_probability,
        window=confidence_window,
        z=confidence_z,
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probability_ci["lower"].to_numpy(dtype=float),
            name="Forecast score CI lower",
            mode="lines",
            yaxis="y2",
            line=dict(width=0),
            hoverinfo="skip",
            showlegend=False,
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probability_ci["upper"].to_numpy(dtype=float),
            name="Forecast score confidence band",
            mode="lines",
            yaxis="y2",
            fill="tonexty",
            fillcolor="rgba(56, 189, 248, 0.14)",
            line=dict(width=0),
            hoverinfo="skip",
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=aligned_probability.to_numpy(dtype=float),
            name=f"{model_name or 'Model'} forecast score",
            mode="lines",
            yaxis="y2",
            line=dict(color="#38BDF8", width=2),
        )
    )
    if return_forecasts is not None:
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=return_forecasts.reindex(data.index).to_numpy(dtype=float),
                name=f"{model_name or 'Model'} return forecast %",
                mode="lines",
                yaxis="y3",
                line=dict(color="#F97316", width=2),
            )
        )
    if threshold is not None and np.isfinite(float(threshold)):
        fig.add_trace(
            go.Scatter(
                x=[plot_x[0], plot_x[-1]],
                y=[float(threshold), float(threshold)],
                name="Signal threshold",
                mode="lines",
                yaxis="y2",
                line=dict(color="#FACC15", width=1, dash="dash"),
                hoverinfo="skip",
            )
        )

    if not signals.empty:
        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers+text",
                name=f"{model_name or 'ML'} Forecast Breakout",
                marker=dict(color="#FACC15", size=12, symbol="circle"),
                text=[f"{prob:.2f}" for prob in signals["ReversalProbability"]],
                textposition="top center",
                hovertext=[
                    (
                        f"ML Buy<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Forecast score: {row['ForecastScore']:.3f}<br>"
                        f"Return forecast: {row['ReturnForecastPct']:.3f}%<br>"
                        f"Switch: {row.get('SwitchKind')} at {row.get('SwitchTurnDate')}<br>"
                        f"Bars from turn: {row.get('BarsFromSwitchTurn')}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Forecast Score",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        yaxis3=dict(
            title="Return Forecast %",
            overlaying="y",
            side="right",
            anchor="free",
            position=0.98,
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=110, t=80, b=55),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    if output_path is not None:
        fig.write_html(str(output_path), include_plotlyjs="cdn")
    return fig


def make_feature_importance_plot(
    *,
    importance: pd.DataFrame,
    title: str,
    output_path: Path | None,
    top_n: int = 25,
) -> Any | None:
    if go is None or importance.empty:
        return None

    plot_frame = importance.head(top_n).copy()
    plot_frame = plot_frame.sort_values("importance_mean", ascending=True)
    fig = go.Figure()
    fig.add_trace(
        go.Bar(
            x=plot_frame["importance_mean"],
            y=plot_frame["feature"],
            orientation="h",
            error_x=dict(
                type="data",
                array=plot_frame["importance_std"].clip(lower=0).to_numpy(dtype=float),
                visible=True,
                color="#94A3B8",
            ),
            marker=dict(color="#38BDF8"),
            hovertext=[
                (
                    f"Feature: {row['feature']}<br>"
                    f"Importance: {row['importance_mean']:.6f}<br>"
                    f"Std: {row['importance_std']:.6f}<br>"
                    f"Rank: {int(row['rank'])}"
                )
                for _, row in plot_frame.iterrows()
            ],
            hoverinfo="text",
            name="Permutation importance",
        )
    )
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=max(520, 24 * len(plot_frame) + 180),
        xaxis=dict(title="Neg-MAE permutation importance"),
        yaxis=dict(title="Feature"),
        margin=dict(l=180, r=50, t=80, b=60),
    )
    if output_path is not None:
        fig.write_html(str(output_path), include_plotlyjs="cdn")
    return fig


def make_hmm_regime_frame(features: pd.DataFrame) -> pd.DataFrame:
    hmm_columns = [
        column
        for column in features.columns
        if column == "hmm_state" or column.startswith("hmm_state_prob_")
    ]
    if not hmm_columns:
        return pd.DataFrame(columns=["Date", "hmm_state"])
    frame = features[hmm_columns].copy()
    frame.insert(0, "Date", features.index)
    return frame


def make_hmm_regime_plot(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    state_summary: pd.DataFrame,
    title: str,
    output_path: Path | None,
) -> Any | None:
    if go is None or "hmm_state" not in features.columns:
        return None

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    states = features["hmm_state"].reindex(data.index)
    state_values = states.to_numpy(dtype=float)
    finite_states = state_values[np.isfinite(state_values)]
    if len(finite_states) == 0:
        return None

    colors = ["#38BDF8", "#FACC15", "#A78BFA", "#34D399", "#FB7185", "#F97316", "#22D3EE"]
    fig = go.Figure()
    for start_idx, end_idx, state in hmm_state_segments(state_values):
        color = colors[int(state) % len(colors)]
        fig.add_shape(
            type="rect",
            xref="x",
            yref="paper",
            x0=start_idx - 0.5,
            x1=end_idx + 0.5,
            y0=0,
            y1=1,
            fillcolor=hex_to_rgba(color, 0.10),
            line=dict(width=0),
            layer="below",
        )

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=state_values,
            name="HMM state",
            mode="lines",
            yaxis="y2",
            line=dict(color="#FACC15", width=2, shape="hv"),
        )
    )

    probability_columns = [
        column for column in features.columns if column.startswith("hmm_state_prob_")
    ]
    if probability_columns:
        max_probability = features[probability_columns].max(axis=1).reindex(data.index)
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=max_probability.to_numpy(dtype=float),
                name="HMM max probability",
                mode="lines",
                yaxis="y3",
                line=dict(color="#38BDF8", width=2),
            )
        )

    if state_summary is not None and not state_summary.empty:
        annotations = []
        summary = state_summary.copy()
        for _, row in summary.iterrows():
            annotations.append(
                f"S{int(row['state'])}: n={int(row['train_count'])}, "
                f"ret={float(row['mean_return']):.4g}, "
                f"vol={float(row['mean_vol']):.4g}, "
                f"dd={float(row['mean_drawdown']):.3g}"
            )
        fig.add_annotation(
            text="<br>".join(annotations),
            xref="paper",
            yref="paper",
            x=0.01,
            y=0.98,
            align="left",
            showarrow=False,
            bgcolor="rgba(15, 23, 42, 0.70)",
            bordercolor="#334155",
            font=dict(size=11),
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="HMM State",
            overlaying="y",
            side="right",
            showgrid=False,
        ),
        yaxis3=dict(
            title="State Confidence",
            overlaying="y",
            side="right",
            anchor="free",
            position=0.98,
            range=[0, 1],
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=110, t=90, b=60),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    if output_path is not None:
        fig.write_html(str(output_path), include_plotlyjs="cdn")
    return fig


def hmm_state_segments(state_values: np.ndarray) -> list[tuple[int, int, int]]:
    segments: list[tuple[int, int, int]] = []
    start_idx: int | None = None
    active_state: int | None = None
    for idx, state in enumerate(state_values):
        if not np.isfinite(state):
            if start_idx is not None and active_state is not None:
                segments.append((start_idx, idx - 1, active_state))
            start_idx = None
            active_state = None
            continue
        state_int = int(state)
        if active_state is None:
            start_idx = idx
            active_state = state_int
            continue
        if state_int != active_state:
            segments.append((int(start_idx), idx - 1, active_state))
            start_idx = idx
            active_state = state_int
    if start_idx is not None and active_state is not None:
        segments.append((start_idx, len(state_values) - 1, active_state))
    return segments


def make_model_comparison_plot(
    *,
    data: pd.DataFrame,
    per_model_probabilities: dict[str, pd.Series],
    per_model_forecasts: dict[str, pd.Series],
    per_model_signals: dict[str, pd.DataFrame],
    per_model_intervals: dict[str, pd.DataFrame],
    per_model_thresholds: dict[str, float],
    title: str,
    output_path: Path | None,
) -> Any | None:
    if go is None or not per_model_probabilities:
        return None

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    colors = ["#38BDF8", "#FACC15", "#A78BFA", "#34D399", "#FB7185", "#F97316"]

    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    for model_idx, (model_name, probabilities) in enumerate(per_model_probabilities.items()):
        color = colors[model_idx % len(colors)]
        threshold = per_model_thresholds.get(model_name, np.nan)
        add_interval_boxes(
            fig,
            per_model_intervals.get(model_name, empty_interval_frame()),
            color=color,
            label=model_name,
            opacity=0.08,
            show_text=False,
        )
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=probabilities.reindex(data.index).to_numpy(dtype=float),
                name=f"{model_name} forecast score",
                mode="lines",
                yaxis="y2",
                line=dict(color=color, width=2),
            )
        )
        if model_name in per_model_forecasts:
            fig.add_trace(
                go.Scatter(
                    x=plot_x,
                    y=per_model_forecasts[model_name].reindex(data.index).to_numpy(dtype=float),
                    name=f"{model_name} return forecast %",
                    mode="lines",
                    yaxis="y3",
                    line=dict(color=color, width=1, dash="dash"),
                    opacity=0.75,
                )
            )
        if threshold is not None and np.isfinite(float(threshold)):
            fig.add_trace(
                go.Scatter(
                    x=[plot_x[0], plot_x[-1]],
                    y=[float(threshold), float(threshold)],
                    name=f"{model_name} threshold",
                    mode="lines",
                    yaxis="y2",
                    line=dict(color=color, width=1, dash="dot"),
                    hoverinfo="skip",
                    showlegend=False,
                )
            )

        signals = per_model_signals.get(model_name, empty_signal_frame())
        if signals is None or signals.empty:
            continue

        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers",
                name=f"{model_name} signals",
                marker=dict(
                    color=color,
                    size=10,
                    symbol="circle-open",
                    line=dict(width=2),
                ),
                hovertext=[
                    (
                        f"Model: {model_name}<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Forecast score: {row['ForecastScore']:.3f}<br>"
                        f"Return forecast: {row['ReturnForecastPct']:.3f}%<br>"
                        f"Threshold: {row['SignalThreshold']:.3f}<br>"
                        f"Switch: {row.get('SwitchKind')} at {row.get('SwitchTurnDate')}<br>"
                        f"Bars from turn: {row.get('BarsFromSwitchTurn')}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=860,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Forecast Score",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        yaxis3=dict(
            title="Return Forecast %",
            overlaying="y",
            side="right",
            anchor="free",
            position=0.98,
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=110, t=90, b=60),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    if output_path is not None:
        fig.write_html(str(output_path), include_plotlyjs="cdn")
    return fig


def run(config: ReversalMLConfig) -> Path:
    analysis = run_analysis(config, save=True)
    print_run_summary(analysis)
    return analysis["run_dir"]


def run_analysis(config: ReversalMLConfig, *, save: bool = True) -> dict[str, Any]:
    data = load_ohlcv(config)
    features = make_feature_frame(data, config)
    labels = make_reversal_labels(data, features, config)
    result = train_reversal_models(data, features, labels, config)
    run_dir = save_outputs(data, result, config) if save else None
    return {
        "config": config,
        "data": data,
        "features": features,
        "labels": labels,
        "result": result,
        "run_dir": run_dir,
        "csv_paths": collect_csv_paths(run_dir, result) if run_dir is not None else {},
        "plot_paths": collect_plot_paths(run_dir, result) if run_dir is not None else {},
        "figures": {},
    }


def run_notebook(
    *,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    csv_path: str | None = None,
    output_dir: str = "outputs/ml_reversal_runs",
    show_plots: bool = True,
    print_outputs: bool = True,
    save: bool = True,
    **overrides: Any,
) -> dict[str, Any]:
    config_values = {
        "symbol": symbol,
        "period": period,
        "interval": interval,
        "csv_path": csv_path,
        "output_dir": output_dir,
    }
    config_values.update(overrides)
    config = ReversalMLConfig(**config_values)
    analysis = run_analysis(config, save=save)
    if print_outputs:
        print_run_summary(analysis)
    if show_plots:
        display_notebook_figures(analysis)
    return analysis


def collect_csv_paths(run_dir: Path | None, result: dict[str, Any]) -> dict[str, Path]:
    if run_dir is None:
        return {}
    paths: dict[str, Path] = {
        "signals": run_dir / "signals.csv",
        "pattern_intervals": run_dir / "pattern_intervals.csv",
        "model_report": run_dir / "model_report.csv",
        "feature_importance": run_dir / "feature_importance.csv",
        "hmm_state_summary": run_dir / "hmm_state_summary.csv",
        "hmm_regimes": run_dir / "hmm_regimes.csv",
        "pca_loadings": run_dir / "pca_loadings.csv",
        "probabilities": run_dir / "probabilities.csv",
        "return_forecasts": run_dir / "return_forecasts.csv",
        "all_model_probabilities": run_dir / "all_model_probabilities.csv",
        "all_model_return_forecasts": run_dir / "all_model_return_forecasts.csv",
        "all_model_pattern_intervals": run_dir / "all_model_pattern_intervals.csv",
    }
    models_dir = run_dir / "models"
    for model_name in result.get("per_model_probabilities", {}):
        safe_model_name = safe_filename(str(model_name))
        paths[f"{model_name}_signals"] = models_dir / f"{safe_model_name}_signals.csv"
        paths[f"{model_name}_pattern_intervals"] = (
            models_dir / f"{safe_model_name}_pattern_intervals.csv"
        )
        paths[f"{model_name}_feature_importance"] = (
            models_dir / f"{safe_model_name}_feature_importance.csv"
        )
        paths[f"{model_name}_return_forecast"] = models_dir / f"{safe_model_name}_return_forecast.csv"
    return paths


def collect_plot_paths(run_dir: Path | None, result: dict[str, Any]) -> dict[str, Path]:
    if run_dir is None:
        return {}
    paths: dict[str, Path] = {
        "best_model": run_dir / "reversal_ml_plot.html",
        "model_comparison": run_dir / "all_models_comparison_plot.html",
        "hmm_regime": run_dir / "hmm_regime_plot.html",
    }
    models_dir = run_dir / "models"
    for model_name in result.get("per_model_probabilities", {}):
        safe_model_name = safe_filename(str(model_name))
        paths[f"{model_name}_plot"] = models_dir / f"{safe_model_name}_plot.html"
        paths[f"{model_name}_feature_importance"] = (
            models_dir / f"{safe_model_name}_feature_importance.html"
        )
    return paths


def print_run_summary(analysis: dict[str, Any]) -> None:
    config = analysis["config"]
    result = analysis["result"]
    labels = analysis.get("labels")
    run_dir = analysis.get("run_dir")

    print("ML reversal run")
    print(f"Run dir: {run_dir if run_dir is not None else 'not saved'}")
    print(
        f"Symbol: {config.symbol} | Period: {config.period} | Interval: {config.interval} | "
        f"HMM: {'on' if config.use_hmm else 'off'}"
    )
    print(
        f"Best model: {result['best_model_name']} | "
        f"Return threshold: {result['return_threshold_pct']:.4f}% | "
        f"Forecast score threshold: {result['threshold']:.4f}"
    )
    print(f"Pattern intervals: {len(result['intervals'])} | Point signals: {len(result['signals'])}")
    if isinstance(labels, pd.DataFrame) and "reversal_label" in labels.columns:
        positive_labels = int(labels["reversal_label"].fillna(0).sum())
        labeled_rows = int(labels["reversal_label"].notna().sum())
        print(f"Switch labels: {positive_labels} positive / {labeled_rows} labeled rows")

    if not result["model_report"].empty:
        print("\nModel report:")
        print(result["model_report"].to_string(index=False))

    if not result["intervals"].empty:
        preview_columns = [
            column
            for column in [
                "StartDate",
                "EndDate",
                "TurnDate",
                "ProbabilityPeakDate",
                "SwitchKind",
                "Bars",
                "ReturnForecastPctPeak",
                "ReturnForecastPctMean",
                "ForecastScoreMax",
                "ForecastScoreMean",
                "IntervalScore",
                "BoxLow",
                "BoxHigh",
                "ShapeHint",
            ]
            if column in result["intervals"].columns
        ]
        print("\nTop pattern intervals:")
        print(
            result["intervals"]
            .sort_values("IntervalScore", ascending=False)
            .head(10)[preview_columns]
            .to_string(index=False)
        )

    csv_paths = analysis.get("csv_paths", {})
    if csv_paths:
        print("\nCSV files:")
        for name, path in csv_paths.items():
            status = "created" if path.exists() else "expected"
            print(f"  {name}: {path} [{status}]")

    plot_paths = analysis.get("plot_paths", {})
    if plot_paths:
        print("\nPlot files:")
        if go is None:
            print("  plotly is not installed, so HTML plot files were skipped.")
        for name, path in plot_paths.items():
            status = "created" if path.exists() else "not created"
            print(f"  {name}: {path} [{status}]")


def build_notebook_figures(analysis: dict[str, Any]) -> dict[str, Any]:
    if go is None:
        warnings.warn("plotly is not installed, so notebook figures cannot be displayed.")
        return {}

    config = analysis["config"]
    data = analysis["data"]
    result = analysis["result"]
    figures: dict[str, Any] = {}

    best_figure = make_plot(
        data=data,
        probabilities=result["probabilities"],
        return_forecasts=result["forecasts"],
        signals=result["signals"],
        intervals=result["intervals"],
        title=f"{config.symbol} ML Return Forecast Breakouts",
        output_path=None,
        model_name=f"best: {result['best_model_name']}",
        threshold=result["threshold"],
        confidence_window=config.confidence_window,
        confidence_z=config.confidence_z,
    )
    if best_figure is not None:
        figures["best_model"] = best_figure

    comparison_figure = make_model_comparison_plot(
        data=data,
        per_model_probabilities=result["per_model_probabilities"],
        per_model_forecasts=result["per_model_forecasts"],
        per_model_signals=result["per_model_signals"],
        per_model_intervals=result["per_model_intervals"],
        per_model_thresholds=result["per_model_thresholds"],
        title=f"{config.symbol} ML Reversal Model Comparison",
        output_path=None,
    )
    if comparison_figure is not None:
        figures["model_comparison"] = comparison_figure

    hmm_figure = make_hmm_regime_plot(
        data=data,
        features=result["features"],
        state_summary=result["hmm_state_summary"],
        title=f"{config.symbol} HMM Regime Map",
        output_path=None,
    )
    if hmm_figure is not None:
        figures["hmm_regime"] = hmm_figure

    for model_name, probabilities in result.get("per_model_probabilities", {}).items():
        forecasts = result["per_model_forecasts"].get(model_name)
        signals = result["per_model_signals"].get(model_name, empty_signal_frame())
        intervals = result["per_model_intervals"].get(model_name, empty_interval_frame())
        threshold = result["per_model_thresholds"].get(model_name, np.nan)
        model_figure = make_plot(
            data=data,
            probabilities=probabilities,
            return_forecasts=forecasts,
            signals=signals,
            intervals=intervals,
            title=f"{config.symbol} {model_name} Return Forecast Breakouts",
            output_path=None,
            model_name=model_name,
            threshold=threshold,
            confidence_window=config.confidence_window,
            confidence_z=config.confidence_z,
        )
        if model_figure is not None:
            figures[f"{model_name}_plot"] = model_figure

        importance = result.get("per_model_feature_importance", {}).get(
            model_name,
            pd.DataFrame(columns=["feature", "importance_mean", "importance_std", "rank"]),
        )
        importance_figure = make_feature_importance_plot(
            importance=importance,
            title=f"{config.symbol} {model_name} XAI Feature Importance",
            output_path=None,
        )
        if importance_figure is not None:
            figures[f"{model_name}_feature_importance"] = importance_figure

    return figures


def display_notebook_figures(analysis: dict[str, Any]) -> None:
    try:
        from IPython.display import display
    except ModuleNotFoundError:
        warnings.warn("IPython unavailable; returning figures in analysis['figures'] without display.")
        analysis["figures"] = build_notebook_figures(analysis)
        return

    figures = build_notebook_figures(analysis)
    analysis["figures"] = figures
    if not figures:
        return

    for name, figure in figures.items():
        print(f"Displaying plot: {name}")
        display(figure)


def display_notebook_plot_links(plot_paths: dict[str, Path]) -> None:
    try:
        from IPython.display import HTML, display
    except ModuleNotFoundError:
        warnings.warn("IPython unavailable; returning plot paths without notebook display.")
        return
    links = "".join(
        f'<li><a href="{path}" target="_blank">{name}</a></li>'
        for name, path in plot_paths.items()
        if path.exists()
    )
    if links:
        display(HTML(f"<b>ML reversal plots</b><ul>{links}</ul>"))


if __name__ == "__main__":
    run_notebook()
```

```python
"""
Notebook usage:

from outputs.ml_reversal_pattern_detector import run_notebook

analysis = run_notebook(
    symbol="THYAO.IS",
    period="60d",
    interval="5m",
    show_plots=True,
)

analysis["result"]["intervals"]
analysis["result"]["forecasts"]
analysis["csv_paths"]
analysis["plot_paths"]
analysis["figures"]
"""

from __future__ import annotations

import json
import math
import warnings
from dataclasses import asdict, dataclass
from datetime import datetime
from pathlib import Path
from typing import Any

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    from sklearn.base import clone
    from sklearn.decomposition import PCA
    from sklearn.ensemble import HistGradientBoostingRegressor, RandomForestRegressor
    from sklearn.impute import SimpleImputer
    from sklearn.inspection import permutation_importance
    from sklearn.linear_model import Ridge
    from sklearn.metrics import (
        mean_absolute_error,
        mean_squared_error,
        precision_recall_curve,
        r2_score,
    )
    from sklearn.model_selection import TimeSeriesSplit
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
except ModuleNotFoundError:
    clone = None
    PCA = None
    HistGradientBoostingRegressor = None
    RandomForestRegressor = None
    SimpleImputer = None
    permutation_importance = None
    Ridge = None
    TimeSeriesSplit = None
    Pipeline = None
    StandardScaler = None
    mean_absolute_error = None
    mean_squared_error = None
    precision_recall_curve = None
    r2_score = None

try:
    from hmmlearn.hmm import GaussianHMM
except ModuleNotFoundError:
    GaussianHMM = None

try:
    from statsmodels.tsa.arima.model import ARIMA
except ModuleNotFoundError:
    ARIMA = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


REQUIRED_COLUMNS = {"open", "high", "low", "close"}


@dataclass
class ReversalMLConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    csv_path: str | None = None
    output_dir: str = "outputs/ml_reversal_runs"
    random_state: int = 42

    label_lookback: int = 80
    label_horizon: int = 40
    min_prior_drawdown_pct: float = 2.0
    max_distance_to_low_atr: float = 2.5
    target_atr: float = 2.0
    stop_atr: float = 1.2
    forecast_horizon: int | None = None
    forecast_target_column: str = "forecast_forward_return_pct"
    forecast_return_threshold_pct: float | None = None
    forecast_score_threshold: float | None = None
    forecast_scale_floor_pct: float = 0.15
    breakout_lookback: int = 8
    breakout_buffer_atr: float = 0.05
    forecast_cross_lookback: int = 3
    forecast_min_score_change: float = 0.08
    forecast_smooth_span: int = 5
    forecast_hysteresis_exit_ratio: float = 0.72
    switch_left_bars: int = 12
    switch_right_bars: int = 5
    switch_confirmation_bars: int = 8
    switch_label_window_bars: int = 1
    switch_min_rebound_atr: float = 0.55
    switch_max_rebound_atr: float = 2.4
    switch_box_left_bars: int = 5
    switch_box_right_bars: int = 10

    test_size_pct: float = 0.25
    n_splits: int = 5
    probability_threshold: float | None = None
    target_precision: float = 0.55
    cooldown_bars: int = 30
    interval_probability_floor: float | None = None
    interval_merge_gap_bars: int = 6
    min_interval_bars: int = 5
    max_interval_bars: int = 90
    interval_expand_atr: float = 0.35
    confidence_window: int = 20
    confidence_z: float = 1.0
    sr_lookback_bars: int = 70
    sr_min_base_bars: int = 12
    sr_base_left_pad_bars: int = 10
    sr_touch_tolerance_atr: float = 0.45
    sr_zone_max_width_atr: float = 0.65
    sr_min_support_touches: int = 2
    sr_min_resistance_touches: int = 2
    sr_touch_cluster_gap_bars: int = 3
    sr_breakout_buffer_atr: float = 0.10
    sr_max_breakout_extension_atr: float = 2.3
    sr_min_height_atr: float = 1.15
    sr_max_height_atr: float = 9.0
    sr_prior_downtrend_window: int = 55
    sr_breakout_volume_z_min: float = -0.50

    model_names: tuple[str, ...] = ("boosting", "ridge", "random_forest")
    use_hmm: bool = True
    hmm_states: int = 4
    use_arima: bool = False
    arima_order: tuple[int, int, int] = (1, 0, 1)
    arima_window: int = 160
    use_pca: bool = False
    pca_variance: float = 0.95
    permutation_repeats: int = 10


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    if df is None or df.empty:
        raise ValueError("OHLCV data is empty.")

    out = df.copy()
    out = out.rename(columns={column: str(column).lower() for column in out.columns})

    if not isinstance(out.index, pd.DatetimeIndex):
        date_column = next(
            (
                column
                for column in ["datetime", "date", "timestamp", "time"]
                if column in out.columns
            ),
            None,
        )
        if date_column is None:
            raise ValueError("Use a DatetimeIndex or include a date/datetime column.")
        out[date_column] = pd.to_datetime(out[date_column], errors="coerce")
        out = out.set_index(date_column)

    missing = REQUIRED_COLUMNS.difference(out.columns)
    if missing:
        raise ValueError(f"Missing columns: {sorted(missing)}")

    out.index = pd.to_datetime(out.index)
    out = out[~out.index.duplicated(keep="last")].sort_index()

    numeric_columns = ["open", "high", "low", "close"]
    if "volume" in out.columns:
        numeric_columns.append("volume")
    for column in numeric_columns:
        out[column] = pd.to_numeric(out[column], errors="coerce")

    out = out.dropna(subset=["open", "high", "low", "close"])
    if "volume" not in out.columns:
        out["volume"] = 0.0
    out["volume"] = out["volume"].fillna(0.0).clip(lower=0.0)

    if out.empty:
        raise ValueError("No valid OHLC rows remain after cleaning.")
    return out


INTRADAY_INTERVALS = {"1m", "2m", "5m", "15m", "30m", "60m", "90m", "1h"}


def fetch_ohlcv(symbol: str, period: str, interval: str) -> pd.DataFrame:
    if Ticker is None:
        raise ImportError(
            "yahooquery is required for online data fetch. "
            "Install with: pip install yahooquery"
        )

    errors: list[str] = []
    for candidate_period in data_period_attempts(period=period, interval=interval):
        try:
            raw = Ticker(symbol).history(period=candidate_period, interval=interval)
            frame = normalize_yahoo_history(raw, symbol=symbol)
            data = prepare_ohlcv(frame)
            if candidate_period != period:
                warnings.warn(
                    f"No usable data for period={period!r}, interval={interval!r}; "
                    f"using period={candidate_period!r} instead."
                )
            return data
        except Exception as exc:
            errors.append(f"{candidate_period}: {type(exc).__name__}: {exc}")

    raise ValueError(
        f"No data returned for {symbol} with interval={interval}. "
        "For Yahoo intraday data, use a shorter period such as 60d/30d/7d "
        "or a wider interval such as 1d. Attempts: "
        + " | ".join(errors)
    )


def data_period_attempts(*, period: str, interval: str) -> list[str]:
    attempts = [str(period)]
    if str(interval).lower() in INTRADAY_INTERVALS:
        attempts.extend(["60d", "30d", "7d", "5d", "1mo"])
    else:
        attempts.extend(["2y", "1y", "6mo", "3mo"])
    return list(dict.fromkeys(attempts))


def normalize_yahoo_history(raw: Any, *, symbol: str) -> pd.DataFrame:
    if raw is None:
        raise ValueError("empty response")
    if isinstance(raw, dict):
        message = raw.get(symbol, raw)
        raise ValueError(f"Yahoo returned an error response: {message}")
    if isinstance(raw, pd.Series):
        raise ValueError(f"Yahoo returned a Series instead of OHLCV data: {raw.to_dict()}")
    if not isinstance(raw, pd.DataFrame):
        raise ValueError(f"Unexpected Yahoo response type: {type(raw).__name__}")
    if raw.empty:
        raise ValueError("empty response")

    frame = raw.copy()
    if isinstance(frame.index, pd.MultiIndex):
        if "symbol" in frame.index.names:
            frame = frame.reset_index(level="symbol", drop=True)
        else:
            frame = frame.reset_index()
    return frame


def load_ohlcv(config: ReversalMLConfig) -> pd.DataFrame:
    if config.csv_path:
        return prepare_ohlcv(pd.read_csv(config.csv_path))
    return fetch_ohlcv(config.symbol, config.period, config.interval)


def make_feature_frame(data: pd.DataFrame, config: ReversalMLConfig) -> pd.DataFrame:
    close = data["close"].astype(float)
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    open_ = data["open"].astype(float)
    volume = data["volume"].astype(float)

    features = pd.DataFrame(index=data.index)
    log_close = np.log(close)
    log_return = log_close.diff()

    features["log_return_1"] = log_return
    for window in [2, 3, 5, 10, 20, 40, 80]:
        features[f"log_return_sum_{window}"] = log_return.rolling(window).sum()
        features[f"realized_vol_{window}"] = log_return.rolling(window).std()

    atr = average_true_range(data, period=14)
    features["atr"] = atr
    features["atr_pct"] = atr / close.replace(0, np.nan) * 100.0
    features["atr_pct_change_5"] = features["atr_pct"].pct_change(5)

    features["rsi_7"] = rsi(close, period=7)
    features["rsi_14"] = rsi(close, period=14)
    macd_line, macd_signal, macd_hist = macd(close)
    features["macd_line"] = macd_line
    features["macd_signal"] = macd_signal
    features["macd_hist"] = macd_hist
    features["stoch_k_14"] = stochastic_k(high, low, close, period=14)
    features["stoch_d_3"] = features["stoch_k_14"].rolling(3).mean()

    feature_windows = sorted({20, 50, 80, int(config.label_lookback)})
    for window in feature_windows:
        roll_high = high.rolling(window).max()
        roll_low = low.rolling(window).min()
        roll_mid = close.rolling(window).mean()
        roll_std = close.rolling(window).std()

        features[f"drawdown_from_high_{window}"] = (roll_high - close) / roll_high * 100.0
        features[f"rebound_from_low_{window}"] = (close - roll_low) / roll_low * 100.0
        features[f"dist_to_low_atr_{window}"] = (close - roll_low) / atr.replace(0, np.nan)
        features[f"dist_to_high_atr_{window}"] = (roll_high - close) / atr.replace(0, np.nan)
        features[f"bollinger_z_{window}"] = (close - roll_mid) / roll_std.replace(0, np.nan)
        features[f"range_pct_{window}"] = (roll_high - roll_low) / roll_mid.replace(0, np.nan) * 100.0
        features[f"slope_pct_{window}"] = rolling_slope_pct(close, window=window)

    for window in [5, 8, 13, 21]:
        prior_high = high.rolling(window).max().shift(1)
        recent_low = low.rolling(window).min()
        recent_high = high.rolling(window).max()
        recent_range = (recent_high - recent_low).replace(0, np.nan)
        features[f"reclaim_from_low_atr_{window}"] = (close - recent_low) / atr.replace(0, np.nan)
        features[f"break_above_prior_high_atr_{window}"] = (close - prior_high) / atr.replace(0, np.nan)
        features[f"low_range_position_{window}"] = (close - recent_low) / recent_range
        features[f"short_turn_score_{window}"] = (
            features[f"reclaim_from_low_atr_{window}"]
            + features[f"break_above_prior_high_atr_{window}"].clip(lower=-2.0, upper=2.0)
        )

    features["volatility_expansion_5_20"] = (
        features["realized_vol_5"] / features["realized_vol_20"].replace(0, np.nan)
    )
    features["volatility_compression_20_80"] = (
        features["realized_vol_20"] / features["realized_vol_80"].replace(0, np.nan)
    )
    features["range_compression_20_80"] = (
        features["range_pct_20"] / features["range_pct_80"].replace(0, np.nan)
    )

    candle_range = (high - low).replace(0, np.nan)
    features["body_pct"] = (close - open_) / open_.replace(0, np.nan) * 100.0
    features["body_to_range"] = (close - open_).abs() / candle_range
    features["upper_wick_to_range"] = (high - np.maximum(open_, close)) / candle_range
    features["lower_wick_to_range"] = (np.minimum(open_, close) - low) / candle_range
    features["close_position_in_range"] = (close - low) / candle_range
    features["gap_pct"] = (open_ - close.shift(1)) / close.shift(1).replace(0, np.nan) * 100.0
    features["green_streak"] = signed_streak(close >= open_)

    features["log_volume"] = np.log1p(volume)
    features["volume_z_20"] = rolling_zscore(features["log_volume"], window=20)
    features["volume_ratio_20"] = volume / volume.rolling(20).mean().replace(0, np.nan)
    up_volume = volume.where(close >= open_, 0.0)
    down_volume = volume.where(close < open_, 0.0)
    rolling_volume_20 = volume.rolling(20).sum().replace(0, np.nan)
    features["up_volume_share_20"] = up_volume.rolling(20).sum() / rolling_volume_20
    features["down_volume_share_20"] = down_volume.rolling(20).sum() / rolling_volume_20
    features["volume_expansion_5_20"] = (
        volume.rolling(5).mean() / volume.rolling(20).mean().replace(0, np.nan)
    )
    obv = on_balance_volume(close, volume)
    features["obv_slope_20"] = rolling_slope_pct(obv, window=20)

    features["prior_down_context"] = (
        (features[f"drawdown_from_high_{config.label_lookback}"] >= config.min_prior_drawdown_pct)
        & (features[f"dist_to_low_atr_{config.label_lookback}"] <= config.max_distance_to_low_atr)
    ).astype(float)

    if config.use_arima:
        arima_features = walk_forward_arima_features(
            log_return,
            order=config.arima_order,
            window=config.arima_window,
        )
        features = features.join(arima_features)

    features = features.replace([np.inf, -np.inf], np.nan)
    return features


def make_reversal_labels(
    data: pd.DataFrame,
    features: pd.DataFrame,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    open_ = data["open"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    atr = features["atr"].to_numpy(dtype=float)
    prior_drawdown = features[f"drawdown_from_high_{config.label_lookback}"].to_numpy(dtype=float)
    dist_to_low = features[f"dist_to_low_atr_{config.label_lookback}"].to_numpy(dtype=float)

    labels = pd.DataFrame(index=data.index)
    labels["reversal_label"] = np.nan
    labels["label_target_price"] = np.nan
    labels["label_stop_price"] = np.nan
    labels["label_hit_bar_idx"] = np.nan
    labels["label_hit_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["label_hit_type"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["forecast_forward_return_pct"] = np.nan
    labels["forecast_max_return_pct"] = np.nan
    labels["forecast_min_return_pct"] = np.nan
    labels["forecast_return_atr"] = np.nan
    labels["forecast_horizon_bars"] = np.nan
    labels["prior_drawdown_pct"] = prior_drawdown
    labels["distance_to_low_atr"] = dist_to_low
    labels["switch_kind"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["switch_turn_bar_idx"] = np.nan
    labels["switch_turn_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["switch_confirm_bar_idx"] = np.nan
    labels["switch_confirm_date"] = pd.Series([None] * len(labels), index=labels.index, dtype=object)
    labels["bars_from_switch_turn"] = np.nan
    labels["switch_rebound_atr"] = np.nan

    n = len(data)
    switch_events = detect_bullish_switch_events(
        data=data,
        features=features,
        open_=open_,
        high=high,
        low=low,
        close=close,
        atr=atr,
        prior_drawdown=prior_drawdown,
        dist_to_low=dist_to_low,
        config=config,
    )
    switch_lookup = make_switch_label_lookup(switch_events, n=n, config=config)

    forecast_horizon = int(config.forecast_horizon or config.label_horizon)
    for bar_idx in range(n):
        if bar_idx + config.label_horizon >= n:
            continue
        if not np.isfinite(atr[bar_idx]) or atr[bar_idx] <= 0:
            continue

        forecast_end_idx = min(n - 1, bar_idx + forecast_horizon)
        if forecast_end_idx > bar_idx and close[bar_idx] != 0:
            future_slice = slice(bar_idx + 1, forecast_end_idx + 1)
            forward_return_pct = (close[forecast_end_idx] / close[bar_idx] - 1.0) * 100.0
            max_return_pct = (finite_max(high[future_slice]) / close[bar_idx] - 1.0) * 100.0
            min_return_pct = (finite_min(low[future_slice]) / close[bar_idx] - 1.0) * 100.0
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_forward_return_pct")] = forward_return_pct
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_max_return_pct")] = max_return_pct
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_min_return_pct")] = min_return_pct
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_return_atr")] = (
                close[forecast_end_idx] - close[bar_idx]
            ) / atr[bar_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("forecast_horizon_bars")] = forecast_end_idx - bar_idx

        target_price = close[bar_idx] + config.target_atr * atr[bar_idx]
        stop_price = close[bar_idx] - config.stop_atr * atr[bar_idx]
        labels.iloc[bar_idx, labels.columns.get_loc("label_target_price")] = target_price
        labels.iloc[bar_idx, labels.columns.get_loc("label_stop_price")] = stop_price

        in_reversal_context = (
            np.isfinite(prior_drawdown[bar_idx])
            and np.isfinite(dist_to_low[bar_idx])
            and prior_drawdown[bar_idx] >= config.min_prior_drawdown_pct
            and dist_to_low[bar_idx] <= config.max_distance_to_low_atr
        )
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=bar_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        switch_event = switch_lookup.get(bar_idx)
        label = int(
            switch_event is not None
            and in_reversal_context
            and hit_type == "target"
        )
        labels.iloc[bar_idx, labels.columns.get_loc("reversal_label")] = label
        if hit_idx is not None:
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_bar_idx")] = int(hit_idx)
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_date")] = data.index[int(hit_idx)]
            labels.iloc[bar_idx, labels.columns.get_loc("label_hit_type")] = hit_type
        if switch_event is not None:
            turn_idx = int(switch_event["turn_idx"])
            confirm_idx = int(switch_event["confirm_idx"])
            labels.iloc[bar_idx, labels.columns.get_loc("switch_kind")] = switch_event["kind"]
            labels.iloc[bar_idx, labels.columns.get_loc("switch_turn_bar_idx")] = turn_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_turn_date")] = data.index[turn_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("switch_confirm_bar_idx")] = confirm_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_confirm_date")] = data.index[confirm_idx]
            labels.iloc[bar_idx, labels.columns.get_loc("bars_from_switch_turn")] = bar_idx - turn_idx
            labels.iloc[bar_idx, labels.columns.get_loc("switch_rebound_atr")] = switch_event["rebound_atr"]

    return labels


def detect_bullish_switch_events(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr: np.ndarray,
    prior_drawdown: np.ndarray,
    dist_to_low: np.ndarray,
    config: ReversalMLConfig,
) -> list[dict[str, Any]]:
    prior_context = features.get("prior_down_context", pd.Series(0.0, index=features.index)).to_numpy(
        dtype=float
    )
    left_bars = max(3, int(config.switch_left_bars))
    right_bars = max(1, int(config.switch_right_bars))
    confirmation_bars = max(right_bars, int(config.switch_confirmation_bars))
    n = len(data)
    max_future_need = max(right_bars, confirmation_bars, int(config.label_horizon))
    events: list[dict[str, Any]] = []

    for turn_idx in range(left_bars, n - max_future_need):
        if not np.isfinite(atr[turn_idx]) or atr[turn_idx] <= 0:
            continue
        if not is_bullish_switch_context(
            turn_idx=turn_idx,
            close=close,
            high=high,
            prior_context=prior_context,
            prior_drawdown=prior_drawdown,
            dist_to_low=dist_to_low,
            config=config,
        ):
            continue
        if not is_local_valley(low, turn_idx=turn_idx, left_bars=left_bars, right_bars=right_bars):
            continue

        confirm_idx = find_bullish_switch_confirmation(
            turn_idx=turn_idx,
            open_=open_,
            high=high,
            low=low,
            close=close,
            atr=atr,
            config=config,
        )
        if confirm_idx is None or confirm_idx + config.label_horizon >= n:
            continue

        target_price = close[confirm_idx] + config.target_atr * atr[confirm_idx]
        stop_price = close[confirm_idx] - config.stop_atr * atr[confirm_idx]
        hit_type, hit_idx = first_long_barrier_hit(
            high=high,
            low=low,
            start_idx=confirm_idx,
            horizon=config.label_horizon,
            target_price=target_price,
            stop_price=stop_price,
        )
        if hit_type != "target":
            continue

        events.append(
            {
                "kind": "valley",
                "turn_idx": int(turn_idx),
                "confirm_idx": int(confirm_idx),
                "hit_idx": int(hit_idx) if hit_idx is not None else None,
                "rebound_atr": float((close[confirm_idx] - low[turn_idx]) / atr[turn_idx]),
                "turn_price": float(low[turn_idx]),
                "confirm_price": float(close[confirm_idx]),
            }
        )

    return suppress_overlapping_switch_events(
        events,
        min_gap=max(2, int(config.switch_confirmation_bars)),
    )


def is_bullish_switch_context(
    *,
    turn_idx: int,
    close: np.ndarray,
    high: np.ndarray,
    prior_context: np.ndarray,
    prior_drawdown: np.ndarray,
    dist_to_low: np.ndarray,
    config: ReversalMLConfig,
) -> bool:
    context_start = max(0, int(turn_idx) - int(config.label_lookback))
    context_seen = finite_max(prior_context[context_start : turn_idx + 1])
    if not np.isfinite(context_seen) or context_seen <= 0:
        return False
    if not np.isfinite(prior_drawdown[turn_idx]) or not np.isfinite(dist_to_low[turn_idx]):
        return False
    if prior_drawdown[turn_idx] < config.min_prior_drawdown_pct:
        return False
    if dist_to_low[turn_idx] > config.max_distance_to_low_atr:
        return False

    approach_start = max(0, int(turn_idx) - max(5, int(config.switch_left_bars)))
    if turn_idx - approach_start < 4:
        return False
    approach_high = finite_max(high[approach_start : turn_idx + 1])
    if not np.isfinite(approach_high) or approach_high <= 0:
        return False
    approach_drawdown = (approach_high - close[turn_idx]) / approach_high * 100.0
    if approach_drawdown < config.min_prior_drawdown_pct * 0.35:
        return False
    return close[turn_idx] <= np.nanmedian(close[approach_start : turn_idx + 1])


def is_local_valley(
    low: np.ndarray,
    *,
    turn_idx: int,
    left_bars: int,
    right_bars: int,
) -> bool:
    start = max(0, int(turn_idx) - int(left_bars))
    end = min(len(low) - 1, int(turn_idx) + int(right_bars))
    window = low[start : end + 1]
    if len(window) == 0 or not np.isfinite(low[turn_idx]):
        return False
    valley = finite_min(window)
    if not np.isfinite(valley):
        return False
    if low[turn_idx] > valley + max(abs(valley) * 1e-8, 1e-10):
        return False
    right_window = low[turn_idx + 1 : end + 1]
    return len(right_window) == 0 or finite_min(right_window) > low[turn_idx] - max(abs(low[turn_idx]) * 1e-8, 1e-10)


def find_bullish_switch_confirmation(
    *,
    turn_idx: int,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr: np.ndarray,
    config: ReversalMLConfig,
) -> int | None:
    atr_at_turn = float(atr[turn_idx])
    if not np.isfinite(atr_at_turn) or atr_at_turn <= 0:
        return None
    end_idx = min(len(close) - 1, int(turn_idx) + int(config.switch_confirmation_bars))
    rebound_level = low[turn_idx] + config.switch_min_rebound_atr * atr_at_turn
    micro_start = max(0, int(turn_idx) - max(3, int(config.switch_left_bars) // 2))
    micro_break_level = finite_max(high[micro_start:turn_idx])

    for idx in range(int(turn_idx) + 1, end_idx + 1):
        rebound_atr = (close[idx] - low[turn_idx]) / atr_at_turn
        if rebound_atr < config.switch_min_rebound_atr:
            continue
        if rebound_atr > config.switch_max_rebound_atr:
            continue
        closes_green = close[idx] > open_[idx]
        breaks_previous_high = idx > 0 and close[idx] > high[idx - 1]
        breaks_micro_high = np.isfinite(micro_break_level) and close[idx] > micro_break_level
        reclaims_valley = close[idx] >= rebound_level
        if reclaims_valley and (closes_green or breaks_previous_high or breaks_micro_high):
            return int(idx)
    return None


def make_switch_label_lookup(
    events: list[dict[str, Any]],
    *,
    n: int,
    config: ReversalMLConfig,
) -> dict[int, dict[str, Any]]:
    lookup: dict[int, dict[str, Any]] = {}
    radius = max(0, int(config.switch_label_window_bars))
    for event in events:
        anchor_idx = int(event["confirm_idx"])
        turn_idx = int(event["turn_idx"])
        for label_idx in range(anchor_idx - radius, anchor_idx + radius + 1):
            if label_idx < turn_idx or label_idx < 0 or label_idx >= n:
                continue
            current = lookup.get(label_idx)
            if current is None:
                lookup[label_idx] = event
                continue
            current_distance = abs(label_idx - int(current["confirm_idx"]))
            new_distance = abs(label_idx - anchor_idx)
            if new_distance < current_distance:
                lookup[label_idx] = event
    return lookup


def suppress_overlapping_switch_events(
    events: list[dict[str, Any]],
    *,
    min_gap: int,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []
    for event in sorted(events, key=lambda item: int(item["confirm_idx"])):
        if not selected:
            selected.append(event)
            continue
        previous = selected[-1]
        if int(event["confirm_idx"]) - int(previous["confirm_idx"]) > min_gap:
            selected.append(event)
            continue
        if float(event["turn_price"]) < float(previous["turn_price"]):
            selected[-1] = event
    return selected


def first_long_barrier_hit(
    *,
    high: np.ndarray,
    low: np.ndarray,
    start_idx: int,
    horizon: int,
    target_price: float,
    stop_price: float,
) -> tuple[str, int | None]:
    max_idx = min(len(high) - 1, int(start_idx) + int(horizon))
    for idx in range(int(start_idx) + 1, max_idx + 1):
        stop_hit = low[idx] <= stop_price
        target_hit = high[idx] >= target_price
        if stop_hit and target_hit:
            return "stop", idx
        if stop_hit:
            return "stop", idx
        if target_hit:
            return "target", idx
    return "timeout", None


def add_hmm_features(
    features: pd.DataFrame,
    train_index: pd.Index,
    config: ReversalMLConfig,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    state_summary_columns = ["state", "train_count", "mean_return", "mean_vol", "mean_drawdown"]
    if not config.use_hmm:
        return features, pd.DataFrame(columns=state_summary_columns)
    if GaussianHMM is None or SimpleImputer is None or StandardScaler is None:
        warnings.warn("hmmlearn/sklearn unavailable; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    hmm_columns = [
        column
        for column in [
            "log_return_1",
            "realized_vol_10",
            "atr_pct",
            "volume_z_20",
            f"drawdown_from_high_{config.label_lookback}",
            f"dist_to_low_atr_{config.label_lookback}",
        ]
        if column in features.columns
    ]
    if len(hmm_columns) < 3:
        return features, pd.DataFrame(columns=state_summary_columns)

    train = features.loc[train_index, hmm_columns]
    if train.dropna(how="all").shape[0] < config.hmm_states * 20:
        warnings.warn("Not enough training rows for HMM; HMM regime features skipped.")
        return features, pd.DataFrame(columns=state_summary_columns)

    imputer = SimpleImputer(strategy="median")
    scaler = StandardScaler()
    train_values = scaler.fit_transform(imputer.fit_transform(train))
    all_values = scaler.transform(imputer.transform(features[hmm_columns]))

    hmm = GaussianHMM(
        n_components=config.hmm_states,
        covariance_type="diag",
        n_iter=200,
        tol=1e-4,
        random_state=config.random_state,
    )
    try:
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            hmm.fit(train_values)
            probabilities = hmm.predict_proba(all_values)
    except Exception as exc:
        warnings.warn(f"HMM regime features skipped after fit failure: {type(exc).__name__}: {exc}")
        return features, pd.DataFrame(columns=state_summary_columns)
    states = probabilities.argmax(axis=1)

    out = features.copy()
    out["hmm_state"] = states.astype(float)
    for state in range(config.hmm_states):
        out[f"hmm_state_prob_{state}"] = probabilities[:, state]

    state_rows: list[dict[str, Any]] = []
    train_augmented = out.loc[train_index].copy()
    for state in range(config.hmm_states):
        subset = train_augmented[train_augmented["hmm_state"] == state]
        state_rows.append(
            {
                "state": state,
                "train_count": int(len(subset)),
                "mean_return": float(subset["log_return_1"].mean()),
                "mean_vol": float(subset["realized_vol_10"].mean()),
                "mean_drawdown": float(
                    subset[f"drawdown_from_high_{config.label_lookback}"].mean()
                ),
            }
        )
    return out, pd.DataFrame(state_rows, columns=state_summary_columns)


def train_reversal_models(
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    config: ReversalMLConfig,
) -> dict[str, Any]:
    require_sklearn()

    target_column = config.forecast_target_column
    if target_column not in labels.columns:
        raise ValueError(f"Unknown forecast target column: {target_column}")

    labeled = features.join(labels[[target_column]], how="inner")
    labeled = labeled.dropna(subset=[target_column])
    if len(labeled) < 200:
        raise ValueError("Not enough labeled rows. Increase period or use a higher timeframe.")

    split_idx = int(len(labeled) * (1.0 - config.test_size_pct))
    split_idx = min(max(split_idx, 50), len(labeled) - 20)
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]

    features_with_hmm, hmm_summary = add_hmm_features(features, train_index, config)
    labeled = features_with_hmm.join(labels[[target_column]], how="inner")
    labeled = labeled.dropna(subset=[target_column])
    y = labeled[target_column].astype(float)
    feature_columns = [
        column
        for column in labeled.columns
        if column != target_column and pd.api.types.is_numeric_dtype(labeled[column])
    ]
    X = labeled[feature_columns]
    split_idx = int(len(labeled) * (1.0 - config.test_size_pct))
    split_idx = min(max(split_idx, 50), len(labeled) - 20)
    train_index = labeled.index[:split_idx]
    test_index = labeled.index[split_idx:]
    X_train = X.loc[train_index]
    y_train = y.loc[train_index]
    X_test = X.loc[test_index]
    y_test = y.loc[test_index]
    if y_train.dropna().nunique() < 2:
        raise ValueError(
            "Training split has no return variation. Use more history or a higher timeframe."
        )

    X_train_clip, clip_bounds = fit_winsorize(X_train)
    X_test_clip = apply_winsorize(X_test, clip_bounds)
    X_all_clip = apply_winsorize(features_with_hmm[feature_columns], clip_bounds)
    return_threshold_pct = choose_return_threshold(y_train, config)
    forecast_score_scale = choose_forecast_score_scale(y_train, config)
    train_sample_weight = make_return_sample_weight(labels, y_train, train_index, config)

    models = build_models(config)
    results: list[dict[str, Any]] = []
    fitted_models: dict[str, Pipeline] = {}

    for name, model in models.items():
        fitted = fit_pipeline_with_weights(
            model,
            X_train_clip,
            y_train,
            sample_weight=train_sample_weight,
        )
        fitted_models[name] = fitted
        test_forecast = predict_return(fitted, X_test_clip)
        metrics = evaluate_return_forecasts(
            y_test,
            test_forecast,
            return_threshold_pct=return_threshold_pct,
        )
        cv_metrics = time_series_cv_score(
            model,
            X_train_clip,
            y_train,
            n_splits=config.n_splits,
            return_threshold_pct=return_threshold_pct,
        )
        result = {"model": name, **metrics, **cv_metrics}
        results.append(result)

    model_report = pd.DataFrame(results)
    best_name = choose_best_model(model_report)
    per_model_probabilities: dict[str, pd.Series] = {}
    per_model_forecasts: dict[str, pd.Series] = {}
    per_model_thresholds: dict[str, float] = {}
    per_model_feature_importance: dict[str, pd.DataFrame] = {}
    per_model_signals: dict[str, pd.DataFrame] = {}
    per_model_intervals: dict[str, pd.DataFrame] = {}

    for name, fitted in fitted_models.items():
        all_forecast = predict_return(fitted, X_all_clip)
        test_forecast = predict_return(fitted, X_test_clip)
        all_score = forecast_to_score(
            all_forecast,
            return_threshold_pct=return_threshold_pct,
            scale_pct=forecast_score_scale,
        )
        raw_score_series = pd.Series(
            all_score,
            index=features_with_hmm.index,
            name=f"{name}_forecast_score_raw",
        )
        score_series = smooth_forecast_score(raw_score_series, config).rename(f"{name}_forecast_score")
        test_score = score_series.reindex(test_index).to_numpy(dtype=float)
        model_threshold = config.forecast_score_threshold
        if model_threshold is None:
            model_threshold = config.probability_threshold
        if model_threshold is None:
            model_threshold = choose_forecast_score_threshold(
                y_test,
                test_score,
                return_threshold_pct=return_threshold_pct,
                target_precision=config.target_precision,
            )

        forecast_series = pd.Series(
            all_forecast,
            index=features_with_hmm.index,
            name=f"{name}_return_forecast_pct",
        )
        importance = permutation_xai(
            fitted,
            X_test_clip,
            y_test,
            feature_columns,
            repeats=config.permutation_repeats,
            random_state=config.random_state,
        )
        signals = build_signal_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=score_series,
            return_forecasts=forecast_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not signals.empty:
            signals.insert(0, "Model", name)
        intervals = build_interval_table(
            data=data,
            features=features_with_hmm,
            labels=labels,
            probabilities=score_series,
            return_forecasts=forecast_series,
            feature_columns=feature_columns,
            feature_importance=importance,
            train_features=X_train_clip,
            threshold=float(model_threshold),
            config=config,
        )
        if not intervals.empty:
            intervals.insert(0, "Model", name)

        per_model_probabilities[name] = score_series
        per_model_forecasts[name] = forecast_series
        per_model_thresholds[name] = float(model_threshold)
        per_model_feature_importance[name] = importance
        per_model_signals[name] = signals
        per_model_intervals[name] = intervals

    best_model = fitted_models[best_name]
    threshold = per_model_thresholds[best_name]
    probabilities = per_model_probabilities[best_name].rename("forecast_score")
    forecasts = per_model_forecasts[best_name].rename("return_forecast_pct")
    signals = per_model_signals[best_name].copy()
    if "Model" in signals.columns:
        signals = signals.drop(columns=["Model"])
    intervals = per_model_intervals[best_name].copy()
    if "Model" in intervals.columns:
        intervals = intervals.drop(columns=["Model"])
    feature_importance = per_model_feature_importance[best_name]
    pca_loadings = extract_pca_loadings(best_model, feature_columns)

    return {
        "features": features_with_hmm,
        "labels": labels,
        "feature_columns": feature_columns,
        "model_report": model_report,
        "best_model_name": best_name,
        "best_model": best_model,
        "probabilities": probabilities,
        "forecasts": forecasts,
        "threshold": threshold,
        "return_threshold_pct": return_threshold_pct,
        "forecast_score_scale": forecast_score_scale,
        "signals": signals,
        "intervals": intervals,
        "feature_importance": feature_importance,
        "per_model_probabilities": per_model_probabilities,
        "per_model_forecasts": per_model_forecasts,
        "per_model_thresholds": per_model_thresholds,
        "per_model_signals": per_model_signals,
        "per_model_intervals": per_model_intervals,
        "per_model_feature_importance": per_model_feature_importance,
        "pca_loadings": pca_loadings,
        "hmm_state_summary": hmm_summary,
    }


def require_sklearn() -> None:
    if Pipeline is None:
        raise ImportError(
            "scikit-learn is required for ML training. Install with: pip install scikit-learn"
        )


def build_models(config: ReversalMLConfig) -> dict[str, Pipeline]:
    requested = {name.strip().lower() for name in config.model_names}
    models: dict[str, Pipeline] = {}

    if "boosting" in requested:
        models["boosting"] = make_pipeline(
            HistGradientBoostingRegressor(
                learning_rate=0.035,
                max_iter=700,
                max_leaf_nodes=9,
                min_samples_leaf=35,
                l2_regularization=0.25,
                early_stopping=True,
                validation_fraction=0.15,
                n_iter_no_change=30,
                random_state=config.random_state,
            ),
            config,
            scale=False,
        )
    if "ridge" in requested:
        models["ridge"] = make_pipeline(
            Ridge(
                alpha=5.0,
                random_state=config.random_state,
            ),
            config,
            scale=True,
        )
    if "random_forest" in requested:
        models["random_forest"] = make_pipeline(
            RandomForestRegressor(
                n_estimators=700,
                max_depth=5,
                min_samples_leaf=35,
                max_features=0.70,
                random_state=config.random_state,
                n_jobs=-1,
            ),
            config,
            scale=False,
        )
    if not models:
        raise ValueError("No valid model names requested.")
    return models


def make_pipeline(estimator: Any, config: ReversalMLConfig, *, scale: bool) -> Pipeline:
    steps: list[tuple[str, Any]] = [("imputer", SimpleImputer(strategy="median"))]
    if scale or config.use_pca:
        steps.append(("scaler", StandardScaler()))
    if config.use_pca:
        steps.append(("pca", PCA(n_components=config.pca_variance, svd_solver="full")))
    steps.append(("model", estimator))
    return Pipeline(steps)


def fit_pipeline_with_weights(
    model: Pipeline,
    X: pd.DataFrame,
    y: pd.Series,
    *,
    sample_weight: np.ndarray | pd.Series | None = None,
) -> Pipeline:
    try:
        return model.fit(X, y, model__sample_weight=sample_weight)
    except TypeError:
        return model.fit(X, y)


def predict_return(model: Pipeline, X: pd.DataFrame) -> np.ndarray:
    return np.asarray(model.predict(X), dtype=float)


def evaluate_return_forecasts(
    y_true: pd.Series,
    forecast: np.ndarray,
    *,
    return_threshold_pct: float,
) -> dict[str, float]:
    y_values = y_true.to_numpy(dtype=float)
    forecast_values = np.asarray(forecast, dtype=float)
    finite_mask = np.isfinite(y_values) & np.isfinite(forecast_values)
    out: dict[str, float] = {
        "test_mae": np.nan,
        "test_rmse": np.nan,
        "test_r2": np.nan,
        "test_correlation": np.nan,
        "test_directional_accuracy": np.nan,
        "test_actual_positive_rate": np.nan,
        "test_predicted_positive_rate": np.nan,
        "test_positive_precision": np.nan,
        "test_positive_recall": np.nan,
    }
    if finite_mask.sum() < 3:
        return out

    y_values = y_values[finite_mask]
    forecast_values = forecast_values[finite_mask]
    out["test_mae"] = float(mean_absolute_error(y_values, forecast_values))
    out["test_rmse"] = float(math.sqrt(mean_squared_error(y_values, forecast_values)))
    out["test_r2"] = float(r2_score(y_values, forecast_values))
    if np.std(y_values) > 0 and np.std(forecast_values) > 0:
        out["test_correlation"] = float(np.corrcoef(y_values, forecast_values)[0, 1])
    out["test_directional_accuracy"] = float(np.mean(np.sign(y_values) == np.sign(forecast_values)))

    actual_positive = y_values >= float(return_threshold_pct)
    predicted_positive = forecast_values >= float(return_threshold_pct)
    out["test_actual_positive_rate"] = float(np.mean(actual_positive))
    out["test_predicted_positive_rate"] = float(np.mean(predicted_positive))
    true_positive = float(np.sum(actual_positive & predicted_positive))
    out["test_positive_precision"] = true_positive / max(float(np.sum(predicted_positive)), 1.0)
    out["test_positive_recall"] = true_positive / max(float(np.sum(actual_positive)), 1.0)
    return out


def time_series_cv_score(
    model: Pipeline,
    X: pd.DataFrame,
    y: pd.Series,
    *,
    n_splits: int,
    return_threshold_pct: float,
) -> dict[str, float]:
    out = {"cv_mae": np.nan, "cv_rmse": np.nan, "cv_correlation": np.nan}
    if len(X) < 200 or y.dropna().nunique() < 2:
        return out

    split_count = min(n_splits, max(2, len(X) // 200))
    splitter = TimeSeriesSplit(n_splits=split_count)
    mae_scores: list[float] = []
    rmse_scores: list[float] = []
    corr_scores: list[float] = []
    for train_idx, test_idx in splitter.split(X):
        y_train = y.iloc[train_idx]
        y_valid = y.iloc[test_idx]
        if y_train.dropna().nunique() < 2:
            continue
        fitted = fit_pipeline_with_weights(clone(model), X.iloc[train_idx], y_train)
        forecast = predict_return(fitted, X.iloc[test_idx])
        metrics = evaluate_return_forecasts(
            y_valid,
            forecast,
            return_threshold_pct=return_threshold_pct,
        )
        if np.isfinite(metrics["test_mae"]):
            mae_scores.append(metrics["test_mae"])
        if np.isfinite(metrics["test_rmse"]):
            rmse_scores.append(metrics["test_rmse"])
        if np.isfinite(metrics["test_correlation"]):
            corr_scores.append(metrics["test_correlation"])

    if mae_scores:
        out["cv_mae"] = float(np.mean(mae_scores))
    if rmse_scores:
        out["cv_rmse"] = float(np.mean(rmse_scores))
    if corr_scores:
        out["cv_correlation"] = float(np.mean(corr_scores))
    return out


def choose_best_model(model_report: pd.DataFrame) -> str:
    ranking = model_report.copy()
    mae_rank = ranking["test_mae"].rank(ascending=True, pct=True).fillna(1.0)
    ranking["_rank_score"] = (
        ranking["test_correlation"].fillna(0.0)
        + ranking["test_directional_accuracy"].fillna(0.0)
        - mae_rank
    )
    ranking = ranking.sort_values(["_rank_score", "test_rmse"], ascending=[False, True])
    return str(ranking.iloc[0]["model"])


def choose_forecast_score_threshold(
    y_true: pd.Series,
    forecast_score: np.ndarray,
    *,
    return_threshold_pct: float,
    target_precision: float,
) -> float:
    y_values = y_true.to_numpy(dtype=float)
    score_values = np.asarray(forecast_score, dtype=float)
    finite_mask = np.isfinite(y_values) & np.isfinite(score_values)
    if finite_mask.sum() < 3:
        return 0.60
    actual_positive = y_values[finite_mask] >= float(return_threshold_pct)
    score_values = score_values[finite_mask]
    if len(np.unique(actual_positive)) < 2 or precision_recall_curve is None:
        return 0.60

    precision, recall, thresholds = precision_recall_curve(actual_positive.astype(int), score_values)
    candidates: list[tuple[float, float, float]] = []
    for threshold, prec, rec in zip(thresholds, precision[:-1], recall[:-1]):
        if prec >= target_precision and rec > 0:
            candidates.append((float(threshold), float(prec), float(rec)))
    if not candidates:
        return float(np.quantile(score_values, 0.80))
    candidates.sort(key=lambda item: (item[2], item[1]), reverse=True)
    return float(candidates[0][0])


def choose_return_threshold(y_train: pd.Series, config: ReversalMLConfig) -> float:
    if config.forecast_return_threshold_pct is not None:
        return float(config.forecast_return_threshold_pct)

    values = y_train.dropna().to_numpy(dtype=float)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return float(config.forecast_scale_floor_pct)
    positive_values = values[values > 0]
    if len(positive_values) >= 10:
        threshold = float(np.quantile(positive_values, 0.40))
    else:
        threshold = float(np.quantile(values, 0.65))
    return max(float(config.forecast_scale_floor_pct), threshold)


def choose_forecast_score_scale(y_train: pd.Series, config: ReversalMLConfig) -> float:
    values = y_train.dropna().to_numpy(dtype=float)
    values = values[np.isfinite(values)]
    if len(values) < 5:
        return float(config.forecast_scale_floor_pct)
    q25, q75 = np.quantile(values, [0.25, 0.75])
    robust_scale = float((q75 - q25) / 2.0)
    std_scale = float(np.std(values))
    scale = max(robust_scale, std_scale * 0.35, float(config.forecast_scale_floor_pct), 1e-6)
    return float(scale)


def forecast_to_score(
    forecast: np.ndarray | pd.Series,
    *,
    return_threshold_pct: float,
    scale_pct: float,
) -> np.ndarray:
    values = np.asarray(forecast, dtype=float)
    z = (values - float(return_threshold_pct)) / max(float(scale_pct), 1e-6)
    z = np.clip(z, -30.0, 30.0)
    return 1.0 / (1.0 + np.exp(-z))


def make_return_sample_weight(
    labels: pd.DataFrame,
    y_train: pd.Series,
    train_index: pd.Index,
    config: ReversalMLConfig,
) -> np.ndarray:
    weights = np.ones(len(y_train), dtype=float)
    aligned_labels = labels.reindex(train_index)
    if "switch_kind" in aligned_labels.columns:
        weights += aligned_labels["switch_kind"].notna().to_numpy(dtype=float) * 1.5
    if "prior_drawdown_pct" in aligned_labels.columns:
        drawdown = aligned_labels["prior_drawdown_pct"].fillna(0.0).to_numpy(dtype=float)
        weights += np.clip(drawdown / max(config.min_prior_drawdown_pct, 1e-6), 0.0, 2.0) * 0.35
    values = y_train.to_numpy(dtype=float)
    finite_values = values[np.isfinite(values)]
    if len(finite_values) > 5:
        scale = max(float(np.quantile(np.abs(finite_values), 0.75)), config.forecast_scale_floor_pct)
        weights += np.clip(np.abs(values) / scale, 0.0, 2.0) * 0.25
    return weights


def permutation_xai(
    model: Pipeline,
    X_test: pd.DataFrame,
    y_test: pd.Series,
    feature_columns: list[str],
    *,
    repeats: int,
    random_state: int,
) -> pd.DataFrame:
    columns = ["feature", "importance_mean", "importance_std", "rank"]
    if len(X_test) < 5 or permutation_importance is None:
        return pd.DataFrame(columns=columns)

    result = permutation_importance(
        model,
        X_test,
        y_test,
        n_repeats=repeats,
        random_state=random_state,
        scoring="neg_mean_absolute_error",
        n_jobs=-1,
    )
    importance = pd.DataFrame(
        {
            "feature": feature_columns,
            "importance_mean": result.importances_mean,
            "importance_std": result.importances_std,
        }
    )
    importance = importance.sort_values("importance_mean", ascending=False).reset_index(drop=True)
    importance["rank"] = np.arange(1, len(importance) + 1)
    return importance[columns]


def extract_pca_loadings(model: Pipeline, feature_columns: list[str]) -> pd.DataFrame:
    if "pca" not in model.named_steps:
        return pd.DataFrame(columns=["component", "feature", "loading", "abs_loading", "rank"])

    pca = model.named_steps["pca"]
    rows: list[dict[str, Any]] = []
    for component_idx, component in enumerate(pca.components_):
        for feature, loading in zip(feature_columns, component):
            rows.append(
                {
                    "component": f"PC{component_idx + 1}",
                    "feature": feature,
                    "loading": float(loading),
                    "abs_loading": float(abs(loading)),
                }
            )
    out = pd.DataFrame(rows)
    out["rank"] = out.groupby("component")["abs_loading"].rank(ascending=False, method="first")
    return out.sort_values(["component", "rank"]).reset_index(drop=True)


def build_signal_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    return_forecasts: pd.Series | None,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    aligned_probability = smooth_forecast_score(
        probabilities.reindex(features.index).astype(float),
        config,
    )
    aligned_return_forecast = (
        return_forecasts.reindex(features.index).astype(float)
        if return_forecasts is not None
        else pd.Series(np.nan, index=features.index)
    )
    floor = config.interval_probability_floor
    if floor is None:
        floor = max(0.05, min(0.95, float(threshold) * float(config.forecast_hysteresis_exit_ratio)))

    confidence = probability_confidence_frame(
        aligned_probability,
        window=config.confidence_window,
        z=config.confidence_z,
    )
    setup_lookup = support_resistance_breakout_setups(data=data, features=features, config=config)
    sr_breakout_mask = support_resistance_breakout_mask(features.index, setup_lookup)
    forecast_turn_filter = return_forecast_turn_filter(
        forecast_score=aligned_probability,
        return_forecast=aligned_return_forecast,
        threshold=float(threshold),
        floor=float(floor),
        config=config,
    )
    candidate_mask = (
        sr_breakout_mask
        & forecast_turn_filter.fillna(False)
        & (
            (aligned_probability >= float(threshold))
            | (
                (confidence["upper"] >= float(threshold))
                & (aligned_probability >= float(floor))
            )
        ).fillna(False)
    )
    candidate_probability = aligned_probability[candidate_mask].dropna()
    if candidate_probability.empty:
        return empty_signal_frame()

    selected_indexes = non_max_suppression_1d(
        candidate_probability,
        cooldown_bars=config.cooldown_bars,
        all_index=features.index,
    )

    train_stats = make_train_feature_stats(train_features)
    rows: list[dict[str, Any]] = []
    atr = features["atr"]
    for timestamp in selected_indexes:
        bar_idx = int(data.index.get_loc(timestamp))
        setup = setup_lookup.get(bar_idx, {})
        entry_price = float(data.loc[timestamp, "close"])
        return_forecast_pct = (
            float(return_forecasts.loc[timestamp])
            if return_forecasts is not None and timestamp in return_forecasts.index
            else np.nan
        )
        target_price = entry_price + config.target_atr * float(atr.loc[timestamp])
        stop_price = entry_price - config.stop_atr * float(atr.loc[timestamp])
        label_row = labels.loc[timestamp] if timestamp in labels.index else pd.Series(dtype=object)
        hit_idx = label_row.get("label_hit_bar_idx", np.nan)
        hit_date = label_row.get("label_hit_date", pd.NaT)
        hit_type = label_row.get("label_hit_type", None)
        realized_label = label_row.get("reversal_label", np.nan)
        switch_kind = label_row.get("switch_kind", None)
        switch_turn_idx = label_row.get("switch_turn_bar_idx", np.nan)
        switch_turn_date = label_row.get("switch_turn_date", pd.NaT)
        switch_confirm_idx = label_row.get("switch_confirm_bar_idx", np.nan)
        switch_confirm_date = label_row.get("switch_confirm_date", pd.NaT)
        bars_from_switch_turn = label_row.get("bars_from_switch_turn", np.nan)
        top_drivers = local_driver_summary(
            features.loc[timestamp, feature_columns],
            feature_importance,
            train_stats,
            top_n=6,
        )
        rows.append(
            {
                "EntryBarIdx": bar_idx,
                "EntryDate": timestamp,
                "EntryPrice": entry_price,
                "ReturnForecastPct": return_forecast_pct,
                "ForecastScore": float(aligned_probability.loc[timestamp]),
                "ReversalProbability": float(aligned_probability.loc[timestamp]),
                "SignalThreshold": float(threshold),
                "SupportZoneLow": setup.get("support_zone_low", np.nan),
                "SupportZoneHigh": setup.get("support_zone_high", np.nan),
                "ResistanceZoneLow": setup.get("resistance_zone_low", np.nan),
                "ResistanceZoneHigh": setup.get("resistance_zone_high", np.nan),
                "BreakoutLevel": setup.get("breakout_level", np.nan),
                "SupportTouches": setup.get("support_touches", np.nan),
                "ResistanceTouches": setup.get("resistance_touches", np.nan),
                "BaseRangeAtr": setup.get("base_range_atr", np.nan),
                "BreakoutStrengthAtr": setup.get("breakout_strength_atr", np.nan),
                "TargetPrice": target_price,
                "StopPrice": stop_price,
                "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                "OutcomeKnown": is_finite_number(realized_label),
                "OutcomeLabel": realized_label,
                "HitBarIdx": int(hit_idx) if is_finite_number(hit_idx) else np.nan,
                "HitDate": hit_date,
                "HitType": hit_type,
                "SwitchKind": switch_kind,
                "SwitchTurnBarIdx": int(switch_turn_idx) if is_finite_number(switch_turn_idx) else np.nan,
                "SwitchTurnDate": switch_turn_date,
                "SwitchConfirmBarIdx": int(switch_confirm_idx) if is_finite_number(switch_confirm_idx) else np.nan,
                "SwitchConfirmDate": switch_confirm_date,
                "BarsFromSwitchTurn": bars_from_switch_turn,
                "ShapeHint": classify_shape_hint(data, bar_idx),
                "TopDrivers": top_drivers,
            }
        )
    return pd.DataFrame(rows).sort_values("EntryBarIdx").reset_index(drop=True)


def build_interval_table(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    labels: pd.DataFrame,
    probabilities: pd.Series,
    return_forecasts: pd.Series | None,
    feature_columns: list[str],
    feature_importance: pd.DataFrame,
    train_features: pd.DataFrame,
    threshold: float,
    config: ReversalMLConfig,
) -> pd.DataFrame:
    aligned_probability = smooth_forecast_score(
        probabilities.reindex(features.index).astype(float),
        config,
    )
    aligned_return_forecast = (
        return_forecasts.reindex(features.index).astype(float)
        if return_forecasts is not None
        else pd.Series(np.nan, index=features.index)
    )
    confidence = probability_confidence_frame(
        aligned_probability,
        window=config.confidence_window,
        z=config.confidence_z,
    )
    floor = config.interval_probability_floor
    if floor is None:
        floor = max(0.05, min(0.95, float(threshold) * float(config.forecast_hysteresis_exit_ratio)))

    prior_context = features.get("prior_down_context", pd.Series(0.0, index=features.index))
    context_seen = prior_context.fillna(0.0).rolling(
        config.label_lookback,
        min_periods=1,
    ).max()
    setup_lookup = support_resistance_breakout_setups(data=data, features=features, config=config)
    sr_breakout_mask = support_resistance_breakout_mask(features.index, setup_lookup)
    forecast_turn_filter = return_forecast_turn_filter(
        forecast_score=aligned_probability,
        return_forecast=aligned_return_forecast,
        threshold=float(threshold),
        floor=float(floor),
        config=config,
    )
    candidate_signal = (
        (aligned_probability >= floor)
        | (
            (confidence["upper"] >= float(threshold))
            & (confidence["mean"] >= floor * 0.85)
        )
    )
    candidate_mask = (
        candidate_signal.fillna(False)
        & (context_seen > 0)
        & sr_breakout_mask
        & forecast_turn_filter.fillna(False)
    )
    candidate_scores = aligned_probability[candidate_mask].dropna()
    if candidate_scores.empty:
        return empty_interval_frame()

    selected_breakouts = non_max_suppression_1d(
        candidate_scores,
        cooldown_bars=config.cooldown_bars,
        all_index=features.index,
    )
    rows: list[dict[str, Any]] = []
    train_stats = make_train_feature_stats(train_features)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    atr = features["atr"].reindex(data.index).to_numpy(dtype=float)
    min_bars = max(1, int(config.min_interval_bars))
    max_bars = max(min_bars, int(config.max_interval_bars))

    for breakout_timestamp in selected_breakouts:
        breakout_idx = int(features.index.get_loc(breakout_timestamp))
        setup = setup_lookup.get(breakout_idx)
        if setup is None:
            continue

        start_idx = int(setup["start_idx"])
        end_idx = int(setup["breakout_idx"])
        if end_idx - start_idx + 1 < min_bars:
            continue
        if end_idx - start_idx + 1 > max_bars:
            start_idx = max(0, end_idx - max_bars + 1)

        interval_probability = aligned_probability.iloc[start_idx : end_idx + 1].dropna()
        if interval_probability.empty:
            continue
        interval_forecast = aligned_return_forecast.iloc[start_idx : end_idx + 1].dropna()
        probability_mean = float(interval_probability.mean())
        probability_max = float(interval_probability.max())
        probability_peak_timestamp = interval_probability.idxmax()
        probability_peak_idx = int(features.index.get_loc(probability_peak_timestamp))
        forecast_mean = float(interval_forecast.mean()) if not interval_forecast.empty else np.nan
        forecast_max = float(interval_forecast.max()) if not interval_forecast.empty else np.nan
        forecast_peak = (
            float(aligned_return_forecast.iloc[probability_peak_idx])
            if np.isfinite(aligned_return_forecast.iloc[probability_peak_idx])
            else np.nan
        )
        interval_confidence_low, interval_confidence_high = probability_interval_bounds(
            interval_probability,
            z=config.confidence_z,
        )
        if probability_max < float(floor) and interval_confidence_high < float(threshold):
            continue

        support_low = float(setup["support_zone_low"])
        support_high = float(setup["support_zone_high"])
        resistance_low = float(setup["resistance_zone_low"])
        resistance_high = float(setup["resistance_zone_high"])
        if not all(np.isfinite(value) for value in [support_low, support_high, resistance_low, resistance_high]):
            continue

        entry_price = float(close[end_idx])
        interval_atr = finite_mean(atr[start_idx : end_idx + 1])
        if not np.isfinite(interval_atr) or interval_atr <= 0:
            interval_atr = max((resistance_high - support_low) * 0.10, abs(entry_price) * 0.001, 1e-12)
        price_pad = max(0.0, float(config.interval_expand_atr)) * interval_atr
        target_price = entry_price + config.target_atr * interval_atr
        stop_price = float(setup["support_zone_low"]) - config.stop_atr * interval_atr * 0.35
        peak_label = labels.loc[breakout_timestamp] if breakout_timestamp in labels.index else pd.Series(dtype=object)
        hit_idx = peak_label.get("label_hit_bar_idx", np.nan)
        realized_label = peak_label.get("reversal_label", np.nan)
        top_drivers = local_driver_summary(
            features.loc[breakout_timestamp, feature_columns],
            feature_importance,
            train_stats,
            top_n=6,
        )

        volume_z_max = finite_max(series_values(features, "volume_z_20", start_idx, end_idx))
        volume_ratio_max = finite_max(series_values(features, "volume_ratio_20", start_idx, end_idx))
        volatility_expansion = finite_mean(
            series_values(features, "volatility_expansion_5_20", start_idx, end_idx)
        )
        volatility_compression = finite_mean(
            series_values(features, "volatility_compression_20_80", start_idx, end_idx)
        )
        range_compression = finite_mean(
            series_values(features, "range_compression_20_80", start_idx, end_idx)
        )
        context_ratio = float(prior_context.iloc[start_idx : end_idx + 1].fillna(0.0).mean())
        interval_score = score_interval(
            probability_mean=probability_mean,
            probability_max=probability_max,
            context_ratio=context_ratio,
            volume_z_max=volume_z_max,
            volatility_expansion=volatility_expansion,
            volatility_compression=volatility_compression,
            confidence_low=interval_confidence_low,
            confidence_high=interval_confidence_high,
            support_touches=float(setup["support_touches"]),
            resistance_touches=float(setup["resistance_touches"]),
            breakout_strength_atr=float(setup["breakout_strength_atr"]),
            base_range_atr=float(setup["base_range_atr"]),
        )

        rows.append(
            {
                "StartBarIdx": int(start_idx),
                "EndBarIdx": int(end_idx),
                "PeakBarIdx": int(probability_peak_idx),
                "ProbabilityPeakBarIdx": int(probability_peak_idx),
                "TurnBarIdx": int(setup["support_anchor_idx"]),
                "SwitchKind": "sr_breakout_base",
                "StartDate": data.index[start_idx],
                "EndDate": data.index[end_idx],
                "PeakDate": probability_peak_timestamp,
                "ProbabilityPeakDate": probability_peak_timestamp,
                "TurnDate": data.index[int(setup["support_anchor_idx"])],
                "BreakoutBarIdx": int(end_idx),
                "BreakoutDate": data.index[end_idx],
                "BreakoutLevel": float(setup["breakout_level"]),
                "Bars": int(end_idx - start_idx + 1),
                "SupportLow": support_low,
                "ResistanceHigh": resistance_high,
                "SupportZoneLow": support_low,
                "SupportZoneHigh": support_high,
                "SupportLevel": float(setup["support_level"]),
                "ResistanceZoneLow": resistance_low,
                "ResistanceZoneHigh": resistance_high,
                "ResistanceLevel": float(setup["resistance_level"]),
                "SupportTouches": int(setup["support_touches"]),
                "ResistanceTouches": int(setup["resistance_touches"]),
                "BaseRangeAtr": float(setup["base_range_atr"]),
                "PriorDrawdownPct": float(setup["prior_drawdown_pct"]),
                "BreakoutStrengthAtr": float(setup["breakout_strength_atr"]),
                "BreakoutVolumeZ": float(setup["breakout_volume_z"]),
                "BoxLow": support_low - price_pad,
                "BoxHigh": max(resistance_high, float(setup["breakout_level"])) + price_pad,
                "EntryBarIdx": int(end_idx),
                "EntryDate": data.index[end_idx],
                "EntryPrice": entry_price,
                "TargetPrice": target_price,
                "StopPrice": stop_price,
                "RiskReward": (target_price - entry_price) / max(entry_price - stop_price, 1e-12),
                "ReturnForecastPctMean": forecast_mean,
                "ReturnForecastPctMax": forecast_max,
                "ReturnForecastPctPeak": forecast_peak,
                "ForecastScoreMean": probability_mean,
                "ForecastScoreMax": probability_max,
                "ProbabilityMean": probability_mean,
                "ProbabilityMax": probability_max,
                "ProbabilityConfidenceLow": interval_confidence_low,
                "ProbabilityConfidenceHigh": interval_confidence_high,
                "SignalThreshold": float(threshold),
                "VolumeZMax": volume_z_max,
                "VolumeRatioMax": volume_ratio_max,
                "VolatilityExpansionMean": volatility_expansion,
                "VolatilityCompressionMean": volatility_compression,
                "RangeCompressionMean": range_compression,
                "PriorDownContextRatio": context_ratio,
                "IntervalScore": interval_score,
                "OutcomeKnown": is_finite_number(realized_label),
                "OutcomeLabel": realized_label,
                "HitBarIdx": int(hit_idx) if is_finite_number(hit_idx) else np.nan,
                "HitDate": peak_label.get("label_hit_date", pd.NaT),
                "HitType": peak_label.get("label_hit_type", None),
                "ShapeHint": classify_shape_hint(data, probability_peak_idx),
                "TopDrivers": top_drivers,
            }
        )

    if not rows:
        return empty_interval_frame()
    rows = suppress_overlapping_interval_rows(rows)
    return pd.DataFrame(rows).sort_values("StartBarIdx").reset_index(drop=True)


def smooth_forecast_score(score: pd.Series, config: ReversalMLConfig) -> pd.Series:
    values = score.astype(float)
    span = max(1, int(config.forecast_smooth_span))
    if span <= 1:
        return values
    return values.ewm(span=span, adjust=False, min_periods=1).mean()


def return_forecast_turn_filter(
    *,
    forecast_score: pd.Series,
    return_forecast: pd.Series,
    threshold: float,
    floor: float,
    config: ReversalMLConfig,
) -> pd.Series:
    cross_lookback = max(1, int(config.forecast_cross_lookback))
    score_cross = (forecast_score >= float(threshold)) & (forecast_score.shift(cross_lookback) <= float(floor))
    score_rising = forecast_score.diff(cross_lookback) >= float(config.forecast_min_score_change) * 0.50
    if return_forecast.dropna().empty:
        return (score_cross | score_rising).fillna(False)

    smoothed_return = return_forecast.rolling(
        max(2, int(config.forecast_smooth_span)),
        min_periods=1,
    ).median()
    return_turning = (
        (smoothed_return > 0)
        | (smoothed_return.diff(cross_lookback) > 0)
        | score_cross
    )
    return (return_turning | score_rising).fillna(False)


def support_resistance_breakout_mask(
    index: pd.Index,
    setups: dict[int, dict[str, Any]],
) -> pd.Series:
    mask = pd.Series(False, index=index)
    for breakout_idx in setups:
        if 0 <= int(breakout_idx) < len(mask):
            mask.iloc[int(breakout_idx)] = True
    return mask


def support_resistance_breakout_setups(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    config: ReversalMLConfig,
) -> dict[int, dict[str, Any]]:
    setups: dict[int, dict[str, Any]] = {}
    min_base_bars = max(int(config.sr_min_base_bars), int(config.min_interval_bars), 4)
    for breakout_idx in range(min_base_bars + 1, len(data)):
        setup = find_support_resistance_breakout_setup(
            data=data,
            features=features,
            breakout_idx=breakout_idx,
            config=config,
        )
        if setup is not None:
            setups[int(breakout_idx)] = setup
    return setups


def find_support_resistance_breakout_setup(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    breakout_idx: int,
    config: ReversalMLConfig,
) -> dict[str, Any] | None:
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    atr = features["atr"].reindex(data.index).to_numpy(dtype=float)
    volume_z = features.get("volume_z_20", pd.Series(np.nan, index=features.index)).reindex(data.index).to_numpy(
        dtype=float
    )

    if breakout_idx <= 1 or breakout_idx >= len(data):
        return None
    atr_value = finite_mean(atr[max(0, breakout_idx - 10) : breakout_idx + 1])
    if not np.isfinite(atr_value) or atr_value <= 0:
        recent_range = finite_mean(
            high[max(0, breakout_idx - 10) : breakout_idx + 1]
            - low[max(0, breakout_idx - 10) : breakout_idx + 1]
        )
        if not np.isfinite(recent_range) or recent_range <= 0:
            recent_range = 0.0
        atr_value = max(recent_range, abs(close[breakout_idx]) * 0.001, 1e-12)

    min_base_bars = max(int(config.sr_min_base_bars), int(config.min_interval_bars), 4)
    lookback = max(min_base_bars + 2, int(config.sr_lookback_bars))
    pre_end = int(breakout_idx) - 1
    search_start = max(0, int(breakout_idx) - lookback)
    if pre_end - search_start + 1 < min_base_bars:
        return None

    window_low = low[search_start : pre_end + 1]
    window_high = high[search_start : pre_end + 1]
    if len(window_low) == 0 or not np.isfinite(window_low).any() or not np.isfinite(window_high).any():
        return None

    support_floor = finite_min(window_low)
    if not np.isfinite(support_floor):
        return None

    price_reference = max(abs(float(close[breakout_idx])), 1e-12)
    touch_tolerance = max(float(config.sr_touch_tolerance_atr) * atr_value, price_reference * 0.001)
    support_mask = window_low <= support_floor + touch_tolerance
    support_touch_positions = np.flatnonzero(support_mask)
    if len(support_touch_positions) == 0:
        return None
    support_touches = count_touch_clusters(
        support_mask,
        min_gap=max(1, int(config.sr_touch_cluster_gap_bars)),
    )
    if support_touches < max(1, int(config.sr_min_support_touches)):
        return None

    support_anchor_idx = search_start + int(support_touch_positions[np.argmin(window_low[support_touch_positions])])
    first_support_idx = search_start + int(support_touch_positions[0])
    base_start = max(search_start, first_support_idx - int(config.sr_base_left_pad_bars))
    if pre_end - base_start + 1 < min_base_bars:
        base_start = max(search_start, pre_end - min_base_bars + 1)

    base_low = low[base_start : pre_end + 1]
    base_high = high[base_start : pre_end + 1]
    if len(base_low) < min_base_bars:
        return None

    support_candidates = base_low[base_low <= support_floor + touch_tolerance]
    if len(support_candidates) == 0:
        return None
    support_level = float(np.nanmedian(support_candidates))
    zone_width_limit = max(float(config.sr_zone_max_width_atr) * atr_value, price_reference * 0.001)
    zone_half_width = max(price_reference * 0.0005, min(zone_width_limit * 0.50, touch_tolerance * 0.60))
    support_zone_low = support_level - zone_half_width
    support_zone_high = support_level + zone_half_width

    high_quantile = float(np.nanquantile(base_high, 0.82))
    resistance_candidates = base_high[base_high >= high_quantile - touch_tolerance]
    if len(resistance_candidates) == 0:
        return None
    resistance_level = float(np.nanmedian(resistance_candidates))
    resistance_zone_low = resistance_level - zone_half_width
    resistance_zone_high = resistance_level + zone_half_width
    resistance_touches = count_touch_clusters(
        base_high >= resistance_level - touch_tolerance,
        min_gap=max(1, int(config.sr_touch_cluster_gap_bars)),
    )
    if resistance_touches < max(1, int(config.sr_min_resistance_touches)):
        return None

    base_height = resistance_level - support_level
    base_range_atr = base_height / max(atr_value, 1e-12)
    if base_range_atr < float(config.sr_min_height_atr):
        return None
    if base_range_atr > float(config.sr_max_height_atr):
        return None

    breakout_level = resistance_zone_high + float(config.sr_breakout_buffer_atr) * atr_value
    if close[breakout_idx] < breakout_level:
        return None
    if close[breakout_idx - 1] > breakout_level:
        return None
    breakout_strength_atr = (close[breakout_idx] - resistance_level) / max(atr_value, 1e-12)
    if breakout_strength_atr > float(config.sr_max_breakout_extension_atr):
        return None

    breakout_volume_z = float(volume_z[breakout_idx]) if np.isfinite(volume_z[breakout_idx]) else np.nan
    if np.isfinite(breakout_volume_z) and breakout_volume_z < float(config.sr_breakout_volume_z_min):
        return None

    prior_start = max(0, base_start - int(config.sr_prior_downtrend_window))
    prior_high = finite_max(high[prior_start : support_anchor_idx + 1])
    if not np.isfinite(prior_high) or prior_high <= 0:
        return None
    prior_drawdown_pct = (prior_high - support_floor) / prior_high * 100.0
    if prior_drawdown_pct < float(config.min_prior_drawdown_pct):
        return None

    recent_base_closes = close[base_start : pre_end + 1]
    if finite_max(recent_base_closes) > breakout_level + touch_tolerance:
        return None

    return {
        "start_idx": int(base_start),
        "end_idx": int(pre_end),
        "breakout_idx": int(breakout_idx),
        "support_anchor_idx": int(support_anchor_idx),
        "support_level": float(support_level),
        "support_zone_low": float(support_zone_low),
        "support_zone_high": float(support_zone_high),
        "resistance_level": float(resistance_level),
        "resistance_zone_low": float(resistance_zone_low),
        "resistance_zone_high": float(resistance_zone_high),
        "breakout_level": float(breakout_level),
        "support_touches": int(support_touches),
        "resistance_touches": int(resistance_touches),
        "base_range_atr": float(base_range_atr),
        "breakout_strength_atr": float(breakout_strength_atr),
        "breakout_volume_z": float(breakout_volume_z),
        "prior_drawdown_pct": float(prior_drawdown_pct),
    }


def count_touch_clusters(mask: np.ndarray | pd.Series, *, min_gap: int) -> int:
    values = np.asarray(mask, dtype=bool)
    count = 0
    previous_touch_idx = -10_000
    in_cluster = False
    for idx, active in enumerate(values):
        if active and not in_cluster:
            if idx - previous_touch_idx >= int(min_gap):
                count += 1
            previous_touch_idx = idx
            in_cluster = True
        elif not active:
            in_cluster = False
    return count


def forecast_breakout_filter(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    forecast_score: pd.Series,
    return_forecast: pd.Series,
    threshold: float,
    floor: float,
    config: ReversalMLConfig,
) -> pd.Series:
    close = data["close"].astype(float).reindex(features.index)
    high = data["high"].astype(float).reindex(features.index)
    low = data["low"].astype(float).reindex(features.index)
    atr = features["atr"].astype(float).reindex(features.index)
    lookback = max(2, int(config.breakout_lookback))
    cross_lookback = max(1, int(config.forecast_cross_lookback))

    prior_resistance = high.rolling(lookback, min_periods=max(2, lookback // 2)).max().shift(1)
    reclaim_buffer = max(0.0, float(config.breakout_buffer_atr)) * atr
    resistance_reclaim = close >= (prior_resistance - reclaim_buffer)

    reclaim_column = f"reclaim_from_low_atr_{8 if 'reclaim_from_low_atr_8' in features.columns else 5}"
    low_position_column = f"low_range_position_{13 if 'low_range_position_13' in features.columns else 8}"
    valley_rebound = features.get(reclaim_column, pd.Series(np.nan, index=features.index)).astype(float) >= (
        config.switch_min_rebound_atr * 0.75
    )
    near_low = (
        features.get(
            f"dist_to_low_atr_{config.label_lookback}",
            pd.Series(np.nan, index=features.index),
        ).astype(float)
        <= config.max_distance_to_low_atr * 1.25
    ) | (
        features.get(low_position_column, pd.Series(np.nan, index=features.index)).astype(float) <= 0.65
    )

    score_cross = (forecast_score >= threshold) & (forecast_score.shift(cross_lookback) <= floor)
    score_rising = forecast_score.diff(cross_lookback) >= float(config.forecast_min_score_change)
    forecast_turning_positive = (
        (return_forecast > 0)
        | (return_forecast.diff(cross_lookback) > 0)
        | score_cross
    )
    return (
        (resistance_reclaim | valley_rebound | score_cross | score_rising)
        & near_low.fillna(False)
        & forecast_turning_positive.fillna(False)
    )


def empty_signal_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "ReturnForecastPct",
            "ForecastScore",
            "ReversalProbability",
            "SignalThreshold",
            "SupportZoneLow",
            "SupportZoneHigh",
            "ResistanceZoneLow",
            "ResistanceZoneHigh",
            "BreakoutLevel",
            "SupportTouches",
            "ResistanceTouches",
            "BaseRangeAtr",
            "BreakoutStrengthAtr",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "SwitchKind",
            "SwitchTurnBarIdx",
            "SwitchTurnDate",
            "SwitchConfirmBarIdx",
            "SwitchConfirmDate",
            "BarsFromSwitchTurn",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def empty_interval_frame() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "StartBarIdx",
            "EndBarIdx",
            "PeakBarIdx",
            "ProbabilityPeakBarIdx",
            "TurnBarIdx",
            "SwitchKind",
            "StartDate",
            "EndDate",
            "PeakDate",
            "ProbabilityPeakDate",
            "TurnDate",
            "BreakoutBarIdx",
            "BreakoutDate",
            "BreakoutLevel",
            "Bars",
            "SupportLow",
            "ResistanceHigh",
            "SupportZoneLow",
            "SupportZoneHigh",
            "SupportLevel",
            "ResistanceZoneLow",
            "ResistanceZoneHigh",
            "ResistanceLevel",
            "SupportTouches",
            "ResistanceTouches",
            "BaseRangeAtr",
            "PriorDrawdownPct",
            "BreakoutStrengthAtr",
            "BreakoutVolumeZ",
            "BoxLow",
            "BoxHigh",
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "TargetPrice",
            "StopPrice",
            "RiskReward",
            "ReturnForecastPctMean",
            "ReturnForecastPctMax",
            "ReturnForecastPctPeak",
            "ForecastScoreMean",
            "ForecastScoreMax",
            "ProbabilityMean",
            "ProbabilityMax",
            "ProbabilityConfidenceLow",
            "ProbabilityConfidenceHigh",
            "SignalThreshold",
            "VolumeZMax",
            "VolumeRatioMax",
            "VolatilityExpansionMean",
            "VolatilityCompressionMean",
            "RangeCompressionMean",
            "PriorDownContextRatio",
            "IntervalScore",
            "OutcomeKnown",
            "OutcomeLabel",
            "HitBarIdx",
            "HitDate",
            "HitType",
            "ShapeHint",
            "TopDrivers",
        ]
    )


def non_max_suppression_1d(
    probability: pd.Series,
    *,
    cooldown_bars: int,
    all_index: pd.Index,
) -> list[pd.Timestamp]:
    ranked = probability.sort_values(ascending=False)
    selected_positions: list[int] = []
    selected_timestamps: list[pd.Timestamp] = []

    for timestamp, _ in ranked.items():
        position = int(all_index.get_loc(timestamp))
        if any(abs(position - selected) <= cooldown_bars for selected in selected_positions):
            continue
        selected_positions.append(position)
        selected_timestamps.append(timestamp)

    selected_timestamps.sort(key=lambda ts: int(all_index.get_loc(ts)))
    return selected_timestamps


def probability_confidence_frame(probability: pd.Series, *, window: int, z: float) -> pd.DataFrame:
    values = probability.astype(float)
    window = max(2, int(window))
    mean = values.rolling(window, min_periods=max(2, window // 4)).mean()
    std = values.rolling(window, min_periods=max(2, window // 4)).std().fillna(0.0)
    count = values.rolling(window, min_periods=max(2, window // 4)).count().replace(0, np.nan)
    stderr = std / np.sqrt(count)
    return pd.DataFrame(
        {
            "mean": mean,
            "lower": (mean - float(z) * stderr).clip(0.0, 1.0),
            "upper": (mean + float(z) * stderr).clip(0.0, 1.0),
        },
        index=probability.index,
    )


def probability_interval_bounds(probability: pd.Series, *, z: float) -> tuple[float, float]:
    values = probability.dropna().to_numpy(dtype=float)
    if len(values) == 0:
        return np.nan, np.nan
    mean = float(np.mean(values))
    if len(values) == 1:
        return mean, mean
    stderr = float(np.std(values, ddof=1) / math.sqrt(len(values)))
    return max(0.0, mean - float(z) * stderr), min(1.0, mean + float(z) * stderr)


def boolean_segments(mask: np.ndarray) -> list[tuple[int, int]]:
    segments: list[tuple[int, int]] = []
    start: int | None = None
    for idx, active in enumerate(mask):
        if active and start is None:
            start = idx
        elif not active and start is not None:
            segments.append((start, idx - 1))
            start = None
    if start is not None:
        segments.append((start, len(mask) - 1))
    return segments


def merge_boolean_segments(
    segments: list[tuple[int, int]],
    *,
    max_gap: int,
) -> list[tuple[int, int]]:
    if not segments:
        return []
    merged: list[tuple[int, int]] = [segments[0]]
    for start, end in segments[1:]:
        previous_start, previous_end = merged[-1]
        if start - previous_end - 1 <= max_gap:
            merged[-1] = (previous_start, end)
        else:
            merged.append((start, end))
    return merged


def split_segment(
    start: int,
    end: int,
    *,
    max_bars: int,
    min_bars: int,
) -> list[tuple[int, int]]:
    length = end - start + 1
    if length <= max_bars:
        return [(start, end)]

    chunks: list[tuple[int, int]] = []
    overlap = max(min_bars, max_bars // 4)
    chunk_start = start
    while chunk_start <= end:
        chunk_end = min(end, chunk_start + max_bars - 1)
        chunks.append((chunk_start, chunk_end))
        if chunk_end >= end:
            break
        chunk_start = max(chunk_start + 1, chunk_end - overlap + 1)
    return chunks


def series_values(features: pd.DataFrame, column: str, start_idx: int, end_idx: int) -> np.ndarray:
    if column not in features.columns:
        return np.array([], dtype=float)
    return features[column].iloc[start_idx : end_idx + 1].to_numpy(dtype=float)


def is_finite_number(value: Any) -> bool:
    try:
        return bool(np.isfinite(float(value)))
    except (TypeError, ValueError):
        return False


def finite_mean(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.mean(array))


def finite_min(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.min(array))


def finite_max(values: np.ndarray | pd.Series) -> float:
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if len(array) == 0:
        return np.nan
    return float(np.max(array))


def score_interval(
    *,
    probability_mean: float,
    probability_max: float,
    context_ratio: float,
    volume_z_max: float,
    volatility_expansion: float,
    volatility_compression: float,
    confidence_low: float,
    confidence_high: float,
    support_touches: float = np.nan,
    resistance_touches: float = np.nan,
    breakout_strength_atr: float = np.nan,
    base_range_atr: float = np.nan,
) -> float:
    volume_score = np.tanh(max(0.0, 0.0 if not np.isfinite(volume_z_max) else volume_z_max) / 3.0)
    expansion_score = np.tanh(
        max(0.0, (0.0 if not np.isfinite(volatility_expansion) else volatility_expansion) - 1.0)
    )
    compression_score = 1.0 - min(
        1.0,
        max(0.0, 0.0 if not np.isfinite(volatility_compression) else volatility_compression),
    )
    confidence_width = max(
        0.0,
        (0.0 if not np.isfinite(confidence_high) else confidence_high)
        - (0.0 if not np.isfinite(confidence_low) else confidence_low),
    )
    stability_score = 1.0 - min(1.0, confidence_width)
    touch_total = (
        max(0.0, 0.0 if not np.isfinite(support_touches) else support_touches)
        + max(0.0, 0.0 if not np.isfinite(resistance_touches) else resistance_touches)
    )
    structure_score = min(1.0, touch_total / 5.0)
    breakout_score = np.tanh(max(0.0, 0.0 if not np.isfinite(breakout_strength_atr) else breakout_strength_atr))
    if np.isfinite(base_range_atr):
        height_score = 1.0 - min(1.0, abs(float(base_range_atr) - 3.0) / 6.0)
    else:
        height_score = 0.0
    score = (
        0.30 * probability_max
        + 0.20 * probability_mean
        + 0.12 * min(1.0, max(0.0, context_ratio))
        + 0.13 * structure_score
        + 0.09 * breakout_score
        + 0.06 * height_score
        + 0.04 * volume_score
        + 0.03 * expansion_score
        + 0.02 * compression_score
        + 0.01 * stability_score
    )
    return float(score)


def suppress_overlapping_interval_rows(
    rows: list[dict[str, Any]],
    *,
    max_overlap_ratio: float = 0.55,
) -> list[dict[str, Any]]:
    selected: list[dict[str, Any]] = []
    for row in sorted(rows, key=lambda item: float(item["IntervalScore"]), reverse=True):
        start = int(row["StartBarIdx"])
        end = int(row["EndBarIdx"])
        length = max(1, end - start + 1)
        overlaps = False
        for existing in selected:
            existing_start = int(existing["StartBarIdx"])
            existing_end = int(existing["EndBarIdx"])
            existing_length = max(1, existing_end - existing_start + 1)
            overlap = max(0, min(end, existing_end) - max(start, existing_start) + 1)
            if overlap / min(length, existing_length) >= max_overlap_ratio:
                overlaps = True
                break
        if not overlaps:
            selected.append(row)
    return selected


def make_train_feature_stats(train_features: pd.DataFrame) -> pd.DataFrame:
    stats = pd.DataFrame(index=train_features.columns)
    stats["median"] = train_features.median(numeric_only=True)
    q25 = train_features.quantile(0.25, numeric_only=True)
    q75 = train_features.quantile(0.75, numeric_only=True)
    stats["iqr"] = (q75 - q25).replace(0, np.nan)
    return stats


def local_driver_summary(
    row: pd.Series,
    feature_importance: pd.DataFrame,
    train_stats: pd.DataFrame,
    *,
    top_n: int,
) -> str:
    if feature_importance.empty:
        return ""

    importance_map = feature_importance.set_index("feature")["importance_mean"].to_dict()
    scored: list[tuple[float, str]] = []
    for feature, value in row.items():
        if feature not in importance_map or feature not in train_stats.index:
            continue
        if not np.isfinite(value):
            continue
        iqr = train_stats.loc[feature, "iqr"]
        median = train_stats.loc[feature, "median"]
        if not np.isfinite(iqr) or iqr == 0:
            continue
        robust_z = (float(value) - float(median)) / float(iqr)
        score = abs(float(importance_map[feature]) * robust_z)
        if score <= 0:
            continue
        scored.append((score, f"{feature}={float(value):.4g} z={robust_z:.2f}"))

    scored.sort(reverse=True)
    return " | ".join(text for _, text in scored[:top_n])


def classify_shape_hint(data: pd.DataFrame, bar_idx: int, *, lookback: int = 80) -> str:
    start = max(0, int(bar_idx) - lookback)
    if int(bar_idx) - start < 20:
        return "ML Bullish Reversal"

    window = data.iloc[start : int(bar_idx) + 1]
    lows = window["low"].to_numpy(dtype=float)
    highs = window["high"].to_numpy(dtype=float)
    close = window["close"].to_numpy(dtype=float)
    atr = float(average_true_range(window, period=14).iloc[-1])
    if not np.isfinite(atr) or atr <= 0:
        atr = max(float(np.nanstd(close)), 1e-12)

    pivot_lows = simple_pivots(lows, mode="low", order=3)
    pivot_highs = simple_pivots(highs, mode="high", order=3)
    if len(pivot_lows) >= 2:
        l1, l2 = pivot_lows[-2], pivot_lows[-1]
        low_diff_atr = abs(lows[l2] - lows[l1]) / atr
        if l2 > l1 and low_diff_atr <= 1.0:
            between_high = np.max(highs[l1:l2 + 1])
            if between_high - max(lows[l1], lows[l2]) >= atr:
                return "ML Double Bottom-like"

    if len(pivot_lows) >= 3 and len(pivot_highs) >= 2:
        left, head, right = pivot_lows[-3], pivot_lows[-2], pivot_lows[-1]
        if left < head < right and lows[head] < min(lows[left], lows[right]):
            shoulder_diff_atr = abs(lows[left] - lows[right]) / atr
            if shoulder_diff_atr <= 1.5:
                return "ML Inverse H&S-like"

    return "ML Bullish Reversal"


def simple_pivots(values: np.ndarray, *, mode: str, order: int) -> list[int]:
    pivots: list[int] = []
    for idx in range(order, len(values) - order):
        left = values[idx - order : idx]
        right = values[idx + 1 : idx + order + 1]
        if mode == "low" and values[idx] <= np.min(left) and values[idx] <= np.min(right):
            pivots.append(idx)
        if mode == "high" and values[idx] >= np.max(left) and values[idx] >= np.max(right):
            pivots.append(idx)
    return pivots


def fit_winsorize(
    X_train: pd.DataFrame,
    *,
    lower: float = 0.01,
    upper: float = 0.99,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    bounds = pd.DataFrame(index=X_train.columns)
    bounds["lower"] = X_train.quantile(lower, numeric_only=True)
    bounds["upper"] = X_train.quantile(upper, numeric_only=True)
    return apply_winsorize(X_train, bounds), bounds


def apply_winsorize(X: pd.DataFrame, bounds: pd.DataFrame) -> pd.DataFrame:
    out = X.copy()
    for column in out.columns:
        if column not in bounds.index:
            continue
        lower = bounds.loc[column, "lower"]
        upper = bounds.loc[column, "upper"]
        if np.isfinite(lower) and np.isfinite(upper) and lower < upper:
            out[column] = out[column].clip(lower=lower, upper=upper)
    return out


def average_true_range(data: pd.DataFrame, *, period: int = 14) -> pd.Series:
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    close = data["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [
            high - low,
            (high - previous_close).abs(),
            (low - previous_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    return true_range.rolling(period, min_periods=1).mean()


def rsi(close: pd.Series, *, period: int = 14) -> pd.Series:
    delta = close.diff()
    gains = delta.clip(lower=0.0)
    losses = -delta.clip(upper=0.0)
    avg_gain = gains.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    avg_loss = losses.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    rs = avg_gain / avg_loss.replace(0, np.nan)
    return 100.0 - (100.0 / (1.0 + rs))


def macd(close: pd.Series) -> tuple[pd.Series, pd.Series, pd.Series]:
    fast = close.ewm(span=12, adjust=False).mean()
    slow = close.ewm(span=26, adjust=False).mean()
    line = fast - slow
    signal = line.ewm(span=9, adjust=False).mean()
    return line, signal, line - signal


def stochastic_k(
    high: pd.Series,
    low: pd.Series,
    close: pd.Series,
    *,
    period: int,
) -> pd.Series:
    lowest = low.rolling(period).min()
    highest = high.rolling(period).max()
    return (close - lowest) / (highest - lowest).replace(0, np.nan) * 100.0


def rolling_zscore(series: pd.Series, *, window: int) -> pd.Series:
    mean = series.rolling(window).mean()
    std = series.rolling(window).std()
    return (series - mean) / std.replace(0, np.nan)


def rolling_slope_pct(series: pd.Series, *, window: int) -> pd.Series:
    x = np.arange(window, dtype=float)
    x = x - x.mean()

    def slope(values: np.ndarray) -> float:
        if np.isnan(values).any():
            return np.nan
        y = values.astype(float)
        y = y - y.mean()
        denom = float(np.dot(x, x))
        if denom == 0:
            return np.nan
        raw_slope = float(np.dot(x, y) / denom)
        return raw_slope / max(abs(float(values[-1])), 1e-12) * 100.0

    return series.rolling(window).apply(slope, raw=True)


def signed_streak(condition: pd.Series) -> pd.Series:
    values = condition.fillna(False).to_numpy(dtype=bool)
    streaks = np.zeros(len(values), dtype=float)
    current = 0
    previous = None
    for idx, value in enumerate(values):
        if previous is None or value != previous:
            current = 1
        else:
            current += 1
        streaks[idx] = current if value else -current
        previous = value
    return pd.Series(streaks, index=condition.index)


def on_balance_volume(close: pd.Series, volume: pd.Series) -> pd.Series:
    direction = np.sign(close.diff()).fillna(0.0)
    return (direction * volume).cumsum()


def walk_forward_arima_features(
    returns: pd.Series,
    *,
    order: tuple[int, int, int],
    window: int,
) -> pd.DataFrame:
    columns = ["arima_forecast_return", "arima_residual", "arima_residual_z_40"]
    out = pd.DataFrame(np.nan, index=returns.index, columns=columns)
    if ARIMA is None:
        warnings.warn("statsmodels unavailable; ARIMA features skipped.")
        return out

    values = returns.astype(float)
    for idx in range(window, len(values)):
        history = values.iloc[idx - window : idx].dropna()
        if len(history) < max(30, sum(order) * 10):
            continue
        try:
            with warnings.catch_warnings():
                warnings.simplefilter("ignore")
                result = ARIMA(history, order=order).fit()
            forecast = float(result.forecast(1).iloc[0])
            residual = float(values.iloc[idx] - forecast)
            out.iloc[idx, out.columns.get_loc("arima_forecast_return")] = forecast
            out.iloc[idx, out.columns.get_loc("arima_residual")] = residual
        except Exception:
            continue
    out["arima_residual_z_40"] = rolling_zscore(out["arima_residual"], window=40)
    return out


def save_outputs(
    data: pd.DataFrame,
    result: dict[str, Any],
    config: ReversalMLConfig,
) -> Path:
    run_name = f"{config.symbol.replace('.', '_')}_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
    run_dir = Path(config.output_dir) / run_name
    run_dir.mkdir(parents=True, exist_ok=True)
    if go is None:
        warnings.warn("plotly unavailable; HTML plot files skipped. Install with: pip install plotly")

    result["signals"].to_csv(run_dir / "signals.csv", index=False)
    result["intervals"].to_csv(run_dir / "pattern_intervals.csv", index=False)
    result["model_report"].to_csv(run_dir / "model_report.csv", index=False)
    result["feature_importance"].to_csv(run_dir / "feature_importance.csv", index=False)
    result["hmm_state_summary"].to_csv(run_dir / "hmm_state_summary.csv", index=False)
    result["pca_loadings"].to_csv(run_dir / "pca_loadings.csv", index=False)

    probability_frame = pd.DataFrame(
        {
            "Date": result["probabilities"].index,
            "ForecastScore": result["probabilities"].to_numpy(dtype=float),
            "ReversalProbability": result["probabilities"].to_numpy(dtype=float),
            "ReturnForecastPct": result["forecasts"].reindex(result["probabilities"].index).to_numpy(dtype=float),
        }
    )
    probability_frame.to_csv(run_dir / "probabilities.csv", index=False)
    probability_frame.to_csv(run_dir / "return_forecasts.csv", index=False)

    all_probability_frame = pd.DataFrame({"Date": data.index})
    for model_name, probabilities in result["per_model_probabilities"].items():
        all_probability_frame[f"{model_name}_forecast_score"] = probabilities.reindex(data.index).to_numpy(dtype=float)
        all_probability_frame[f"{model_name}_return_forecast_pct"] = (
            result["per_model_forecasts"][model_name].reindex(data.index).to_numpy(dtype=float)
        )
    all_probability_frame.to_csv(run_dir / "all_model_probabilities.csv", index=False)
    all_probability_frame.to_csv(run_dir / "all_model_return_forecasts.csv", index=False)

    hmm_regimes = make_hmm_regime_frame(result["features"])
    hmm_regimes.to_csv(run_dir / "hmm_regimes.csv", index=False)

    all_interval_frames = [
        intervals
        for intervals in result["per_model_intervals"].values()
        if intervals is not None and not intervals.empty
    ]
    if all_interval_frames:
        pd.concat(all_interval_frames, ignore_index=True).to_csv(
            run_dir / "all_model_pattern_intervals.csv",
            index=False,
        )
    else:
        empty_interval_frame().to_csv(run_dir / "all_model_pattern_intervals.csv", index=False)

    models_dir = run_dir / "models"
    models_dir.mkdir(exist_ok=True)
    for model_name, probabilities in result["per_model_probabilities"].items():
        safe_model_name = safe_filename(model_name)
        forecasts = result["per_model_forecasts"][model_name]
        signals = result["per_model_signals"].get(model_name, empty_signal_frame())
        importance = result["per_model_feature_importance"].get(
            model_name,
            pd.DataFrame(columns=["feature", "importance_mean", "importance_std", "rank"]),
        )
        threshold = result["per_model_thresholds"].get(model_name, np.nan)
        intervals = result["per_model_intervals"].get(model_name, empty_interval_frame())
        signals.to_csv(models_dir / f"{safe_model_name}_signals.csv", index=False)
        intervals.to_csv(models_dir / f"{safe_model_name}_pattern_intervals.csv", index=False)
        importance.to_csv(models_dir / f"{safe_model_name}_feature_importance.csv", index=False)
        pd.DataFrame(
            {
                "Date": data.index,
                "ForecastScore": probabilities.reindex(data.index).to_numpy(dtype=float),
                "ReturnForecastPct": forecasts.reindex(data.index).to_numpy(dtype=float),
            }
        ).to_csv(models_dir / f"{safe_model_name}_return_forecast.csv", index=False)
        make_plot(
            data=data,
            probabilities=probabilities,
            return_forecasts=forecasts,
            signals=signals,
            intervals=intervals,
            title=f"{config.symbol} {model_name} ML Reversal Signals",
            output_path=models_dir / f"{safe_model_name}_plot.html",
            model_name=model_name,
            threshold=threshold,
            confidence_window=config.confidence_window,
            confidence_z=config.confidence_z,
        )
        make_feature_importance_plot(
            importance=importance,
            title=f"{config.symbol} {model_name} XAI Feature Importance",
            output_path=models_dir / f"{safe_model_name}_feature_importance.html",
        )

    with open(run_dir / "config.json", "w", encoding="utf-8") as handle:
        json.dump(
            {
                **asdict(config),
                "best_model_name": result["best_model_name"],
                "threshold": result["threshold"],
                "return_threshold_pct": result["return_threshold_pct"],
                "forecast_score_scale": result["forecast_score_scale"],
            },
            handle,
            indent=2,
            default=str,
        )

    plot_path = run_dir / "reversal_ml_plot.html"
    make_plot(
        data=data,
        probabilities=result["probabilities"],
        return_forecasts=result["forecasts"],
        signals=result["signals"],
        intervals=result["intervals"],
        title=f"{config.symbol} ML Reversal Signals",
        output_path=plot_path,
        model_name=f"best: {result['best_model_name']}",
        threshold=result["threshold"],
        confidence_window=config.confidence_window,
        confidence_z=config.confidence_z,
    )
    make_model_comparison_plot(
        data=data,
        per_model_probabilities=result["per_model_probabilities"],
        per_model_forecasts=result["per_model_forecasts"],
        per_model_signals=result["per_model_signals"],
        per_model_intervals=result["per_model_intervals"],
        per_model_thresholds=result["per_model_thresholds"],
        title=f"{config.symbol} ML Reversal Model Comparison",
        output_path=run_dir / "all_models_comparison_plot.html",
    )
    make_hmm_regime_plot(
        data=data,
        features=result["features"],
        state_summary=result["hmm_state_summary"],
        title=f"{config.symbol} HMM Regime Map",
        output_path=run_dir / "hmm_regime_plot.html",
    )
    return run_dir


def safe_filename(value: str) -> str:
    safe = "".join(character if character.isalnum() or character in {"-", "_"} else "_" for character in value)
    return safe.strip("_") or "model"


def hex_to_rgba(hex_color: str, alpha: float) -> str:
    value = hex_color.lstrip("#")
    if len(value) != 6:
        return f"rgba(56, 189, 248, {alpha})"
    red, green, blue = (int(value[idx : idx + 2], 16) for idx in (0, 2, 4))
    return f"rgba({red}, {green}, {blue}, {alpha})"


def add_interval_boxes(
    fig: Any,
    intervals: pd.DataFrame | None,
    *,
    color: str,
    label: str,
    opacity: float,
    show_text: bool,
) -> None:
    if intervals is None or intervals.empty:
        return

    center_x: list[float] = []
    center_y: list[float] = []
    hover_text: list[str] = []
    visible_text: list[str] = []
    for interval_number, (_, row) in enumerate(intervals.iterrows(), start=1):
        try:
            start_idx = int(row["StartBarIdx"])
            end_idx = int(row["EndBarIdx"])
            box_low = float(row["BoxLow"])
            box_high = float(row["BoxHigh"])
        except (KeyError, TypeError, ValueError):
            continue
        if not all(np.isfinite(value) for value in [start_idx, end_idx, box_low, box_high]):
            continue

        fig.add_shape(
            type="rect",
            xref="x",
            yref="y",
            x0=start_idx - 0.5,
            x1=end_idx + 0.5,
            y0=box_low,
            y1=box_high,
            fillcolor=hex_to_rgba(color, opacity),
            line=dict(color=color, width=2),
            layer="below",
        )
        if {
            "SupportZoneLow",
            "SupportZoneHigh",
            "ResistanceZoneLow",
            "ResistanceZoneHigh",
        }.issubset(intervals.columns):
            support_low = float(row.get("SupportZoneLow", np.nan))
            support_high = float(row.get("SupportZoneHigh", np.nan))
            resistance_low = float(row.get("ResistanceZoneLow", np.nan))
            resistance_high = float(row.get("ResistanceZoneHigh", np.nan))
            if np.isfinite(support_low) and np.isfinite(support_high) and support_high >= support_low:
                fig.add_shape(
                    type="rect",
                    xref="x",
                    yref="y",
                    x0=start_idx - 0.5,
                    x1=end_idx + 0.5,
                    y0=support_low,
                    y1=support_high,
                    fillcolor="rgba(34, 197, 94, 0.18)",
                    line=dict(color="#22C55E", width=1),
                    layer="below",
                )
            if np.isfinite(resistance_low) and np.isfinite(resistance_high) and resistance_high >= resistance_low:
                fig.add_shape(
                    type="rect",
                    xref="x",
                    yref="y",
                    x0=start_idx - 0.5,
                    x1=end_idx + 0.5,
                    y0=resistance_low,
                    y1=resistance_high,
                    fillcolor="rgba(248, 113, 113, 0.16)",
                    line=dict(color="#F87171", width=1),
                    layer="below",
                )
        breakout_level = float(row.get("BreakoutLevel", np.nan))
        if np.isfinite(breakout_level):
            fig.add_shape(
                type="line",
                xref="x",
                yref="y",
                x0=start_idx - 0.5,
                x1=end_idx + 0.5,
                y0=breakout_level,
                y1=breakout_level,
                line=dict(color="#E5E7EB", width=2),
                layer="above",
            )
        center_x.append((start_idx + end_idx) / 2)
        center_y.append(box_high)
        probability_max = float(row.get("ProbabilityMax", np.nan))
        probability_mean = float(row.get("ProbabilityMean", np.nan))
        visible_text.append(f"SR Box {interval_number} {probability_max:.2f}")
        hover_text.append(
            (
                f"{label} interval<br>"
                f"Start: {row.get('StartDate')}<br>"
                f"End: {row.get('EndDate')}<br>"
                f"Turn: {row.get('TurnDate')}<br>"
                f"Breakout: {row.get('BreakoutDate')}<br>"
                f"Breakout level: {float(row.get('BreakoutLevel', np.nan)):.4f}<br>"
                f"Support zone: {float(row.get('SupportZoneLow', np.nan)):.4f}"
                f" - {float(row.get('SupportZoneHigh', np.nan)):.4f}<br>"
                f"Resistance zone: {float(row.get('ResistanceZoneLow', np.nan)):.4f}"
                f" - {float(row.get('ResistanceZoneHigh', np.nan)):.4f}<br>"
                f"Touches S/R: {row.get('SupportTouches')}/{row.get('ResistanceTouches')}<br>"
                f"Probability peak: {row.get('ProbabilityPeakDate', row.get('PeakDate'))}<br>"
                f"Switch kind: {row.get('SwitchKind')}<br>"
                f"Return forecast peak: {float(row.get('ReturnForecastPctPeak', np.nan)):.3f}%<br>"
                f"Return forecast mean: {float(row.get('ReturnForecastPctMean', np.nan)):.3f}%<br>"
                f"Mean forecast score: {probability_mean:.3f}<br>"
                f"Max forecast score: {probability_max:.3f}<br>"
                f"Score CI: {float(row.get('ProbabilityConfidenceLow', np.nan)):.3f}"
                f" - {float(row.get('ProbabilityConfidenceHigh', np.nan)):.3f}<br>"
                f"Volume z max: {float(row.get('VolumeZMax', np.nan)):.3f}<br>"
                f"Vol expansion: {float(row.get('VolatilityExpansionMean', np.nan)):.3f}<br>"
                f"Score: {float(row.get('IntervalScore', np.nan)):.3f}<br>"
                f"Shape: {row.get('ShapeHint')}<br>"
                f"Drivers: {row.get('TopDrivers')}"
            )
        )

    if not center_x:
        return
    mode = "markers+text" if show_text else "markers"
    fig.add_trace(
        go.Scatter(
            x=center_x,
            y=center_y,
            mode=mode,
            name=f"{label} pattern interval",
            marker=dict(
                color=color,
                size=11,
                symbol="square-open",
                line=dict(width=2),
            ),
            text=visible_text if show_text else None,
            textposition="top center",
            hovertext=hover_text,
            hoverinfo="text",
        )
    )


def make_plot(
    *,
    data: pd.DataFrame,
    probabilities: pd.Series,
    return_forecasts: pd.Series | None,
    signals: pd.DataFrame,
    intervals: pd.DataFrame | None = None,
    title: str,
    output_path: Path | None,
    model_name: str | None = None,
    threshold: float | None = None,
    confidence_window: int = 20,
    confidence_z: float = 1.0,
) -> Any | None:
    if go is None:
        return None

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )
    add_interval_boxes(
        fig,
        intervals,
        color="#FACC15",
        label=model_name or "ML",
        opacity=0.16,
        show_text=True,
    )
    aligned_probability = probabilities.reindex(data.index).astype(float)
    probability_ci = probability_confidence_frame(
        aligned_probability,
        window=confidence_window,
        z=confidence_z,
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probability_ci["lower"].to_numpy(dtype=float),
            name="Forecast score CI lower",
            mode="lines",
            yaxis="y2",
            line=dict(width=0),
            hoverinfo="skip",
            showlegend=False,
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=probability_ci["upper"].to_numpy(dtype=float),
            name="Forecast score confidence band",
            mode="lines",
            yaxis="y2",
            fill="tonexty",
            fillcolor="rgba(56, 189, 248, 0.14)",
            line=dict(width=0),
            hoverinfo="skip",
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=aligned_probability.to_numpy(dtype=float),
            name=f"{model_name or 'Model'} forecast score",
            mode="lines",
            yaxis="y2",
            line=dict(color="#38BDF8", width=2),
        )
    )
    if return_forecasts is not None:
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=return_forecasts.reindex(data.index).to_numpy(dtype=float),
                name=f"{model_name or 'Model'} return forecast %",
                mode="lines",
                yaxis="y3",
                line=dict(color="#F97316", width=2),
            )
        )
    if threshold is not None and np.isfinite(float(threshold)):
        fig.add_trace(
            go.Scatter(
                x=[plot_x[0], plot_x[-1]],
                y=[float(threshold), float(threshold)],
                name="Signal threshold",
                mode="lines",
                yaxis="y2",
                line=dict(color="#FACC15", width=1, dash="dash"),
                hoverinfo="skip",
            )
        )

    if not signals.empty:
        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers+text",
                name=f"{model_name or 'ML'} Forecast Breakout",
                marker=dict(color="#FACC15", size=12, symbol="circle"),
                text=[f"{prob:.2f}" for prob in signals["ReversalProbability"]],
                textposition="top center",
                hovertext=[
                    (
                        f"ML Buy<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Forecast score: {row['ForecastScore']:.3f}<br>"
                        f"Return forecast: {row['ReturnForecastPct']:.3f}%<br>"
                        f"Breakout level: {float(row.get('BreakoutLevel', np.nan)):.4f}<br>"
                        f"Support touches: {row.get('SupportTouches')}<br>"
                        f"Resistance touches: {row.get('ResistanceTouches')}<br>"
                        f"Breakout strength ATR: {float(row.get('BreakoutStrengthAtr', np.nan)):.3f}<br>"
                        f"Switch: {row.get('SwitchKind')} at {row.get('SwitchTurnDate')}<br>"
                        f"Bars from turn: {row.get('BarsFromSwitchTurn')}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Forecast Score",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        yaxis3=dict(
            title="Return Forecast %",
            overlaying="y",
            side="right",
            anchor="free",
            position=0.98,
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=110, t=80, b=55),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    if output_path is not None:
        fig.write_html(str(output_path), include_plotlyjs="cdn")
    return fig


def make_feature_importance_plot(
    *,
    importance: pd.DataFrame,
    title: str,
    output_path: Path | None,
    top_n: int = 25,
) -> Any | None:
    if go is None or importance.empty:
        return None

    plot_frame = importance.head(top_n).copy()
    plot_frame = plot_frame.sort_values("importance_mean", ascending=True)
    fig = go.Figure()
    fig.add_trace(
        go.Bar(
            x=plot_frame["importance_mean"],
            y=plot_frame["feature"],
            orientation="h",
            error_x=dict(
                type="data",
                array=plot_frame["importance_std"].clip(lower=0).to_numpy(dtype=float),
                visible=True,
                color="#94A3B8",
            ),
            marker=dict(color="#38BDF8"),
            hovertext=[
                (
                    f"Feature: {row['feature']}<br>"
                    f"Importance: {row['importance_mean']:.6f}<br>"
                    f"Std: {row['importance_std']:.6f}<br>"
                    f"Rank: {int(row['rank'])}"
                )
                for _, row in plot_frame.iterrows()
            ],
            hoverinfo="text",
            name="Permutation importance",
        )
    )
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=max(520, 24 * len(plot_frame) + 180),
        xaxis=dict(title="Neg-MAE permutation importance"),
        yaxis=dict(title="Feature"),
        margin=dict(l=180, r=50, t=80, b=60),
    )
    if output_path is not None:
        fig.write_html(str(output_path), include_plotlyjs="cdn")
    return fig


def make_hmm_regime_frame(features: pd.DataFrame) -> pd.DataFrame:
    hmm_columns = [
        column
        for column in features.columns
        if column == "hmm_state" or column.startswith("hmm_state_prob_")
    ]
    if not hmm_columns:
        return pd.DataFrame(columns=["Date", "hmm_state"])
    frame = features[hmm_columns].copy()
    frame.insert(0, "Date", features.index)
    return frame


def make_hmm_regime_plot(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    state_summary: pd.DataFrame,
    title: str,
    output_path: Path | None,
) -> Any | None:
    if go is None or "hmm_state" not in features.columns:
        return None

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    states = features["hmm_state"].reindex(data.index)
    state_values = states.to_numpy(dtype=float)
    finite_states = state_values[np.isfinite(state_values)]
    if len(finite_states) == 0:
        return None

    colors = ["#38BDF8", "#FACC15", "#A78BFA", "#34D399", "#FB7185", "#F97316", "#22D3EE"]
    fig = go.Figure()
    for start_idx, end_idx, state in hmm_state_segments(state_values):
        color = colors[int(state) % len(colors)]
        fig.add_shape(
            type="rect",
            xref="x",
            yref="paper",
            x0=start_idx - 0.5,
            x1=end_idx + 0.5,
            y0=0,
            y1=1,
            fillcolor=hex_to_rgba(color, 0.10),
            line=dict(width=0),
            layer="below",
        )

    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )
    fig.add_trace(
        go.Scatter(
            x=plot_x,
            y=state_values,
            name="HMM state",
            mode="lines",
            yaxis="y2",
            line=dict(color="#FACC15", width=2, shape="hv"),
        )
    )

    probability_columns = [
        column for column in features.columns if column.startswith("hmm_state_prob_")
    ]
    if probability_columns:
        max_probability = features[probability_columns].max(axis=1).reindex(data.index)
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=max_probability.to_numpy(dtype=float),
                name="HMM max probability",
                mode="lines",
                yaxis="y3",
                line=dict(color="#38BDF8", width=2),
            )
        )

    if state_summary is not None and not state_summary.empty:
        annotations = []
        summary = state_summary.copy()
        for _, row in summary.iterrows():
            annotations.append(
                f"S{int(row['state'])}: n={int(row['train_count'])}, "
                f"ret={float(row['mean_return']):.4g}, "
                f"vol={float(row['mean_vol']):.4g}, "
                f"dd={float(row['mean_drawdown']):.3g}"
            )
        fig.add_annotation(
            text="<br>".join(annotations),
            xref="paper",
            yref="paper",
            x=0.01,
            y=0.98,
            align="left",
            showarrow=False,
            bgcolor="rgba(15, 23, 42, 0.70)",
            bordercolor="#334155",
            font=dict(size=11),
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="HMM State",
            overlaying="y",
            side="right",
            showgrid=False,
        ),
        yaxis3=dict(
            title="State Confidence",
            overlaying="y",
            side="right",
            anchor="free",
            position=0.98,
            range=[0, 1],
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=110, t=90, b=60),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    if output_path is not None:
        fig.write_html(str(output_path), include_plotlyjs="cdn")
    return fig


def hmm_state_segments(state_values: np.ndarray) -> list[tuple[int, int, int]]:
    segments: list[tuple[int, int, int]] = []
    start_idx: int | None = None
    active_state: int | None = None
    for idx, state in enumerate(state_values):
        if not np.isfinite(state):
            if start_idx is not None and active_state is not None:
                segments.append((start_idx, idx - 1, active_state))
            start_idx = None
            active_state = None
            continue
        state_int = int(state)
        if active_state is None:
            start_idx = idx
            active_state = state_int
            continue
        if state_int != active_state:
            segments.append((int(start_idx), idx - 1, active_state))
            start_idx = idx
            active_state = state_int
    if start_idx is not None and active_state is not None:
        segments.append((start_idx, len(state_values) - 1, active_state))
    return segments


def make_model_comparison_plot(
    *,
    data: pd.DataFrame,
    per_model_probabilities: dict[str, pd.Series],
    per_model_forecasts: dict[str, pd.Series],
    per_model_signals: dict[str, pd.DataFrame],
    per_model_intervals: dict[str, pd.DataFrame],
    per_model_thresholds: dict[str, float],
    title: str,
    output_path: Path | None,
) -> Any | None:
    if go is None or not per_model_probabilities:
        return None

    plot_x = np.arange(len(data))
    date_labels = data.index.strftime("%Y-%m-%d %H:%M").to_numpy()
    colors = ["#38BDF8", "#FACC15", "#A78BFA", "#34D399", "#FB7185", "#F97316"]

    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=plot_x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=date_labels,
            hovertext=date_labels,
            hoverinfo="x+y+text",
        )
    )

    for model_idx, (model_name, probabilities) in enumerate(per_model_probabilities.items()):
        color = colors[model_idx % len(colors)]
        threshold = per_model_thresholds.get(model_name, np.nan)
        add_interval_boxes(
            fig,
            per_model_intervals.get(model_name, empty_interval_frame()),
            color=color,
            label=model_name,
            opacity=0.08,
            show_text=False,
        )
        fig.add_trace(
            go.Scatter(
                x=plot_x,
                y=probabilities.reindex(data.index).to_numpy(dtype=float),
                name=f"{model_name} forecast score",
                mode="lines",
                yaxis="y2",
                line=dict(color=color, width=2),
            )
        )
        if model_name in per_model_forecasts:
            fig.add_trace(
                go.Scatter(
                    x=plot_x,
                    y=per_model_forecasts[model_name].reindex(data.index).to_numpy(dtype=float),
                    name=f"{model_name} return forecast %",
                    mode="lines",
                    yaxis="y3",
                    line=dict(color=color, width=1, dash="dash"),
                    opacity=0.75,
                )
            )
        if threshold is not None and np.isfinite(float(threshold)):
            fig.add_trace(
                go.Scatter(
                    x=[plot_x[0], plot_x[-1]],
                    y=[float(threshold), float(threshold)],
                    name=f"{model_name} threshold",
                    mode="lines",
                    yaxis="y2",
                    line=dict(color=color, width=1, dash="dot"),
                    hoverinfo="skip",
                    showlegend=False,
                )
            )

        signals = per_model_signals.get(model_name, empty_signal_frame())
        if signals is None or signals.empty:
            continue

        signal_x = signals["EntryBarIdx"].astype(int).to_numpy()
        fig.add_trace(
            go.Scatter(
                x=signal_x,
                y=signals["EntryPrice"],
                mode="markers",
                name=f"{model_name} signals",
                marker=dict(
                    color=color,
                    size=10,
                    symbol="circle-open",
                    line=dict(width=2),
                ),
                hovertext=[
                    (
                        f"Model: {model_name}<br>"
                        f"Date: {row['EntryDate']}<br>"
                        f"Forecast score: {row['ForecastScore']:.3f}<br>"
                        f"Return forecast: {row['ReturnForecastPct']:.3f}%<br>"
                        f"Threshold: {row['SignalThreshold']:.3f}<br>"
                        f"Breakout level: {float(row.get('BreakoutLevel', np.nan)):.4f}<br>"
                        f"Support touches: {row.get('SupportTouches')}<br>"
                        f"Resistance touches: {row.get('ResistanceTouches')}<br>"
                        f"Breakout strength ATR: {float(row.get('BreakoutStrengthAtr', np.nan)):.3f}<br>"
                        f"Switch: {row.get('SwitchKind')} at {row.get('SwitchTurnDate')}<br>"
                        f"Bars from turn: {row.get('BarsFromSwitchTurn')}<br>"
                        f"Shape: {row['ShapeHint']}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Drivers: {row['TopDrivers']}"
                    )
                    for _, row in signals.iterrows()
                ],
                hoverinfo="text",
            )
        )

    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=860,
        xaxis_rangeslider_visible=False,
        yaxis=dict(title="Price"),
        yaxis2=dict(
            title="Forecast Score",
            overlaying="y",
            side="right",
            range=[0, 1],
            showgrid=False,
        ),
        yaxis3=dict(
            title="Return Forecast %",
            overlaying="y",
            side="right",
            anchor="free",
            position=0.98,
            showgrid=False,
        ),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=55, r=110, t=90, b=60),
    )
    tick_count = min(12, len(data))
    tick_indexes = np.linspace(0, len(data) - 1, tick_count, dtype=int)
    fig.update_xaxes(
        title_text="Bars",
        tickmode="array",
        tickvals=tick_indexes,
        ticktext=date_labels[tick_indexes],
        tickangle=-30,
    )
    if output_path is not None:
        fig.write_html(str(output_path), include_plotlyjs="cdn")
    return fig


def run(config: ReversalMLConfig) -> Path:
    analysis = run_analysis(config, save=True)
    print_run_summary(analysis)
    return analysis["run_dir"]


def run_analysis(config: ReversalMLConfig, *, save: bool = True) -> dict[str, Any]:
    data = load_ohlcv(config)
    features = make_feature_frame(data, config)
    labels = make_reversal_labels(data, features, config)
    result = train_reversal_models(data, features, labels, config)
    run_dir = save_outputs(data, result, config) if save else None
    return {
        "config": config,
        "data": data,
        "features": features,
        "labels": labels,
        "result": result,
        "run_dir": run_dir,
        "csv_paths": collect_csv_paths(run_dir, result) if run_dir is not None else {},
        "plot_paths": collect_plot_paths(run_dir, result) if run_dir is not None else {},
        "figures": {},
    }


def run_notebook(
    *,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    csv_path: str | None = None,
    output_dir: str = "outputs/ml_reversal_runs",
    show_plots: bool = True,
    print_outputs: bool = True,
    save: bool = True,
    **overrides: Any,
) -> dict[str, Any]:
    config_values = {
        "symbol": symbol,
        "period": period,
        "interval": interval,
        "csv_path": csv_path,
        "output_dir": output_dir,
    }
    config_values.update(overrides)
    config = ReversalMLConfig(**config_values)
    analysis = run_analysis(config, save=save)
    if print_outputs:
        print_run_summary(analysis)
    if show_plots:
        display_notebook_figures(analysis)
    return analysis


def collect_csv_paths(run_dir: Path | None, result: dict[str, Any]) -> dict[str, Path]:
    if run_dir is None:
        return {}
    paths: dict[str, Path] = {
        "signals": run_dir / "signals.csv",
        "pattern_intervals": run_dir / "pattern_intervals.csv",
        "model_report": run_dir / "model_report.csv",
        "feature_importance": run_dir / "feature_importance.csv",
        "hmm_state_summary": run_dir / "hmm_state_summary.csv",
        "hmm_regimes": run_dir / "hmm_regimes.csv",
        "pca_loadings": run_dir / "pca_loadings.csv",
        "probabilities": run_dir / "probabilities.csv",
        "return_forecasts": run_dir / "return_forecasts.csv",
        "all_model_probabilities": run_dir / "all_model_probabilities.csv",
        "all_model_return_forecasts": run_dir / "all_model_return_forecasts.csv",
        "all_model_pattern_intervals": run_dir / "all_model_pattern_intervals.csv",
    }
    models_dir = run_dir / "models"
    for model_name in result.get("per_model_probabilities", {}):
        safe_model_name = safe_filename(str(model_name))
        paths[f"{model_name}_signals"] = models_dir / f"{safe_model_name}_signals.csv"
        paths[f"{model_name}_pattern_intervals"] = (
            models_dir / f"{safe_model_name}_pattern_intervals.csv"
        )
        paths[f"{model_name}_feature_importance"] = (
            models_dir / f"{safe_model_name}_feature_importance.csv"
        )
        paths[f"{model_name}_return_forecast"] = models_dir / f"{safe_model_name}_return_forecast.csv"
    return paths


def collect_plot_paths(run_dir: Path | None, result: dict[str, Any]) -> dict[str, Path]:
    if run_dir is None:
        return {}
    paths: dict[str, Path] = {
        "best_model": run_dir / "reversal_ml_plot.html",
        "model_comparison": run_dir / "all_models_comparison_plot.html",
        "hmm_regime": run_dir / "hmm_regime_plot.html",
    }
    models_dir = run_dir / "models"
    for model_name in result.get("per_model_probabilities", {}):
        safe_model_name = safe_filename(str(model_name))
        paths[f"{model_name}_plot"] = models_dir / f"{safe_model_name}_plot.html"
        paths[f"{model_name}_feature_importance"] = (
            models_dir / f"{safe_model_name}_feature_importance.html"
        )
    return paths


def print_run_summary(analysis: dict[str, Any]) -> None:
    config = analysis["config"]
    result = analysis["result"]
    labels = analysis.get("labels")
    run_dir = analysis.get("run_dir")

    print("ML reversal run")
    print(f"Run dir: {run_dir if run_dir is not None else 'not saved'}")
    print(
        f"Symbol: {config.symbol} | Period: {config.period} | Interval: {config.interval} | "
        f"HMM: {'on' if config.use_hmm else 'off'}"
    )
    print(
        f"Best model: {result['best_model_name']} | "
        f"Return threshold: {result['return_threshold_pct']:.4f}% | "
        f"Forecast score threshold: {result['threshold']:.4f}"
    )
    print(f"Pattern intervals: {len(result['intervals'])} | Point signals: {len(result['signals'])}")
    if isinstance(labels, pd.DataFrame) and "reversal_label" in labels.columns:
        positive_labels = int(labels["reversal_label"].fillna(0).sum())
        labeled_rows = int(labels["reversal_label"].notna().sum())
        print(f"Switch labels: {positive_labels} positive / {labeled_rows} labeled rows")

    if not result["model_report"].empty:
        print("\nModel report:")
        print(result["model_report"].to_string(index=False))

    if not result["intervals"].empty:
        preview_columns = [
            column
            for column in [
                "StartDate",
                "EndDate",
                "TurnDate",
                "BreakoutDate",
                "ProbabilityPeakDate",
                "SwitchKind",
                "Bars",
                "BreakoutLevel",
                "SupportTouches",
                "ResistanceTouches",
                "BaseRangeAtr",
                "BreakoutStrengthAtr",
                "ReturnForecastPctPeak",
                "ReturnForecastPctMean",
                "ForecastScoreMax",
                "ForecastScoreMean",
                "IntervalScore",
                "BoxLow",
                "BoxHigh",
                "ShapeHint",
            ]
            if column in result["intervals"].columns
        ]
        print("\nTop pattern intervals:")
        print(
            result["intervals"]
            .sort_values("IntervalScore", ascending=False)
            .head(10)[preview_columns]
            .to_string(index=False)
        )

    csv_paths = analysis.get("csv_paths", {})
    if csv_paths:
        print("\nCSV files:")
        for name, path in csv_paths.items():
            status = "created" if path.exists() else "expected"
            print(f"  {name}: {path} [{status}]")

    plot_paths = analysis.get("plot_paths", {})
    if plot_paths:
        print("\nPlot files:")
        if go is None:
            print("  plotly is not installed, so HTML plot files were skipped.")
        for name, path in plot_paths.items():
            status = "created" if path.exists() else "not created"
            print(f"  {name}: {path} [{status}]")


def build_notebook_figures(analysis: dict[str, Any]) -> dict[str, Any]:
    if go is None:
        warnings.warn("plotly is not installed, so notebook figures cannot be displayed.")
        return {}

    config = analysis["config"]
    data = analysis["data"]
    result = analysis["result"]
    figures: dict[str, Any] = {}

    best_figure = make_plot(
        data=data,
        probabilities=result["probabilities"],
        return_forecasts=result["forecasts"],
        signals=result["signals"],
        intervals=result["intervals"],
        title=f"{config.symbol} ML Return Forecast Breakouts",
        output_path=None,
        model_name=f"best: {result['best_model_name']}",
        threshold=result["threshold"],
        confidence_window=config.confidence_window,
        confidence_z=config.confidence_z,
    )
    if best_figure is not None:
        figures["best_model"] = best_figure

    comparison_figure = make_model_comparison_plot(
        data=data,
        per_model_probabilities=result["per_model_probabilities"],
        per_model_forecasts=result["per_model_forecasts"],
        per_model_signals=result["per_model_signals"],
        per_model_intervals=result["per_model_intervals"],
        per_model_thresholds=result["per_model_thresholds"],
        title=f"{config.symbol} ML Reversal Model Comparison",
        output_path=None,
    )
    if comparison_figure is not None:
        figures["model_comparison"] = comparison_figure

    hmm_figure = make_hmm_regime_plot(
        data=data,
        features=result["features"],
        state_summary=result["hmm_state_summary"],
        title=f"{config.symbol} HMM Regime Map",
        output_path=None,
    )
    if hmm_figure is not None:
        figures["hmm_regime"] = hmm_figure

    for model_name, probabilities in result.get("per_model_probabilities", {}).items():
        forecasts = result["per_model_forecasts"].get(model_name)
        signals = result["per_model_signals"].get(model_name, empty_signal_frame())
        intervals = result["per_model_intervals"].get(model_name, empty_interval_frame())
        threshold = result["per_model_thresholds"].get(model_name, np.nan)
        model_figure = make_plot(
            data=data,
            probabilities=probabilities,
            return_forecasts=forecasts,
            signals=signals,
            intervals=intervals,
            title=f"{config.symbol} {model_name} Return Forecast Breakouts",
            output_path=None,
            model_name=model_name,
            threshold=threshold,
            confidence_window=config.confidence_window,
            confidence_z=config.confidence_z,
        )
        if model_figure is not None:
            figures[f"{model_name}_plot"] = model_figure

        importance = result.get("per_model_feature_importance", {}).get(
            model_name,
            pd.DataFrame(columns=["feature", "importance_mean", "importance_std", "rank"]),
        )
        importance_figure = make_feature_importance_plot(
            importance=importance,
            title=f"{config.symbol} {model_name} XAI Feature Importance",
            output_path=None,
        )
        if importance_figure is not None:
            figures[f"{model_name}_feature_importance"] = importance_figure

    return figures


def display_notebook_figures(analysis: dict[str, Any]) -> None:
    try:
        from IPython.display import display
    except ModuleNotFoundError:
        warnings.warn("IPython unavailable; returning figures in analysis['figures'] without display.")
        analysis["figures"] = build_notebook_figures(analysis)
        return

    figures = build_notebook_figures(analysis)
    analysis["figures"] = figures
    if not figures:
        return

    for name, figure in figures.items():
        print(f"Displaying plot: {name}")
        display(figure)


def display_notebook_plot_links(plot_paths: dict[str, Path]) -> None:
    try:
        from IPython.display import HTML, display
    except ModuleNotFoundError:
        warnings.warn("IPython unavailable; returning plot paths without notebook display.")
        return
    links = "".join(
        f'<li><a href="{path}" target="_blank">{name}</a></li>'
        for name, path in plot_paths.items()
        if path.exists()
    )
    if links:
        display(HTML(f"<b>ML reversal plots</b><ul>{links}</ul>"))


if __name__ == "__main__":
    run_notebook()
```

```python
from __future__ import annotations

from dataclasses import dataclass
from pathlib import Path
from typing import Any
import warnings

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


@dataclass
class BoxConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"

    lookback: int = 70
    min_base_bars: int = 14
    left_pad: int = 8
    atr_period: int = 14

    touch_tolerance_atr: float = 0.45
    zone_width_atr: float = 0.30
    min_support_touches: int = 2
    min_resistance_touches: int = 2
    touch_cluster_gap: int = 3

    breakout_buffer_atr: float = 0.10
    max_breakout_extension_atr: float = 2.30
    min_box_height_atr: float = 1.10
    max_box_height_atr: float = 9.00

    prior_downtrend_window: int = 55
    min_prior_drawdown_pct: float = 2.00

    outcome_horizon: int = 40
    target_atr: float = 2.00
    stop_atr: float = 1.20


def run_step01(
    *,
    csv_path: str | None = None,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    show_plot: bool = True,
    save_csv_path: str | None = None,
    **overrides: Any,
) -> dict[str, Any]:
    config = BoxConfig(symbol=symbol, period=period, interval=interval, **overrides)
    data = load_ohlcv(csv_path=csv_path, config=config)
    features = make_basic_features(data, config)
    boxes = find_pseudo_reversal_boxes(data, features, config)

    if save_csv_path is not None:
        Path(save_csv_path).parent.mkdir(parents=True, exist_ok=True)
        boxes.to_csv(save_csv_path, index=False)

    fig = plot_pseudo_boxes(data, boxes, title=f"{config.symbol} pseudo reversal boxes")
    if show_plot and fig is not None:
        display_figure(fig)

    print("step01 pseudo boxes")
    print(f"bars: {len(data)} | boxes: {len(boxes)}")
    if len(boxes):
        preview = [
            "StartDate",
            "EndDate",
            "BreakoutDate",
            "SupportTouches",
            "ResistanceTouches",
            "BoxHeightAtr",
            "BreakoutStrengthAtr",
            "Outcome",
        ]
        print(boxes[preview].head(20).to_string(index=False))

    return {"config": config, "data": data, "features": features, "boxes": boxes, "fig": fig}


def load_ohlcv(*, csv_path: str | None, config: BoxConfig) -> pd.DataFrame:
    if csv_path is not None:
        return prepare_ohlcv(pd.read_csv(csv_path))

    if Ticker is None:
        raise ImportError("Install yahooquery or pass csv_path.")

    raw = Ticker(config.symbol).history(period=config.period, interval=config.interval)
    if isinstance(raw, pd.DataFrame) and isinstance(raw.index, pd.MultiIndex):
        if "symbol" in raw.index.names:
            raw = raw.reset_index(level="symbol", drop=True)
        else:
            raw = raw.reset_index()
    return prepare_ohlcv(raw)


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    data = df.copy()
    data = data.rename(columns={column: str(column).lower() for column in data.columns})

    if not isinstance(data.index, pd.DatetimeIndex):
        date_col = next(
            (col for col in ["datetime", "date", "timestamp", "time"] if col in data.columns),
            None,
        )
        if date_col is None:
            raise ValueError("Use DatetimeIndex or include datetime/date/timestamp column.")
        data[date_col] = pd.to_datetime(data[date_col], errors="coerce")
        data = data.set_index(date_col)

    required = {"open", "high", "low", "close"}
    missing = required.difference(data.columns)
    if missing:
        raise ValueError(f"Missing OHLC columns: {sorted(missing)}")

    data.index = pd.to_datetime(data.index)
    data = data[~data.index.duplicated(keep="last")].sort_index()

    for col in ["open", "high", "low", "close", "volume"]:
        if col not in data.columns:
            data[col] = 0.0
        data[col] = pd.to_numeric(data[col], errors="coerce")

    data = data.dropna(subset=["open", "high", "low", "close"])
    data["volume"] = data["volume"].fillna(0.0).clip(lower=0.0)
    return data[["open", "high", "low", "close", "volume"]]


def make_basic_features(data: pd.DataFrame, config: BoxConfig) -> pd.DataFrame:
    out = pd.DataFrame(index=data.index)
    close = data["close"].astype(float)
    volume = data["volume"].astype(float)

    out["atr"] = atr(data, period=config.atr_period)
    out["atr_pct"] = out["atr"] / close.replace(0, np.nan) * 100.0
    out["log_return"] = np.log(close).diff()
    out["volume_z_20"] = zscore(np.log1p(volume), window=20)
    out["drawdown_lookback"] = (
        data["high"].rolling(config.prior_downtrend_window).max() - close
    ) / data["high"].rolling(config.prior_downtrend_window).max().replace(0, np.nan) * 100.0
    return out.replace([np.inf, -np.inf], np.nan)


def find_pseudo_reversal_boxes(
    data: pd.DataFrame,
    features: pd.DataFrame,
    config: BoxConfig,
) -> pd.DataFrame:
    rows: list[dict[str, Any]] = []
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    atr_values = features["atr"].to_numpy(dtype=float)
    volume_z = features["volume_z_20"].to_numpy(dtype=float)

    start_idx = max(config.lookback, config.min_base_bars + 2)
    for breakout_idx in range(start_idx, len(data) - config.outcome_horizon):
        box = find_one_box(
            breakout_idx=breakout_idx,
            high=high,
            low=low,
            close=close,
            atr_values=atr_values,
            volume_z=volume_z,
            config=config,
        )
        if box is None:
            continue

        outcome = first_target_or_stop(
            high=high,
            low=low,
            start_idx=breakout_idx,
            horizon=config.outcome_horizon,
            target=close[breakout_idx] + config.target_atr * box["Atr"],
            stop=box["SupportZoneLow"] - 0.35 * config.stop_atr * box["Atr"],
        )
        rows.append(
            {
                **box,
                "StartDate": data.index[int(box["StartIdx"])],
                "EndDate": data.index[int(box["EndIdx"])],
                "BreakoutDate": data.index[int(box["BreakoutIdx"])],
                "Outcome": outcome["kind"],
                "OutcomeBarIdx": outcome["idx"],
                "OutcomeDate": data.index[int(outcome["idx"])] if outcome["idx"] is not None else pd.NaT,
            }
        )

    if not rows:
        return empty_boxes()
    boxes = pd.DataFrame(rows)
    boxes = suppress_overlaps(boxes)
    return boxes.sort_values("StartIdx").reset_index(drop=True)


def find_one_box(
    *,
    breakout_idx: int,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr_values: np.ndarray,
    volume_z: np.ndarray,
    config: BoxConfig,
) -> dict[str, Any] | None:
    atr_value = finite_mean(atr_values[max(0, breakout_idx - 10) : breakout_idx + 1])
    if not np.isfinite(atr_value) or atr_value <= 0:
        atr_value = max(abs(close[breakout_idx]) * 0.001, 1e-12)

    pre_end = breakout_idx - 1
    search_start = max(0, breakout_idx - config.lookback)
    window_low = low[search_start : pre_end + 1]
    window_high = high[search_start : pre_end + 1]
    if len(window_low) < config.min_base_bars:
        return None

    support_floor = finite_min(window_low)
    if not np.isfinite(support_floor):
        return None

    tolerance = max(config.touch_tolerance_atr * atr_value, abs(close[breakout_idx]) * 0.001)
    support_mask = window_low <= support_floor + tolerance
    if count_touch_clusters(support_mask, gap=config.touch_cluster_gap) < config.min_support_touches:
        return None

    support_positions = np.flatnonzero(support_mask)
    support_anchor = search_start + int(support_positions[np.argmin(window_low[support_positions])])
    first_support = search_start + int(support_positions[0])
    base_start = max(search_start, first_support - config.left_pad)
    if pre_end - base_start + 1 < config.min_base_bars:
        base_start = max(search_start, pre_end - config.min_base_bars + 1)

    base_low = low[base_start : pre_end + 1]
    base_high = high[base_start : pre_end + 1]
    if len(base_low) < config.min_base_bars:
        return None

    support_candidates = base_low[base_low <= support_floor + tolerance]
    if len(support_candidates) == 0:
        return None
    support_level = float(np.nanmedian(support_candidates))

    resistance_cut = float(np.nanquantile(base_high, 0.82))
    resistance_candidates = base_high[base_high >= resistance_cut - tolerance]
    if len(resistance_candidates) == 0:
        return None
    resistance_level = float(np.nanmedian(resistance_candidates))

    resistance_mask = base_high >= resistance_level - tolerance
    if count_touch_clusters(resistance_mask, gap=config.touch_cluster_gap) < config.min_resistance_touches:
        return None

    box_height_atr = (resistance_level - support_level) / max(atr_value, 1e-12)
    if box_height_atr < config.min_box_height_atr or box_height_atr > config.max_box_height_atr:
        return None

    zone_half_width = max(config.zone_width_atr * atr_value, abs(close[breakout_idx]) * 0.0005)
    support_low = support_level - zone_half_width
    support_high = support_level + zone_half_width
    resistance_low = resistance_level - zone_half_width
    resistance_high = resistance_level + zone_half_width

    breakout_level = resistance_high + config.breakout_buffer_atr * atr_value
    if close[breakout_idx] < breakout_level:
        return None
    if close[breakout_idx - 1] > breakout_level:
        return None

    breakout_strength = (close[breakout_idx] - resistance_level) / max(atr_value, 1e-12)
    if breakout_strength > config.max_breakout_extension_atr:
        return None

    prior_start = max(0, base_start - config.prior_downtrend_window)
    prior_high = finite_max(high[prior_start : support_anchor + 1])
    prior_drawdown = (prior_high - support_floor) / prior_high * 100.0 if prior_high > 0 else np.nan
    if not np.isfinite(prior_drawdown) or prior_drawdown < config.min_prior_drawdown_pct:
        return None

    return {
        "StartIdx": int(base_start),
        "EndIdx": int(breakout_idx),
        "BreakoutIdx": int(breakout_idx),
        "SupportAnchorIdx": int(support_anchor),
        "SupportLevel": float(support_level),
        "SupportZoneLow": float(support_low),
        "SupportZoneHigh": float(support_high),
        "ResistanceLevel": float(resistance_level),
        "ResistanceZoneLow": float(resistance_low),
        "ResistanceZoneHigh": float(resistance_high),
        "BreakoutLevel": float(breakout_level),
        "BoxLow": float(support_low),
        "BoxHigh": float(max(resistance_high, close[breakout_idx])),
        "Atr": float(atr_value),
        "BoxHeightAtr": float(box_height_atr),
        "SupportTouches": int(count_touch_clusters(base_low <= support_floor + tolerance, gap=config.touch_cluster_gap)),
        "ResistanceTouches": int(count_touch_clusters(resistance_mask, gap=config.touch_cluster_gap)),
        "BreakoutStrengthAtr": float(breakout_strength),
        "BreakoutVolumeZ": float(volume_z[breakout_idx]) if np.isfinite(volume_z[breakout_idx]) else np.nan,
        "PriorDrawdownPct": float(prior_drawdown),
    }


def plot_pseudo_boxes(data: pd.DataFrame, boxes: pd.DataFrame, *, title: str):
    if go is None:
        warnings.warn("plotly is not installed; skipping plot.")
        return None

    x = np.arange(len(data))
    labels = data.index.strftime("%Y-%m-%d %H:%M")
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=labels,
            hovertext=labels,
            hoverinfo="text+y",
        )
    )

    for number, row in boxes.iterrows():
        start = int(row["StartIdx"])
        end = int(row["EndIdx"])
        fig.add_shape(
            type="rect",
            x0=start - 0.5,
            x1=end + 0.5,
            y0=float(row["BoxLow"]),
            y1=float(row["BoxHigh"]),
            fillcolor="rgba(250, 204, 21, 0.14)",
            line=dict(color="#FACC15", width=2),
            layer="below",
        )
        fig.add_shape(
            type="rect",
            x0=start - 0.5,
            x1=end + 0.5,
            y0=float(row["SupportZoneLow"]),
            y1=float(row["SupportZoneHigh"]),
            fillcolor="rgba(34, 197, 94, 0.22)",
            line=dict(color="#22C55E", width=1),
            layer="below",
        )
        fig.add_shape(
            type="rect",
            x0=start - 0.5,
            x1=end + 0.5,
            y0=float(row["ResistanceZoneLow"]),
            y1=float(row["ResistanceZoneHigh"]),
            fillcolor="rgba(248, 113, 113, 0.18)",
            line=dict(color="#F87171", width=1),
            layer="below",
        )
        fig.add_shape(
            type="line",
            x0=start - 0.5,
            x1=end + 0.5,
            y0=float(row["BreakoutLevel"]),
            y1=float(row["BreakoutLevel"]),
            line=dict(color="#E5E7EB", width=2),
        )
        fig.add_trace(
            go.Scatter(
                x=[int(row["BreakoutIdx"])],
                y=[float(data["close"].iloc[int(row["BreakoutIdx"])])],
                mode="markers+text",
                marker=dict(color="#FACC15", size=11),
                text=[f"Box {number + 1}"],
                textposition="top center",
                name="Pseudo box",
                hovertext=[
                    (
                        f"Start: {row['StartDate']}<br>"
                        f"Breakout: {row['BreakoutDate']}<br>"
                        f"S touches: {row['SupportTouches']}<br>"
                        f"R touches: {row['ResistanceTouches']}<br>"
                        f"Height ATR: {row['BoxHeightAtr']:.2f}<br>"
                        f"Breakout ATR: {row['BreakoutStrengthAtr']:.2f}<br>"
                        f"Outcome: {row['Outcome']}"
                    )
                ],
                hoverinfo="text",
                showlegend=False,
            )
        )

    ticks = np.linspace(0, len(data) - 1, min(12, len(data)), dtype=int)
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=780,
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", y=1.02, x=1, xanchor="right"),
        margin=dict(l=55, r=40, t=80, b=60),
    )
    fig.update_xaxes(tickmode="array", tickvals=ticks, ticktext=labels[ticks], tickangle=-30)
    return fig


def display_figure(fig: Any) -> None:
    try:
        from IPython.display import display
    except ModuleNotFoundError:
        return
    display(fig)


def first_target_or_stop(
    *,
    high: np.ndarray,
    low: np.ndarray,
    start_idx: int,
    horizon: int,
    target: float,
    stop: float,
) -> dict[str, Any]:
    end = min(len(high) - 1, start_idx + horizon)
    for idx in range(start_idx + 1, end + 1):
        hit_target = high[idx] >= target
        hit_stop = low[idx] <= stop
        if hit_stop and hit_target:
            return {"kind": "stop_first", "idx": int(idx)}
        if hit_stop:
            return {"kind": "stop", "idx": int(idx)}
        if hit_target:
            return {"kind": "target", "idx": int(idx)}
    return {"kind": "timeout", "idx": None}


def suppress_overlaps(boxes: pd.DataFrame, max_overlap: float = 0.55) -> pd.DataFrame:
    selected: list[pd.Series] = []
    quality = boxes.copy()
    quality["_outcome_score"] = quality["Outcome"].map({"target": 3, "timeout": 2, "stop": 1, "stop_first": 0})
    quality["_outcome_score"] = quality["_outcome_score"].fillna(0)
    ranked = quality.sort_values(
        ["_outcome_score", "BreakoutStrengthAtr", "BoxHeightAtr"],
        ascending=[False, False, True],
    )
    for _, row in ranked.iterrows():
        start = int(row["StartIdx"])
        end = int(row["EndIdx"])
        length = max(1, end - start + 1)
        keep = True
        for other in selected:
            other_start = int(other["StartIdx"])
            other_end = int(other["EndIdx"])
            other_length = max(1, other_end - other_start + 1)
            overlap = max(0, min(end, other_end) - max(start, other_start) + 1)
            if overlap / min(length, other_length) >= max_overlap:
                keep = False
                break
        if keep:
            selected.append(row)
    if not selected:
        return empty_boxes()
    return pd.DataFrame(selected).drop(columns=["_outcome_score"], errors="ignore")


def empty_boxes() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "StartIdx",
            "EndIdx",
            "BreakoutIdx",
            "SupportAnchorIdx",
            "StartDate",
            "EndDate",
            "BreakoutDate",
            "SupportLevel",
            "SupportZoneLow",
            "SupportZoneHigh",
            "ResistanceLevel",
            "ResistanceZoneLow",
            "ResistanceZoneHigh",
            "BreakoutLevel",
            "BoxLow",
            "BoxHigh",
            "Atr",
            "BoxHeightAtr",
            "SupportTouches",
            "ResistanceTouches",
            "BreakoutStrengthAtr",
            "BreakoutVolumeZ",
            "PriorDrawdownPct",
            "Outcome",
            "OutcomeBarIdx",
            "OutcomeDate",
        ]
    )


def atr(data: pd.DataFrame, *, period: int) -> pd.Series:
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    close = data["close"].astype(float)
    prev_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [
            high - low,
            (high - prev_close).abs(),
            (low - prev_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    return true_range.rolling(period, min_periods=1).mean()


def zscore(series: pd.Series, *, window: int) -> pd.Series:
    mean = series.rolling(window).mean()
    std = series.rolling(window).std()
    return (series - mean) / std.replace(0, np.nan)


def finite_min(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.min(arr)) if len(arr) else np.nan


def finite_max(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.max(arr)) if len(arr) else np.nan


def finite_mean(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.mean(arr)) if len(arr) else np.nan


def count_touch_clusters(mask: np.ndarray | pd.Series, *, gap: int) -> int:
    values = np.asarray(mask, dtype=bool)
    count = 0
    last_touch = -10_000
    active_cluster = False
    for idx, value in enumerate(values):
        if value and not active_cluster:
            if idx - last_touch >= gap:
                count += 1
            last_touch = idx
            active_cluster = True
        elif not value:
            active_cluster = False
    return count
```

```python
from __future__ import annotations

from dataclasses import dataclass
from typing import Any

import numpy as np
import pandas as pd



@dataclass
class WindowConfig:
    window_size: int = 128
    stride: int = 4
    min_box_overlap: float = 0.35
    clip_value: float = 10.0


def run_step02(
    *,
    step01_result: dict[str, Any] | None = None,
    data: pd.DataFrame | None = None,
    boxes: pd.DataFrame | None = None,
    csv_path: str | None = None,
    boxes_csv_path: str | None = None,
    **overrides: Any,
) -> dict[str, Any]:
    config = WindowConfig(**overrides)

    if step01_result is not None:
        data = step01_result["data"]
        boxes = step01_result["boxes"]
    if data is None and csv_path is not None:
        data = prepare_ohlcv(pd.read_csv(csv_path))
    if boxes is None and boxes_csv_path is not None:
        boxes = pd.read_csv(boxes_csv_path)
    if data is None or boxes is None:
        raise ValueError("Pass step01_result or pass both data and boxes.")

    features = make_window_features(data, clip_value=config.clip_value)
    dataset = make_detector_windows(data=data, features=features, boxes=boxes, config=config)

    print("step02 detector windows")
    print(f"X: {dataset['X'].shape}")
    print(f"objectness positives: {(dataset['objectness'] > 0).sum()} / {len(dataset['objectness'])}")
    print("channels:", dataset["channels"])
    return dataset


def make_window_features(data: pd.DataFrame, *, clip_value: float = 10.0) -> pd.DataFrame:
    close = data["close"].astype(float)
    open_ = data["open"].astype(float)
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    volume = data["volume"].astype(float)
    atr_values = atr(data, period=14).replace(0, np.nan)
    candle_range = (high - low).replace(0, np.nan)

    features = pd.DataFrame(index=data.index)
    features["log_return"] = np.log(close).diff()
    features["open_ret_pct"] = (open_ / close.shift(1) - 1.0) * 100.0
    features["high_ret_pct"] = (high / close.shift(1) - 1.0) * 100.0
    features["low_ret_pct"] = (low / close.shift(1) - 1.0) * 100.0
    features["close_ret_pct"] = (close / close.shift(1) - 1.0) * 100.0
    features["range_atr"] = (high - low) / atr_values
    features["body_atr"] = (close - open_) / atr_values
    features["upper_wick_atr"] = (high - np.maximum(open_, close)) / atr_values
    features["lower_wick_atr"] = (np.minimum(open_, close) - low) / atr_values
    features["close_position"] = (close - low) / candle_range
    features["volume_z_20"] = zscore(np.log1p(volume), window=20)
    features["atr_pct"] = atr_values / close.replace(0, np.nan) * 100.0

    features = features.replace([np.inf, -np.inf], np.nan).fillna(0.0)
    return features.clip(lower=-clip_value, upper=clip_value)


def make_detector_windows(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    boxes: pd.DataFrame,
    config: WindowConfig,
) -> dict[str, Any]:
    window_size = int(config.window_size)
    stride = int(config.stride)
    if len(features) < window_size:
        raise ValueError("Not enough bars for the requested window_size.")

    X_rows: list[np.ndarray] = []
    objectness: list[float] = []
    box_targets: list[list[float]] = []
    meta_rows: list[dict[str, Any]] = []
    channels = list(features.columns)

    for start in range(0, len(features) - window_size + 1, stride):
        end = start + window_size - 1
        best = best_box_for_window(boxes, start=start, end=end, min_overlap=config.min_box_overlap)

        window = features.iloc[start : end + 1].to_numpy(dtype=np.float32)
        X_rows.append(window)

        if best is None:
            objectness.append(0.0)
            box_targets.append([0.0, 0.0, 0.0])
            meta_rows.append(
                {
                    "WindowStartIdx": start,
                    "WindowEndIdx": end,
                    "WindowStartDate": data.index[start],
                    "WindowEndDate": data.index[end],
                    "HasBox": False,
                    "BoxIdx": np.nan,
                    "Outcome": None,
                }
            )
            continue

        row_idx, box = best
        box_start = int(box["StartIdx"])
        box_end = int(box["EndIdx"])
        breakout = int(box["BreakoutIdx"])
        outcome_score = soft_outcome_score(str(box.get("Outcome", "")))

        center_rel = ((box_start + box_end) / 2.0 - start) / max(window_size - 1, 1)
        length_rel = (box_end - box_start + 1) / window_size
        breakout_rel = (breakout - start) / max(window_size - 1, 1)

        objectness.append(float(outcome_score))
        box_targets.append(
            [
                float(np.clip(center_rel, 0.0, 1.0)),
                float(np.clip(length_rel, 0.0, 1.0)),
                float(np.clip(breakout_rel, 0.0, 1.0)),
            ]
        )
        meta_rows.append(
            {
                "WindowStartIdx": start,
                "WindowEndIdx": end,
                "WindowStartDate": data.index[start],
                "WindowEndDate": data.index[end],
                "HasBox": True,
                "BoxIdx": int(row_idx),
                "BoxStartIdx": box_start,
                "BoxEndIdx": box_end,
                "BreakoutIdx": breakout,
                "Outcome": box.get("Outcome", None),
                "Objectness": float(outcome_score),
            }
        )

    return {
        "X": np.stack(X_rows, axis=0),
        "objectness": np.asarray(objectness, dtype=np.float32),
        "box_targets": np.asarray(box_targets, dtype=np.float32),
        "meta": pd.DataFrame(meta_rows),
        "features": features,
        "channels": channels,
        "config": config,
    }


def best_box_for_window(
    boxes: pd.DataFrame,
    *,
    start: int,
    end: int,
    min_overlap: float,
) -> tuple[int, pd.Series] | None:
    if boxes is None or boxes.empty:
        return None

    candidates: list[tuple[float, int, pd.Series]] = []
    for row_idx, box in boxes.iterrows():
        box_start = int(box["StartIdx"])
        box_end = int(box["EndIdx"])
        overlap = max(0, min(end, box_end) - max(start, box_start) + 1)
        box_len = max(1, box_end - box_start + 1)
        overlap_ratio = overlap / box_len
        breakout_inside = start <= int(box["BreakoutIdx"]) <= end
        if overlap_ratio >= min_overlap and breakout_inside:
            score = overlap_ratio * soft_outcome_score(str(box.get("Outcome", "")))
            candidates.append((score, int(row_idx), box))

    if not candidates:
        return None
    candidates.sort(key=lambda item: item[0], reverse=True)
    _, row_idx, box = candidates[0]
    return row_idx, box


def soft_outcome_score(outcome: str) -> float:
    if outcome == "target":
        return 1.0
    if outcome == "timeout":
        return 0.45
    if outcome == "stop":
        return 0.10
    return 0.0

```

```python
from __future__ import annotations

from dataclasses import dataclass
from pathlib import Path
from typing import Any
import warnings

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


@dataclass
class BoxConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"

    lookback: int = 70
    min_base_bars: int = 14
    left_pad: int = 8
    atr_period: int = 14

    touch_tolerance_atr: float = 0.45
    zone_width_atr: float = 0.30
    min_support_touches: int = 2
    min_resistance_touches: int = 2
    touch_cluster_gap: int = 3

    breakout_buffer_atr: float = 0.10
    max_breakout_extension_atr: float = 2.30
    min_box_height_atr: float = 1.10
    max_box_height_atr: float = 9.00

    prior_downtrend_window: int = 55
    min_prior_drawdown_pct: float = 2.00

    structure_order: int = 4
    structure_min_move_atr: float = 0.85
    pre_structure_lookback: int = 90
    post_structure_bars: int = 28
    require_bullish_choch: bool = True
    include_post_bos_structure: bool = True

    outcome_horizon: int = 40
    target_atr: float = 2.00
    stop_atr: float = 1.20


def run_step01(
    *,
    csv_path: str | None = None,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    show_plot: bool = True,
    save_csv_path: str | None = None,
    **overrides: Any,
) -> dict[str, Any]:
    config = BoxConfig(symbol=symbol, period=period, interval=interval, **overrides)
    data = load_ohlcv(csv_path=csv_path, config=config)
    features = make_basic_features(data, config)
    boxes = find_pseudo_reversal_boxes(data, features, config)

    if save_csv_path is not None:
        Path(save_csv_path).parent.mkdir(parents=True, exist_ok=True)
        boxes.to_csv(save_csv_path, index=False)

    fig = plot_pseudo_boxes(data, boxes, title=f"{config.symbol} pseudo reversal boxes")
    if show_plot and fig is not None:
        display_figure(fig)

    print("step01 pseudo boxes")
    print(f"bars: {len(data)} | boxes: {len(boxes)}")
    if len(boxes):
        preview = [
            "StartDate",
            "EndDate",
            "BreakoutDate",
            "HasBullishChoCh",
            "PreLowLabel",
            "PreHighLabel",
            "PostLowLabel",
            "PostHighLabel",
            "SupportTouches",
            "ResistanceTouches",
            "BoxHeightAtr",
            "BreakoutStrengthAtr",
            "Outcome",
        ]
        print(boxes[preview].head(20).to_string(index=False))

    return {"config": config, "data": data, "features": features, "boxes": boxes, "fig": fig}


def load_ohlcv(*, csv_path: str | None, config: BoxConfig) -> pd.DataFrame:
    if csv_path is not None:
        return prepare_ohlcv(pd.read_csv(csv_path))

    if Ticker is None:
        raise ImportError("Install yahooquery or pass csv_path.")

    raw = Ticker(config.symbol).history(period=config.period, interval=config.interval)
    if isinstance(raw, pd.DataFrame) and isinstance(raw.index, pd.MultiIndex):
        if "symbol" in raw.index.names:
            raw = raw.reset_index(level="symbol", drop=True)
        else:
            raw = raw.reset_index()
    return prepare_ohlcv(raw)


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    data = df.copy()
    data = data.rename(columns={column: str(column).lower() for column in data.columns})

    if not isinstance(data.index, pd.DatetimeIndex):
        date_col = next(
            (col for col in ["datetime", "date", "timestamp", "time"] if col in data.columns),
            None,
        )
        if date_col is None:
            raise ValueError("Use DatetimeIndex or include datetime/date/timestamp column.")
        data[date_col] = pd.to_datetime(data[date_col], errors="coerce")
        data = data.set_index(date_col)

    required = {"open", "high", "low", "close"}
    missing = required.difference(data.columns)
    if missing:
        raise ValueError(f"Missing OHLC columns: {sorted(missing)}")

    data.index = pd.to_datetime(data.index)
    data = data[~data.index.duplicated(keep="last")].sort_index()

    for col in ["open", "high", "low", "close", "volume"]:
        if col not in data.columns:
            data[col] = 0.0
        data[col] = pd.to_numeric(data[col], errors="coerce")

    data = data.dropna(subset=["open", "high", "low", "close"])
    data["volume"] = data["volume"].fillna(0.0).clip(lower=0.0)
    return data[["open", "high", "low", "close", "volume"]]


def make_basic_features(data: pd.DataFrame, config: BoxConfig) -> pd.DataFrame:
    out = pd.DataFrame(index=data.index)
    close = data["close"].astype(float)
    volume = data["volume"].astype(float)

    out["atr"] = atr(data, period=config.atr_period)
    out["atr_pct"] = out["atr"] / close.replace(0, np.nan) * 100.0
    out["log_return"] = np.log(close).diff()
    out["volume_z_20"] = zscore(np.log1p(volume), window=20)
    out["drawdown_lookback"] = (
        data["high"].rolling(config.prior_downtrend_window).max() - close
    ) / data["high"].rolling(config.prior_downtrend_window).max().replace(0, np.nan) * 100.0
    return out.replace([np.inf, -np.inf], np.nan)


def find_pseudo_reversal_boxes(
    data: pd.DataFrame,
    features: pd.DataFrame,
    config: BoxConfig,
) -> pd.DataFrame:
    rows: list[dict[str, Any]] = []
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    atr_values = features["atr"].to_numpy(dtype=float)
    volume_z = features["volume_z_20"].to_numpy(dtype=float)
    structure = structural_pivots(
        high=high,
        low=low,
        atr_values=atr_values,
        config=config,
    )

    start_idx = max(config.lookback, config.min_base_bars + 2)
    for breakout_idx in range(start_idx, len(data) - config.outcome_horizon):
        box = find_one_box(
            breakout_idx=breakout_idx,
            high=high,
            low=low,
            close=close,
            atr_values=atr_values,
            volume_z=volume_z,
            structure=structure,
            config=config,
        )
        if box is None:
            continue

        outcome = first_target_or_stop(
            high=high,
            low=low,
            start_idx=breakout_idx,
            horizon=config.outcome_horizon,
            target=close[breakout_idx] + config.target_atr * box["Atr"],
            stop=box["SupportZoneLow"] - 0.35 * config.stop_atr * box["Atr"],
        )
        rows.append(
            {
                **box,
                "StartDate": data.index[int(box["StartIdx"])],
                "EndDate": data.index[int(box["EndIdx"])],
                "BreakoutDate": data.index[int(box["BreakoutIdx"])],
                "Outcome": outcome["kind"],
                "OutcomeBarIdx": outcome["idx"],
                "OutcomeDate": data.index[int(outcome["idx"])] if outcome["idx"] is not None else pd.NaT,
            }
        )

    if not rows:
        return empty_boxes()
    boxes = pd.DataFrame(rows)
    boxes = suppress_overlaps(boxes)
    return boxes.sort_values("StartIdx").reset_index(drop=True)


def find_one_box(
    *,
    breakout_idx: int,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr_values: np.ndarray,
    volume_z: np.ndarray,
    structure: pd.DataFrame,
    config: BoxConfig,
) -> dict[str, Any] | None:
    atr_value = finite_mean(atr_values[max(0, breakout_idx - 10) : breakout_idx + 1])
    if not np.isfinite(atr_value) or atr_value <= 0:
        atr_value = max(abs(close[breakout_idx]) * 0.001, 1e-12)

    pre_end = breakout_idx - 1
    search_start = max(0, breakout_idx - config.lookback)
    window_low = low[search_start : pre_end + 1]
    window_high = high[search_start : pre_end + 1]
    if len(window_low) < config.min_base_bars:
        return None

    support_floor = finite_min(window_low)
    if not np.isfinite(support_floor):
        return None

    tolerance = max(config.touch_tolerance_atr * atr_value, abs(close[breakout_idx]) * 0.001)
    support_mask = window_low <= support_floor + tolerance
    if count_touch_clusters(support_mask, gap=config.touch_cluster_gap) < config.min_support_touches:
        return None

    support_positions = np.flatnonzero(support_mask)
    support_anchor = search_start + int(support_positions[np.argmin(window_low[support_positions])])
    first_support = search_start + int(support_positions[0])
    base_start = max(search_start, first_support - config.left_pad)
    if pre_end - base_start + 1 < config.min_base_bars:
        base_start = max(search_start, pre_end - config.min_base_bars + 1)

    base_low = low[base_start : pre_end + 1]
    base_high = high[base_start : pre_end + 1]
    if len(base_low) < config.min_base_bars:
        return None

    support_candidates = base_low[base_low <= support_floor + tolerance]
    if len(support_candidates) == 0:
        return None
    support_level = float(np.nanmedian(support_candidates))

    resistance_cut = float(np.nanquantile(base_high, 0.82))
    resistance_candidates = base_high[base_high >= resistance_cut - tolerance]
    if len(resistance_candidates) == 0:
        return None
    resistance_level = float(np.nanmedian(resistance_candidates))

    resistance_mask = base_high >= resistance_level - tolerance
    if count_touch_clusters(resistance_mask, gap=config.touch_cluster_gap) < config.min_resistance_touches:
        return None

    box_height_atr = (resistance_level - support_level) / max(atr_value, 1e-12)
    if box_height_atr < config.min_box_height_atr or box_height_atr > config.max_box_height_atr:
        return None

    zone_half_width = max(config.zone_width_atr * atr_value, abs(close[breakout_idx]) * 0.0005)
    support_low = support_level - zone_half_width
    support_high = support_level + zone_half_width
    resistance_low = resistance_level - zone_half_width
    resistance_high = resistance_level + zone_half_width

    breakout_level = resistance_high + config.breakout_buffer_atr * atr_value
    if close[breakout_idx] < breakout_level:
        return None
    if close[breakout_idx - 1] > breakout_level:
        return None

    breakout_strength = (close[breakout_idx] - resistance_level) / max(atr_value, 1e-12)
    if breakout_strength > config.max_breakout_extension_atr:
        return None

    prior_start = max(0, base_start - config.prior_downtrend_window)
    prior_high = finite_max(high[prior_start : support_anchor + 1])
    prior_drawdown = (prior_high - support_floor) / prior_high * 100.0 if prior_high > 0 else np.nan
    if not np.isfinite(prior_drawdown) or prior_drawdown < config.min_prior_drawdown_pct:
        return None

    structure_context = structure_context_for_box(
        structure=structure,
        base_start=base_start,
        support_anchor=support_anchor,
        breakout_idx=breakout_idx,
        breakout_price=float(close[breakout_idx]),
        breakout_buffer=float(config.breakout_buffer_atr) * atr_value,
        config=config,
    )
    if config.require_bullish_choch and not structure_context["HasBullishChoCh"]:
        return None

    expanded_start = min(
        base_start,
        int_or_default(structure_context.get("PreLowIdx"), base_start),
        int_or_default(structure_context.get("PreHighIdx"), base_start),
    )
    expanded_end = breakout_idx
    if config.include_post_bos_structure:
        expanded_end = max(
            expanded_end,
            int_or_default(structure_context.get("PostLowIdx"), breakout_idx),
            int_or_default(structure_context.get("PostHighIdx"), breakout_idx),
        )

    return {
        "StartIdx": int(expanded_start),
        "EndIdx": int(expanded_end),
        "BaseStartIdx": int(base_start),
        "BaseEndIdx": int(pre_end),
        "BreakoutIdx": int(breakout_idx),
        "SupportAnchorIdx": int(support_anchor),
        **structure_context,
        "SupportLevel": float(support_level),
        "SupportZoneLow": float(support_low),
        "SupportZoneHigh": float(support_high),
        "ResistanceLevel": float(resistance_level),
        "ResistanceZoneLow": float(resistance_low),
        "ResistanceZoneHigh": float(resistance_high),
        "BreakoutLevel": float(breakout_level),
        "BoxLow": float(support_low),
        "BoxHigh": float(max(resistance_high, close[breakout_idx])),
        "Atr": float(atr_value),
        "BoxHeightAtr": float(box_height_atr),
        "SupportTouches": int(count_touch_clusters(base_low <= support_floor + tolerance, gap=config.touch_cluster_gap)),
        "ResistanceTouches": int(count_touch_clusters(resistance_mask, gap=config.touch_cluster_gap)),
        "BreakoutStrengthAtr": float(breakout_strength),
        "BreakoutVolumeZ": float(volume_z[breakout_idx]) if np.isfinite(volume_z[breakout_idx]) else np.nan,
        "PriorDrawdownPct": float(prior_drawdown),
    }


def structural_pivots(
    *,
    high: np.ndarray,
    low: np.ndarray,
    atr_values: np.ndarray,
    config: BoxConfig,
) -> pd.DataFrame:
    order = max(2, int(config.structure_order))
    raw: list[dict[str, Any]] = []

    for idx in range(order, len(high) - order):
        high_window = high[idx - order : idx + order + 1]
        low_window = low[idx - order : idx + order + 1]
        if not np.isfinite(high[idx]) or not np.isfinite(low[idx]):
            continue

        if high[idx] >= finite_max(high_window):
            raw.append({"Idx": int(idx), "Type": "H", "Price": float(high[idx])})
        if low[idx] <= finite_min(low_window):
            raw.append({"Idx": int(idx), "Type": "L", "Price": float(low[idx])})

    if not raw:
        return empty_structure()

    raw.sort(key=lambda item: int(item["Idx"]))
    pivots: list[dict[str, Any]] = []
    for pivot in raw:
        if not pivots:
            pivots.append(pivot)
            continue

        previous = pivots[-1]
        if pivot["Type"] == previous["Type"]:
            if is_more_extreme_same_type(pivot, previous):
                pivots[-1] = pivot
            continue

        atr_here = atr_values[int(pivot["Idx"])]
        if not np.isfinite(atr_here) or atr_here <= 0:
            atr_here = max(abs(float(pivot["Price"])) * 0.001, 1e-12)
        move_atr = abs(float(pivot["Price"]) - float(previous["Price"])) / max(float(atr_here), 1e-12)
        if move_atr >= float(config.structure_min_move_atr):
            pivots.append(pivot)

    labeled: list[dict[str, Any]] = []
    previous_high = np.nan
    previous_low = np.nan
    for pivot in pivots:
        price = float(pivot["Price"])
        if pivot["Type"] == "H":
            label = "H" if not np.isfinite(previous_high) else ("HH" if price > previous_high else "LH")
            previous_high = price
        else:
            label = "L" if not np.isfinite(previous_low) else ("HL" if price > previous_low else "LL")
            previous_low = price
        labeled.append({**pivot, "Label": label})

    return pd.DataFrame(labeled, columns=["Idx", "Type", "Price", "Label"])


def structure_context_for_box(
    *,
    structure: pd.DataFrame,
    base_start: int,
    support_anchor: int,
    breakout_idx: int,
    breakout_price: float,
    breakout_buffer: float,
    config: BoxConfig,
) -> dict[str, Any]:
    context = empty_structure_context()
    if structure is None or structure.empty:
        return context

    pre_start = max(0, int(support_anchor) - int(config.pre_structure_lookback))
    pre = structure[(structure["Idx"] >= pre_start) & (structure["Idx"] < int(breakout_idx))]
    pre_lows = pre[pre["Type"] == "L"]
    pre_highs = pre[pre["Type"] == "H"]

    pre_low = pick_last_labeled(pre_lows, preferred_labels={"LL"})
    pre_high = pick_last_labeled(pre_highs, preferred_labels={"LH", "HH", "H"})

    if pre_low is not None:
        context.update(
            {
                "PreLowIdx": int(pre_low["Idx"]),
                "PreLowPrice": float(pre_low["Price"]),
                "PreLowLabel": str(pre_low["Label"]),
            }
        )
    if pre_high is not None:
        choch_level = float(pre_high["Price"])
        has_choch = bool(float(breakout_price) > choch_level + float(breakout_buffer) * 0.25)
        context.update(
            {
                "PreHighIdx": int(pre_high["Idx"]),
                "PreHighPrice": choch_level,
                "PreHighLabel": str(pre_high["Label"]),
                "ChoChIdx": int(breakout_idx) if has_choch else np.nan,
                "ChoChLevel": choch_level,
                "BOSIdx": int(breakout_idx) if has_choch else np.nan,
                "BOSLevel": choch_level,
                "HasBullishChoCh": has_choch,
            }
        )

    if bool(context["HasBullishChoCh"]):
        post_end = int(breakout_idx) + int(config.post_structure_bars)
        post = structure[(structure["Idx"] > int(breakout_idx)) & (structure["Idx"] <= post_end)]
        post_lows = post[post["Type"] == "L"]
        post_highs = post[post["Type"] == "H"]
        post_low = pick_first_labeled(post_lows, preferred_labels={"HL", "LL", "L"})
        post_high_start = int(post_low["Idx"]) if post_low is not None else int(breakout_idx)
        post_highs = post_highs[post_highs["Idx"] > post_high_start]
        post_high = pick_first_labeled(post_highs, preferred_labels={"HH", "H", "LH"})

        if post_low is not None:
            context.update(
                {
                    "PostLowIdx": int(post_low["Idx"]),
                    "PostLowPrice": float(post_low["Price"]),
                    "PostLowLabel": str(post_low["Label"]),
                }
            )
        if post_high is not None:
            context.update(
                {
                    "PostHighIdx": int(post_high["Idx"]),
                    "PostHighPrice": float(post_high["Price"]),
                    "PostHighLabel": str(post_high["Label"]),
                }
            )

    context["StructureStartIdx"] = min(
        int_or_default(context.get("PreLowIdx"), base_start),
        int_or_default(context.get("PreHighIdx"), base_start),
        int(base_start),
    )
    context["StructureEndIdx"] = max(
        int_or_default(context.get("PostLowIdx"), breakout_idx),
        int_or_default(context.get("PostHighIdx"), breakout_idx),
        int(breakout_idx),
    )
    return context


def empty_structure_context() -> dict[str, Any]:
    return {
        "StructureStartIdx": np.nan,
        "StructureEndIdx": np.nan,
        "PreLowIdx": np.nan,
        "PreLowPrice": np.nan,
        "PreLowLabel": None,
        "PreHighIdx": np.nan,
        "PreHighPrice": np.nan,
        "PreHighLabel": None,
        "ChoChIdx": np.nan,
        "ChoChLevel": np.nan,
        "BOSIdx": np.nan,
        "BOSLevel": np.nan,
        "HasBullishChoCh": False,
        "PostLowIdx": np.nan,
        "PostLowPrice": np.nan,
        "PostLowLabel": None,
        "PostHighIdx": np.nan,
        "PostHighPrice": np.nan,
        "PostHighLabel": None,
    }


def empty_structure() -> pd.DataFrame:
    return pd.DataFrame(columns=["Idx", "Type", "Price", "Label"])


def pick_last_labeled(frame: pd.DataFrame, *, preferred_labels: set[str]) -> pd.Series | None:
    if frame.empty:
        return None
    preferred = frame[frame["Label"].isin(preferred_labels)]
    if not preferred.empty:
        return preferred.iloc[-1]
    return frame.iloc[-1]


def pick_first_labeled(frame: pd.DataFrame, *, preferred_labels: set[str]) -> pd.Series | None:
    if frame.empty:
        return None
    preferred = frame[frame["Label"].isin(preferred_labels)]
    if not preferred.empty:
        return preferred.iloc[0]
    return frame.iloc[0]


def is_more_extreme_same_type(candidate: dict[str, Any], previous: dict[str, Any]) -> bool:
    if candidate["Type"] == "H":
        return float(candidate["Price"]) > float(previous["Price"])
    return float(candidate["Price"]) < float(previous["Price"])


def int_or_default(value: Any, default: int) -> int:
    try:
        if np.isfinite(float(value)):
            return int(value)
    except (TypeError, ValueError):
        pass
    return int(default)


def plot_pseudo_boxes(data: pd.DataFrame, boxes: pd.DataFrame, *, title: str):
    if go is None:
        warnings.warn("plotly is not installed; skipping plot.")
        return None

    x = np.arange(len(data))
    labels = data.index.strftime("%Y-%m-%d %H:%M")
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=labels,
            hovertext=labels,
            hoverinfo="text+y",
        )
    )

    for number, row in boxes.iterrows():
        start = int(row["StartIdx"])
        end = int(row["EndIdx"])
        base_start = int(row.get("BaseStartIdx", start))
        breakout = int(row["BreakoutIdx"])
        fig.add_shape(
            type="rect",
            x0=start - 0.5,
            x1=end + 0.5,
            y0=float(row["BoxLow"]),
            y1=float(row["BoxHigh"]),
            fillcolor="rgba(250, 204, 21, 0.14)",
            line=dict(color="#FACC15", width=2),
            layer="below",
        )
        fig.add_shape(
            type="rect",
            x0=base_start - 0.5,
            x1=breakout + 0.5,
            y0=float(row["SupportZoneLow"]),
            y1=float(row["SupportZoneHigh"]),
            fillcolor="rgba(34, 197, 94, 0.22)",
            line=dict(color="#22C55E", width=1),
            layer="below",
        )
        fig.add_shape(
            type="rect",
            x0=base_start - 0.5,
            x1=breakout + 0.5,
            y0=float(row["ResistanceZoneLow"]),
            y1=float(row["ResistanceZoneHigh"]),
            fillcolor="rgba(248, 113, 113, 0.18)",
            line=dict(color="#F87171", width=1),
            layer="below",
        )
        fig.add_shape(
            type="line",
            x0=base_start - 0.5,
            x1=breakout + 0.5,
            y0=float(row["BreakoutLevel"]),
            y1=float(row["BreakoutLevel"]),
            line=dict(color="#E5E7EB", width=2),
        )
        fig.add_trace(
            go.Scatter(
                x=[int(row["BreakoutIdx"])],
                y=[float(data["close"].iloc[int(row["BreakoutIdx"])])],
                mode="markers+text",
                marker=dict(color="#FACC15", size=11),
                text=[f"Box {number + 1}"],
                textposition="top center",
                name="Pseudo box",
                hovertext=[
                    (
                        f"Start: {row['StartDate']}<br>"
                        f"Breakout: {row['BreakoutDate']}<br>"
                        f"ChoCH: {row.get('HasBullishChoCh')}<br>"
                        f"Pre L/H: {row.get('PreLowLabel')} / {row.get('PreHighLabel')}<br>"
                        f"Post L/H: {row.get('PostLowLabel')} / {row.get('PostHighLabel')}<br>"
                        f"S touches: {row['SupportTouches']}<br>"
                        f"R touches: {row['ResistanceTouches']}<br>"
                        f"Height ATR: {row['BoxHeightAtr']:.2f}<br>"
                        f"Breakout ATR: {row['BreakoutStrengthAtr']:.2f}<br>"
                        f"Outcome: {row['Outcome']}"
                    )
                ],
                hoverinfo="text",
                showlegend=False,
            )
        )
        add_structure_marker(fig, data, row, idx_col="PreLowIdx", price_col="PreLowPrice", text_col="PreLowLabel")
        add_structure_marker(fig, data, row, idx_col="PreHighIdx", price_col="PreHighPrice", text_col="PreHighLabel")
        add_structure_marker(fig, data, row, idx_col="ChoChIdx", price_col="ChoChLevel", text="ChoCH/BOS")
        add_structure_marker(fig, data, row, idx_col="PostLowIdx", price_col="PostLowPrice", text_col="PostLowLabel")
        add_structure_marker(fig, data, row, idx_col="PostHighIdx", price_col="PostHighPrice", text_col="PostHighLabel")

    ticks = np.linspace(0, len(data) - 1, min(12, len(data)), dtype=int)
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=780,
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", y=1.02, x=1, xanchor="right"),
        margin=dict(l=55, r=40, t=80, b=60),
    )
    fig.update_xaxes(tickmode="array", tickvals=ticks, ticktext=labels[ticks], tickangle=-30)
    return fig


def add_structure_marker(
    fig: Any,
    data: pd.DataFrame,
    row: pd.Series,
    *,
    idx_col: str,
    price_col: str,
    text_col: str | None = None,
    text: str | None = None,
) -> None:
    idx = row.get(idx_col, np.nan)
    price = row.get(price_col, np.nan)
    try:
        if not np.isfinite(float(idx)) or not np.isfinite(float(price)):
            return
    except (TypeError, ValueError):
        return

    marker_idx = int(idx)
    if marker_idx < 0 or marker_idx >= len(data):
        return
    label = text if text is not None else str(row.get(text_col, "S"))
    fig.add_trace(
        go.Scatter(
            x=[marker_idx],
            y=[float(price)],
            mode="markers+text",
            marker=dict(color="#38BDF8", size=8, symbol="diamond"),
            text=[label],
            textposition="bottom center",
            hovertext=[f"{label}<br>{data.index[marker_idx]}<br>{float(price):.4f}"],
            hoverinfo="text",
            showlegend=False,
        )
    )


def display_figure(fig: Any) -> None:
    try:
        from IPython.display import display
    except ModuleNotFoundError:
        return
    display(fig)


def first_target_or_stop(
    *,
    high: np.ndarray,
    low: np.ndarray,
    start_idx: int,
    horizon: int,
    target: float,
    stop: float,
) -> dict[str, Any]:
    end = min(len(high) - 1, start_idx + horizon)
    for idx in range(start_idx + 1, end + 1):
        hit_target = high[idx] >= target
        hit_stop = low[idx] <= stop
        if hit_stop and hit_target:
            return {"kind": "stop_first", "idx": int(idx)}
        if hit_stop:
            return {"kind": "stop", "idx": int(idx)}
        if hit_target:
            return {"kind": "target", "idx": int(idx)}
    return {"kind": "timeout", "idx": None}


def suppress_overlaps(boxes: pd.DataFrame, max_overlap: float = 0.55) -> pd.DataFrame:
    selected: list[pd.Series] = []
    quality = boxes.copy()
    quality["_outcome_score"] = quality["Outcome"].map({"target": 3, "timeout": 2, "stop": 1, "stop_first": 0})
    quality["_outcome_score"] = quality["_outcome_score"].fillna(0)
    ranked = quality.sort_values(
        ["_outcome_score", "BreakoutStrengthAtr", "BoxHeightAtr"],
        ascending=[False, False, True],
    )
    for _, row in ranked.iterrows():
        start = int(row["StartIdx"])
        end = int(row["EndIdx"])
        length = max(1, end - start + 1)
        keep = True
        for other in selected:
            other_start = int(other["StartIdx"])
            other_end = int(other["EndIdx"])
            other_length = max(1, other_end - other_start + 1)
            overlap = max(0, min(end, other_end) - max(start, other_start) + 1)
            if overlap / min(length, other_length) >= max_overlap:
                keep = False
                break
        if keep:
            selected.append(row)
    if not selected:
        return empty_boxes()
    return pd.DataFrame(selected).drop(columns=["_outcome_score"], errors="ignore")


def empty_boxes() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "StartIdx",
            "EndIdx",
            "BaseStartIdx",
            "BaseEndIdx",
            "BreakoutIdx",
            "SupportAnchorIdx",
            "StartDate",
            "EndDate",
            "BreakoutDate",
            "StructureStartIdx",
            "StructureEndIdx",
            "PreLowIdx",
            "PreLowPrice",
            "PreLowLabel",
            "PreHighIdx",
            "PreHighPrice",
            "PreHighLabel",
            "ChoChIdx",
            "ChoChLevel",
            "BOSIdx",
            "BOSLevel",
            "HasBullishChoCh",
            "PostLowIdx",
            "PostLowPrice",
            "PostLowLabel",
            "PostHighIdx",
            "PostHighPrice",
            "PostHighLabel",
            "SupportLevel",
            "SupportZoneLow",
            "SupportZoneHigh",
            "ResistanceLevel",
            "ResistanceZoneLow",
            "ResistanceZoneHigh",
            "BreakoutLevel",
            "BoxLow",
            "BoxHigh",
            "Atr",
            "BoxHeightAtr",
            "SupportTouches",
            "ResistanceTouches",
            "BreakoutStrengthAtr",
            "BreakoutVolumeZ",
            "PriorDrawdownPct",
            "Outcome",
            "OutcomeBarIdx",
            "OutcomeDate",
        ]
    )


def atr(data: pd.DataFrame, *, period: int) -> pd.Series:
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    close = data["close"].astype(float)
    prev_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [
            high - low,
            (high - prev_close).abs(),
            (low - prev_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    return true_range.rolling(period, min_periods=1).mean()


def zscore(series: pd.Series, *, window: int) -> pd.Series:
    mean = series.rolling(window).mean()
    std = series.rolling(window).std()
    return (series - mean) / std.replace(0, np.nan)


def finite_min(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.min(arr)) if len(arr) else np.nan


def finite_max(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.max(arr)) if len(arr) else np.nan


def finite_mean(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.mean(arr)) if len(arr) else np.nan


def count_touch_clusters(mask: np.ndarray | pd.Series, *, gap: int) -> int:
    values = np.asarray(mask, dtype=bool)
    count = 0
    last_touch = -10_000
    active_cluster = False
    for idx, value in enumerate(values):
        if value and not active_cluster:
            if idx - last_touch >= gap:
                count += 1
            last_touch = idx
            active_cluster = True
        elif not value:
            active_cluster = False
    return count
```

```python
from __future__ import annotations

from dataclasses import dataclass
from typing import Any

import numpy as np
import pandas as pd



@dataclass
class WindowConfig:
    window_size: int = 128
    stride: int = 4
    min_box_overlap: float = 0.35
    clip_value: float = 10.0


def run_step02(
    *,
    step01_result: dict[str, Any] | None = None,
    data: pd.DataFrame | None = None,
    boxes: pd.DataFrame | None = None,
    csv_path: str | None = None,
    boxes_csv_path: str | None = None,
    **overrides: Any,
) -> dict[str, Any]:
    config = WindowConfig(**overrides)

    if step01_result is not None:
        data = step01_result["data"]
        boxes = step01_result["boxes"]
    if data is None and csv_path is not None:
        data = prepare_ohlcv(pd.read_csv(csv_path))
    if boxes is None and boxes_csv_path is not None:
        boxes = pd.read_csv(boxes_csv_path)
    if data is None or boxes is None:
        raise ValueError("Pass step01_result or pass both data and boxes.")

    features = make_window_features(data, clip_value=config.clip_value)
    dataset = make_detector_windows(data=data, features=features, boxes=boxes, config=config)

    print("step02 detector windows")
    print(f"X: {dataset['X'].shape}")
    print(f"objectness positives: {(dataset['objectness'] > 0).sum()} / {len(dataset['objectness'])}")
    print("channels:", dataset["channels"])
    return dataset


def make_window_features(data: pd.DataFrame, *, clip_value: float = 10.0) -> pd.DataFrame:
    close = data["close"].astype(float)
    open_ = data["open"].astype(float)
    high = data["high"].astype(float)
    low = data["low"].astype(float)
    volume = data["volume"].astype(float)
    atr_values = atr(data, period=14).replace(0, np.nan)
    candle_range = (high - low).replace(0, np.nan)

    features = pd.DataFrame(index=data.index)
    features["log_return"] = np.log(close).diff()
    features["open_ret_pct"] = (open_ / close.shift(1) - 1.0) * 100.0
    features["high_ret_pct"] = (high / close.shift(1) - 1.0) * 100.0
    features["low_ret_pct"] = (low / close.shift(1) - 1.0) * 100.0
    features["close_ret_pct"] = (close / close.shift(1) - 1.0) * 100.0
    features["range_atr"] = (high - low) / atr_values
    features["body_atr"] = (close - open_) / atr_values
    features["upper_wick_atr"] = (high - np.maximum(open_, close)) / atr_values
    features["lower_wick_atr"] = (np.minimum(open_, close) - low) / atr_values
    features["close_position"] = (close - low) / candle_range
    features["volume_z_20"] = zscore(np.log1p(volume), window=20)
    features["atr_pct"] = atr_values / close.replace(0, np.nan) * 100.0

    features = features.replace([np.inf, -np.inf], np.nan).fillna(0.0)
    return features.clip(lower=-clip_value, upper=clip_value)


def make_detector_windows(
    *,
    data: pd.DataFrame,
    features: pd.DataFrame,
    boxes: pd.DataFrame,
    config: WindowConfig,
) -> dict[str, Any]:
    window_size = int(config.window_size)
    stride = int(config.stride)
    if len(features) < window_size:
        raise ValueError("Not enough bars for the requested window_size.")

    X_rows: list[np.ndarray] = []
    objectness: list[float] = []
    box_targets: list[list[float]] = []
    meta_rows: list[dict[str, Any]] = []
    channels = list(features.columns)

    for start in range(0, len(features) - window_size + 1, stride):
        end = start + window_size - 1
        best = best_box_for_window(boxes, start=start, end=end, min_overlap=config.min_box_overlap)

        window = features.iloc[start : end + 1].to_numpy(dtype=np.float32)
        X_rows.append(window)

        if best is None:
            objectness.append(0.0)
            box_targets.append([0.0, 0.0, 0.0])
            meta_rows.append(
                {
                    "WindowStartIdx": start,
                    "WindowEndIdx": end,
                    "WindowStartDate": data.index[start],
                    "WindowEndDate": data.index[end],
                    "HasBox": False,
                    "BoxIdx": np.nan,
                    "Outcome": None,
                }
            )
            continue

        row_idx, box = best
        box_start = int(box["StartIdx"])
        box_end = int(box["EndIdx"])
        breakout = int(box["BreakoutIdx"])
        outcome_score = soft_outcome_score(str(box.get("Outcome", "")))

        center_rel = ((box_start + box_end) / 2.0 - start) / max(window_size - 1, 1)
        length_rel = (box_end - box_start + 1) / window_size
        breakout_rel = (breakout - start) / max(window_size - 1, 1)

        objectness.append(float(outcome_score))
        box_targets.append(
            [
                float(np.clip(center_rel, 0.0, 1.0)),
                float(np.clip(length_rel, 0.0, 1.0)),
                float(np.clip(breakout_rel, 0.0, 1.0)),
            ]
        )
        meta_rows.append(
            {
                "WindowStartIdx": start,
                "WindowEndIdx": end,
                "WindowStartDate": data.index[start],
                "WindowEndDate": data.index[end],
                "HasBox": True,
                "BoxIdx": int(row_idx),
                "BoxStartIdx": box_start,
                "BoxEndIdx": box_end,
                "BreakoutIdx": breakout,
                "Outcome": box.get("Outcome", None),
                "Objectness": float(outcome_score),
            }
        )

    return {
        "X": np.stack(X_rows, axis=0),
        "objectness": np.asarray(objectness, dtype=np.float32),
        "box_targets": np.asarray(box_targets, dtype=np.float32),
        "meta": pd.DataFrame(meta_rows),
        "features": features,
        "channels": channels,
        "config": config,
    }


def best_box_for_window(
    boxes: pd.DataFrame,
    *,
    start: int,
    end: int,
    min_overlap: float,
) -> tuple[int, pd.Series] | None:
    if boxes is None or boxes.empty:
        return None

    candidates: list[tuple[float, int, pd.Series]] = []
    for row_idx, box in boxes.iterrows():
        box_start = int(box["StartIdx"])
        box_end = int(box["EndIdx"])
        overlap = max(0, min(end, box_end) - max(start, box_start) + 1)
        box_len = max(1, box_end - box_start + 1)
        overlap_ratio = overlap / box_len
        breakout_inside = start <= int(box["BreakoutIdx"]) <= end
        if overlap_ratio >= min_overlap and breakout_inside:
            score = overlap_ratio * soft_outcome_score(str(box.get("Outcome", "")))
            candidates.append((score, int(row_idx), box))

    if not candidates:
        return None
    candidates.sort(key=lambda item: item[0], reverse=True)
    _, row_idx, box = candidates[0]
    return row_idx, box


def soft_outcome_score(outcome: str) -> float:
    if outcome == "target":
        return 1.0
    if outcome == "timeout":
        return 0.45
    if outcome == "stop":
        return 0.10
    return 0.0

```

```python
from __future__ import annotations

from dataclasses import dataclass
from pathlib import Path
from typing import Any
import warnings

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


@dataclass
class CRTTBSConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"

    atr_period: int = 14
    min_anchor_range_atr: float = 0.55
    max_anchor_range_atr: float = 8.0
    min_sweep_atr: float = 0.05
    reentry_depth: float = 0.05
    require_sweep_close_outside: bool = True
    require_full_body_outside: bool = False

    context_lookback: int = 20
    min_context_move_atr: float = 0.0

    outcome_horizon: int = 30
    stop_buffer_atr: float = 0.15
    cooldown_bars: int = 6


def run_crt_tbs(
    *,
    csv_path: str | None = None,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    show_plot: bool = True,
    save_csv_path: str | None = None,
    **overrides: Any,
) -> dict[str, Any]:
    config = CRTTBSConfig(symbol=symbol, period=period, interval=interval, **overrides)
    data = load_ohlcv(csv_path=csv_path, config=config)
    data = add_atr(data, period=config.atr_period)
    setups = detect_crt_tbs(data, config)

    if save_csv_path is not None:
        Path(save_csv_path).parent.mkdir(parents=True, exist_ok=True)
        setups.to_csv(save_csv_path, index=False)

    fig = plot_crt_tbs(data, setups, title=f"{config.symbol} CRT TBS")
    if show_plot and fig is not None:
        display_figure(fig)

    print("CRT TBS")
    print(f"bars: {len(data)} | setups: {len(setups)}")
    if len(setups):
        preview = [
            "Direction",
            "AnchorDate",
            "SweepDate",
            "ReclaimDate",
            "AnchorLow",
            "AnchorHigh",
            "EntryPrice",
            "TargetPrice",
            "StopPrice",
            "Outcome",
        ]
        print(setups[preview].head(30).to_string(index=False))

    return {"config": config, "data": data, "setups": setups, "fig": fig}


def load_ohlcv(*, csv_path: str | None, config: CRTTBSConfig) -> pd.DataFrame:
    if csv_path is not None:
        return prepare_ohlcv(pd.read_csv(csv_path))
    if Ticker is None:
        raise ImportError("Install yahooquery or pass csv_path.")

    raw = Ticker(config.symbol).history(period=config.period, interval=config.interval)
    if isinstance(raw, pd.DataFrame) and isinstance(raw.index, pd.MultiIndex):
        if "symbol" in raw.index.names:
            raw = raw.reset_index(level="symbol", drop=True)
        else:
            raw = raw.reset_index()
    return prepare_ohlcv(raw)


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    data = df.copy()
    data = data.rename(columns={column: str(column).lower() for column in data.columns})

    if not isinstance(data.index, pd.DatetimeIndex):
        date_col = next(
            (col for col in ["datetime", "date", "timestamp", "time"] if col in data.columns),
            None,
        )
        if date_col is None:
            raise ValueError("Use DatetimeIndex or include datetime/date/timestamp column.")
        data[date_col] = pd.to_datetime(data[date_col], errors="coerce")
        data = data.set_index(date_col)

    required = {"open", "high", "low", "close"}
    missing = required.difference(data.columns)
    if missing:
        raise ValueError(f"Missing OHLC columns: {sorted(missing)}")

    data.index = pd.to_datetime(data.index)
    data = data[~data.index.duplicated(keep="last")].sort_index()
    if "volume" not in data.columns:
        data["volume"] = 0.0

    for col in ["open", "high", "low", "close", "volume"]:
        data[col] = pd.to_numeric(data[col], errors="coerce")

    data = data.dropna(subset=["open", "high", "low", "close"])
    data["volume"] = data["volume"].fillna(0.0).clip(lower=0.0)
    return data[["open", "high", "low", "close", "volume"]]


def add_atr(data: pd.DataFrame, *, period: int) -> pd.DataFrame:
    out = data.copy()
    high = out["high"].astype(float)
    low = out["low"].astype(float)
    close = out["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [
            high - low,
            (high - previous_close).abs(),
            (low - previous_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    out["atr"] = true_range.rolling(period, min_periods=1).mean()
    return out


def detect_crt_tbs(data: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    rows: list[dict[str, Any]] = []
    open_ = data["open"].to_numpy(dtype=float)
    high = data["high"].to_numpy(dtype=float)
    low = data["low"].to_numpy(dtype=float)
    close = data["close"].to_numpy(dtype=float)
    atr = data["atr"].to_numpy(dtype=float)

    last_accepted = -10_000
    end_idx = len(data) - int(config.outcome_horizon)
    for reclaim_idx in range(2, max(2, end_idx)):
        anchor_idx = reclaim_idx - 2
        sweep_idx = reclaim_idx - 1
        if reclaim_idx - last_accepted <= int(config.cooldown_bars):
            continue

        setup = classify_three_candle_tbs(
            anchor_idx=anchor_idx,
            sweep_idx=sweep_idx,
            reclaim_idx=reclaim_idx,
            open_=open_,
            high=high,
            low=low,
            close=close,
            atr=atr,
            config=config,
        )
        if setup is None:
            continue

        outcome = first_target_or_stop(
            high=high,
            low=low,
            start_idx=reclaim_idx,
            horizon=config.outcome_horizon,
            target=setup["TargetPrice"],
            stop=setup["StopPrice"],
        )
        rows.append(
            {
                **setup,
                "AnchorDate": data.index[anchor_idx],
                "SweepDate": data.index[sweep_idx],
                "ReclaimDate": data.index[reclaim_idx],
                "Outcome": outcome["kind"],
                "OutcomeBarIdx": outcome["idx"],
                "OutcomeDate": data.index[int(outcome["idx"])] if outcome["idx"] is not None else pd.NaT,
            }
        )
        last_accepted = reclaim_idx

    if not rows:
        return empty_setups()
    return pd.DataFrame(rows).reset_index(drop=True)


def classify_three_candle_tbs(
    *,
    anchor_idx: int,
    sweep_idx: int,
    reclaim_idx: int,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr: np.ndarray,
    config: CRTTBSConfig,
) -> dict[str, Any] | None:
    anchor_high = float(high[anchor_idx])
    anchor_low = float(low[anchor_idx])
    anchor_range = anchor_high - anchor_low
    atr_value = float(atr[anchor_idx])
    if not np.isfinite(atr_value) or atr_value <= 0:
        atr_value = max(anchor_range, abs(float(close[anchor_idx])) * 0.001, 1e-12)
    if anchor_range <= 0:
        return None

    range_atr = anchor_range / max(atr_value, 1e-12)
    if range_atr < config.min_anchor_range_atr or range_atr > config.max_anchor_range_atr:
        return None

    bullish = bullish_tbs(
        anchor_low=anchor_low,
        anchor_high=anchor_high,
        sweep_idx=sweep_idx,
        reclaim_idx=reclaim_idx,
        open_=open_,
        high=high,
        low=low,
        close=close,
        atr_value=atr_value,
        config=config,
    )
    bearish = bearish_tbs(
        anchor_low=anchor_low,
        anchor_high=anchor_high,
        sweep_idx=sweep_idx,
        reclaim_idx=reclaim_idx,
        open_=open_,
        high=high,
        low=low,
        close=close,
        atr_value=atr_value,
        config=config,
    )

    if bullish:
        direction = "bullish"
        entry = float(close[reclaim_idx])
        target = anchor_high
        mid_target = anchor_low + 0.5 * anchor_range
        stop = min(float(low[sweep_idx]), float(low[reclaim_idx])) - config.stop_buffer_atr * atr_value
        sweep_price = float(low[sweep_idx])
        context_move = prior_context_move_atr(
            close=close,
            anchor_idx=anchor_idx,
            direction=direction,
            atr_value=atr_value,
            config=config,
        )
    elif bearish:
        direction = "bearish"
        entry = float(close[reclaim_idx])
        target = anchor_low
        mid_target = anchor_high - 0.5 * anchor_range
        stop = max(float(high[sweep_idx]), float(high[reclaim_idx])) + config.stop_buffer_atr * atr_value
        sweep_price = float(high[sweep_idx])
        context_move = prior_context_move_atr(
            close=close,
            anchor_idx=anchor_idx,
            direction=direction,
            atr_value=atr_value,
            config=config,
        )
    else:
        return None

    if context_move < float(config.min_context_move_atr):
        return None

    return {
        "Direction": direction,
        "AnchorIdx": int(anchor_idx),
        "SweepIdx": int(sweep_idx),
        "ReclaimIdx": int(reclaim_idx),
        "AnchorLow": float(anchor_low),
        "AnchorHigh": float(anchor_high),
        "AnchorRangeAtr": float(range_atr),
        "SweepPrice": float(sweep_price),
        "EntryPrice": float(entry),
        "MidTargetPrice": float(mid_target),
        "TargetPrice": float(target),
        "StopPrice": float(stop),
        "PriorContextMoveAtr": float(context_move),
        "RiskReward": float(abs(target - entry) / max(abs(entry - stop), 1e-12)),
    }


def bullish_tbs(
    *,
    anchor_low: float,
    anchor_high: float,
    sweep_idx: int,
    reclaim_idx: int,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr_value: float,
    config: CRTTBSConfig,
) -> bool:
    buffer = float(config.min_sweep_atr) * atr_value
    swept_low = low[sweep_idx] < anchor_low - buffer
    if not swept_low:
        return False

    if config.require_full_body_outside:
        body_outside = max(open_[sweep_idx], close[sweep_idx]) < anchor_low
    elif config.require_sweep_close_outside:
        body_outside = close[sweep_idx] < anchor_low
    else:
        body_outside = min(open_[sweep_idx], close[sweep_idx]) < anchor_low
    if not body_outside:
        return False

    reentry_level = anchor_low + float(config.reentry_depth) * (anchor_high - anchor_low)
    closes_back_inside = reentry_level <= close[reclaim_idx] <= anchor_high
    return bool(closes_back_inside)


def bearish_tbs(
    *,
    anchor_low: float,
    anchor_high: float,
    sweep_idx: int,
    reclaim_idx: int,
    open_: np.ndarray,
    high: np.ndarray,
    low: np.ndarray,
    close: np.ndarray,
    atr_value: float,
    config: CRTTBSConfig,
) -> bool:
    buffer = float(config.min_sweep_atr) * atr_value
    swept_high = high[sweep_idx] > anchor_high + buffer
    if not swept_high:
        return False

    if config.require_full_body_outside:
        body_outside = min(open_[sweep_idx], close[sweep_idx]) > anchor_high
    elif config.require_sweep_close_outside:
        body_outside = close[sweep_idx] > anchor_high
    else:
        body_outside = max(open_[sweep_idx], close[sweep_idx]) > anchor_high
    if not body_outside:
        return False

    reentry_level = anchor_high - float(config.reentry_depth) * (anchor_high - anchor_low)
    closes_back_inside = anchor_low <= close[reclaim_idx] <= reentry_level
    return bool(closes_back_inside)


def prior_context_move_atr(
    *,
    close: np.ndarray,
    anchor_idx: int,
    direction: str,
    atr_value: float,
    config: CRTTBSConfig,
) -> float:
    start = max(0, int(anchor_idx) - int(config.context_lookback))
    context = close[start : anchor_idx + 1]
    if len(context) < 3:
        return 0.0
    if direction == "bullish":
        return float((np.nanmax(context) - close[anchor_idx]) / max(atr_value, 1e-12))
    return float((close[anchor_idx] - np.nanmin(context)) / max(atr_value, 1e-12))


def first_target_or_stop(
    *,
    high: np.ndarray,
    low: np.ndarray,
    start_idx: int,
    horizon: int,
    target: float,
    stop: float,
) -> dict[str, Any]:
    direction = "bullish" if target > stop else "bearish"
    end = min(len(high) - 1, start_idx + int(horizon))
    for idx in range(start_idx + 1, end + 1):
        if direction == "bullish":
            stop_hit = low[idx] <= stop
            target_hit = high[idx] >= target
        else:
            stop_hit = high[idx] >= stop
            target_hit = low[idx] <= target
        if stop_hit and target_hit:
            return {"kind": "stop_first", "idx": int(idx)}
        if stop_hit:
            return {"kind": "stop", "idx": int(idx)}
        if target_hit:
            return {"kind": "target", "idx": int(idx)}
    return {"kind": "timeout", "idx": None}


def plot_crt_tbs(data: pd.DataFrame, setups: pd.DataFrame, *, title: str):
    if go is None:
        warnings.warn("plotly is not installed; skipping plot.")
        return None

    x = np.arange(len(data))
    labels = data.index.strftime("%Y-%m-%d %H:%M")
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=labels,
            hovertext=labels,
            hoverinfo="text+y",
        )
    )

    for number, row in setups.iterrows():
        color = "#22C55E" if row["Direction"] == "bullish" else "#EF4444"
        fill = "rgba(34, 197, 94, 0.13)" if row["Direction"] == "bullish" else "rgba(239, 68, 68, 0.13)"
        anchor_idx = int(row["AnchorIdx"])
        reclaim_idx = int(row["ReclaimIdx"])
        sweep_idx = int(row["SweepIdx"])
        fig.add_shape(
            type="rect",
            x0=anchor_idx - 0.5,
            x1=reclaim_idx + 0.5,
            y0=float(row["AnchorLow"]),
            y1=float(row["AnchorHigh"]),
            fillcolor=fill,
            line=dict(color=color, width=2),
            layer="below",
        )
        add_price_line(fig, anchor_idx, reclaim_idx, row["AnchorHigh"], color=color, name="CRT high")
        add_price_line(fig, anchor_idx, reclaim_idx, row["AnchorLow"], color=color, name="CRT low")
        add_price_line(fig, reclaim_idx, min(len(data) - 1, reclaim_idx + 12), row["TargetPrice"], color="#38BDF8", name="target")
        add_price_line(fig, reclaim_idx, min(len(data) - 1, reclaim_idx + 12), row["StopPrice"], color="#FACC15", name="stop")

        fig.add_trace(
            go.Scatter(
                x=[sweep_idx, reclaim_idx],
                y=[float(row["SweepPrice"]), float(row["EntryPrice"])],
                mode="markers+text",
                marker=dict(color=[color, "#FACC15"], size=[10, 11]),
                text=["TBS", "Reclaim"],
                textposition="top center",
                name="CRT TBS",
                hovertext=[
                    f"TBS sweep<br>{row['SweepDate']}<br>{row['Direction']}",
                    (
                        f"Reclaim<br>{row['ReclaimDate']}<br>"
                        f"Entry: {row['EntryPrice']:.4f}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>"
                        f"Stop: {row['StopPrice']:.4f}<br>"
                        f"Outcome: {row['Outcome']}"
                    ),
                ],
                hoverinfo="text",
                showlegend=(number == 0),
            )
        )

    ticks = np.linspace(0, len(data) - 1, min(12, len(data)), dtype=int)
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=780,
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", y=1.02, x=1, xanchor="right"),
        margin=dict(l=55, r=40, t=80, b=60),
    )
    fig.update_xaxes(tickmode="array", tickvals=ticks, ticktext=labels[ticks], tickangle=-30)
    return fig


def add_price_line(fig: Any, start_idx: int, end_idx: int, price: float, *, color: str, name: str) -> None:
    fig.add_shape(
        type="line",
        x0=int(start_idx) - 0.5,
        x1=int(end_idx) + 0.5,
        y0=float(price),
        y1=float(price),
        line=dict(color=color, width=1, dash="dash"),
    )


def display_figure(fig: Any) -> None:
    try:
        from IPython.display import display
    except ModuleNotFoundError:
        return
    display(fig)


def empty_setups() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "Direction",
            "AnchorIdx",
            "SweepIdx",
            "ReclaimIdx",
            "AnchorDate",
            "SweepDate",
            "ReclaimDate",
            "AnchorLow",
            "AnchorHigh",
            "AnchorRangeAtr",
            "SweepPrice",
            "EntryPrice",
            "MidTargetPrice",
            "TargetPrice",
            "StopPrice",
            "PriorContextMoveAtr",
            "RiskReward",
            "Outcome",
            "OutcomeBarIdx",
            "OutcomeDate",
        ]
    )

```

```python
from __future__ import annotations

from dataclasses import dataclass
from pathlib import Path
from typing import Any
from zoneinfo import ZoneInfo
import warnings

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None


@dataclass
class CRTTBSConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    data_timezone: str = "Europe/Istanbul"
    timezone: str = "America/New_York"

    main_session: str = "0830-0945"
    secondary_session: str = "0945-1600"
    search_24x7: bool = True
    maximum_crts_per_day: int = 4
    secondary_cooldown_bars: int = 20
    setup_expiration_bars: int = 120

    atr_length: int = 14
    secondary_range_atr: float = 1.50
    secondary_body_percent: float = 65.0
    secondary_close_extreme: float = 0.80
    require_secondary_expansion: bool = True
    secondary_only_after_main: bool = True
    require_secondary_range_expansion: bool = True

    swing_left: int = 3
    swing_right: int = 3
    structure_lookback: int = 20
    sweep_ticks: float = 1.0
    tick_size: float = 0.01
    require_close_inside: bool = True
    bos_break_atr: float = 0.05
    max_bars_to_choch: int = 20
    invalidation_close_atr: float = 0.10

    enable_breakout_retest: bool = True
    breakout_body_atr: float = 0.70
    breakout_body_percent: float = 65.0
    breakout_close_atr: float = 0.10
    breakout_retest_bars: int = 20
    require_breakout_fvg: bool = False
    breakout_invalidation_atr: float = 0.15

    show_important_fvg: bool = True
    minimum_fvg_atr: float = 0.15
    displacement_body_atr: float = 0.70
    displacement_body_percent: float = 65.0
    maximum_important_fvg: int = 1

    show_order_block: bool = True
    order_block_lookback: int = 12
    minimum_choch_displacement_atr: float = 0.50

    entry_body_percent: float = 45.0
    entry_confirmation_bars: int = 15
    retest_tolerance_atr: float = 0.20
    accept_bos_retest: bool = True
    accept_fvg_retest: bool = True
    accept_order_block_retest: bool = True
    require_fvg_or_ob: bool = True
    require_close_beyond_bos: bool = True
    require_midpoint_alignment: bool = True
    require_ema_alignment: bool = False
    fast_ema_length: int = 20
    slow_ema_length: int = 50
    minimum_potential_score: int = 3

    enable_sweep_rejection_entry: bool = True
    sweep_rejection_bars: int = 4
    minimum_rejection_wick_percent: float = 35.0

    enable_high_potential_zones: bool = True
    important_zone_tolerance_atr: float = 0.15

    result_risk_reward: float = 1.5
    result_stop_buffer_atr: float = 0.10


def run_crt_tbs(
    *,
    csv_path: str | None = None,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    show_plot: bool = True,
    save_csv_path: str | None = None,
    **overrides: Any,
) -> dict[str, Any]:
    config = CRTTBSConfig(symbol=symbol, period=period, interval=interval, **overrides)
    data = load_ohlcv(csv_path=csv_path, config=config)
    data = add_indicators(data, config)
    result = run_pine_state_machine(data, config)

    if save_csv_path is not None:
        Path(save_csv_path).parent.mkdir(parents=True, exist_ok=True)
        result["entries"].to_csv(save_csv_path, index=False)

    fig = plot_crt_tbs_state_machine(data, result, title=f"{config.symbol} CRT TBS Pine-style")
    if show_plot and fig is not None:
        display_figure(fig)

    print("CRT TBS Pine-style")
    print(f"bars: {len(data)} | CRTs: {len(result['crts'])} | events: {len(result['events'])} | entries: {len(result['entries'])}")
    if len(result["entries"]):
        preview = [
            "Signal",
            "Model",
            "Component",
            "Potential",
            "EntryDate",
            "EntryPrice",
            "StopPrice",
            "TargetPrice",
            "Result",
        ]
        print(result["entries"][preview].head(30).to_string(index=False))

    return {"config": config, "data": data, "fig": fig, **result, "setups": result["entries"]}


def load_ohlcv(*, csv_path: str | None, config: CRTTBSConfig) -> pd.DataFrame:
    if csv_path is not None:
        return prepare_ohlcv(pd.read_csv(csv_path))
    if Ticker is None:
        raise ImportError("Install yahooquery or pass csv_path.")

    raw = Ticker(config.symbol).history(period=config.period, interval=config.interval)
    if isinstance(raw, pd.DataFrame) and isinstance(raw.index, pd.MultiIndex):
        if "symbol" in raw.index.names:
            raw = raw.reset_index(level="symbol", drop=True)
        else:
            raw = raw.reset_index()
    return prepare_ohlcv(raw)


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    data = df.copy()
    data = data.rename(columns={column: str(column).lower() for column in data.columns})
    if not isinstance(data.index, pd.DatetimeIndex):
        date_col = next(
            (col for col in ["datetime", "date", "timestamp", "time"] if col in data.columns),
            None,
        )
        if date_col is None:
            raise ValueError("Use DatetimeIndex or include datetime/date/timestamp column.")
        data[date_col] = pd.to_datetime(data[date_col], errors="coerce")
        data = data.set_index(date_col)

    required = {"open", "high", "low", "close"}
    missing = required.difference(data.columns)
    if missing:
        raise ValueError(f"Missing OHLC columns: {sorted(missing)}")

    data.index = pd.to_datetime(data.index)
    data = data[~data.index.duplicated(keep="last")].sort_index()
    if "volume" not in data.columns:
        data["volume"] = 0.0
    for col in ["open", "high", "low", "close", "volume"]:
        data[col] = pd.to_numeric(data[col], errors="coerce")
    data = data.dropna(subset=["open", "high", "low", "close"])
    data["volume"] = data["volume"].fillna(0.0).clip(lower=0.0)
    return data[["open", "high", "low", "close", "volume"]]


def add_indicators(data: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    out = data.copy()
    high = out["high"].astype(float)
    low = out["low"].astype(float)
    open_ = out["open"].astype(float)
    close = out["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [high - low, (high - previous_close).abs(), (low - previous_close).abs()],
        axis=1,
    ).max(axis=1)
    out["atr"] = true_range.rolling(config.atr_length, min_periods=1).mean()
    out["candle_range"] = high - low
    out["candle_body"] = (close - open_).abs()
    out["body_ratio"] = out["candle_body"] / out["candle_range"].replace(0, np.nan)
    out["body_ratio"] = out["body_ratio"].fillna(0.0)
    out["close_location"] = (close - low) / out["candle_range"].replace(0, np.nan)
    out["close_location"] = out["close_location"].fillna(0.5)
    out["range_to_atr"] = out["candle_range"] / out["atr"].replace(0, np.nan)
    out["fast_ema"] = close.ewm(span=config.fast_ema_length, adjust=False, min_periods=1).mean()
    out["slow_ema"] = close.ewm(span=config.slow_ema_length, adjust=False, min_periods=1).mean()
    return out.replace([np.inf, -np.inf], np.nan)


def run_pine_state_machine(data: pd.DataFrame, config: CRTTBSConfig) -> dict[str, pd.DataFrame]:
    o = data["open"].to_numpy(dtype=float)
    h = data["high"].to_numpy(dtype=float)
    l = data["low"].to_numpy(dtype=float)
    c = data["close"].to_numpy(dtype=float)
    atr = data["atr"].to_numpy(dtype=float)
    body = data["candle_body"].to_numpy(dtype=float)
    body_ratio = data["body_ratio"].to_numpy(dtype=float)
    candle_range = data["candle_range"].to_numpy(dtype=float)
    close_location = data["close_location"].to_numpy(dtype=float)
    range_to_atr = data["range_to_atr"].to_numpy(dtype=float)
    fast_ema = data["fast_ema"].to_numpy(dtype=float)
    slow_ema = data["slow_ema"].to_numpy(dtype=float)

    tz_index = index_in_timezone(data.index, config)
    inside_main = np.array([in_session(ts, config.main_session) for ts in tz_index], dtype=bool)
    inside_secondary = np.array(
        [config.search_24x7 or in_session(ts, config.secondary_session) for ts in tz_index],
        dtype=bool,
    )
    date_keys = np.array([date_key(ts) for ts in tz_index], dtype=int)

    state = new_state()
    crts: list[dict[str, Any]] = []
    events: list[dict[str, Any]] = []
    entries: list[dict[str, Any]] = []
    active_trades: list[dict[str, Any]] = []

    best_main_range = np.nan
    best_main_high = np.nan
    best_main_low = np.nan
    best_main_bar = None
    main_crt_created_today = False
    crts_today = 0
    last_crt_creation_bar = None
    last_swing_high = np.nan
    last_swing_low = np.nan
    next_trade_id = 1

    for i in range(len(data)):
        new_trading_day = i > 0 and date_keys[i] != date_keys[i - 1]
        main_session_started = bool(inside_main[i] and (i == 0 or not inside_main[i - 1]))
        main_session_finished = bool((not inside_main[i]) and i > 0 and inside_main[i - 1])

        if new_trading_day:
            state = new_state()
            main_crt_created_today = False
            crts_today = 0
            last_crt_creation_bar = None
            best_main_range = np.nan
            best_main_high = np.nan
            best_main_low = np.nan
            best_main_bar = None

        if new_trading_day or main_session_started:
            best_main_range = np.nan
            best_main_high = np.nan
            best_main_low = np.nan
            best_main_bar = None

        if inside_main[i] and (not np.isfinite(best_main_range) or candle_range[i] > best_main_range):
            best_main_range = candle_range[i]
            best_main_high = h[i]
            best_main_low = l[i]
            best_main_bar = i

        pivot_high, pivot_low = confirmed_pivots(
            i=i,
            high=h,
            low=l,
            left=config.swing_left,
            right=config.swing_right,
        )
        if np.isfinite(pivot_high):
            last_swing_high = pivot_high
        if np.isfinite(pivot_low):
            last_swing_low = pivot_low

        create_main_crt = (
            main_session_finished
            and not main_crt_created_today
            and np.isfinite(best_main_high)
            and np.isfinite(best_main_low)
            and best_main_bar is not None
        )
        if create_main_crt:
            main_crt_created_today = True
            crts_today += 1
            state = activate_crt(
                crt_high=float(best_main_high),
                crt_low=float(best_main_low),
                crt_start_bar=int(best_main_bar),
                activation_bar=i,
                crt_type="PRINCIPAL",
                crt_number=crts_today,
            )
            last_crt_creation_bar = i
            crts.append(crt_record(data, state))
            add_event(events, data, i, state, "CRT", "PRINCIPAL", price=state["active_mid"])

        if state["setup_state"] >= 1 and state["setup_state"] <= 4:
            if i - state["active_activation_bar"] > config.setup_expiration_bars:
                add_event(events, data, i, state, "EXPIRED", "setup_expiration")
                state["setup_state"] = 6
                state["status"] = "EXPIRADO"

        raw_bull_fvg, raw_bear_fvg = raw_fvg(i, h, l, atr, config)
        update_important_fvg(i, o, h, l, c, atr, body, body_ratio, raw_bull_fvg, raw_bear_fvg, state, config, events, data)

        bullish_sweep, bearish_sweep = process_tbs_sweep(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            state=state,
            last_swing_high=last_swing_high,
            last_swing_low=last_swing_low,
            config=config,
            events=events,
            data=data,
        )

        process_real_breakout(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body=body,
            body_ratio=body_ratio,
            raw_bull_fvg=raw_bull_fvg,
            raw_bear_fvg=raw_bear_fvg,
            bullish_sweep=bullish_sweep,
            bearish_sweep=bearish_sweep,
            state=state,
            config=config,
            events=events,
            data=data,
        )

        invalidate_before_choch(i, c, atr, state, config, events, data)
        process_choch(i, o, h, l, c, atr, body, state, config, events, data)
        process_bos_level(i, o, h, l, c, state)
        process_bos_confirmation(i, c, atr, state, config, events, data)
        invalidate_order_block(i, c, state, events, data)

        confirmation = evaluate_entries(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body_ratio=body_ratio,
            fast_ema=fast_ema,
            slow_ema=slow_ema,
            state=state,
            config=config,
        )
        if confirmation is not None:
            state["setup_state"] = 5
            add_event(events, data, i, state, "ENTRY", confirmation["model"], price=float(c[i]))
            entry = entry_record(data, i, state, confirmation, c, atr, config, next_trade_id)
            next_trade_id += 1
            entries.append(entry)
            active_trades.append(entry.copy())

        update_trades(i, h, l, data, active_trades, entries)

        if state["setup_state"] == 4 and i - state.get("bos_bar", i) > config.entry_confirmation_bars:
            add_event(events, data, i, state, "EXPIRED", "entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"
        if state["setup_state"] == 8 and i - state.get("breakout_bar", i) > config.breakout_retest_bars:
            add_event(events, data, i, state, "EXPIRED", "breakout_entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"

        if create_secondary_crt_allowed(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            candle_range=candle_range,
            body_ratio=body_ratio,
            close_location=close_location,
            range_to_atr=range_to_atr,
            inside_secondary=inside_secondary,
            inside_main=inside_main,
            main_session_finished=main_session_finished,
            state=state,
            crts_today=crts_today,
            main_crt_created_today=main_crt_created_today,
            last_crt_creation_bar=last_crt_creation_bar,
            config=config,
        ):
            crts_today += 1
            state = activate_crt(
                crt_high=float(h[i]),
                crt_low=float(l[i]),
                crt_start_bar=i,
                activation_bar=i,
                crt_type="SECUNDARIO",
                crt_number=crts_today,
            )
            last_crt_creation_bar = i
            crts.append(crt_record(data, state))
            add_event(events, data, i, state, "CRT", "SECUNDARIO", price=state["active_mid"])

    return {
        "crts": pd.DataFrame(crts),
        "events": pd.DataFrame(events),
        "entries": pd.DataFrame(entries) if entries else empty_entries(),
        "trades": pd.DataFrame(entries) if entries else empty_entries(),
    }


def new_state() -> dict[str, Any]:
    return {
        "setup_state": 0,
        "setup_direction": 0,
        "active_high": np.nan,
        "active_low": np.nan,
        "active_mid": np.nan,
        "active_start_bar": None,
        "active_activation_bar": None,
        "active_type": "",
        "active_number": np.nan,
        "sweep_bar": None,
        "low_swept": False,
        "high_swept": False,
        "choch_level": np.nan,
        "choch_bar": None,
        "impulse_extreme": np.nan,
        "bos_level": np.nan,
        "bos_bar": None,
        "pullback_detected": False,
        "first_retest_consumed": False,
        "breakout_level": np.nan,
        "breakout_bar": None,
        "breakout_had_fvg": False,
        "breakout_retest_consumed": False,
        "important_fvg_count": 0,
        "bullish_fvg_top": np.nan,
        "bullish_fvg_bottom": np.nan,
        "bullish_fvg_creation_bar": None,
        "bearish_fvg_top": np.nan,
        "bearish_fvg_bottom": np.nan,
        "bearish_fvg_creation_bar": None,
        "bullish_ob_high": np.nan,
        "bullish_ob_low": np.nan,
        "bullish_ob_creation_bar": None,
        "bearish_ob_high": np.nan,
        "bearish_ob_low": np.nan,
        "bearish_ob_creation_bar": None,
        "bullish_ob_created": False,
        "bearish_ob_created": False,
        "status": "Esperando",
    }


def activate_crt(*, crt_high: float, crt_low: float, crt_start_bar: int, activation_bar: int, crt_type: str, crt_number: int) -> dict[str, Any]:
    state = new_state()
    state.update(
        {
            "setup_state": 1,
            "active_high": float(crt_high),
            "active_low": float(crt_low),
            "active_mid": (float(crt_high) + float(crt_low)) / 2.0,
            "active_start_bar": int(crt_start_bar),
            "active_activation_bar": int(activation_bar),
            "active_type": crt_type,
            "active_number": int(crt_number),
            "status": "Esperando TBS",
        }
    )
    return state


def process_tbs_sweep(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    state: dict[str, Any],
    last_swing_high: float,
    last_swing_low: float,
    config: CRTTBSConfig,
    events: list[dict[str, Any]],
    data: pd.DataFrame,
) -> tuple[bool, bool]:
    if state["setup_state"] != 1 or i <= state["active_activation_bar"]:
        return False, False
    minimum_sweep = float(config.sweep_ticks) * float(config.tick_size)
    low_sweep_raw = l[i] <= state["active_low"] - minimum_sweep
    high_sweep_raw = h[i] >= state["active_high"] + minimum_sweep
    if low_sweep_raw:
        state["low_swept"] = True
    if high_sweep_raw:
        state["high_swept"] = True

    bullish_sweep = low_sweep_raw and not high_sweep_raw and (not config.require_close_inside or c[i] > state["active_low"])
    bearish_sweep = high_sweep_raw and not low_sweep_raw and (not config.require_close_inside or c[i] < state["active_high"])
    if bullish_sweep:
        fallback_high = finite_max(h[max(0, i - config.structure_lookback) : i])
        state["setup_state"] = 2
        state["setup_direction"] = 1
        state["sweep_bar"] = i
        state["choch_level"] = last_swing_high if np.isfinite(last_swing_high) else fallback_high
        state["status"] = "Esperando CHoCH"
        add_event(events, data, i, state, "TBS", "bullish_sweep", price=float(l[i]), level=state["choch_level"])
    elif bearish_sweep:
        fallback_low = finite_min(l[max(0, i - config.structure_lookback) : i])
        state["setup_state"] = 2
        state["setup_direction"] = -1
        state["sweep_bar"] = i
        state["choch_level"] = last_swing_low if np.isfinite(last_swing_low) else fallback_low
        state["status"] = "Esperando CHoCH"
        add_event(events, data, i, state, "TBS", "bearish_sweep", price=float(h[i]), level=state["choch_level"])
    return bool(bullish_sweep), bool(bearish_sweep)


def process_real_breakout(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    atr: np.ndarray,
    body: np.ndarray,
    body_ratio: np.ndarray,
    raw_bull_fvg: bool,
    raw_bear_fvg: bool,
    bullish_sweep: bool,
    bearish_sweep: bool,
    state: dict[str, Any],
    config: CRTTBSConfig,
    events: list[dict[str, Any]],
    data: pd.DataFrame,
) -> None:
    if not config.enable_breakout_retest or state["setup_state"] != 1 or i <= state["active_activation_bar"]:
        return
    breakout_margin = atr[i] * config.breakout_close_atr
    strong_body = body[i] >= atr[i] * config.breakout_body_atr and body_ratio[i] * 100.0 >= config.breakout_body_percent
    bullish_breakout = c[i] > state["active_high"] + breakout_margin and strong_body and not bullish_sweep and not bearish_sweep
    bearish_breakout = c[i] < state["active_low"] - breakout_margin and strong_body and not bullish_sweep and not bearish_sweep
    if bullish_breakout:
        state["setup_state"] = 8
        state["setup_direction"] = 1
        state["breakout_level"] = state["active_high"]
        state["breakout_bar"] = i
        state["breakout_had_fvg"] = bool(raw_bull_fvg or (i > 0 and raw_fvg(i - 1, h, l, atr, config)[0]))
        state["breakout_retest_consumed"] = False
        state["status"] = "Breakout: esperando retest"
        add_event(events, data, i, state, "BREAKOUT", "bullish_breakout", price=float(c[i]), level=state["breakout_level"])
    elif bearish_breakout:
        state["setup_state"] = 8
        state["setup_direction"] = -1
        state["breakout_level"] = state["active_low"]
        state["breakout_bar"] = i
        state["breakout_had_fvg"] = bool(raw_bear_fvg or (i > 0 and raw_fvg(i - 1, h, l, atr, config)[1]))
        state["breakout_retest_consumed"] = False
        state["status"] = "Breakout: esperando retest"
        add_event(events, data, i, state, "BREAKOUT", "bearish_breakout", price=float(c[i]), level=state["breakout_level"])


def invalidate_before_choch(i: int, c: np.ndarray, atr: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 2:
        return
    both_sides = state["low_swept"] and state["high_swept"]
    time_expired = state["sweep_bar"] is not None and i - state["sweep_bar"] > config.max_bars_to_choch
    bullish_opposite = state["setup_direction"] == 1 and c[i] < state["active_low"] - atr[i] * config.invalidation_close_atr
    bearish_opposite = state["setup_direction"] == -1 and c[i] > state["active_high"] + atr[i] * config.invalidation_close_atr
    if both_sides or time_expired or bullish_opposite or bearish_opposite:
        state["setup_state"] = 7
        state["status"] = "INVALIDADO"
        add_event(events, data, i, state, "INVALIDATED", "before_choch")


def process_choch(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, atr: np.ndarray, body: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 2 or state["sweep_bar"] is None or i <= state["sweep_bar"] or not np.isfinite(state["choch_level"]):
        return
    current_body_atr = body[i] / max(atr[i], 1e-12)
    bullish = state["setup_direction"] == 1 and c[i] > state["choch_level"] and current_body_atr >= config.minimum_choch_displacement_atr
    bearish = state["setup_direction"] == -1 and c[i] < state["choch_level"] and current_body_atr >= config.minimum_choch_displacement_atr
    if bullish or bearish:
        state["setup_state"] = 3
        state["choch_bar"] = i
        state["impulse_extreme"] = h[i] if bullish else l[i]
        state["pullback_detected"] = False
        state["status"] = "Esperando retroceso"
        create_order_block(i, o, h, l, c, state, config)
        add_event(events, data, i, state, "CHoCH", "bullish" if bullish else "bearish", price=float(c[i]), level=state["choch_level"])


def process_bos_level(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any]) -> None:
    if state["setup_state"] != 3:
        return
    if state["setup_direction"] == 1 and not state["pullback_detected"]:
        state["impulse_extreme"] = max(state["impulse_extreme"], h[i]) if np.isfinite(state["impulse_extreme"]) else h[i]
        bullish_pullback = state["choch_bar"] is not None and i > state["choch_bar"] and (c[i] < o[i] or l[i] < l[i - 1])
        if bullish_pullback:
            state["pullback_detected"] = True
            state["bos_level"] = state["impulse_extreme"]
            state["status"] = "Esperando BOS"
    elif state["setup_direction"] == -1 and not state["pullback_detected"]:
        state["impulse_extreme"] = min(state["impulse_extreme"], l[i]) if np.isfinite(state["impulse_extreme"]) else l[i]
        bearish_pullback = state["choch_bar"] is not None and i > state["choch_bar"] and (c[i] > o[i] or h[i] > h[i - 1])
        if bearish_pullback:
            state["pullback_detected"] = True
            state["bos_level"] = state["impulse_extreme"]
            state["status"] = "Esperando BOS"


def process_bos_confirmation(i: int, c: np.ndarray, atr: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 3 or not state["pullback_detected"] or not np.isfinite(state["bos_level"]):
        return
    distance = atr[i] * config.bos_break_atr
    bullish = state["setup_direction"] == 1 and c[i] > state["bos_level"] + distance
    bearish = state["setup_direction"] == -1 and c[i] < state["bos_level"] - distance
    if bullish or bearish:
        state["setup_state"] = 4
        state["bos_bar"] = i
        state["first_retest_consumed"] = False
        state["status"] = "Esperando entrada"
        add_event(events, data, i, state, "BOS", "bullish" if bullish else "bearish", price=float(c[i]), level=state["bos_level"])


def evaluate_entries(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    atr: np.ndarray,
    body_ratio: np.ndarray,
    fast_ema: np.ndarray,
    slow_ema: np.ndarray,
    state: dict[str, Any],
    config: CRTTBSConfig,
) -> dict[str, Any] | None:
    current_atr = safe_atr_value(atr[i])
    (
        bullish_rejection_candle,
        bearish_rejection_candle,
        bullish_rejection,
        bearish_rejection,
        bullish_rejection_score,
        bearish_rejection_score,
    ) = rejection_signals(i, o, h, l, c, body_ratio, fast_ema, slow_ema, state, config)
    tbs = tbs_entry_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config)
    breakout = breakout_retest_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config)
    zone = zone_confluence_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config, bullish_rejection_candle, bearish_rejection_candle)

    if tbs is not None:
        return tbs
    if breakout is not None:
        return breakout
    if bullish_rejection and state["setup_direction"] == 1:
        return {"direction": 1, "signal": "BUY", "component": "BARRIDA + RECHAZO", "model": "TBS REJECTION", "potential": bullish_rejection_score}
    if bearish_rejection and state["setup_direction"] == -1:
        return {"direction": -1, "signal": "SELL", "component": "BARRIDA + RECHAZO", "model": "TBS REJECTION", "potential": bearish_rejection_score}
    if zone is not None:
        return zone
    return None


def tbs_entry_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    valid = state["setup_state"] == 4 and state["bos_bar"] is not None and i > state["bos_bar"] and i - state["bos_bar"] <= config.entry_confirmation_bars and not state["first_retest_consumed"]
    if not valid:
        return None
    tolerance = current_atr * config.retest_tolerance_atr
    bullish_bos_retest = state["setup_direction"] == 1 and config.accept_bos_retest and l[i] <= state["bos_level"] + tolerance and h[i] >= state["bos_level"] - tolerance
    bearish_bos_retest = state["setup_direction"] == -1 and config.accept_bos_retest and h[i] >= state["bos_level"] - tolerance and l[i] <= state["bos_level"] + tolerance
    bullish_fvg_retest = state["setup_direction"] == 1 and config.accept_fvg_retest and state["bullish_fvg_creation_bar"] is not None and state["bullish_fvg_creation_bar"] > state["sweep_bar"] and l[i] <= state["bullish_fvg_top"] + tolerance and h[i] >= state["bullish_fvg_bottom"] - tolerance
    bearish_fvg_retest = state["setup_direction"] == -1 and config.accept_fvg_retest and state["bearish_fvg_creation_bar"] is not None and state["bearish_fvg_creation_bar"] > state["sweep_bar"] and h[i] >= state["bearish_fvg_bottom"] - tolerance and l[i] <= state["bearish_fvg_top"] + tolerance
    bullish_ob_retest = state["setup_direction"] == 1 and config.accept_order_block_retest and state["bullish_ob_created"] and state["bullish_ob_creation_bar"] >= state["sweep_bar"] and l[i] <= state["bullish_ob_high"] + tolerance and h[i] >= state["bullish_ob_low"] - tolerance and c[i] >= state["bullish_ob_low"]
    bearish_ob_retest = state["setup_direction"] == -1 and config.accept_order_block_retest and state["bearish_ob_created"] and state["bearish_ob_creation_bar"] >= state["sweep_bar"] and h[i] >= state["bearish_ob_low"] - tolerance and l[i] <= state["bearish_ob_high"] + tolerance and c[i] <= state["bearish_ob_high"]

    bullish_primary = bullish_ob_retest or bullish_fvg_retest
    bearish_primary = bearish_ob_retest or bearish_fvg_retest
    bullish_zone_accepted = bullish_primary if config.require_fvg_or_ob else (bullish_primary or bullish_bos_retest)
    bearish_zone_accepted = bearish_primary if config.require_fvg_or_ob else (bearish_primary or bearish_bos_retest)
    bullish_confirm = c[i] > o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] > c[i - 1]
    bearish_confirm = c[i] < o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] < c[i - 1]
    bullish_bos_close = (not config.require_close_beyond_bos) or c[i] > state["bos_level"]
    bearish_bos_close = (not config.require_close_beyond_bos) or c[i] < state["bos_level"]
    bullish_mid = (not config.require_midpoint_alignment) or c[i] > state["active_mid"]
    bearish_mid = (not config.require_midpoint_alignment) or c[i] < state["active_mid"]
    bullish_ema = (not config.require_ema_alignment) or (fast_ema[i] > slow_ema[i] and c[i] > fast_ema[i])
    bearish_ema = (not config.require_ema_alignment) or (fast_ema[i] < slow_ema[i] and c[i] < fast_ema[i])
    bullish_score = int(bullish_ob_retest) + int(bullish_fvg_retest) + int(bullish_bos_close) + int(bullish_mid) + int(bullish_ema)
    bearish_score = int(bearish_ob_retest) + int(bearish_fvg_retest) + int(bearish_bos_close) + int(bearish_mid) + int(bearish_ema)
    if bullish_zone_accepted or bearish_zone_accepted:
        state["first_retest_consumed"] = True

    if state["setup_direction"] == 1 and bullish_zone_accepted and bullish_confirm and bullish_bos_close and bullish_mid and bullish_ema and bullish_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if bullish_ob_retest else ("FVG" if bullish_fvg_retest else "BOS")
        return {"direction": 1, "signal": "BUY", "component": component, "model": "TBS REVERSIÓN", "potential": bullish_score}
    if state["setup_direction"] == -1 and bearish_zone_accepted and bearish_confirm and bearish_bos_close and bearish_mid and bearish_ema and bearish_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if bearish_ob_retest else ("FVG" if bearish_fvg_retest else "BOS")
        return {"direction": -1, "signal": "SELL", "component": component, "model": "TBS REVERSIÓN", "potential": bearish_score}
    return None


def breakout_retest_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    valid = state["setup_state"] == 8 and state["breakout_bar"] is not None and i > state["breakout_bar"] and i - state["breakout_bar"] <= config.breakout_retest_bars and not state["breakout_retest_consumed"]
    if not valid:
        return None
    tolerance = current_atr * config.retest_tolerance_atr
    bullish_touch = state["setup_direction"] == 1 and l[i] <= state["breakout_level"] + tolerance and h[i] >= state["breakout_level"] - tolerance
    bearish_touch = state["setup_direction"] == -1 and h[i] >= state["breakout_level"] - tolerance and l[i] <= state["breakout_level"] + tolerance
    bullish_confirm = bullish_touch and c[i] > state["breakout_level"] and c[i] > o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] > c[i - 1]
    bearish_confirm = bearish_touch and c[i] < state["breakout_level"] and c[i] < o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] < c[i - 1]
    bullish_ema = (not config.require_ema_alignment) or (fast_ema[i] > slow_ema[i] and c[i] > fast_ema[i])
    bearish_ema = (not config.require_ema_alignment) or (fast_ema[i] < slow_ema[i] and c[i] < fast_ema[i])
    if bullish_touch or bearish_touch:
        state["breakout_retest_consumed"] = True
    if state["setup_direction"] == 1 and c[i] < state["active_high"] - current_atr * config.breakout_invalidation_atr:
        state["setup_state"] = 7
        return None
    if state["setup_direction"] == -1 and c[i] > state["active_low"] + current_atr * config.breakout_invalidation_atr:
        state["setup_state"] = 7
        return None
    bullish_score = int(state["breakout_had_fvg"]) + int(c[i] > state["active_high"]) + int(bullish_ema) + int(body_ratio[i] * 100.0 >= config.entry_body_percent) + 1
    bearish_score = int(state["breakout_had_fvg"]) + int(c[i] < state["active_low"]) + int(bearish_ema) + int(body_ratio[i] * 100.0 >= config.entry_body_percent) + 1
    if state["setup_direction"] == 1 and bullish_confirm and bullish_ema and (not config.require_breakout_fvg or state["breakout_had_fvg"]) and bullish_score >= config.minimum_potential_score:
        return {"direction": 1, "signal": "BUY", "component": "CRT HIGH RETEST", "model": "BREAKOUT RETEST", "potential": bullish_score}
    if state["setup_direction"] == -1 and bearish_confirm and bearish_ema and (not config.require_breakout_fvg or state["breakout_had_fvg"]) and bearish_score >= config.minimum_potential_score:
        return {"direction": -1, "signal": "SELL", "component": "CRT LOW RETEST", "model": "BREAKOUT RETEST", "potential": bearish_score}
    return None


def rejection_signals(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> tuple[bool, bool, bool, bool, int, int]:
    candle_range = max(h[i] - l[i], 1e-12)
    lower_wick_pct = (min(o[i], c[i]) - l[i]) / candle_range * 100.0
    upper_wick_pct = (h[i] - max(o[i], c[i])) / candle_range * 100.0
    bullish_candle = c[i] > o[i] and lower_wick_pct >= config.minimum_rejection_wick_percent and c[i] > state.get("active_low", np.nan) and c[i] > c[i - 1]
    bearish_candle = c[i] < o[i] and upper_wick_pct >= config.minimum_rejection_wick_percent and c[i] < state.get("active_high", np.nan) and c[i] < c[i - 1]
    valid_bullish = config.enable_sweep_rejection_entry and state["setup_state"] == 2 and state["setup_direction"] == 1 and state["sweep_bar"] is not None and i >= state["sweep_bar"] and i - state["sweep_bar"] <= config.sweep_rejection_bars
    valid_bearish = config.enable_sweep_rejection_entry and state["setup_state"] == 2 and state["setup_direction"] == -1 and state["sweep_bar"] is not None and i >= state["sweep_bar"] and i - state["sweep_bar"] <= config.sweep_rejection_bars
    bullish_score = int(c[i] > state.get("active_low", np.nan)) + int(lower_wick_pct >= config.minimum_rejection_wick_percent) + int(c[i] > o[i]) + int(c[i] > c[i - 1]) + int((not config.require_ema_alignment) or fast_ema[i] > slow_ema[i])
    bearish_score = int(c[i] < state.get("active_high", np.nan)) + int(upper_wick_pct >= config.minimum_rejection_wick_percent) + int(c[i] < o[i]) + int(c[i] < c[i - 1]) + int((not config.require_ema_alignment) or fast_ema[i] < slow_ema[i])
    confirmed_bullish = valid_bullish and bullish_candle and bullish_score >= config.minimum_potential_score
    confirmed_bearish = valid_bearish and bearish_candle and bearish_score >= config.minimum_potential_score
    return bullish_candle, bearish_candle, confirmed_bullish, confirmed_bearish, bullish_score, bearish_score


def zone_confluence_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig, bullish_rejection_candle: bool, bearish_rejection_candle: bool) -> dict[str, Any] | None:
    if not config.enable_high_potential_zones or state["setup_state"] not in {1, 2, 3, 4}:
        return None
    tolerance = current_atr * config.important_zone_tolerance_atr
    price_at_bull_ob = state["bullish_ob_created"] and l[i] <= state["bullish_ob_high"] + tolerance and h[i] >= state["bullish_ob_low"] - tolerance and c[i] >= state["bullish_ob_low"]
    price_at_bear_ob = state["bearish_ob_created"] and h[i] >= state["bearish_ob_low"] - tolerance and l[i] <= state["bearish_ob_high"] + tolerance and c[i] <= state["bearish_ob_high"]
    price_at_bull_fvg = state["bullish_fvg_creation_bar"] is not None and l[i] <= state["bullish_fvg_top"] + tolerance and h[i] >= state["bullish_fvg_bottom"] - tolerance
    price_at_bear_fvg = state["bearish_fvg_creation_bar"] is not None and h[i] >= state["bearish_fvg_bottom"] - tolerance and l[i] <= state["bearish_fvg_top"] + tolerance
    price_at_low_liq = np.isfinite(state["active_low"]) and l[i] <= state["active_low"] + tolerance and c[i] > state["active_low"]
    price_at_high_liq = np.isfinite(state["active_high"]) and h[i] >= state["active_high"] - tolerance and c[i] < state["active_high"]
    bull_score = int(price_at_bull_ob) + int(price_at_bull_fvg) + int(price_at_low_liq) + int(bullish_rejection_candle) + int((not config.require_ema_alignment) or fast_ema[i] > slow_ema[i])
    bear_score = int(price_at_bear_ob) + int(price_at_bear_fvg) + int(price_at_high_liq) + int(bearish_rejection_candle) + int((not config.require_ema_alignment) or fast_ema[i] < slow_ema[i])
    if (price_at_bull_ob or price_at_bull_fvg or price_at_low_liq) and bullish_rejection_candle and bull_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if price_at_bull_ob else ("FVG" if price_at_bull_fvg else "LIQUIDEZ CRT")
        return {"direction": 1, "signal": "BUY", "component": component, "model": "ZONE CONFLUENCE", "potential": bull_score}
    if (price_at_bear_ob or price_at_bear_fvg or price_at_high_liq) and bearish_rejection_candle and bear_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if price_at_bear_ob else ("FVG" if price_at_bear_fvg else "LIQUIDEZ CRT")
        return {"direction": -1, "signal": "SELL", "component": component, "model": "ZONE CONFLUENCE", "potential": bear_score}
    return None


def invalidate_order_block(i: int, c: np.ndarray, state: dict[str, Any], events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] < 3 or state["setup_state"] > 4:
        return
    bullish_broken = (
        state["setup_direction"] == 1
        and state["bullish_ob_created"]
        and np.isfinite(state["bullish_ob_low"])
        and c[i] < state["bullish_ob_low"]
    )
    bearish_broken = (
        state["setup_direction"] == -1
        and state["bearish_ob_created"]
        and np.isfinite(state["bearish_ob_high"])
        and c[i] > state["bearish_ob_high"]
    )
    if bullish_broken or bearish_broken:
        state["setup_state"] = 7
        state["status"] = "INVALIDADO"
        add_event(events, data, i, state, "INVALIDATED", "order_block_broken")


def create_order_block(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> None:
    if not config.show_order_block or state["sweep_bar"] is None:
        return
    max_search = min(config.order_block_lookback, i - state["sweep_bar"])
    for offset in range(1, max_search + 1):
        idx = i - offset
        if state["setup_direction"] == 1 and c[idx] < o[idx]:
            state["bullish_ob_high"] = h[idx]
            state["bullish_ob_low"] = l[idx]
            state["bullish_ob_creation_bar"] = idx
            state["bullish_ob_created"] = True
            return
        if state["setup_direction"] == -1 and c[idx] > o[idx]:
            state["bearish_ob_high"] = h[idx]
            state["bearish_ob_low"] = l[idx]
            state["bearish_ob_creation_bar"] = idx
            state["bearish_ob_created"] = True
            return


def update_important_fvg(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, atr: np.ndarray, body: np.ndarray, body_ratio: np.ndarray, raw_bull: bool, raw_bear: bool, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if i < 2 or not config.show_important_fvg or state["important_fvg_count"] >= config.maximum_important_fvg:
        return
    center_body = abs(c[i - 1] - o[i - 1])
    center_range = max(h[i - 1] - l[i - 1], 1e-12)
    center_body_ratio = center_body / center_range
    displacement = center_body >= atr[i - 1] * config.displacement_body_atr and center_body_ratio * 100.0 >= config.displacement_body_percent
    state_ok = state["setup_state"] in {2, 3, 4, 8}
    impulse_started = (state["sweep_bar"] is not None and i > state["sweep_bar"]) or (state["breakout_bar"] is not None and i >= state["breakout_bar"])
    if state["setup_direction"] == 1 and state_ok and impulse_started and raw_bull and displacement:
        mid = (l[i] + h[i - 2]) / 2.0
        if abs(c[i] - mid) <= atr[i] * 2.0:
            state["bullish_fvg_top"] = l[i]
            state["bullish_fvg_bottom"] = h[i - 2]
            state["bullish_fvg_creation_bar"] = i
            state["important_fvg_count"] += 1
            add_event(events, data, i, state, "FVG", "bullish", price=mid)
    if state["setup_direction"] == -1 and state_ok and impulse_started and raw_bear and displacement:
        mid = (l[i - 2] + h[i]) / 2.0
        if abs(c[i] - mid) <= atr[i] * 2.0:
            state["bearish_fvg_top"] = l[i - 2]
            state["bearish_fvg_bottom"] = h[i]
            state["bearish_fvg_creation_bar"] = i
            state["important_fvg_count"] += 1
            add_event(events, data, i, state, "FVG", "bearish", price=mid)


def create_secondary_crt_allowed(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    candle_range: np.ndarray,
    body_ratio: np.ndarray,
    close_location: np.ndarray,
    range_to_atr: np.ndarray,
    inside_secondary: np.ndarray,
    inside_main: np.ndarray,
    main_session_finished: bool,
    state: dict[str, Any],
    crts_today: int,
    main_crt_created_today: bool,
    last_crt_creation_bar: int | None,
    config: CRTTBSConfig,
) -> bool:
    secondary_bull_quality = c[i] > o[i] and close_location[i] >= config.secondary_close_extreme
    secondary_bear_quality = c[i] < o[i] and close_location[i] <= 1.0 - config.secondary_close_extreme
    secondary_expansion = (not config.require_secondary_expansion) or h[i] > h[i - 1] or l[i] < l[i - 1]
    secondary_range_expansion = (not config.require_secondary_range_expansion) or candle_range[i] > candle_range[i - 1]
    valid_secondary = inside_secondary[i] and range_to_atr[i] >= config.secondary_range_atr and body_ratio[i] * 100.0 >= config.secondary_body_percent and (secondary_bull_quality or secondary_bear_quality) and secondary_expansion and secondary_range_expansion
    previous_finished = state["setup_state"] in {0, 5, 6, 7}
    cooldown = last_crt_creation_bar is None or i - last_crt_creation_bar >= config.secondary_cooldown_bars
    time_allowed = (not config.secondary_only_after_main) or main_crt_created_today
    return valid_secondary and time_allowed and not inside_main[i] and previous_finished and cooldown and crts_today < config.maximum_crts_per_day and not main_session_finished


def entry_record(data: pd.DataFrame, i: int, state: dict[str, Any], confirmation: dict[str, Any], c: np.ndarray, atr: np.ndarray, config: CRTTBSConfig, trade_id: int) -> dict[str, Any]:
    direction = int(confirmation["direction"])
    entry = float(c[i])
    structural_stop = state["active_low"] if direction == 1 else state["active_high"]
    if confirmation["component"] == "ORDER BLOCK":
        if direction == 1 and np.isfinite(state["bullish_ob_low"]):
            structural_stop = state["bullish_ob_low"]
        elif direction == -1 and np.isfinite(state["bearish_ob_high"]):
            structural_stop = state["bearish_ob_high"]
    stop = structural_stop - atr[i] * config.result_stop_buffer_atr if direction == 1 else structural_stop + atr[i] * config.result_stop_buffer_atr
    risk = abs(entry - stop)
    target = entry + risk * config.result_risk_reward if direction == 1 else entry - risk * config.result_risk_reward
    return {
        "TradeId": trade_id,
        "CRTNumber": state["active_number"],
        "CRTType": state["active_type"],
        "Signal": confirmation["signal"],
        "Direction": direction,
        "Model": confirmation["model"],
        "Component": confirmation["component"],
        "Potential": confirmation["potential"],
        "EntryBarIdx": i,
        "EntryDate": data.index[i],
        "EntryPrice": entry,
        "StopPrice": float(stop),
        "TargetPrice": float(target),
        "Result": "ACTIVE",
        "ResultBarIdx": np.nan,
        "ResultDate": pd.NaT,
        "ActiveCRTHigh": state["active_high"],
        "ActiveCRTLow": state["active_low"],
        "TBSBarIdx": state["sweep_bar"],
        "ChoChBarIdx": state["choch_bar"],
        "BOSBarIdx": state["bos_bar"],
        "BreakoutBarIdx": state["breakout_bar"],
    }


def update_trades(i: int, h: np.ndarray, l: np.ndarray, data: pd.DataFrame, active_trades: list[dict[str, Any]], entries: list[dict[str, Any]]) -> None:
    remaining = []
    for trade in active_trades:
        if i <= trade["EntryBarIdx"]:
            remaining.append(trade)
            continue
        direction = int(trade["Direction"])
        target_hit = h[i] >= trade["TargetPrice"] if direction == 1 else l[i] <= trade["TargetPrice"]
        stop_hit = l[i] <= trade["StopPrice"] if direction == 1 else h[i] >= trade["StopPrice"]
        if target_hit or stop_hit:
            result = "WIN" if target_hit and not stop_hit else "FAILD"
            trade["Result"] = result
            trade["ResultBarIdx"] = i
            trade["ResultDate"] = data.index[i]
            for row in entries:
                if row["TradeId"] == trade["TradeId"]:
                    row.update({"Result": result, "ResultBarIdx": i, "ResultDate": data.index[i]})
                    break
        else:
            remaining.append(trade)
    active_trades[:] = remaining


def add_event(events: list[dict[str, Any]], data: pd.DataFrame, i: int, state: dict[str, Any], event: str, detail: str, *, price: float | None = None, level: float | None = None) -> None:
    events.append(
        {
            "BarIdx": i,
            "Date": data.index[i],
            "Event": event,
            "Detail": detail,
            "CRTNumber": state.get("active_number", np.nan),
            "CRTType": state.get("active_type", ""),
            "Direction": state.get("setup_direction", 0),
            "Price": np.nan if price is None else float(price),
            "Level": np.nan if level is None or not np.isfinite(level) else float(level),
            "State": state.get("setup_state", 0),
        }
    )


def crt_record(data: pd.DataFrame, state: dict[str, Any]) -> dict[str, Any]:
    return {
        "CRTNumber": state["active_number"],
        "CRTType": state["active_type"],
        "StartBarIdx": state["active_start_bar"],
        "ActivationBarIdx": state["active_activation_bar"],
        "StartDate": data.index[state["active_start_bar"]],
        "ActivationDate": data.index[state["active_activation_bar"]],
        "CRTHigh": state["active_high"],
        "CRTLow": state["active_low"],
        "CRTMid": state["active_mid"],
    }


def confirmed_pivots(i: int, high: np.ndarray, low: np.ndarray, *, left: int, right: int) -> tuple[float, float]:
    if i < left + right:
        return np.nan, np.nan
    pivot_idx = i - right
    start = pivot_idx - left
    end = pivot_idx + right + 1
    pivot_high = high[pivot_idx] if high[pivot_idx] >= finite_max(high[start:end]) else np.nan
    pivot_low = low[pivot_idx] if low[pivot_idx] <= finite_min(low[start:end]) else np.nan
    return pivot_high, pivot_low


def raw_fvg(i: int, high: np.ndarray, low: np.ndarray, atr: np.ndarray, config: CRTTBSConfig) -> tuple[bool, bool]:
    if i < 2:
        return False, False
    bullish = low[i] > high[i - 2] and low[i] - high[i - 2] >= atr[i] * config.minimum_fvg_atr
    bearish = high[i] < low[i - 2] and low[i - 2] - high[i] >= atr[i] * config.minimum_fvg_atr
    return bool(bullish), bool(bearish)


def index_in_timezone(index: pd.DatetimeIndex, config: CRTTBSConfig) -> pd.DatetimeIndex:
    idx = pd.DatetimeIndex(index)
    if idx.tz is None:
        idx = idx.tz_localize(ZoneInfo(config.data_timezone), nonexistent="shift_forward", ambiguous="NaT")
    return idx.tz_convert(ZoneInfo(config.timezone))


def in_session(ts: pd.Timestamp, session: str) -> bool:
    if pd.isna(ts):
        return False
    start_text, end_text = session.split("-")
    start = int(start_text[:2]) * 60 + int(start_text[2:])
    end = int(end_text[:2]) * 60 + int(end_text[2:])
    now = int(ts.hour) * 60 + int(ts.minute)
    if start <= end:
        return start <= now < end
    return now >= start or now < end


def date_key(ts: pd.Timestamp) -> int:
    if pd.isna(ts):
        return -1
    return int(ts.year) * 10000 + int(ts.month) * 100 + int(ts.day)


def safe_atr_value(value: float) -> float:
    return float(value) if np.isfinite(value) and value > 0 else 1e-12


def finite_min(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.min(arr)) if len(arr) else np.nan


def finite_max(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.max(arr)) if len(arr) else np.nan


def empty_entries() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "TradeId",
            "CRTNumber",
            "CRTType",
            "Signal",
            "Direction",
            "Model",
            "Component",
            "Potential",
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "StopPrice",
            "TargetPrice",
            "Result",
            "ResultBarIdx",
            "ResultDate",
        ]
    )


def plot_crt_tbs_state_machine(data: pd.DataFrame, result: dict[str, pd.DataFrame], *, title: str):
    if go is None:
        warnings.warn("plotly is not installed; skipping plot.")
        return None
    x = np.arange(len(data))
    labels = data.index.strftime("%Y-%m-%d %H:%M")
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=labels,
            hovertext=labels,
            hoverinfo="text+y",
        )
    )
    for _, row in result["crts"].iterrows():
        start = int(row["StartBarIdx"])
        end = min(len(data) - 1, int(row["ActivationBarIdx"]) + 80)
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTHigh"], y1=row["CRTHigh"], line=dict(color="#E5E7EB", width=1))
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTLow"], y1=row["CRTLow"], line=dict(color="#E5E7EB", width=1))
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTMid"], y1=row["CRTMid"], line=dict(color="#94A3B8", width=1, dash="dot"))
    event_colors = {"TBS": "#FACC15", "CHoCH": "#38BDF8", "BOS": "#22C55E", "BREAKOUT": "#F97316", "ENTRY": "#A78BFA", "INVALIDATED": "#EF4444", "EXPIRED": "#64748B", "FVG": "#CBD5E1", "CRT": "#E5E7EB"}
    if not result["events"].empty:
        for event_name, group in result["events"].groupby("Event"):
            prices = []
            for _, row in group.iterrows():
                if np.isfinite(row.get("Price", np.nan)):
                    prices.append(float(row["Price"]))
                elif np.isfinite(row.get("Level", np.nan)):
                    prices.append(float(row["Level"]))
                else:
                    prices.append(float(data["close"].iloc[int(row["BarIdx"])]))
            fig.add_trace(
                go.Scatter(
                    x=group["BarIdx"],
                    y=prices,
                    mode="markers+text",
                    name=event_name,
                    marker=dict(color=event_colors.get(event_name, "#E5E7EB"), size=9),
                    text=group["Event"],
                    textposition="top center",
                    hovertext=[
                        f"{row['Event']} {row['Detail']}<br>{row['Date']}<br>CRT #{row['CRTNumber']}<br>Level: {row['Level']}"
                        for _, row in group.iterrows()
                    ],
                    hoverinfo="text",
                )
            )
    if not result["entries"].empty:
        fig.add_trace(
            go.Scatter(
                x=result["entries"]["EntryBarIdx"],
                y=result["entries"]["EntryPrice"],
                mode="markers+text",
                name="Entries",
                marker=dict(color="#A78BFA", size=13, symbol="star"),
                text=result["entries"]["Signal"],
                textposition="bottom center",
                hovertext=[
                    (
                        f"{row['Signal']}<br>{row['Model']} / {row['Component']}<br>"
                        f"Entry: {row['EntryPrice']:.4f}<br>Stop: {row['StopPrice']:.4f}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>Result: {row['Result']}"
                    )
                    for _, row in result["entries"].iterrows()
                ],
                hoverinfo="text",
            )
        )
    ticks = np.linspace(0, len(data) - 1, min(12, len(data)), dtype=int)
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", y=1.02, x=1, xanchor="right"),
        margin=dict(l=55, r=40, t=80, b=60),
    )
    fig.update_xaxes(tickmode="array", tickvals=ticks, ticktext=labels[ticks], tickangle=-30)
    return fig


def display_figure(fig: Any) -> None:
    try:
        from IPython.display import display
    except ModuleNotFoundError:
        return
    display(fig)
```

```python
from __future__ import annotations

from dataclasses import dataclass
from pathlib import Path
from typing import Any
from zoneinfo import ZoneInfo
import warnings

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None

try:
    import vectorbt as vbt
    VBT_IMPORT_ERROR = None
except Exception as exc:
    vbt = None
    VBT_IMPORT_ERROR = exc


@dataclass
class CRTTBSConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    data_timezone: str = "Europe/Istanbul"
    timezone: str = "America/New_York"

    main_session: str = "0830-0945"
    secondary_session: str = "0945-1600"
    search_24x7: bool = True
    maximum_crts_per_day: int = 4
    secondary_cooldown_bars: int = 20
    setup_expiration_bars: int = 120

    atr_length: int = 14
    secondary_range_atr: float = 1.50
    secondary_body_percent: float = 65.0
    secondary_close_extreme: float = 0.80
    require_secondary_expansion: bool = True
    secondary_only_after_main: bool = True
    require_secondary_range_expansion: bool = True

    swing_left: int = 3
    swing_right: int = 3
    structure_lookback: int = 20
    sweep_ticks: float = 1.0
    tick_size: float = 0.01
    require_close_inside: bool = True
    bos_break_atr: float = 0.05
    max_bars_to_choch: int = 20
    invalidation_close_atr: float = 0.10

    enable_breakout_retest: bool = True
    breakout_body_atr: float = 0.70
    breakout_body_percent: float = 65.0
    breakout_close_atr: float = 0.10
    breakout_retest_bars: int = 20
    require_breakout_fvg: bool = False
    breakout_invalidation_atr: float = 0.15

    show_important_fvg: bool = True
    minimum_fvg_atr: float = 0.15
    displacement_body_atr: float = 0.70
    displacement_body_percent: float = 65.0
    maximum_important_fvg: int = 1

    show_order_block: bool = True
    order_block_lookback: int = 12
    minimum_choch_displacement_atr: float = 0.50

    entry_body_percent: float = 45.0
    entry_confirmation_bars: int = 15
    retest_tolerance_atr: float = 0.20
    accept_bos_retest: bool = True
    accept_fvg_retest: bool = True
    accept_order_block_retest: bool = True
    require_fvg_or_ob: bool = True
    require_close_beyond_bos: bool = True
    require_midpoint_alignment: bool = True
    require_ema_alignment: bool = False
    fast_ema_length: int = 20
    slow_ema_length: int = 50
    minimum_potential_score: int = 3

    enable_sweep_rejection_entry: bool = True
    sweep_rejection_bars: int = 4
    minimum_rejection_wick_percent: float = 35.0

    enable_high_potential_zones: bool = True
    important_zone_tolerance_atr: float = 0.15

    result_risk_reward: float = 1.5
    result_stop_buffer_atr: float = 0.10

    backtest_engine: str = "orders"
    backtest_price_column: str = "close"
    backtest_init_cash: float = 100_000.0
    backtest_size: float = 1.0
    backtest_size_type: str = "amount"
    backtest_fees: float = 0.0
    backtest_fixed_fees: float = 0.0
    backtest_slippage: float = 0.0
    backtest_shift_signals: bool = False
    backtest_upon_opposite_entry: str = "close"
    backtest_accumulate: bool = False
    backtest_freq: str | None = None
    backtest_raise_missing_vectorbt: bool = False


def run_crt_tbs(
    *,
    csv_path: str | None = None,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    show_plot: bool = True,
    run_backtest: bool = True,
    show_backtest_plot: bool = True,
    save_csv_path: str | None = None,
    **overrides: Any,
) -> dict[str, Any]:
    config = CRTTBSConfig(symbol=symbol, period=period, interval=interval, **overrides)
    data = load_ohlcv(csv_path=csv_path, config=config)
    data = add_indicators(data, config)
    result = run_pine_state_machine(data, config)
    backtest = run_vectorbt_backtest(
        data,
        result,
        config,
        show_plot=show_plot and show_backtest_plot,
    ) if run_backtest else empty_backtest_result(data, result, config)

    if save_csv_path is not None:
        Path(save_csv_path).parent.mkdir(parents=True, exist_ok=True)
        result["entries"].to_csv(save_csv_path, index=False)

    fig = plot_crt_tbs_state_machine(data, result, title=f"{config.symbol} CRT TBS Pine-style")
    if show_plot and fig is not None:
        display_figure(fig)

    print("CRT TBS Pine-style")
    print(f"bars: {len(data)} | CRTs: {len(result['crts'])} | events: {len(result['events'])} | entries: {len(result['entries'])}")
    if len(result["entries"]):
        preview = [
            "Signal",
            "Model",
            "Component",
            "Potential",
            "EntryDate",
            "EntryPrice",
            "StopPrice",
            "TargetPrice",
            "Result",
        ]
        print(result["entries"][preview].head(30).to_string(index=False))
    if run_backtest:
        print("VectorBT backtest")
        print(
            f"engine: {config.backtest_engine} | "
            f"long entries: {int(backtest['signals']['long_entries'].sum())} | "
            f"short entries: {int(backtest['signals']['short_entries'].sum())} | "
            f"orders: {len(backtest['orders'])}"
        )
        if backtest["portfolio"] is None:
            print("portfolio: vectorbt is not installed in this Python environment")
        elif len(backtest["stats"]):
            preferred = [
                "Start",
                "End",
                "Period",
                "Start Value",
                "End Value",
                "Total Return [%]",
                "Benchmark Return [%]",
                "Max Drawdown [%]",
                "Total Trades",
                "Win Rate [%]",
                "Profit Factor",
                "Sharpe Ratio",
            ]
            stats = backtest["stats"]
            available = [name for name in preferred if name in stats.index]
            print(stats.loc[available].to_string())

    return {
        "config": config,
        "data": data,
        "fig": fig,
        **result,
        "setups": result["entries"],
        "backtest": backtest,
        "backtest_signals": backtest["signals"],
        "backtest_orders": backtest["orders"],
        "portfolio": backtest["portfolio"],
        "portfolio_stats": backtest["stats"],
        "backtest_fig": backtest["fig"],
    }


def load_ohlcv(*, csv_path: str | None, config: CRTTBSConfig) -> pd.DataFrame:
    if csv_path is not None:
        return prepare_ohlcv(pd.read_csv(csv_path))
    if Ticker is None:
        raise ImportError("Install yahooquery or pass csv_path.")

    raw = Ticker(config.symbol).history(period=config.period, interval=config.interval)
    if isinstance(raw, pd.DataFrame) and isinstance(raw.index, pd.MultiIndex):
        if "symbol" in raw.index.names:
            raw = raw.reset_index(level="symbol", drop=True)
        else:
            raw = raw.reset_index()
    return prepare_ohlcv(raw)


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    data = df.copy()
    data = data.rename(columns={column: str(column).lower() for column in data.columns})
    if not isinstance(data.index, pd.DatetimeIndex):
        date_col = next(
            (col for col in ["datetime", "date", "timestamp", "time"] if col in data.columns),
            None,
        )
        if date_col is None:
            raise ValueError("Use DatetimeIndex or include datetime/date/timestamp column.")
        data[date_col] = pd.to_datetime(data[date_col], errors="coerce")
        data = data.set_index(date_col)

    required = {"open", "high", "low", "close"}
    missing = required.difference(data.columns)
    if missing:
        raise ValueError(f"Missing OHLC columns: {sorted(missing)}")

    data.index = pd.to_datetime(data.index)
    data = data[~data.index.duplicated(keep="last")].sort_index()
    if "volume" not in data.columns:
        data["volume"] = 0.0
    for col in ["open", "high", "low", "close", "volume"]:
        data[col] = pd.to_numeric(data[col], errors="coerce")
    data = data.dropna(subset=["open", "high", "low", "close"])
    data["volume"] = data["volume"].fillna(0.0).clip(lower=0.0)
    return data[["open", "high", "low", "close", "volume"]]


def add_indicators(data: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    out = data.copy()
    high = out["high"].astype(float)
    low = out["low"].astype(float)
    open_ = out["open"].astype(float)
    close = out["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [high - low, (high - previous_close).abs(), (low - previous_close).abs()],
        axis=1,
    ).max(axis=1)
    out["atr"] = true_range.rolling(config.atr_length, min_periods=1).mean()
    out["candle_range"] = high - low
    out["candle_body"] = (close - open_).abs()
    out["body_ratio"] = out["candle_body"] / out["candle_range"].replace(0, np.nan)
    out["body_ratio"] = out["body_ratio"].fillna(0.0)
    out["close_location"] = (close - low) / out["candle_range"].replace(0, np.nan)
    out["close_location"] = out["close_location"].fillna(0.5)
    out["range_to_atr"] = out["candle_range"] / out["atr"].replace(0, np.nan)
    out["fast_ema"] = close.ewm(span=config.fast_ema_length, adjust=False, min_periods=1).mean()
    out["slow_ema"] = close.ewm(span=config.slow_ema_length, adjust=False, min_periods=1).mean()
    return out.replace([np.inf, -np.inf], np.nan)


def run_pine_state_machine(data: pd.DataFrame, config: CRTTBSConfig) -> dict[str, pd.DataFrame]:
    o = data["open"].to_numpy(dtype=float)
    h = data["high"].to_numpy(dtype=float)
    l = data["low"].to_numpy(dtype=float)
    c = data["close"].to_numpy(dtype=float)
    atr = data["atr"].to_numpy(dtype=float)
    body = data["candle_body"].to_numpy(dtype=float)
    body_ratio = data["body_ratio"].to_numpy(dtype=float)
    candle_range = data["candle_range"].to_numpy(dtype=float)
    close_location = data["close_location"].to_numpy(dtype=float)
    range_to_atr = data["range_to_atr"].to_numpy(dtype=float)
    fast_ema = data["fast_ema"].to_numpy(dtype=float)
    slow_ema = data["slow_ema"].to_numpy(dtype=float)

    tz_index = index_in_timezone(data.index, config)
    inside_main = np.array([in_session(ts, config.main_session) for ts in tz_index], dtype=bool)
    inside_secondary = np.array(
        [config.search_24x7 or in_session(ts, config.secondary_session) for ts in tz_index],
        dtype=bool,
    )
    date_keys = np.array([date_key(ts) for ts in tz_index], dtype=int)

    state = new_state()
    crts: list[dict[str, Any]] = []
    events: list[dict[str, Any]] = []
    entries: list[dict[str, Any]] = []
    active_trades: list[dict[str, Any]] = []

    best_main_range = np.nan
    best_main_high = np.nan
    best_main_low = np.nan
    best_main_bar = None
    main_crt_created_today = False
    crts_today = 0
    last_crt_creation_bar = None
    last_swing_high = np.nan
    last_swing_low = np.nan
    next_trade_id = 1

    for i in range(len(data)):
        new_trading_day = i > 0 and date_keys[i] != date_keys[i - 1]
        main_session_started = bool(inside_main[i] and (i == 0 or not inside_main[i - 1]))
        main_session_finished = bool((not inside_main[i]) and i > 0 and inside_main[i - 1])

        if new_trading_day:
            state = new_state()
            main_crt_created_today = False
            crts_today = 0
            last_crt_creation_bar = None
            best_main_range = np.nan
            best_main_high = np.nan
            best_main_low = np.nan
            best_main_bar = None

        if new_trading_day or main_session_started:
            best_main_range = np.nan
            best_main_high = np.nan
            best_main_low = np.nan
            best_main_bar = None

        if inside_main[i] and (not np.isfinite(best_main_range) or candle_range[i] > best_main_range):
            best_main_range = candle_range[i]
            best_main_high = h[i]
            best_main_low = l[i]
            best_main_bar = i

        pivot_high, pivot_low = confirmed_pivots(
            i=i,
            high=h,
            low=l,
            left=config.swing_left,
            right=config.swing_right,
        )
        if np.isfinite(pivot_high):
            last_swing_high = pivot_high
        if np.isfinite(pivot_low):
            last_swing_low = pivot_low

        create_main_crt = (
            main_session_finished
            and not main_crt_created_today
            and np.isfinite(best_main_high)
            and np.isfinite(best_main_low)
            and best_main_bar is not None
        )
        if create_main_crt:
            main_crt_created_today = True
            crts_today += 1
            state = activate_crt(
                crt_high=float(best_main_high),
                crt_low=float(best_main_low),
                crt_start_bar=int(best_main_bar),
                activation_bar=i,
                crt_type="PRINCIPAL",
                crt_number=crts_today,
            )
            last_crt_creation_bar = i
            crts.append(crt_record(data, state))
            add_event(events, data, i, state, "CRT", "PRINCIPAL", price=state["active_mid"])

        if state["setup_state"] >= 1 and state["setup_state"] <= 4:
            if i - state["active_activation_bar"] > config.setup_expiration_bars:
                add_event(events, data, i, state, "EXPIRED", "setup_expiration")
                state["setup_state"] = 6
                state["status"] = "EXPIRADO"

        raw_bull_fvg, raw_bear_fvg = raw_fvg(i, h, l, atr, config)
        update_important_fvg(i, o, h, l, c, atr, body, body_ratio, raw_bull_fvg, raw_bear_fvg, state, config, events, data)

        bullish_sweep, bearish_sweep = process_tbs_sweep(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            state=state,
            last_swing_high=last_swing_high,
            last_swing_low=last_swing_low,
            config=config,
            events=events,
            data=data,
        )

        process_real_breakout(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body=body,
            body_ratio=body_ratio,
            raw_bull_fvg=raw_bull_fvg,
            raw_bear_fvg=raw_bear_fvg,
            bullish_sweep=bullish_sweep,
            bearish_sweep=bearish_sweep,
            state=state,
            config=config,
            events=events,
            data=data,
        )

        invalidate_before_choch(i, c, atr, state, config, events, data)
        process_choch(i, o, h, l, c, atr, body, state, config, events, data)
        process_bos_level(i, o, h, l, c, state)
        process_bos_confirmation(i, c, atr, state, config, events, data)
        invalidate_order_block(i, c, state, events, data)

        confirmation = evaluate_entries(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body_ratio=body_ratio,
            fast_ema=fast_ema,
            slow_ema=slow_ema,
            state=state,
            config=config,
        )
        if confirmation is not None:
            state["setup_state"] = 5
            add_event(events, data, i, state, "ENTRY", confirmation["model"], price=float(c[i]))
            entry = entry_record(data, i, state, confirmation, c, atr, config, next_trade_id)
            next_trade_id += 1
            entries.append(entry)
            active_trades.append(entry.copy())

        update_trades(i, h, l, data, active_trades, entries)

        if state["setup_state"] == 4 and i - state.get("bos_bar", i) > config.entry_confirmation_bars:
            add_event(events, data, i, state, "EXPIRED", "entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"
        if state["setup_state"] == 8 and i - state.get("breakout_bar", i) > config.breakout_retest_bars:
            add_event(events, data, i, state, "EXPIRED", "breakout_entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"

        if create_secondary_crt_allowed(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            candle_range=candle_range,
            body_ratio=body_ratio,
            close_location=close_location,
            range_to_atr=range_to_atr,
            inside_secondary=inside_secondary,
            inside_main=inside_main,
            main_session_finished=main_session_finished,
            state=state,
            crts_today=crts_today,
            main_crt_created_today=main_crt_created_today,
            last_crt_creation_bar=last_crt_creation_bar,
            config=config,
        ):
            crts_today += 1
            state = activate_crt(
                crt_high=float(h[i]),
                crt_low=float(l[i]),
                crt_start_bar=i,
                activation_bar=i,
                crt_type="SECUNDARIO",
                crt_number=crts_today,
            )
            last_crt_creation_bar = i
            crts.append(crt_record(data, state))
            add_event(events, data, i, state, "CRT", "SECUNDARIO", price=state["active_mid"])

    return {
        "crts": pd.DataFrame(crts),
        "events": pd.DataFrame(events),
        "entries": pd.DataFrame(entries) if entries else empty_entries(),
        "trades": pd.DataFrame(entries) if entries else empty_entries(),
    }


def new_state() -> dict[str, Any]:
    return {
        "setup_state": 0,
        "setup_direction": 0,
        "active_high": np.nan,
        "active_low": np.nan,
        "active_mid": np.nan,
        "active_start_bar": None,
        "active_activation_bar": None,
        "active_type": "",
        "active_number": np.nan,
        "sweep_bar": None,
        "low_swept": False,
        "high_swept": False,
        "choch_level": np.nan,
        "choch_bar": None,
        "impulse_extreme": np.nan,
        "bos_level": np.nan,
        "bos_bar": None,
        "pullback_detected": False,
        "first_retest_consumed": False,
        "breakout_level": np.nan,
        "breakout_bar": None,
        "breakout_had_fvg": False,
        "breakout_retest_consumed": False,
        "important_fvg_count": 0,
        "bullish_fvg_top": np.nan,
        "bullish_fvg_bottom": np.nan,
        "bullish_fvg_creation_bar": None,
        "bearish_fvg_top": np.nan,
        "bearish_fvg_bottom": np.nan,
        "bearish_fvg_creation_bar": None,
        "bullish_ob_high": np.nan,
        "bullish_ob_low": np.nan,
        "bullish_ob_creation_bar": None,
        "bearish_ob_high": np.nan,
        "bearish_ob_low": np.nan,
        "bearish_ob_creation_bar": None,
        "bullish_ob_created": False,
        "bearish_ob_created": False,
        "status": "Esperando",
    }


def activate_crt(*, crt_high: float, crt_low: float, crt_start_bar: int, activation_bar: int, crt_type: str, crt_number: int) -> dict[str, Any]:
    state = new_state()
    state.update(
        {
            "setup_state": 1,
            "active_high": float(crt_high),
            "active_low": float(crt_low),
            "active_mid": (float(crt_high) + float(crt_low)) / 2.0,
            "active_start_bar": int(crt_start_bar),
            "active_activation_bar": int(activation_bar),
            "active_type": crt_type,
            "active_number": int(crt_number),
            "status": "Esperando TBS",
        }
    )
    return state


def process_tbs_sweep(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    state: dict[str, Any],
    last_swing_high: float,
    last_swing_low: float,
    config: CRTTBSConfig,
    events: list[dict[str, Any]],
    data: pd.DataFrame,
) -> tuple[bool, bool]:
    if state["setup_state"] != 1 or i <= state["active_activation_bar"]:
        return False, False
    minimum_sweep = float(config.sweep_ticks) * float(config.tick_size)
    low_sweep_raw = l[i] <= state["active_low"] - minimum_sweep
    high_sweep_raw = h[i] >= state["active_high"] + minimum_sweep
    if low_sweep_raw:
        state["low_swept"] = True
    if high_sweep_raw:
        state["high_swept"] = True

    bullish_sweep = low_sweep_raw and not high_sweep_raw and (not config.require_close_inside or c[i] > state["active_low"])
    bearish_sweep = high_sweep_raw and not low_sweep_raw and (not config.require_close_inside or c[i] < state["active_high"])
    if bullish_sweep:
        fallback_high = finite_max(h[max(0, i - config.structure_lookback) : i])
        state["setup_state"] = 2
        state["setup_direction"] = 1
        state["sweep_bar"] = i
        state["choch_level"] = last_swing_high if np.isfinite(last_swing_high) else fallback_high
        state["status"] = "Esperando CHoCH"
        add_event(events, data, i, state, "TBS", "bullish_sweep", price=float(l[i]), level=state["choch_level"])
    elif bearish_sweep:
        fallback_low = finite_min(l[max(0, i - config.structure_lookback) : i])
        state["setup_state"] = 2
        state["setup_direction"] = -1
        state["sweep_bar"] = i
        state["choch_level"] = last_swing_low if np.isfinite(last_swing_low) else fallback_low
        state["status"] = "Esperando CHoCH"
        add_event(events, data, i, state, "TBS", "bearish_sweep", price=float(h[i]), level=state["choch_level"])
    return bool(bullish_sweep), bool(bearish_sweep)


def process_real_breakout(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    atr: np.ndarray,
    body: np.ndarray,
    body_ratio: np.ndarray,
    raw_bull_fvg: bool,
    raw_bear_fvg: bool,
    bullish_sweep: bool,
    bearish_sweep: bool,
    state: dict[str, Any],
    config: CRTTBSConfig,
    events: list[dict[str, Any]],
    data: pd.DataFrame,
) -> None:
    if not config.enable_breakout_retest or state["setup_state"] != 1 or i <= state["active_activation_bar"]:
        return
    breakout_margin = atr[i] * config.breakout_close_atr
    strong_body = body[i] >= atr[i] * config.breakout_body_atr and body_ratio[i] * 100.0 >= config.breakout_body_percent
    bullish_breakout = c[i] > state["active_high"] + breakout_margin and strong_body and not bullish_sweep and not bearish_sweep
    bearish_breakout = c[i] < state["active_low"] - breakout_margin and strong_body and not bullish_sweep and not bearish_sweep
    if bullish_breakout:
        state["setup_state"] = 8
        state["setup_direction"] = 1
        state["breakout_level"] = state["active_high"]
        state["breakout_bar"] = i
        state["breakout_had_fvg"] = bool(raw_bull_fvg or (i > 0 and raw_fvg(i - 1, h, l, atr, config)[0]))
        state["breakout_retest_consumed"] = False
        state["status"] = "Breakout: esperando retest"
        add_event(events, data, i, state, "BREAKOUT", "bullish_breakout", price=float(c[i]), level=state["breakout_level"])
    elif bearish_breakout:
        state["setup_state"] = 8
        state["setup_direction"] = -1
        state["breakout_level"] = state["active_low"]
        state["breakout_bar"] = i
        state["breakout_had_fvg"] = bool(raw_bear_fvg or (i > 0 and raw_fvg(i - 1, h, l, atr, config)[1]))
        state["breakout_retest_consumed"] = False
        state["status"] = "Breakout: esperando retest"
        add_event(events, data, i, state, "BREAKOUT", "bearish_breakout", price=float(c[i]), level=state["breakout_level"])


def invalidate_before_choch(i: int, c: np.ndarray, atr: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 2:
        return
    both_sides = state["low_swept"] and state["high_swept"]
    time_expired = state["sweep_bar"] is not None and i - state["sweep_bar"] > config.max_bars_to_choch
    bullish_opposite = state["setup_direction"] == 1 and c[i] < state["active_low"] - atr[i] * config.invalidation_close_atr
    bearish_opposite = state["setup_direction"] == -1 and c[i] > state["active_high"] + atr[i] * config.invalidation_close_atr
    if both_sides or time_expired or bullish_opposite or bearish_opposite:
        state["setup_state"] = 7
        state["status"] = "INVALIDADO"
        add_event(events, data, i, state, "INVALIDATED", "before_choch")


def process_choch(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, atr: np.ndarray, body: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 2 or state["sweep_bar"] is None or i <= state["sweep_bar"] or not np.isfinite(state["choch_level"]):
        return
    current_body_atr = body[i] / max(atr[i], 1e-12)
    bullish = state["setup_direction"] == 1 and c[i] > state["choch_level"] and current_body_atr >= config.minimum_choch_displacement_atr
    bearish = state["setup_direction"] == -1 and c[i] < state["choch_level"] and current_body_atr >= config.minimum_choch_displacement_atr
    if bullish or bearish:
        state["setup_state"] = 3
        state["choch_bar"] = i
        state["impulse_extreme"] = h[i] if bullish else l[i]
        state["pullback_detected"] = False
        state["status"] = "Esperando retroceso"
        create_order_block(i, o, h, l, c, state, config)
        add_event(events, data, i, state, "CHoCH", "bullish" if bullish else "bearish", price=float(c[i]), level=state["choch_level"])


def process_bos_level(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any]) -> None:
    if state["setup_state"] != 3:
        return
    if state["setup_direction"] == 1 and not state["pullback_detected"]:
        state["impulse_extreme"] = max(state["impulse_extreme"], h[i]) if np.isfinite(state["impulse_extreme"]) else h[i]
        bullish_pullback = state["choch_bar"] is not None and i > state["choch_bar"] and (c[i] < o[i] or l[i] < l[i - 1])
        if bullish_pullback:
            state["pullback_detected"] = True
            state["bos_level"] = state["impulse_extreme"]
            state["status"] = "Esperando BOS"
    elif state["setup_direction"] == -1 and not state["pullback_detected"]:
        state["impulse_extreme"] = min(state["impulse_extreme"], l[i]) if np.isfinite(state["impulse_extreme"]) else l[i]
        bearish_pullback = state["choch_bar"] is not None and i > state["choch_bar"] and (c[i] > o[i] or h[i] > h[i - 1])
        if bearish_pullback:
            state["pullback_detected"] = True
            state["bos_level"] = state["impulse_extreme"]
            state["status"] = "Esperando BOS"


def process_bos_confirmation(i: int, c: np.ndarray, atr: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 3 or not state["pullback_detected"] or not np.isfinite(state["bos_level"]):
        return
    distance = atr[i] * config.bos_break_atr
    bullish = state["setup_direction"] == 1 and c[i] > state["bos_level"] + distance
    bearish = state["setup_direction"] == -1 and c[i] < state["bos_level"] - distance
    if bullish or bearish:
        state["setup_state"] = 4
        state["bos_bar"] = i
        state["first_retest_consumed"] = False
        state["status"] = "Esperando entrada"
        add_event(events, data, i, state, "BOS", "bullish" if bullish else "bearish", price=float(c[i]), level=state["bos_level"])


def evaluate_entries(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    atr: np.ndarray,
    body_ratio: np.ndarray,
    fast_ema: np.ndarray,
    slow_ema: np.ndarray,
    state: dict[str, Any],
    config: CRTTBSConfig,
) -> dict[str, Any] | None:
    current_atr = safe_atr_value(atr[i])
    (
        bullish_rejection_candle,
        bearish_rejection_candle,
        bullish_rejection,
        bearish_rejection,
        bullish_rejection_score,
        bearish_rejection_score,
    ) = rejection_signals(i, o, h, l, c, body_ratio, fast_ema, slow_ema, state, config)
    tbs = tbs_entry_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config)
    breakout = breakout_retest_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config)
    zone = zone_confluence_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config, bullish_rejection_candle, bearish_rejection_candle)

    if tbs is not None:
        return tbs
    if breakout is not None:
        return breakout
    if bullish_rejection and state["setup_direction"] == 1:
        return {"direction": 1, "signal": "BUY", "component": "BARRIDA + RECHAZO", "model": "TBS REJECTION", "potential": bullish_rejection_score}
    if bearish_rejection and state["setup_direction"] == -1:
        return {"direction": -1, "signal": "SELL", "component": "BARRIDA + RECHAZO", "model": "TBS REJECTION", "potential": bearish_rejection_score}
    if zone is not None:
        return zone
    return None


def tbs_entry_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    valid = state["setup_state"] == 4 and state["bos_bar"] is not None and i > state["bos_bar"] and i - state["bos_bar"] <= config.entry_confirmation_bars and not state["first_retest_consumed"]
    if not valid:
        return None
    tolerance = current_atr * config.retest_tolerance_atr
    bullish_bos_retest = state["setup_direction"] == 1 and config.accept_bos_retest and l[i] <= state["bos_level"] + tolerance and h[i] >= state["bos_level"] - tolerance
    bearish_bos_retest = state["setup_direction"] == -1 and config.accept_bos_retest and h[i] >= state["bos_level"] - tolerance and l[i] <= state["bos_level"] + tolerance
    bullish_fvg_retest = state["setup_direction"] == 1 and config.accept_fvg_retest and state["bullish_fvg_creation_bar"] is not None and state["bullish_fvg_creation_bar"] > state["sweep_bar"] and l[i] <= state["bullish_fvg_top"] + tolerance and h[i] >= state["bullish_fvg_bottom"] - tolerance
    bearish_fvg_retest = state["setup_direction"] == -1 and config.accept_fvg_retest and state["bearish_fvg_creation_bar"] is not None and state["bearish_fvg_creation_bar"] > state["sweep_bar"] and h[i] >= state["bearish_fvg_bottom"] - tolerance and l[i] <= state["bearish_fvg_top"] + tolerance
    bullish_ob_retest = state["setup_direction"] == 1 and config.accept_order_block_retest and state["bullish_ob_created"] and state["bullish_ob_creation_bar"] >= state["sweep_bar"] and l[i] <= state["bullish_ob_high"] + tolerance and h[i] >= state["bullish_ob_low"] - tolerance and c[i] >= state["bullish_ob_low"]
    bearish_ob_retest = state["setup_direction"] == -1 and config.accept_order_block_retest and state["bearish_ob_created"] and state["bearish_ob_creation_bar"] >= state["sweep_bar"] and h[i] >= state["bearish_ob_low"] - tolerance and l[i] <= state["bearish_ob_high"] + tolerance and c[i] <= state["bearish_ob_high"]

    bullish_primary = bullish_ob_retest or bullish_fvg_retest
    bearish_primary = bearish_ob_retest or bearish_fvg_retest
    bullish_zone_accepted = bullish_primary if config.require_fvg_or_ob else (bullish_primary or bullish_bos_retest)
    bearish_zone_accepted = bearish_primary if config.require_fvg_or_ob else (bearish_primary or bearish_bos_retest)
    bullish_confirm = c[i] > o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] > c[i - 1]
    bearish_confirm = c[i] < o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] < c[i - 1]
    bullish_bos_close = (not config.require_close_beyond_bos) or c[i] > state["bos_level"]
    bearish_bos_close = (not config.require_close_beyond_bos) or c[i] < state["bos_level"]
    bullish_mid = (not config.require_midpoint_alignment) or c[i] > state["active_mid"]
    bearish_mid = (not config.require_midpoint_alignment) or c[i] < state["active_mid"]
    bullish_ema = (not config.require_ema_alignment) or (fast_ema[i] > slow_ema[i] and c[i] > fast_ema[i])
    bearish_ema = (not config.require_ema_alignment) or (fast_ema[i] < slow_ema[i] and c[i] < fast_ema[i])
    bullish_score = int(bullish_ob_retest) + int(bullish_fvg_retest) + int(bullish_bos_close) + int(bullish_mid) + int(bullish_ema)
    bearish_score = int(bearish_ob_retest) + int(bearish_fvg_retest) + int(bearish_bos_close) + int(bearish_mid) + int(bearish_ema)
    if bullish_zone_accepted or bearish_zone_accepted:
        state["first_retest_consumed"] = True

    if state["setup_direction"] == 1 and bullish_zone_accepted and bullish_confirm and bullish_bos_close and bullish_mid and bullish_ema and bullish_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if bullish_ob_retest else ("FVG" if bullish_fvg_retest else "BOS")
        return {"direction": 1, "signal": "BUY", "component": component, "model": "TBS REVERSIÓN", "potential": bullish_score}
    if state["setup_direction"] == -1 and bearish_zone_accepted and bearish_confirm and bearish_bos_close and bearish_mid and bearish_ema and bearish_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if bearish_ob_retest else ("FVG" if bearish_fvg_retest else "BOS")
        return {"direction": -1, "signal": "SELL", "component": component, "model": "TBS REVERSIÓN", "potential": bearish_score}
    return None


def breakout_retest_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    valid = state["setup_state"] == 8 and state["breakout_bar"] is not None and i > state["breakout_bar"] and i - state["breakout_bar"] <= config.breakout_retest_bars and not state["breakout_retest_consumed"]
    if not valid:
        return None
    tolerance = current_atr * config.retest_tolerance_atr
    bullish_touch = state["setup_direction"] == 1 and l[i] <= state["breakout_level"] + tolerance and h[i] >= state["breakout_level"] - tolerance
    bearish_touch = state["setup_direction"] == -1 and h[i] >= state["breakout_level"] - tolerance and l[i] <= state["breakout_level"] + tolerance
    bullish_confirm = bullish_touch and c[i] > state["breakout_level"] and c[i] > o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] > c[i - 1]
    bearish_confirm = bearish_touch and c[i] < state["breakout_level"] and c[i] < o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] < c[i - 1]
    bullish_ema = (not config.require_ema_alignment) or (fast_ema[i] > slow_ema[i] and c[i] > fast_ema[i])
    bearish_ema = (not config.require_ema_alignment) or (fast_ema[i] < slow_ema[i] and c[i] < fast_ema[i])
    if bullish_touch or bearish_touch:
        state["breakout_retest_consumed"] = True
    if state["setup_direction"] == 1 and c[i] < state["active_high"] - current_atr * config.breakout_invalidation_atr:
        state["setup_state"] = 7
        return None
    if state["setup_direction"] == -1 and c[i] > state["active_low"] + current_atr * config.breakout_invalidation_atr:
        state["setup_state"] = 7
        return None
    bullish_score = int(state["breakout_had_fvg"]) + int(c[i] > state["active_high"]) + int(bullish_ema) + int(body_ratio[i] * 100.0 >= config.entry_body_percent) + 1
    bearish_score = int(state["breakout_had_fvg"]) + int(c[i] < state["active_low"]) + int(bearish_ema) + int(body_ratio[i] * 100.0 >= config.entry_body_percent) + 1
    if state["setup_direction"] == 1 and bullish_confirm and bullish_ema and (not config.require_breakout_fvg or state["breakout_had_fvg"]) and bullish_score >= config.minimum_potential_score:
        return {"direction": 1, "signal": "BUY", "component": "CRT HIGH RETEST", "model": "BREAKOUT RETEST", "potential": bullish_score}
    if state["setup_direction"] == -1 and bearish_confirm and bearish_ema and (not config.require_breakout_fvg or state["breakout_had_fvg"]) and bearish_score >= config.minimum_potential_score:
        return {"direction": -1, "signal": "SELL", "component": "CRT LOW RETEST", "model": "BREAKOUT RETEST", "potential": bearish_score}
    return None


def rejection_signals(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> tuple[bool, bool, bool, bool, int, int]:
    candle_range = max(h[i] - l[i], 1e-12)
    lower_wick_pct = (min(o[i], c[i]) - l[i]) / candle_range * 100.0
    upper_wick_pct = (h[i] - max(o[i], c[i])) / candle_range * 100.0
    bullish_candle = c[i] > o[i] and lower_wick_pct >= config.minimum_rejection_wick_percent and c[i] > state.get("active_low", np.nan) and c[i] > c[i - 1]
    bearish_candle = c[i] < o[i] and upper_wick_pct >= config.minimum_rejection_wick_percent and c[i] < state.get("active_high", np.nan) and c[i] < c[i - 1]
    valid_bullish = config.enable_sweep_rejection_entry and state["setup_state"] == 2 and state["setup_direction"] == 1 and state["sweep_bar"] is not None and i >= state["sweep_bar"] and i - state["sweep_bar"] <= config.sweep_rejection_bars
    valid_bearish = config.enable_sweep_rejection_entry and state["setup_state"] == 2 and state["setup_direction"] == -1 and state["sweep_bar"] is not None and i >= state["sweep_bar"] and i - state["sweep_bar"] <= config.sweep_rejection_bars
    bullish_score = int(c[i] > state.get("active_low", np.nan)) + int(lower_wick_pct >= config.minimum_rejection_wick_percent) + int(c[i] > o[i]) + int(c[i] > c[i - 1]) + int((not config.require_ema_alignment) or fast_ema[i] > slow_ema[i])
    bearish_score = int(c[i] < state.get("active_high", np.nan)) + int(upper_wick_pct >= config.minimum_rejection_wick_percent) + int(c[i] < o[i]) + int(c[i] < c[i - 1]) + int((not config.require_ema_alignment) or fast_ema[i] < slow_ema[i])
    confirmed_bullish = valid_bullish and bullish_candle and bullish_score >= config.minimum_potential_score
    confirmed_bearish = valid_bearish and bearish_candle and bearish_score >= config.minimum_potential_score
    return bullish_candle, bearish_candle, confirmed_bullish, confirmed_bearish, bullish_score, bearish_score


def zone_confluence_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig, bullish_rejection_candle: bool, bearish_rejection_candle: bool) -> dict[str, Any] | None:
    if not config.enable_high_potential_zones or state["setup_state"] not in {1, 2, 3, 4}:
        return None
    tolerance = current_atr * config.important_zone_tolerance_atr
    price_at_bull_ob = state["bullish_ob_created"] and l[i] <= state["bullish_ob_high"] + tolerance and h[i] >= state["bullish_ob_low"] - tolerance and c[i] >= state["bullish_ob_low"]
    price_at_bear_ob = state["bearish_ob_created"] and h[i] >= state["bearish_ob_low"] - tolerance and l[i] <= state["bearish_ob_high"] + tolerance and c[i] <= state["bearish_ob_high"]
    price_at_bull_fvg = state["bullish_fvg_creation_bar"] is not None and l[i] <= state["bullish_fvg_top"] + tolerance and h[i] >= state["bullish_fvg_bottom"] - tolerance
    price_at_bear_fvg = state["bearish_fvg_creation_bar"] is not None and h[i] >= state["bearish_fvg_bottom"] - tolerance and l[i] <= state["bearish_fvg_top"] + tolerance
    price_at_low_liq = np.isfinite(state["active_low"]) and l[i] <= state["active_low"] + tolerance and c[i] > state["active_low"]
    price_at_high_liq = np.isfinite(state["active_high"]) and h[i] >= state["active_high"] - tolerance and c[i] < state["active_high"]
    bull_score = int(price_at_bull_ob) + int(price_at_bull_fvg) + int(price_at_low_liq) + int(bullish_rejection_candle) + int((not config.require_ema_alignment) or fast_ema[i] > slow_ema[i])
    bear_score = int(price_at_bear_ob) + int(price_at_bear_fvg) + int(price_at_high_liq) + int(bearish_rejection_candle) + int((not config.require_ema_alignment) or fast_ema[i] < slow_ema[i])
    if (price_at_bull_ob or price_at_bull_fvg or price_at_low_liq) and bullish_rejection_candle and bull_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if price_at_bull_ob else ("FVG" if price_at_bull_fvg else "LIQUIDEZ CRT")
        return {"direction": 1, "signal": "BUY", "component": component, "model": "ZONE CONFLUENCE", "potential": bull_score}
    if (price_at_bear_ob or price_at_bear_fvg or price_at_high_liq) and bearish_rejection_candle and bear_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if price_at_bear_ob else ("FVG" if price_at_bear_fvg else "LIQUIDEZ CRT")
        return {"direction": -1, "signal": "SELL", "component": component, "model": "ZONE CONFLUENCE", "potential": bear_score}
    return None


def invalidate_order_block(i: int, c: np.ndarray, state: dict[str, Any], events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] < 3 or state["setup_state"] > 4:
        return
    bullish_broken = (
        state["setup_direction"] == 1
        and state["bullish_ob_created"]
        and np.isfinite(state["bullish_ob_low"])
        and c[i] < state["bullish_ob_low"]
    )
    bearish_broken = (
        state["setup_direction"] == -1
        and state["bearish_ob_created"]
        and np.isfinite(state["bearish_ob_high"])
        and c[i] > state["bearish_ob_high"]
    )
    if bullish_broken or bearish_broken:
        state["setup_state"] = 7
        state["status"] = "INVALIDADO"
        add_event(events, data, i, state, "INVALIDATED", "order_block_broken")


def create_order_block(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> None:
    if not config.show_order_block or state["sweep_bar"] is None:
        return
    max_search = min(config.order_block_lookback, i - state["sweep_bar"])
    for offset in range(1, max_search + 1):
        idx = i - offset
        if state["setup_direction"] == 1 and c[idx] < o[idx]:
            state["bullish_ob_high"] = h[idx]
            state["bullish_ob_low"] = l[idx]
            state["bullish_ob_creation_bar"] = idx
            state["bullish_ob_created"] = True
            return
        if state["setup_direction"] == -1 and c[idx] > o[idx]:
            state["bearish_ob_high"] = h[idx]
            state["bearish_ob_low"] = l[idx]
            state["bearish_ob_creation_bar"] = idx
            state["bearish_ob_created"] = True
            return


def update_important_fvg(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, atr: np.ndarray, body: np.ndarray, body_ratio: np.ndarray, raw_bull: bool, raw_bear: bool, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if i < 2 or not config.show_important_fvg or state["important_fvg_count"] >= config.maximum_important_fvg:
        return
    center_body = abs(c[i - 1] - o[i - 1])
    center_range = max(h[i - 1] - l[i - 1], 1e-12)
    center_body_ratio = center_body / center_range
    displacement = center_body >= atr[i - 1] * config.displacement_body_atr and center_body_ratio * 100.0 >= config.displacement_body_percent
    state_ok = state["setup_state"] in {2, 3, 4, 8}
    impulse_started = (state["sweep_bar"] is not None and i > state["sweep_bar"]) or (state["breakout_bar"] is not None and i >= state["breakout_bar"])
    if state["setup_direction"] == 1 and state_ok and impulse_started and raw_bull and displacement:
        mid = (l[i] + h[i - 2]) / 2.0
        if abs(c[i] - mid) <= atr[i] * 2.0:
            state["bullish_fvg_top"] = l[i]
            state["bullish_fvg_bottom"] = h[i - 2]
            state["bullish_fvg_creation_bar"] = i
            state["important_fvg_count"] += 1
            add_event(events, data, i, state, "FVG", "bullish", price=mid)
    if state["setup_direction"] == -1 and state_ok and impulse_started and raw_bear and displacement:
        mid = (l[i - 2] + h[i]) / 2.0
        if abs(c[i] - mid) <= atr[i] * 2.0:
            state["bearish_fvg_top"] = l[i - 2]
            state["bearish_fvg_bottom"] = h[i]
            state["bearish_fvg_creation_bar"] = i
            state["important_fvg_count"] += 1
            add_event(events, data, i, state, "FVG", "bearish", price=mid)


def create_secondary_crt_allowed(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    candle_range: np.ndarray,
    body_ratio: np.ndarray,
    close_location: np.ndarray,
    range_to_atr: np.ndarray,
    inside_secondary: np.ndarray,
    inside_main: np.ndarray,
    main_session_finished: bool,
    state: dict[str, Any],
    crts_today: int,
    main_crt_created_today: bool,
    last_crt_creation_bar: int | None,
    config: CRTTBSConfig,
) -> bool:
    secondary_bull_quality = c[i] > o[i] and close_location[i] >= config.secondary_close_extreme
    secondary_bear_quality = c[i] < o[i] and close_location[i] <= 1.0 - config.secondary_close_extreme
    secondary_expansion = (not config.require_secondary_expansion) or h[i] > h[i - 1] or l[i] < l[i - 1]
    secondary_range_expansion = (not config.require_secondary_range_expansion) or candle_range[i] > candle_range[i - 1]
    valid_secondary = inside_secondary[i] and range_to_atr[i] >= config.secondary_range_atr and body_ratio[i] * 100.0 >= config.secondary_body_percent and (secondary_bull_quality or secondary_bear_quality) and secondary_expansion and secondary_range_expansion
    previous_finished = state["setup_state"] in {0, 5, 6, 7}
    cooldown = last_crt_creation_bar is None or i - last_crt_creation_bar >= config.secondary_cooldown_bars
    time_allowed = (not config.secondary_only_after_main) or main_crt_created_today
    return valid_secondary and time_allowed and not inside_main[i] and previous_finished and cooldown and crts_today < config.maximum_crts_per_day and not main_session_finished


def entry_record(data: pd.DataFrame, i: int, state: dict[str, Any], confirmation: dict[str, Any], c: np.ndarray, atr: np.ndarray, config: CRTTBSConfig, trade_id: int) -> dict[str, Any]:
    direction = int(confirmation["direction"])
    entry = float(c[i])
    structural_stop = state["active_low"] if direction == 1 else state["active_high"]
    if confirmation["component"] == "ORDER BLOCK":
        if direction == 1 and np.isfinite(state["bullish_ob_low"]):
            structural_stop = state["bullish_ob_low"]
        elif direction == -1 and np.isfinite(state["bearish_ob_high"]):
            structural_stop = state["bearish_ob_high"]
    stop = structural_stop - atr[i] * config.result_stop_buffer_atr if direction == 1 else structural_stop + atr[i] * config.result_stop_buffer_atr
    risk = abs(entry - stop)
    target = entry + risk * config.result_risk_reward if direction == 1 else entry - risk * config.result_risk_reward
    return {
        "TradeId": trade_id,
        "CRTNumber": state["active_number"],
        "CRTType": state["active_type"],
        "Signal": confirmation["signal"],
        "Direction": direction,
        "Model": confirmation["model"],
        "Component": confirmation["component"],
        "Potential": confirmation["potential"],
        "EntryBarIdx": i,
        "EntryDate": data.index[i],
        "EntryPrice": entry,
        "StopPrice": float(stop),
        "TargetPrice": float(target),
        "Result": "ACTIVE",
        "ResultBarIdx": np.nan,
        "ResultDate": pd.NaT,
        "ActiveCRTHigh": state["active_high"],
        "ActiveCRTLow": state["active_low"],
        "TBSBarIdx": state["sweep_bar"],
        "ChoChBarIdx": state["choch_bar"],
        "BOSBarIdx": state["bos_bar"],
        "BreakoutBarIdx": state["breakout_bar"],
    }


def update_trades(i: int, h: np.ndarray, l: np.ndarray, data: pd.DataFrame, active_trades: list[dict[str, Any]], entries: list[dict[str, Any]]) -> None:
    remaining = []
    for trade in active_trades:
        if i <= trade["EntryBarIdx"]:
            remaining.append(trade)
            continue
        direction = int(trade["Direction"])
        target_hit = h[i] >= trade["TargetPrice"] if direction == 1 else l[i] <= trade["TargetPrice"]
        stop_hit = l[i] <= trade["StopPrice"] if direction == 1 else h[i] >= trade["StopPrice"]
        if target_hit or stop_hit:
            result = "WIN" if target_hit and not stop_hit else "FAILD"
            trade["Result"] = result
            trade["ResultBarIdx"] = i
            trade["ResultDate"] = data.index[i]
            for row in entries:
                if row["TradeId"] == trade["TradeId"]:
                    row.update({"Result": result, "ResultBarIdx": i, "ResultDate": data.index[i]})
                    break
        else:
            remaining.append(trade)
    active_trades[:] = remaining


def add_event(events: list[dict[str, Any]], data: pd.DataFrame, i: int, state: dict[str, Any], event: str, detail: str, *, price: float | None = None, level: float | None = None) -> None:
    events.append(
        {
            "BarIdx": i,
            "Date": data.index[i],
            "Event": event,
            "Detail": detail,
            "CRTNumber": state.get("active_number", np.nan),
            "CRTType": state.get("active_type", ""),
            "Direction": state.get("setup_direction", 0),
            "Price": np.nan if price is None else float(price),
            "Level": np.nan if level is None or not np.isfinite(level) else float(level),
            "State": state.get("setup_state", 0),
        }
    )


def crt_record(data: pd.DataFrame, state: dict[str, Any]) -> dict[str, Any]:
    return {
        "CRTNumber": state["active_number"],
        "CRTType": state["active_type"],
        "StartBarIdx": state["active_start_bar"],
        "ActivationBarIdx": state["active_activation_bar"],
        "StartDate": data.index[state["active_start_bar"]],
        "ActivationDate": data.index[state["active_activation_bar"]],
        "CRTHigh": state["active_high"],
        "CRTLow": state["active_low"],
        "CRTMid": state["active_mid"],
    }


def confirmed_pivots(i: int, high: np.ndarray, low: np.ndarray, *, left: int, right: int) -> tuple[float, float]:
    if i < left + right:
        return np.nan, np.nan
    pivot_idx = i - right
    start = pivot_idx - left
    end = pivot_idx + right + 1
    pivot_high = high[pivot_idx] if high[pivot_idx] >= finite_max(high[start:end]) else np.nan
    pivot_low = low[pivot_idx] if low[pivot_idx] <= finite_min(low[start:end]) else np.nan
    return pivot_high, pivot_low


def raw_fvg(i: int, high: np.ndarray, low: np.ndarray, atr: np.ndarray, config: CRTTBSConfig) -> tuple[bool, bool]:
    if i < 2:
        return False, False
    bullish = low[i] > high[i - 2] and low[i] - high[i - 2] >= atr[i] * config.minimum_fvg_atr
    bearish = high[i] < low[i - 2] and low[i - 2] - high[i] >= atr[i] * config.minimum_fvg_atr
    return bool(bullish), bool(bearish)


def index_in_timezone(index: pd.DatetimeIndex, config: CRTTBSConfig) -> pd.DatetimeIndex:
    idx = pd.DatetimeIndex(index)
    if idx.tz is None:
        idx = idx.tz_localize(ZoneInfo(config.data_timezone), nonexistent="shift_forward", ambiguous="NaT")
    return idx.tz_convert(ZoneInfo(config.timezone))


def in_session(ts: pd.Timestamp, session: str) -> bool:
    if pd.isna(ts):
        return False
    start_text, end_text = session.split("-")
    start = int(start_text[:2]) * 60 + int(start_text[2:])
    end = int(end_text[:2]) * 60 + int(end_text[2:])
    now = int(ts.hour) * 60 + int(ts.minute)
    if start <= end:
        return start <= now < end
    return now >= start or now < end


def date_key(ts: pd.Timestamp) -> int:
    if pd.isna(ts):
        return -1
    return int(ts.year) * 10000 + int(ts.month) * 100 + int(ts.day)


def safe_atr_value(value: float) -> float:
    return float(value) if np.isfinite(value) and value > 0 else 1e-12


def finite_min(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.min(arr)) if len(arr) else np.nan


def finite_max(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.max(arr)) if len(arr) else np.nan


def empty_entries() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "TradeId",
            "CRTNumber",
            "CRTType",
            "Signal",
            "Direction",
            "Model",
            "Component",
            "Potential",
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "StopPrice",
            "TargetPrice",
            "Result",
            "ResultBarIdx",
            "ResultDate",
        ]
    )


def run_vectorbt_backtest(
    data: pd.DataFrame,
    result: dict[str, pd.DataFrame],
    config: CRTTBSConfig,
    *,
    show_plot: bool,
) -> dict[str, Any]:
    signals = build_vectorbt_signals(data, result["entries"], config)
    orders = build_vectorbt_order_frame(data, result["entries"], config)
    if vbt is None:
        detail = "" if VBT_IMPORT_ERROR is None else f" ({type(VBT_IMPORT_ERROR).__name__}: {VBT_IMPORT_ERROR})"
        message = f"Install/fix vectorbt to run the portfolio backtest: pip install vectorbt{detail}"
        if config.backtest_raise_missing_vectorbt:
            raise ImportError(message)
        warnings.warn(message)
        return {"signals": signals, "orders": orders, "portfolio": None, "stats": pd.Series(dtype=object), "fig": None}

    engine = config.backtest_engine.lower().strip()
    if engine == "orders":
        portfolio = vectorbt_from_orders(data, orders, config)
    elif engine == "signals":
        portfolio = vectorbt_from_signals(data, signals, config)
    else:
        raise ValueError("backtest_engine must be 'orders' or 'signals'.")

    stats = portfolio_stats(portfolio)
    fig = None
    if show_plot:
        fig = portfolio_plot(portfolio)
        if fig is not None:
            display_figure(fig)
    return {"signals": signals, "orders": orders, "portfolio": portfolio, "stats": stats, "fig": fig}


def empty_backtest_result(data: pd.DataFrame, result: dict[str, pd.DataFrame], config: CRTTBSConfig) -> dict[str, Any]:
    signals = empty_vectorbt_signals(data)
    orders = build_vectorbt_order_frame(data, result["entries"], config) if "entries" in result else empty_order_frame()
    return {"signals": signals, "orders": orders, "portfolio": None, "stats": pd.Series(dtype=object), "fig": None}


def build_vectorbt_signals(data: pd.DataFrame, entries: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    signals = empty_vectorbt_signals(data)
    if entries.empty:
        return signals

    for _, row in entries.iterrows():
        raw_entry_idx = int(row["EntryBarIdx"])
        entry_idx = raw_entry_idx + int(config.backtest_shift_signals)
        if entry_idx >= len(data):
            continue
        direction = int(row["Direction"])
        entry_price = float(data[config.backtest_price_column].iloc[entry_idx])
        stop_price = float(row["StopPrice"])
        target_price = float(row["TargetPrice"])
        if direction == 1:
            signals.iat[entry_idx, signals.columns.get_loc("long_entries")] = True
            signals.iat[entry_idx, signals.columns.get_loc("sl_stop")] = max((entry_price - stop_price) / entry_price, 0.0)
            signals.iat[entry_idx, signals.columns.get_loc("tp_stop")] = max((target_price - entry_price) / entry_price, 0.0)
        else:
            signals.iat[entry_idx, signals.columns.get_loc("short_entries")] = True
            signals.iat[entry_idx, signals.columns.get_loc("sl_stop")] = max((stop_price - entry_price) / entry_price, 0.0)
            signals.iat[entry_idx, signals.columns.get_loc("tp_stop")] = max((entry_price - target_price) / entry_price, 0.0)
        signals.iat[entry_idx, signals.columns.get_loc("trade_id")] = int(row["TradeId"])
        signals.iat[entry_idx, signals.columns.get_loc("entry_price")] = entry_price
        signals.iat[entry_idx, signals.columns.get_loc("stop_price")] = stop_price
        signals.iat[entry_idx, signals.columns.get_loc("target_price")] = target_price

    return signals


def empty_vectorbt_signals(data: pd.DataFrame) -> pd.DataFrame:
    return pd.DataFrame(
        {
            "long_entries": pd.Series(False, index=data.index, dtype=bool),
            "long_exits": pd.Series(False, index=data.index, dtype=bool),
            "short_entries": pd.Series(False, index=data.index, dtype=bool),
            "short_exits": pd.Series(False, index=data.index, dtype=bool),
            "sl_stop": pd.Series(np.nan, index=data.index, dtype=float),
            "tp_stop": pd.Series(np.nan, index=data.index, dtype=float),
            "trade_id": pd.Series(np.nan, index=data.index, dtype=float),
            "entry_price": pd.Series(np.nan, index=data.index, dtype=float),
            "stop_price": pd.Series(np.nan, index=data.index, dtype=float),
            "target_price": pd.Series(np.nan, index=data.index, dtype=float),
        }
    )


def build_vectorbt_order_frame(data: pd.DataFrame, entries: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    if entries.empty:
        return empty_order_frame()
    rows: list[dict[str, Any]] = []
    for _, entry in entries.iterrows():
        entry_idx = int(entry["EntryBarIdx"]) + int(config.backtest_shift_signals)
        if entry_idx >= len(data):
            continue
        direction = int(entry["Direction"])
        size = float(config.backtest_size) * direction
        rows.append(
            {
                "TradeId": int(entry["TradeId"]),
                "BarIdx": entry_idx,
                "Date": data.index[entry_idx],
                "Size": size,
                "Price": float(data[config.backtest_price_column].iloc[entry_idx]),
                "Reason": "ENTRY",
                "Signal": entry["Signal"],
                "Model": entry["Model"],
            }
        )
        if pd.notna(entry["ResultBarIdx"]):
            exit_idx = int(entry["ResultBarIdx"])
            if exit_idx < len(data) and exit_idx > entry_idx:
                exit_price = float(entry["TargetPrice"]) if entry["Result"] == "WIN" else float(entry["StopPrice"])
                rows.append(
                    {
                        "TradeId": int(entry["TradeId"]),
                        "BarIdx": exit_idx,
                        "Date": data.index[exit_idx],
                        "Size": -size,
                        "Price": exit_price,
                        "Reason": str(entry["Result"]),
                        "Signal": entry["Signal"],
                        "Model": entry["Model"],
                    }
                )
    if not rows:
        return empty_order_frame()
    return pd.DataFrame(rows).sort_values(["BarIdx", "TradeId", "Reason"]).reset_index(drop=True)


def empty_order_frame() -> pd.DataFrame:
    return pd.DataFrame(columns=["TradeId", "BarIdx", "Date", "Size", "Price", "Reason", "Signal", "Model"])


def vectorbt_from_orders(data: pd.DataFrame, orders: pd.DataFrame, config: CRTTBSConfig):
    order_size = pd.Series(np.nan, index=data.index, dtype=float)
    order_price = data[config.backtest_price_column].astype(float).copy()
    if not orders.empty:
        for bar_idx, group in orders.groupby("BarIdx", sort=True):
            bar_idx = int(bar_idx)
            size_sum = float(group["Size"].sum())
            if size_sum == 0.0:
                continue
            weights = group["Size"].abs()
            order_size.iloc[bar_idx] = size_sum
            order_price.iloc[bar_idx] = float((group["Price"] * weights).sum() / weights.sum())
    return vbt.Portfolio.from_orders(
        close=data["close"].astype(float),
        size=order_size,
        price=order_price,
        size_type=config.backtest_size_type,
        direction="both",
        fees=config.backtest_fees,
        fixed_fees=config.backtest_fixed_fees,
        slippage=config.backtest_slippage,
        init_cash=config.backtest_init_cash,
        freq=vectorbt_freq(data, config),
    )


def vectorbt_from_signals(data: pd.DataFrame, signals: pd.DataFrame, config: CRTTBSConfig):
    price = data[config.backtest_price_column].astype(float)
    return vbt.Portfolio.from_signals(
        close=data["close"].astype(float),
        entries=signals["long_entries"],
        exits=signals["long_exits"],
        short_entries=signals["short_entries"],
        short_exits=signals["short_exits"],
        size=config.backtest_size,
        size_type=config.backtest_size_type,
        price=price,
        fees=config.backtest_fees,
        fixed_fees=config.backtest_fixed_fees,
        slippage=config.backtest_slippage,
        open=data["open"].astype(float),
        high=data["high"].astype(float),
        low=data["low"].astype(float),
        sl_stop=signals["sl_stop"],
        tp_stop=signals["tp_stop"],
        sl_trail=False,
        upon_opposite_entry=config.backtest_upon_opposite_entry,
        accumulate=config.backtest_accumulate,
        init_cash=config.backtest_init_cash,
        freq=vectorbt_freq(data, config),
    )


def vectorbt_freq(data: pd.DataFrame, config: CRTTBSConfig) -> str | None:
    if config.backtest_freq is not None:
        return config.backtest_freq
    try:
        return pd.infer_freq(data.index)
    except (TypeError, ValueError):
        return None


def portfolio_stats(portfolio: Any) -> pd.Series:
    try:
        return portfolio.stats()
    except Exception as exc:
        warnings.warn(f"vectorbt stats failed: {exc}")
        return pd.Series(dtype=object)


def portfolio_plot(portfolio: Any) -> Any:
    try:
        return portfolio.plot()
    except Exception as exc:
        warnings.warn(f"vectorbt plot failed: {exc}")
        return None


def plot_crt_tbs_state_machine(data: pd.DataFrame, result: dict[str, pd.DataFrame], *, title: str):
    if go is None:
        warnings.warn("plotly is not installed; skipping plot.")
        return None
    x = np.arange(len(data))
    labels = data.index.strftime("%Y-%m-%d %H:%M")
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=labels,
            hovertext=labels,
            hoverinfo="text+y",
        )
    )
    for _, row in result["crts"].iterrows():
        start = int(row["StartBarIdx"])
        end = min(len(data) - 1, int(row["ActivationBarIdx"]) + 80)
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTHigh"], y1=row["CRTHigh"], line=dict(color="#E5E7EB", width=1))
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTLow"], y1=row["CRTLow"], line=dict(color="#E5E7EB", width=1))
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTMid"], y1=row["CRTMid"], line=dict(color="#94A3B8", width=1, dash="dot"))
    event_colors = {"TBS": "#FACC15", "CHoCH": "#38BDF8", "BOS": "#22C55E", "BREAKOUT": "#F97316", "ENTRY": "#A78BFA", "INVALIDATED": "#EF4444", "EXPIRED": "#64748B", "FVG": "#CBD5E1", "CRT": "#E5E7EB"}
    if not result["events"].empty:
        for event_name, group in result["events"].groupby("Event"):
            prices = []
            for _, row in group.iterrows():
                if np.isfinite(row.get("Price", np.nan)):
                    prices.append(float(row["Price"]))
                elif np.isfinite(row.get("Level", np.nan)):
                    prices.append(float(row["Level"]))
                else:
                    prices.append(float(data["close"].iloc[int(row["BarIdx"])]))
            fig.add_trace(
                go.Scatter(
                    x=group["BarIdx"],
                    y=prices,
                    mode="markers+text",
                    name=event_name,
                    marker=dict(color=event_colors.get(event_name, "#E5E7EB"), size=9),
                    text=group["Event"],
                    textposition="top center",
                    hovertext=[
                        f"{row['Event']} {row['Detail']}<br>{row['Date']}<br>CRT #{row['CRTNumber']}<br>Level: {row['Level']}"
                        for _, row in group.iterrows()
                    ],
                    hoverinfo="text",
                )
            )
    if not result["entries"].empty:
        fig.add_trace(
            go.Scatter(
                x=result["entries"]["EntryBarIdx"],
                y=result["entries"]["EntryPrice"],
                mode="markers+text",
                name="Entries",
                marker=dict(color="#A78BFA", size=13, symbol="star"),
                text=result["entries"]["Signal"],
                textposition="bottom center",
                hovertext=[
                    (
                        f"{row['Signal']}<br>{row['Model']} / {row['Component']}<br>"
                        f"Entry: {row['EntryPrice']:.4f}<br>Stop: {row['StopPrice']:.4f}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>Result: {row['Result']}"
                    )
                    for _, row in result["entries"].iterrows()
                ],
                hoverinfo="text",
            )
        )
    ticks = np.linspace(0, len(data) - 1, min(12, len(data)), dtype=int)
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", y=1.02, x=1, xanchor="right"),
        margin=dict(l=55, r=40, t=80, b=60),
    )
    fig.update_xaxes(tickmode="array", tickvals=ticks, ticktext=labels[ticks], tickangle=-30)
    return fig


def display_figure(fig: Any) -> None:
    try:
        from IPython.display import display
    except ModuleNotFoundError:
        return
    display(fig)
```

```python
from __future__ import annotations

from dataclasses import dataclass
from pathlib import Path
from typing import Any
from zoneinfo import ZoneInfo
import warnings

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None

try:
    import vectorbt as vbt
    VBT_IMPORT_ERROR = None
except Exception as exc:
    vbt = None
    VBT_IMPORT_ERROR = exc


@dataclass
class CRTTBSConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    data_timezone: str = "Europe/Istanbul"
    timezone: str = "America/New_York"

    main_session: str = "0830-0945"
    secondary_session: str = "0945-1600"
    search_24x7: bool = True
    maximum_crts_per_day: int = 4
    secondary_cooldown_bars: int = 20
    setup_expiration_bars: int = 120

    atr_length: int = 14
    secondary_range_atr: float = 1.50
    secondary_body_percent: float = 65.0
    secondary_close_extreme: float = 0.80
    require_secondary_expansion: bool = True
    secondary_only_after_main: bool = True
    require_secondary_range_expansion: bool = True

    swing_left: int = 3
    swing_right: int = 3
    structure_lookback: int = 20
    sweep_ticks: float = 1.0
    tick_size: float = 0.01
    require_close_inside: bool = True
    bos_break_atr: float = 0.05
    max_bars_to_choch: int = 20
    invalidation_close_atr: float = 0.10

    enable_breakout_retest: bool = True
    breakout_body_atr: float = 0.70
    breakout_body_percent: float = 65.0
    breakout_close_atr: float = 0.10
    breakout_retest_bars: int = 20
    require_breakout_fvg: bool = False
    breakout_invalidation_atr: float = 0.15

    show_important_fvg: bool = True
    minimum_fvg_atr: float = 0.15
    displacement_body_atr: float = 0.70
    displacement_body_percent: float = 65.0
    maximum_important_fvg: int = 1

    show_order_block: bool = True
    order_block_lookback: int = 12
    minimum_choch_displacement_atr: float = 0.50

    entry_body_percent: float = 45.0
    entry_confirmation_bars: int = 15
    retest_tolerance_atr: float = 0.20
    accept_bos_retest: bool = True
    accept_fvg_retest: bool = True
    accept_order_block_retest: bool = True
    require_fvg_or_ob: bool = True
    require_close_beyond_bos: bool = True
    require_midpoint_alignment: bool = True
    require_ema_alignment: bool = False
    fast_ema_length: int = 20
    slow_ema_length: int = 50
    minimum_potential_score: int = 3

    enable_sweep_rejection_entry: bool = True
    sweep_rejection_bars: int = 4
    minimum_rejection_wick_percent: float = 35.0

    enable_high_potential_zones: bool = True
    important_zone_tolerance_atr: float = 0.15

    result_risk_reward: float = 1.5
    result_stop_buffer_atr: float = 0.10
    result_target_mode: str = "crt"

    use_mtf_crt: bool = True
    htf_intervals: tuple[str, ...] = ("4h",)
    htf_csv_path: str | None = None
    htf_min_range_atr: float = 0.0
    htf_target_zone_fraction: float = 0.05
    htf_allow_both_side_sweep: bool = False
    mtf_entry_mode: str = "tbs_touch"
    use_wick_touch: bool = True
    use_body_touch: bool = True
    require_internal_crt_confirmation: bool = False
    internal_crt_enabled: bool = True
    internal_crt_inside_parent: bool = True

    backtest_engine: str = "orders"
    backtest_price_column: str = "close"
    backtest_init_cash: float = 100_000.0
    backtest_size: float = 1.0
    backtest_size_type: str = "amount"
    backtest_fees: float = 0.0
    backtest_fixed_fees: float = 0.0
    backtest_slippage: float = 0.0
    backtest_shift_signals: bool = False
    backtest_upon_opposite_entry: str = "close"
    backtest_accumulate: bool = False
    backtest_freq: str | None = None
    backtest_raise_missing_vectorbt: bool = False


def run_crt_tbs(
    *,
    csv_path: str | None = None,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    show_plot: bool = True,
    run_backtest: bool = True,
    show_backtest_plot: bool = True,
    save_csv_path: str | None = None,
    **overrides: Any,
) -> dict[str, Any]:
    config = CRTTBSConfig(symbol=symbol, period=period, interval=interval, **overrides)
    data = load_ohlcv(csv_path=csv_path, config=config)
    data = add_indicators(data, config)
    result = run_mtf_crt_tbs_state_machine(data, config) if config.use_mtf_crt else run_pine_state_machine(data, config)
    backtest = run_vectorbt_backtest(
        data,
        result,
        config,
        show_plot=show_plot and show_backtest_plot,
    ) if run_backtest else empty_backtest_result(data, result, config)

    if save_csv_path is not None:
        Path(save_csv_path).parent.mkdir(parents=True, exist_ok=True)
        result["entries"].to_csv(save_csv_path, index=False)

    mode_text = f"MTF CRT {','.join(map(str, config.htf_intervals))} -> {config.interval} TBS" if config.use_mtf_crt else "CRT TBS Pine-style"
    fig = plot_crt_tbs_state_machine(data, result, title=f"{config.symbol} {mode_text}")
    if show_plot and fig is not None:
        display_figure(fig)

    print(mode_text)
    print(
        f"bars: {len(data)} | HTF CRTs: {len(result.get('htf_crts', []))} | "
        f"active CRTs: {len(result['crts'])} | internal CRTs: {len(result.get('internal_crts', []))} | "
        f"events: {len(result['events'])} | entries: {len(result['entries'])}"
    )
    if len(result["entries"]):
        preview = [
            "Signal",
            "Model",
            "Component",
            "Potential",
            "EntryDate",
            "EntryPrice",
            "StopPrice",
            "TargetPrice",
            "Result",
        ]
        print(result["entries"][preview].head(30).to_string(index=False))
    if run_backtest:
        print("VectorBT backtest")
        print(
            f"engine: {config.backtest_engine} | "
            f"long entries: {int(backtest['signals']['long_entries'].sum())} | "
            f"short entries: {int(backtest['signals']['short_entries'].sum())} | "
            f"orders: {len(backtest['orders'])}"
        )
        if backtest["portfolio"] is None:
            print("portfolio: vectorbt is not installed in this Python environment")
        elif len(backtest["stats"]):
            preferred = [
                "Start",
                "End",
                "Period",
                "Start Value",
                "End Value",
                "Total Return [%]",
                "Benchmark Return [%]",
                "Max Drawdown [%]",
                "Total Trades",
                "Win Rate [%]",
                "Profit Factor",
                "Sharpe Ratio",
            ]
            stats = backtest["stats"]
            available = [name for name in preferred if name in stats.index]
            print(stats.loc[available].to_string())

    return {
        "config": config,
        "data": data,
        "fig": fig,
        **result,
        "setups": result["entries"],
        "backtest": backtest,
        "backtest_signals": backtest["signals"],
        "backtest_orders": backtest["orders"],
        "portfolio": backtest["portfolio"],
        "portfolio_stats": backtest["stats"],
        "backtest_fig": backtest["fig"],
    }


def load_ohlcv(*, csv_path: str | None, config: CRTTBSConfig) -> pd.DataFrame:
    if csv_path is not None:
        return prepare_ohlcv(pd.read_csv(csv_path))
    if Ticker is None:
        raise ImportError("Install yahooquery or pass csv_path.")

    raw = Ticker(config.symbol).history(period=config.period, interval=config.interval)
    if isinstance(raw, pd.DataFrame) and isinstance(raw.index, pd.MultiIndex):
        if "symbol" in raw.index.names:
            raw = raw.reset_index(level="symbol", drop=True)
        else:
            raw = raw.reset_index()
    return prepare_ohlcv(raw)


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    data = df.copy()
    data = data.rename(columns={column: str(column).lower() for column in data.columns})
    if not isinstance(data.index, pd.DatetimeIndex):
        date_col = next(
            (col for col in ["datetime", "date", "timestamp", "time"] if col in data.columns),
            None,
        )
        if date_col is None:
            raise ValueError("Use DatetimeIndex or include datetime/date/timestamp column.")
        data[date_col] = pd.to_datetime(data[date_col], errors="coerce")
        data = data.set_index(date_col)

    required = {"open", "high", "low", "close"}
    missing = required.difference(data.columns)
    if missing:
        raise ValueError(f"Missing OHLC columns: {sorted(missing)}")

    data.index = pd.to_datetime(data.index)
    data = data[~data.index.duplicated(keep="last")].sort_index()
    if "volume" not in data.columns:
        data["volume"] = 0.0
    for col in ["open", "high", "low", "close", "volume"]:
        data[col] = pd.to_numeric(data[col], errors="coerce")
    data = data.dropna(subset=["open", "high", "low", "close"])
    data["volume"] = data["volume"].fillna(0.0).clip(lower=0.0)
    return data[["open", "high", "low", "close", "volume"]]


def add_indicators(data: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    out = data.copy()
    high = out["high"].astype(float)
    low = out["low"].astype(float)
    open_ = out["open"].astype(float)
    close = out["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [high - low, (high - previous_close).abs(), (low - previous_close).abs()],
        axis=1,
    ).max(axis=1)
    out["atr"] = true_range.rolling(config.atr_length, min_periods=1).mean()
    out["candle_range"] = high - low
    out["candle_body"] = (close - open_).abs()
    out["body_ratio"] = out["candle_body"] / out["candle_range"].replace(0, np.nan)
    out["body_ratio"] = out["body_ratio"].fillna(0.0)
    out["close_location"] = (close - low) / out["candle_range"].replace(0, np.nan)
    out["close_location"] = out["close_location"].fillna(0.5)
    out["range_to_atr"] = out["candle_range"] / out["atr"].replace(0, np.nan)
    out["fast_ema"] = close.ewm(span=config.fast_ema_length, adjust=False, min_periods=1).mean()
    out["slow_ema"] = close.ewm(span=config.slow_ema_length, adjust=False, min_periods=1).mean()
    return out.replace([np.inf, -np.inf], np.nan)


def run_mtf_crt_tbs_state_machine(data: pd.DataFrame, config: CRTTBSConfig) -> dict[str, pd.DataFrame]:
    o = data["open"].to_numpy(dtype=float)
    h = data["high"].to_numpy(dtype=float)
    l = data["low"].to_numpy(dtype=float)
    c = data["close"].to_numpy(dtype=float)
    atr = data["atr"].to_numpy(dtype=float)
    body = data["candle_body"].to_numpy(dtype=float)
    body_ratio = data["body_ratio"].to_numpy(dtype=float)
    fast_ema = data["fast_ema"].to_numpy(dtype=float)
    slow_ema = data["slow_ema"].to_numpy(dtype=float)

    htf_crts = detect_all_htf_crts(data, config)
    activations: dict[int, list[dict[str, Any]]] = {}
    for _, setup in htf_crts.iterrows():
        activations.setdefault(int(setup["ActivationBarIdx"]), []).append(setup.to_dict())

    state = new_state()
    crts: list[dict[str, Any]] = []
    events: list[dict[str, Any]] = []
    entries: list[dict[str, Any]] = []
    internal_crts: list[dict[str, Any]] = []
    active_trades: list[dict[str, Any]] = []

    last_swing_high = np.nan
    last_swing_low = np.nan
    next_trade_id = 1

    for i in range(len(data)):
        pivot_high, pivot_low = confirmed_pivots(
            i=i,
            high=h,
            low=l,
            left=config.swing_left,
            right=config.swing_right,
        )
        if np.isfinite(pivot_high):
            last_swing_high = pivot_high
        if np.isfinite(pivot_low):
            last_swing_low = pivot_low

        if i in activations and state["setup_state"] in {0, 5, 6, 7}:
            setup = choose_htf_setup(activations[i])
            state = activate_crt(
                crt_high=float(setup["RangeHigh"]),
                crt_low=float(setup["RangeLow"]),
                crt_start_bar=int(setup["RangeStartBarIdx"]),
                activation_bar=i,
                crt_type=f"HTF-{setup['Timeframe']}",
                crt_number=len(crts) + 1,
                allowed_direction=int(setup["Direction"]),
                target_price=float(setup["TargetPrice"]),
                target_zone_low=float(setup["TargetZoneLow"]),
                target_zone_high=float(setup["TargetZoneHigh"]),
                timeframe=str(setup["Timeframe"]),
                is_mtf=True,
            )
            crts.append(crt_record(data, state))
            add_event(
                events,
                data,
                i,
                state,
                "HTF_CRT",
                str(setup["Signal"]),
                price=state["active_mid"],
                level=state["active_target"],
            )

        if state["setup_state"] >= 1 and state["setup_state"] <= 4:
            if i - state["active_activation_bar"] > config.setup_expiration_bars:
                add_event(events, data, i, state, "EXPIRED", "setup_expiration")
                state["setup_state"] = 6
                state["status"] = "EXPIRADO"

        internal = detect_internal_crt(i, o, h, l, c, state, config)
        if internal is not None:
            state["internal_crt_seen"] = True
            state["internal_crt_bar"] = i
            internal_crts.append(internal_crt_record(data, i, state, internal))
            add_event(
                events,
                data,
                i,
                state,
                "INTERNAL_CRT",
                internal["signal"],
                price=internal["price"],
                level=internal["target"],
            )

        direct_confirmation = evaluate_mtf_tbs_touch_entry(i, o, h, l, c, state, config)
        if direct_confirmation is not None:
            state["setup_state"] = 5
            add_event(events, data, i, state, "LTF_TBS", direct_confirmation["component"], price=float(c[i]), level=state["active_target"])
            add_event(events, data, i, state, "ENTRY", direct_confirmation["model"], price=float(c[i]), level=state["active_target"])
            entry = entry_record(data, i, state, direct_confirmation, c, atr, config, next_trade_id)
            next_trade_id += 1
            entries.append(entry)
            active_trades.append(entry.copy())
            update_trades(i, h, l, data, active_trades, entries)
            continue

        if state.get("is_mtf_crt", False) and mtf_direct_entry_mode(config):
            update_trades(i, h, l, data, active_trades, entries)
            continue

        raw_bull_fvg, raw_bear_fvg = raw_fvg(i, h, l, atr, config)
        update_important_fvg(i, o, h, l, c, atr, body, body_ratio, raw_bull_fvg, raw_bear_fvg, state, config, events, data)

        bullish_sweep, bearish_sweep = process_tbs_sweep(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            state=state,
            last_swing_high=last_swing_high,
            last_swing_low=last_swing_low,
            config=config,
            events=events,
            data=data,
        )

        process_real_breakout(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body=body,
            body_ratio=body_ratio,
            raw_bull_fvg=raw_bull_fvg,
            raw_bear_fvg=raw_bear_fvg,
            bullish_sweep=bullish_sweep,
            bearish_sweep=bearish_sweep,
            state=state,
            config=config,
            events=events,
            data=data,
        )

        invalidate_before_choch(i, c, atr, state, config, events, data)
        process_choch(i, o, h, l, c, atr, body, state, config, events, data)
        process_bos_level(i, o, h, l, c, state)
        process_bos_confirmation(i, c, atr, state, config, events, data)
        invalidate_order_block(i, c, state, events, data)

        confirmation = evaluate_entries(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body_ratio=body_ratio,
            fast_ema=fast_ema,
            slow_ema=slow_ema,
            state=state,
            config=config,
        )
        if confirmation is not None:
            state["setup_state"] = 5
            add_event(events, data, i, state, "ENTRY", confirmation["model"], price=float(c[i]))
            entry = entry_record(data, i, state, confirmation, c, atr, config, next_trade_id)
            next_trade_id += 1
            entries.append(entry)
            active_trades.append(entry.copy())

        update_trades(i, h, l, data, active_trades, entries)

        if state["setup_state"] == 4 and i - state.get("bos_bar", i) > config.entry_confirmation_bars:
            add_event(events, data, i, state, "EXPIRED", "entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"
        if state["setup_state"] == 8 and i - state.get("breakout_bar", i) > config.breakout_retest_bars:
            add_event(events, data, i, state, "EXPIRED", "breakout_entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"

    return {
        "htf_crts": htf_crts,
        "internal_crts": pd.DataFrame(internal_crts) if internal_crts else empty_internal_crts(),
        "crts": pd.DataFrame(crts),
        "events": pd.DataFrame(events),
        "entries": pd.DataFrame(entries) if entries else empty_entries(),
        "trades": pd.DataFrame(entries) if entries else empty_entries(),
    }


def detect_all_htf_crts(data: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    rows: list[dict[str, Any]] = []
    for timeframe in config.htf_intervals:
        htf = load_or_resample_htf(data, config, timeframe)
        rows.extend(detect_htf_crts(data, htf, str(timeframe), config))
    if not rows:
        return empty_htf_crts()
    out = pd.DataFrame(rows).sort_values(["ActivationBarIdx", "TimeframeMinutes"], ascending=[True, False])
    out["SetupId"] = np.arange(1, len(out) + 1)
    return out.reset_index(drop=True)


def load_or_resample_htf(data: pd.DataFrame, config: CRTTBSConfig, timeframe: str) -> pd.DataFrame:
    if config.htf_csv_path is not None and len(config.htf_intervals) == 1:
        return align_index_timezone(prepare_ohlcv(pd.read_csv(config.htf_csv_path)), data.index)
    return resample_ohlcv(data, timeframe)


def align_index_timezone(frame: pd.DataFrame, target_index: pd.DatetimeIndex) -> pd.DataFrame:
    out = frame.copy()
    source = pd.DatetimeIndex(out.index)
    target = pd.DatetimeIndex(target_index)
    if target.tz is not None and source.tz is None:
        source = source.tz_localize(target.tz)
    elif target.tz is not None and source.tz is not None:
        source = source.tz_convert(target.tz)
    elif target.tz is None and source.tz is not None:
        source = source.tz_localize(None)
    out.index = source
    return out


def resample_ohlcv(data: pd.DataFrame, timeframe: str) -> pd.DataFrame:
    rule = timeframe_to_pandas_rule(timeframe)
    out = data.resample(rule, label="right", closed="right").agg(
        open=("open", "first"),
        high=("high", "max"),
        low=("low", "min"),
        close=("close", "last"),
        volume=("volume", "sum"),
    )
    return out.dropna(subset=["open", "high", "low", "close"])


def detect_htf_crts(data: pd.DataFrame, htf: pd.DataFrame, timeframe: str, config: CRTTBSConfig) -> list[dict[str, Any]]:
    rows: list[dict[str, Any]] = []
    if len(htf) < 2:
        return rows
    minutes = timeframe_to_minutes(timeframe)
    htf_atr = htf_atr_series(htf, config.atr_length)
    for j in range(1, len(htf)):
        previous = htf.iloc[j - 1]
        current = htf.iloc[j]
        previous_range = float(previous["high"] - previous["low"])
        if previous_range <= 0:
            continue
        if config.htf_min_range_atr > 0:
            current_atr = float(htf_atr.iloc[j])
            if np.isfinite(current_atr) and previous_range < current_atr * config.htf_min_range_atr:
                continue

        swept_low = float(current["low"]) <= float(previous["low"]) - config.tick_size * config.sweep_ticks
        swept_high = float(current["high"]) >= float(previous["high"]) + config.tick_size * config.sweep_ticks
        closed_inside = float(previous["low"]) < float(current["close"]) < float(previous["high"])
        if not closed_inside:
            continue
        if swept_low and swept_high and not config.htf_allow_both_side_sweep:
            continue

        direction = 0
        signal = ""
        target = np.nan
        if swept_low and not swept_high:
            direction = 1
            signal = "Bullish HTF TS"
            target = float(previous["high"])
        elif swept_high and not swept_low:
            direction = -1
            signal = "Bearish HTF TS"
            target = float(previous["low"])
        elif swept_low and swept_high:
            close_location = (float(current["close"]) - float(previous["low"])) / previous_range
            direction = 1 if close_location >= 0.5 else -1
            signal = "Bullish HTF TS" if direction == 1 else "Bearish HTF TS"
            target = float(previous["high"]) if direction == 1 else float(previous["low"])
        if direction == 0:
            continue

        confirm_time = htf.index[j]
        activation_idx = int(data.index.searchsorted(confirm_time, side="right"))
        if activation_idx >= len(data):
            continue
        previous_start_time = previous_htf_start_time(htf.index[j - 1], timeframe)
        start_idx = int(data.index.searchsorted(previous_start_time, side="left"))
        zone_low, zone_high = target_zone(
            target=target,
            direction=direction,
            range_high=float(previous["high"]),
            range_low=float(previous["low"]),
            fraction=config.htf_target_zone_fraction,
        )
        rows.append(
            {
                "SetupId": np.nan,
                "Timeframe": str(timeframe),
                "TimeframeMinutes": minutes,
                "Direction": direction,
                "Signal": signal,
                "RangeHigh": float(previous["high"]),
                "RangeLow": float(previous["low"]),
                "RangeMid": (float(previous["high"]) + float(previous["low"])) / 2.0,
                "TargetPrice": target,
                "TargetZoneLow": zone_low,
                "TargetZoneHigh": zone_high,
                "SourceRangeDate": htf.index[j - 1],
                "ConfirmDate": confirm_time,
                "ActivationDate": data.index[activation_idx],
                "RangeStartBarIdx": max(0, min(start_idx, len(data) - 1)),
                "ActivationBarIdx": activation_idx,
            }
        )
    return rows


def htf_atr_series(htf: pd.DataFrame, length: int) -> pd.Series:
    high = htf["high"].astype(float)
    low = htf["low"].astype(float)
    close = htf["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat([high - low, (high - previous_close).abs(), (low - previous_close).abs()], axis=1).max(axis=1)
    return true_range.rolling(length, min_periods=1).mean()


def choose_htf_setup(setups: list[dict[str, Any]]) -> dict[str, Any]:
    return sorted(setups, key=lambda row: float(row.get("TimeframeMinutes", 0.0)), reverse=True)[0]


def detect_internal_crt(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    if not config.internal_crt_enabled or i <= 0 or not state.get("is_mtf_crt", False):
        return None
    if state["setup_state"] not in {1, 2, 3, 4}:
        return None
    if config.internal_crt_inside_parent:
        previous_inside = h[i - 1] <= state["active_high"] and l[i - 1] >= state["active_low"]
        close_inside_parent = state["active_low"] < c[i] < state["active_high"]
        if not previous_inside or not close_inside_parent:
            return None

    prev_high = h[i - 1]
    prev_low = l[i - 1]
    swept_low = l[i] <= prev_low - config.tick_size * config.sweep_ticks
    swept_high = h[i] >= prev_high + config.tick_size * config.sweep_ticks
    closed_inside_prev = prev_low < c[i] < prev_high
    if not closed_inside_prev:
        return None

    allowed = int(state.get("allowed_direction", 0))
    bullish = swept_low and not swept_high and allowed in {0, 1}
    bearish = swept_high and not swept_low and allowed in {0, -1}
    if bullish:
        return {"direction": 1, "signal": "Bullish internal TS", "price": float(l[i]), "target": float(prev_high)}
    if bearish:
        return {"direction": -1, "signal": "Bearish internal TS", "price": float(h[i]), "target": float(prev_low)}
    return None


def evaluate_mtf_tbs_touch_entry(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    if not mtf_direct_entry_mode(config):
        return None
    if not state.get("is_mtf_crt", False) or state["setup_state"] != 1 or i <= state["active_activation_bar"]:
        return None
    if config.require_internal_crt_confirmation and not state.get("internal_crt_seen", False):
        return None

    allowed = int(state.get("allowed_direction", 0))
    bullish_wick_touch = l[i] < state["active_low"] and c[i] > state["active_low"] and h[i] < state["active_high"]
    bullish_body_touch = o[i] < state["active_low"] and c[i] > state["active_low"]
    bearish_wick_touch = h[i] > state["active_high"] and c[i] < state["active_high"] and l[i] > state["active_low"]
    bearish_body_touch = o[i] > state["active_high"] and c[i] < state["active_high"]
    bullish_touch = allowed in {0, 1} and ((config.use_wick_touch and bullish_wick_touch) or (config.use_body_touch and bullish_body_touch))
    bearish_touch = allowed in {0, -1} and ((config.use_wick_touch and bearish_wick_touch) or (config.use_body_touch and bearish_body_touch))

    if bullish_touch:
        state["setup_direction"] = 1
        state["sweep_bar"] = i
        state["low_swept"] = True
        state["last_sweep_price"] = float(l[i])
        score = 3 + int(c[i] > state["active_mid"]) + int(state.get("internal_crt_seen", False))
        return {"direction": 1, "signal": "BUY", "component": "HTF RANGE LOW", "model": "MTF TBS TOUCH", "potential": min(score, 5)}
    if bearish_touch:
        state["setup_direction"] = -1
        state["sweep_bar"] = i
        state["high_swept"] = True
        state["last_sweep_price"] = float(h[i])
        score = 3 + int(c[i] < state["active_mid"]) + int(state.get("internal_crt_seen", False))
        return {"direction": -1, "signal": "SELL", "component": "HTF RANGE HIGH", "model": "MTF TBS TOUCH", "potential": min(score, 5)}
    return None


def mtf_direct_entry_mode(config: CRTTBSConfig) -> bool:
    return config.mtf_entry_mode.lower().strip() in {"tbs_touch", "touch", "direct"}


def internal_crt_record(data: pd.DataFrame, i: int, state: dict[str, Any], internal: dict[str, Any]) -> dict[str, Any]:
    return {
        "BarIdx": i,
        "Date": data.index[i],
        "CRTNumber": state.get("active_number", np.nan),
        "Timeframe": state.get("active_timeframe", ""),
        "Direction": internal["direction"],
        "Signal": internal["signal"],
        "Price": internal["price"],
        "Target": internal["target"],
    }


def target_zone(*, target: float, direction: int, range_high: float, range_low: float, fraction: float) -> tuple[float, float]:
    width = max(abs(range_high - range_low) * fraction, 1e-12)
    if direction == 1:
        return float(target - width), float(target)
    return float(target), float(target + width)


def timeframe_to_pandas_rule(timeframe: str) -> str:
    text = str(timeframe).strip()
    lower = text.lower()
    if lower in {"d", "1d", "day", "daily"}:
        return "1D"
    if lower in {"w", "1w", "week", "weekly"}:
        return "1W"
    if lower in {"m", "1m", "month", "monthly"}:
        return "1ME"
    if lower.endswith("min"):
        return lower
    if lower.endswith("h"):
        return lower
    if lower.isdigit():
        return f"{lower}min"
    return text


def timeframe_to_minutes(timeframe: str) -> float:
    text = str(timeframe).strip().lower()
    if text.isdigit():
        return float(text)
    if text.endswith("min"):
        return float(text[:-3])
    if text.endswith("h"):
        return float(text[:-1]) * 60.0
    if text in {"d", "1d", "day", "daily"}:
        return 1440.0
    if text in {"w", "1w", "week", "weekly"}:
        return 10080.0
    if text in {"m", "1m", "month", "monthly"}:
        return 43200.0
    return 0.0


def previous_htf_start_time(timestamp: pd.Timestamp, timeframe: str) -> pd.Timestamp:
    minutes = timeframe_to_minutes(timeframe)
    if minutes <= 0:
        return timestamp
    return timestamp - pd.Timedelta(minutes=minutes)


def empty_htf_crts() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "SetupId",
            "Timeframe",
            "TimeframeMinutes",
            "Direction",
            "Signal",
            "RangeHigh",
            "RangeLow",
            "RangeMid",
            "TargetPrice",
            "TargetZoneLow",
            "TargetZoneHigh",
            "SourceRangeDate",
            "ConfirmDate",
            "ActivationDate",
            "RangeStartBarIdx",
            "ActivationBarIdx",
        ]
    )


def empty_internal_crts() -> pd.DataFrame:
    return pd.DataFrame(columns=["BarIdx", "Date", "CRTNumber", "Timeframe", "Direction", "Signal", "Price", "Target"])


def run_pine_state_machine(data: pd.DataFrame, config: CRTTBSConfig) -> dict[str, pd.DataFrame]:
    o = data["open"].to_numpy(dtype=float)
    h = data["high"].to_numpy(dtype=float)
    l = data["low"].to_numpy(dtype=float)
    c = data["close"].to_numpy(dtype=float)
    atr = data["atr"].to_numpy(dtype=float)
    body = data["candle_body"].to_numpy(dtype=float)
    body_ratio = data["body_ratio"].to_numpy(dtype=float)
    candle_range = data["candle_range"].to_numpy(dtype=float)
    close_location = data["close_location"].to_numpy(dtype=float)
    range_to_atr = data["range_to_atr"].to_numpy(dtype=float)
    fast_ema = data["fast_ema"].to_numpy(dtype=float)
    slow_ema = data["slow_ema"].to_numpy(dtype=float)

    tz_index = index_in_timezone(data.index, config)
    inside_main = np.array([in_session(ts, config.main_session) for ts in tz_index], dtype=bool)
    inside_secondary = np.array(
        [config.search_24x7 or in_session(ts, config.secondary_session) for ts in tz_index],
        dtype=bool,
    )
    date_keys = np.array([date_key(ts) for ts in tz_index], dtype=int)

    state = new_state()
    crts: list[dict[str, Any]] = []
    events: list[dict[str, Any]] = []
    entries: list[dict[str, Any]] = []
    active_trades: list[dict[str, Any]] = []

    best_main_range = np.nan
    best_main_high = np.nan
    best_main_low = np.nan
    best_main_bar = None
    main_crt_created_today = False
    crts_today = 0
    last_crt_creation_bar = None
    last_swing_high = np.nan
    last_swing_low = np.nan
    next_trade_id = 1

    for i in range(len(data)):
        new_trading_day = i > 0 and date_keys[i] != date_keys[i - 1]
        main_session_started = bool(inside_main[i] and (i == 0 or not inside_main[i - 1]))
        main_session_finished = bool((not inside_main[i]) and i > 0 and inside_main[i - 1])

        if new_trading_day:
            state = new_state()
            main_crt_created_today = False
            crts_today = 0
            last_crt_creation_bar = None
            best_main_range = np.nan
            best_main_high = np.nan
            best_main_low = np.nan
            best_main_bar = None

        if new_trading_day or main_session_started:
            best_main_range = np.nan
            best_main_high = np.nan
            best_main_low = np.nan
            best_main_bar = None

        if inside_main[i] and (not np.isfinite(best_main_range) or candle_range[i] > best_main_range):
            best_main_range = candle_range[i]
            best_main_high = h[i]
            best_main_low = l[i]
            best_main_bar = i

        pivot_high, pivot_low = confirmed_pivots(
            i=i,
            high=h,
            low=l,
            left=config.swing_left,
            right=config.swing_right,
        )
        if np.isfinite(pivot_high):
            last_swing_high = pivot_high
        if np.isfinite(pivot_low):
            last_swing_low = pivot_low

        create_main_crt = (
            main_session_finished
            and not main_crt_created_today
            and np.isfinite(best_main_high)
            and np.isfinite(best_main_low)
            and best_main_bar is not None
        )
        if create_main_crt:
            main_crt_created_today = True
            crts_today += 1
            state = activate_crt(
                crt_high=float(best_main_high),
                crt_low=float(best_main_low),
                crt_start_bar=int(best_main_bar),
                activation_bar=i,
                crt_type="PRINCIPAL",
                crt_number=crts_today,
            )
            last_crt_creation_bar = i
            crts.append(crt_record(data, state))
            add_event(events, data, i, state, "CRT", "PRINCIPAL", price=state["active_mid"])

        if state["setup_state"] >= 1 and state["setup_state"] <= 4:
            if i - state["active_activation_bar"] > config.setup_expiration_bars:
                add_event(events, data, i, state, "EXPIRED", "setup_expiration")
                state["setup_state"] = 6
                state["status"] = "EXPIRADO"

        raw_bull_fvg, raw_bear_fvg = raw_fvg(i, h, l, atr, config)
        update_important_fvg(i, o, h, l, c, atr, body, body_ratio, raw_bull_fvg, raw_bear_fvg, state, config, events, data)

        bullish_sweep, bearish_sweep = process_tbs_sweep(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            state=state,
            last_swing_high=last_swing_high,
            last_swing_low=last_swing_low,
            config=config,
            events=events,
            data=data,
        )

        process_real_breakout(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body=body,
            body_ratio=body_ratio,
            raw_bull_fvg=raw_bull_fvg,
            raw_bear_fvg=raw_bear_fvg,
            bullish_sweep=bullish_sweep,
            bearish_sweep=bearish_sweep,
            state=state,
            config=config,
            events=events,
            data=data,
        )

        invalidate_before_choch(i, c, atr, state, config, events, data)
        process_choch(i, o, h, l, c, atr, body, state, config, events, data)
        process_bos_level(i, o, h, l, c, state)
        process_bos_confirmation(i, c, atr, state, config, events, data)
        invalidate_order_block(i, c, state, events, data)

        confirmation = evaluate_entries(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body_ratio=body_ratio,
            fast_ema=fast_ema,
            slow_ema=slow_ema,
            state=state,
            config=config,
        )
        if confirmation is not None:
            state["setup_state"] = 5
            add_event(events, data, i, state, "ENTRY", confirmation["model"], price=float(c[i]))
            entry = entry_record(data, i, state, confirmation, c, atr, config, next_trade_id)
            next_trade_id += 1
            entries.append(entry)
            active_trades.append(entry.copy())

        update_trades(i, h, l, data, active_trades, entries)

        if state["setup_state"] == 4 and i - state.get("bos_bar", i) > config.entry_confirmation_bars:
            add_event(events, data, i, state, "EXPIRED", "entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"
        if state["setup_state"] == 8 and i - state.get("breakout_bar", i) > config.breakout_retest_bars:
            add_event(events, data, i, state, "EXPIRED", "breakout_entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"

        if create_secondary_crt_allowed(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            candle_range=candle_range,
            body_ratio=body_ratio,
            close_location=close_location,
            range_to_atr=range_to_atr,
            inside_secondary=inside_secondary,
            inside_main=inside_main,
            main_session_finished=main_session_finished,
            state=state,
            crts_today=crts_today,
            main_crt_created_today=main_crt_created_today,
            last_crt_creation_bar=last_crt_creation_bar,
            config=config,
        ):
            crts_today += 1
            state = activate_crt(
                crt_high=float(h[i]),
                crt_low=float(l[i]),
                crt_start_bar=i,
                activation_bar=i,
                crt_type="SECUNDARIO",
                crt_number=crts_today,
            )
            last_crt_creation_bar = i
            crts.append(crt_record(data, state))
            add_event(events, data, i, state, "CRT", "SECUNDARIO", price=state["active_mid"])

    return {
        "crts": pd.DataFrame(crts),
        "events": pd.DataFrame(events),
        "entries": pd.DataFrame(entries) if entries else empty_entries(),
        "trades": pd.DataFrame(entries) if entries else empty_entries(),
    }


def new_state() -> dict[str, Any]:
    return {
        "setup_state": 0,
        "setup_direction": 0,
        "active_high": np.nan,
        "active_low": np.nan,
        "active_mid": np.nan,
        "active_target": np.nan,
        "target_zone_low": np.nan,
        "target_zone_high": np.nan,
        "active_start_bar": None,
        "active_activation_bar": None,
        "active_type": "",
        "active_timeframe": "",
        "active_number": np.nan,
        "allowed_direction": 0,
        "is_mtf_crt": False,
        "internal_crt_seen": False,
        "internal_crt_bar": None,
        "last_sweep_price": np.nan,
        "sweep_bar": None,
        "low_swept": False,
        "high_swept": False,
        "choch_level": np.nan,
        "choch_bar": None,
        "impulse_extreme": np.nan,
        "bos_level": np.nan,
        "bos_bar": None,
        "pullback_detected": False,
        "first_retest_consumed": False,
        "breakout_level": np.nan,
        "breakout_bar": None,
        "breakout_had_fvg": False,
        "breakout_retest_consumed": False,
        "important_fvg_count": 0,
        "bullish_fvg_top": np.nan,
        "bullish_fvg_bottom": np.nan,
        "bullish_fvg_creation_bar": None,
        "bearish_fvg_top": np.nan,
        "bearish_fvg_bottom": np.nan,
        "bearish_fvg_creation_bar": None,
        "bullish_ob_high": np.nan,
        "bullish_ob_low": np.nan,
        "bullish_ob_creation_bar": None,
        "bearish_ob_high": np.nan,
        "bearish_ob_low": np.nan,
        "bearish_ob_creation_bar": None,
        "bullish_ob_created": False,
        "bearish_ob_created": False,
        "status": "Esperando",
    }


def activate_crt(
    *,
    crt_high: float,
    crt_low: float,
    crt_start_bar: int,
    activation_bar: int,
    crt_type: str,
    crt_number: int,
    allowed_direction: int = 0,
    target_price: float = np.nan,
    target_zone_low: float = np.nan,
    target_zone_high: float = np.nan,
    timeframe: str = "",
    is_mtf: bool = False,
) -> dict[str, Any]:
    state = new_state()
    mid = (float(crt_high) + float(crt_low)) / 2.0
    if not np.isfinite(target_price):
        if allowed_direction == 1:
            target_price = float(crt_high)
        elif allowed_direction == -1:
            target_price = float(crt_low)
    state.update(
        {
            "setup_state": 1,
            "active_high": float(crt_high),
            "active_low": float(crt_low),
            "active_mid": mid,
            "active_target": float(target_price) if np.isfinite(target_price) else np.nan,
            "target_zone_low": float(target_zone_low) if np.isfinite(target_zone_low) else np.nan,
            "target_zone_high": float(target_zone_high) if np.isfinite(target_zone_high) else np.nan,
            "active_start_bar": int(crt_start_bar),
            "active_activation_bar": int(activation_bar),
            "active_type": crt_type,
            "active_timeframe": timeframe,
            "active_number": int(crt_number),
            "allowed_direction": int(allowed_direction),
            "is_mtf_crt": bool(is_mtf),
            "status": "Esperando TBS",
        }
    )
    return state


def process_tbs_sweep(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    state: dict[str, Any],
    last_swing_high: float,
    last_swing_low: float,
    config: CRTTBSConfig,
    events: list[dict[str, Any]],
    data: pd.DataFrame,
) -> tuple[bool, bool]:
    if state["setup_state"] != 1 or i <= state["active_activation_bar"]:
        return False, False
    minimum_sweep = float(config.sweep_ticks) * float(config.tick_size)
    low_sweep_raw = l[i] <= state["active_low"] - minimum_sweep
    high_sweep_raw = h[i] >= state["active_high"] + minimum_sweep
    if low_sweep_raw:
        state["low_swept"] = True
    if high_sweep_raw:
        state["high_swept"] = True

    allowed_direction = int(state.get("allowed_direction", 0))
    bullish_sweep = low_sweep_raw and not high_sweep_raw and allowed_direction in {0, 1} and (not config.require_close_inside or c[i] > state["active_low"])
    bearish_sweep = high_sweep_raw and not low_sweep_raw and allowed_direction in {0, -1} and (not config.require_close_inside or c[i] < state["active_high"])
    if bullish_sweep:
        fallback_high = finite_max(h[max(0, i - config.structure_lookback) : i])
        state["setup_state"] = 2
        state["setup_direction"] = 1
        state["sweep_bar"] = i
        state["last_sweep_price"] = float(l[i])
        state["choch_level"] = last_swing_high if np.isfinite(last_swing_high) else fallback_high
        state["status"] = "Esperando CHoCH"
        add_event(events, data, i, state, "TBS", "bullish_sweep", price=float(l[i]), level=state["choch_level"])
    elif bearish_sweep:
        fallback_low = finite_min(l[max(0, i - config.structure_lookback) : i])
        state["setup_state"] = 2
        state["setup_direction"] = -1
        state["sweep_bar"] = i
        state["last_sweep_price"] = float(h[i])
        state["choch_level"] = last_swing_low if np.isfinite(last_swing_low) else fallback_low
        state["status"] = "Esperando CHoCH"
        add_event(events, data, i, state, "TBS", "bearish_sweep", price=float(h[i]), level=state["choch_level"])
    return bool(bullish_sweep), bool(bearish_sweep)


def process_real_breakout(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    atr: np.ndarray,
    body: np.ndarray,
    body_ratio: np.ndarray,
    raw_bull_fvg: bool,
    raw_bear_fvg: bool,
    bullish_sweep: bool,
    bearish_sweep: bool,
    state: dict[str, Any],
    config: CRTTBSConfig,
    events: list[dict[str, Any]],
    data: pd.DataFrame,
) -> None:
    if not config.enable_breakout_retest or state["setup_state"] != 1 or i <= state["active_activation_bar"]:
        return
    breakout_margin = atr[i] * config.breakout_close_atr
    strong_body = body[i] >= atr[i] * config.breakout_body_atr and body_ratio[i] * 100.0 >= config.breakout_body_percent
    bullish_breakout = c[i] > state["active_high"] + breakout_margin and strong_body and not bullish_sweep and not bearish_sweep
    bearish_breakout = c[i] < state["active_low"] - breakout_margin and strong_body and not bullish_sweep and not bearish_sweep
    if bullish_breakout:
        state["setup_state"] = 8
        state["setup_direction"] = 1
        state["breakout_level"] = state["active_high"]
        state["breakout_bar"] = i
        state["breakout_had_fvg"] = bool(raw_bull_fvg or (i > 0 and raw_fvg(i - 1, h, l, atr, config)[0]))
        state["breakout_retest_consumed"] = False
        state["status"] = "Breakout: esperando retest"
        add_event(events, data, i, state, "BREAKOUT", "bullish_breakout", price=float(c[i]), level=state["breakout_level"])
    elif bearish_breakout:
        state["setup_state"] = 8
        state["setup_direction"] = -1
        state["breakout_level"] = state["active_low"]
        state["breakout_bar"] = i
        state["breakout_had_fvg"] = bool(raw_bear_fvg or (i > 0 and raw_fvg(i - 1, h, l, atr, config)[1]))
        state["breakout_retest_consumed"] = False
        state["status"] = "Breakout: esperando retest"
        add_event(events, data, i, state, "BREAKOUT", "bearish_breakout", price=float(c[i]), level=state["breakout_level"])


def invalidate_before_choch(i: int, c: np.ndarray, atr: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 2:
        return
    both_sides = state["low_swept"] and state["high_swept"]
    time_expired = state["sweep_bar"] is not None and i - state["sweep_bar"] > config.max_bars_to_choch
    bullish_opposite = state["setup_direction"] == 1 and c[i] < state["active_low"] - atr[i] * config.invalidation_close_atr
    bearish_opposite = state["setup_direction"] == -1 and c[i] > state["active_high"] + atr[i] * config.invalidation_close_atr
    if both_sides or time_expired or bullish_opposite or bearish_opposite:
        state["setup_state"] = 7
        state["status"] = "INVALIDADO"
        add_event(events, data, i, state, "INVALIDATED", "before_choch")


def process_choch(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, atr: np.ndarray, body: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 2 or state["sweep_bar"] is None or i <= state["sweep_bar"] or not np.isfinite(state["choch_level"]):
        return
    current_body_atr = body[i] / max(atr[i], 1e-12)
    bullish = state["setup_direction"] == 1 and c[i] > state["choch_level"] and current_body_atr >= config.minimum_choch_displacement_atr
    bearish = state["setup_direction"] == -1 and c[i] < state["choch_level"] and current_body_atr >= config.minimum_choch_displacement_atr
    if bullish or bearish:
        state["setup_state"] = 3
        state["choch_bar"] = i
        state["impulse_extreme"] = h[i] if bullish else l[i]
        state["pullback_detected"] = False
        state["status"] = "Esperando retroceso"
        create_order_block(i, o, h, l, c, state, config)
        add_event(events, data, i, state, "CHoCH", "bullish" if bullish else "bearish", price=float(c[i]), level=state["choch_level"])


def process_bos_level(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any]) -> None:
    if state["setup_state"] != 3:
        return
    if state["setup_direction"] == 1 and not state["pullback_detected"]:
        state["impulse_extreme"] = max(state["impulse_extreme"], h[i]) if np.isfinite(state["impulse_extreme"]) else h[i]
        bullish_pullback = state["choch_bar"] is not None and i > state["choch_bar"] and (c[i] < o[i] or l[i] < l[i - 1])
        if bullish_pullback:
            state["pullback_detected"] = True
            state["bos_level"] = state["impulse_extreme"]
            state["status"] = "Esperando BOS"
    elif state["setup_direction"] == -1 and not state["pullback_detected"]:
        state["impulse_extreme"] = min(state["impulse_extreme"], l[i]) if np.isfinite(state["impulse_extreme"]) else l[i]
        bearish_pullback = state["choch_bar"] is not None and i > state["choch_bar"] and (c[i] > o[i] or h[i] > h[i - 1])
        if bearish_pullback:
            state["pullback_detected"] = True
            state["bos_level"] = state["impulse_extreme"]
            state["status"] = "Esperando BOS"


def process_bos_confirmation(i: int, c: np.ndarray, atr: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 3 or not state["pullback_detected"] or not np.isfinite(state["bos_level"]):
        return
    distance = atr[i] * config.bos_break_atr
    bullish = state["setup_direction"] == 1 and c[i] > state["bos_level"] + distance
    bearish = state["setup_direction"] == -1 and c[i] < state["bos_level"] - distance
    if bullish or bearish:
        state["setup_state"] = 4
        state["bos_bar"] = i
        state["first_retest_consumed"] = False
        state["status"] = "Esperando entrada"
        add_event(events, data, i, state, "BOS", "bullish" if bullish else "bearish", price=float(c[i]), level=state["bos_level"])


def evaluate_entries(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    atr: np.ndarray,
    body_ratio: np.ndarray,
    fast_ema: np.ndarray,
    slow_ema: np.ndarray,
    state: dict[str, Any],
    config: CRTTBSConfig,
) -> dict[str, Any] | None:
    if config.require_internal_crt_confirmation and state.get("is_mtf_crt", False) and not state.get("internal_crt_seen", False):
        return None
    current_atr = safe_atr_value(atr[i])
    (
        bullish_rejection_candle,
        bearish_rejection_candle,
        bullish_rejection,
        bearish_rejection,
        bullish_rejection_score,
        bearish_rejection_score,
    ) = rejection_signals(i, o, h, l, c, body_ratio, fast_ema, slow_ema, state, config)
    tbs = tbs_entry_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config)
    breakout = breakout_retest_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config)
    zone = zone_confluence_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config, bullish_rejection_candle, bearish_rejection_candle)

    if tbs is not None:
        return tbs
    if breakout is not None:
        return breakout
    if bullish_rejection and state["setup_direction"] == 1:
        return {"direction": 1, "signal": "BUY", "component": "BARRIDA + RECHAZO", "model": "TBS REJECTION", "potential": bullish_rejection_score}
    if bearish_rejection and state["setup_direction"] == -1:
        return {"direction": -1, "signal": "SELL", "component": "BARRIDA + RECHAZO", "model": "TBS REJECTION", "potential": bearish_rejection_score}
    if zone is not None:
        return zone
    return None


def tbs_entry_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    valid = state["setup_state"] == 4 and state["bos_bar"] is not None and i > state["bos_bar"] and i - state["bos_bar"] <= config.entry_confirmation_bars and not state["first_retest_consumed"]
    if not valid:
        return None
    tolerance = current_atr * config.retest_tolerance_atr
    bullish_bos_retest = state["setup_direction"] == 1 and config.accept_bos_retest and l[i] <= state["bos_level"] + tolerance and h[i] >= state["bos_level"] - tolerance
    bearish_bos_retest = state["setup_direction"] == -1 and config.accept_bos_retest and h[i] >= state["bos_level"] - tolerance and l[i] <= state["bos_level"] + tolerance
    bullish_fvg_retest = state["setup_direction"] == 1 and config.accept_fvg_retest and state["bullish_fvg_creation_bar"] is not None and state["bullish_fvg_creation_bar"] > state["sweep_bar"] and l[i] <= state["bullish_fvg_top"] + tolerance and h[i] >= state["bullish_fvg_bottom"] - tolerance
    bearish_fvg_retest = state["setup_direction"] == -1 and config.accept_fvg_retest and state["bearish_fvg_creation_bar"] is not None and state["bearish_fvg_creation_bar"] > state["sweep_bar"] and h[i] >= state["bearish_fvg_bottom"] - tolerance and l[i] <= state["bearish_fvg_top"] + tolerance
    bullish_ob_retest = state["setup_direction"] == 1 and config.accept_order_block_retest and state["bullish_ob_created"] and state["bullish_ob_creation_bar"] >= state["sweep_bar"] and l[i] <= state["bullish_ob_high"] + tolerance and h[i] >= state["bullish_ob_low"] - tolerance and c[i] >= state["bullish_ob_low"]
    bearish_ob_retest = state["setup_direction"] == -1 and config.accept_order_block_retest and state["bearish_ob_created"] and state["bearish_ob_creation_bar"] >= state["sweep_bar"] and h[i] >= state["bearish_ob_low"] - tolerance and l[i] <= state["bearish_ob_high"] + tolerance and c[i] <= state["bearish_ob_high"]

    bullish_primary = bullish_ob_retest or bullish_fvg_retest
    bearish_primary = bearish_ob_retest or bearish_fvg_retest
    bullish_zone_accepted = bullish_primary if config.require_fvg_or_ob else (bullish_primary or bullish_bos_retest)
    bearish_zone_accepted = bearish_primary if config.require_fvg_or_ob else (bearish_primary or bearish_bos_retest)
    bullish_confirm = c[i] > o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] > c[i - 1]
    bearish_confirm = c[i] < o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] < c[i - 1]
    bullish_bos_close = (not config.require_close_beyond_bos) or c[i] > state["bos_level"]
    bearish_bos_close = (not config.require_close_beyond_bos) or c[i] < state["bos_level"]
    bullish_mid = (not config.require_midpoint_alignment) or c[i] > state["active_mid"]
    bearish_mid = (not config.require_midpoint_alignment) or c[i] < state["active_mid"]
    bullish_ema = (not config.require_ema_alignment) or (fast_ema[i] > slow_ema[i] and c[i] > fast_ema[i])
    bearish_ema = (not config.require_ema_alignment) or (fast_ema[i] < slow_ema[i] and c[i] < fast_ema[i])
    bullish_score = int(bullish_ob_retest) + int(bullish_fvg_retest) + int(bullish_bos_close) + int(bullish_mid) + int(bullish_ema)
    bearish_score = int(bearish_ob_retest) + int(bearish_fvg_retest) + int(bearish_bos_close) + int(bearish_mid) + int(bearish_ema)
    if bullish_zone_accepted or bearish_zone_accepted:
        state["first_retest_consumed"] = True

    if state["setup_direction"] == 1 and bullish_zone_accepted and bullish_confirm and bullish_bos_close and bullish_mid and bullish_ema and bullish_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if bullish_ob_retest else ("FVG" if bullish_fvg_retest else "BOS")
        return {"direction": 1, "signal": "BUY", "component": component, "model": "TBS REVERSIÓN", "potential": bullish_score}
    if state["setup_direction"] == -1 and bearish_zone_accepted and bearish_confirm and bearish_bos_close and bearish_mid and bearish_ema and bearish_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if bearish_ob_retest else ("FVG" if bearish_fvg_retest else "BOS")
        return {"direction": -1, "signal": "SELL", "component": component, "model": "TBS REVERSIÓN", "potential": bearish_score}
    return None


def breakout_retest_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    valid = state["setup_state"] == 8 and state["breakout_bar"] is not None and i > state["breakout_bar"] and i - state["breakout_bar"] <= config.breakout_retest_bars and not state["breakout_retest_consumed"]
    if not valid:
        return None
    tolerance = current_atr * config.retest_tolerance_atr
    bullish_touch = state["setup_direction"] == 1 and l[i] <= state["breakout_level"] + tolerance and h[i] >= state["breakout_level"] - tolerance
    bearish_touch = state["setup_direction"] == -1 and h[i] >= state["breakout_level"] - tolerance and l[i] <= state["breakout_level"] + tolerance
    bullish_confirm = bullish_touch and c[i] > state["breakout_level"] and c[i] > o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] > c[i - 1]
    bearish_confirm = bearish_touch and c[i] < state["breakout_level"] and c[i] < o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] < c[i - 1]
    bullish_ema = (not config.require_ema_alignment) or (fast_ema[i] > slow_ema[i] and c[i] > fast_ema[i])
    bearish_ema = (not config.require_ema_alignment) or (fast_ema[i] < slow_ema[i] and c[i] < fast_ema[i])
    if bullish_touch or bearish_touch:
        state["breakout_retest_consumed"] = True
    if state["setup_direction"] == 1 and c[i] < state["active_high"] - current_atr * config.breakout_invalidation_atr:
        state["setup_state"] = 7
        return None
    if state["setup_direction"] == -1 and c[i] > state["active_low"] + current_atr * config.breakout_invalidation_atr:
        state["setup_state"] = 7
        return None
    bullish_score = int(state["breakout_had_fvg"]) + int(c[i] > state["active_high"]) + int(bullish_ema) + int(body_ratio[i] * 100.0 >= config.entry_body_percent) + 1
    bearish_score = int(state["breakout_had_fvg"]) + int(c[i] < state["active_low"]) + int(bearish_ema) + int(body_ratio[i] * 100.0 >= config.entry_body_percent) + 1
    if state["setup_direction"] == 1 and bullish_confirm and bullish_ema and (not config.require_breakout_fvg or state["breakout_had_fvg"]) and bullish_score >= config.minimum_potential_score:
        return {"direction": 1, "signal": "BUY", "component": "CRT HIGH RETEST", "model": "BREAKOUT RETEST", "potential": bullish_score}
    if state["setup_direction"] == -1 and bearish_confirm and bearish_ema and (not config.require_breakout_fvg or state["breakout_had_fvg"]) and bearish_score >= config.minimum_potential_score:
        return {"direction": -1, "signal": "SELL", "component": "CRT LOW RETEST", "model": "BREAKOUT RETEST", "potential": bearish_score}
    return None


def rejection_signals(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> tuple[bool, bool, bool, bool, int, int]:
    candle_range = max(h[i] - l[i], 1e-12)
    lower_wick_pct = (min(o[i], c[i]) - l[i]) / candle_range * 100.0
    upper_wick_pct = (h[i] - max(o[i], c[i])) / candle_range * 100.0
    bullish_candle = c[i] > o[i] and lower_wick_pct >= config.minimum_rejection_wick_percent and c[i] > state.get("active_low", np.nan) and c[i] > c[i - 1]
    bearish_candle = c[i] < o[i] and upper_wick_pct >= config.minimum_rejection_wick_percent and c[i] < state.get("active_high", np.nan) and c[i] < c[i - 1]
    valid_bullish = config.enable_sweep_rejection_entry and state["setup_state"] == 2 and state["setup_direction"] == 1 and state["sweep_bar"] is not None and i >= state["sweep_bar"] and i - state["sweep_bar"] <= config.sweep_rejection_bars
    valid_bearish = config.enable_sweep_rejection_entry and state["setup_state"] == 2 and state["setup_direction"] == -1 and state["sweep_bar"] is not None and i >= state["sweep_bar"] and i - state["sweep_bar"] <= config.sweep_rejection_bars
    bullish_score = int(c[i] > state.get("active_low", np.nan)) + int(lower_wick_pct >= config.minimum_rejection_wick_percent) + int(c[i] > o[i]) + int(c[i] > c[i - 1]) + int((not config.require_ema_alignment) or fast_ema[i] > slow_ema[i])
    bearish_score = int(c[i] < state.get("active_high", np.nan)) + int(upper_wick_pct >= config.minimum_rejection_wick_percent) + int(c[i] < o[i]) + int(c[i] < c[i - 1]) + int((not config.require_ema_alignment) or fast_ema[i] < slow_ema[i])
    confirmed_bullish = valid_bullish and bullish_candle and bullish_score >= config.minimum_potential_score
    confirmed_bearish = valid_bearish and bearish_candle and bearish_score >= config.minimum_potential_score
    return bullish_candle, bearish_candle, confirmed_bullish, confirmed_bearish, bullish_score, bearish_score


def zone_confluence_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig, bullish_rejection_candle: bool, bearish_rejection_candle: bool) -> dict[str, Any] | None:
    if not config.enable_high_potential_zones or state["setup_state"] not in {1, 2, 3, 4}:
        return None
    tolerance = current_atr * config.important_zone_tolerance_atr
    price_at_bull_ob = state["bullish_ob_created"] and l[i] <= state["bullish_ob_high"] + tolerance and h[i] >= state["bullish_ob_low"] - tolerance and c[i] >= state["bullish_ob_low"]
    price_at_bear_ob = state["bearish_ob_created"] and h[i] >= state["bearish_ob_low"] - tolerance and l[i] <= state["bearish_ob_high"] + tolerance and c[i] <= state["bearish_ob_high"]
    price_at_bull_fvg = state["bullish_fvg_creation_bar"] is not None and l[i] <= state["bullish_fvg_top"] + tolerance and h[i] >= state["bullish_fvg_bottom"] - tolerance
    price_at_bear_fvg = state["bearish_fvg_creation_bar"] is not None and h[i] >= state["bearish_fvg_bottom"] - tolerance and l[i] <= state["bearish_fvg_top"] + tolerance
    price_at_low_liq = np.isfinite(state["active_low"]) and l[i] <= state["active_low"] + tolerance and c[i] > state["active_low"]
    price_at_high_liq = np.isfinite(state["active_high"]) and h[i] >= state["active_high"] - tolerance and c[i] < state["active_high"]
    bull_score = int(price_at_bull_ob) + int(price_at_bull_fvg) + int(price_at_low_liq) + int(bullish_rejection_candle) + int((not config.require_ema_alignment) or fast_ema[i] > slow_ema[i])
    bear_score = int(price_at_bear_ob) + int(price_at_bear_fvg) + int(price_at_high_liq) + int(bearish_rejection_candle) + int((not config.require_ema_alignment) or fast_ema[i] < slow_ema[i])
    if (price_at_bull_ob or price_at_bull_fvg or price_at_low_liq) and bullish_rejection_candle and bull_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if price_at_bull_ob else ("FVG" if price_at_bull_fvg else "LIQUIDEZ CRT")
        return {"direction": 1, "signal": "BUY", "component": component, "model": "ZONE CONFLUENCE", "potential": bull_score}
    if (price_at_bear_ob or price_at_bear_fvg or price_at_high_liq) and bearish_rejection_candle and bear_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if price_at_bear_ob else ("FVG" if price_at_bear_fvg else "LIQUIDEZ CRT")
        return {"direction": -1, "signal": "SELL", "component": component, "model": "ZONE CONFLUENCE", "potential": bear_score}
    return None


def invalidate_order_block(i: int, c: np.ndarray, state: dict[str, Any], events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] < 3 or state["setup_state"] > 4:
        return
    bullish_broken = (
        state["setup_direction"] == 1
        and state["bullish_ob_created"]
        and np.isfinite(state["bullish_ob_low"])
        and c[i] < state["bullish_ob_low"]
    )
    bearish_broken = (
        state["setup_direction"] == -1
        and state["bearish_ob_created"]
        and np.isfinite(state["bearish_ob_high"])
        and c[i] > state["bearish_ob_high"]
    )
    if bullish_broken or bearish_broken:
        state["setup_state"] = 7
        state["status"] = "INVALIDADO"
        add_event(events, data, i, state, "INVALIDATED", "order_block_broken")


def create_order_block(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> None:
    if not config.show_order_block or state["sweep_bar"] is None:
        return
    max_search = min(config.order_block_lookback, i - state["sweep_bar"])
    for offset in range(1, max_search + 1):
        idx = i - offset
        if state["setup_direction"] == 1 and c[idx] < o[idx]:
            state["bullish_ob_high"] = h[idx]
            state["bullish_ob_low"] = l[idx]
            state["bullish_ob_creation_bar"] = idx
            state["bullish_ob_created"] = True
            return
        if state["setup_direction"] == -1 and c[idx] > o[idx]:
            state["bearish_ob_high"] = h[idx]
            state["bearish_ob_low"] = l[idx]
            state["bearish_ob_creation_bar"] = idx
            state["bearish_ob_created"] = True
            return


def update_important_fvg(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, atr: np.ndarray, body: np.ndarray, body_ratio: np.ndarray, raw_bull: bool, raw_bear: bool, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if i < 2 or not config.show_important_fvg or state["important_fvg_count"] >= config.maximum_important_fvg:
        return
    center_body = abs(c[i - 1] - o[i - 1])
    center_range = max(h[i - 1] - l[i - 1], 1e-12)
    center_body_ratio = center_body / center_range
    displacement = center_body >= atr[i - 1] * config.displacement_body_atr and center_body_ratio * 100.0 >= config.displacement_body_percent
    state_ok = state["setup_state"] in {2, 3, 4, 8}
    impulse_started = (state["sweep_bar"] is not None and i > state["sweep_bar"]) or (state["breakout_bar"] is not None and i >= state["breakout_bar"])
    if state["setup_direction"] == 1 and state_ok and impulse_started and raw_bull and displacement:
        mid = (l[i] + h[i - 2]) / 2.0
        if abs(c[i] - mid) <= atr[i] * 2.0:
            state["bullish_fvg_top"] = l[i]
            state["bullish_fvg_bottom"] = h[i - 2]
            state["bullish_fvg_creation_bar"] = i
            state["important_fvg_count"] += 1
            add_event(events, data, i, state, "FVG", "bullish", price=mid)
    if state["setup_direction"] == -1 and state_ok and impulse_started and raw_bear and displacement:
        mid = (l[i - 2] + h[i]) / 2.0
        if abs(c[i] - mid) <= atr[i] * 2.0:
            state["bearish_fvg_top"] = l[i - 2]
            state["bearish_fvg_bottom"] = h[i]
            state["bearish_fvg_creation_bar"] = i
            state["important_fvg_count"] += 1
            add_event(events, data, i, state, "FVG", "bearish", price=mid)


def create_secondary_crt_allowed(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    candle_range: np.ndarray,
    body_ratio: np.ndarray,
    close_location: np.ndarray,
    range_to_atr: np.ndarray,
    inside_secondary: np.ndarray,
    inside_main: np.ndarray,
    main_session_finished: bool,
    state: dict[str, Any],
    crts_today: int,
    main_crt_created_today: bool,
    last_crt_creation_bar: int | None,
    config: CRTTBSConfig,
) -> bool:
    secondary_bull_quality = c[i] > o[i] and close_location[i] >= config.secondary_close_extreme
    secondary_bear_quality = c[i] < o[i] and close_location[i] <= 1.0 - config.secondary_close_extreme
    secondary_expansion = (not config.require_secondary_expansion) or h[i] > h[i - 1] or l[i] < l[i - 1]
    secondary_range_expansion = (not config.require_secondary_range_expansion) or candle_range[i] > candle_range[i - 1]
    valid_secondary = inside_secondary[i] and range_to_atr[i] >= config.secondary_range_atr and body_ratio[i] * 100.0 >= config.secondary_body_percent and (secondary_bull_quality or secondary_bear_quality) and secondary_expansion and secondary_range_expansion
    previous_finished = state["setup_state"] in {0, 5, 6, 7}
    cooldown = last_crt_creation_bar is None or i - last_crt_creation_bar >= config.secondary_cooldown_bars
    time_allowed = (not config.secondary_only_after_main) or main_crt_created_today
    return valid_secondary and time_allowed and not inside_main[i] and previous_finished and cooldown and crts_today < config.maximum_crts_per_day and not main_session_finished


def entry_record(data: pd.DataFrame, i: int, state: dict[str, Any], confirmation: dict[str, Any], c: np.ndarray, atr: np.ndarray, config: CRTTBSConfig, trade_id: int) -> dict[str, Any]:
    direction = int(confirmation["direction"])
    entry = float(c[i])
    structural_stop = state["active_low"] if direction == 1 else state["active_high"]
    if np.isfinite(state.get("last_sweep_price", np.nan)):
        structural_stop = min(structural_stop, state["last_sweep_price"]) if direction == 1 else max(structural_stop, state["last_sweep_price"])
    if confirmation["component"] == "ORDER BLOCK":
        if direction == 1 and np.isfinite(state["bullish_ob_low"]):
            structural_stop = state["bullish_ob_low"]
        elif direction == -1 and np.isfinite(state["bearish_ob_high"]):
            structural_stop = state["bearish_ob_high"]
    stop = structural_stop - atr[i] * config.result_stop_buffer_atr if direction == 1 else structural_stop + atr[i] * config.result_stop_buffer_atr
    risk = abs(entry - stop)
    target = entry + risk * config.result_risk_reward if direction == 1 else entry - risk * config.result_risk_reward
    if config.result_target_mode.lower().strip() == "crt" and np.isfinite(state.get("active_target", np.nan)):
        crt_target = float(state["active_target"])
        if (direction == 1 and crt_target > entry) or (direction == -1 and crt_target < entry):
            target = crt_target
    return {
        "TradeId": trade_id,
        "CRTNumber": state["active_number"],
        "CRTType": state["active_type"],
        "Signal": confirmation["signal"],
        "Direction": direction,
        "Model": confirmation["model"],
        "Component": confirmation["component"],
        "Potential": confirmation["potential"],
        "EntryBarIdx": i,
        "EntryDate": data.index[i],
        "EntryPrice": entry,
        "StopPrice": float(stop),
        "TargetPrice": float(target),
        "Result": "ACTIVE",
        "ResultBarIdx": np.nan,
        "ResultDate": pd.NaT,
        "ActiveCRTHigh": state["active_high"],
        "ActiveCRTLow": state["active_low"],
        "TBSBarIdx": state["sweep_bar"],
        "ChoChBarIdx": state["choch_bar"],
        "BOSBarIdx": state["bos_bar"],
        "BreakoutBarIdx": state["breakout_bar"],
        "HTFTimeframe": state.get("active_timeframe", ""),
        "ActiveTargetPrice": state.get("active_target", np.nan),
        "TargetZoneLow": state.get("target_zone_low", np.nan),
        "TargetZoneHigh": state.get("target_zone_high", np.nan),
        "InternalCRTSeen": bool(state.get("internal_crt_seen", False)),
        "InternalCRTBarIdx": state.get("internal_crt_bar", None),
    }


def update_trades(i: int, h: np.ndarray, l: np.ndarray, data: pd.DataFrame, active_trades: list[dict[str, Any]], entries: list[dict[str, Any]]) -> None:
    remaining = []
    for trade in active_trades:
        if i <= trade["EntryBarIdx"]:
            remaining.append(trade)
            continue
        direction = int(trade["Direction"])
        target_hit = h[i] >= trade["TargetPrice"] if direction == 1 else l[i] <= trade["TargetPrice"]
        stop_hit = l[i] <= trade["StopPrice"] if direction == 1 else h[i] >= trade["StopPrice"]
        if target_hit or stop_hit:
            result = "WIN" if target_hit and not stop_hit else "FAILD"
            trade["Result"] = result
            trade["ResultBarIdx"] = i
            trade["ResultDate"] = data.index[i]
            for row in entries:
                if row["TradeId"] == trade["TradeId"]:
                    row.update({"Result": result, "ResultBarIdx": i, "ResultDate": data.index[i]})
                    break
        else:
            remaining.append(trade)
    active_trades[:] = remaining


def add_event(events: list[dict[str, Any]], data: pd.DataFrame, i: int, state: dict[str, Any], event: str, detail: str, *, price: float | None = None, level: float | None = None) -> None:
    events.append(
        {
            "BarIdx": i,
            "Date": data.index[i],
            "Event": event,
            "Detail": detail,
            "CRTNumber": state.get("active_number", np.nan),
            "CRTType": state.get("active_type", ""),
            "Direction": state.get("setup_direction", 0),
            "Price": np.nan if price is None else float(price),
            "Level": np.nan if level is None or not np.isfinite(level) else float(level),
            "State": state.get("setup_state", 0),
        }
    )


def crt_record(data: pd.DataFrame, state: dict[str, Any]) -> dict[str, Any]:
    return {
        "CRTNumber": state["active_number"],
        "CRTType": state["active_type"],
        "Timeframe": state.get("active_timeframe", ""),
        "Direction": state.get("allowed_direction", 0),
        "StartBarIdx": state["active_start_bar"],
        "ActivationBarIdx": state["active_activation_bar"],
        "StartDate": data.index[state["active_start_bar"]],
        "ActivationDate": data.index[state["active_activation_bar"]],
        "CRTHigh": state["active_high"],
        "CRTLow": state["active_low"],
        "CRTMid": state["active_mid"],
        "TargetPrice": state.get("active_target", np.nan),
        "TargetZoneLow": state.get("target_zone_low", np.nan),
        "TargetZoneHigh": state.get("target_zone_high", np.nan),
        "IsMTF": bool(state.get("is_mtf_crt", False)),
    }


def confirmed_pivots(i: int, high: np.ndarray, low: np.ndarray, *, left: int, right: int) -> tuple[float, float]:
    if i < left + right:
        return np.nan, np.nan
    pivot_idx = i - right
    start = pivot_idx - left
    end = pivot_idx + right + 1
    pivot_high = high[pivot_idx] if high[pivot_idx] >= finite_max(high[start:end]) else np.nan
    pivot_low = low[pivot_idx] if low[pivot_idx] <= finite_min(low[start:end]) else np.nan
    return pivot_high, pivot_low


def raw_fvg(i: int, high: np.ndarray, low: np.ndarray, atr: np.ndarray, config: CRTTBSConfig) -> tuple[bool, bool]:
    if i < 2:
        return False, False
    bullish = low[i] > high[i - 2] and low[i] - high[i - 2] >= atr[i] * config.minimum_fvg_atr
    bearish = high[i] < low[i - 2] and low[i - 2] - high[i] >= atr[i] * config.minimum_fvg_atr
    return bool(bullish), bool(bearish)


def index_in_timezone(index: pd.DatetimeIndex, config: CRTTBSConfig) -> pd.DatetimeIndex:
    idx = pd.DatetimeIndex(index)
    if idx.tz is None:
        idx = idx.tz_localize(ZoneInfo(config.data_timezone), nonexistent="shift_forward", ambiguous="NaT")
    return idx.tz_convert(ZoneInfo(config.timezone))


def in_session(ts: pd.Timestamp, session: str) -> bool:
    if pd.isna(ts):
        return False
    start_text, end_text = session.split("-")
    start = int(start_text[:2]) * 60 + int(start_text[2:])
    end = int(end_text[:2]) * 60 + int(end_text[2:])
    now = int(ts.hour) * 60 + int(ts.minute)
    if start <= end:
        return start <= now < end
    return now >= start or now < end


def date_key(ts: pd.Timestamp) -> int:
    if pd.isna(ts):
        return -1
    return int(ts.year) * 10000 + int(ts.month) * 100 + int(ts.day)


def safe_atr_value(value: float) -> float:
    return float(value) if np.isfinite(value) and value > 0 else 1e-12


def finite_min(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.min(arr)) if len(arr) else np.nan


def finite_max(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.max(arr)) if len(arr) else np.nan


def empty_entries() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "TradeId",
            "CRTNumber",
            "CRTType",
            "Signal",
            "Direction",
            "Model",
            "Component",
            "Potential",
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "StopPrice",
            "TargetPrice",
            "Result",
            "ResultBarIdx",
            "ResultDate",
            "ActiveCRTHigh",
            "ActiveCRTLow",
            "TBSBarIdx",
            "ChoChBarIdx",
            "BOSBarIdx",
            "BreakoutBarIdx",
            "HTFTimeframe",
            "ActiveTargetPrice",
            "TargetZoneLow",
            "TargetZoneHigh",
            "InternalCRTSeen",
            "InternalCRTBarIdx",
        ]
    )


def run_vectorbt_backtest(
    data: pd.DataFrame,
    result: dict[str, pd.DataFrame],
    config: CRTTBSConfig,
    *,
    show_plot: bool,
) -> dict[str, Any]:
    signals = build_vectorbt_signals(data, result["entries"], config)
    orders = build_vectorbt_order_frame(data, result["entries"], config)
    if vbt is None:
        detail = "" if VBT_IMPORT_ERROR is None else f" ({type(VBT_IMPORT_ERROR).__name__}: {VBT_IMPORT_ERROR})"
        message = f"Install/fix vectorbt to run the portfolio backtest: pip install vectorbt{detail}"
        if config.backtest_raise_missing_vectorbt:
            raise ImportError(message)
        warnings.warn(message)
        return {"signals": signals, "orders": orders, "portfolio": None, "stats": pd.Series(dtype=object), "fig": None}

    engine = config.backtest_engine.lower().strip()
    if engine == "orders":
        portfolio = vectorbt_from_orders(data, orders, config)
    elif engine == "signals":
        portfolio = vectorbt_from_signals(data, signals, config)
    else:
        raise ValueError("backtest_engine must be 'orders' or 'signals'.")

    stats = portfolio_stats(portfolio)
    fig = None
    if show_plot:
        fig = portfolio_plot(portfolio)
        if fig is not None:
            display_figure(fig)
    return {"signals": signals, "orders": orders, "portfolio": portfolio, "stats": stats, "fig": fig}


def empty_backtest_result(data: pd.DataFrame, result: dict[str, pd.DataFrame], config: CRTTBSConfig) -> dict[str, Any]:
    signals = empty_vectorbt_signals(data)
    orders = build_vectorbt_order_frame(data, result["entries"], config) if "entries" in result else empty_order_frame()
    return {"signals": signals, "orders": orders, "portfolio": None, "stats": pd.Series(dtype=object), "fig": None}


def build_vectorbt_signals(data: pd.DataFrame, entries: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    signals = empty_vectorbt_signals(data)
    if entries.empty:
        return signals

    for _, row in entries.iterrows():
        raw_entry_idx = int(row["EntryBarIdx"])
        entry_idx = raw_entry_idx + int(config.backtest_shift_signals)
        if entry_idx >= len(data):
            continue
        direction = int(row["Direction"])
        entry_price = float(data[config.backtest_price_column].iloc[entry_idx])
        stop_price = float(row["StopPrice"])
        target_price = float(row["TargetPrice"])
        if direction == 1:
            signals.iat[entry_idx, signals.columns.get_loc("long_entries")] = True
            signals.iat[entry_idx, signals.columns.get_loc("sl_stop")] = max((entry_price - stop_price) / entry_price, 0.0)
            signals.iat[entry_idx, signals.columns.get_loc("tp_stop")] = max((target_price - entry_price) / entry_price, 0.0)
        else:
            signals.iat[entry_idx, signals.columns.get_loc("short_entries")] = True
            signals.iat[entry_idx, signals.columns.get_loc("sl_stop")] = max((stop_price - entry_price) / entry_price, 0.0)
            signals.iat[entry_idx, signals.columns.get_loc("tp_stop")] = max((entry_price - target_price) / entry_price, 0.0)
        signals.iat[entry_idx, signals.columns.get_loc("trade_id")] = int(row["TradeId"])
        signals.iat[entry_idx, signals.columns.get_loc("entry_price")] = entry_price
        signals.iat[entry_idx, signals.columns.get_loc("stop_price")] = stop_price
        signals.iat[entry_idx, signals.columns.get_loc("target_price")] = target_price

    return signals


def empty_vectorbt_signals(data: pd.DataFrame) -> pd.DataFrame:
    return pd.DataFrame(
        {
            "long_entries": pd.Series(False, index=data.index, dtype=bool),
            "long_exits": pd.Series(False, index=data.index, dtype=bool),
            "short_entries": pd.Series(False, index=data.index, dtype=bool),
            "short_exits": pd.Series(False, index=data.index, dtype=bool),
            "sl_stop": pd.Series(np.nan, index=data.index, dtype=float),
            "tp_stop": pd.Series(np.nan, index=data.index, dtype=float),
            "trade_id": pd.Series(np.nan, index=data.index, dtype=float),
            "entry_price": pd.Series(np.nan, index=data.index, dtype=float),
            "stop_price": pd.Series(np.nan, index=data.index, dtype=float),
            "target_price": pd.Series(np.nan, index=data.index, dtype=float),
        }
    )


def build_vectorbt_order_frame(data: pd.DataFrame, entries: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    if entries.empty:
        return empty_order_frame()
    rows: list[dict[str, Any]] = []
    for _, entry in entries.iterrows():
        entry_idx = int(entry["EntryBarIdx"]) + int(config.backtest_shift_signals)
        if entry_idx >= len(data):
            continue
        direction = int(entry["Direction"])
        size = float(config.backtest_size) * direction
        rows.append(
            {
                "TradeId": int(entry["TradeId"]),
                "BarIdx": entry_idx,
                "Date": data.index[entry_idx],
                "Size": size,
                "Price": float(data[config.backtest_price_column].iloc[entry_idx]),
                "Reason": "ENTRY",
                "Signal": entry["Signal"],
                "Model": entry["Model"],
            }
        )
        if pd.notna(entry["ResultBarIdx"]):
            exit_idx = int(entry["ResultBarIdx"])
            if exit_idx < len(data) and exit_idx > entry_idx:
                exit_price = float(entry["TargetPrice"]) if entry["Result"] == "WIN" else float(entry["StopPrice"])
                rows.append(
                    {
                        "TradeId": int(entry["TradeId"]),
                        "BarIdx": exit_idx,
                        "Date": data.index[exit_idx],
                        "Size": -size,
                        "Price": exit_price,
                        "Reason": str(entry["Result"]),
                        "Signal": entry["Signal"],
                        "Model": entry["Model"],
                    }
                )
    if not rows:
        return empty_order_frame()
    return pd.DataFrame(rows).sort_values(["BarIdx", "TradeId", "Reason"]).reset_index(drop=True)


def empty_order_frame() -> pd.DataFrame:
    return pd.DataFrame(columns=["TradeId", "BarIdx", "Date", "Size", "Price", "Reason", "Signal", "Model"])


def vectorbt_from_orders(data: pd.DataFrame, orders: pd.DataFrame, config: CRTTBSConfig):
    order_size = pd.Series(np.nan, index=data.index, dtype=float)
    order_price = data[config.backtest_price_column].astype(float).copy()
    if not orders.empty:
        for bar_idx, group in orders.groupby("BarIdx", sort=True):
            bar_idx = int(bar_idx)
            size_sum = float(group["Size"].sum())
            if size_sum == 0.0:
                continue
            weights = group["Size"].abs()
            order_size.iloc[bar_idx] = size_sum
            order_price.iloc[bar_idx] = float((group["Price"] * weights).sum() / weights.sum())
    return vbt.Portfolio.from_orders(
        close=data["close"].astype(float),
        size=order_size,
        price=order_price,
        size_type=config.backtest_size_type,
        direction="both",
        fees=config.backtest_fees,
        fixed_fees=config.backtest_fixed_fees,
        slippage=config.backtest_slippage,
        init_cash=config.backtest_init_cash,
        freq=vectorbt_freq(data, config),
    )


def vectorbt_from_signals(data: pd.DataFrame, signals: pd.DataFrame, config: CRTTBSConfig):
    price = data[config.backtest_price_column].astype(float)
    return vbt.Portfolio.from_signals(
        close=data["close"].astype(float),
        entries=signals["long_entries"],
        exits=signals["long_exits"],
        short_entries=signals["short_entries"],
        short_exits=signals["short_exits"],
        size=config.backtest_size,
        size_type=config.backtest_size_type,
        price=price,
        fees=config.backtest_fees,
        fixed_fees=config.backtest_fixed_fees,
        slippage=config.backtest_slippage,
        open=data["open"].astype(float),
        high=data["high"].astype(float),
        low=data["low"].astype(float),
        sl_stop=signals["sl_stop"],
        tp_stop=signals["tp_stop"],
        sl_trail=False,
        upon_opposite_entry=config.backtest_upon_opposite_entry,
        accumulate=config.backtest_accumulate,
        init_cash=config.backtest_init_cash,
        freq=vectorbt_freq(data, config),
    )


def vectorbt_freq(data: pd.DataFrame, config: CRTTBSConfig) -> str | None:
    if config.backtest_freq is not None:
        return config.backtest_freq
    try:
        return pd.infer_freq(data.index)
    except (TypeError, ValueError):
        return None


def portfolio_stats(portfolio: Any) -> pd.Series:
    try:
        return portfolio.stats()
    except Exception as exc:
        warnings.warn(f"vectorbt stats failed: {exc}")
        return pd.Series(dtype=object)


def portfolio_plot(portfolio: Any) -> Any:
    try:
        return portfolio.plot()
    except Exception as exc:
        warnings.warn(f"vectorbt plot failed: {exc}")
        return None


def plot_crt_tbs_state_machine(data: pd.DataFrame, result: dict[str, pd.DataFrame], *, title: str):
    if go is None:
        warnings.warn("plotly is not installed; skipping plot.")
        return None
    x = np.arange(len(data))
    labels = data.index.strftime("%Y-%m-%d %H:%M")
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=labels,
            hovertext=labels,
            hoverinfo="text+y",
        )
    )
    for _, row in result["crts"].iterrows():
        start = int(row["StartBarIdx"])
        end = min(len(data) - 1, int(row["ActivationBarIdx"]) + 80)
        is_mtf = bool(row.get("IsMTF", False))
        line_color = "#38BDF8" if is_mtf else "#E5E7EB"
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTHigh"], y1=row["CRTHigh"], line=dict(color=line_color, width=2 if is_mtf else 1))
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTLow"], y1=row["CRTLow"], line=dict(color=line_color, width=2 if is_mtf else 1))
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTMid"], y1=row["CRTMid"], line=dict(color="#94A3B8", width=1, dash="dot"))
        if np.isfinite(row.get("TargetZoneLow", np.nan)) and np.isfinite(row.get("TargetZoneHigh", np.nan)):
            fig.add_shape(
                type="rect",
                x0=int(row["ActivationBarIdx"]),
                x1=end,
                y0=float(row["TargetZoneLow"]),
                y1=float(row["TargetZoneHigh"]),
                fillcolor="rgba(250, 204, 21, 0.16)",
                line=dict(color="#FACC15", width=1, dash="dot"),
            )
            fig.add_annotation(
                x=end,
                y=float(row["TargetPrice"]),
                text=f"{row.get('Timeframe', '')} Target",
                showarrow=False,
                font=dict(color="#FACC15", size=10),
                xanchor="left",
            )
    event_colors = {"HTF_CRT": "#38BDF8", "INTERNAL_CRT": "#FACC15", "TBS": "#FACC15", "CHoCH": "#38BDF8", "BOS": "#22C55E", "BREAKOUT": "#F97316", "ENTRY": "#A78BFA", "INVALIDATED": "#EF4444", "EXPIRED": "#64748B", "FVG": "#CBD5E1", "CRT": "#E5E7EB"}
    if not result["events"].empty:
        for event_name, group in result["events"].groupby("Event"):
            prices = []
            for _, row in group.iterrows():
                if np.isfinite(row.get("Price", np.nan)):
                    prices.append(float(row["Price"]))
                elif np.isfinite(row.get("Level", np.nan)):
                    prices.append(float(row["Level"]))
                else:
                    prices.append(float(data["close"].iloc[int(row["BarIdx"])]))
            fig.add_trace(
                go.Scatter(
                    x=group["BarIdx"],
                    y=prices,
                    mode="markers+text",
                    name=event_name,
                    marker=dict(color=event_colors.get(event_name, "#E5E7EB"), size=9),
                    text=group["Event"],
                    textposition="top center",
                    hovertext=[
                        f"{row['Event']} {row['Detail']}<br>{row['Date']}<br>CRT #{row['CRTNumber']}<br>Level: {row['Level']}"
                        for _, row in group.iterrows()
                    ],
                    hoverinfo="text",
                )
            )
    if not result["entries"].empty:
        fig.add_trace(
            go.Scatter(
                x=result["entries"]["EntryBarIdx"],
                y=result["entries"]["EntryPrice"],
                mode="markers+text",
                name="Entries",
                marker=dict(color="#A78BFA", size=13, symbol="star"),
                text=result["entries"]["Signal"],
                textposition="bottom center",
                hovertext=[
                    (
                        f"{row['Signal']}<br>{row['Model']} / {row['Component']}<br>"
                        f"Entry: {row['EntryPrice']:.4f}<br>Stop: {row['StopPrice']:.4f}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>Result: {row['Result']}"
                    )
                    for _, row in result["entries"].iterrows()
                ],
                hoverinfo="text",
            )
        )
    ticks = np.linspace(0, len(data) - 1, min(12, len(data)), dtype=int)
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", y=1.02, x=1, xanchor="right"),
        margin=dict(l=55, r=40, t=80, b=60),
    )
    fig.update_xaxes(tickmode="array", tickvals=ticks, ticktext=labels[ticks], tickangle=-30)
    return fig


def display_figure(fig: Any) -> None:
    try:
        from IPython.display import display
    except ModuleNotFoundError:
        return
    display(fig)
```

```python
from __future__ import annotations

from dataclasses import dataclass
from pathlib import Path
from typing import Any
from zoneinfo import ZoneInfo
import warnings

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ModuleNotFoundError:
    Ticker = None

try:
    import plotly.graph_objects as go
except ModuleNotFoundError:
    go = None

try:
    import vectorbt as vbt
    VBT_IMPORT_ERROR = None
except Exception as exc:
    vbt = None
    VBT_IMPORT_ERROR = exc


@dataclass
class CRTTBSConfig:
    symbol: str = "THYAO.IS"
    period: str = "60d"
    interval: str = "5m"
    data_timezone: str = "Europe/Istanbul"
    timezone: str = "America/New_York"

    main_session: str = "0830-0945"
    secondary_session: str = "0945-1600"
    search_24x7: bool = True
    maximum_crts_per_day: int = 4
    secondary_cooldown_bars: int = 20
    setup_expiration_bars: int = 120

    atr_length: int = 14
    secondary_range_atr: float = 1.50
    secondary_body_percent: float = 65.0
    secondary_close_extreme: float = 0.80
    require_secondary_expansion: bool = True
    secondary_only_after_main: bool = True
    require_secondary_range_expansion: bool = True

    swing_left: int = 3
    swing_right: int = 3
    structure_lookback: int = 20
    sweep_ticks: float = 1.0
    tick_size: float = 0.01
    require_close_inside: bool = True
    bos_break_atr: float = 0.05
    max_bars_to_choch: int = 20
    invalidation_close_atr: float = 0.10

    enable_breakout_retest: bool = True
    breakout_body_atr: float = 0.70
    breakout_body_percent: float = 65.0
    breakout_close_atr: float = 0.10
    breakout_retest_bars: int = 20
    require_breakout_fvg: bool = False
    breakout_invalidation_atr: float = 0.15

    show_important_fvg: bool = True
    minimum_fvg_atr: float = 0.15
    displacement_body_atr: float = 0.70
    displacement_body_percent: float = 65.0
    maximum_important_fvg: int = 1

    show_order_block: bool = True
    order_block_lookback: int = 12
    minimum_choch_displacement_atr: float = 0.50

    entry_body_percent: float = 45.0
    entry_confirmation_bars: int = 15
    retest_tolerance_atr: float = 0.20
    accept_bos_retest: bool = True
    accept_fvg_retest: bool = True
    accept_order_block_retest: bool = True
    require_fvg_or_ob: bool = True
    require_close_beyond_bos: bool = True
    require_midpoint_alignment: bool = True
    require_ema_alignment: bool = False
    fast_ema_length: int = 20
    slow_ema_length: int = 50
    minimum_potential_score: int = 3

    enable_sweep_rejection_entry: bool = True
    sweep_rejection_bars: int = 4
    minimum_rejection_wick_percent: float = 35.0

    enable_high_potential_zones: bool = True
    important_zone_tolerance_atr: float = 0.15

    result_risk_reward: float = 1.5
    result_stop_buffer_atr: float = 0.10
    result_target_mode: str = "crt"

    use_mtf_crt: bool = True
    htf_intervals: tuple[str, ...] | str = ("1h",)
    htf_fetch_mode: str = "yahoo"
    htf_period: str | None = None
    htf_csv_path: str | None = None
    htf_resample_fallback: bool = False
    htf_min_range_atr: float = 0.0
    htf_target_zone_fraction: float = 0.05
    htf_allow_both_side_sweep: bool = False
    mtf_entry_mode: str = "tbs_touch"
    use_wick_touch: bool = True
    use_body_touch: bool = True
    require_internal_crt_confirmation: bool = False
    internal_crt_enabled: bool = True
    internal_crt_inside_parent: bool = True

    backtest_engine: str = "orders"
    backtest_price_column: str = "close"
    backtest_init_cash: float = 100_000.0
    backtest_size: float = 1.0
    backtest_size_type: str = "amount"
    backtest_fees: float = 0.0
    backtest_fixed_fees: float = 0.0
    backtest_slippage: float = 0.0
    backtest_shift_signals: bool = False
    backtest_upon_opposite_entry: str = "close"
    backtest_accumulate: bool = False
    backtest_freq: str | None = None
    backtest_raise_missing_vectorbt: bool = False


def run_crt_tbs(
    *,
    csv_path: str | None = None,
    symbol: str = "THYAO.IS",
    period: str = "60d",
    interval: str = "5m",
    show_plot: bool = True,
    run_backtest: bool = True,
    show_backtest_plot: bool = True,
    save_csv_path: str | None = None,
    **overrides: Any,
) -> dict[str, Any]:
    config = CRTTBSConfig(symbol=symbol, period=period, interval=interval, **overrides)
    data = load_ohlcv(csv_path=csv_path, config=config)
    data = add_indicators(data, config)
    result = run_mtf_crt_tbs_state_machine(data, config) if config.use_mtf_crt else run_pine_state_machine(data, config)
    backtest = run_vectorbt_backtest(
        data,
        result,
        config,
        show_plot=show_plot and show_backtest_plot,
    ) if run_backtest else empty_backtest_result(data, result, config)

    if save_csv_path is not None:
        Path(save_csv_path).parent.mkdir(parents=True, exist_ok=True)
        result["entries"].to_csv(save_csv_path, index=False)

    mode_text = f"MTF CRT {','.join(map(str, configured_htf_intervals(config)))} -> {config.interval} TBS" if config.use_mtf_crt else "CRT TBS Pine-style"
    fig = plot_crt_tbs_state_machine(data, result, title=f"{config.symbol} {mode_text}")
    if show_plot and fig is not None:
        display_figure(fig)

    print(mode_text)
    print(
        f"bars: {len(data)} | HTF CRTs: {len(result.get('htf_crts', []))} | "
        f"active CRTs: {len(result['crts'])} | internal CRTs: {len(result.get('internal_crts', []))} | "
        f"events: {len(result['events'])} | entries: {len(result['entries'])}"
    )
    if len(result["entries"]):
        preview = [
            "Signal",
            "Model",
            "Component",
            "Potential",
            "EntryDate",
            "EntryPrice",
            "StopPrice",
            "TargetPrice",
            "Result",
        ]
        print(result["entries"][preview].head(30).to_string(index=False))
    if run_backtest:
        print("VectorBT backtest")
        print(
            f"engine: {config.backtest_engine} | "
            f"long entries: {int(backtest['signals']['long_entries'].sum())} | "
            f"short entries: {int(backtest['signals']['short_entries'].sum())} | "
            f"orders: {len(backtest['orders'])}"
        )
        if backtest["portfolio"] is None:
            print("portfolio: vectorbt is not installed in this Python environment")
        elif len(backtest["stats"]):
            preferred = [
                "Start",
                "End",
                "Period",
                "Start Value",
                "End Value",
                "Total Return [%]",
                "Benchmark Return [%]",
                "Max Drawdown [%]",
                "Total Trades",
                "Win Rate [%]",
                "Profit Factor",
                "Sharpe Ratio",
            ]
            stats = backtest["stats"]
            available = [name for name in preferred if name in stats.index]
            print(stats.loc[available].to_string())

    return {
        "config": config,
        "data": data,
        "fig": fig,
        **result,
        "setups": result["entries"],
        "backtest": backtest,
        "backtest_signals": backtest["signals"],
        "backtest_orders": backtest["orders"],
        "portfolio": backtest["portfolio"],
        "portfolio_stats": backtest["stats"],
        "backtest_fig": backtest["fig"],
    }


def load_ohlcv(*, csv_path: str | None, config: CRTTBSConfig) -> pd.DataFrame:
    if csv_path is not None:
        return prepare_ohlcv(pd.read_csv(csv_path))
    return fetch_yahoo_ohlcv(config.symbol, config.period, config.interval)


def fetch_yahoo_ohlcv(symbol: str, period: str, interval: str) -> pd.DataFrame:
    if Ticker is None:
        raise ImportError("Install yahooquery or pass csv_path.")

    raw = Ticker(symbol).history(period=period, interval=normalize_yahoo_interval(interval))
    if not isinstance(raw, pd.DataFrame) or raw.empty:
        raise ValueError(f"No data returned from Yahoo for {symbol} period={period} interval={interval}.")
    if "adjclose" in raw.columns and not {"open", "high", "low", "close"}.issubset(raw.columns):
        raise ValueError(f"Yahoo returned non-OHLC data for {symbol} period={period} interval={interval}.")
    if isinstance(raw, pd.DataFrame) and isinstance(raw.index, pd.MultiIndex):
        if "symbol" in raw.index.names:
            raw = raw.reset_index(level="symbol", drop=True)
        else:
            raw = raw.reset_index()
    return prepare_ohlcv(raw)


def prepare_ohlcv(df: pd.DataFrame) -> pd.DataFrame:
    data = df.copy()
    data = data.rename(columns={column: str(column).lower() for column in data.columns})
    if not isinstance(data.index, pd.DatetimeIndex):
        date_col = next(
            (col for col in ["datetime", "date", "timestamp", "time"] if col in data.columns),
            None,
        )
        if date_col is None:
            raise ValueError("Use DatetimeIndex or include datetime/date/timestamp column.")
        data[date_col] = pd.to_datetime(data[date_col], errors="coerce")
        data = data.set_index(date_col)

    required = {"open", "high", "low", "close"}
    missing = required.difference(data.columns)
    if missing:
        raise ValueError(f"Missing OHLC columns: {sorted(missing)}")

    data.index = pd.to_datetime(data.index)
    data = data[~data.index.duplicated(keep="last")].sort_index()
    if "volume" not in data.columns:
        data["volume"] = 0.0
    for col in ["open", "high", "low", "close", "volume"]:
        data[col] = pd.to_numeric(data[col], errors="coerce")
    data = data.dropna(subset=["open", "high", "low", "close"])
    data["volume"] = data["volume"].fillna(0.0).clip(lower=0.0)
    return data[["open", "high", "low", "close", "volume"]]


def add_indicators(data: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    out = data.copy()
    high = out["high"].astype(float)
    low = out["low"].astype(float)
    open_ = out["open"].astype(float)
    close = out["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat(
        [high - low, (high - previous_close).abs(), (low - previous_close).abs()],
        axis=1,
    ).max(axis=1)
    out["atr"] = true_range.rolling(config.atr_length, min_periods=1).mean()
    out["candle_range"] = high - low
    out["candle_body"] = (close - open_).abs()
    out["body_ratio"] = out["candle_body"] / out["candle_range"].replace(0, np.nan)
    out["body_ratio"] = out["body_ratio"].fillna(0.0)
    out["close_location"] = (close - low) / out["candle_range"].replace(0, np.nan)
    out["close_location"] = out["close_location"].fillna(0.5)
    out["range_to_atr"] = out["candle_range"] / out["atr"].replace(0, np.nan)
    out["fast_ema"] = close.ewm(span=config.fast_ema_length, adjust=False, min_periods=1).mean()
    out["slow_ema"] = close.ewm(span=config.slow_ema_length, adjust=False, min_periods=1).mean()
    return out.replace([np.inf, -np.inf], np.nan)


def run_mtf_crt_tbs_state_machine(data: pd.DataFrame, config: CRTTBSConfig) -> dict[str, pd.DataFrame]:
    o = data["open"].to_numpy(dtype=float)
    h = data["high"].to_numpy(dtype=float)
    l = data["low"].to_numpy(dtype=float)
    c = data["close"].to_numpy(dtype=float)
    atr = data["atr"].to_numpy(dtype=float)
    body = data["candle_body"].to_numpy(dtype=float)
    body_ratio = data["body_ratio"].to_numpy(dtype=float)
    fast_ema = data["fast_ema"].to_numpy(dtype=float)
    slow_ema = data["slow_ema"].to_numpy(dtype=float)

    htf_data = load_all_htf_data(data, config)
    htf_crts = detect_all_htf_crts(data, config, htf_data)
    activations: dict[int, list[dict[str, Any]]] = {}
    for _, setup in htf_crts.iterrows():
        activations.setdefault(int(setup["ActivationBarIdx"]), []).append(setup.to_dict())

    state = new_state()
    crts: list[dict[str, Any]] = []
    events: list[dict[str, Any]] = []
    entries: list[dict[str, Any]] = []
    internal_crts: list[dict[str, Any]] = []
    active_trades: list[dict[str, Any]] = []

    last_swing_high = np.nan
    last_swing_low = np.nan
    next_trade_id = 1

    for i in range(len(data)):
        pivot_high, pivot_low = confirmed_pivots(
            i=i,
            high=h,
            low=l,
            left=config.swing_left,
            right=config.swing_right,
        )
        if np.isfinite(pivot_high):
            last_swing_high = pivot_high
        if np.isfinite(pivot_low):
            last_swing_low = pivot_low

        if i in activations and state["setup_state"] in {0, 5, 6, 7}:
            setup = choose_htf_setup(activations[i])
            state = activate_crt(
                crt_high=float(setup["RangeHigh"]),
                crt_low=float(setup["RangeLow"]),
                crt_start_bar=int(setup["RangeStartBarIdx"]),
                activation_bar=i,
                crt_type=f"HTF-{setup['Timeframe']}",
                crt_number=len(crts) + 1,
                allowed_direction=int(setup["Direction"]),
                target_price=float(setup["TargetPrice"]),
                target_zone_low=float(setup["TargetZoneLow"]),
                target_zone_high=float(setup["TargetZoneHigh"]),
                timeframe=str(setup["Timeframe"]),
                is_mtf=True,
            )
            crts.append(crt_record(data, state))
            add_event(
                events,
                data,
                i,
                state,
                "HTF_CRT",
                str(setup["Signal"]),
                price=state["active_mid"],
                level=state["active_target"],
            )

        if state["setup_state"] >= 1 and state["setup_state"] <= 4:
            if i - state["active_activation_bar"] > config.setup_expiration_bars:
                add_event(events, data, i, state, "EXPIRED", "setup_expiration")
                state["setup_state"] = 6
                state["status"] = "EXPIRADO"

        internal = detect_internal_crt(i, o, h, l, c, state, config)
        if internal is not None:
            state["internal_crt_seen"] = True
            state["internal_crt_bar"] = i
            internal_crts.append(internal_crt_record(data, i, state, internal))
            add_event(
                events,
                data,
                i,
                state,
                "INTERNAL_CRT",
                internal["signal"],
                price=internal["price"],
                level=internal["target"],
            )

        direct_confirmation = evaluate_mtf_tbs_touch_entry(i, o, h, l, c, state, config)
        if direct_confirmation is not None:
            state["setup_state"] = 5
            add_event(events, data, i, state, "LTF_TBS", direct_confirmation["component"], price=float(c[i]), level=state["active_target"])
            add_event(events, data, i, state, "ENTRY", direct_confirmation["model"], price=float(c[i]), level=state["active_target"])
            entry = entry_record(data, i, state, direct_confirmation, c, atr, config, next_trade_id)
            next_trade_id += 1
            entries.append(entry)
            active_trades.append(entry.copy())
            update_trades(i, h, l, data, active_trades, entries)
            continue

        if state.get("is_mtf_crt", False) and mtf_direct_entry_mode(config):
            update_trades(i, h, l, data, active_trades, entries)
            continue

        raw_bull_fvg, raw_bear_fvg = raw_fvg(i, h, l, atr, config)
        update_important_fvg(i, o, h, l, c, atr, body, body_ratio, raw_bull_fvg, raw_bear_fvg, state, config, events, data)

        bullish_sweep, bearish_sweep = process_tbs_sweep(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            state=state,
            last_swing_high=last_swing_high,
            last_swing_low=last_swing_low,
            config=config,
            events=events,
            data=data,
        )

        process_real_breakout(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body=body,
            body_ratio=body_ratio,
            raw_bull_fvg=raw_bull_fvg,
            raw_bear_fvg=raw_bear_fvg,
            bullish_sweep=bullish_sweep,
            bearish_sweep=bearish_sweep,
            state=state,
            config=config,
            events=events,
            data=data,
        )

        invalidate_before_choch(i, c, atr, state, config, events, data)
        process_choch(i, o, h, l, c, atr, body, state, config, events, data)
        process_bos_level(i, o, h, l, c, state)
        process_bos_confirmation(i, c, atr, state, config, events, data)
        invalidate_order_block(i, c, state, events, data)

        confirmation = evaluate_entries(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body_ratio=body_ratio,
            fast_ema=fast_ema,
            slow_ema=slow_ema,
            state=state,
            config=config,
        )
        if confirmation is not None:
            state["setup_state"] = 5
            add_event(events, data, i, state, "ENTRY", confirmation["model"], price=float(c[i]))
            entry = entry_record(data, i, state, confirmation, c, atr, config, next_trade_id)
            next_trade_id += 1
            entries.append(entry)
            active_trades.append(entry.copy())

        update_trades(i, h, l, data, active_trades, entries)

        if state["setup_state"] == 4 and i - state.get("bos_bar", i) > config.entry_confirmation_bars:
            add_event(events, data, i, state, "EXPIRED", "entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"
        if state["setup_state"] == 8 and i - state.get("breakout_bar", i) > config.breakout_retest_bars:
            add_event(events, data, i, state, "EXPIRED", "breakout_entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"

    return {
        "htf_crts": htf_crts,
        "htf_data": htf_data,
        "internal_crts": pd.DataFrame(internal_crts) if internal_crts else empty_internal_crts(),
        "crts": pd.DataFrame(crts),
        "events": pd.DataFrame(events),
        "entries": pd.DataFrame(entries) if entries else empty_entries(),
        "trades": pd.DataFrame(entries) if entries else empty_entries(),
    }


def load_all_htf_data(data: pd.DataFrame, config: CRTTBSConfig) -> dict[str, pd.DataFrame]:
    return {str(timeframe): load_htf_ohlcv(data, config, str(timeframe)) for timeframe in configured_htf_intervals(config)}


def detect_all_htf_crts(data: pd.DataFrame, config: CRTTBSConfig, htf_data: dict[str, pd.DataFrame] | None = None) -> pd.DataFrame:
    htf_data = htf_data if htf_data is not None else load_all_htf_data(data, config)
    rows: list[dict[str, Any]] = []
    for timeframe, htf in htf_data.items():
        rows.extend(detect_htf_crts(data, htf, str(timeframe), config))
    if not rows:
        return empty_htf_crts()
    out = pd.DataFrame(rows).sort_values(["ActivationBarIdx", "TimeframeMinutes"], ascending=[True, False])
    out["SetupId"] = np.arange(1, len(out) + 1)
    return out.reset_index(drop=True)


def load_htf_ohlcv(data: pd.DataFrame, config: CRTTBSConfig, timeframe: str) -> pd.DataFrame:
    if config.htf_csv_path is not None and len(configured_htf_intervals(config)) == 1:
        return align_index_timezone(prepare_ohlcv(pd.read_csv(config.htf_csv_path)), data.index)
    mode = config.htf_fetch_mode.lower().strip()
    if mode == "yahoo":
        try:
            htf = fetch_yahoo_ohlcv(
                symbol=config.symbol,
                period=config.htf_period or config.period,
                interval=normalize_yahoo_interval(timeframe),
            )
        except ValueError:
            if not config.htf_resample_fallback:
                raise
            htf = resample_ohlcv(data, timeframe)
        return align_index_timezone(htf, data.index)
    if mode == "resample":
        return resample_ohlcv(data, timeframe)
    if mode == "csv":
        if config.htf_csv_path is None:
            raise ValueError("Set htf_csv_path when htf_fetch_mode='csv'.")
        return align_index_timezone(prepare_ohlcv(pd.read_csv(config.htf_csv_path)), data.index)
    raise ValueError("htf_fetch_mode must be 'yahoo', 'csv', or 'resample'.")


def configured_htf_intervals(config: CRTTBSConfig) -> tuple[str, ...]:
    if isinstance(config.htf_intervals, str):
        return (config.htf_intervals,)
    return tuple(str(item) for item in config.htf_intervals)


def normalize_yahoo_interval(interval: str) -> str:
    text = str(interval).strip().lower()
    aliases = {
        "1": "1m",
        "2": "2m",
        "5": "5m",
        "15": "15m",
        "30": "30m",
        "60": "60m",
        "90": "90m",
        "d": "1d",
        "day": "1d",
        "daily": "1d",
        "w": "1wk",
        "week": "1wk",
        "weekly": "1wk",
        "m": "1mo",
        "month": "1mo",
        "monthly": "1mo",
    }
    normalized = aliases.get(text, text)
    supported = {"1m", "2m", "5m", "15m", "30m", "60m", "90m", "1h", "1d", "5d", "1wk", "1mo", "3mo"}
    if normalized not in supported:
        raise ValueError(
            f"Yahoo native interval '{interval}' is not supported. "
            "Use one of 1m, 2m, 5m, 15m, 30m, 60m, 90m, 1h, 1d, 5d, 1wk, 1mo, 3mo; "
            "or pass real HTF candles with htf_csv_path; or explicitly set htf_fetch_mode='resample'."
        )
    return normalized


def align_index_timezone(frame: pd.DataFrame, target_index: pd.DatetimeIndex) -> pd.DataFrame:
    out = frame.copy()
    source = pd.DatetimeIndex(out.index)
    target = pd.DatetimeIndex(target_index)
    if target.tz is not None and source.tz is None:
        source = source.tz_localize(target.tz)
    elif target.tz is not None and source.tz is not None:
        source = source.tz_convert(target.tz)
    elif target.tz is None and source.tz is not None:
        source = source.tz_localize(None)
    out.index = source
    return out


def resample_ohlcv(data: pd.DataFrame, timeframe: str) -> pd.DataFrame:
    rule = timeframe_to_pandas_rule(timeframe)
    out = data.resample(rule, label="right", closed="right").agg(
        open=("open", "first"),
        high=("high", "max"),
        low=("low", "min"),
        close=("close", "last"),
        volume=("volume", "sum"),
    )
    return out.dropna(subset=["open", "high", "low", "close"])


def detect_htf_crts(data: pd.DataFrame, htf: pd.DataFrame, timeframe: str, config: CRTTBSConfig) -> list[dict[str, Any]]:
    rows: list[dict[str, Any]] = []
    if len(htf) < 2:
        return rows
    minutes = timeframe_to_minutes(timeframe)
    htf_atr = htf_atr_series(htf, config.atr_length)
    for j in range(1, len(htf)):
        previous = htf.iloc[j - 1]
        current = htf.iloc[j]
        previous_range = float(previous["high"] - previous["low"])
        if previous_range <= 0:
            continue
        if config.htf_min_range_atr > 0:
            current_atr = float(htf_atr.iloc[j])
            if np.isfinite(current_atr) and previous_range < current_atr * config.htf_min_range_atr:
                continue

        swept_low = float(current["low"]) <= float(previous["low"]) - config.tick_size * config.sweep_ticks
        swept_high = float(current["high"]) >= float(previous["high"]) + config.tick_size * config.sweep_ticks
        closed_inside = float(previous["low"]) < float(current["close"]) < float(previous["high"])
        if not closed_inside:
            continue
        if swept_low and swept_high and not config.htf_allow_both_side_sweep:
            continue

        direction = 0
        signal = ""
        target = np.nan
        if swept_low and not swept_high:
            direction = 1
            signal = "Bullish HTF TS"
            target = float(previous["high"])
        elif swept_high and not swept_low:
            direction = -1
            signal = "Bearish HTF TS"
            target = float(previous["low"])
        elif swept_low and swept_high:
            close_location = (float(current["close"]) - float(previous["low"])) / previous_range
            direction = 1 if close_location >= 0.5 else -1
            signal = "Bullish HTF TS" if direction == 1 else "Bearish HTF TS"
            target = float(previous["high"]) if direction == 1 else float(previous["low"])
        if direction == 0:
            continue

        confirm_time = htf.index[j]
        activation_idx = int(data.index.searchsorted(confirm_time, side="right"))
        if activation_idx >= len(data):
            continue
        previous_start_time = previous_htf_start_time(htf.index[j - 1], timeframe)
        start_idx = int(data.index.searchsorted(previous_start_time, side="left"))
        zone_low, zone_high = target_zone(
            target=target,
            direction=direction,
            range_high=float(previous["high"]),
            range_low=float(previous["low"]),
            fraction=config.htf_target_zone_fraction,
        )
        rows.append(
            {
                "SetupId": np.nan,
                "Timeframe": str(timeframe),
                "TimeframeMinutes": minutes,
                "Direction": direction,
                "Signal": signal,
                "RangeHigh": float(previous["high"]),
                "RangeLow": float(previous["low"]),
                "RangeMid": (float(previous["high"]) + float(previous["low"])) / 2.0,
                "TargetPrice": target,
                "TargetZoneLow": zone_low,
                "TargetZoneHigh": zone_high,
                "SourceRangeDate": htf.index[j - 1],
                "ConfirmDate": confirm_time,
                "ActivationDate": data.index[activation_idx],
                "RangeStartBarIdx": max(0, min(start_idx, len(data) - 1)),
                "ActivationBarIdx": activation_idx,
            }
        )
    return rows


def htf_atr_series(htf: pd.DataFrame, length: int) -> pd.Series:
    high = htf["high"].astype(float)
    low = htf["low"].astype(float)
    close = htf["close"].astype(float)
    previous_close = close.shift(1).fillna(close)
    true_range = pd.concat([high - low, (high - previous_close).abs(), (low - previous_close).abs()], axis=1).max(axis=1)
    return true_range.rolling(length, min_periods=1).mean()


def choose_htf_setup(setups: list[dict[str, Any]]) -> dict[str, Any]:
    return sorted(setups, key=lambda row: float(row.get("TimeframeMinutes", 0.0)), reverse=True)[0]


def detect_internal_crt(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    if not config.internal_crt_enabled or i <= 0 or not state.get("is_mtf_crt", False):
        return None
    if state["setup_state"] not in {1, 2, 3, 4}:
        return None
    if config.internal_crt_inside_parent:
        previous_inside = h[i - 1] <= state["active_high"] and l[i - 1] >= state["active_low"]
        close_inside_parent = state["active_low"] < c[i] < state["active_high"]
        if not previous_inside or not close_inside_parent:
            return None

    prev_high = h[i - 1]
    prev_low = l[i - 1]
    swept_low = l[i] <= prev_low - config.tick_size * config.sweep_ticks
    swept_high = h[i] >= prev_high + config.tick_size * config.sweep_ticks
    closed_inside_prev = prev_low < c[i] < prev_high
    if not closed_inside_prev:
        return None

    allowed = int(state.get("allowed_direction", 0))
    bullish = swept_low and not swept_high and allowed in {0, 1}
    bearish = swept_high and not swept_low and allowed in {0, -1}
    if bullish:
        return {"direction": 1, "signal": "Bullish internal TS", "price": float(l[i]), "target": float(prev_high)}
    if bearish:
        return {"direction": -1, "signal": "Bearish internal TS", "price": float(h[i]), "target": float(prev_low)}
    return None


def evaluate_mtf_tbs_touch_entry(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    if not mtf_direct_entry_mode(config):
        return None
    if not state.get("is_mtf_crt", False) or state["setup_state"] != 1 or i <= state["active_activation_bar"]:
        return None
    if config.require_internal_crt_confirmation and not state.get("internal_crt_seen", False):
        return None

    allowed = int(state.get("allowed_direction", 0))
    bullish_wick_touch = l[i] < state["active_low"] and c[i] > state["active_low"] and h[i] < state["active_high"]
    bullish_body_touch = o[i] < state["active_low"] and c[i] > state["active_low"]
    bearish_wick_touch = h[i] > state["active_high"] and c[i] < state["active_high"] and l[i] > state["active_low"]
    bearish_body_touch = o[i] > state["active_high"] and c[i] < state["active_high"]
    bullish_touch = allowed in {0, 1} and ((config.use_wick_touch and bullish_wick_touch) or (config.use_body_touch and bullish_body_touch))
    bearish_touch = allowed in {0, -1} and ((config.use_wick_touch and bearish_wick_touch) or (config.use_body_touch and bearish_body_touch))

    if bullish_touch:
        state["setup_direction"] = 1
        state["sweep_bar"] = i
        state["low_swept"] = True
        state["last_sweep_price"] = float(l[i])
        score = 3 + int(c[i] > state["active_mid"]) + int(state.get("internal_crt_seen", False))
        return {"direction": 1, "signal": "BUY", "component": "HTF RANGE LOW", "model": "MTF TBS TOUCH", "potential": min(score, 5)}
    if bearish_touch:
        state["setup_direction"] = -1
        state["sweep_bar"] = i
        state["high_swept"] = True
        state["last_sweep_price"] = float(h[i])
        score = 3 + int(c[i] < state["active_mid"]) + int(state.get("internal_crt_seen", False))
        return {"direction": -1, "signal": "SELL", "component": "HTF RANGE HIGH", "model": "MTF TBS TOUCH", "potential": min(score, 5)}
    return None


def mtf_direct_entry_mode(config: CRTTBSConfig) -> bool:
    return config.mtf_entry_mode.lower().strip() in {"tbs_touch", "touch", "direct"}


def internal_crt_record(data: pd.DataFrame, i: int, state: dict[str, Any], internal: dict[str, Any]) -> dict[str, Any]:
    return {
        "BarIdx": i,
        "Date": data.index[i],
        "CRTNumber": state.get("active_number", np.nan),
        "Timeframe": state.get("active_timeframe", ""),
        "Direction": internal["direction"],
        "Signal": internal["signal"],
        "Price": internal["price"],
        "Target": internal["target"],
    }


def target_zone(*, target: float, direction: int, range_high: float, range_low: float, fraction: float) -> tuple[float, float]:
    width = max(abs(range_high - range_low) * fraction, 1e-12)
    if direction == 1:
        return float(target - width), float(target)
    return float(target), float(target + width)


def timeframe_to_pandas_rule(timeframe: str) -> str:
    text = str(timeframe).strip()
    lower = text.lower()
    if lower in {"d", "1d", "day", "daily"}:
        return "1D"
    if lower in {"w", "1w", "week", "weekly"}:
        return "1W"
    if lower in {"m", "1m", "month", "monthly"}:
        return "1ME"
    if lower.endswith("min"):
        return lower
    if lower.endswith("h"):
        return lower
    if lower.isdigit():
        return f"{lower}min"
    return text


def timeframe_to_minutes(timeframe: str) -> float:
    text = str(timeframe).strip().lower()
    if text.isdigit():
        return float(text)
    if text.endswith("min"):
        return float(text[:-3])
    if text.endswith("h"):
        return float(text[:-1]) * 60.0
    if text in {"d", "1d", "day", "daily"}:
        return 1440.0
    if text in {"w", "1w", "week", "weekly"}:
        return 10080.0
    if text in {"m", "1m", "month", "monthly"}:
        return 43200.0
    return 0.0


def previous_htf_start_time(timestamp: pd.Timestamp, timeframe: str) -> pd.Timestamp:
    minutes = timeframe_to_minutes(timeframe)
    if minutes <= 0:
        return timestamp
    return timestamp - pd.Timedelta(minutes=minutes)


def empty_htf_crts() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "SetupId",
            "Timeframe",
            "TimeframeMinutes",
            "Direction",
            "Signal",
            "RangeHigh",
            "RangeLow",
            "RangeMid",
            "TargetPrice",
            "TargetZoneLow",
            "TargetZoneHigh",
            "SourceRangeDate",
            "ConfirmDate",
            "ActivationDate",
            "RangeStartBarIdx",
            "ActivationBarIdx",
        ]
    )


def empty_internal_crts() -> pd.DataFrame:
    return pd.DataFrame(columns=["BarIdx", "Date", "CRTNumber", "Timeframe", "Direction", "Signal", "Price", "Target"])


def run_pine_state_machine(data: pd.DataFrame, config: CRTTBSConfig) -> dict[str, pd.DataFrame]:
    o = data["open"].to_numpy(dtype=float)
    h = data["high"].to_numpy(dtype=float)
    l = data["low"].to_numpy(dtype=float)
    c = data["close"].to_numpy(dtype=float)
    atr = data["atr"].to_numpy(dtype=float)
    body = data["candle_body"].to_numpy(dtype=float)
    body_ratio = data["body_ratio"].to_numpy(dtype=float)
    candle_range = data["candle_range"].to_numpy(dtype=float)
    close_location = data["close_location"].to_numpy(dtype=float)
    range_to_atr = data["range_to_atr"].to_numpy(dtype=float)
    fast_ema = data["fast_ema"].to_numpy(dtype=float)
    slow_ema = data["slow_ema"].to_numpy(dtype=float)

    tz_index = index_in_timezone(data.index, config)
    inside_main = np.array([in_session(ts, config.main_session) for ts in tz_index], dtype=bool)
    inside_secondary = np.array(
        [config.search_24x7 or in_session(ts, config.secondary_session) for ts in tz_index],
        dtype=bool,
    )
    date_keys = np.array([date_key(ts) for ts in tz_index], dtype=int)

    state = new_state()
    crts: list[dict[str, Any]] = []
    events: list[dict[str, Any]] = []
    entries: list[dict[str, Any]] = []
    active_trades: list[dict[str, Any]] = []

    best_main_range = np.nan
    best_main_high = np.nan
    best_main_low = np.nan
    best_main_bar = None
    main_crt_created_today = False
    crts_today = 0
    last_crt_creation_bar = None
    last_swing_high = np.nan
    last_swing_low = np.nan
    next_trade_id = 1

    for i in range(len(data)):
        new_trading_day = i > 0 and date_keys[i] != date_keys[i - 1]
        main_session_started = bool(inside_main[i] and (i == 0 or not inside_main[i - 1]))
        main_session_finished = bool((not inside_main[i]) and i > 0 and inside_main[i - 1])

        if new_trading_day:
            state = new_state()
            main_crt_created_today = False
            crts_today = 0
            last_crt_creation_bar = None
            best_main_range = np.nan
            best_main_high = np.nan
            best_main_low = np.nan
            best_main_bar = None

        if new_trading_day or main_session_started:
            best_main_range = np.nan
            best_main_high = np.nan
            best_main_low = np.nan
            best_main_bar = None

        if inside_main[i] and (not np.isfinite(best_main_range) or candle_range[i] > best_main_range):
            best_main_range = candle_range[i]
            best_main_high = h[i]
            best_main_low = l[i]
            best_main_bar = i

        pivot_high, pivot_low = confirmed_pivots(
            i=i,
            high=h,
            low=l,
            left=config.swing_left,
            right=config.swing_right,
        )
        if np.isfinite(pivot_high):
            last_swing_high = pivot_high
        if np.isfinite(pivot_low):
            last_swing_low = pivot_low

        create_main_crt = (
            main_session_finished
            and not main_crt_created_today
            and np.isfinite(best_main_high)
            and np.isfinite(best_main_low)
            and best_main_bar is not None
        )
        if create_main_crt:
            main_crt_created_today = True
            crts_today += 1
            state = activate_crt(
                crt_high=float(best_main_high),
                crt_low=float(best_main_low),
                crt_start_bar=int(best_main_bar),
                activation_bar=i,
                crt_type="PRINCIPAL",
                crt_number=crts_today,
            )
            last_crt_creation_bar = i
            crts.append(crt_record(data, state))
            add_event(events, data, i, state, "CRT", "PRINCIPAL", price=state["active_mid"])

        if state["setup_state"] >= 1 and state["setup_state"] <= 4:
            if i - state["active_activation_bar"] > config.setup_expiration_bars:
                add_event(events, data, i, state, "EXPIRED", "setup_expiration")
                state["setup_state"] = 6
                state["status"] = "EXPIRADO"

        raw_bull_fvg, raw_bear_fvg = raw_fvg(i, h, l, atr, config)
        update_important_fvg(i, o, h, l, c, atr, body, body_ratio, raw_bull_fvg, raw_bear_fvg, state, config, events, data)

        bullish_sweep, bearish_sweep = process_tbs_sweep(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            state=state,
            last_swing_high=last_swing_high,
            last_swing_low=last_swing_low,
            config=config,
            events=events,
            data=data,
        )

        process_real_breakout(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body=body,
            body_ratio=body_ratio,
            raw_bull_fvg=raw_bull_fvg,
            raw_bear_fvg=raw_bear_fvg,
            bullish_sweep=bullish_sweep,
            bearish_sweep=bearish_sweep,
            state=state,
            config=config,
            events=events,
            data=data,
        )

        invalidate_before_choch(i, c, atr, state, config, events, data)
        process_choch(i, o, h, l, c, atr, body, state, config, events, data)
        process_bos_level(i, o, h, l, c, state)
        process_bos_confirmation(i, c, atr, state, config, events, data)
        invalidate_order_block(i, c, state, events, data)

        confirmation = evaluate_entries(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            atr=atr,
            body_ratio=body_ratio,
            fast_ema=fast_ema,
            slow_ema=slow_ema,
            state=state,
            config=config,
        )
        if confirmation is not None:
            state["setup_state"] = 5
            add_event(events, data, i, state, "ENTRY", confirmation["model"], price=float(c[i]))
            entry = entry_record(data, i, state, confirmation, c, atr, config, next_trade_id)
            next_trade_id += 1
            entries.append(entry)
            active_trades.append(entry.copy())

        update_trades(i, h, l, data, active_trades, entries)

        if state["setup_state"] == 4 and i - state.get("bos_bar", i) > config.entry_confirmation_bars:
            add_event(events, data, i, state, "EXPIRED", "entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"
        if state["setup_state"] == 8 and i - state.get("breakout_bar", i) > config.breakout_retest_bars:
            add_event(events, data, i, state, "EXPIRED", "breakout_entry_expired")
            state["setup_state"] = 6
            state["status"] = "EXPIRADO"

        if create_secondary_crt_allowed(
            i=i,
            o=o,
            h=h,
            l=l,
            c=c,
            candle_range=candle_range,
            body_ratio=body_ratio,
            close_location=close_location,
            range_to_atr=range_to_atr,
            inside_secondary=inside_secondary,
            inside_main=inside_main,
            main_session_finished=main_session_finished,
            state=state,
            crts_today=crts_today,
            main_crt_created_today=main_crt_created_today,
            last_crt_creation_bar=last_crt_creation_bar,
            config=config,
        ):
            crts_today += 1
            state = activate_crt(
                crt_high=float(h[i]),
                crt_low=float(l[i]),
                crt_start_bar=i,
                activation_bar=i,
                crt_type="SECUNDARIO",
                crt_number=crts_today,
            )
            last_crt_creation_bar = i
            crts.append(crt_record(data, state))
            add_event(events, data, i, state, "CRT", "SECUNDARIO", price=state["active_mid"])

    return {
        "crts": pd.DataFrame(crts),
        "events": pd.DataFrame(events),
        "entries": pd.DataFrame(entries) if entries else empty_entries(),
        "trades": pd.DataFrame(entries) if entries else empty_entries(),
    }


def new_state() -> dict[str, Any]:
    return {
        "setup_state": 0,
        "setup_direction": 0,
        "active_high": np.nan,
        "active_low": np.nan,
        "active_mid": np.nan,
        "active_target": np.nan,
        "target_zone_low": np.nan,
        "target_zone_high": np.nan,
        "active_start_bar": None,
        "active_activation_bar": None,
        "active_type": "",
        "active_timeframe": "",
        "active_number": np.nan,
        "allowed_direction": 0,
        "is_mtf_crt": False,
        "internal_crt_seen": False,
        "internal_crt_bar": None,
        "last_sweep_price": np.nan,
        "sweep_bar": None,
        "low_swept": False,
        "high_swept": False,
        "choch_level": np.nan,
        "choch_bar": None,
        "impulse_extreme": np.nan,
        "bos_level": np.nan,
        "bos_bar": None,
        "pullback_detected": False,
        "first_retest_consumed": False,
        "breakout_level": np.nan,
        "breakout_bar": None,
        "breakout_had_fvg": False,
        "breakout_retest_consumed": False,
        "important_fvg_count": 0,
        "bullish_fvg_top": np.nan,
        "bullish_fvg_bottom": np.nan,
        "bullish_fvg_creation_bar": None,
        "bearish_fvg_top": np.nan,
        "bearish_fvg_bottom": np.nan,
        "bearish_fvg_creation_bar": None,
        "bullish_ob_high": np.nan,
        "bullish_ob_low": np.nan,
        "bullish_ob_creation_bar": None,
        "bearish_ob_high": np.nan,
        "bearish_ob_low": np.nan,
        "bearish_ob_creation_bar": None,
        "bullish_ob_created": False,
        "bearish_ob_created": False,
        "status": "Esperando",
    }


def activate_crt(
    *,
    crt_high: float,
    crt_low: float,
    crt_start_bar: int,
    activation_bar: int,
    crt_type: str,
    crt_number: int,
    allowed_direction: int = 0,
    target_price: float = np.nan,
    target_zone_low: float = np.nan,
    target_zone_high: float = np.nan,
    timeframe: str = "",
    is_mtf: bool = False,
) -> dict[str, Any]:
    state = new_state()
    mid = (float(crt_high) + float(crt_low)) / 2.0
    if not np.isfinite(target_price):
        if allowed_direction == 1:
            target_price = float(crt_high)
        elif allowed_direction == -1:
            target_price = float(crt_low)
    state.update(
        {
            "setup_state": 1,
            "active_high": float(crt_high),
            "active_low": float(crt_low),
            "active_mid": mid,
            "active_target": float(target_price) if np.isfinite(target_price) else np.nan,
            "target_zone_low": float(target_zone_low) if np.isfinite(target_zone_low) else np.nan,
            "target_zone_high": float(target_zone_high) if np.isfinite(target_zone_high) else np.nan,
            "active_start_bar": int(crt_start_bar),
            "active_activation_bar": int(activation_bar),
            "active_type": crt_type,
            "active_timeframe": timeframe,
            "active_number": int(crt_number),
            "allowed_direction": int(allowed_direction),
            "is_mtf_crt": bool(is_mtf),
            "status": "Esperando TBS",
        }
    )
    return state


def process_tbs_sweep(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    state: dict[str, Any],
    last_swing_high: float,
    last_swing_low: float,
    config: CRTTBSConfig,
    events: list[dict[str, Any]],
    data: pd.DataFrame,
) -> tuple[bool, bool]:
    if state["setup_state"] != 1 or i <= state["active_activation_bar"]:
        return False, False
    minimum_sweep = float(config.sweep_ticks) * float(config.tick_size)
    low_sweep_raw = l[i] <= state["active_low"] - minimum_sweep
    high_sweep_raw = h[i] >= state["active_high"] + minimum_sweep
    if low_sweep_raw:
        state["low_swept"] = True
    if high_sweep_raw:
        state["high_swept"] = True

    allowed_direction = int(state.get("allowed_direction", 0))
    bullish_sweep = low_sweep_raw and not high_sweep_raw and allowed_direction in {0, 1} and (not config.require_close_inside or c[i] > state["active_low"])
    bearish_sweep = high_sweep_raw and not low_sweep_raw and allowed_direction in {0, -1} and (not config.require_close_inside or c[i] < state["active_high"])
    if bullish_sweep:
        fallback_high = finite_max(h[max(0, i - config.structure_lookback) : i])
        state["setup_state"] = 2
        state["setup_direction"] = 1
        state["sweep_bar"] = i
        state["last_sweep_price"] = float(l[i])
        state["choch_level"] = last_swing_high if np.isfinite(last_swing_high) else fallback_high
        state["status"] = "Esperando CHoCH"
        add_event(events, data, i, state, "TBS", "bullish_sweep", price=float(l[i]), level=state["choch_level"])
    elif bearish_sweep:
        fallback_low = finite_min(l[max(0, i - config.structure_lookback) : i])
        state["setup_state"] = 2
        state["setup_direction"] = -1
        state["sweep_bar"] = i
        state["last_sweep_price"] = float(h[i])
        state["choch_level"] = last_swing_low if np.isfinite(last_swing_low) else fallback_low
        state["status"] = "Esperando CHoCH"
        add_event(events, data, i, state, "TBS", "bearish_sweep", price=float(h[i]), level=state["choch_level"])
    return bool(bullish_sweep), bool(bearish_sweep)


def process_real_breakout(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    atr: np.ndarray,
    body: np.ndarray,
    body_ratio: np.ndarray,
    raw_bull_fvg: bool,
    raw_bear_fvg: bool,
    bullish_sweep: bool,
    bearish_sweep: bool,
    state: dict[str, Any],
    config: CRTTBSConfig,
    events: list[dict[str, Any]],
    data: pd.DataFrame,
) -> None:
    if not config.enable_breakout_retest or state["setup_state"] != 1 or i <= state["active_activation_bar"]:
        return
    breakout_margin = atr[i] * config.breakout_close_atr
    strong_body = body[i] >= atr[i] * config.breakout_body_atr and body_ratio[i] * 100.0 >= config.breakout_body_percent
    bullish_breakout = c[i] > state["active_high"] + breakout_margin and strong_body and not bullish_sweep and not bearish_sweep
    bearish_breakout = c[i] < state["active_low"] - breakout_margin and strong_body and not bullish_sweep and not bearish_sweep
    if bullish_breakout:
        state["setup_state"] = 8
        state["setup_direction"] = 1
        state["breakout_level"] = state["active_high"]
        state["breakout_bar"] = i
        state["breakout_had_fvg"] = bool(raw_bull_fvg or (i > 0 and raw_fvg(i - 1, h, l, atr, config)[0]))
        state["breakout_retest_consumed"] = False
        state["status"] = "Breakout: esperando retest"
        add_event(events, data, i, state, "BREAKOUT", "bullish_breakout", price=float(c[i]), level=state["breakout_level"])
    elif bearish_breakout:
        state["setup_state"] = 8
        state["setup_direction"] = -1
        state["breakout_level"] = state["active_low"]
        state["breakout_bar"] = i
        state["breakout_had_fvg"] = bool(raw_bear_fvg or (i > 0 and raw_fvg(i - 1, h, l, atr, config)[1]))
        state["breakout_retest_consumed"] = False
        state["status"] = "Breakout: esperando retest"
        add_event(events, data, i, state, "BREAKOUT", "bearish_breakout", price=float(c[i]), level=state["breakout_level"])


def invalidate_before_choch(i: int, c: np.ndarray, atr: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 2:
        return
    both_sides = state["low_swept"] and state["high_swept"]
    time_expired = state["sweep_bar"] is not None and i - state["sweep_bar"] > config.max_bars_to_choch
    bullish_opposite = state["setup_direction"] == 1 and c[i] < state["active_low"] - atr[i] * config.invalidation_close_atr
    bearish_opposite = state["setup_direction"] == -1 and c[i] > state["active_high"] + atr[i] * config.invalidation_close_atr
    if both_sides or time_expired or bullish_opposite or bearish_opposite:
        state["setup_state"] = 7
        state["status"] = "INVALIDADO"
        add_event(events, data, i, state, "INVALIDATED", "before_choch")


def process_choch(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, atr: np.ndarray, body: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 2 or state["sweep_bar"] is None or i <= state["sweep_bar"] or not np.isfinite(state["choch_level"]):
        return
    current_body_atr = body[i] / max(atr[i], 1e-12)
    bullish = state["setup_direction"] == 1 and c[i] > state["choch_level"] and current_body_atr >= config.minimum_choch_displacement_atr
    bearish = state["setup_direction"] == -1 and c[i] < state["choch_level"] and current_body_atr >= config.minimum_choch_displacement_atr
    if bullish or bearish:
        state["setup_state"] = 3
        state["choch_bar"] = i
        state["impulse_extreme"] = h[i] if bullish else l[i]
        state["pullback_detected"] = False
        state["status"] = "Esperando retroceso"
        create_order_block(i, o, h, l, c, state, config)
        add_event(events, data, i, state, "CHoCH", "bullish" if bullish else "bearish", price=float(c[i]), level=state["choch_level"])


def process_bos_level(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any]) -> None:
    if state["setup_state"] != 3:
        return
    if state["setup_direction"] == 1 and not state["pullback_detected"]:
        state["impulse_extreme"] = max(state["impulse_extreme"], h[i]) if np.isfinite(state["impulse_extreme"]) else h[i]
        bullish_pullback = state["choch_bar"] is not None and i > state["choch_bar"] and (c[i] < o[i] or l[i] < l[i - 1])
        if bullish_pullback:
            state["pullback_detected"] = True
            state["bos_level"] = state["impulse_extreme"]
            state["status"] = "Esperando BOS"
    elif state["setup_direction"] == -1 and not state["pullback_detected"]:
        state["impulse_extreme"] = min(state["impulse_extreme"], l[i]) if np.isfinite(state["impulse_extreme"]) else l[i]
        bearish_pullback = state["choch_bar"] is not None and i > state["choch_bar"] and (c[i] > o[i] or h[i] > h[i - 1])
        if bearish_pullback:
            state["pullback_detected"] = True
            state["bos_level"] = state["impulse_extreme"]
            state["status"] = "Esperando BOS"


def process_bos_confirmation(i: int, c: np.ndarray, atr: np.ndarray, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] != 3 or not state["pullback_detected"] or not np.isfinite(state["bos_level"]):
        return
    distance = atr[i] * config.bos_break_atr
    bullish = state["setup_direction"] == 1 and c[i] > state["bos_level"] + distance
    bearish = state["setup_direction"] == -1 and c[i] < state["bos_level"] - distance
    if bullish or bearish:
        state["setup_state"] = 4
        state["bos_bar"] = i
        state["first_retest_consumed"] = False
        state["status"] = "Esperando entrada"
        add_event(events, data, i, state, "BOS", "bullish" if bullish else "bearish", price=float(c[i]), level=state["bos_level"])


def evaluate_entries(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    atr: np.ndarray,
    body_ratio: np.ndarray,
    fast_ema: np.ndarray,
    slow_ema: np.ndarray,
    state: dict[str, Any],
    config: CRTTBSConfig,
) -> dict[str, Any] | None:
    if config.require_internal_crt_confirmation and state.get("is_mtf_crt", False) and not state.get("internal_crt_seen", False):
        return None
    current_atr = safe_atr_value(atr[i])
    (
        bullish_rejection_candle,
        bearish_rejection_candle,
        bullish_rejection,
        bearish_rejection,
        bullish_rejection_score,
        bearish_rejection_score,
    ) = rejection_signals(i, o, h, l, c, body_ratio, fast_ema, slow_ema, state, config)
    tbs = tbs_entry_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config)
    breakout = breakout_retest_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config)
    zone = zone_confluence_signal(i, o, h, l, c, body_ratio, fast_ema, slow_ema, current_atr, state, config, bullish_rejection_candle, bearish_rejection_candle)

    if tbs is not None:
        return tbs
    if breakout is not None:
        return breakout
    if bullish_rejection and state["setup_direction"] == 1:
        return {"direction": 1, "signal": "BUY", "component": "BARRIDA + RECHAZO", "model": "TBS REJECTION", "potential": bullish_rejection_score}
    if bearish_rejection and state["setup_direction"] == -1:
        return {"direction": -1, "signal": "SELL", "component": "BARRIDA + RECHAZO", "model": "TBS REJECTION", "potential": bearish_rejection_score}
    if zone is not None:
        return zone
    return None


def tbs_entry_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    valid = state["setup_state"] == 4 and state["bos_bar"] is not None and i > state["bos_bar"] and i - state["bos_bar"] <= config.entry_confirmation_bars and not state["first_retest_consumed"]
    if not valid:
        return None
    tolerance = current_atr * config.retest_tolerance_atr
    bullish_bos_retest = state["setup_direction"] == 1 and config.accept_bos_retest and l[i] <= state["bos_level"] + tolerance and h[i] >= state["bos_level"] - tolerance
    bearish_bos_retest = state["setup_direction"] == -1 and config.accept_bos_retest and h[i] >= state["bos_level"] - tolerance and l[i] <= state["bos_level"] + tolerance
    bullish_fvg_retest = state["setup_direction"] == 1 and config.accept_fvg_retest and state["bullish_fvg_creation_bar"] is not None and state["bullish_fvg_creation_bar"] > state["sweep_bar"] and l[i] <= state["bullish_fvg_top"] + tolerance and h[i] >= state["bullish_fvg_bottom"] - tolerance
    bearish_fvg_retest = state["setup_direction"] == -1 and config.accept_fvg_retest and state["bearish_fvg_creation_bar"] is not None and state["bearish_fvg_creation_bar"] > state["sweep_bar"] and h[i] >= state["bearish_fvg_bottom"] - tolerance and l[i] <= state["bearish_fvg_top"] + tolerance
    bullish_ob_retest = state["setup_direction"] == 1 and config.accept_order_block_retest and state["bullish_ob_created"] and state["bullish_ob_creation_bar"] >= state["sweep_bar"] and l[i] <= state["bullish_ob_high"] + tolerance and h[i] >= state["bullish_ob_low"] - tolerance and c[i] >= state["bullish_ob_low"]
    bearish_ob_retest = state["setup_direction"] == -1 and config.accept_order_block_retest and state["bearish_ob_created"] and state["bearish_ob_creation_bar"] >= state["sweep_bar"] and h[i] >= state["bearish_ob_low"] - tolerance and l[i] <= state["bearish_ob_high"] + tolerance and c[i] <= state["bearish_ob_high"]

    bullish_primary = bullish_ob_retest or bullish_fvg_retest
    bearish_primary = bearish_ob_retest or bearish_fvg_retest
    bullish_zone_accepted = bullish_primary if config.require_fvg_or_ob else (bullish_primary or bullish_bos_retest)
    bearish_zone_accepted = bearish_primary if config.require_fvg_or_ob else (bearish_primary or bearish_bos_retest)
    bullish_confirm = c[i] > o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] > c[i - 1]
    bearish_confirm = c[i] < o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] < c[i - 1]
    bullish_bos_close = (not config.require_close_beyond_bos) or c[i] > state["bos_level"]
    bearish_bos_close = (not config.require_close_beyond_bos) or c[i] < state["bos_level"]
    bullish_mid = (not config.require_midpoint_alignment) or c[i] > state["active_mid"]
    bearish_mid = (not config.require_midpoint_alignment) or c[i] < state["active_mid"]
    bullish_ema = (not config.require_ema_alignment) or (fast_ema[i] > slow_ema[i] and c[i] > fast_ema[i])
    bearish_ema = (not config.require_ema_alignment) or (fast_ema[i] < slow_ema[i] and c[i] < fast_ema[i])
    bullish_score = int(bullish_ob_retest) + int(bullish_fvg_retest) + int(bullish_bos_close) + int(bullish_mid) + int(bullish_ema)
    bearish_score = int(bearish_ob_retest) + int(bearish_fvg_retest) + int(bearish_bos_close) + int(bearish_mid) + int(bearish_ema)
    if bullish_zone_accepted or bearish_zone_accepted:
        state["first_retest_consumed"] = True

    if state["setup_direction"] == 1 and bullish_zone_accepted and bullish_confirm and bullish_bos_close and bullish_mid and bullish_ema and bullish_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if bullish_ob_retest else ("FVG" if bullish_fvg_retest else "BOS")
        return {"direction": 1, "signal": "BUY", "component": component, "model": "TBS REVERSIÓN", "potential": bullish_score}
    if state["setup_direction"] == -1 and bearish_zone_accepted and bearish_confirm and bearish_bos_close and bearish_mid and bearish_ema and bearish_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if bearish_ob_retest else ("FVG" if bearish_fvg_retest else "BOS")
        return {"direction": -1, "signal": "SELL", "component": component, "model": "TBS REVERSIÓN", "potential": bearish_score}
    return None


def breakout_retest_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig) -> dict[str, Any] | None:
    valid = state["setup_state"] == 8 and state["breakout_bar"] is not None and i > state["breakout_bar"] and i - state["breakout_bar"] <= config.breakout_retest_bars and not state["breakout_retest_consumed"]
    if not valid:
        return None
    tolerance = current_atr * config.retest_tolerance_atr
    bullish_touch = state["setup_direction"] == 1 and l[i] <= state["breakout_level"] + tolerance and h[i] >= state["breakout_level"] - tolerance
    bearish_touch = state["setup_direction"] == -1 and h[i] >= state["breakout_level"] - tolerance and l[i] <= state["breakout_level"] + tolerance
    bullish_confirm = bullish_touch and c[i] > state["breakout_level"] and c[i] > o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] > c[i - 1]
    bearish_confirm = bearish_touch and c[i] < state["breakout_level"] and c[i] < o[i] and body_ratio[i] * 100.0 >= config.entry_body_percent and c[i] < c[i - 1]
    bullish_ema = (not config.require_ema_alignment) or (fast_ema[i] > slow_ema[i] and c[i] > fast_ema[i])
    bearish_ema = (not config.require_ema_alignment) or (fast_ema[i] < slow_ema[i] and c[i] < fast_ema[i])
    if bullish_touch or bearish_touch:
        state["breakout_retest_consumed"] = True
    if state["setup_direction"] == 1 and c[i] < state["active_high"] - current_atr * config.breakout_invalidation_atr:
        state["setup_state"] = 7
        return None
    if state["setup_direction"] == -1 and c[i] > state["active_low"] + current_atr * config.breakout_invalidation_atr:
        state["setup_state"] = 7
        return None
    bullish_score = int(state["breakout_had_fvg"]) + int(c[i] > state["active_high"]) + int(bullish_ema) + int(body_ratio[i] * 100.0 >= config.entry_body_percent) + 1
    bearish_score = int(state["breakout_had_fvg"]) + int(c[i] < state["active_low"]) + int(bearish_ema) + int(body_ratio[i] * 100.0 >= config.entry_body_percent) + 1
    if state["setup_direction"] == 1 and bullish_confirm and bullish_ema and (not config.require_breakout_fvg or state["breakout_had_fvg"]) and bullish_score >= config.minimum_potential_score:
        return {"direction": 1, "signal": "BUY", "component": "CRT HIGH RETEST", "model": "BREAKOUT RETEST", "potential": bullish_score}
    if state["setup_direction"] == -1 and bearish_confirm and bearish_ema and (not config.require_breakout_fvg or state["breakout_had_fvg"]) and bearish_score >= config.minimum_potential_score:
        return {"direction": -1, "signal": "SELL", "component": "CRT LOW RETEST", "model": "BREAKOUT RETEST", "potential": bearish_score}
    return None


def rejection_signals(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> tuple[bool, bool, bool, bool, int, int]:
    candle_range = max(h[i] - l[i], 1e-12)
    lower_wick_pct = (min(o[i], c[i]) - l[i]) / candle_range * 100.0
    upper_wick_pct = (h[i] - max(o[i], c[i])) / candle_range * 100.0
    bullish_candle = c[i] > o[i] and lower_wick_pct >= config.minimum_rejection_wick_percent and c[i] > state.get("active_low", np.nan) and c[i] > c[i - 1]
    bearish_candle = c[i] < o[i] and upper_wick_pct >= config.minimum_rejection_wick_percent and c[i] < state.get("active_high", np.nan) and c[i] < c[i - 1]
    valid_bullish = config.enable_sweep_rejection_entry and state["setup_state"] == 2 and state["setup_direction"] == 1 and state["sweep_bar"] is not None and i >= state["sweep_bar"] and i - state["sweep_bar"] <= config.sweep_rejection_bars
    valid_bearish = config.enable_sweep_rejection_entry and state["setup_state"] == 2 and state["setup_direction"] == -1 and state["sweep_bar"] is not None and i >= state["sweep_bar"] and i - state["sweep_bar"] <= config.sweep_rejection_bars
    bullish_score = int(c[i] > state.get("active_low", np.nan)) + int(lower_wick_pct >= config.minimum_rejection_wick_percent) + int(c[i] > o[i]) + int(c[i] > c[i - 1]) + int((not config.require_ema_alignment) or fast_ema[i] > slow_ema[i])
    bearish_score = int(c[i] < state.get("active_high", np.nan)) + int(upper_wick_pct >= config.minimum_rejection_wick_percent) + int(c[i] < o[i]) + int(c[i] < c[i - 1]) + int((not config.require_ema_alignment) or fast_ema[i] < slow_ema[i])
    confirmed_bullish = valid_bullish and bullish_candle and bullish_score >= config.minimum_potential_score
    confirmed_bearish = valid_bearish and bearish_candle and bearish_score >= config.minimum_potential_score
    return bullish_candle, bearish_candle, confirmed_bullish, confirmed_bearish, bullish_score, bearish_score


def zone_confluence_signal(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, body_ratio: np.ndarray, fast_ema: np.ndarray, slow_ema: np.ndarray, current_atr: float, state: dict[str, Any], config: CRTTBSConfig, bullish_rejection_candle: bool, bearish_rejection_candle: bool) -> dict[str, Any] | None:
    if not config.enable_high_potential_zones or state["setup_state"] not in {1, 2, 3, 4}:
        return None
    tolerance = current_atr * config.important_zone_tolerance_atr
    price_at_bull_ob = state["bullish_ob_created"] and l[i] <= state["bullish_ob_high"] + tolerance and h[i] >= state["bullish_ob_low"] - tolerance and c[i] >= state["bullish_ob_low"]
    price_at_bear_ob = state["bearish_ob_created"] and h[i] >= state["bearish_ob_low"] - tolerance and l[i] <= state["bearish_ob_high"] + tolerance and c[i] <= state["bearish_ob_high"]
    price_at_bull_fvg = state["bullish_fvg_creation_bar"] is not None and l[i] <= state["bullish_fvg_top"] + tolerance and h[i] >= state["bullish_fvg_bottom"] - tolerance
    price_at_bear_fvg = state["bearish_fvg_creation_bar"] is not None and h[i] >= state["bearish_fvg_bottom"] - tolerance and l[i] <= state["bearish_fvg_top"] + tolerance
    price_at_low_liq = np.isfinite(state["active_low"]) and l[i] <= state["active_low"] + tolerance and c[i] > state["active_low"]
    price_at_high_liq = np.isfinite(state["active_high"]) and h[i] >= state["active_high"] - tolerance and c[i] < state["active_high"]
    bull_score = int(price_at_bull_ob) + int(price_at_bull_fvg) + int(price_at_low_liq) + int(bullish_rejection_candle) + int((not config.require_ema_alignment) or fast_ema[i] > slow_ema[i])
    bear_score = int(price_at_bear_ob) + int(price_at_bear_fvg) + int(price_at_high_liq) + int(bearish_rejection_candle) + int((not config.require_ema_alignment) or fast_ema[i] < slow_ema[i])
    if (price_at_bull_ob or price_at_bull_fvg or price_at_low_liq) and bullish_rejection_candle and bull_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if price_at_bull_ob else ("FVG" if price_at_bull_fvg else "LIQUIDEZ CRT")
        return {"direction": 1, "signal": "BUY", "component": component, "model": "ZONE CONFLUENCE", "potential": bull_score}
    if (price_at_bear_ob or price_at_bear_fvg or price_at_high_liq) and bearish_rejection_candle and bear_score >= config.minimum_potential_score:
        component = "ORDER BLOCK" if price_at_bear_ob else ("FVG" if price_at_bear_fvg else "LIQUIDEZ CRT")
        return {"direction": -1, "signal": "SELL", "component": component, "model": "ZONE CONFLUENCE", "potential": bear_score}
    return None


def invalidate_order_block(i: int, c: np.ndarray, state: dict[str, Any], events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if state["setup_state"] < 3 or state["setup_state"] > 4:
        return
    bullish_broken = (
        state["setup_direction"] == 1
        and state["bullish_ob_created"]
        and np.isfinite(state["bullish_ob_low"])
        and c[i] < state["bullish_ob_low"]
    )
    bearish_broken = (
        state["setup_direction"] == -1
        and state["bearish_ob_created"]
        and np.isfinite(state["bearish_ob_high"])
        and c[i] > state["bearish_ob_high"]
    )
    if bullish_broken or bearish_broken:
        state["setup_state"] = 7
        state["status"] = "INVALIDADO"
        add_event(events, data, i, state, "INVALIDATED", "order_block_broken")


def create_order_block(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, state: dict[str, Any], config: CRTTBSConfig) -> None:
    if not config.show_order_block or state["sweep_bar"] is None:
        return
    max_search = min(config.order_block_lookback, i - state["sweep_bar"])
    for offset in range(1, max_search + 1):
        idx = i - offset
        if state["setup_direction"] == 1 and c[idx] < o[idx]:
            state["bullish_ob_high"] = h[idx]
            state["bullish_ob_low"] = l[idx]
            state["bullish_ob_creation_bar"] = idx
            state["bullish_ob_created"] = True
            return
        if state["setup_direction"] == -1 and c[idx] > o[idx]:
            state["bearish_ob_high"] = h[idx]
            state["bearish_ob_low"] = l[idx]
            state["bearish_ob_creation_bar"] = idx
            state["bearish_ob_created"] = True
            return


def update_important_fvg(i: int, o: np.ndarray, h: np.ndarray, l: np.ndarray, c: np.ndarray, atr: np.ndarray, body: np.ndarray, body_ratio: np.ndarray, raw_bull: bool, raw_bear: bool, state: dict[str, Any], config: CRTTBSConfig, events: list[dict[str, Any]], data: pd.DataFrame) -> None:
    if i < 2 or not config.show_important_fvg or state["important_fvg_count"] >= config.maximum_important_fvg:
        return
    center_body = abs(c[i - 1] - o[i - 1])
    center_range = max(h[i - 1] - l[i - 1], 1e-12)
    center_body_ratio = center_body / center_range
    displacement = center_body >= atr[i - 1] * config.displacement_body_atr and center_body_ratio * 100.0 >= config.displacement_body_percent
    state_ok = state["setup_state"] in {2, 3, 4, 8}
    impulse_started = (state["sweep_bar"] is not None and i > state["sweep_bar"]) or (state["breakout_bar"] is not None and i >= state["breakout_bar"])
    if state["setup_direction"] == 1 and state_ok and impulse_started and raw_bull and displacement:
        mid = (l[i] + h[i - 2]) / 2.0
        if abs(c[i] - mid) <= atr[i] * 2.0:
            state["bullish_fvg_top"] = l[i]
            state["bullish_fvg_bottom"] = h[i - 2]
            state["bullish_fvg_creation_bar"] = i
            state["important_fvg_count"] += 1
            add_event(events, data, i, state, "FVG", "bullish", price=mid)
    if state["setup_direction"] == -1 and state_ok and impulse_started and raw_bear and displacement:
        mid = (l[i - 2] + h[i]) / 2.0
        if abs(c[i] - mid) <= atr[i] * 2.0:
            state["bearish_fvg_top"] = l[i - 2]
            state["bearish_fvg_bottom"] = h[i]
            state["bearish_fvg_creation_bar"] = i
            state["important_fvg_count"] += 1
            add_event(events, data, i, state, "FVG", "bearish", price=mid)


def create_secondary_crt_allowed(
    *,
    i: int,
    o: np.ndarray,
    h: np.ndarray,
    l: np.ndarray,
    c: np.ndarray,
    candle_range: np.ndarray,
    body_ratio: np.ndarray,
    close_location: np.ndarray,
    range_to_atr: np.ndarray,
    inside_secondary: np.ndarray,
    inside_main: np.ndarray,
    main_session_finished: bool,
    state: dict[str, Any],
    crts_today: int,
    main_crt_created_today: bool,
    last_crt_creation_bar: int | None,
    config: CRTTBSConfig,
) -> bool:
    secondary_bull_quality = c[i] > o[i] and close_location[i] >= config.secondary_close_extreme
    secondary_bear_quality = c[i] < o[i] and close_location[i] <= 1.0 - config.secondary_close_extreme
    secondary_expansion = (not config.require_secondary_expansion) or h[i] > h[i - 1] or l[i] < l[i - 1]
    secondary_range_expansion = (not config.require_secondary_range_expansion) or candle_range[i] > candle_range[i - 1]
    valid_secondary = inside_secondary[i] and range_to_atr[i] >= config.secondary_range_atr and body_ratio[i] * 100.0 >= config.secondary_body_percent and (secondary_bull_quality or secondary_bear_quality) and secondary_expansion and secondary_range_expansion
    previous_finished = state["setup_state"] in {0, 5, 6, 7}
    cooldown = last_crt_creation_bar is None or i - last_crt_creation_bar >= config.secondary_cooldown_bars
    time_allowed = (not config.secondary_only_after_main) or main_crt_created_today
    return valid_secondary and time_allowed and not inside_main[i] and previous_finished and cooldown and crts_today < config.maximum_crts_per_day and not main_session_finished


def entry_record(data: pd.DataFrame, i: int, state: dict[str, Any], confirmation: dict[str, Any], c: np.ndarray, atr: np.ndarray, config: CRTTBSConfig, trade_id: int) -> dict[str, Any]:
    direction = int(confirmation["direction"])
    entry = float(c[i])
    structural_stop = state["active_low"] if direction == 1 else state["active_high"]
    if np.isfinite(state.get("last_sweep_price", np.nan)):
        structural_stop = min(structural_stop, state["last_sweep_price"]) if direction == 1 else max(structural_stop, state["last_sweep_price"])
    if confirmation["component"] == "ORDER BLOCK":
        if direction == 1 and np.isfinite(state["bullish_ob_low"]):
            structural_stop = state["bullish_ob_low"]
        elif direction == -1 and np.isfinite(state["bearish_ob_high"]):
            structural_stop = state["bearish_ob_high"]
    stop = structural_stop - atr[i] * config.result_stop_buffer_atr if direction == 1 else structural_stop + atr[i] * config.result_stop_buffer_atr
    risk = abs(entry - stop)
    target = entry + risk * config.result_risk_reward if direction == 1 else entry - risk * config.result_risk_reward
    if config.result_target_mode.lower().strip() == "crt" and np.isfinite(state.get("active_target", np.nan)):
        crt_target = float(state["active_target"])
        if (direction == 1 and crt_target > entry) or (direction == -1 and crt_target < entry):
            target = crt_target
    return {
        "TradeId": trade_id,
        "CRTNumber": state["active_number"],
        "CRTType": state["active_type"],
        "Signal": confirmation["signal"],
        "Direction": direction,
        "Model": confirmation["model"],
        "Component": confirmation["component"],
        "Potential": confirmation["potential"],
        "EntryBarIdx": i,
        "EntryDate": data.index[i],
        "EntryPrice": entry,
        "StopPrice": float(stop),
        "TargetPrice": float(target),
        "Result": "ACTIVE",
        "ResultBarIdx": np.nan,
        "ResultDate": pd.NaT,
        "ActiveCRTHigh": state["active_high"],
        "ActiveCRTLow": state["active_low"],
        "TBSBarIdx": state["sweep_bar"],
        "ChoChBarIdx": state["choch_bar"],
        "BOSBarIdx": state["bos_bar"],
        "BreakoutBarIdx": state["breakout_bar"],
        "HTFTimeframe": state.get("active_timeframe", ""),
        "ActiveTargetPrice": state.get("active_target", np.nan),
        "TargetZoneLow": state.get("target_zone_low", np.nan),
        "TargetZoneHigh": state.get("target_zone_high", np.nan),
        "InternalCRTSeen": bool(state.get("internal_crt_seen", False)),
        "InternalCRTBarIdx": state.get("internal_crt_bar", None),
    }


def update_trades(i: int, h: np.ndarray, l: np.ndarray, data: pd.DataFrame, active_trades: list[dict[str, Any]], entries: list[dict[str, Any]]) -> None:
    remaining = []
    for trade in active_trades:
        if i <= trade["EntryBarIdx"]:
            remaining.append(trade)
            continue
        direction = int(trade["Direction"])
        target_hit = h[i] >= trade["TargetPrice"] if direction == 1 else l[i] <= trade["TargetPrice"]
        stop_hit = l[i] <= trade["StopPrice"] if direction == 1 else h[i] >= trade["StopPrice"]
        if target_hit or stop_hit:
            result = "WIN" if target_hit and not stop_hit else "FAILD"
            trade["Result"] = result
            trade["ResultBarIdx"] = i
            trade["ResultDate"] = data.index[i]
            for row in entries:
                if row["TradeId"] == trade["TradeId"]:
                    row.update({"Result": result, "ResultBarIdx": i, "ResultDate": data.index[i]})
                    break
        else:
            remaining.append(trade)
    active_trades[:] = remaining


def add_event(events: list[dict[str, Any]], data: pd.DataFrame, i: int, state: dict[str, Any], event: str, detail: str, *, price: float | None = None, level: float | None = None) -> None:
    events.append(
        {
            "BarIdx": i,
            "Date": data.index[i],
            "Event": event,
            "Detail": detail,
            "CRTNumber": state.get("active_number", np.nan),
            "CRTType": state.get("active_type", ""),
            "Direction": state.get("setup_direction", 0),
            "Price": np.nan if price is None else float(price),
            "Level": np.nan if level is None or not np.isfinite(level) else float(level),
            "State": state.get("setup_state", 0),
        }
    )


def crt_record(data: pd.DataFrame, state: dict[str, Any]) -> dict[str, Any]:
    return {
        "CRTNumber": state["active_number"],
        "CRTType": state["active_type"],
        "Timeframe": state.get("active_timeframe", ""),
        "Direction": state.get("allowed_direction", 0),
        "StartBarIdx": state["active_start_bar"],
        "ActivationBarIdx": state["active_activation_bar"],
        "StartDate": data.index[state["active_start_bar"]],
        "ActivationDate": data.index[state["active_activation_bar"]],
        "CRTHigh": state["active_high"],
        "CRTLow": state["active_low"],
        "CRTMid": state["active_mid"],
        "TargetPrice": state.get("active_target", np.nan),
        "TargetZoneLow": state.get("target_zone_low", np.nan),
        "TargetZoneHigh": state.get("target_zone_high", np.nan),
        "IsMTF": bool(state.get("is_mtf_crt", False)),
    }


def confirmed_pivots(i: int, high: np.ndarray, low: np.ndarray, *, left: int, right: int) -> tuple[float, float]:
    if i < left + right:
        return np.nan, np.nan
    pivot_idx = i - right
    start = pivot_idx - left
    end = pivot_idx + right + 1
    pivot_high = high[pivot_idx] if high[pivot_idx] >= finite_max(high[start:end]) else np.nan
    pivot_low = low[pivot_idx] if low[pivot_idx] <= finite_min(low[start:end]) else np.nan
    return pivot_high, pivot_low


def raw_fvg(i: int, high: np.ndarray, low: np.ndarray, atr: np.ndarray, config: CRTTBSConfig) -> tuple[bool, bool]:
    if i < 2:
        return False, False
    bullish = low[i] > high[i - 2] and low[i] - high[i - 2] >= atr[i] * config.minimum_fvg_atr
    bearish = high[i] < low[i - 2] and low[i - 2] - high[i] >= atr[i] * config.minimum_fvg_atr
    return bool(bullish), bool(bearish)


def index_in_timezone(index: pd.DatetimeIndex, config: CRTTBSConfig) -> pd.DatetimeIndex:
    idx = pd.DatetimeIndex(index)
    if idx.tz is None:
        idx = idx.tz_localize(ZoneInfo(config.data_timezone), nonexistent="shift_forward", ambiguous="NaT")
    return idx.tz_convert(ZoneInfo(config.timezone))


def in_session(ts: pd.Timestamp, session: str) -> bool:
    if pd.isna(ts):
        return False
    start_text, end_text = session.split("-")
    start = int(start_text[:2]) * 60 + int(start_text[2:])
    end = int(end_text[:2]) * 60 + int(end_text[2:])
    now = int(ts.hour) * 60 + int(ts.minute)
    if start <= end:
        return start <= now < end
    return now >= start or now < end


def date_key(ts: pd.Timestamp) -> int:
    if pd.isna(ts):
        return -1
    return int(ts.year) * 10000 + int(ts.month) * 100 + int(ts.day)


def safe_atr_value(value: float) -> float:
    return float(value) if np.isfinite(value) and value > 0 else 1e-12


def finite_min(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.min(arr)) if len(arr) else np.nan


def finite_max(values: np.ndarray | pd.Series) -> float:
    arr = np.asarray(values, dtype=float)
    arr = arr[np.isfinite(arr)]
    return float(np.max(arr)) if len(arr) else np.nan


def empty_entries() -> pd.DataFrame:
    return pd.DataFrame(
        columns=[
            "TradeId",
            "CRTNumber",
            "CRTType",
            "Signal",
            "Direction",
            "Model",
            "Component",
            "Potential",
            "EntryBarIdx",
            "EntryDate",
            "EntryPrice",
            "StopPrice",
            "TargetPrice",
            "Result",
            "ResultBarIdx",
            "ResultDate",
            "ActiveCRTHigh",
            "ActiveCRTLow",
            "TBSBarIdx",
            "ChoChBarIdx",
            "BOSBarIdx",
            "BreakoutBarIdx",
            "HTFTimeframe",
            "ActiveTargetPrice",
            "TargetZoneLow",
            "TargetZoneHigh",
            "InternalCRTSeen",
            "InternalCRTBarIdx",
        ]
    )


def run_vectorbt_backtest(
    data: pd.DataFrame,
    result: dict[str, pd.DataFrame],
    config: CRTTBSConfig,
    *,
    show_plot: bool,
) -> dict[str, Any]:
    signals = build_vectorbt_signals(data, result["entries"], config)
    orders = build_vectorbt_order_frame(data, result["entries"], config)
    if vbt is None:
        detail = "" if VBT_IMPORT_ERROR is None else f" ({type(VBT_IMPORT_ERROR).__name__}: {VBT_IMPORT_ERROR})"
        message = f"Install/fix vectorbt to run the portfolio backtest: pip install vectorbt{detail}"
        if config.backtest_raise_missing_vectorbt:
            raise ImportError(message)
        warnings.warn(message)
        return {"signals": signals, "orders": orders, "portfolio": None, "stats": pd.Series(dtype=object), "fig": None}

    engine = config.backtest_engine.lower().strip()
    if engine == "orders":
        portfolio = vectorbt_from_orders(data, orders, config)
    elif engine == "signals":
        portfolio = vectorbt_from_signals(data, signals, config)
    else:
        raise ValueError("backtest_engine must be 'orders' or 'signals'.")

    stats = portfolio_stats(portfolio)
    fig = None
    if show_plot:
        fig = portfolio_plot(portfolio)
        if fig is not None:
            display_figure(fig)
    return {"signals": signals, "orders": orders, "portfolio": portfolio, "stats": stats, "fig": fig}


def empty_backtest_result(data: pd.DataFrame, result: dict[str, pd.DataFrame], config: CRTTBSConfig) -> dict[str, Any]:
    signals = empty_vectorbt_signals(data)
    orders = build_vectorbt_order_frame(data, result["entries"], config) if "entries" in result else empty_order_frame()
    return {"signals": signals, "orders": orders, "portfolio": None, "stats": pd.Series(dtype=object), "fig": None}


def build_vectorbt_signals(data: pd.DataFrame, entries: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    signals = empty_vectorbt_signals(data)
    if entries.empty:
        return signals

    for _, row in entries.iterrows():
        raw_entry_idx = int(row["EntryBarIdx"])
        entry_idx = raw_entry_idx + int(config.backtest_shift_signals)
        if entry_idx >= len(data):
            continue
        direction = int(row["Direction"])
        entry_price = float(data[config.backtest_price_column].iloc[entry_idx])
        stop_price = float(row["StopPrice"])
        target_price = float(row["TargetPrice"])
        if direction == 1:
            signals.iat[entry_idx, signals.columns.get_loc("long_entries")] = True
            signals.iat[entry_idx, signals.columns.get_loc("sl_stop")] = max((entry_price - stop_price) / entry_price, 0.0)
            signals.iat[entry_idx, signals.columns.get_loc("tp_stop")] = max((target_price - entry_price) / entry_price, 0.0)
        else:
            signals.iat[entry_idx, signals.columns.get_loc("short_entries")] = True
            signals.iat[entry_idx, signals.columns.get_loc("sl_stop")] = max((stop_price - entry_price) / entry_price, 0.0)
            signals.iat[entry_idx, signals.columns.get_loc("tp_stop")] = max((entry_price - target_price) / entry_price, 0.0)
        signals.iat[entry_idx, signals.columns.get_loc("trade_id")] = int(row["TradeId"])
        signals.iat[entry_idx, signals.columns.get_loc("entry_price")] = entry_price
        signals.iat[entry_idx, signals.columns.get_loc("stop_price")] = stop_price
        signals.iat[entry_idx, signals.columns.get_loc("target_price")] = target_price

    return signals


def empty_vectorbt_signals(data: pd.DataFrame) -> pd.DataFrame:
    return pd.DataFrame(
        {
            "long_entries": pd.Series(False, index=data.index, dtype=bool),
            "long_exits": pd.Series(False, index=data.index, dtype=bool),
            "short_entries": pd.Series(False, index=data.index, dtype=bool),
            "short_exits": pd.Series(False, index=data.index, dtype=bool),
            "sl_stop": pd.Series(np.nan, index=data.index, dtype=float),
            "tp_stop": pd.Series(np.nan, index=data.index, dtype=float),
            "trade_id": pd.Series(np.nan, index=data.index, dtype=float),
            "entry_price": pd.Series(np.nan, index=data.index, dtype=float),
            "stop_price": pd.Series(np.nan, index=data.index, dtype=float),
            "target_price": pd.Series(np.nan, index=data.index, dtype=float),
        }
    )


def build_vectorbt_order_frame(data: pd.DataFrame, entries: pd.DataFrame, config: CRTTBSConfig) -> pd.DataFrame:
    if entries.empty:
        return empty_order_frame()
    rows: list[dict[str, Any]] = []
    for _, entry in entries.iterrows():
        entry_idx = int(entry["EntryBarIdx"]) + int(config.backtest_shift_signals)
        if entry_idx >= len(data):
            continue
        direction = int(entry["Direction"])
        size = float(config.backtest_size) * direction
        rows.append(
            {
                "TradeId": int(entry["TradeId"]),
                "BarIdx": entry_idx,
                "Date": data.index[entry_idx],
                "Size": size,
                "Price": float(data[config.backtest_price_column].iloc[entry_idx]),
                "Reason": "ENTRY",
                "Signal": entry["Signal"],
                "Model": entry["Model"],
            }
        )
        if pd.notna(entry["ResultBarIdx"]):
            exit_idx = int(entry["ResultBarIdx"])
            if exit_idx < len(data) and exit_idx > entry_idx:
                exit_price = float(entry["TargetPrice"]) if entry["Result"] == "WIN" else float(entry["StopPrice"])
                rows.append(
                    {
                        "TradeId": int(entry["TradeId"]),
                        "BarIdx": exit_idx,
                        "Date": data.index[exit_idx],
                        "Size": -size,
                        "Price": exit_price,
                        "Reason": str(entry["Result"]),
                        "Signal": entry["Signal"],
                        "Model": entry["Model"],
                    }
                )
    if not rows:
        return empty_order_frame()
    return pd.DataFrame(rows).sort_values(["BarIdx", "TradeId", "Reason"]).reset_index(drop=True)


def empty_order_frame() -> pd.DataFrame:
    return pd.DataFrame(columns=["TradeId", "BarIdx", "Date", "Size", "Price", "Reason", "Signal", "Model"])


def vectorbt_from_orders(data: pd.DataFrame, orders: pd.DataFrame, config: CRTTBSConfig):
    order_size = pd.Series(np.nan, index=data.index, dtype=float)
    order_price = data[config.backtest_price_column].astype(float).copy()
    if not orders.empty:
        for bar_idx, group in orders.groupby("BarIdx", sort=True):
            bar_idx = int(bar_idx)
            size_sum = float(group["Size"].sum())
            if size_sum == 0.0:
                continue
            weights = group["Size"].abs()
            order_size.iloc[bar_idx] = size_sum
            order_price.iloc[bar_idx] = float((group["Price"] * weights).sum() / weights.sum())
    return vbt.Portfolio.from_orders(
        close=data["close"].astype(float),
        size=order_size,
        price=order_price,
        size_type=config.backtest_size_type,
        direction="both",
        fees=config.backtest_fees,
        fixed_fees=config.backtest_fixed_fees,
        slippage=config.backtest_slippage,
        init_cash=config.backtest_init_cash,
        freq=vectorbt_freq(data, config),
    )


def vectorbt_from_signals(data: pd.DataFrame, signals: pd.DataFrame, config: CRTTBSConfig):
    price = data[config.backtest_price_column].astype(float)
    return vbt.Portfolio.from_signals(
        close=data["close"].astype(float),
        entries=signals["long_entries"],
        exits=signals["long_exits"],
        short_entries=signals["short_entries"],
        short_exits=signals["short_exits"],
        size=config.backtest_size,
        size_type=config.backtest_size_type,
        price=price,
        fees=config.backtest_fees,
        fixed_fees=config.backtest_fixed_fees,
        slippage=config.backtest_slippage,
        open=data["open"].astype(float),
        high=data["high"].astype(float),
        low=data["low"].astype(float),
        sl_stop=signals["sl_stop"],
        tp_stop=signals["tp_stop"],
        sl_trail=False,
        upon_opposite_entry=config.backtest_upon_opposite_entry,
        accumulate=config.backtest_accumulate,
        init_cash=config.backtest_init_cash,
        freq=vectorbt_freq(data, config),
    )


def vectorbt_freq(data: pd.DataFrame, config: CRTTBSConfig) -> str | None:
    if config.backtest_freq is not None:
        return config.backtest_freq
    try:
        return pd.infer_freq(data.index)
    except (TypeError, ValueError):
        return None


def portfolio_stats(portfolio: Any) -> pd.Series:
    try:
        return portfolio.stats()
    except Exception as exc:
        warnings.warn(f"vectorbt stats failed: {exc}")
        return pd.Series(dtype=object)


def portfolio_plot(portfolio: Any) -> Any:
    try:
        return portfolio.plot()
    except Exception as exc:
        warnings.warn(f"vectorbt plot failed: {exc}")
        return None


def plot_crt_tbs_state_machine(data: pd.DataFrame, result: dict[str, pd.DataFrame], *, title: str):
    if go is None:
        warnings.warn("plotly is not installed; skipping plot.")
        return None
    x = np.arange(len(data))
    labels = data.index.strftime("%Y-%m-%d %H:%M")
    fig = go.Figure()
    fig.add_trace(
        go.Candlestick(
            x=x,
            open=data["open"],
            high=data["high"],
            low=data["low"],
            close=data["close"],
            name="Candles",
            increasing_line_color="#22C55E",
            decreasing_line_color="#EF4444",
            customdata=labels,
            hovertext=labels,
            hoverinfo="text+y",
        )
    )
    for _, row in result["crts"].iterrows():
        start = int(row["StartBarIdx"])
        end = min(len(data) - 1, int(row["ActivationBarIdx"]) + 80)
        is_mtf = bool(row.get("IsMTF", False))
        line_color = "#38BDF8" if is_mtf else "#E5E7EB"
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTHigh"], y1=row["CRTHigh"], line=dict(color=line_color, width=2 if is_mtf else 1))
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTLow"], y1=row["CRTLow"], line=dict(color=line_color, width=2 if is_mtf else 1))
        fig.add_shape(type="line", x0=start, x1=end, y0=row["CRTMid"], y1=row["CRTMid"], line=dict(color="#94A3B8", width=1, dash="dot"))
        if np.isfinite(row.get("TargetZoneLow", np.nan)) and np.isfinite(row.get("TargetZoneHigh", np.nan)):
            fig.add_shape(
                type="rect",
                x0=int(row["ActivationBarIdx"]),
                x1=end,
                y0=float(row["TargetZoneLow"]),
                y1=float(row["TargetZoneHigh"]),
                fillcolor="rgba(250, 204, 21, 0.16)",
                line=dict(color="#FACC15", width=1, dash="dot"),
            )
            fig.add_annotation(
                x=end,
                y=float(row["TargetPrice"]),
                text=f"{row.get('Timeframe', '')} Target",
                showarrow=False,
                font=dict(color="#FACC15", size=10),
                xanchor="left",
            )
    event_colors = {"HTF_CRT": "#38BDF8", "INTERNAL_CRT": "#FACC15", "TBS": "#FACC15", "CHoCH": "#38BDF8", "BOS": "#22C55E", "BREAKOUT": "#F97316", "ENTRY": "#A78BFA", "INVALIDATED": "#EF4444", "EXPIRED": "#64748B", "FVG": "#CBD5E1", "CRT": "#E5E7EB"}
    if not result["events"].empty:
        for event_name, group in result["events"].groupby("Event"):
            prices = []
            for _, row in group.iterrows():
                if np.isfinite(row.get("Price", np.nan)):
                    prices.append(float(row["Price"]))
                elif np.isfinite(row.get("Level", np.nan)):
                    prices.append(float(row["Level"]))
                else:
                    prices.append(float(data["close"].iloc[int(row["BarIdx"])]))
            fig.add_trace(
                go.Scatter(
                    x=group["BarIdx"],
                    y=prices,
                    mode="markers+text",
                    name=event_name,
                    marker=dict(color=event_colors.get(event_name, "#E5E7EB"), size=9),
                    text=group["Event"],
                    textposition="top center",
                    hovertext=[
                        f"{row['Event']} {row['Detail']}<br>{row['Date']}<br>CRT #{row['CRTNumber']}<br>Level: {row['Level']}"
                        for _, row in group.iterrows()
                    ],
                    hoverinfo="text",
                )
            )
    if not result["entries"].empty:
        fig.add_trace(
            go.Scatter(
                x=result["entries"]["EntryBarIdx"],
                y=result["entries"]["EntryPrice"],
                mode="markers+text",
                name="Entries",
                marker=dict(color="#A78BFA", size=13, symbol="star"),
                text=result["entries"]["Signal"],
                textposition="bottom center",
                hovertext=[
                    (
                        f"{row['Signal']}<br>{row['Model']} / {row['Component']}<br>"
                        f"Entry: {row['EntryPrice']:.4f}<br>Stop: {row['StopPrice']:.4f}<br>"
                        f"Target: {row['TargetPrice']:.4f}<br>Result: {row['Result']}"
                    )
                    for _, row in result["entries"].iterrows()
                ],
                hoverinfo="text",
            )
        )
    ticks = np.linspace(0, len(data) - 1, min(12, len(data)), dtype=int)
    fig.update_layout(
        title=title,
        template="plotly_dark",
        height=820,
        xaxis_rangeslider_visible=False,
        legend=dict(orientation="h", y=1.02, x=1, xanchor="right"),
        margin=dict(l=55, r=40, t=80, b=60),
    )
    fig.update_xaxes(tickmode="array", tickvals=ticks, ticktext=labels[ticks], tickangle=-30)
    return fig


def display_figure(fig: Any) -> None:
    try:
        from IPython.display import display
    except ModuleNotFoundError:
        return
    display(fig)
```

```python
#!/usr/bin/env python3
"""
CRT + TBS (Candle Range Theory + Turtle Body Soup) backtest.

Default model implemented here (strict, non-repainting C1/C2 execution model):

1) The previous completed HTF candle (default 4H) defines the reference range:
   CRH = reference high, CRL = reference low, EQ = 50% midpoint.
2) On the LTF (default 5m), a manipulation candle must have its full BODY
   beyond one side of that range.
3) The next LTF candle must close back inside the reference range.
4) Entry is placed at the NEXT bar's open to avoid close-to-close lookahead.
5) Stop is beyond the manipulation candle (wick or body, configurable).
6) Target is either the reference midpoint or the opposite boundary.

The implementation deliberately uses only completed candles for signals.
It includes fees, slippage, risk-based sizing, a leverage cap, conservative
same-bar SL/TP handling, CSV exports and an equity chart.

Install:
    python -m pip install yahooquery pandas numpy matplotlib

Examples:
    python crt_tbs_yahooquery.py --symbols THYAO.IS --period 60d
    python crt_tbs_yahooquery.py --symbols THYAO.IS TUPRS.IS --target midpoint
    python crt_tbs_yahooquery.py --symbols BTC-USD --session-filter off \
        --timezone UTC --htf-offset 0h --period 60d

Important:
Yahoo Finance is an unofficial data source through yahooquery. Intraday data
availability is limited by Yahoo. Always validate timestamps, missing bars,
fees, spread and execution assumptions before relying on a backtest.
"""

from __future__ import annotations

import argparse
import sys
from dataclasses import asdict, dataclass
from pathlib import Path
from typing import Literal, Optional

import numpy as np
import pandas as pd

try:
    from yahooquery import Ticker
except ImportError as exc:  # pragma: no cover - user-facing dependency check
    raise SystemExit(
        "yahooquery bulunamadı. Kurulum: "
        "python -m pip install yahooquery pandas numpy matplotlib"
    ) from exc


TargetMode = Literal["midpoint", "opposite"]
StopMode = Literal["wick", "body"]
TBSMode = Literal["strict_body", "close_only"]
IntrabarPriority = Literal["stop", "target"]


@dataclass(frozen=True)
class StrategyConfig:
    # Data / timeframe
    interval: str = "5m"
    period: str = "60d"
    timezone: str = "Europe/Istanbul"
    htf_rule: str = "4h"
    htf_offset: str = "10h"  # BIST: 4H buckets beginning at 10:00
    min_htf_completeness: float = 0.65

    # Optional exchange-session filtering (BIST defaults)
    session_filter: bool = True
    session_start: str = "10:00"
    session_end: str = "18:10"

    # Pattern definition
    tbs_mode: TBSMode = "strict_body"
    require_parent_opposite_color: bool = False
    require_confirmation_color: bool = False
    exclude_double_sweep: bool = True
    one_signal_per_reference_side: bool = True

    # Trade construction
    target_mode: TargetMode = "opposite"
    stop_mode: StopMode = "wick"
    atr_period: int = 14
    stop_buffer_atr: float = 0.05
    stop_buffer_pct: float = 0.0002
    min_rr: float = 0.80
    max_holding_bars: int = 192
    intrabar_priority: IntrabarPriority = "stop"

    # Risk / costs
    initial_capital: float = 100_000.0
    risk_per_trade: float = 0.01
    max_leverage: float = 1.0
    fee_rate_per_side: float = 0.0005
    slippage_bps: float = 1.0

    # Output
    output_dir: str = "crt_tbs_output"


@dataclass
class PendingOrder:
    direction: Literal["long", "short"]
    signal_time: pd.Timestamp
    entry_bar_pos: int
    ref_bucket: pd.Timestamp
    ref_high: float
    ref_low: float
    ref_mid: float
    manipulation_time: pd.Timestamp
    manipulation_open: float
    manipulation_high: float
    manipulation_low: float
    manipulation_close: float
    atr: float


@dataclass
class OpenTrade:
    direction: Literal["long", "short"]
    signal_time: pd.Timestamp
    entry_time: pd.Timestamp
    ref_bucket: pd.Timestamp
    ref_high: float
    ref_low: float
    ref_mid: float
    manipulation_time: pd.Timestamp
    entry_price: float
    stop_price: float
    target_price: float
    quantity: float
    initial_risk_cash: float
    entry_fee: float
    entry_bar_pos: int


@dataclass
class ClosedTrade:
    symbol: str
    direction: str
    signal_time: pd.Timestamp
    entry_time: pd.Timestamp
    exit_time: pd.Timestamp
    manipulation_time: pd.Timestamp
    ref_bucket: pd.Timestamp
    entry_price: float
    stop_price: float
    target_price: float
    exit_price: float
    quantity: float
    gross_pnl: float
    fees: float
    net_pnl: float
    r_multiple: float
    bars_held: int
    exit_reason: str
    ref_high: float
    ref_low: float
    ref_mid: float


def parse_args() -> argparse.Namespace:
    parser = argparse.ArgumentParser(
        description="YahooQuery ile CRT + TBS sinyali ve backtesti"
    )
    parser.add_argument(
        "--symbols",
        nargs="+",
        default=["THYAO.IS"],
        help="Yahoo sembolleri. Örn: THYAO.IS TUPRS.IS BTC-USD",
    )
    parser.add_argument("--period", default="60d")
    parser.add_argument("--interval", default="5m")
    parser.add_argument("--timezone", default="Europe/Istanbul")
    parser.add_argument("--htf", dest="htf_rule", default="4h")
    parser.add_argument("--htf-offset", default="10h")
    parser.add_argument(
        "--session-filter", choices=["on", "off"], default="on"
    )
    parser.add_argument("--session-start", default="10:00")
    parser.add_argument("--session-end", default="18:10")
    parser.add_argument(
        "--target", choices=["midpoint", "opposite"], default="opposite"
    )
    parser.add_argument("--stop-mode", choices=["wick", "body"], default="wick")
    parser.add_argument(
        "--tbs-mode",
        choices=["strict_body", "close_only"],
        default="strict_body",
    )
    parser.add_argument("--initial-capital", type=float, default=100_000.0)
    parser.add_argument("--risk", type=float, default=0.01)
    parser.add_argument("--max-leverage", type=float, default=1.0)
    parser.add_argument("--fee", type=float, default=0.0005)
    parser.add_argument("--slippage-bps", type=float, default=1.0)
    parser.add_argument("--min-rr", type=float, default=0.80)
    parser.add_argument("--max-holding-bars", type=int, default=192)
    parser.add_argument("--output-dir", default="crt_tbs_output")
    return parser.parse_args()


def validate_config(cfg: StrategyConfig) -> None:
    if not 0 < cfg.risk_per_trade <= 1:
        raise ValueError("risk_per_trade 0 ile 1 arasında olmalı.")
    if cfg.max_leverage <= 0:
        raise ValueError("max_leverage pozitif olmalı.")
    if not 0 <= cfg.fee_rate_per_side < 1:
        raise ValueError("fee_rate_per_side geçersiz.")
    if cfg.max_holding_bars < 1:
        raise ValueError("max_holding_bars en az 1 olmalı.")
    if not 0 < cfg.min_htf_completeness <= 1:
        raise ValueError("min_htf_completeness 0 ile 1 arasında olmalı.")


def download_yahooquery(symbol: str, cfg: StrategyConfig) -> pd.DataFrame:
    """Download and normalize OHLCV from yahooquery."""
    ticker = Ticker(symbol, asynchronous=False)
    raw = ticker.history(
        period=cfg.period,
        interval=cfg.interval,
        adj_timezone=True,
        adj_ohlc=False,
    )

    if raw is None or not isinstance(raw, pd.DataFrame) or raw.empty:
        raise RuntimeError(f"{symbol}: YahooQuery boş veri döndürdü.")

    df = raw.copy()
    if isinstance(df.index, pd.MultiIndex):
        symbol_level = "symbol" if "symbol" in df.index.names else df.index.names[0]
        try:
            df = df.xs(symbol, level=symbol_level)
        except KeyError:
            # Some Yahoo responses normalize symbol capitalization.
            available = df.index.get_level_values(symbol_level).unique().tolist()
            if len(available) != 1:
                raise RuntimeError(
                    f"{symbol}: MultiIndex içinden sembol seçilemedi. Mevcut: {available}"
                )
            df = df.xs(available[0], level=symbol_level)

    df.index = pd.to_datetime(df.index)
    if df.index.tz is None:
        # Defensive fallback. Yahooquery with adj_timezone=True normally returns
        # exchange-local timestamps, but naive values are treated as UTC here.
        df.index = df.index.tz_localize("UTC")

    df.index = df.index.tz_convert(cfg.timezone)
    df.index.name = "datetime"

    df.columns = [str(c).lower().replace(" ", "_") for c in df.columns]
    required = ["open", "high", "low", "close", "volume"]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise RuntimeError(f"{symbol}: eksik sütunlar: {missing}")

    df = df[required].apply(pd.to_numeric, errors="coerce")
    df = df.replace([np.inf, -np.inf], np.nan).dropna(subset=required)
    df = df[~df.index.duplicated(keep="last")].sort_index()
    df = df[(df["high"] >= df[["open", "close", "low"]].max(axis=1))]
    df = df[(df["low"] <= df[["open", "close", "high"]].min(axis=1))]
    df = df[df["volume"] >= 0]

    if cfg.session_filter:
        df = df.between_time(
            cfg.session_start,
            cfg.session_end,
            inclusive="both",
        )

    if len(df) < 100:
        raise RuntimeError(
            f"{symbol}: filtrelerden sonra yalnızca {len(df)} bar kaldı."
        )
    return df


def add_atr(df: pd.DataFrame, period: int) -> pd.DataFrame:
    out = df.copy()
    prev_close = out["close"].shift(1)
    tr = pd.concat(
        [
            out["high"] - out["low"],
            (out["high"] - prev_close).abs(),
            (out["low"] - prev_close).abs(),
        ],
        axis=1,
    ).max(axis=1)
    out["atr"] = tr.ewm(alpha=1 / period, adjust=False, min_periods=period).mean()
    return out


def add_htf_reference(df: pd.DataFrame, cfg: StrategyConfig) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Map each LTF bar to its current HTF bucket and attach ONLY the previous
    completed HTF candle as the reference. This prevents HTF lookahead.
    """
    out = df.copy()
    offset = pd.Timedelta(cfg.htf_offset)
    htf_delta = pd.Timedelta(cfg.htf_rule)
    ltf_delta = pd.Timedelta(cfg.interval)

    # Works for fixed intraday intervals such as 5m -> 4h.
    out["htf_bucket"] = (out.index - offset).floor(cfg.htf_rule) + offset

    grouped = out.groupby("htf_bucket", sort=True)
    htf = grouped.agg(
        open=("open", "first"),
        high=("high", "max"),
        low=("low", "min"),
        close=("close", "last"),
        volume=("volume", "sum"),
        ltf_bars=("close", "size"),
        start_time=("close", lambda x: x.index[0]),
        end_time=("close", lambda x: x.index[-1]),
    )

    expected_bars = max(1, int(round(htf_delta / ltf_delta)))
    htf["is_complete_enough"] = (
        htf["ltf_bars"] >= expected_bars * cfg.min_htf_completeness
    )

    # Diagnostic: completed classic CRT on HTF.
    prev_high = htf["high"].shift(1)
    prev_low = htf["low"].shift(1)
    inside_close = htf["close"].between(prev_low, prev_high, inclusive="both")
    bull_sweep = htf["low"] < prev_low
    bear_sweep = htf["high"] > prev_high
    if cfg.exclude_double_sweep:
        htf["bullish_crt"] = bull_sweep & ~bear_sweep & inside_close
        htf["bearish_crt"] = bear_sweep & ~bull_sweep & inside_close
    else:
        htf["bullish_crt"] = bull_sweep & inside_close
        htf["bearish_crt"] = bear_sweep & inside_close

    # Only valid, sufficiently complete HTF candles may become references.
    ref = htf[
        ["open", "high", "low", "close", "is_complete_enough"]
    ].copy()
    ref["ref_bucket"] = ref.index
    ref = ref.shift(1).add_prefix("ref_")

    out = out.join(ref, on="htf_bucket")
    out["ref_mid"] = (out["ref_high"] + out["ref_low"]) / 2.0
    out["ref_valid"] = out["ref_is_complete_enough"].eq(True)
    return out, htf


def add_tbs_signals(df: pd.DataFrame, cfg: StrategyConfig) -> pd.DataFrame:
    """Create strict two-candle TBS confirmation signals on the LTF."""
    out = df.copy()

    prev_open = out["open"].shift(1)
    prev_close = out["close"].shift(1)
    prev_high = out["high"].shift(1)
    prev_low = out["low"].shift(1)
    prev_ref_low = out["ref_low"].shift(1)
    prev_ref_high = out["ref_high"].shift(1)

    same_reference = (
        out["ref_ref_bucket"].eq(out["ref_ref_bucket"].shift(1))
        & out["htf_bucket"].eq(out["htf_bucket"].shift(1))
        & out["ref_valid"]
        & out["ref_valid"].shift(1).eq(True)
    )

    if cfg.tbs_mode == "strict_body":
        body_below = (prev_open < prev_ref_low) & (prev_close < prev_ref_low)
        body_above = (prev_open > prev_ref_high) & (prev_close > prev_ref_high)
    else:
        body_below = prev_close < prev_ref_low
        body_above = prev_close > prev_ref_high

    confirmation_inside = out["close"].between(
        out["ref_low"], out["ref_high"], inclusive="both"
    )
    reclaim_low = out["close"] > out["ref_low"]
    reclaim_high = out["close"] < out["ref_high"]

    long_signal = same_reference & body_below & reclaim_low & confirmation_inside
    short_signal = same_reference & body_above & reclaim_high & confirmation_inside

    if cfg.exclude_double_sweep:
        # Exclude manipulation bars that traversed both sides of the HTF range.
        long_signal &= ~(prev_high > prev_ref_high)
        short_signal &= ~(prev_low < prev_ref_low)

    if cfg.require_parent_opposite_color:
        # Long prefers bearish parent; short prefers bullish parent.
        long_signal &= out["ref_close"] < out["ref_open"]
        short_signal &= out["ref_close"] > out["ref_open"]

    if cfg.require_confirmation_color:
        long_signal &= out["close"] > out["open"]
        short_signal &= out["close"] < out["open"]

    out["long_signal"] = long_signal.fillna(False)
    out["short_signal"] = short_signal.fillna(False)
    out["manipulation_time"] = out.index.to_series().shift(1)
    return out


def adverse_slippage(price: float, direction: str, bps: float, is_entry: bool) -> float:
    slip = bps / 10_000.0
    if direction == "long":
        return price * (1 + slip if is_entry else 1 - slip)
    return price * (1 - slip if is_entry else 1 + slip)


def build_pending_order(df: pd.DataFrame, signal_pos: int, direction: str) -> PendingOrder:
    signal = df.iloc[signal_pos]
    manip = df.iloc[signal_pos - 1]
    return PendingOrder(
        direction=direction,  # type: ignore[arg-type]
        signal_time=df.index[signal_pos],
        entry_bar_pos=signal_pos + 1,
        ref_bucket=signal["ref_ref_bucket"],
        ref_high=float(signal["ref_high"]),
        ref_low=float(signal["ref_low"]),
        ref_mid=float(signal["ref_mid"]),
        manipulation_time=df.index[signal_pos - 1],
        manipulation_open=float(manip["open"]),
        manipulation_high=float(manip["high"]),
        manipulation_low=float(manip["low"]),
        manipulation_close=float(manip["close"]),
        atr=float(signal["atr"]) if pd.notna(signal["atr"]) else np.nan,
    )


def create_trade(
    order: PendingOrder,
    bar: pd.Series,
    bar_time: pd.Timestamp,
    equity: float,
    cfg: StrategyConfig,
) -> Optional[OpenTrade]:
    raw_entry = float(bar["open"])
    entry = adverse_slippage(raw_entry, order.direction, cfg.slippage_bps, True)

    atr_buffer = 0.0 if not np.isfinite(order.atr) else order.atr * cfg.stop_buffer_atr
    pct_buffer = entry * cfg.stop_buffer_pct
    buffer = max(atr_buffer, pct_buffer)

    if order.direction == "long":
        base_stop = (
            order.manipulation_low
            if cfg.stop_mode == "wick"
            else min(order.manipulation_open, order.manipulation_close)
        )
        stop = base_stop - buffer
        target = order.ref_mid if cfg.target_mode == "midpoint" else order.ref_high
        risk_per_unit = entry - stop
        reward_per_unit = target - entry
    else:
        base_stop = (
            order.manipulation_high
            if cfg.stop_mode == "wick"
            else max(order.manipulation_open, order.manipulation_close)
        )
        stop = base_stop + buffer
        target = order.ref_mid if cfg.target_mode == "midpoint" else order.ref_low
        risk_per_unit = stop - entry
        reward_per_unit = entry - target

    if risk_per_unit <= 0 or reward_per_unit <= 0:
        return None
    rr = reward_per_unit / risk_per_unit
    if rr < cfg.min_rr:
        return None

    risk_budget = equity * cfg.risk_per_trade
    qty_by_risk = risk_budget / risk_per_unit
    qty_by_leverage = (equity * cfg.max_leverage) / entry
    quantity = min(qty_by_risk, qty_by_leverage)
    if not np.isfinite(quantity) or quantity <= 0:
        return None

    initial_risk_cash = risk_per_unit * quantity
    entry_fee = entry * quantity * cfg.fee_rate_per_side
    return OpenTrade(
        direction=order.direction,
        signal_time=order.signal_time,
        entry_time=bar_time,
        ref_bucket=order.ref_bucket,
        ref_high=order.ref_high,
        ref_low=order.ref_low,
        ref_mid=order.ref_mid,
        manipulation_time=order.manipulation_time,
        entry_price=entry,
        stop_price=stop,
        target_price=target,
        quantity=quantity,
        initial_risk_cash=initial_risk_cash,
        entry_fee=entry_fee,
        entry_bar_pos=order.entry_bar_pos,
    )


def choose_exit(
    trade: OpenTrade,
    bar: pd.Series,
    current_pos: int,
    cfg: StrategyConfig,
) -> tuple[Optional[float], Optional[str]]:
    high = float(bar["high"])
    low = float(bar["low"])

    if trade.direction == "long":
        stop_hit = low <= trade.stop_price
        target_hit = high >= trade.target_price
    else:
        stop_hit = high >= trade.stop_price
        target_hit = low <= trade.target_price

    if stop_hit and target_hit:
        if cfg.intrabar_priority == "stop":
            return trade.stop_price, "same_bar_stop_first"
        return trade.target_price, "same_bar_target_first"
    if stop_hit:
        return trade.stop_price, "stop"
    if target_hit:
        return trade.target_price, "target"
    if current_pos - trade.entry_bar_pos + 1 >= cfg.max_holding_bars:
        return float(bar["close"]), "time_exit"
    return None, None


def close_trade(
    symbol: str,
    trade: OpenTrade,
    raw_exit_price: float,
    exit_time: pd.Timestamp,
    exit_reason: str,
    current_pos: int,
    cfg: StrategyConfig,
) -> ClosedTrade:
    exit_price = adverse_slippage(
        raw_exit_price, trade.direction, cfg.slippage_bps, False
    )
    if trade.direction == "long":
        gross = (exit_price - trade.entry_price) * trade.quantity
    else:
        gross = (trade.entry_price - exit_price) * trade.quantity

    exit_fee = exit_price * trade.quantity * cfg.fee_rate_per_side
    fees = trade.entry_fee + exit_fee
    net = gross - fees
    r_multiple = net / trade.initial_risk_cash if trade.initial_risk_cash > 0 else np.nan

    return ClosedTrade(
        symbol=symbol,
        direction=trade.direction,
        signal_time=trade.signal_time,
        entry_time=trade.entry_time,
        exit_time=exit_time,
        manipulation_time=trade.manipulation_time,
        ref_bucket=trade.ref_bucket,
        entry_price=trade.entry_price,
        stop_price=trade.stop_price,
        target_price=trade.target_price,
        exit_price=exit_price,
        quantity=trade.quantity,
        gross_pnl=gross,
        fees=fees,
        net_pnl=net,
        r_multiple=r_multiple,
        bars_held=current_pos - trade.entry_bar_pos + 1,
        exit_reason=exit_reason,
        ref_high=trade.ref_high,
        ref_low=trade.ref_low,
        ref_mid=trade.ref_mid,
    )


def mark_to_market(equity: float, trade: Optional[OpenTrade], close: float) -> float:
    if trade is None:
        return equity
    if trade.direction == "long":
        unrealized = (close - trade.entry_price) * trade.quantity
    else:
        unrealized = (trade.entry_price - close) * trade.quantity
    return equity + unrealized - trade.entry_fee


def run_backtest(
    symbol: str, df: pd.DataFrame, cfg: StrategyConfig
) -> tuple[pd.DataFrame, pd.DataFrame, dict[str, float]]:
    equity = cfg.initial_capital
    pending: Optional[PendingOrder] = None
    open_trade: Optional[OpenTrade] = None
    closed: list[ClosedTrade] = []
    equity_points: list[dict[str, object]] = []
    consumed: set[tuple[pd.Timestamp, str]] = set()

    for pos in range(len(df)):
        ts = df.index[pos]
        bar = df.iloc[pos]

        # Execute previously confirmed signal at next bar open.
        if pending is not None and pos == pending.entry_bar_pos and open_trade is None:
            open_trade = create_trade(pending, bar, ts, equity, cfg)
            pending = None

        # Manage active position from its entry bar onward.
        if open_trade is not None:
            raw_exit, reason = choose_exit(open_trade, bar, pos, cfg)
            if raw_exit is not None and reason is not None:
                record = close_trade(
                    symbol,
                    open_trade,
                    raw_exit,
                    ts,
                    reason,
                    pos,
                    cfg,
                )
                equity += record.net_pnl
                closed.append(record)
                open_trade = None

        # Generate a pending order after the current bar has closed.
        if open_trade is None and pending is None and pos >= 1 and pos + 1 < len(df):
            direction: Optional[str] = None
            if bool(bar["long_signal"]):
                direction = "long"
            elif bool(bar["short_signal"]):
                direction = "short"

            if direction is not None:
                ref_bucket = bar["ref_ref_bucket"]
                key = (ref_bucket, direction)
                if not cfg.one_signal_per_reference_side or key not in consumed:
                    pending = build_pending_order(df, pos, direction)
                    consumed.add(key)

        mtm = mark_to_market(equity, open_trade, float(bar["close"]))
        equity_points.append(
            {
                "datetime": ts,
                "realized_equity": equity,
                "mtm_equity": mtm,
                "position": 0 if open_trade is None else (1 if open_trade.direction == "long" else -1),
            }
        )

    # Close any still-open trade at the final close.
    if open_trade is not None:
        pos = len(df) - 1
        ts = df.index[pos]
        record = close_trade(
            symbol,
            open_trade,
            float(df.iloc[pos]["close"]),
            ts,
            "end_of_data",
            pos,
            cfg,
        )
        equity += record.net_pnl
        closed.append(record)
        equity_points[-1]["realized_equity"] = equity
        equity_points[-1]["mtm_equity"] = equity
        equity_points[-1]["position"] = 0

    trades_df = pd.DataFrame([asdict(t) for t in closed])
    equity_df = pd.DataFrame(equity_points).set_index("datetime")
    metrics = compute_metrics(trades_df, equity_df, cfg.initial_capital)
    return trades_df, equity_df, metrics


def compute_metrics(
    trades: pd.DataFrame, equity: pd.DataFrame, initial_capital: float
) -> dict[str, float]:
    final_equity = (
        float(equity["realized_equity"].iloc[-1]) if not equity.empty else initial_capital
    )
    total_return = final_equity / initial_capital - 1

    mtm = equity["mtm_equity"].astype(float)
    running_max = mtm.cummax()
    drawdown = mtm / running_max - 1
    max_drawdown = float(drawdown.min()) if not drawdown.empty else 0.0

    daily = mtm.resample("1D").last().dropna().pct_change().dropna()
    sharpe = (
        float(np.sqrt(252) * daily.mean() / daily.std(ddof=1))
        if len(daily) > 1 and daily.std(ddof=1) > 0
        else np.nan
    )

    if trades.empty:
        return {
            "trades": 0.0,
            "wins": 0.0,
            "losses": 0.0,
            "win_rate": np.nan,
            "net_pnl": 0.0,
            "final_equity": final_equity,
            "total_return": total_return,
            "max_drawdown": max_drawdown,
            "profit_factor": np.nan,
            "avg_r": np.nan,
            "median_r": np.nan,
            "expectancy": np.nan,
            "sharpe_daily": sharpe,
        }

    wins = trades[trades["net_pnl"] > 0]
    losses = trades[trades["net_pnl"] <= 0]
    gross_profit = wins["net_pnl"].sum()
    gross_loss = -losses["net_pnl"].sum()
    profit_factor = gross_profit / gross_loss if gross_loss > 0 else np.inf

    return {
        "trades": float(len(trades)),
        "wins": float(len(wins)),
        "losses": float(len(losses)),
        "win_rate": float(len(wins) / len(trades)),
        "net_pnl": float(trades["net_pnl"].sum()),
        "final_equity": final_equity,
        "total_return": total_return,
        "max_drawdown": max_drawdown,
        "profit_factor": float(profit_factor),
        "avg_r": float(trades["r_multiple"].mean()),
        "median_r": float(trades["r_multiple"].median()),
        "expectancy": float(trades["net_pnl"].mean()),
        "sharpe_daily": sharpe,
    }


def save_plot(symbol: str, equity: pd.DataFrame, output_path: Path) -> None:
    try:
        import matplotlib.pyplot as plt
    except ImportError:
        print("matplotlib yok; equity grafiği atlandı.", file=sys.stderr)
        return

    fig, ax = plt.subplots(figsize=(12, 5))
    ax.plot(equity.index, equity["mtm_equity"], label="Mark-to-market equity")
    ax.plot(
        equity.index,
        equity["realized_equity"],
        label="Realized equity",
        alpha=0.75,
    )
    ax.set_title(f"{symbol} — CRT + TBS Equity")
    ax.set_xlabel("Date")
    ax.set_ylabel("Equity")
    ax.grid(True, alpha=0.25)
    ax.legend()
    fig.tight_layout()
    fig.savefig(output_path, dpi=150)
    plt.close(fig)


def safe_symbol_name(symbol: str) -> str:
    return "".join(ch if ch.isalnum() or ch in "-_." else "_" for ch in symbol)


def format_pct(value: float) -> str:
    return "n/a" if not np.isfinite(value) else f"{value * 100:.2f}%"


def print_metrics(symbol: str, metrics: dict[str, float]) -> None:
    pf = metrics["profit_factor"]
    pf_text = "n/a" if np.isnan(pf) else ("∞" if np.isinf(pf) else f"{pf:.2f}")
    print("\n" + "=" * 72)
    print(f"{symbol} — CRT + TBS BACKTEST")
    print("=" * 72)
    print(f"Trade sayısı       : {int(metrics['trades'])}")
    print(f"Kazanan / Kaybeden : {int(metrics['wins'])} / {int(metrics['losses'])}")
    print(f"Win rate           : {format_pct(metrics['win_rate'])}")
    print(f"Net PnL            : {metrics['net_pnl']:.2f}")
    print(f"Final equity       : {metrics['final_equity']:.2f}")
    print(f"Toplam getiri      : {format_pct(metrics['total_return'])}")
    print(f"Maks. drawdown     : {format_pct(metrics['max_drawdown'])}")
    print(f"Profit factor      : {pf_text}")
    print(f"Ortalama R         : {metrics['avg_r']:.3f}")
    print(f"Medyan R           : {metrics['median_r']:.3f}")
    print(f"Trade başı beklenti: {metrics['expectancy']:.2f}")
    print(f"Günlük Sharpe      : {metrics['sharpe_daily']:.3f}")


def process_symbol(symbol: str, cfg: StrategyConfig) -> dict[str, object]:
    print(f"\n{symbol}: veri indiriliyor...")
    raw = download_yahooquery(symbol, cfg)
    data = add_atr(raw, cfg.atr_period)
    data, htf = add_htf_reference(data, cfg)
    data = add_tbs_signals(data, cfg)

    trades, equity, metrics = run_backtest(symbol, data, cfg)
    print_metrics(symbol, metrics)

    out_dir = Path(cfg.output_dir)
    out_dir.mkdir(parents=True, exist_ok=True)
    stem = safe_symbol_name(symbol)

    data.to_csv(out_dir / f"{stem}_bars_signals.csv")
    htf.to_csv(out_dir / f"{stem}_htf_crt.csv")
    trades.to_csv(out_dir / f"{stem}_trades.csv", index=False)
    equity.to_csv(out_dir / f"{stem}_equity.csv")
    save_plot(symbol, equity, out_dir / f"{stem}_equity.png")

    latest_signals = data.loc[data["long_signal"] | data["short_signal"]].tail(10)
    if not latest_signals.empty:
        cols = [
            "open",
            "high",
            "low",
            "close",
            "ref_low",
            "ref_mid",
            "ref_high",
            "long_signal",
            "short_signal",
        ]
        print("\nSon TBS sinyalleri:")
        print(latest_signals[cols].to_string())
    else:
        print("\nSeçili dönemde TBS sinyali bulunamadı.")

    return {"symbol": symbol, **metrics}


def main() -> int:
    args = parse_args()
    cfg = StrategyConfig(
        interval=args.interval,
        period=args.period,
        timezone=args.timezone,
        htf_rule=args.htf_rule,
        htf_offset=args.htf_offset,
        session_filter=args.session_filter == "on",
        session_start=args.session_start,
        session_end=args.session_end,
        target_mode=args.target,
        stop_mode=args.stop_mode,
        tbs_mode=args.tbs_mode,
        initial_capital=args.initial_capital,
        risk_per_trade=args.risk,
        max_leverage=args.max_leverage,
        fee_rate_per_side=args.fee,
        slippage_bps=args.slippage_bps,
        min_rr=args.min_rr,
        max_holding_bars=args.max_holding_bars,
        output_dir=args.output_dir,
    )
    validate_config(cfg)

    summaries: list[dict[str, object]] = []
    failures: list[tuple[str, str]] = []
    for symbol in args.symbols:
        try:
            summaries.append(process_symbol(symbol, cfg))
        except Exception as exc:  # keep batch processing other symbols
            failures.append((symbol, str(exc)))
            print(f"\n{symbol}: HATA — {exc}", file=sys.stderr)

    out_dir = Path(cfg.output_dir)
    out_dir.mkdir(parents=True, exist_ok=True)
    if summaries:
        summary_df = pd.DataFrame(summaries)
        summary_df.to_csv(out_dir / "summary.csv", index=False)
        print("\nTOPLU ÖZET")
        display_cols = [
            "symbol",
            "trades",
            "win_rate",
            "net_pnl",
            "total_return",
            "max_drawdown",
            "profit_factor",
            "avg_r",
        ]
        print(summary_df[display_cols].to_string(index=False))

    if failures:
        print("\nBaşarısız semboller:", file=sys.stderr)
        for symbol, message in failures:
            print(f"- {symbol}: {message}", file=sys.stderr)

    return 0 if summaries else 1


if __name__ == "__main__":
    raise SystemExit(main())
```
