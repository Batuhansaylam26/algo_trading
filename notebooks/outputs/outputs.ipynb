```python
from pathlib import Path
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    from IPython.display import display, Image
except Exception:
    display = print
    Image = None


PROJECT_ROOT = Path(
    "/Users/batuhansaylam/Documents/Codex/2026-07-04/a/outputs/yahooquery_lakehouse_revamp"
)

ARTIFACT_ROOTS = [
    PROJECT_ROOT / "artifacts" / "stock_close_training",
    PROJECT_ROOT / "kedro_project" / "artifacts" / "stock_close_training",
]

ARTIFACT_ROOTS = [path for path in ARTIFACT_ROOTS if path.exists()]

print("Artifact roots:")
for root in ARTIFACT_ROOTS:
    print(root)


def title(text: str):
    print("\n" + "=" * 90)
    print(text)
    print("=" * 90)


def infer_family_tier(path: Path, root: Path) -> tuple[str | None, str | None]:
    rel = path.relative_to(root)
    parts = rel.parts
    family = parts[0] if len(parts) > 0 else None
    tier = parts[1] if len(parts) > 1 else None
    return family, tier


def read_metric_files(filename: str) -> pd.DataFrame:
    frames = []

    for root in ARTIFACT_ROOTS:
        for path in sorted(root.glob(f"**/{filename}")):
            try:
                df = pd.read_csv(path)
                family, tier = infer_family_tier(path, root)
                df["_artifact_root"] = str(root)
                df["_source_file"] = str(path.relative_to(root))
                df["_family"] = family
                df["_tier"] = tier
                frames.append(df)
            except Exception as exc:
                print(f"Skipped {path}: {exc}")

    if not frames:
        return pd.DataFrame()

    return pd.concat(frames, ignore_index=True, sort=False)


def clean_metric_columns(df: pd.DataFrame) -> pd.DataFrame:
    if df.empty:
        return df

    df = df.copy()
    df.columns = [str(col).strip() for col in df.columns]

    numeric_cols = [
        "mae",
        "rmse",
        "mape",
        "r2",
        "rows",
        "long_accuracy",
        "long_precision",
        "long_recall",
        "long_signal_rate",
    ]

    for col in numeric_cols:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")

    if "model" in df.columns:
        df = df[df["model"].astype(str).ne("previous_actual_close")]

    return df


def show_image(path: Path, width: int = 1100):
    if path.exists() and Image is not None:
        display(Image(filename=str(path), width=width))
    elif path.exists():
        print(path)
    else:
        print(f"Missing image: {path}")


def find_forecast_plot(row: pd.Series) -> Path | None:
    root = Path(row["_artifact_root"])
    family = row["_family"]
    tier = row["_tier"]
    ticker = str(row.get("unique_id", ""))
    model = str(row.get("model", ""))

    candidates = []

    if family == "pecnet":
        candidates.extend(
            [
                root / family / tier / "plots" / ticker / "forecasts" / f"{model}_forecast.png",
                root / family / tier / "plots" / ticker / "forecasts" / "PECNet_forecast.png",
                root / family / tier / "tickers" / ticker / "plots" / ticker / "forecasts" / "PECNet_forecast.png",
            ]
        )
    else:
        candidates.extend(
            [
                root / family / tier / "plots" / "forecasts" / f"{model}_forecast.png",
                root / family / tier / "plots" / ticker / "forecasts" / f"{model}_forecast.png",
            ]
        )

    for candidate in candidates:
        if candidate.exists():
            return candidate

    return None


def find_confusion_plot(row: pd.Series) -> Path | None:
    root = Path(row["_artifact_root"])
    family = row["_family"]
    tier = row["_tier"]
    ticker = str(row.get("unique_id", ""))
    model = str(row.get("model", ""))

    candidates = [
        root / family / tier / "plots" / ticker / "confusion_matrices" / f"{model}_long_confusion_matrix.png",
        root / family / tier / "plots" / ticker / "confusion_matrices" / "PECNet_long_confusion_matrix.png",
    ]

    for candidate in candidates:
        if candidate.exists():
            return candidate

    return None


title("Load Forecasting Metrics")

regression_metrics = clean_metric_columns(read_metric_files("regression_metrics.csv"))
long_metrics = clean_metric_columns(read_metric_files("long_direction_metrics.csv"))
feature_selection = read_metric_files("feature_selection.csv")

print(f"Regression metric rows: {len(regression_metrics):,}")
print(f"Long-direction metric rows: {len(long_metrics):,}")
print(f"Feature-selection rows: {len(feature_selection):,}")

display(regression_metrics.head(20))
display(long_metrics.head(20))
display(feature_selection.head(20))


title("PECNet Regression Performance")

pecnet_regression = regression_metrics[
    regression_metrics["_family"].eq("pecnet")
].copy()

pecnet_cols = [
    "_tier",
    "_family",
    "unique_id",
    "model",
    "mae",
    "rmse",
    "mape",
    "r2",
    "rows",
    "_source_file",
    "_artifact_root",
]

pecnet_regression = pecnet_regression[
    [col for col in pecnet_cols if col in pecnet_regression.columns]
].sort_values(["unique_id", "rmse", "mae"], ascending=[True, True, True])

display(pecnet_regression)

best_pecnet_by_rmse = (
    pecnet_regression
    .dropna(subset=["rmse"])
    .sort_values(["unique_id", "rmse", "mae"], ascending=[True, True, True])
    .groupby("unique_id", as_index=False)
    .head(1)
)

best_pecnet_by_r2 = (
    pecnet_regression
    .dropna(subset=["r2"])
    .sort_values(["unique_id", "r2", "rmse"], ascending=[True, False, True])
    .groupby("unique_id", as_index=False)
    .head(1)
)

print("Best PECNet by RMSE")
display(best_pecnet_by_rmse)

print("Best PECNet by R2")
display(best_pecnet_by_r2)


title("All Forecasting Models")

comparison_cols = [
    "_tier",
    "_family",
    "unique_id",
    "model",
    "mae",
    "rmse",
    "mape",
    "r2",
    "rows",
    "_source_file",
]

comparison = regression_metrics[
    [col for col in comparison_cols if col in regression_metrics.columns]
].copy()

comparison = comparison.sort_values(
    ["unique_id", "rmse", "mae"],
    ascending=[True, True, True],
)

display(comparison)


title("Ticker-Level Regression Metric Plots")

for ticker, ticker_df in comparison.groupby("unique_id", observed=True):
    ticker_df = ticker_df.copy()
    ticker_df["label"] = (
        ticker_df["_tier"].astype(str)
        + " | "
        + ticker_df["_family"].astype(str)
        + " | "
        + ticker_df["model"].astype(str)
    )

    fig, axes = plt.subplots(2, 2, figsize=(18, 10))
    fig.suptitle(f"{ticker}: All Regression Metrics by Model and Tier", fontsize=16)

    metric_specs = [
        ("mae", "MAE", True),
        ("rmse", "RMSE", True),
        ("mape", "MAPE", True),
        ("r2", "R2", False),
    ]

    for ax, (metric, metric_title, lower_is_better) in zip(axes.ravel(), metric_specs):
        metric_df = ticker_df.dropna(subset=[metric]).copy()
        metric_df = metric_df.sort_values(metric, ascending=lower_is_better)

        color = "#2563eb" if metric != "r2" else "#16a34a"
        ax.barh(metric_df["label"], metric_df[metric], color=color, alpha=0.85)
        ax.set_title(metric_title)
        ax.grid(axis="x", alpha=0.25)

        for index, value in enumerate(metric_df[metric]):
            ax.text(value, index, f" {value:.3f}", va="center", fontsize=8)

        if lower_is_better:
            ax.invert_yaxis()

    plt.tight_layout()
    plt.show()


title("Best PECNet Forecast Plots")

for _, row in best_pecnet_by_rmse.iterrows():
    plot_path = find_forecast_plot(row)
    print(f"{row['unique_id']} | {row['_tier']} | PECNet")
    if plot_path:
        show_image(plot_path)
    else:
        print("Forecast plot not found.")


title("Best PECNet Confusion Matrices")

for _, row in best_pecnet_by_rmse.iterrows():
    plot_path = find_confusion_plot(row)
    print(f"{row['unique_id']} | {row['_tier']} | PECNet")
    if plot_path:
        show_image(plot_path, width=650)
    else:
        print("Confusion matrix not found.")


title("PECNet Feature Selection")

if feature_selection.empty:
    print("No PECNet feature-selection CSV found.")
else:
    display(feature_selection)

for root in ARTIFACT_ROOTS:
    heatmaps = sorted((root / "pecnet").glob("tier*/feature_selection/*.png"))
    for heatmap in heatmaps:
        print(heatmap)
        show_image(heatmap, width=900)


title("Traditional Gap Trading From TimescaleDB")

TIMESCALE_CONFIG = {
    "host": os.getenv("TIMESCALE_HOST", "127.0.0.1"),
    "port": int(os.getenv("TIMESCALE_PORT", "5432")),
    "dbname": os.getenv("TIMESCALE_DB", "dataops"),
    "user": os.getenv("TIMESCALE_USER", "dataops"),
    "password": os.getenv("TIMESCALE_PASSWORD", "dataops"),
}

CONVENTIONAL_GAP_SQL = """
SELECT *
FROM feature_store.conventional_gap_trading
ORDER BY symbol, date;
"""


def load_conventional_gap_trading_from_timescale(config: dict) -> pd.DataFrame:
    try:
        from sqlalchemy import create_engine

        url = (
            f"postgresql+psycopg2://{config['user']}:{config['password']}"
            f"@{config['host']}:{config['port']}/{config['dbname']}"
        )
        engine = create_engine(url)
        return pd.read_sql_query(CONVENTIONAL_GAP_SQL, engine)
    except Exception as exc:
        print(f"SQLAlchemy path failed, trying psycopg2 directly: {exc}")

    import psycopg2

    with psycopg2.connect(connect_timeout=5, **config) as connection:
        return pd.read_sql_query(CONVENTIONAL_GAP_SQL, connection)


try:
    conventional_gap_df = load_conventional_gap_trading_from_timescale(TIMESCALE_CONFIG)
    conventional_gap_df["date"] = pd.to_datetime(
        conventional_gap_df["date"],
        utc=True,
    ).dt.tz_localize(None)
    conventional_gap_df = conventional_gap_df.sort_values(
        ["symbol", "date"]
    ).reset_index(drop=True)

    print(f"Conventional gap rows: {len(conventional_gap_df):,}")
    print(f"Symbols: {conventional_gap_df['symbol'].nunique()}")
    print(
        f"Date range: {conventional_gap_df['date'].min()} "
        f"-> {conventional_gap_df['date'].max()}"
    )

    display(conventional_gap_df.head(20))
except Exception as exc:
    conventional_gap_df = pd.DataFrame()
    print(f"Could not load conventional gap trading table: {exc}")


if not conventional_gap_df.empty:
    title("Traditional Gap Trading Metrics")

    gap = conventional_gap_df.copy()

    gap["prev_high"] = gap.groupby("symbol")["high"].shift(1)
    gap["prev_low"] = gap.groupby("symbol")["low"].shift(1)
    gap["next_close"] = gap.groupby("symbol")["close"].shift(-1)
    gap["next_return_pct"] = (gap["next_close"] / gap["close"] - 1.0) * 100.0

    long_signals = {
        "Breakaway Up",
        "Runaway Up",
        "Common Down",
        "Exhaustion Down",
    }

    short_signals = {
        "Breakaway Down",
        "Runaway Down",
        "Common Up",
        "Exhaustion Up",
    }

    gap["strategy_side"] = np.select(
        [
            gap["Gap_Type"].isin(long_signals),
            gap["Gap_Type"].isin(short_signals),
        ],
        [
            "long",
            "short",
        ],
        default="no_trade",
    )

    gap["strategy_return_pct"] = np.select(
        [
            gap["strategy_side"].eq("long"),
            gap["strategy_side"].eq("short"),
        ],
        [
            gap["next_return_pct"],
            -gap["next_return_pct"],
        ],
        default=np.nan,
    )

    gap["strategy_win"] = gap["strategy_return_pct"] > 0

    gap["common_fill_target"] = np.select(
        [
            gap["Gap_Type"].eq("Common Up"),
            gap["Gap_Type"].eq("Common Down"),
        ],
        [
            gap["prev_high"],
            gap["prev_low"],
        ],
        default=np.nan,
    )

    gap["common_gap_filled_same_day"] = np.select(
        [
            gap["Gap_Type"].eq("Common Up"),
            gap["Gap_Type"].eq("Common Down"),
        ],
        [
            gap["low"] <= gap["prev_high"],
            gap["high"] >= gap["prev_low"],
        ],
        default=np.nan,
    )

    gap_type_summary = (
        gap.groupby(["symbol", "Gap_Type"], observed=True)
        .agg(
            signals=("Gap_Type", "size"),
            avg_next_return_pct=("next_return_pct", "mean"),
            avg_strategy_return_pct=("strategy_return_pct", "mean"),
            strategy_win_rate=("strategy_win", "mean"),
        )
        .reset_index()
        .sort_values(["symbol", "signals"], ascending=[True, False])
    )

    common_gap_fill_summary = (
        gap[gap["Gap_Type"].isin(["Common Up", "Common Down"])]
        .groupby(["symbol", "Gap_Type"], observed=True)
        .agg(
            signals=("Gap_Type", "size"),
            same_day_fill_rate=("common_gap_filled_same_day", "mean"),
            avg_strategy_return_pct=("strategy_return_pct", "mean"),
            strategy_win_rate=("strategy_win", "mean"),
        )
        .reset_index()
        .sort_values(["symbol", "Gap_Type"])
    )

    directional_strategy_summary = (
        gap[gap["strategy_side"].ne("no_trade")]
        .groupby(["symbol", "strategy_side"], observed=True)
        .agg(
            trades=("strategy_side", "size"),
            avg_strategy_return_pct=("strategy_return_pct", "mean"),
            median_strategy_return_pct=("strategy_return_pct", "median"),
            strategy_win_rate=("strategy_win", "mean"),
        )
        .reset_index()
        .sort_values(["symbol", "strategy_side"])
    )

    print("Gap type summary")
    display(gap_type_summary)

    print("Common gap fill summary")
    display(common_gap_fill_summary)

    print("Directional strategy summary")
    display(directional_strategy_summary)


if not conventional_gap_df.empty:
    title("Traditional Gap Trading Plots")

    for symbol, symbol_df in gap.groupby("symbol", observed=True):
        fig, axes = plt.subplots(1, 3, figsize=(20, 5))
        fig.suptitle(f"{symbol}: Conventional Gap Trading Evidence", fontsize=15)

        signal_counts = (
            symbol_df["Gap_Type"]
            .value_counts()
            .drop(labels=["None"], errors="ignore")
            .sort_values()
        )

        signal_counts.plot(kind="barh", ax=axes[0], color="#3b82f6")
        axes[0].set_title("Gap Signal Counts")
        axes[0].set_xlabel("Count")
        axes[0].grid(axis="x", alpha=0.25)

        symbol_common = common_gap_fill_summary[
            common_gap_fill_summary["symbol"].eq(symbol)
        ]

        if not symbol_common.empty:
            axes[1].bar(
                symbol_common["Gap_Type"],
                symbol_common["same_day_fill_rate"],
                color="#10b981",
            )
            axes[1].set_ylim(0, 1)
            axes[1].set_title("Common Gap Same-Day Fill Rate")
            axes[1].set_ylabel("Fill Rate")
            axes[1].grid(axis="y", alpha=0.25)
        else:
            axes[1].set_title("No Common Gap Rows")
            axes[1].axis("off")

        symbol_strategy = directional_strategy_summary[
            directional_strategy_summary["symbol"].eq(symbol)
        ]

        if not symbol_strategy.empty:
            axes[2].bar(
                symbol_strategy["strategy_side"],
                symbol_strategy["strategy_win_rate"],
                color="#f97316",
            )
            axes[2].set_ylim(0, 1)
            axes[2].set_title("Directional Strategy Win Rate")
            axes[2].set_ylabel("Win Rate")
            axes[2].grid(axis="y", alpha=0.25)
        else:
            axes[2].set_title("No Tradable Gap Rows")
            axes[2].axis("off")

        plt.tight_layout()
        plt.show()


title("Article-Level Comparison Table")

paper_comparison = comparison.copy()
paper_comparison["evidence_type"] = "forecasting_model"

if not conventional_gap_df.empty:
    gap_article_rows = directional_strategy_summary.copy()
    gap_article_rows = gap_article_rows.rename(
        columns={
            "symbol": "unique_id",
            "trades": "rows",
        }
    )
    gap_article_rows["_tier"] = "traditional"
    gap_article_rows["_family"] = "gap_trading"
    gap_article_rows["model"] = "Conventional Gap Rules"
    gap_article_rows["evidence_type"] = "trading_rule"
    gap_article_rows["mae"] = np.nan
    gap_article_rows["rmse"] = np.nan
    gap_article_rows["mape"] = np.nan
    gap_article_rows["r2"] = np.nan

    paper_comparison = pd.concat(
        [
            paper_comparison,
            gap_article_rows[
                [
                    "_tier",
                    "_family",
                    "unique_id",
                    "model",
                    "mae",
                    "rmse",
                    "mape",
                    "r2",
                    "rows",
                    "strategy_side",
                    "avg_strategy_return_pct",
                    "median_strategy_return_pct",
                    "strategy_win_rate",
                    "evidence_type",
                ]
            ],
        ],
        ignore_index=True,
        sort=False,
    )

display(paper_comparison)


title("Auto Article Notes")

for _, row in best_pecnet_by_rmse.iterrows():
    print(
        f"- For {row['unique_id']}, the best PECNet configuration by RMSE is "
        f"{row['_tier']} with MAE={row['mae']:.3f}, RMSE={row['rmse']:.3f}, "
        f"MAPE={row['mape']:.3f}, and R2={row['r2']:.3f}."
    )

if not conventional_gap_df.empty:
    for _, row in directional_strategy_summary.iterrows():
        print(
            f"- For {row['symbol']}, the traditional gap trading {row['strategy_side']} "
            f"side produced {int(row['trades'])} trades, an average next-day strategy "
            f"return of {row['avg_strategy_return_pct']:.3f}%, and a win rate of "
            f"{row['strategy_win_rate']:.3f}."
        )

print("- PECNet should be presented as a learned forecasting model.")
print("- Conventional gap trading should be presented as a rule-based trading baseline.")
print("- The comparison should separate forecasting accuracy from trading-rule profitability.")
print("- Common gaps test mean reversion; breakaway and runaway gaps test continuation.")
print("- The strongest article argument is whether PECNet captures temporal structure more consistently than handcrafted gap rules.")
```

```python
# Clean paper-quality forecast plots from MLflow artifacts
# Uses MLflow prediction/data tables and redraws figures from scratch.

from pathlib import Path
import os
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import mlflow
from mlflow.tracking import MlflowClient

try:
    from IPython.display import display
except Exception:
    display = print


# -----------------------------
# Environment
# -----------------------------

# Local Mac:
os.environ.setdefault("MLFLOW_TRACKING_URI", "http://127.0.0.1:5001")
os.environ.setdefault("MLFLOW_S3_ENDPOINT_URL", "http://127.0.0.1:9000")

# Container içindeysen üstteki iki satırı şuna çevir:
# os.environ["MLFLOW_TRACKING_URI"] = "http://host.docker.internal:5001"
# os.environ["MLFLOW_S3_ENDPOINT_URL"] = "http://host.docker.internal:9000"

os.environ.setdefault("AWS_ACCESS_KEY_ID", "admin")
os.environ.setdefault("AWS_SECRET_ACCESS_KEY", "admin1234")
os.environ.setdefault("AWS_DEFAULT_REGION", "us-east-1")

MLFLOW_TRACKING_URI = os.environ["MLFLOW_TRACKING_URI"]
mlflow.set_tracking_uri(MLFLOW_TRACKING_URI)
client = MlflowClient(tracking_uri=MLFLOW_TRACKING_URI)

OUTPUT_DIR = Path.cwd() / "paper_mlflow_forecast_plots"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

TRAIN_TAIL = 120
FALLBACK_TRAIN_FROM_TIMESCALE = True

SELECTED_MODELS = [
    {"unique_id": "AAPL", "family": "pecnet", "tier": "tier8", "model": "PECNet"},
    {"unique_id": "BMW.DE", "family": "pecnet", "tier": "tier7", "model": "PECNet"},
    {"unique_id": "AAPL", "family": "statsforecast", "tier": "tier1", "model": "AutoARIMA"},
    {"unique_id": "AAPL", "family": "statsforecast", "tier": "tier1", "model": "AutoRegressive"},
    {"unique_id": "BMW.DE", "family": "statsforecast", "tier": "tier2", "model": "AutoARIMA"},
    {"unique_id": "BMW.DE", "family": "statsforecast", "tier": "tier2", "model": "AutoRegressive"},
]

print("MLflow:", MLFLOW_TRACKING_URI)
print("S3 endpoint:", os.environ["MLFLOW_S3_ENDPOINT_URL"])
print("Output:", OUTPUT_DIR)


# -----------------------------
# Helpers
# -----------------------------

def title(text):
    print("\n" + "=" * 90)
    print(text)
    print("=" * 90)


def normalize_dates(series):
    return pd.to_datetime(series, errors="coerce", utc=True).dt.tz_localize(None)


def list_all_runs():
    rows = []

    for experiment in client.search_experiments():
        runs = client.search_runs(
            experiment_ids=[experiment.experiment_id],
            filter_string="",
            order_by=["attributes.start_time DESC"],
            max_results=5000,
        )

        for run in runs:
            rows.append(
                {
                    "experiment_id": experiment.experiment_id,
                    "experiment_name": experiment.name,
                    "run_id": run.info.run_id,
                    "run_name": run.data.tags.get("mlflow.runName", ""),
                    "status": run.info.status,
                    "start_time": run.info.start_time or 0,
                    "family": run.data.tags.get("model_family") or run.data.params.get("model_family") or "",
                    "tier": run.data.tags.get("tier_name") or run.data.params.get("tier_name") or "",
                    "artifact_uri": run.info.artifact_uri,
                }
            )

    return pd.DataFrame(rows)


def artifact_path_for_predictions(family, tier, unique_id):
    if family == "pecnet":
        return f"pecnet/{tier}/predictions/{unique_id}.json"
    return f"{family}/{tier}/evaluation/predictions.json"


def artifact_path_for_train(family, tier):
    return f"{family}/{tier}/datasets/train.json"


def artifact_path_for_test(family, tier):
    return f"{family}/{tier}/datasets/test.json"


def download_artifact(run_id, artifact_path):
    try:
        return Path(client.download_artifacts(run_id, artifact_path))
    except Exception:
        return None


def read_mlflow_table(path):
    with path.open("r", encoding="utf-8") as file:
        raw = json.load(file)

    if isinstance(raw, list):
        return pd.DataFrame(raw)

    if isinstance(raw, dict):
        if "columns" in raw and "data" in raw:
            return pd.DataFrame(raw["data"], columns=raw["columns"])
        if "data" in raw and isinstance(raw["data"], list):
            return pd.DataFrame(raw["data"])
        return pd.DataFrame([raw])

    return pd.read_json(path)


def artifact_table(run_id, artifact_path):
    path = download_artifact(run_id, artifact_path)
    if path is None:
        return pd.DataFrame()

    try:
        return read_mlflow_table(path)
    except Exception as exc:
        print(f"Could not read artifact {artifact_path}: {exc}")
        return pd.DataFrame()


def normalize_frame(df):
    if df.empty:
        return pd.DataFrame(columns=["unique_id", "ds", "y"])

    df = df.copy()
    df.columns = [str(c).strip() for c in df.columns]

    rename_map = {}
    if "date" in df.columns and "ds" not in df.columns:
        rename_map["date"] = "ds"
    if "symbol" in df.columns and "unique_id" not in df.columns:
        rename_map["symbol"] = "unique_id"
    if "actual" in df.columns and "y" not in df.columns:
        rename_map["actual"] = "y"
    if "target_close" in df.columns and "y" not in df.columns:
        rename_map["target_close"] = "y"
    if "close" in df.columns and "y" not in df.columns:
        rename_map["close"] = "y"

    df = df.rename(columns=rename_map)

    if "ds" in df.columns:
        df["ds"] = normalize_dates(df["ds"])

    for col in df.columns:
        if col not in {"unique_id", "ds", "model", "family", "tier"}:
            converted = pd.to_numeric(df[col], errors="coerce")
            if converted.notna().sum() > 0:
                df[col] = converted

    sort_cols = [c for c in ["unique_id", "ds"] if c in df.columns]
    if sort_cols:
        df = df.sort_values(sort_cols)

    return df.reset_index(drop=True)


def ticker_slice(df, unique_id):
    if df.empty:
        return df
    if "unique_id" not in df.columns:
        return df
    return df[df["unique_id"].astype(str).eq(unique_id)].copy()


def model_prediction_columns(df):
    ignored = {
        "unique_id",
        "ds",
        "y",
        "actual",
        "cutoff",
        "model",
        "family",
        "tier",
        "_tier",
        "_family",
    }

    interval_tokens = [
        "-lo-",
        "-hi-",
        "_lo_",
        "_hi_",
        "_lower_",
        "_upper_",
    ]

    cols = []
    for col in df.columns:
        if col in ignored:
            continue
        if any(token in col for token in interval_tokens):
            continue
        if pd.api.types.is_numeric_dtype(df[col]):
            cols.append(col)

    return cols


def choose_prediction_column(pred_df, requested_model):
    if requested_model in pred_df.columns:
        return requested_model

    lower_map = {str(c).lower(): c for c in pred_df.columns}
    if requested_model.lower() in lower_map:
        return lower_map[requested_model.lower()]

    cols = model_prediction_columns(pred_df)
    if not cols:
        raise ValueError(f"No prediction column found. Columns={list(pred_df.columns)}")

    print(f"Requested model '{requested_model}' not found. Using '{cols[0]}'")
    return cols[0]


def interval_columns_for(df, model_col):
    intervals = {}

    for level in [80, 90, 95]:
        lo_candidates = [
            f"{model_col}-lo-{level}",
            f"{model_col}_lo_{level}",
            f"{model_col}_lower_{level}",
        ]
        hi_candidates = [
            f"{model_col}-hi-{level}",
            f"{model_col}_hi_{level}",
            f"{model_col}_upper_{level}",
        ]

        lo_col = next((c for c in lo_candidates if c in df.columns), None)
        hi_col = next((c for c in hi_candidates if c in df.columns), None)

        if lo_col and hi_col:
            intervals[level] = (lo_col, hi_col)

    return intervals


def candidate_runs(family, tier, unique_id):
    runs = list_all_runs()
    if runs.empty:
        raise ValueError("No MLflow runs found.")

    family_lower = family.lower()

    mask_family = (
        runs["family"].astype(str).str.lower().eq(family_lower)
        | runs["run_name"].astype(str).str.lower().str.contains(family_lower, na=False)
        | runs["experiment_name"].astype(str).str.lower().str.contains(family_lower, na=False)
    )

    mask_tier = (
        runs["tier"].astype(str).eq(tier)
        | runs["run_name"].astype(str).str.contains(tier, na=False)
        | runs["experiment_name"].astype(str).str.contains(tier, na=False)
    )

    candidates = runs[mask_family & mask_tier].copy()

    if candidates.empty:
        print("Available runs:")
        display(runs[["experiment_name", "run_name", "family", "tier", "status", "run_id"]].head(120))
        raise ValueError(f"No candidate runs for family={family}, tier={tier}")

    candidates["ticker_match"] = candidates["run_name"].astype(str).str.contains(unique_id, na=False)
    candidates["finished_rank"] = np.where(candidates["status"].eq("FINISHED"), 0, 1)

    return candidates.sort_values(
        ["ticker_match", "finished_rank", "start_time"],
        ascending=[False, True, False],
    )


def load_run_data(family, tier, unique_id):
    pred_path = artifact_path_for_predictions(family, tier, unique_id)
    candidates = candidate_runs(family, tier, unique_id)

    last_error = None

    for _, run in candidates.iterrows():
        run_id = run["run_id"]

        pred_df = artifact_table(run_id, pred_path)
        pred_df = normalize_frame(pred_df)

        if pred_df.empty:
            continue

        if "unique_id" in pred_df.columns:
            pred_ticker = ticker_slice(pred_df, unique_id)
            if pred_ticker.empty:
                continue

        train_df = normalize_frame(artifact_table(run_id, artifact_path_for_train(family, tier)))
        test_df = normalize_frame(artifact_table(run_id, artifact_path_for_test(family, tier)))

        print(
            f"Using run | {unique_id} | {family} | {tier} | "
            f"{run['run_name']} | {run_id}"
        )

        return {
            "run": run,
            "run_id": run_id,
            "predictions": pred_df,
            "train": train_df,
            "test": test_df,
        }

    raise ValueError(
        f"No run with prediction artifact found for {unique_id} | {family} | {tier}. "
        f"Expected artifact: {pred_path}. Last error={last_error}"
    )


# -----------------------------
# Optional Timescale fallback
# -----------------------------

def load_actual_close_from_timescale(unique_id):
    config = {
        "host": os.getenv("TIMESCALE_HOST", "127.0.0.1"),
        "port": int(os.getenv("TIMESCALE_PORT", "5432")),
        "dbname": os.getenv("TIMESCALE_DB", "dataops"),
        "user": os.getenv("TIMESCALE_USER", "dataops"),
        "password": os.getenv("TIMESCALE_PASSWORD", "dataops"),
    }

    sql = """
    SELECT unique_id, ds, y
    FROM feature_store.stock_close_model_dataset
    WHERE unique_id = %(unique_id)s
    ORDER BY ds;
    """

    try:
        from sqlalchemy import create_engine

        url = (
            f"postgresql+psycopg2://{config['user']}:{config['password']}"
            f"@{config['host']}:{config['port']}/{config['dbname']}"
        )
        engine = create_engine(url)
        df = pd.read_sql_query(sql, engine, params={"unique_id": unique_id})
        return normalize_frame(df)
    except Exception as exc:
        print(f"Timescale fallback failed for {unique_id}: {exc}")
        return pd.DataFrame(columns=["unique_id", "ds", "y"])


def build_plot_frame(family, tier, unique_id, model, train_df, test_df, pred_df):
    pred_ticker = ticker_slice(pred_df, unique_id)

    if pred_ticker.empty:
        raise ValueError(
            f"No prediction rows for {unique_id}. "
            f"Prediction columns={list(pred_df.columns)}"
        )

    pred_col = choose_prediction_column(pred_ticker, model)

    if "ds" not in pred_ticker.columns:
        raise ValueError(f"Prediction table has no ds column. Columns={list(pred_ticker.columns)}")

    if "y" not in pred_ticker.columns or pred_ticker["y"].isna().all():
        test_ticker = ticker_slice(test_df, unique_id)

        if not test_ticker.empty and {"unique_id", "ds", "y"}.issubset(test_ticker.columns):
            pred_ticker = pred_ticker.drop(columns=["y"], errors="ignore").merge(
                test_ticker[["unique_id", "ds", "y"]],
                on=["unique_id", "ds"],
                how="left",
            )

        if ("y" not in pred_ticker.columns or pred_ticker["y"].isna().all()) and FALLBACK_TRAIN_FROM_TIMESCALE:
            actual_series = load_actual_close_from_timescale(unique_id)
            if not actual_series.empty:
                pred_ticker = pred_ticker.drop(columns=["y"], errors="ignore").merge(
                    actual_series[["unique_id", "ds", "y"]],
                    on=["unique_id", "ds"],
                    how="left",
                )

    if "y" not in pred_ticker.columns:
        raise ValueError("Could not find actual y values for test period.")

    train_ticker = ticker_slice(train_df, unique_id)

    if (
        train_ticker.empty
        or not {"unique_id", "ds", "y"}.issubset(train_ticker.columns)
        or train_ticker["y"].isna().all()
    ):
        print(
            f"Train artifact missing/incomplete for {unique_id} | {family} | {tier}. "
            "Trying Timescale fallback."
        )

        if FALLBACK_TRAIN_FROM_TIMESCALE:
            actual_series = load_actual_close_from_timescale(unique_id)
            if not actual_series.empty:
                test_start = pred_ticker["ds"].min()
                train_ticker = actual_series[actual_series["ds"] < test_start].copy()
            else:
                train_ticker = pd.DataFrame(columns=["unique_id", "ds", "y"])
        else:
            train_ticker = pd.DataFrame(columns=["unique_id", "ds", "y"])

    pred_ticker = pred_ticker.sort_values("ds").reset_index(drop=True)

    if "ds" in train_ticker.columns:
        train_ticker = train_ticker.sort_values("ds").reset_index(drop=True)
    else:
        train_ticker = pd.DataFrame(columns=["unique_id", "ds", "y"])

    return train_ticker, pred_ticker, pred_col


# -----------------------------
# Plotting
# -----------------------------

def plot_single_forecast(item, show_intervals=True, save=True):
    unique_id = item["unique_id"]
    family = item["family"]
    tier = item["tier"]
    model = item["model"]
    train_df = item["train"].tail(TRAIN_TAIL)
    test_df = item["test_joined"]
    pred_col = item["pred_col"]

    fig, ax = plt.subplots(figsize=(18, 5))

    if not train_df.empty:
        ax.plot(
            train_df["ds"],
            train_df["y"],
            color="#64748b",
            linewidth=2.0,
            label=f"Train actual (last {TRAIN_TAIL})",
        )

    ax.plot(
        test_df["ds"],
        test_df["y"],
        color="#111827",
        linewidth=2.0,
        marker="o",
        markersize=3,
        label="Test actual",
    )

    if show_intervals:
        intervals = interval_columns_for(test_df, pred_col)
        for level, (lo_col, hi_col) in sorted(intervals.items(), reverse=True):
            ax.fill_between(
                test_df["ds"],
                test_df[lo_col].astype(float),
                test_df[hi_col].astype(float),
                alpha=0.16 if level >= 95 else 0.24,
                label=f"{level}% prediction interval",
            )

    ax.plot(
        test_df["ds"],
        test_df[pred_col],
        color="#f97316",
        linewidth=2.0,
        marker="x",
        markersize=4,
        label="Prediction",
    )

    ax.axvline(
        test_df["ds"].min(),
        color="#2563eb",
        linestyle="--",
        linewidth=1.2,
        alpha=0.75,
        label="Test start",
    )

    ax.set_title(f"{unique_id} | {model} | {tier} | {family}", fontsize=14)
    ax.set_xlabel("Date", fontsize=12)
    ax.set_ylabel("Close", fontsize=12)
    ax.grid(True, alpha=0.22)
    ax.legend(loc="best", frameon=True)
    fig.autofmt_xdate()
    plt.tight_layout()

    if save:
        filename = f"{unique_id}_{family}_{tier}_{model}_clean_forecast.png".replace("/", "_")
        path = OUTPUT_DIR / filename
        fig.savefig(path, dpi=240, bbox_inches="tight")
        print(f"Saved: {path}")

    plt.show()


def plot_overlay(unique_id, loaded_models, show_intervals=True):
    fig, ax = plt.subplots(figsize=(18, 5))

    colors = ["#f97316", "#7c3aed", "#2563eb", "#dc2626", "#16a34a", "#0891b2"]
    actual_drawn = False
    test_start_drawn = False

    ticker_models = [item for item in loaded_models if item["unique_id"] == unique_id]

    for idx, item in enumerate(ticker_models):
        test_df = item["test_joined"]
        pred_col = item["pred_col"]
        color = colors[idx % len(colors)]

        label_base = f"{item['tier']} | {item['family']} | {item['model']}"

        if not actual_drawn:
            ax.plot(
                test_df["ds"],
                test_df["y"],
                color="#111827",
                linewidth=2.2,
                marker="o",
                markersize=3,
                label="Test actual",
            )
            actual_drawn = True

        if show_intervals:
            intervals = interval_columns_for(test_df, pred_col)
            for level, (lo_col, hi_col) in sorted(intervals.items(), reverse=True):
                ax.fill_between(
                    test_df["ds"],
                    test_df[lo_col].astype(float),
                    test_df[hi_col].astype(float),
                    color=color,
                    alpha=0.08 if level >= 95 else 0.13,
                    label=f"{label_base} {level}% interval",
                )

        ax.plot(
            test_df["ds"],
            test_df[pred_col],
            color=color,
            linewidth=2.0,
            marker="x",
            markersize=3,
            label=f"{label_base} prediction",
        )

        if not test_start_drawn:
            ax.axvline(
                test_df["ds"].min(),
                color="#2563eb",
                linestyle="--",
                linewidth=1.1,
                alpha=0.7,
                label="Test start",
            )
            test_start_drawn = True

    ax.set_title(
        f"{unique_id}: Selected Forecast Overlay"
        + (" with Intervals" if show_intervals else " Point Forecasts Only"),
        fontsize=14,
    )
    ax.set_xlabel("Date", fontsize=12)
    ax.set_ylabel("Close", fontsize=12)
    ax.grid(True, alpha=0.22)
    ax.legend(loc="best", fontsize=8, frameon=True)
    fig.autofmt_xdate()
    plt.tight_layout()

    suffix = "with_intervals" if show_intervals else "point_only"
    path = OUTPUT_DIR / f"{unique_id}_forecast_overlay_{suffix}.png"
    fig.savefig(path, dpi=240, bbox_inches="tight")
    print(f"Saved: {path}")

    plt.show()


def calculate_metrics(item):
    df = item["test_joined"].copy()
    pred_col = item["pred_col"]

    y_true = pd.to_numeric(df["y"], errors="coerce")
    y_pred = pd.to_numeric(df[pred_col], errors="coerce")
    mask = y_true.notna() & y_pred.notna()

    y_true = y_true[mask]
    y_pred = y_pred[mask]

    mae = np.mean(np.abs(y_true - y_pred))
    rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100

    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - y_true.mean()) ** 2)
    r2 = 1 - ss_res / ss_tot if ss_tot else np.nan

    return {
        "unique_id": item["unique_id"],
        "family": item["family"],
        "tier": item["tier"],
        "model": item["model"],
        "run_id": item["run_id"],
        "rows": int(mask.sum()),
        "mae": float(mae),
        "rmse": float(rmse),
        "mape": float(mape),
        "r2": float(r2),
    }


# -----------------------------
# Execute
# -----------------------------

title("Load selected model outputs from MLflow")

loaded_models = []

for spec in SELECTED_MODELS:
    unique_id = spec["unique_id"]
    family = spec["family"]
    tier = spec["tier"]
    model = spec["model"]

    print("\n" + "-" * 90)
    print(f"Loading {unique_id} | {family} | {tier} | {model}")

    data = load_run_data(family, tier, unique_id)

    train_df, test_joined_df, pred_col = build_plot_frame(
        family=family,
        tier=tier,
        unique_id=unique_id,
        model=model,
        train_df=data["train"],
        test_df=data["test"],
        pred_df=data["predictions"],
    )

    item = {
        **spec,
        "run_id": data["run_id"],
        "train": train_df,
        "test_joined": test_joined_df,
        "pred_col": pred_col,
    }

    loaded_models.append(item)
    plot_single_forecast(item, show_intervals=True, save=True)


title("Combined grid")

if loaded_models:
    n = len(loaded_models)
    rows = int(np.ceil(n / 2))
    fig, axes = plt.subplots(rows, 2, figsize=(20, 4.8 * rows), squeeze=False)
    axes = axes.ravel()

    for ax, item in zip(axes, loaded_models):
        train_df = item["train"].tail(TRAIN_TAIL)
        test_df = item["test_joined"]
        pred_col = item["pred_col"]

        if not train_df.empty:
            ax.plot(
                train_df["ds"],
                train_df["y"],
                color="#64748b",
                linewidth=1.8,
                label=f"Train actual (last {TRAIN_TAIL})",
            )

        ax.plot(
            test_df["ds"],
            test_df["y"],
            color="#111827",
            linewidth=1.8,
            marker="o",
            markersize=2.5,
            label="Test actual",
        )

        intervals = interval_columns_for(test_df, pred_col)
        for level, (lo_col, hi_col) in sorted(intervals.items(), reverse=True):
            ax.fill_between(
                test_df["ds"],
                test_df[lo_col].astype(float),
                test_df[hi_col].astype(float),
                alpha=0.14 if level >= 95 else 0.22,
                label=f"{level}% interval",
            )

        ax.plot(
            test_df["ds"],
            test_df[pred_col],
            color="#f97316",
            linewidth=1.8,
            marker="x",
            markersize=3,
            label="Prediction",
        )

        ax.axvline(
            test_df["ds"].min(),
            color="#2563eb",
            linestyle="--",
            linewidth=1,
            alpha=0.7,
            label="Test start",
        )

        ax.set_title(
            f"{item['unique_id']} | {item['model']} | {item['tier']} | {item['family']}",
            fontsize=11,
        )
        ax.set_xlabel("Date")
        ax.set_ylabel("Close")
        ax.grid(True, alpha=0.22)
        ax.legend(fontsize=7, loc="best")

    for ax in axes[len(loaded_models):]:
        ax.axis("off")

    plt.tight_layout()
    grid_path = OUTPUT_DIR / "selected_models_clean_forecast_grid.png"
    fig.savefig(grid_path, dpi=240, bbox_inches="tight")
    print(f"Saved: {grid_path}")
    plt.show()


title("Ticker overlays")

for unique_id in sorted({item["unique_id"] for item in loaded_models}):
    plot_overlay(unique_id, loaded_models, show_intervals=True)
    plot_overlay(unique_id, loaded_models, show_intervals=False)


title("Metrics recalculated from plotted data")

paper_metrics = pd.DataFrame([calculate_metrics(item) for item in loaded_models])
paper_metrics = paper_metrics.sort_values(["unique_id", "rmse"], ascending=[True, True])

display(paper_metrics)

metrics_path = OUTPUT_DIR / "selected_models_clean_metrics.csv"
paper_metrics.to_csv(metrics_path, index=False)
print(f"Saved: {metrics_path}")
```

```python
import os
import json
import re
from pathlib import Path
from datetime import datetime

from IPython.display import display
from mlflow.tracking import MlflowClient

os.environ.setdefault("MLFLOW_TRACKING_URI", "http://127.0.0.1:5001")
os.environ.setdefault("MLFLOW_S3_ENDPOINT_URL", "http://127.0.0.1:9000")
os.environ.setdefault("AWS_ACCESS_KEY_ID", "admin")
os.environ.setdefault("AWS_SECRET_ACCESS_KEY", "admin1234")
os.environ.setdefault("AWS_DEFAULT_REGION", "us-east-1")

client = MlflowClient(tracking_uri=os.environ["MLFLOW_TRACKING_URI"])

SELECTED_MODELS = [
    {"ticker": "AAPL", "family": "pecnet", "tier": "tier8", "model": "PECNet"},
    {"ticker": "AAPL", "family": "pecnet", "tier": "tier4", "model": "PECNet"},
    {"ticker": "AAPL", "family": "statsforecast", "tier": "tier1", "model": "AutoARIMA"},
    {"ticker": "BMW.DE", "family": "pecnet", "tier": "tier7", "model": "PECNet"},
    {"ticker": "BMW.DE", "family": "statsforecast", "tier": "tier2", "model": "AutoARIMA"},
    {"ticker": "BMW.DE", "family": "statsforecast", "tier": "tier2", "model": "AutoRegressive"},
]


def safe_name(value: str) -> str:
    return re.sub(r"[^a-zA-Z0-9_.-]+", "_", str(value))


def expected_run_name(selection: dict) -> str:
    if selection["family"] == "statsforecast":
        return f"stock-close-{selection['tier']}-statsforecast"
    if selection["family"] == "pecnet":
        return f"pecnet-{selection['tier']}-{safe_name(selection['ticker'])}"
    return ""


def artifact_paths(selection: dict) -> dict:
    tier = selection["tier"]
    ticker = selection["ticker"]

    if selection["family"] == "statsforecast":
        return {
            "train": f"statsforecast/{tier}/datasets/train.json",
            "test": f"statsforecast/{tier}/datasets/test.json",
            "predictions": f"statsforecast/{tier}/evaluation/predictions.json",
        }

    return {
        "train": f"pecnet/{tier}/tickers/{safe_name(ticker)}/datasets/train.json",
        "test": f"pecnet/{tier}/tickers/{safe_name(ticker)}/datasets/test.json",
        "predictions": f"pecnet/{tier}/predictions/{ticker}.json",
    }


def load_json_artifact(run_id: str, artifact_path: str):
    try:
        local_path = client.download_artifacts(run_id, artifact_path)
        with open(local_path, "r", encoding="utf-8") as file:
            return json.load(file)
    except Exception as exc:
        return {"_error": str(exc)}


def rows_from_payload(payload):
    if isinstance(payload, dict) and "_error" in payload:
        return []

    if isinstance(payload, list):
        return payload

    if isinstance(payload, dict):
        if "columns" in payload and "data" in payload:
            columns = payload["columns"]
            return [dict(zip(columns, row)) for row in payload["data"]]

        if isinstance(payload.get("data"), list):
            return payload["data"]

        return [payload]

    return []


def parse_date(value):
    if value is None:
        return None

    text = str(value).strip()
    if not text:
        return None

    text = text.replace("Z", "+00:00")

    try:
        return datetime.fromisoformat(text).date()
    except Exception:
        pass

    for fmt in ("%Y-%m-%d", "%Y-%m-%d %H:%M:%S", "%Y/%m/%d"):
        try:
            return datetime.strptime(text[:19], fmt).date()
        except Exception:
            continue

    return None


def extract_date(row: dict):
    for key in ("ds", "date", "event_timestamp", "created_timestamp"):
        if key in row:
            parsed = parse_date(row[key])
            if parsed is not None:
                return parsed
    return None


def extract_symbol(row: dict):
    for key in ("unique_id", "symbol", "ticker"):
        if key in row:
            return str(row[key])
    return None


def summarize_payload(payload, ticker: str | None = None):
    rows = rows_from_payload(payload)

    if ticker is not None:
        filtered = []
        for row in rows:
            symbol = extract_symbol(row)
            if symbol is None or symbol == ticker:
                filtered.append(row)
        rows = filtered

    dates = [extract_date(row) for row in rows]
    dates = [date for date in dates if date is not None]

    return {
        "rows": len(rows),
        "start_date": min(dates).isoformat() if dates else None,
        "end_date": max(dates).isoformat() if dates else None,
    }


def find_matching_run(selection: dict):
    wanted_run_name = expected_run_name(selection)
    candidates = []

    for experiment in client.search_experiments():
        runs = client.search_runs(
            experiment_ids=[experiment.experiment_id],
            order_by=["attributes.start_time DESC"],
            max_results=200,
        )

        for run in runs:
            run_name = run.data.tags.get("mlflow.runName", "")
            tier = run.data.tags.get("tier_name") or run.data.params.get("tier_name") or ""
            family = run.data.tags.get("model_family") or run.data.params.get("model_family") or ""

            haystack = " ".join(
                [experiment.name, run_name, tier, family]
            ).lower()

            if selection["tier"].lower() not in haystack:
                continue

            if selection["family"].lower() not in haystack:
                continue

            score = 0
            if run_name == wanted_run_name:
                score += 10
            if selection["ticker"] in run_name:
                score += 5
            if run.info.status == "FINISHED":
                score += 1

            candidates.append(
                {
                    "score": score,
                    "start_time": run.info.start_time or 0,
                    "experiment": experiment.name,
                    "run_id": run.info.run_id,
                    "run_name": run_name,
                    "status": run.info.status,
                }
            )

    if not candidates:
        return None

    return sorted(
        candidates,
        key=lambda row: (row["score"], row["start_time"]),
        reverse=True,
    )[0]


summary_rows = []

for selection in SELECTED_MODELS:
    run = find_matching_run(selection)

    if run is None:
        summary_rows.append(
            {
                **selection,
                "artifact_kind": "run",
                "rows": None,
                "start_date": None,
                "end_date": None,
                "run_name": None,
                "run_id": None,
                "note": "run not found",
            }
        )
        continue

    for kind, path in artifact_paths(selection).items():
        payload = load_json_artifact(run["run_id"], path)
        summary = summarize_payload(payload, ticker=selection["ticker"])

        note = "ok"
        if isinstance(payload, dict) and "_error" in payload:
            note = payload["_error"]
        elif summary["rows"] == 0:
            note = "empty or ticker not found"

        summary_rows.append(
            {
                **selection,
                "artifact_kind": kind,
                **summary,
                "experiment": run["experiment"],
                "run_name": run["run_name"],
                "run_id": run["run_id"],
                "artifact_path": path,
                "note": note,
            }
        )


# Display without pandas dependency
try:
    import pandas as pd

    date_summary = pd.DataFrame(summary_rows)
    display(
        date_summary[
            [
                "ticker",
                "family",
                "tier",
                "model",
                "artifact_kind",
                "rows",
                "start_date",
                "end_date",
                "run_name",
                "note",
            ]
        ]
    )
except Exception:
    for row in summary_rows:
        print(row)
```

```python
from pathlib import Path
import os
import re
import json

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

from mlflow.tracking import MlflowClient


# ============================================================
# Config
# ============================================================

EXCEL_PATH = Path(
    "/Users/batuhansaylam/Documents/Codex/2026-07-04/a/outputs/"
    "yahooquery_lakehouse_revamp/artifacts/artifact_json_export.xlsx"
)

OUTPUT_DIR = Path(
    "/Users/batuhansaylam/Documents/Codex/2026-07-04/a/outputs/"
    "yahooquery_lakehouse_revamp/notebooks/outputs/paper_top5_forecast_plots"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

TOP_N = 5
EXCLUDE_TIERS = {"tier9"}
TRAIN_TAIL = 120
SAVE_PLOTS = True
USE_TIMESCALE_TRAIN_FALLBACK = True

os.environ.setdefault("MLFLOW_TRACKING_URI", "http://127.0.0.1:5001")
os.environ.setdefault("MLFLOW_S3_ENDPOINT_URL", "http://127.0.0.1:9000")
os.environ.setdefault("AWS_ACCESS_KEY_ID", "admin")
os.environ.setdefault("AWS_SECRET_ACCESS_KEY", "admin1234")
os.environ.setdefault("AWS_DEFAULT_REGION", "us-east-1")

os.environ.setdefault("TIMESCALE_HOST", "127.0.0.1")
os.environ.setdefault("TIMESCALE_PORT", "5432")
os.environ.setdefault("TIMESCALE_DB", "dataops")
os.environ.setdefault("TIMESCALE_USER", "dataops")
os.environ.setdefault("TIMESCALE_PASSWORD", "dataops")

client = MlflowClient(tracking_uri=os.environ["MLFLOW_TRACKING_URI"])


# ============================================================
# Metric Table Loading
# ============================================================

def clean_column_name(column: str) -> str:
    column = str(column).replace("\xa0", " ").strip().lower()
    column = re.sub(r"\s+", "_", column)
    return column


def parse_number(value):
    if pd.isna(value):
        return np.nan
    if isinstance(value, (int, float, np.number)):
        return float(value)

    text = str(value).replace("\xa0", " ").strip().replace(" ", "")

    if "," in text and "." not in text:
        text = text.replace(",", ".")
    else:
        text = text.replace(",", "")

    return pd.to_numeric(text, errors="coerce")


def normalize_metric_frame(frame: pd.DataFrame) -> pd.DataFrame:
    frame = frame.copy()
    frame.columns = [clean_column_name(column) for column in frame.columns]

    rename_map = {
        "_tier": "tier",
        "_family": "family",
        "ticker": "unique_id",
        "symbol": "unique_id",
        "r^2": "r2",
    }
    frame = frame.rename(columns={k: v for k, v in rename_map.items() if k in frame.columns})

    required = {"tier", "unique_id", "model", "mae", "rmse", "mape", "r2"}
    if not required.issubset(frame.columns):
        return pd.DataFrame()

    if "family" not in frame.columns:
        frame["family"] = "unknown"

    for column in ["mae", "rmse", "mape", "r2", "rows"]:
        if column in frame.columns:
            frame[column] = frame[column].map(parse_number)

    frame["tier"] = frame["tier"].astype(str).str.lower().str.strip()
    frame["family"] = frame["family"].astype(str).str.lower().str.strip()
    frame["unique_id"] = frame["unique_id"].astype(str).str.strip()
    frame["model"] = frame["model"].astype(str).str.strip()

    frame = frame[~frame["tier"].isin(EXCLUDE_TIERS)].copy()
    frame = frame.dropna(subset=["unique_id", "model", "mae", "rmse", "mape", "r2"])

    return frame


def load_regression_metrics_from_excel(excel_path: Path) -> pd.DataFrame:
    sheets = pd.read_excel(excel_path, sheet_name=None)

    frames = []
    for sheet_name, sheet_frame in sheets.items():
        normalized = normalize_metric_frame(sheet_frame)
        if normalized.empty:
            continue
        normalized.insert(0, "source_sheet", sheet_name)
        frames.append(normalized)

    if not frames:
        raise ValueError("No regression metric sheet found in Excel.")

    metrics = pd.concat(frames, ignore_index=True, sort=False)

    dedupe_cols = ["tier", "family", "unique_id", "model", "mae", "rmse", "mape", "r2"]
    metrics = metrics.drop_duplicates(subset=[c for c in dedupe_cols if c in metrics.columns])

    return metrics.reset_index(drop=True)


def top_models_by_ticker(metrics: pd.DataFrame, top_n: int = 5) -> pd.DataFrame:
    return (
        metrics.sort_values(
            ["unique_id",  "mape", "mae", "r2",],
            ascending=[True,  True, True, False],
        )
        .groupby("unique_id", observed=True)
        .head(top_n)
        .reset_index(drop=True)
    )


# ============================================================
# MLflow Artifact Helpers
# ============================================================

def safe_name(value: str) -> str:
    return re.sub(r"[^a-zA-Z0-9_.-]+", "_", str(value))


def expected_run_name(row: pd.Series) -> str:
    if row["family"] == "statsforecast":
        return f"stock-close-{row['tier']}-statsforecast"
    if row["family"] == "mlforecast":
        return f"stock-close-{row['tier']}-automlforecast"
    if row["family"] == "pecnet":
        return f"pecnet-{row['tier']}-{safe_name(row['unique_id'])}"
    return ""


def prediction_artifact_path(row: pd.Series) -> str:
    tier = row["tier"]
    ticker = row["unique_id"]

    if row["family"] == "statsforecast":
        return f"statsforecast/{tier}/evaluation/predictions.json"
    if row["family"] == "mlforecast":
        return f"mlforecast/{tier}/evaluation/predictions.json"
    if row["family"] == "pecnet":
        return f"pecnet/{tier}/predictions/{ticker}.json"

    raise ValueError(f"Unsupported family: {row['family']}")


def train_artifact_path(row: pd.Series) -> str:
    tier = row["tier"]
    ticker = row["unique_id"]

    if row["family"] in {"statsforecast", "mlforecast"}:
        return f"{row['family']}/{tier}/datasets/train.json"
    if row["family"] == "pecnet":
        return f"pecnet/{tier}/tickers/{safe_name(ticker)}/datasets/train.json"

    raise ValueError(f"Unsupported family: {row['family']}")


def test_artifact_path(row: pd.Series) -> str:
    tier = row["tier"]
    ticker = row["unique_id"]

    if row["family"] in {"statsforecast", "mlforecast"}:
        return f"{row['family']}/{tier}/datasets/test.json"
    if row["family"] == "pecnet":
        return f"pecnet/{tier}/tickers/{safe_name(ticker)}/datasets/test.json"

    raise ValueError(f"Unsupported family: {row['family']}")


def read_json_table(path: str | Path) -> pd.DataFrame:
    with open(path, "r", encoding="utf-8") as file:
        payload = json.load(file)

    if isinstance(payload, list):
        return pd.DataFrame(payload)

    if isinstance(payload, dict):
        if "columns" in payload and "data" in payload:
            return pd.DataFrame(payload["data"], columns=payload["columns"])
        if isinstance(payload.get("data"), list):
            return pd.DataFrame(payload["data"])
        return pd.DataFrame([payload])

    return pd.DataFrame()


def download_artifact_table(run_id: str, artifact_path: str) -> pd.DataFrame:
    try:
        local_path = client.download_artifacts(run_id, artifact_path)
        return read_json_table(local_path)
    except Exception as exc:
        print(f"Artifact missing: {artifact_path} | {exc}")
        return pd.DataFrame()


def normalize_frame(frame: pd.DataFrame) -> pd.DataFrame:
    if frame.empty:
        return frame

    frame = frame.copy()
    frame.columns = [str(column).strip() for column in frame.columns]

    rename_map = {}
    if "date" in frame.columns and "ds" not in frame.columns:
        rename_map["date"] = "ds"
    if "symbol" in frame.columns and "unique_id" not in frame.columns:
        rename_map["symbol"] = "unique_id"
    if "actual" in frame.columns and "y" not in frame.columns:
        rename_map["actual"] = "y"
    if "target_close" in frame.columns and "y" not in frame.columns:
        rename_map["target_close"] = "y"
    if "close" in frame.columns and "y" not in frame.columns:
        rename_map["close"] = "y"

    frame = frame.rename(columns=rename_map)

    if "ds" in frame.columns:
        frame["ds"] = pd.to_datetime(frame["ds"], errors="coerce", utc=True).dt.tz_localize(None)

    for column in frame.columns:
        if column not in {"unique_id", "ds", "model", "family", "tier"}:
            converted = pd.to_numeric(frame[column], errors="coerce")
            if converted.notna().sum() > 0:
                frame[column] = converted

    sort_cols = [column for column in ["unique_id", "ds"] if column in frame.columns]
    if sort_cols:
        frame = frame.sort_values(sort_cols)

    return frame.reset_index(drop=True)


def ticker_slice(frame: pd.DataFrame, ticker: str) -> pd.DataFrame:
    if frame.empty:
        return frame
    if "unique_id" not in frame.columns:
        return frame
    return frame[frame["unique_id"].astype(str).eq(ticker)].copy()


def find_matching_run(row: pd.Series):
    wanted_name = expected_run_name(row)
    family = row["family"]
    tier = row["tier"]
    ticker = row["unique_id"]

    candidates = []

    for experiment in client.search_experiments():
        runs = client.search_runs(
            experiment_ids=[experiment.experiment_id],
            order_by=["attributes.start_time DESC"],
            max_results=300,
        )

        for run in runs:
            run_name = run.data.tags.get("mlflow.runName", "")
            run_family = run.data.tags.get("model_family") or run.data.params.get("model_family") or ""
            run_tier = run.data.tags.get("tier_name") or run.data.params.get("tier_name") or ""

            haystack = " ".join([experiment.name, run_name, run_family, run_tier]).lower()

            if tier.lower() not in haystack:
                continue
            if family.lower() not in haystack:
                continue

            score = 0
            if run_name == wanted_name:
                score += 20
            if ticker in run_name:
                score += 8
            if run.info.status == "FINISHED":
                score += 2

            candidates.append((score, run.info.start_time or 0, experiment.name, run))

    if not candidates:
        return None, None

    candidates = sorted(candidates, key=lambda item: (item[0], item[1]), reverse=True)
    return candidates[0][3], candidates[0][2]


# ============================================================
# Timescale Train Fallback
# ============================================================

def load_actual_close_from_timescale(ticker: str) -> pd.DataFrame:
    try:
        from sqlalchemy import create_engine, text
    except Exception as exc:
        print(f"SQLAlchemy not available. Timescale fallback disabled: {exc}")
        return pd.DataFrame(columns=["unique_id", "ds", "y"])

    host = os.environ["TIMESCALE_HOST"]
    port = os.environ["TIMESCALE_PORT"]
    db = os.environ["TIMESCALE_DB"]
    user = os.environ["TIMESCALE_USER"]
    password = os.environ["TIMESCALE_PASSWORD"]

    url = f"postgresql+psycopg2://{user}:{password}@{host}:{port}/{db}"

    queries = [
        """
        SELECT unique_id, ds, y
        FROM feature_store.stock_close_model_dataset
        WHERE unique_id = :ticker
        ORDER BY ds
        """,
        """
        SELECT symbol AS unique_id, date AS ds, close AS y
        FROM feature_store.conventional_gap_trading
        WHERE symbol = :ticker
        ORDER BY date
        """,
    ]

    for query in queries:
        try:
            engine = create_engine(url)
            frame = pd.read_sql_query(text(query), engine, params={"ticker": ticker})
            frame = normalize_frame(frame)
            if not frame.empty:
                print(f"Timescale fallback loaded for {ticker}: rows={len(frame)}")
                return frame
        except Exception as exc:
            print(f"Timescale fallback query failed for {ticker}: {exc}")

    return pd.DataFrame(columns=["unique_id", "ds", "y"])


def build_train_tail(
    ticker: str,
    train: pd.DataFrame,
    predictions: pd.DataFrame,
    train_tail: int = 120,
) -> pd.DataFrame:
    if not train.empty and {"ds", "y"}.issubset(train.columns):
        clean_train = train.dropna(subset=["ds", "y"]).sort_values("ds")
        if not clean_train.empty:
            return clean_train.tail(train_tail)

    if USE_TIMESCALE_TRAIN_FALLBACK and not predictions.empty and "ds" in predictions.columns:
        actual_series = load_actual_close_from_timescale(ticker)
        if not actual_series.empty:
            test_start = predictions["ds"].min()
            fallback_train = (
                actual_series[actual_series["ds"] < test_start]
                .dropna(subset=["ds", "y"])
                .sort_values("ds")
                .tail(train_tail)
            )
            if not fallback_train.empty:
                return fallback_train

    return pd.DataFrame(columns=["ds", "y"])


# ============================================================
# Prediction / Interval Helpers
# ============================================================

def prediction_column(frame: pd.DataFrame, model: str) -> str:
    if model in frame.columns:
        return model

    lower_map = {str(column).lower(): column for column in frame.columns}
    if model.lower() in lower_map:
        return lower_map[model.lower()]

    ignored = {"unique_id", "ds", "y", "actual", "cutoff", "model", "family", "tier"}
    interval_tokens = ["-lo-", "-hi-", "_lo_", "_hi_", "_lower_", "_upper_"]

    candidates = []
    for column in frame.columns:
        if column in ignored:
            continue
        if any(token in column for token in interval_tokens):
            continue
        if pd.api.types.is_numeric_dtype(frame[column]):
            candidates.append(column)

    if not candidates:
        raise ValueError(f"No prediction column found. Columns={list(frame.columns)}")

    print(f"Model column {model} not found. Using {candidates[0]}")
    return candidates[0]


def interval_columns(frame: pd.DataFrame, model_col: str) -> dict[int, tuple[str, str]]:
    intervals = {}

    for level in [80, 90, 95]:
        candidates = [
            (f"{model_col}-lo-{level}", f"{model_col}-hi-{level}"),
            (f"{model_col}_lo_{level}", f"{model_col}_hi_{level}"),
            (f"{model_col}_lower_{level}", f"{model_col}_upper_{level}"),
        ]

        for lo_col, hi_col in candidates:
            if lo_col in frame.columns and hi_col in frame.columns:
                intervals[level] = (lo_col, hi_col)
                break

    return intervals


def build_plot_data(row: pd.Series) -> dict:
    run, experiment_name = find_matching_run(row)

    if run is None:
        raise ValueError(
            f"No MLflow run found for {row['unique_id']} | {row['family']} | {row['tier']} | {row['model']}"
        )

    run_id = run.info.run_id

    predictions = normalize_frame(
        download_artifact_table(run_id, prediction_artifact_path(row))
    )
    train = normalize_frame(
        download_artifact_table(run_id, train_artifact_path(row))
    )
    test = normalize_frame(
        download_artifact_table(run_id, test_artifact_path(row))
    )

    ticker = row["unique_id"]
    predictions = ticker_slice(predictions, ticker)
    train = ticker_slice(train, ticker)
    test = ticker_slice(test, ticker)

    if predictions.empty:
        raise ValueError(f"No prediction rows found for {ticker} | {row['family']} | {row['tier']}")

    pred_col = prediction_column(predictions, row["model"])

    if "y" not in predictions.columns or predictions["y"].isna().all():
        if not test.empty and {"unique_id", "ds", "y"}.issubset(test.columns):
            predictions = predictions.drop(columns=["y"], errors="ignore").merge(
                test[["unique_id", "ds", "y"]],
                on=["unique_id", "ds"],
                how="left",
            )

    predictions = predictions.dropna(subset=["ds"]).sort_values("ds").reset_index(drop=True)
    train = train.dropna(subset=["ds"]).sort_values("ds").reset_index(drop=True) if not train.empty else train

    train_tail = build_train_tail(
        ticker=ticker,
        train=train,
        predictions=predictions,
        train_tail=TRAIN_TAIL,
    )

    return {
        "run_id": run_id,
        "experiment_name": experiment_name,
        "run_name": run.data.tags.get("mlflow.runName", ""),
        "selection": row.to_dict(),
        "train": train,
        "train_tail": train_tail,
        "predictions": predictions,
        "pred_col": pred_col,
    }


# ============================================================
# Plotting
# ============================================================

def plot_forecast_like_paper(result: dict) -> None:
    selection = result["selection"]
    ticker = selection["unique_id"]
    family = selection["family"]
    tier = selection["tier"]
    model = selection["model"]

    train = result["train"].copy()
    train_tail = result["train_tail"].copy()
    predictions = result["predictions"].copy()
    pred_col = result["pred_col"]

    fig, ax = plt.subplots(figsize=(18, 5))

    print(
        f"{ticker} | {family} | {tier} | {model} | "
        f"train_rows={len(train)} | train_tail_rows={len(train_tail)} | "
        f"prediction_rows={len(predictions)} | "
        f"prediction_start={predictions['ds'].min().date()} | "
        f"prediction_end={predictions['ds'].max().date()}"
    )

    if not train_tail.empty:
        ax.plot(
            train_tail["ds"],
            train_tail["y"],
            color="#64748b",
            linewidth=2.0,
            label=f"Train actual (last {TRAIN_TAIL})",
        )
    else:
        print(f"WARNING: Train actual could not be plotted for {ticker} | {family} | {tier} | {model}")

    ax.plot(
        predictions["ds"],
        predictions["y"],
        color="#111827",
        linewidth=2.0,
        marker="o",
        markersize=3,
        label="Test actual",
    )

    if family == "statsforecast":
        available_intervals = interval_columns(predictions, pred_col)
        for level, (lo_col, hi_col) in sorted(available_intervals.items(), reverse=True):
            ax.fill_between(
                predictions["ds"],
                predictions[lo_col].astype(float),
                predictions[hi_col].astype(float),
                alpha=0.16 if level >= 95 else 0.24,
                label=f"{level}% prediction interval",
            )

    ax.plot(
        predictions["ds"],
        predictions[pred_col],
        color="#f97316",
        linewidth=2.0,
        marker="x",
        markersize=4,
        label="Prediction",
    )

    ax.axvline(
        predictions["ds"].min(),
        color="#2563eb",
        linestyle="--",
        linewidth=1.2,
        alpha=0.75,
        label="Test start",
    )

    ax.set_title(f"{ticker} | {model} | {tier} | {family}", fontsize=14)
    ax.set_xlabel("Date", fontsize=12)
    ax.set_ylabel("Close", fontsize=12)
    ax.grid(True, alpha=0.22)
    ax.legend(loc="best", frameon=True)
    fig.autofmt_xdate()
    plt.tight_layout()

    if SAVE_PLOTS:
        filename = f"{ticker}_{family}_{tier}_{model}_top5_forecast.png"
        filename = safe_name(filename)
        path = OUTPUT_DIR / filename
        fig.savefig(path, dpi=240, bbox_inches="tight")
        print(f"Saved: {path}")

    plt.show()


# ============================================================
# Execute
# ============================================================

metrics = load_regression_metrics_from_excel(EXCEL_PATH)
top5 = top_models_by_ticker(metrics, TOP_N)

print("Top 5 models per ticker, excluding:", sorted(EXCLUDE_TIERS))
display(
    top5[
        ["tier", "family", "unique_id", "model", "mae", "rmse", "mape", "r2"]
        + (["rows"] if "rows" in top5.columns else [])
    ]
)

loaded_results = []

for _, row in top5.iterrows():
    print("\n" + "=" * 90)
    print(f"Loading and plotting: {row['unique_id']} | {row['family']} | {row['tier']} | {row['model']}")
    print("=" * 90)

    try:
        result = build_plot_data(row)
        loaded_results.append(result)
        plot_forecast_like_paper(result)
    except Exception as exc:
        print(f"Skipped because plot data could not be loaded: {exc}")
```
